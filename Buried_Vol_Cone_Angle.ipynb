```python
# ============================
# INSTALL (run once)
# ============================
!pip install morfeus-ml pandas numpy

# ============================
# IMPORTS
# ============================
import os
import warnings
import pandas as pd

from morfeus import BuriedVolume, ConeAngle, read_xyz

# suppress cone angle fallback warnings
warnings.filterwarnings("ignore")

# ============================
# PATH TO YOUR XYZ FILES
# ============================
xyz_folder = "/content/drive/MyDrive/xyz_files"

xyz_files = [f for f in os.listdir(xyz_folder) if f.endswith(".xyz")]

# ============================
# FUNCTION: AUTO DETECT METAL
# ============================
def get_core_atoms(mol, metal="Pd"):
    """
    Automatically finds metal atom index in xyz.
    Change metal="Ni", "Pt", etc. if needed.
    """
    atoms = mol.atoms
    core = [i for i, a in enumerate(atoms) if a.symbol == metal]

    if len(core) == 0:
        raise ValueError(f"{metal} not found in structure")

    return core

# ============================
# STORAGE
# ============================
results = []

# ============================
# MAIN LOOP
# ============================
for file in xyz_files:
    path = os.path.join(xyz_folder, file)

    try:
        mol = read_xyz(path)

        # ---- core atom detection ----
        core_atoms = get_core_atoms(mol, metal="Pd")  # change if needed

        # ---- Buried Volume ----
        bv = BuriedVolume(
            mol,
            core=core_atoms,
            radius=3.5
        )

        vbur = bv.percent_buried_volume()
        distal = bv.distal_volume

        # ---- Cone Angle ----
        ca = ConeAngle(mol)
        cone = ca.cone_angle

        # ---- store results ----
        results.append({
            "file": file,
            "Vbur (%)": vbur,
            "Distal Volume": distal,
            "Cone Angle (deg)": cone
        })

        print(f"Processed: {file}")

    except Exception as e:
        print(f"Error in {file}: {e}")

# ============================
# SAVE RESULTS
# ============================
df = pd.DataFrame(results)
output_file = os.path.join(xyz_folder, "steric_results.csv")
df.to_csv(output_file, index=False)

print("\nDONE ✔")
print("Saved at:", output_file)
```
