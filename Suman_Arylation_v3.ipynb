```python
# ============================================================
# Pd-CATALYZED γ-ARYLATION
# COMPLETE LIGAND-DESCRIPTOR ANALYSIS
# CORRELATION OF ALL DESCRIPTORS VS REACTION YIELD
#
# ============================================================
# ATOM CONVENTION
# ============================================================
#
# Pd atom         = 1
# Substrate atoms = 2–16
# Ligand atoms    = 17–last atom
#
# ============================================================
# DONOR RULE
# ============================================================
#
# Search ONLY ligand atoms (17 onward).
# Candidate donor elements: N and O.
# Accepted donor:
#     2.00 Å <= Pd–N/O distance <= 2.30 Å
#
# ============================================================
# CORRELATION RULE
# ============================================================
#
# Every numerical molecular descriptor is correlated with:
#     Gamma_Arylation_Yield_percent
#
# No descriptor-vs-descriptor correlation matrix is calculated.
#
# ============================================================
# IMPORTANT xTB NOTE
# ============================================================
#
# Set MOLECULAR_CHARGE and N_UNPAIRED to the correct charge
# and spin state of your Pd complexes.
#
# If every ligand complex does NOT have the same charge/spin,
# use a metadata table for charge and multiplicity instead.
# ============================================================


from google.colab import drive
drive.mount('/content/drive', force_remount=True)

# ============================================================
# 1. INSTALL REQUIRED PYTHON PACKAGES
# ============================================================

!pip install -q morfeus-ml pandas numpy openpyxl matplotlib seaborn scipy


# ============================================================
# 2. INSTALL xTB EXECUTABLE
#
# Required for Morfeus XTB calculations.
# ============================================================

# ============================================================
# INSTALL xTB IN GOOGLE COLAB FROM OFFICIAL PRECOMPILED BINARY
# ============================================================

!rm -rf xtb-dist xtb-*.tar.xz

!wget -q --show-progress \
    https://github.com/grimme-lab/xtb/releases/download/v6.7.1/xtb-6.7.1-linux-x86_64.tar.xz \
    -O xtb-6.7.1-linux-x86_64.tar.xz

!tar -xJf xtb-6.7.1-linux-x86_64.tar.xz

!find . -type f -name xtb -path "*bin/xtb"

!chmod +x ./xtb-dist/bin/xtb

import os
os.environ["PATH"] = (
    os.path.abspath("./xtb-dist/bin")
    + ":"
    + os.environ["PATH"]
)

os.environ["XTBHOME"] = os.path.abspath("./xtb-dist")

!which xtb
!xtb --version


# ============================================================
# 3. IMPORTS
# ============================================================

import os
import re
import glob
import shutil
import warnings
import subprocess

import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns

from scipy.stats import pearsonr, spearmanr, pointbiserialr

from morfeus import (
    Sterimol,
    BuriedVolume,
    BiteAngle,
    Dispersion,
    SASA,
    XTB
)


warnings.filterwarnings("ignore")


# ============================================================
# 4. PATHS
# ============================================================

XYZ_FOLDER = "/content/drive/MyDrive/Suman_files"

OUTPUT_EXCEL = (
    "/content/drive/MyDrive/"
    "Pd_Gamma_Arylation_All_Descriptors_vs_Yield.xlsx"
)

PEARSON_FIG = (
    "/content/drive/MyDrive/"
    "Pd_Gamma_Arylation_Pearson_Descriptors_vs_Yield.png"
)

SPEARMAN_FIG = (
    "/content/drive/MyDrive/"
    "Pd_Gamma_Arylation_Spearman_Descriptors_vs_Yield.png"
)

SCATTER_FOLDER = (
    "/content/drive/MyDrive/"
    "Pd_Gamma_Arylation_Descriptor_vs_Yield_Scatterplots"
)


# ============================================================
# 5. ATOM DEFINITIONS
# ============================================================

PD_ATOM = 1

SUBSTRATE_START = 2
SUBSTRATE_END = 16

LIGAND_START = 17


# ============================================================
# 6. DONOR-DETECTION SETTINGS
# ============================================================

DONOR_ELEMENTS = {"N", "O"}

PD_DONOR_MIN_DISTANCE = 2.00
PD_DONOR_MAX_DISTANCE = 2.30


# ============================================================
# 7. MORFEUS DESCRIPTOR SETTINGS
# ============================================================

VBUR_RADIUS = 3.5

SASA_PROBE_RADIUS = 1.4

STERIMOL_RADII_TYPE = "crc"
VBUR_RADII_TYPE = "bondi"

STERIMOL_ROTATION_VECTORS = 3600


# ============================================================
# 8. xTB SETTINGS
#
# IMPORTANT:
# Change these if your Pd complexes are charged/open-shell.
# ============================================================

MOLECULAR_CHARGE = 0

N_UNPAIRED = 0

XTB_METHOD = 2

REACTION_SOLVENT = "hfip"


# ============================================================
# 9. GAMMA-ARYLATION YIELDS
#
# Yield values are percentages.
# The ligand identifier must occur in each XYZ filename:
#
# L7.xyz
# Pd_L7.xyz
# Pd_L7_optimized.xyz
# L16_complex.xyz
# ============================================================

yield_dict = {

    "L7": 0,
    "L8": 0,
    "L9": 0,
    "L10": 0,
    "L11": 0,
    "L12": 0,
    "L13": 0,
    "L14": 0,
    "L15": 0,

    "L16": 12,
    "L17": 53,
    "L18": 46,
    "L19": 67,
    "L20": 40,
    "L21": 31,

    "L22": 0,
    "L23": 0,
    "L24": 0,
    "L25": 0
}

YIELD_COLUMN = "Gamma_Arylation_Yield_percent"


# ============================================================
# 10. VALIDATION SETTINGS
#
# Broad screening intervals only.
# A warning does NOT automatically invalidate a descriptor.
# ============================================================

BITE_ANGLE_RANGE = (40.0, 180.0)

BITE_ANGLE_TOLERANCE = 0.01

STERIMOL_B1_RANGE = (0.0, 20.0)
STERIMOL_B5_RANGE = (0.0, 30.0)
STERIMOL_L_RANGE = (0.0, 50.0)

VBUR_RANGE = (0.0, 100.0)

SASA_AREA_RANGE = (0.0, 5000.0)
SASA_VOLUME_RANGE = (0.0, 50000.0)

DISPERSION_AREA_RANGE = (0.0, 5000.0)
DISPERSION_VOLUME_RANGE = (0.0, 50000.0)
PINT_RANGE = (0.0, 1000.0)

LIGAND_EXTENSION_RANGE = (0.0, 100.0)

PYRAMIDALIZATION_RANGE = (-30.0, 100.0)

XTB_CHARGE_SUM_TOLERANCE = 0.10

REPEAT_CALCULATION_TOLERANCE = 1e-8

MIN_CORRELATION_OBSERVATIONS = 4

TOP_N_SCATTER_PLOTS = 10


# ============================================================
# 11. CREATE OUTPUT FOLDER FOR SCATTER PLOTS
# ============================================================

os.makedirs(
    SCATTER_FOLDER,
    exist_ok=True
)


# ============================================================
# 12. CHECK WHETHER xTB EXECUTABLE IS AVAILABLE
# ============================================================

def check_xtb_available():

    xtb_path = shutil.which("xtb")

    if xtb_path is None:
        return False, "xtb executable not found in PATH."

    try:

        version_text = subprocess.check_output(
            ["xtb", "--version"],
            stderr=subprocess.STDOUT,
            text=True,
            timeout=30
        ).strip()

        return True, version_text

    except Exception as error:

        return (
            False,
            f"xTB executable check failed: {error}"
        )


XTB_AVAILABLE, XTB_INFO = check_xtb_available()

print("=" * 90)
print("xTB CHECK")
print("=" * 90)
print(f"xTB available: {XTB_AVAILABLE}")
print(f"xTB information: {XTB_INFO}")
print("=" * 90)


# ============================================================
# 13. XYZ READER
# ============================================================

def read_xyz_file(filename):

    with open(filename, "r") as file:
        raw_lines = file.readlines()

    lines = [
        line.strip()
        for line in raw_lines
        if line.strip()
    ]

    if len(lines) < 3:
        raise ValueError(
            "XYZ file is too short or malformed."
        )

    try:
        n_atoms = int(lines[0])
    except ValueError:
        raise ValueError(
            "The first XYZ line must contain the atom count."
        )

    coordinate_lines = lines[2:2 + n_atoms]

    if len(coordinate_lines) != n_atoms:
        raise ValueError(
            f"XYZ declares {n_atoms} atoms but provides "
            f"{len(coordinate_lines)} coordinate lines."
        )

    elements = []
    coordinates = []

    for line_number, line in enumerate(
        coordinate_lines,
        start=3
    ):

        parts = line.split()

        if len(parts) < 4:
            raise ValueError(
                f"Malformed XYZ coordinate line {line_number}: {line}"
            )

        try:
            x = float(parts[1])
            y = float(parts[2])
            z = float(parts[3])
        except ValueError:
            raise ValueError(
                f"Non-numeric coordinate at line {line_number}: {line}"
            )

        elements.append(parts[0])
        coordinates.append([x, y, z])

    coordinates = np.array(
        coordinates,
        dtype=float
    )

    return elements, coordinates


# ============================================================
# 14. BASIC GEOMETRY FUNCTIONS
# ============================================================

def distance(coord1, coord2):

    return float(
        np.linalg.norm(coord1 - coord2)
    )


def manual_angle(atom_a, center_atom, atom_c):
    """
    Returns angle A-center-C in degrees.
    Used to cross-check Morfeus BiteAngle.
    """

    vector_a = atom_a - center_atom
    vector_c = atom_c - center_atom

    denominator = (
        np.linalg.norm(vector_a)
        * np.linalg.norm(vector_c)
    )

    if denominator == 0:
        return np.nan

    cosine = np.dot(vector_a, vector_c) / denominator

    cosine = np.clip(
        cosine,
        -1.0,
        1.0
    )

    return float(
        np.degrees(
            np.arccos(cosine)
        )
    )


# ============================================================
# 15. GENERAL UTILITY FUNCTIONS
# ============================================================

def safe_mean(values):

    clean_values = [
        value
        for value in values
        if not pd.isna(value)
    ]

    if len(clean_values) == 0:
        return np.nan

    return float(np.mean(clean_values))


def is_finite(value):

    try:
        return np.isfinite(float(value))
    except Exception:
        return False


def format_contacts(contacts):

    if len(contacts) == 0:
        return "None"

    return "; ".join(
        f"{element}{atom_id}:{dist_value:.3f}"
        for atom_id, element, dist_value in contacts
    )


def safe_filename(text):

    return re.sub(
        r"[^A-Za-z0-9_.-]+",
        "_",
        str(text)
    )


def append_validation(
    validation_records,
    descriptor,
    value,
    status,
    message
):

    validation_records.append({
        "Descriptor": descriptor,
        "Value": value,
        "Status": status,
        "Message": message
    })


def validate_numeric_range(
    descriptor,
    value,
    lower,
    upper
):

    if pd.isna(value):

        return (
            "NOT_APPLICABLE",
            f"{descriptor}: no value available."
        )

    if not is_finite(value):

        return (
            "FAIL",
            f"{descriptor}: non-finite value."
        )

    if value < lower or value > upper:

        return (
            "WARNING",
            f"{descriptor}: {value:.6f} is outside "
            f"the screening interval {lower:.3f}–{upper:.3f}."
        )

    return (
        "PASS",
        f"{descriptor}: {value:.6f} is within "
        f"the screening interval."
    )


def merge_status(statuses):

    if "FAIL" in statuses:
        return "FAIL"

    if "WARNING" in statuses:
        return "WARNING"

    if "PASS" in statuses:
        return "PASS"

    return "NOT_APPLICABLE"


# ============================================================
# 16. LIGAND ID AND ATOM LIST FUNCTIONS
# ============================================================

def extract_ligand_id(filename):
    """
    Examples:
    L7.xyz              -> L7
    Pd_L7.xyz           -> L7
    Pd_L16_optimized.xyz -> L16
    """

    basename = os.path.basename(filename)

    stem = os.path.splitext(basename)[0]

    matches = re.findall(
        r"(?<![A-Za-z0-9])(L\d+)(?![A-Za-z0-9])",
        stem,
        flags=re.IGNORECASE
    )

    if len(matches) == 0:
        return np.nan

    unique_matches = sorted(
        set(match.upper() for match in matches)
    )

    if len(unique_matches) == 1:
        return unique_matches[0]

    return np.nan


def get_ligand_atoms(n_atoms):

    if n_atoms < LIGAND_START:
        return []

    return list(
        range(LIGAND_START, n_atoms + 1)
    )


# ============================================================
# 17. STRUCTURE VALIDATION
# ============================================================

def validate_structure(
    elements,
    coordinates
):

    validations = []

    n_atoms = len(elements)

    if n_atoms == len(coordinates):

        append_validation(
            validations,
            "XYZ_Atom_Count",
            n_atoms,
            "PASS",
            "Atom labels and coordinate rows match."
        )

    else:

        append_validation(
            validations,
            "XYZ_Atom_Count",
            n_atoms,
            "FAIL",
            "Atom labels and coordinate rows do not match."
        )

    if n_atoms >= LIGAND_START:

        append_validation(
            validations,
            "Ligand_Boundary",
            LIGAND_START,
            "PASS",
            f"Ligand atom range is {LIGAND_START}–{n_atoms}."
        )

    else:

        append_validation(
            validations,
            "Ligand_Boundary",
            n_atoms,
            "FAIL",
            f"Structure contains fewer than {LIGAND_START} atoms."
        )

    if elements[PD_ATOM - 1] == "Pd":

        append_validation(
            validations,
            "Pd_Identity",
            "Pd",
            "PASS",
            f"Atom {PD_ATOM} is Pd."
        )

    else:

        append_validation(
            validations,
            "Pd_Identity",
            elements[PD_ATOM - 1],
            "FAIL",
            f"Atom {PD_ATOM} is not Pd."
        )

    if np.all(np.isfinite(coordinates)):

        append_validation(
            validations,
            "Coordinate_Finiteness",
            "Finite",
            "PASS",
            "All coordinates are finite."
        )

    else:

        append_validation(
            validations,
            "Coordinate_Finiteness",
            np.nan,
            "FAIL",
            "One or more coordinates are NaN or infinite."
        )

    return validations


# ============================================================
# 18. FIND Pd-COORDINATED LIGAND N/O DONORS
# ============================================================

def find_coordinated_atoms(
    elements,
    coordinates
):

    pd_xyz = coordinates[PD_ATOM - 1]

    ligand_atoms = get_ligand_atoms(
        len(elements)
    )

    accepted_donors = []
    all_ligand_no_contacts = []

    for atom_id in ligand_atoms:

        element = elements[atom_id - 1]

        if element not in DONOR_ELEMENTS:
            continue

        donor_xyz = coordinates[atom_id - 1]

        pd_donor_distance = distance(
            pd_xyz,
            donor_xyz
        )

        contact = (
            atom_id,
            element,
            pd_donor_distance
        )

        all_ligand_no_contacts.append(contact)

        if (
            PD_DONOR_MIN_DISTANCE
            <= pd_donor_distance
            <= PD_DONOR_MAX_DISTANCE
        ):
            accepted_donors.append(contact)

    accepted_donors.sort(
        key=lambda x: x[2]
    )

    all_ligand_no_contacts.sort(
        key=lambda x: x[2]
    )

    return accepted_donors, all_ligand_no_contacts


# ============================================================
# 19. DONOR VALIDATION
# ============================================================

def validate_donors(
    donor_atoms,
    all_ligand_no_contacts
):

    validations = []

    n_donors = len(donor_atoms)

    if n_donors == 0:

        append_validation(
            validations,
            "Donor_Assignment",
            n_donors,
            "FAIL",
            "No ligand N/O donor was found at 2.00–2.30 Å from Pd."
        )

    elif n_donors == 1:

        append_validation(
            validations,
            "Donor_Assignment",
            n_donors,
            "WARNING",
            "One ligand N/O donor found. This may be valid for a monodentate ligand."
        )

    elif n_donors == 2:

        append_validation(
            validations,
            "Donor_Assignment",
            n_donors,
            "PASS",
            "Two ligand N/O donors found within the defined Pd-distance cutoff."
        )

    else:

        append_validation(
            validations,
            "Donor_Assignment",
            n_donors,
            "WARNING",
            "More than two ligand N/O atoms lie within 2.00–2.30 Å of Pd."
        )

    for atom_id, element, distance_value in donor_atoms:

        status, message = validate_numeric_range(
            f"Pd_{element}{atom_id}_Distance",
            distance_value,
            PD_DONOR_MIN_DISTANCE,
            PD_DONOR_MAX_DISTANCE
        )

        append_validation(
            validations,
            f"Pd_{element}{atom_id}_Distance",
            distance_value,
            status,
            message
        )

    nearby_contacts = [
        contact
        for contact in all_ligand_no_contacts
        if 1.80 <= contact[2] <= 2.60
    ]

    if len(nearby_contacts) > len(donor_atoms):

        append_validation(
            validations,
            "Additional_Close_N_O_Contacts",
            len(nearby_contacts),
            "WARNING",
            "Additional ligand N/O atom(s) occur within 1.80–2.60 Å of Pd."
        )

    else:

        append_validation(
            validations,
            "Additional_Close_N_O_Contacts",
            len(nearby_contacts),
            "PASS",
            "No additional close ligand N/O contacts were found."
        )

    return validations


# ============================================================
# 20. COORDINATION DESCRIPTORS
# ============================================================

def calculate_coordination_descriptors(
    donor_atoms
):

    result = {
        "Donor1_Atom": np.nan,
        "Donor1_Element": np.nan,
        "Pd_Donor1_Distance": np.nan,

        "Donor2_Atom": np.nan,
        "Donor2_Element": np.nan,
        "Pd_Donor2_Distance": np.nan,

        "Average_Pd_Donor_Distance": np.nan,
        "Delta_Pd_Donor_Distance": np.nan
    }

    if len(donor_atoms) >= 1:

        donor1_atom, donor1_element, donor1_distance = (
            donor_atoms[0]
        )

        result["Donor1_Atom"] = donor1_atom
        result["Donor1_Element"] = donor1_element
        result["Pd_Donor1_Distance"] = donor1_distance

    if len(donor_atoms) >= 2:

        donor2_atom, donor2_element, donor2_distance = (
            donor_atoms[1]
        )

        result["Donor2_Atom"] = donor2_atom
        result["Donor2_Element"] = donor2_element
        result["Pd_Donor2_Distance"] = donor2_distance

        result["Average_Pd_Donor_Distance"] = (
            donor1_distance + donor2_distance
        ) / 2.0

        result["Delta_Pd_Donor_Distance"] = abs(
            donor1_distance - donor2_distance
        )

    return result


# ============================================================
# 21. BITE ANGLE: MORFEUS + MANUAL CROSS-CHECK
# ============================================================

def calculate_bite_angle(
    coordinates,
    donor_atoms
):

    result = {
        "Bite_Angle_Morfeus": np.nan,
        "Bite_Angle_Manual": np.nan,
        "Bite_Angle_Difference": np.nan,
        "Bite_Angle_Inverted": np.nan
    }

    if len(donor_atoms) != 2:
        return result

    donor1_atom = donor_atoms[0][0]
    donor2_atom = donor_atoms[1][0]

    pd_xyz = coordinates[PD_ATOM - 1]
    donor1_xyz = coordinates[donor1_atom - 1]
    donor2_xyz = coordinates[donor2_atom - 1]

    manual_bite_angle = manual_angle(
        donor1_xyz,
        pd_xyz,
        donor2_xyz
    )

    result["Bite_Angle_Manual"] = manual_bite_angle

    try:

        bite = BiteAngle(
            coordinates,
            PD_ATOM,
            donor1_atom,
            donor2_atom
        )

        morfeus_bite_angle = float(
            bite.angle
        )

        result["Bite_Angle_Morfeus"] = (
            morfeus_bite_angle
        )

        result["Bite_Angle_Inverted"] = bool(
            bite.inverted
        )

        if (
            not pd.isna(morfeus_bite_angle)
            and not pd.isna(manual_bite_angle)
        ):

            result["Bite_Angle_Difference"] = abs(
                morfeus_bite_angle
                - manual_bite_angle
            )

    except Exception as error:

        print(f"BiteAngle failed: {error}")

    return result


def validate_bite_angle(
    bite_result,
    donor_atoms
):

    validations = []

    if len(donor_atoms) != 2:

        append_validation(
            validations,
            "Bite_Angle",
            np.nan,
            "NOT_APPLICABLE",
            "Bite angle requires exactly two accepted ligand donors."
        )

        return validations

    morfeus_value = bite_result[
        "Bite_Angle_Morfeus"
    ]

    manual_value = bite_result[
        "Bite_Angle_Manual"
    ]

    difference = bite_result[
        "Bite_Angle_Difference"
    ]

    inverted = bite_result[
        "Bite_Angle_Inverted"
    ]

    for descriptor, value in [
        ("Bite_Angle_Morfeus", morfeus_value),
        ("Bite_Angle_Manual", manual_value)
    ]:

        status, message = validate_numeric_range(
            descriptor,
            value,
            BITE_ANGLE_RANGE[0],
            BITE_ANGLE_RANGE[1]
        )

        append_validation(
            validations,
            descriptor,
            value,
            status,
            message
        )

    if pd.isna(difference):

        append_validation(
            validations,
            "Bite_Angle_Crosscheck",
            np.nan,
            "FAIL",
            "Morfeus/manual bite-angle comparison could not be completed."
        )

    elif difference <= BITE_ANGLE_TOLERANCE:

        append_validation(
            validations,
            "Bite_Angle_Crosscheck",
            difference,
            "PASS",
            f"Morfeus/manual bite-angle difference is "
            f"{difference:.6f}°, within tolerance."
        )

    else:

        append_validation(
            validations,
            "Bite_Angle_Crosscheck",
            difference,
            "WARNING",
            f"Morfeus/manual bite-angle difference is "
            f"{difference:.6f}°, above the tolerance."
        )

    if inverted is True:

        append_validation(
            validations,
            "Bite_Angle_Inverted",
            morfeus_value,
            "WARNING",
            "Morfeus reported an inverted bite angle; inspect orientation."
        )

    else:

        append_validation(
            validations,
            "Bite_Angle_Inverted",
            inverted,
            "PASS",
            "No inverted bite angle reported."
        )

    return validations


# ============================================================
# 22. STERIMOL
# ============================================================

def calculate_sterimol_for_donor(
    elements,
    coordinates,
    ligand_atoms,
    donor_atom_id
):

    if donor_atom_id is None or pd.isna(donor_atom_id):
        return np.nan, np.nan, np.nan

    donor_atom_id = int(donor_atom_id)

    if donor_atom_id not in ligand_atoms:
        return np.nan, np.nan, np.nan

    keep_atoms = [PD_ATOM] + ligand_atoms

    keep_indices = [
        atom_id - 1
        for atom_id in keep_atoms
    ]

    sub_elements = [
        elements[index]
        for index in keep_indices
    ]

    sub_coordinates = coordinates[
        keep_indices
    ]

    attached_index = (
        keep_atoms.index(donor_atom_id) + 1
    )

    try:

        sterimol = Sterimol(
            sub_elements,
            sub_coordinates,
            dummy_index=1,
            attached_index=attached_index,
            radii_type=STERIMOL_RADII_TYPE,
            n_rot_vectors=STERIMOL_ROTATION_VECTORS
        )

        return (
            float(sterimol.B_1_value),
            float(sterimol.B_5_value),
            float(sterimol.L_value)
        )

    except Exception as error:

        print(
            f"Sterimol failed for donor atom "
            f"{donor_atom_id}: {error}"
        )

        return np.nan, np.nan, np.nan


def validate_sterimol(
    B1,
    B5,
    L,
    donor_label
):

    validations = []

    if (
        pd.isna(B1)
        or pd.isna(B5)
        or pd.isna(L)
    ):

        append_validation(
            validations,
            f"Sterimol_{donor_label}",
            np.nan,
            "NOT_APPLICABLE",
            "Sterimol was not calculated because a donor axis was unavailable."
        )

        return validations

    for descriptor, value, bounds in [
        (
            f"Sterimol_B1_{donor_label}",
            B1,
            STERIMOL_B1_RANGE
        ),
        (
            f"Sterimol_B5_{donor_label}",
            B5,
            STERIMOL_B5_RANGE
        ),
        (
            f"Sterimol_L_{donor_label}",
            L,
            STERIMOL_L_RANGE
        )
    ]:

        status, message = validate_numeric_range(
            descriptor,
            value,
            bounds[0],
            bounds[1]
        )

        append_validation(
            validations,
            descriptor,
            value,
            status,
            message
        )

    if B1 <= B5:

        append_validation(
            validations,
            f"Sterimol_B1_leq_B5_{donor_label}",
            B5 - B1,
            "PASS",
            "Sterimol internal validation passed: B1 <= B5."
        )

    else:

        append_validation(
            validations,
            f"Sterimol_B1_leq_B5_{donor_label}",
            B1 - B5,
            "FAIL",
            "Sterimol internal validation failed: B1 > B5."
        )

    return validations


# ============================================================
# 23. LIGAND-ONLY %VBUR
# ============================================================

def calculate_ligand_vbur(
    elements,
    coordinates,
    ligand_atoms
):

    keep_atoms = [PD_ATOM] + ligand_atoms

    excluded_atoms = [
        atom_id
        for atom_id in range(1, len(elements) + 1)
        if atom_id not in keep_atoms
    ]

    try:

        buried = BuriedVolume(
            elements,
            coordinates,
            metal_index=PD_ATOM,
            excluded_atoms=excluded_atoms,
            radius=VBUR_RADIUS,
            include_hs=False,
            radii_type=VBUR_RADII_TYPE
        )

        return float(
            buried.fraction_buried_volume * 100.0
        )

    except Exception as error:

        print(f"%Vbur failed: {error}")

        return np.nan


def validate_vbur(
    vbur,
    elements,
    coordinates,
    ligand_atoms
):

    validations = []

    status, message = validate_numeric_range(
        "Ligand_Vbur_3.5A_percent",
        vbur,
        VBUR_RANGE[0],
        VBUR_RANGE[1]
    )

    append_validation(
        validations,
        "Ligand_Vbur_3.5A_percent",
        vbur,
        status,
        message
    )

    if pd.isna(vbur):
        return validations

    repeated_vbur = calculate_ligand_vbur(
        elements,
        coordinates,
        ligand_atoms
    )

    if pd.isna(repeated_vbur):

        append_validation(
            validations,
            "Ligand_Vbur_Reproducibility",
            np.nan,
            "FAIL",
            "Repeated %Vbur calculation failed."
        )

    else:

        difference = abs(
            vbur - repeated_vbur
        )

        if difference <= REPEAT_CALCULATION_TOLERANCE:

            append_validation(
                validations,
                "Ligand_Vbur_Reproducibility",
                difference,
                "PASS",
                "Repeated %Vbur calculation gives the identical result."
            )

        else:

            append_validation(
                validations,
                "Ligand_Vbur_Reproducibility",
                difference,
                "WARNING",
                "Repeated %Vbur calculation differs from first result."
            )

    return validations


# ============================================================
# 24. LIGAND EXTENSION
# ============================================================

def calculate_ligand_extension(
    coordinates,
    ligand_atoms,
    donor_atom_id
):

    if donor_atom_id is None or pd.isna(donor_atom_id):
        return np.nan

    if len(ligand_atoms) == 0:
        return np.nan

    donor_xyz = coordinates[
        int(donor_atom_id) - 1
    ]

    all_distances = [
        distance(
            donor_xyz,
            coordinates[atom_id - 1]
        )
        for atom_id in ligand_atoms
    ]

    return float(max(all_distances))


def validate_ligand_extension(
    extension,
    donor_label
):

    status, message = validate_numeric_range(
        f"Ligand_Max_Extension_{donor_label}",
        extension,
        LIGAND_EXTENSION_RANGE[0],
        LIGAND_EXTENSION_RANGE[1]
    )

    return [{
        "Descriptor": f"Ligand_Max_Extension_{donor_label}",
        "Value": extension,
        "Status": status,
        "Message": message
    }]


# ============================================================
# 25. DONOR PYRAMIDALIZATION
#
# Exploratory descriptor only.
# ============================================================

def donor_pyramidalization(
    coordinates,
    ligand_atoms,
    donor_atom_id
):

    if donor_atom_id is None or pd.isna(donor_atom_id):
        return np.nan

    donor_atom_id = int(donor_atom_id)
    donor_index = donor_atom_id - 1

    neighbor_distances = []

    for atom_id in ligand_atoms:

        atom_index = atom_id - 1

        if atom_index == donor_index:
            continue

        neighbor_distances.append(
            (
                atom_index,
                distance(
                    coordinates[donor_index],
                    coordinates[atom_index]
                )
            )
        )

    neighbor_distances.sort(
        key=lambda x: x[1]
    )

    neighbors = [
        atom_index
        for atom_index, _ in neighbor_distances[:3]
    ]

    if len(neighbors) < 3:
        return np.nan

    atom_a = coordinates[neighbors[0]]
    donor = coordinates[donor_index]
    atom_c = coordinates[neighbors[1]]
    atom_d = coordinates[neighbors[2]]

    angle_1 = manual_angle(
        atom_a,
        donor,
        atom_c
    )

    angle_2 = manual_angle(
        atom_a,
        donor,
        atom_d
    )

    angle_3 = manual_angle(
        atom_c,
        donor,
        atom_d
    )

    if any(
        pd.isna(value)
        for value in [angle_1, angle_2, angle_3]
    ):
        return np.nan

    return float(
        360.0
        - angle_1
        - angle_2
        - angle_3
    )


def validate_pyramidalization(
    pyramidalization,
    donor_label,
    donor_element
):

    validations = []

    status, message = validate_numeric_range(
        f"Donor_Pyramidalization_{donor_label}",
        pyramidalization,
        PYRAMIDALIZATION_RANGE[0],
        PYRAMIDALIZATION_RANGE[1]
    )

    append_validation(
        validations,
        f"Donor_Pyramidalization_{donor_label}",
        pyramidalization,
        status,
        message
    )

    if (
        donor_element in {"N", "O"}
        and not pd.isna(pyramidalization)
    ):

        append_validation(
            validations,
            f"Pyramidalization_Interpretation_{donor_label}",
            donor_element,
            "WARNING",
            "Pyramidalization is exploratory and may not be chemically meaningful for aromatic/planar N or O donors."
        )

    return validations


# ============================================================
# 26. LIGAND-ONLY SASA
#
# SASA is calculated on Pd + ligand only.
# Thus substrate atoms do not mask ligand surface area.
# ============================================================

def calculate_ligand_sasa(
    elements,
    coordinates,
    ligand_atoms,
    donor_atoms
):

    result = {
        "Ligand_SASA_Area_A2": np.nan,
        "Ligand_SASA_Volume_A3": np.nan,
        "Pd_SASA_A2": np.nan,
        "Ligand_Atomic_SASA_Sum_A2": np.nan,
        "Ligand_Mean_Atomic_SASA_A2": np.nan,
        "Donor1_SASA_A2": np.nan,
        "Donor2_SASA_A2": np.nan
    }

    if len(ligand_atoms) == 0:
        return result

    keep_atoms = [PD_ATOM] + ligand_atoms

    keep_indices = [
        atom_id - 1
        for atom_id in keep_atoms
    ]

    sub_elements = [
        elements[index]
        for index in keep_indices
    ]

    sub_coordinates = coordinates[
        keep_indices
    ]

    try:

        sasa = SASA(
            sub_elements,
            sub_coordinates,
            probe_radius=SASA_PROBE_RADIUS
        )

        result["Ligand_SASA_Area_A2"] = float(
            sasa.area
        )

        result["Ligand_SASA_Volume_A3"] = float(
            sasa.volume
        )

        result["Pd_SASA_A2"] = float(
            sasa.atom_areas.get(1, np.nan)
        )

        ligand_atom_areas = []

        for reduced_index in range(
            2,
            len(keep_atoms) + 1
        ):

            ligand_atom_areas.append(
                sasa.atom_areas.get(
                    reduced_index,
                    np.nan
                )
            )

        ligand_atom_areas = [
            value
            for value in ligand_atom_areas
            if not pd.isna(value)
        ]

        result["Ligand_Atomic_SASA_Sum_A2"] = float(
            np.sum(ligand_atom_areas)
        )

        result["Ligand_Mean_Atomic_SASA_A2"] = safe_mean(
            ligand_atom_areas
        )

        if len(donor_atoms) >= 1:

            donor1_atom = donor_atoms[0][0]

            donor1_reduced = (
                keep_atoms.index(donor1_atom) + 1
            )

            result["Donor1_SASA_A2"] = float(
                sasa.atom_areas.get(
                    donor1_reduced,
                    np.nan
                )
            )

        if len(donor_atoms) >= 2:

            donor2_atom = donor_atoms[1][0]

            donor2_reduced = (
                keep_atoms.index(donor2_atom) + 1
            )

            result["Donor2_SASA_A2"] = float(
                sasa.atom_areas.get(
                    donor2_reduced,
                    np.nan
                )
            )

    except Exception as error:

        print(f"SASA failed: {error}")

    return result


def validate_sasa(
    sasa_result,
    elements,
    coordinates,
    ligand_atoms,
    donor_atoms
):

    validations = []

    for descriptor, value, bounds in [
        (
            "Ligand_SASA_Area_A2",
            sasa_result["Ligand_SASA_Area_A2"],
            SASA_AREA_RANGE
        ),
        (
            "Ligand_SASA_Volume_A3",
            sasa_result["Ligand_SASA_Volume_A3"],
            SASA_VOLUME_RANGE
        )
    ]:

        status, message = validate_numeric_range(
            descriptor,
            value,
            bounds[0],
            bounds[1]
        )

        append_validation(
            validations,
            descriptor,
            value,
            status,
            message
        )

    total_area = sasa_result["Ligand_SASA_Area_A2"]
    atomic_area_sum = sasa_result["Ligand_Atomic_SASA_Sum_A2"]
    pd_area = sasa_result["Pd_SASA_A2"]

    if (
        not pd.isna(total_area)
        and not pd.isna(atomic_area_sum)
        and not pd.isna(pd_area)
    ):

        difference = abs(
            total_area
            - (atomic_area_sum + pd_area)
        )

        if difference <= 1e-6:

            append_validation(
                validations,
                "SASA_Atomic_Area_Sum_Check",
                difference,
                "PASS",
                "Total SASA agrees with the sum of atomic SASA values."
            )

        else:

            append_validation(
                validations,
                "SASA_Atomic_Area_Sum_Check",
                difference,
                "WARNING",
                "Total SASA differs from the sum of atomic SASA values."
            )

    return validations


# ============================================================
# 27. LIGAND-ONLY DISPERSION DESCRIPTORS
#
# Dispersion is calculated on Pd + ligand only.
# ============================================================

def calculate_ligand_dispersion(
    elements,
    coordinates,
    ligand_atoms,
    donor_atoms
):

    result = {
        "Ligand_Dispersion_Surface_Area_A2": np.nan,
        "Ligand_Dispersion_Surface_Volume_A3": np.nan,
        "Ligand_P_int": np.nan,
        "Pd_P_int": np.nan,
        "Ligand_Atomic_P_int_Sum": np.nan,
        "Ligand_Mean_Atomic_P_int": np.nan,
        "Donor1_P_int": np.nan,
        "Donor2_P_int": np.nan
    }

    if len(ligand_atoms) == 0:
        return result

    keep_atoms = [PD_ATOM] + ligand_atoms

    keep_indices = [
        atom_id - 1
        for atom_id in keep_atoms
    ]

    sub_elements = [
        elements[index]
        for index in keep_indices
    ]

    sub_coordinates = coordinates[
        keep_indices
    ]

    try:

        dispersion = Dispersion(
            sub_elements,
            sub_coordinates
        )

        result[
            "Ligand_Dispersion_Surface_Area_A2"
        ] = float(dispersion.area)

        result[
            "Ligand_Dispersion_Surface_Volume_A3"
        ] = float(dispersion.volume)

        result["Ligand_P_int"] = float(
            dispersion.p_int
        )

        result["Pd_P_int"] = float(
            dispersion.atom_p_int.get(
                1,
                np.nan
            )
        )

        ligand_p_int_values = []

        for reduced_index in range(
            2,
            len(keep_atoms) + 1
        ):

            ligand_p_int_values.append(
                dispersion.atom_p_int.get(
                    reduced_index,
                    np.nan
                )
            )

        ligand_p_int_values = [
            value
            for value in ligand_p_int_values
            if not pd.isna(value)
        ]

        result["Ligand_Atomic_P_int_Sum"] = float(
            np.sum(ligand_p_int_values)
        )

        result["Ligand_Mean_Atomic_P_int"] = safe_mean(
            ligand_p_int_values
        )

        if len(donor_atoms) >= 1:

            donor1_atom = donor_atoms[0][0]

            donor1_reduced = (
                keep_atoms.index(donor1_atom) + 1
            )

            result["Donor1_P_int"] = float(
                dispersion.atom_p_int.get(
                    donor1_reduced,
                    np.nan
                )
            )

        if len(donor_atoms) >= 2:

            donor2_atom = donor_atoms[1][0]

            donor2_reduced = (
                keep_atoms.index(donor2_atom) + 1
            )

            result["Donor2_P_int"] = float(
                dispersion.atom_p_int.get(
                    donor2_reduced,
                    np.nan
                )
            )

    except Exception as error:

        print(f"Dispersion failed: {error}")

    return result


def validate_dispersion(
    dispersion_result
):

    validations = []

    for descriptor, value, bounds in [
        (
            "Ligand_Dispersion_Surface_Area_A2",
            dispersion_result[
                "Ligand_Dispersion_Surface_Area_A2"
            ],
            DISPERSION_AREA_RANGE
        ),
        (
            "Ligand_Dispersion_Surface_Volume_A3",
            dispersion_result[
                "Ligand_Dispersion_Surface_Volume_A3"
            ],
            DISPERSION_VOLUME_RANGE
        ),
        (
            "Ligand_P_int",
            dispersion_result["Ligand_P_int"],
            PINT_RANGE
        )
    ]:

        status, message = validate_numeric_range(
            descriptor,
            value,
            bounds[0],
            bounds[1]
        )

        append_validation(
            validations,
            descriptor,
            value,
            status,
            message
        )

    return validations


# ============================================================
# 28. xTB DESCRIPTORS
#
# xTB is calculated on the FULL COMPLEX:
# Pd + substrate + ligand.
#
# Ligand-relevant values extracted:
# - Pd charge
# - donor charges
# - Pd-donor bond orders
# - donor Fukui indices
# ============================================================

def safe_xtb_call(function):

    try:
        return function()
    except Exception:
        return np.nan


def calculate_xtb_descriptors(
    elements,
    coordinates,
    donor_atoms
):

    result = {

        # xTB calculation metadata

        "XTB_Available": XTB_AVAILABLE,
        "XTB_Method": XTB_METHOD,
        "XTB_Charge": MOLECULAR_CHARGE,
        "XTB_N_Unpaired": N_UNPAIRED,
        "XTB_Solvent": REACTION_SOLVENT,
        "XTB_Run_Status": "NOT_RUN",

        # Whole-complex electronic properties

        "XTB_Energy_Eh": np.nan,
        "XTB_HOMO_Eh": np.nan,
        "XTB_LUMO_Eh": np.nan,
        "XTB_HOMO_LUMO_Gap_Eh": np.nan,

        "XTB_Dipole_X": np.nan,
        "XTB_Dipole_Y": np.nan,
        "XTB_Dipole_Z": np.nan,
        "XTB_Dipole_Magnitude": np.nan,

        "XTB_IP_Corrected_eV": np.nan,
        "XTB_IP_Uncorrected_eV": np.nan,
        "XTB_EA_Corrected_eV": np.nan,
        "XTB_EA_Uncorrected_eV": np.nan,

        "XTB_Global_Electrophilicity": np.nan,
        "XTB_Global_Nucleophilicity": np.nan,
        "XTB_Global_Electrofugality": np.nan,
        "XTB_Global_Nucleofugality": np.nan,

        # HFIP implicit-solvation terms

        "XTB_HFIP_Solvation_Energy_kcal_mol": np.nan,
        "XTB_HFIP_HBond_Correction_kcal_mol": np.nan,

        # Charges

        "XTB_Pd_Charge": np.nan,
        "Donor1_Charge": np.nan,
        "Donor2_Charge": np.nan,
        "Average_Donor_Charge": np.nan,
        "Delta_Donor_Charge": np.nan,
        "XTB_Charge_Sum": np.nan,

        # Pd-donor bond orders

        "Pd_Donor1_Bond_Order": np.nan,
        "Pd_Donor2_Bond_Order": np.nan,
        "Average_Pd_Donor_Bond_Order": np.nan,
        "Delta_Pd_Donor_Bond_Order": np.nan,

        # Donor Fukui functions

        "Donor1_Fukui_Electrophilicity": np.nan,
        "Donor2_Fukui_Electrophilicity": np.nan,
        "Average_Donor_Fukui_Electrophilicity": np.nan,

        "Donor1_Fukui_Nucleophilicity": np.nan,
        "Donor2_Fukui_Nucleophilicity": np.nan,
        "Average_Donor_Fukui_Nucleophilicity": np.nan,

        "Donor1_Fukui_Radical": np.nan,
        "Donor2_Fukui_Radical": np.nan,
        "Average_Donor_Fukui_Radical": np.nan,

        "Donor1_Fukui_Dual": np.nan,
        "Donor2_Fukui_Dual": np.nan,
        "Average_Donor_Fukui_Dual": np.nan
    }

    if not XTB_AVAILABLE:

        result["XTB_Run_Status"] = "XTB_UNAVAILABLE"

        return result

    try:

        xtb = XTB(
            elements,
            coordinates,
            charge=MOLECULAR_CHARGE,
            n_unpaired=N_UNPAIRED,
            solvent=REACTION_SOLVENT,
            method=XTB_METHOD
        )

        result["XTB_Run_Status"] = "SUCCESS"

        energy = safe_xtb_call(
            xtb.get_energy
        )

        homo = safe_xtb_call(
            xtb.get_homo
        )

        lumo = safe_xtb_call(
            xtb.get_lumo
        )

        dipole = safe_xtb_call(
            xtb.get_dipole
        )

        result["XTB_Energy_Eh"] = energy
        result["XTB_HOMO_Eh"] = homo
        result["XTB_LUMO_Eh"] = lumo

        if (
            not pd.isna(homo)
            and not pd.isna(lumo)
        ):

            result["XTB_HOMO_LUMO_Gap_Eh"] = (
                lumo - homo
            )

        if isinstance(dipole, np.ndarray) and len(dipole) == 3:

            result["XTB_Dipole_X"] = float(dipole[0])
            result["XTB_Dipole_Y"] = float(dipole[1])
            result["XTB_Dipole_Z"] = float(dipole[2])

            result["XTB_Dipole_Magnitude"] = float(
                np.linalg.norm(dipole)
            )

        result["XTB_IP_Corrected_eV"] = safe_xtb_call(
            lambda: xtb.get_ip(corrected=True)
        )

        result["XTB_IP_Uncorrected_eV"] = safe_xtb_call(
            lambda: xtb.get_ip(corrected=False)
        )

        result["XTB_EA_Corrected_eV"] = safe_xtb_call(
            lambda: xtb.get_ea(corrected=True)
        )

        result["XTB_EA_Uncorrected_eV"] = safe_xtb_call(
            lambda: xtb.get_ea(corrected=False)
        )

        global_descriptor_names = {
            "XTB_Global_Electrophilicity": "electrophilicity",
            "XTB_Global_Nucleophilicity": "nucleophilicity",
            "XTB_Global_Electrofugality": "electrofugality",
            "XTB_Global_Nucleofugality": "nucleofugality"
        }

        for output_column, descriptor_name in (
            global_descriptor_names.items()
        ):

            result[output_column] = safe_xtb_call(
                lambda name=descriptor_name:
                xtb.get_global_descriptor(name)
            )

        result[
            "XTB_HFIP_Solvation_Energy_kcal_mol"
        ] = safe_xtb_call(
            lambda: xtb.get_solvation_energy(
                unit="kcal/mol"
            )
        )

        result[
            "XTB_HFIP_HBond_Correction_kcal_mol"
        ] = safe_xtb_call(
            lambda: xtb.get_solvation_h_bond_correction(
                unit="kcal/mol"
            )
        )

        charges = safe_xtb_call(
            xtb.get_charges
        )

        if isinstance(charges, dict):

            result["XTB_Pd_Charge"] = charges.get(
                PD_ATOM,
                np.nan
            )

            result["XTB_Charge_Sum"] = float(
                np.sum(
                    list(charges.values())
                )
            )

            donor_charges = []

            if len(donor_atoms) >= 1:

                donor1_atom = donor_atoms[0][0]

                donor1_charge = charges.get(
                    donor1_atom,
                    np.nan
                )

                result["Donor1_Charge"] = donor1_charge

                donor_charges.append(
                    donor1_charge
                )

            if len(donor_atoms) >= 2:

                donor2_atom = donor_atoms[1][0]

                donor2_charge = charges.get(
                    donor2_atom,
                    np.nan
                )

                result["Donor2_Charge"] = donor2_charge

                donor_charges.append(
                    donor2_charge
                )

            result["Average_Donor_Charge"] = safe_mean(
                donor_charges
            )

            if len(donor_charges) == 2:

                result["Delta_Donor_Charge"] = abs(
                    donor_charges[0]
                    - donor_charges[1]
                )

        donor_bond_orders = []

        if len(donor_atoms) >= 1:

            donor1_atom = donor_atoms[0][0]

            donor1_bond_order = safe_xtb_call(
                lambda: xtb.get_bond_order(
                    PD_ATOM,
                    donor1_atom
                )
            )

            result["Pd_Donor1_Bond_Order"] = (
                donor1_bond_order
            )

            donor_bond_orders.append(
                donor1_bond_order
            )

        if len(donor_atoms) >= 2:

            donor2_atom = donor_atoms[1][0]

            donor2_bond_order = safe_xtb_call(
                lambda: xtb.get_bond_order(
                    PD_ATOM,
                    donor2_atom
                )
            )

            result["Pd_Donor2_Bond_Order"] = (
                donor2_bond_order
            )

            donor_bond_orders.append(
                donor2_bond_order
            )

        result["Average_Pd_Donor_Bond_Order"] = safe_mean(
            donor_bond_orders
        )

        if len(donor_bond_orders) == 2:

            result["Delta_Pd_Donor_Bond_Order"] = abs(
                donor_bond_orders[0]
                - donor_bond_orders[1]
            )

        fukui_map = {
            "Electrophilicity": "electrophilicity",
            "Nucleophilicity": "nucleophilicity",
            "Radical": "radical",
            "Dual": "dual"
        }

        donor_ids = [
            donor[0]
            for donor in donor_atoms[:2]
        ]

        for output_label, fukui_name in fukui_map.items():

            fukui_values = safe_xtb_call(
                lambda name=fukui_name:
                xtb.get_fukui(name)
            )

            if isinstance(fukui_values, dict):

                selected_values = []

                if len(donor_ids) >= 1:

                    donor1_value = fukui_values.get(
                        donor_ids[0],
                        np.nan
                    )

                    result[
                        f"Donor1_Fukui_{output_label}"
                    ] = donor1_value

                    selected_values.append(
                        donor1_value
                    )

                if len(donor_ids) >= 2:

                    donor2_value = fukui_values.get(
                        donor_ids[1],
                        np.nan
                    )

                    result[
                        f"Donor2_Fukui_{output_label}"
                    ] = donor2_value

                    selected_values.append(
                        donor2_value
                    )

                result[
                    f"Average_Donor_Fukui_{output_label}"
                ] = safe_mean(selected_values)

    except Exception as error:

        result["XTB_Run_Status"] = (
            f"FAILED: {error}"
        )

    return result


def validate_xtb(
    xtb_result
):

    validations = []

    run_status = xtb_result["XTB_Run_Status"]

    if run_status == "XTB_UNAVAILABLE":

        append_validation(
            validations,
            "XTB_Run",
            np.nan,
            "NOT_APPLICABLE",
            "xTB executable unavailable; xTB descriptors were not calculated."
        )

        return validations

    if run_status != "SUCCESS":

        append_validation(
            validations,
            "XTB_Run",
            np.nan,
            "FAIL",
            f"xTB calculation failed: {run_status}"
        )

        return validations

    append_validation(
        validations,
        "XTB_Run",
        "SUCCESS",
        "PASS",
        "xTB calculation completed successfully."
    )

    energy = xtb_result["XTB_Energy_Eh"]

    if pd.isna(energy):

        append_validation(
            validations,
            "XTB_Energy_Eh",
            np.nan,
            "FAIL",
            "xTB energy was not returned."
        )

    else:

        append_validation(
            validations,
            "XTB_Energy_Eh",
            energy,
            "PASS",
            "xTB energy was returned."
        )

    gap = xtb_result["XTB_HOMO_LUMO_Gap_Eh"]

    if pd.isna(gap):

        append_validation(
            validations,
            "XTB_HOMO_LUMO_Gap_Eh",
            np.nan,
            "WARNING",
            "HOMO-LUMO gap was not available."
        )

    elif gap > 0:

        append_validation(
            validations,
            "XTB_HOMO_LUMO_Gap_Eh",
            gap,
            "PASS",
            "HOMO-LUMO gap is positive."
        )

    else:

        append_validation(
            validations,
            "XTB_HOMO_LUMO_Gap_Eh",
            gap,
            "WARNING",
            "HOMO-LUMO gap is non-positive; verify charge/spin/state."
        )

    charge_sum = xtb_result["XTB_Charge_Sum"]

    if pd.isna(charge_sum):

        append_validation(
            validations,
            "XTB_Charge_Sum",
            np.nan,
            "WARNING",
            "Atomic charge sum was not available."
        )

    else:

        difference = abs(
            charge_sum
            - MOLECULAR_CHARGE
        )

        if difference <= XTB_CHARGE_SUM_TOLERANCE:

            append_validation(
                validations,
                "XTB_Charge_Sum",
                charge_sum,
                "PASS",
                f"Charge sum agrees with specified molecular charge "
                f"within ±{XTB_CHARGE_SUM_TOLERANCE:.2f}."
            )

        else:

            append_validation(
                validations,
                "XTB_Charge_Sum",
                charge_sum,
                "WARNING",
                "Atomic charge sum differs substantially from specified charge."
            )

    hfip_solvation = xtb_result[
        "XTB_HFIP_Solvation_Energy_kcal_mol"
    ]

    if pd.isna(hfip_solvation):

        append_validation(
            validations,
            "XTB_HFIP_Solvation",
            np.nan,
            "WARNING",
            "HFIP solvation energy unavailable; verify solvent keyword support."
        )

    else:

        append_validation(
            validations,
            "XTB_HFIP_Solvation",
            hfip_solvation,
            "PASS",
            "HFIP implicit-solvation energy was calculated."
        )

    return validations


# ============================================================
# 29. COORDINATION CLASSIFICATION
# ============================================================

def classify_coordination(donor_atoms):

    if len(donor_atoms) == 0:
        return "No_ligand_N_or_O_donor_in_cutoff"

    if len(donor_atoms) == 1:
        return f"Monodentate_{donor_atoms[0][1]}"

    if len(donor_atoms) == 2:

        donor_types = "".join(
            donor[1]
            for donor in donor_atoms
        )

        return f"Bidentate_{donor_types}"

    donor_types = "".join(
        donor[1]
        for donor in donor_atoms
    )

    return f"Ambiguous_more_than_2_donors_{donor_types}"


# ============================================================
# STRUCTURE-LEVEL VALIDATION STATUS
#
# Only these items can invalidate the whole structure.
# All other descriptor problems affect only their individual
# descriptor values and do not eliminate the molecule from the
# descriptor-vs-yield analysis.
# ============================================================

CRITICAL_VALIDATION_DESCRIPTORS = {

    # Basic XYZ integrity

    "XYZ_Atom_Count",
    "Ligand_Boundary",
    "Pd_Identity",
    "Coordinate_Finiteness",

    # Required for the descriptor-vs-yield analysis

    "Yield_Matching"
}


def determine_structure_validation(validation_records):

    critical_failures = [

        record
        for record in validation_records
        if (
            record["Descriptor"]
            in CRITICAL_VALIDATION_DESCRIPTORS
            and record["Status"] == "FAIL"
        )
    ]

    optional_failures = [

        record
        for record in validation_records
        if (
            record["Descriptor"]
            not in CRITICAL_VALIDATION_DESCRIPTORS
            and record["Status"] == "FAIL"
        )
    ]

    warnings_found = [

        record
        for record in validation_records
        if record["Status"] == "WARNING"
    ]

    # A true FAIL means that the file itself, Pd identity,
    # ligand boundary, coordinates, or yield matching is invalid.

    if len(critical_failures) > 0:
        return "FAIL"

    # Optional descriptor failures must not discard the entire
    # structure. They become a structure-level warning.

    if len(optional_failures) > 0:
        return "WARNING"

    if len(warnings_found) > 0:
        return "WARNING"

    return "PASS"

# ============================================================
# 30. PROCESS ONE XYZ FILE
# ============================================================

def process_xyz(filename):

    basename = os.path.basename(filename)

    all_validations = []

    # --------------------------------------------------------
    # Read file
    # --------------------------------------------------------

    elements, coordinates = read_xyz_file(filename)

    # --------------------------------------------------------
    # Validate structure assumptions
    # --------------------------------------------------------

    structure_validations = validate_structure(
        elements,
        coordinates
    )

    all_validations.extend(
        structure_validations
    )

    critical_validation_items = {
        "XYZ_Atom_Count",
        "Ligand_Boundary",
        "Pd_Identity",
        "Coordinate_Finiteness"
    }

    critical_failure = any(
        record["Status"] == "FAIL"
        and record["Descriptor"] in critical_validation_items
        for record in structure_validations
    )

    if critical_failure:

        raise ValueError(
            "Critical structure validation failed."
        )

    # --------------------------------------------------------
    # Match yield from filename
    # --------------------------------------------------------

    ligand_id = extract_ligand_id(filename)

    if pd.isna(ligand_id):

        reaction_yield = np.nan

        append_validation(
            all_validations,
            "Yield_Matching",
            np.nan,
            "WARNING",
            "No unique ligand ID (e.g., L7) could be extracted from filename."
        )

    elif ligand_id not in yield_dict:

        reaction_yield = np.nan

        append_validation(
            all_validations,
            "Yield_Matching",
            ligand_id,
            "WARNING",
            f"{ligand_id} is not present in yield_dict."
        )

    else:

        reaction_yield = float(
            yield_dict[ligand_id]
        )

        append_validation(
            all_validations,
            "Yield_Matching",
            reaction_yield,
            "PASS",
            f"{ligand_id} matched to yield {reaction_yield:.1f}%."
        )

    # --------------------------------------------------------
    # Ligand atom list and donor identification
    # --------------------------------------------------------

    ligand_atoms = get_ligand_atoms(
        len(elements)
    )

    donor_atoms, all_ligand_no_contacts = (
        find_coordinated_atoms(
            elements,
            coordinates
        )
    )

    all_validations.extend(
        validate_donors(
            donor_atoms,
            all_ligand_no_contacts
        )
    )

    # --------------------------------------------------------
    # Coordination distances
    # --------------------------------------------------------

    coordination = calculate_coordination_descriptors(
        donor_atoms
    )

    donor1_atom = coordination["Donor1_Atom"]
    donor2_atom = coordination["Donor2_Atom"]

    donor1_element = coordination["Donor1_Element"]
    donor2_element = coordination["Donor2_Element"]

    # --------------------------------------------------------
    # Bite angle
    # --------------------------------------------------------

    bite_result = calculate_bite_angle(
        coordinates,
        donor_atoms
    )

    all_validations.extend(
        validate_bite_angle(
            bite_result,
            donor_atoms
        )
    )

    # --------------------------------------------------------
    # Sterimol
    # --------------------------------------------------------

    B1_D1, B5_D1, L_D1 = calculate_sterimol_for_donor(
        elements,
        coordinates,
        ligand_atoms,
        donor1_atom
    )

    B1_D2, B5_D2, L_D2 = calculate_sterimol_for_donor(
        elements,
        coordinates,
        ligand_atoms,
        donor2_atom
    )

    all_validations.extend(
        validate_sterimol(
            B1_D1,
            B5_D1,
            L_D1,
            "Donor1"
        )
    )

    all_validations.extend(
        validate_sterimol(
            B1_D2,
            B5_D2,
            L_D2,
            "Donor2"
        )
    )

    # --------------------------------------------------------
    # %Vbur
    # --------------------------------------------------------

    ligand_vbur = calculate_ligand_vbur(
        elements,
        coordinates,
        ligand_atoms
    )

    all_validations.extend(
        validate_vbur(
            ligand_vbur,
            elements,
            coordinates,
            ligand_atoms
        )
    )

    # --------------------------------------------------------
    # Ligand extension
    # --------------------------------------------------------

    extension_d1 = calculate_ligand_extension(
        coordinates,
        ligand_atoms,
        donor1_atom
    )

    extension_d2 = calculate_ligand_extension(
        coordinates,
        ligand_atoms,
        donor2_atom
    )

    all_validations.extend(
        validate_ligand_extension(
            extension_d1,
            "Donor1"
        )
    )

    all_validations.extend(
        validate_ligand_extension(
            extension_d2,
            "Donor2"
        )
    )

    # --------------------------------------------------------
    # Donor pyramidalization
    # --------------------------------------------------------

    pyramidalization_d1 = donor_pyramidalization(
        coordinates,
        ligand_atoms,
        donor1_atom
    )

    pyramidalization_d2 = donor_pyramidalization(
        coordinates,
        ligand_atoms,
        donor2_atom
    )

    all_validations.extend(
        validate_pyramidalization(
            pyramidalization_d1,
            "Donor1",
            donor1_element
        )
    )

    all_validations.extend(
        validate_pyramidalization(
            pyramidalization_d2,
            "Donor2",
            donor2_element
        )
    )

    if (
        not pd.isna(pyramidalization_d1)
        and not pd.isna(pyramidalization_d2)
    ):

        delta_pyramidalization = abs(
            pyramidalization_d1
            - pyramidalization_d2
        )

    else:

        delta_pyramidalization = np.nan

    # --------------------------------------------------------
    # Ligand-only SASA
    # --------------------------------------------------------

    sasa_result = calculate_ligand_sasa(
        elements,
        coordinates,
        ligand_atoms,
        donor_atoms
    )

    all_validations.extend(
        validate_sasa(
            sasa_result,
            elements,
            coordinates,
            ligand_atoms,
            donor_atoms
        )
    )

    # --------------------------------------------------------
    # Ligand-only dispersion
    # --------------------------------------------------------

    dispersion_result = calculate_ligand_dispersion(
        elements,
        coordinates,
        ligand_atoms,
        donor_atoms
    )

    all_validations.extend(
        validate_dispersion(
            dispersion_result
        )
    )

    # --------------------------------------------------------
    # xTB descriptors: full Pd complex in HFIP
    # --------------------------------------------------------

    xtb_result = calculate_xtb_descriptors(
        elements,
        coordinates,
        donor_atoms
    )

    all_validations.extend(
        validate_xtb(
            xtb_result
        )
    )

    # --------------------------------------------------------
    # Validation summary
    # --------------------------------------------------------

    statuses = [
        record["Status"]
        for record in all_validations
    ]

    overall_validation = determine_structure_validation(all_validations)

    warning_fail_messages = [
        record["Message"]
        for record in all_validations
        if record["Status"] in {"WARNING", "FAIL"}
    ]

    validation_message_text = (
        "No warnings or failures recorded."
        if len(warning_fail_messages) == 0
        else " | ".join(warning_fail_messages)
    )

    # --------------------------------------------------------
    # Build descriptor result row
    # --------------------------------------------------------

    result = {

        # File / yield identity

        "Filename": basename,
        "Ligand_ID": ligand_id,
        YIELD_COLUMN: reaction_yield,

        # Atom partition

        "N_Atoms_Total": len(elements),
        "Pd_Atom": PD_ATOM,
        "Substrate_Atom_Range": (
            f"{SUBSTRATE_START}-{SUBSTRATE_END}"
        ),
        "Ligand_Atom_Range": (
            f"{LIGAND_START}-{len(elements)}"
        ),
        "N_Ligand_Atoms": len(ligand_atoms),

        # Donor diagnostics

        "N_Ligand_N_O_Atoms": len(
            all_ligand_no_contacts
        ),
        "N_Donor_Atoms_Found": len(
            donor_atoms
        ),
        "Coordination_Classification": (
            classify_coordination(donor_atoms)
        ),
        "All_Ligand_N_O_Contacts_Pd_A": (
            format_contacts(all_ligand_no_contacts)
        ),
        "Accepted_Donor_Contacts_Pd_A": (
            format_contacts(donor_atoms)
        ),

        # Coordination distances

        **coordination,

        # Bite angle

        **bite_result,

        # Sterimol

        "Sterimol_B1_Donor1": B1_D1,
        "Sterimol_B5_Donor1": B5_D1,
        "Sterimol_L_Donor1": L_D1,

        "Sterimol_B1_Donor2": B1_D2,
        "Sterimol_B5_Donor2": B5_D2,
        "Sterimol_L_Donor2": L_D2,

        # %Vbur

        "Ligand_Vbur_3.5A_percent": ligand_vbur,

        # Ligand extension

        "Ligand_Max_Extension_Donor1": extension_d1,
        "Ligand_Max_Extension_Donor2": extension_d2,

        # Donor pyramidalization

        "Donor1_Pyramidalization": pyramidalization_d1,
        "Donor2_Pyramidalization": pyramidalization_d2,
        "Delta_Donor_Pyramidalization": delta_pyramidalization,

        # SASA

        **sasa_result,

        # Dispersion

        **dispersion_result,

        # xTB

        **xtb_result,

        # Validation summary

        "Validation_Overall": overall_validation,

        "Validation_PASS_Count": sum(
            status == "PASS"
            for status in statuses
        ),

        "Validation_WARNING_Count": sum(
            status == "WARNING"
            for status in statuses
        ),

        "Validation_FAIL_Count": sum(
            status == "FAIL"
            for status in statuses
        ),

        "Validation_Warnings_or_Failures": (
            validation_message_text
        )
    }

    file_validation_df = pd.DataFrame(
        all_validations
    )

    file_validation_df.insert(
        0,
        "Filename",
        basename
    )

    return result, file_validation_df


# ============================================================
# 31. FIND XYZ FILES
# ============================================================

xyz_files = sorted(
    glob.glob(
        os.path.join(
            XYZ_FOLDER,
            "*.xyz"
        )
    )
)

print("\n" + "=" * 100)
print(f"Found {len(xyz_files)} XYZ file(s).")
print("=" * 100)

if len(xyz_files) == 0:

    raise FileNotFoundError(
        f"No XYZ files found in:\n{XYZ_FOLDER}"
    )


# ============================================================
# 32. PRE-FLIGHT CHECK
# ============================================================

preflight_rows = []

print("\n" + "=" * 100)
print("PRE-FLIGHT CHECK")
print("=" * 100)

for xyz_file in xyz_files:

    basename = os.path.basename(xyz_file)

    try:

        elements, coordinates = read_xyz_file(
            xyz_file
        )

        n_atoms = len(elements)

        ligand_id = extract_ligand_id(
            xyz_file
        )

        yield_value = yield_dict.get(
            ligand_id,
            np.nan
        )

        ligand_start_element = (
            elements[LIGAND_START - 1]
            if n_atoms >= LIGAND_START
            else "N/A"
        )

        ligand_no_atoms = [
            f"{elements[atom_id - 1]}{atom_id}"
            for atom_id in get_ligand_atoms(n_atoms)
            if elements[atom_id - 1] in DONOR_ELEMENTS
        ]

        status = "PASS"

        if elements[PD_ATOM - 1] != "Pd":
            status = "FAIL"

        if n_atoms < LIGAND_START:
            status = "FAIL"

        if pd.isna(ligand_id):
            status = "WARNING"

        elif ligand_id not in yield_dict:
            status = "WARNING"

        preflight_rows.append({

            "Filename": basename,
            "Ligand_ID": ligand_id,
            YIELD_COLUMN: yield_value,

            "N_Atoms_Total": n_atoms,
            "Atom_1_Element": elements[PD_ATOM - 1],

            f"Atom_{LIGAND_START}_Element": (
                ligand_start_element
            ),

            "Ligand_N_O_Atoms": (
                ", ".join(ligand_no_atoms)
                if ligand_no_atoms
                else "None"
            ),

            "Preflight_Status": status
        })

    except Exception as error:

        preflight_rows.append({

            "Filename": basename,
            "Ligand_ID": np.nan,
            YIELD_COLUMN: np.nan,

            "N_Atoms_Total": np.nan,
            "Atom_1_Element": np.nan,

            f"Atom_{LIGAND_START}_Element": np.nan,

            "Ligand_N_O_Atoms": np.nan,

            "Preflight_Status": "FAIL",

            "Error": str(error)
        })

preflight_df = pd.DataFrame(
    preflight_rows
)

display(preflight_df)


# ============================================================
# 33. PROCESS ALL XYZ FILES
# ============================================================

results = []
validation_tables = []

for i, xyz_file in enumerate(
    xyz_files,
    start=1
):

    basename = os.path.basename(xyz_file)

    print("\n" + "-" * 100)
    print(f"[{i}/{len(xyz_files)}] Processing: {basename}")
    print("-" * 100)

    try:

        result, file_validation_df = process_xyz(
            xyz_file
        )

        results.append(result)
        validation_tables.append(file_validation_df)

        print(
            f"Ligand ID: {result['Ligand_ID']}"
        )

        print(
            f"Yield: {result[YIELD_COLUMN]}"
        )

        print(
            "Accepted donor(s): "
            f"{result['Accepted_Donor_Contacts_Pd_A']}"
        )

        print(
            "Bite angle: "
            f"{result['Bite_Angle_Morfeus']:.3f}°"
            if not pd.isna(
                result["Bite_Angle_Morfeus"]
            )
            else "Bite angle: NaN"
        )

        print(
            "%Vbur: "
            f"{result['Ligand_Vbur_3.5A_percent']:.3f}%"
            if not pd.isna(
                result["Ligand_Vbur_3.5A_percent"]
            )
            else "%Vbur: NaN"
        )

        print(
            f"Validation: {result['Validation_Overall']}"
        )

    except Exception as error:

        print(f"ERROR: {error}")

        results.append({

            "Filename": basename,
            "Ligand_ID": extract_ligand_id(xyz_file),
            YIELD_COLUMN: np.nan,

            "Validation_Overall": "FAIL",
            "Validation_PASS_Count": 0,
            "Validation_WARNING_Count": 0,
            "Validation_FAIL_Count": 1,
            "Validation_Warnings_or_Failures": str(error)
        })

        validation_tables.append(
            pd.DataFrame([{

                "Filename": basename,
                "Descriptor": "Processing",
                "Value": np.nan,
                "Status": "FAIL",
                "Message": str(error)
            }])
        )


descriptor_df = pd.DataFrame(results)

validation_log_df = pd.concat(
    validation_tables,
    ignore_index=True
)


# ============================================================
# 34. DISPLAY DESCRIPTOR RESULTS
# ============================================================

pd.set_option("display.max_columns", None)
pd.set_option("display.width", 300)

print("\nALL DESCRIPTORS")
display(descriptor_df)

print("\nVALIDATION LOG")
display(validation_log_df)


# ============================================================
# 35. BUILD DESCRIPTOR-VS-YIELD DATASET
#
# This is NOT a descriptor correlation matrix.
#
# It contains only:
# - structures with a matched yield
# - structures not marked as validation FAIL
# ============================================================

# ============================================================
# RETAIN ALL STRUCTURES WITH MATCHED YIELD
#
# Optional descriptor failures must NOT remove the complete
# ligand from descriptor-vs-yield correlations.
#
# Each descriptor is independently filtered with dropna()
# during its individual correlation calculation.
# ============================================================

yield_analysis_df = descriptor_df.copy()

yield_analysis_df[YIELD_COLUMN] = pd.to_numeric(
    yield_analysis_df[YIELD_COLUMN],
    errors="coerce"
)

yield_analysis_df = yield_analysis_df.dropna(
    subset=[YIELD_COLUMN]
).copy()

yield_analysis_df = yield_analysis_df.loc[
    (
        yield_analysis_df[YIELD_COLUMN] >= 0
    )
    &
    (
        yield_analysis_df[YIELD_COLUMN] <= 100
    )
].copy()

print(
    "Structures retained for descriptor-vs-yield analysis:",
    len(yield_analysis_df)
)

display(
    yield_analysis_df[
        [
            "Filename",
            "Ligand_ID",
            YIELD_COLUMN,
            "Validation_Overall",
            "Validation_Warnings_or_Failures"
        ]
    ]
)


yield_analysis_df[YIELD_COLUMN] = pd.to_numeric(
    yield_analysis_df[YIELD_COLUMN],
    errors="coerce"
)

yield_analysis_df = yield_analysis_df.dropna(
    subset=[YIELD_COLUMN]
).copy()

yield_analysis_df = yield_analysis_df.loc[
    (
        yield_analysis_df[YIELD_COLUMN] >= 0
    )
    &
    (
        yield_analysis_df[YIELD_COLUMN] <= 100
    )
].copy()

print("\n" + "=" * 100)
print("FINAL DATASET USED FOR DESCRIPTOR-VS-YIELD CORRELATION")
print("=" * 100)

display(
    yield_analysis_df[
        [
            "Filename",
            "Ligand_ID",
            YIELD_COLUMN,
            "Validation_Overall"
        ]
    ]
)

print(
    f"Number of structures with matched yield: "
    f"{len(yield_analysis_df)}"
)

print(
    f"Number of 0% yield structures: "
    f"{int((yield_analysis_df[YIELD_COLUMN] == 0).sum())}"
)

print(
    f"Number of non-zero yield structures: "
    f"{int((yield_analysis_df[YIELD_COLUMN] > 0).sum())}"
)

print("=" * 100)


# ============================================================
# 36. SELECT NUMERIC DESCRIPTORS
#
# Exclude:
# - yield itself
# - atom IDs
# - atom counts
# - validation counts
# - xTB input settings
# - Boolean flags
# ============================================================

EXCLUDED_NUMERIC_COLUMNS = [

    # Response variable

    YIELD_COLUMN,

    # Atom numbering / counts

    "N_Atoms_Total",
    "Pd_Atom",
    "N_Ligand_Atoms",
    "N_Ligand_N_O_Atoms",
    "N_Donor_Atoms_Found",

    "Donor1_Atom",
    "Donor2_Atom",

    # Validation counters

    "Validation_PASS_Count",
    "Validation_WARNING_Count",
    "Validation_FAIL_Count",

    # xTB calculation settings

    "XTB_Method",
    "XTB_Charge",
    "XTB_N_Unpaired",

    # Boolean / diagnostic values, not descriptors

    "XTB_Available",
    "Bite_Angle_Inverted"
]

all_numeric_columns = yield_analysis_df.select_dtypes(
    include=np.number
).columns.tolist()

descriptor_columns = [
    column
    for column in all_numeric_columns
    if column not in EXCLUDED_NUMERIC_COLUMNS
]

descriptor_columns = [
    column
    for column in descriptor_columns
    if yield_analysis_df[column].notna().sum() > 0
    and yield_analysis_df[column].nunique(dropna=True) > 1
]

print("\nNUMERICAL DESCRIPTORS TO CORRELATE WITH YIELD")

for descriptor in descriptor_columns:
    print(f" - {descriptor}")


# ============================================================
# 37. CORRELATION-STRENGTH LABEL
# ============================================================

def correlation_strength_label(value):

    if pd.isna(value):
        return "Not_calculated"

    absolute_value = abs(value)

    if absolute_value >= 0.80:
        return "Very_strong"

    if absolute_value >= 0.60:
        return "Strong"

    if absolute_value >= 0.40:
        return "Moderate"

    if absolute_value >= 0.20:
        return "Weak"

    return "Very_weak"


# ============================================================
# 38. CORRELATE EVERY DESCRIPTOR AGAINST YIELD
#
# Pearson r:
# linear descriptor-yield relationship
#
# Spearman rho:
# rank/monotonic descriptor-yield relationship
#
# Point-biserial r:
# descriptor association with inactive (0%) versus active (>0%)
# ============================================================

yield_correlation_rows = []

for descriptor in descriptor_columns:

    paired_df = yield_analysis_df[
        [descriptor, YIELD_COLUMN]
    ].dropna().copy()

    n_observations = len(paired_df)

    n_unique_descriptor = paired_df[
        descriptor
    ].nunique()

    n_zero_yield = int(
        (paired_df[YIELD_COLUMN] == 0).sum()
    )

    n_nonzero_yield = int(
        (paired_df[YIELD_COLUMN] > 0).sum()
    )

    row = {

        "Descriptor": descriptor,

        "N_Observations": n_observations,

        "N_Unique_Descriptor_Values": (
            n_unique_descriptor
        ),

        "N_Zero_Yield": n_zero_yield,

        "N_Nonzero_Yield": n_nonzero_yield,

        "Pearson_r": np.nan,
        "Pearson_p_value": np.nan,

        "Spearman_rho": np.nan,
        "Spearman_p_value": np.nan,

        "Point_Biserial_r_Active_vs_Zero": np.nan,
        "Point_Biserial_p_value": np.nan,

        "Pearson_Strength": "Not_calculated",
        "Spearman_Strength": "Not_calculated",

        "Correlation_Status": "NOT_CALCULATED",
        "Correlation_Message": ""
    }

    # --------------------------------------------------------
    # Basic data checks
    # --------------------------------------------------------

    if n_observations < MIN_CORRELATION_OBSERVATIONS:

        row["Correlation_Status"] = "WARNING"

        row["Correlation_Message"] = (
            f"Only {n_observations} paired observations; "
            f"minimum requested is {MIN_CORRELATION_OBSERVATIONS}."
        )

        yield_correlation_rows.append(row)

        continue

    if n_unique_descriptor < 2:

        row["Correlation_Status"] = "WARNING"

        row["Correlation_Message"] = (
            "Descriptor is constant in the available data."
        )

        yield_correlation_rows.append(row)

        continue

    if paired_df[YIELD_COLUMN].nunique() < 2:

        row["Correlation_Status"] = "WARNING"

        row["Correlation_Message"] = (
            "Yield is constant in the available data."
        )

        yield_correlation_rows.append(row)

        continue

    # --------------------------------------------------------
    # Pearson correlation: descriptor vs yield
    # --------------------------------------------------------

    try:

        pearson_r, pearson_p = pearsonr(
            paired_df[descriptor],
            paired_df[YIELD_COLUMN]
        )

        row["Pearson_r"] = float(pearson_r)
        row["Pearson_p_value"] = float(pearson_p)

        row["Pearson_Strength"] = (
            correlation_strength_label(pearson_r)
        )

    except Exception as error:

        row["Correlation_Message"] += (
            f"Pearson error: {error}. "
        )

    # --------------------------------------------------------
    # Spearman correlation: descriptor vs yield
    # --------------------------------------------------------

    try:

        spearman_rho, spearman_p = spearmanr(
            paired_df[descriptor],
            paired_df[YIELD_COLUMN]
        )

        row["Spearman_rho"] = float(spearman_rho)
        row["Spearman_p_value"] = float(spearman_p)

        row["Spearman_Strength"] = (
            correlation_strength_label(spearman_rho)
        )

    except Exception as error:

        row["Correlation_Message"] += (
            f"Spearman error: {error}. "
        )

    # --------------------------------------------------------
    # Point-biserial: descriptor vs inactive/active class
    #
    # 0 = zero yield
    # 1 = non-zero yield
    # --------------------------------------------------------

    active_vs_zero = (
        paired_df[YIELD_COLUMN] > 0
    ).astype(int)

    if active_vs_zero.nunique() == 2:

        try:

            point_r, point_p = pointbiserialr(
                active_vs_zero,
                paired_df[descriptor]
            )

            row[
                "Point_Biserial_r_Active_vs_Zero"
            ] = float(point_r)

            row[
                "Point_Biserial_p_value"
            ] = float(point_p)

        except Exception as error:

            row["Correlation_Message"] += (
                f"Point-biserial error: {error}. "
            )

    else:

        row["Correlation_Message"] += (
            "Point-biserial unavailable because only one "
            "yield class is represented. "
        )

    # --------------------------------------------------------
    # Statistical warning / result status
    # --------------------------------------------------------

    if (
        pd.isna(row["Pearson_r"])
        and pd.isna(row["Spearman_rho"])
    ):

        row["Correlation_Status"] = "FAIL"

        row["Correlation_Message"] += (
            "Neither Pearson nor Spearman correlation could be calculated."
        )

    elif n_observations < 8:

        row["Correlation_Status"] = "WARNING"

        row["Correlation_Message"] += (
            "Correlation calculated, but fewer than 8 observations; "
            "interpret as exploratory."
        )

    elif n_zero_yield / n_observations >= 0.50:

        row["Correlation_Status"] = "WARNING"

        row["Correlation_Message"] += (
            "At least half the yields are zero; use scatter plots and "
            "point-biserial results alongside Pearson/Spearman."
        )

    else:

        row["Correlation_Status"] = "PASS"

        row["Correlation_Message"] += (
            "Correlation calculated successfully."
        )

    yield_correlation_rows.append(row)


yield_correlation_df = pd.DataFrame(
    yield_correlation_rows
)

yield_correlation_df["Abs_Pearson_r"] = abs(
    yield_correlation_df["Pearson_r"]
)

yield_correlation_df["Abs_Spearman_rho"] = abs(
    yield_correlation_df["Spearman_rho"]
)

yield_correlation_df = (
    yield_correlation_df
    .sort_values(
        by=[
            "Abs_Spearman_rho",
            "Abs_Pearson_r"
        ],
        ascending=False,
        na_position="last"
    )
    .reset_index(drop=True)
)

print("\n" + "=" * 100)
print("ALL DESCRIPTORS CORRELATED AGAINST γ-ARYLATION YIELD")
print("=" * 100)

display(yield_correlation_df)


# ============================================================
# 39. CORRELATION VALIDATION TABLE
# ============================================================

yield_correlation_check_df = (
    yield_correlation_df[
        [
            "Descriptor",
            "N_Observations",
            "N_Unique_Descriptor_Values",
            "N_Zero_Yield",
            "N_Nonzero_Yield",
            "Correlation_Status",
            "Correlation_Message"
        ]
    ]
    .copy()
)

print("\nYIELD-CORRELATION QUALITY CHECK")
display(yield_correlation_check_df)


# ============================================================
# 40. PLOT SPEARMAN DESCRIPTOR-VS-YIELD CORRELATIONS
# ============================================================

def plot_spearman_correlations(
    correlation_table,
    output_file
):

    plot_df = correlation_table.dropna(
        subset=["Spearman_rho"]
    ).copy()

    if plot_df.empty:

        print(
            "No Spearman correlations available for plotting."
        )

        return

    plot_df = plot_df.sort_values(
        "Spearman_rho",
        ascending=True
    )

    height = max(
        8,
        min(35, 0.38 * len(plot_df))
    )

    plt.figure(figsize=(13, height))

    colors = [
        "#B2182B"
        if value < 0
        else "#2166AC"
        for value in plot_df["Spearman_rho"]
    ]

    sns.barplot(
        data=plot_df,
        x="Spearman_rho",
        y="Descriptor",
        palette=colors
    )

    plt.axvline(
        0,
        color="black",
        linewidth=1
    )

    plt.xlim(-1, 1)

    plt.xlabel(
        "Spearman correlation coefficient, ρ"
    )

    plt.ylabel(
        "Ligand descriptor"
    )

    plt.title(
        "Ligand Descriptors vs γ-Arylation Yield: Spearman Correlation"
    )

    plt.tight_layout()

    plt.savefig(
        output_file,
        dpi=300,
        bbox_inches="tight"
    )

    plt.show()

    plt.close()


plot_spearman_correlations(
    yield_correlation_df,
    SPEARMAN_FIG
)


# ============================================================
# 41. PLOT PEARSON DESCRIPTOR-VS-YIELD CORRELATIONS
# ============================================================

def plot_pearson_correlations(
    correlation_table,
    output_file
):

    plot_df = correlation_table.dropna(
        subset=["Pearson_r"]
    ).copy()

    if plot_df.empty:

        print(
            "No Pearson correlations available for plotting."
        )

        return

    plot_df = plot_df.sort_values(
        "Pearson_r",
        ascending=True
    )

    height = max(
        8,
        min(35, 0.38 * len(plot_df))
    )

    plt.figure(figsize=(13, height))

    colors = [
        "#B2182B"
        if value < 0
        else "#2166AC"
        for value in plot_df["Pearson_r"]
    ]

    sns.barplot(
        data=plot_df,
        x="Pearson_r",
        y="Descriptor",
        palette=colors
    )

    plt.axvline(
        0,
        color="black",
        linewidth=1
    )

    plt.xlim(-1, 1)

    plt.xlabel(
        "Pearson correlation coefficient, r"
    )

    plt.ylabel(
        "Ligand descriptor"
    )

    plt.title(
        "Ligand Descriptors vs γ-Arylation Yield: Pearson Correlation"
    )

    plt.tight_layout()

    plt.savefig(
        output_file,
        dpi=300,
        bbox_inches="tight"
    )

    plt.show()

    plt.close()


plot_pearson_correlations(
    yield_correlation_df,
    PEARSON_FIG
)


# ============================================================
# 42. CREATE SCATTER PLOTS FOR TOP DESCRIPTORS
#
# Top descriptors are selected by absolute Spearman correlation.
# Each point is labeled with its ligand ID.
# ============================================================

def plot_descriptor_vs_yield(
    dataframe,
    descriptor,
    output_folder
):

    plot_df = dataframe[
        [
            "Ligand_ID",
            descriptor,
            YIELD_COLUMN
        ]
    ].dropna().copy()

    if len(plot_df) < MIN_CORRELATION_OBSERVATIONS:
        return

    plt.figure(figsize=(7.5, 5.5))

    sns.regplot(
        data=plot_df,
        x=descriptor,
        y=YIELD_COLUMN,
        scatter_kws={
            "s": 70,
            "alpha": 0.85
        },
        line_kws={
            "color": "black",
            "linewidth": 1.5
        }
    )

    for _, row in plot_df.iterrows():

        plt.annotate(
            str(row["Ligand_ID"]),
            (
                row[descriptor],
                row[YIELD_COLUMN]
            ),
            xytext=(4, 4),
            textcoords="offset points",
            fontsize=8
        )

    plt.title(
        f"{descriptor} vs γ-Arylation Yield"
    )

    plt.xlabel(descriptor)

    plt.ylabel(
        "γ-Arylation yield (%)"
    )

    plt.ylim(-3, 103)

    plt.tight_layout()

    plot_file = os.path.join(
        output_folder,
        f"{safe_filename(descriptor)}_vs_Yield.png"
    )

    plt.savefig(
        plot_file,
        dpi=300,
        bbox_inches="tight"
    )

    plt.show()

    plt.close()


top_descriptors = (
    yield_correlation_df
    .dropna(subset=["Spearman_rho"])
    .head(TOP_N_SCATTER_PLOTS)["Descriptor"]
    .tolist()
)

print("\nTOP DESCRIPTORS SELECTED FOR SCATTER PLOTS:")

for descriptor in top_descriptors:

    print(f" - {descriptor}")

    plot_descriptor_vs_yield(
        yield_analysis_df,
        descriptor,
        SCATTER_FOLDER
    )


# ============================================================
# 43. SAVE EXCEL WORKBOOK
# ============================================================

with pd.ExcelWriter(
    OUTPUT_EXCEL,
    engine="openpyxl"
) as writer:

    preflight_df.to_excel(
        writer,
        sheet_name="Preflight_Check",
        index=False
    )

    descriptor_df.to_excel(
        writer,
        sheet_name="All_Descriptors",
        index=False
    )

    validation_log_df.to_excel(
        writer,
        sheet_name="Validation_Log",
        index=False
    )

    yield_analysis_df.to_excel(
        writer,
        sheet_name="Yield_Analysis_Data",
        index=False
    )

    yield_correlation_df.to_excel(
        writer,
        sheet_name="Descriptor_vs_Yield",
        index=False
    )

    yield_correlation_check_df.to_excel(
        writer,
        sheet_name="Correlation_QC",
        index=False
    )


# ============================================================
# 44. FINAL REPORT
# ============================================================

print("\n" + "=" * 100)
print("COMPLETE: ALL Pd LIGAND DESCRIPTORS CORRELATED WITH YIELD")
print("=" * 100)

print(
    f"Total XYZ structures processed: "
    f"{len(descriptor_df)}"
)

print(
    f"Structures included in descriptor-vs-yield analysis: "
    f"{len(yield_analysis_df)}"
)

print(
    f"Zero-yield structures included: "
    f"{int((yield_analysis_df[YIELD_COLUMN] == 0).sum())}"
)

print(
    f"Non-zero-yield structures included: "
    f"{int((yield_analysis_df[YIELD_COLUMN] > 0).sum())}"
)

print("\nTop 15 descriptors ranked by absolute Spearman rho:")

display(
    yield_correlation_df[
        [
            "Descriptor",
            "N_Observations",
            "Pearson_r",
            "Pearson_p_value",
            "Spearman_rho",
            "Spearman_p_value",
            "Point_Biserial_r_Active_vs_Zero",
            "Point_Biserial_p_value",
            "Correlation_Status"
        ]
    ].head(15)
)

print("\nExcel workbook:")
print(OUTPUT_EXCEL)

print("\nPearson descriptor-vs-yield plot:")
print(PEARSON_FIG)

print("\nSpearman descriptor-vs-yield plot:")
print(SPEARMAN_FIG)

print("\nIndividual descriptor-vs-yield scatter plots:")
print(SCATTER_FOLDER)

print("=" * 100)
```

```python
# ============================================================
# HEATMAPS: YIELD VS ALL DESCRIPTORS
#
# Each heatmap is rectangular:
# Rows    = molecular descriptors
# Column  = gamma-arylation yield
#
# It is NOT a descriptor-vs-descriptor matrix.
# ============================================================


def make_yield_correlation_heatmap(
    correlation_results,
    coefficient_column,
    pvalue_column,
    title,
    output_file
):
    """
    Creates one-column rectangular heatmap.

    Rows:
        molecular descriptors

    Column:
        correlation with gamma-arylation yield

    Cell annotation:
        correlation coefficient plus p-value and n
    """

    plot_df = correlation_results.copy()

    # Keep only descriptors for which the requested
    # correlation coefficient exists.

    plot_df = plot_df.dropna(
        subset=[coefficient_column]
    ).copy()

    if plot_df.empty:

        print(
            f"Skipping heatmap '{title}': "
            "no usable correlation coefficients."
        )

        return

    # Sort from strongest positive to strongest negative.
    # Alternatively, use absolute value if preferred.

    plot_df = plot_df.sort_values(
        coefficient_column,
        ascending=False
    ).reset_index(drop=True)

    # Build a one-column dataframe for the heatmap values.

    heatmap_data = pd.DataFrame({
        "γ-Arylation Yield": plot_df[
            coefficient_column
        ].values
    })

    # Build annotations: coefficient, p-value, number of points.
    # Example:
    # +0.62
    # p=0.004
    # n=19

    annotation_text = []

    for _, row in plot_df.iterrows():

        coefficient = row[coefficient_column]

        p_value = row[pvalue_column]

        n_points = row["N_Observations"]

        if pd.isna(p_value):

            annotation = (
                f"{coefficient:+.2f}\n"
                f"p=NA\n"
                f"n={int(n_points)}"
            )

        else:

            annotation = (
                f"{coefficient:+.2f}\n"
                f"p={p_value:.3f}\n"
                f"n={int(n_points)}"
            )

        annotation_text.append([annotation])

    annotation_array = np.array(
        annotation_text,
        dtype=object
    )

    # Dynamic figure height prevents descriptor labels
    # from overlapping.

    figure_height = max(
        10,
        min(45, len(plot_df) * 0.42)
    )

    plt.figure(
        figsize=(8, figure_height)
    )

    sns.heatmap(
        heatmap_data,
        annot=annotation_array,
        fmt="",
        cmap="coolwarm",
        center=0,
        vmin=-1,
        vmax=1,
        linewidths=0.5,
        linecolor="white",
        cbar_kws={
            "label": "Correlation coefficient",
            "shrink": 0.70
        },
        annot_kws={
            "size": 8
        }
    )

    plt.yticks(
        ticks=np.arange(len(plot_df)) + 0.5,
        labels=plot_df["Descriptor"],
        rotation=0,
        fontsize=8
    )

    plt.xticks(
        rotation=0,
        fontsize=11
    )

    plt.title(
        title,
        fontsize=14,
        pad=16
    )

    plt.tight_layout()

    plt.savefig(
        output_file,
        dpi=300,
        bbox_inches="tight"
    )

    plt.show()

    plt.close()


# ============================================================
# PEARSON HEATMAP:
# EACH DESCRIPTOR VS ACTUAL YIELD PERCENTAGE
# ============================================================

make_yield_correlation_heatmap(
    correlation_results=yield_correlation_df,
    coefficient_column="Pearson_r",
    pvalue_column="Pearson_p_value",
    title=(
        "Pearson Correlation: "
        "γ-Arylation Yield vs All Ligand Descriptors"
    ),
    output_file=YIELD_PEARSON_HEATMAP
)


# ============================================================
# SPEARMAN HEATMAP:
# EACH DESCRIPTOR VS ACTUAL YIELD PERCENTAGE
# ============================================================

make_yield_correlation_heatmap(
    correlation_results=yield_correlation_df,
    coefficient_column="Spearman_rho",
    pvalue_column="Spearman_p_value",
    title=(
        "Spearman Correlation: "
        "γ-Arylation Yield vs All Ligand Descriptors"
    ),
    output_file=YIELD_SPEARMAN_HEATMAP
)
```

```python
# ============================================================
# SPEARMAN P-VALUE HEATMAP
#
# Lower p-values appear darker.
# This is exploratory because the dataset is small and
# yield contains many tied 0% observations.
# ============================================================

def make_yield_pvalue_heatmap(
    correlation_results,
    pvalue_column,
    title,
    output_file
):

    plot_df = correlation_results.copy()

    plot_df = plot_df.dropna(
        subset=[pvalue_column]
    ).copy()

    if plot_df.empty:

        print(
            f"Skipping p-value heatmap '{title}': "
            "no p-values available."
        )

        return

    plot_df = plot_df.sort_values(
        pvalue_column,
        ascending=True
    ).reset_index(drop=True)

    heatmap_data = pd.DataFrame({
        "Spearman p-value": plot_df[
            pvalue_column
        ].values
    })

    annotation_text = []

    for _, row in plot_df.iterrows():

        p_value = row[pvalue_column]

        rho = row["Spearman_rho"]

        n_points = row["N_Observations"]

        annotation = (
            f"p={p_value:.3f}\n"
            f"ρ={rho:+.2f}\n"
            f"n={int(n_points)}"
        )

        annotation_text.append([annotation])

    annotation_array = np.array(
        annotation_text,
        dtype=object
    )

    figure_height = max(
        10,
        min(45, len(plot_df) * 0.42)
    )

    plt.figure(
        figsize=(8, figure_height)
    )

    sns.heatmap(
        heatmap_data,
        annot=annotation_array,
        fmt="",
        cmap="YlOrRd_r",
        vmin=0,
        vmax=1,
        linewidths=0.5,
        linecolor="white",
        cbar_kws={
            "label": "Spearman p-value",
            "shrink": 0.70
        },
        annot_kws={
            "size": 8
        }
    )

    plt.yticks(
        ticks=np.arange(len(plot_df)) + 0.5,
        labels=plot_df["Descriptor"],
        rotation=0,
        fontsize=8
    )

    plt.xticks(
        rotation=0,
        fontsize=11
    )

    plt.title(
        title,
        fontsize=14,
        pad=16
    )

    plt.tight_layout()

    plt.savefig(
        output_file,
        dpi=300,
        bbox_inches="tight"
    )

    plt.show()

    plt.close()


make_yield_pvalue_heatmap(
    correlation_results=yield_correlation_df,
    pvalue_column="Spearman_p_value",
    title=(
        "Spearman p-values: "
        "γ-Arylation Yield vs All Ligand Descriptors"
    ),
    output_file=YIELD_PVALUE_HEATMAP
)
```
