```python
# ============================================================
# Pd γ-ARYLATION
# COMPLETE LIGAND DESCRIPTOR + YIELD CORRELATION ANALYSIS
#
# LIGAND ATOMS : 17 -> LAST
# Pd ATOM       : 1
# Pd-X CUTOFF   : 2.00 - 2.40 Å
#
# PROJECT: Pd γ-arylation ONLY
# ============================================================


# ============================================================
# 1. INSTALL PYTHON PACKAGES
# ============================================================

!pip -q install morfeus-ml pandas numpy scipy openpyxl matplotlib seaborn statsmodels


# ============================================================
# 2. IMPORTS
#
# IMPORTANT: morfeus is deliberately NOT imported here.
# morfeus.XTB wraps its methods with a decorator that checks
# shutil.which("xtb") exactly once, at the moment morfeus.xtb
# is first imported -- not on every call. If xtb isn't already
# on PATH at that instant, every XTB method gets permanently
# stubbed to raise OSError for the rest of this kernel session,
# even after xtb is installed and PATH is fixed afterward. So
# morfeus is imported further below, only after the xTB
# install/PATH section has confirmed the executable is
# reachable via shutil.which.
# ============================================================

import os
import re
import glob
import shutil
import subprocess
import warnings
import itertools

import numpy as np
import pandas as pd

import matplotlib.pyplot as plt
import seaborn as sns

from scipy.stats import pearsonr, spearmanr

warnings.filterwarnings("ignore")


# ============================================================
# 3. GOOGLE DRIVE
# ============================================================

from google.colab import drive

drive.mount("/content/drive")


# ============================================================
# 4. USER SETTINGS
# ============================================================

XYZ_FOLDER = "/content/drive/MyDrive/Suman_files"

OUTPUT_FOLDER = (
    "/content/drive/MyDrive/"
    "Pd_gamma_arylation_RESULTS"
)

os.makedirs(
    OUTPUT_FOLDER,
    exist_ok=True
)


# ------------------------------------------------------------
# Output Excel
# ------------------------------------------------------------

FINAL_EXCEL = os.path.join(
    OUTPUT_FOLDER,
    "Pd_gamma_arylation_COMPLETE_ANALYSIS.xlsx"
)


# ============================================================
# CHEMICAL SETTINGS
# ============================================================

PD_ATOM = 1

LIGAND_START = 17

# Coordination is classified from the actual nearest Pd contacts.
# H is always excluded. Element-specific windows are defined below
# in Section 13. These legacy scalar values are retained only as
# metadata for compatibility with the existing workbook.
PD_DONOR_MIN = 2.00
PD_DONOR_MAX = 2.40

DONOR_ELEMENTS = {"N", "P", "S", "O"}


# ============================================================
# MORFEUS SETTINGS
# ============================================================

SASA_PROBE_RADIUS = 1.4

# Keep only chemically distinct radii to avoid redundant parameters.
# 3.0 Å = proximal Pd pocket; 4.0 Å = near-ligand shell;
# 5.5 Å = broader/distal ligand environment.
VBUR_RADII = [
    3.0,
    4.0,
    5.5
]

DISTAL_VOLUME_RADIUS = 3.5

# ------------------------------------------------------------
# IMPORTANT FIX:
# morfeus.BuriedVolume has NO "n_points" parameter -- it only
# accepts "density" (volume per grid point, in Å³; SMALLER
# density = MORE points = finer resolution). Every BuriedVolume(...)
# call in this script previously passed n_points=VBUR_N_POINTS,
# which raised TypeError: unexpected keyword argument 'n_points'.
# That error was caught by each surrounding try/except, so every
# single %Vbur column, the distal volume, the molecular volume,
# and any directional/quadrant analysis were silently NaN for
# every ligand -- this is why distal volume looked "missing".
#
# morfeus's own default is density=0.001, which for a typical
# r=3.5 A sphere already works out to roughly 180,000 points
# (sphere volume / density) -- finer than the 100,000-point
# target this script was aiming for. VBUR_DENSITY below is set
# even finer than morfeus's own default for extra precision.
# ------------------------------------------------------------

VBUR_DENSITY = 0.0008

CALCULATE_DIRECTIONAL_VBUR = False
CALCULATE_DISPERSION = True

VBUR_RADII_TYPE = "bondi"

SOLID_ANGLE_RADII_TYPE = "bondi"

STERIMOL_RADII_TYPE = "crc"


# ============================================================
# xTB SETTINGS
# ============================================================

XTB_METHOD = 2                 # GFN2-xTB
XTB_CHARGE = 0
XTB_N_UNPAIRED = 0
XTB_PROCESSES = 2


# ============================================================
# OUTLIER SETTINGS
# ============================================================

OUTLIER_FRACTION_LIMIT = 0.10

IQR_MULTIPLIER = 1.5

MAD_Z_LIMIT = 3.5


# ============================================================
# HIGH CORRELATION THRESHOLD
# ============================================================

HIGH_CORR_THRESHOLD = 0.80


# ============================================================
# 5. YIELD DATA
# ============================================================

YIELD_DATA = {

    "L7": 0,
    "L8": 0,
    "L9": 0,
    "L10": 0,
    "L11": 0,
    "L12": 0,
    "L13": 0,
    "L14": 0,
    "L15": 0,

    "L16": 12,
    "L17": 53,
    "L18": 46,
    "L19": 67,
    "L20": 40,
    "L21": 31,

    "L22": 0,
    "L23": 0,
    "L24": 0,
    "L25": 0
}


# ============================================================
# 6. CREATE PLOT FOLDERS
# ============================================================

PLOT_FOLDER = os.path.join(
    OUTPUT_FOLDER,
    "PLOTS"
)

SCATTER_FOLDER = os.path.join(
    PLOT_FOLDER,
    "Yield_vs_Descriptors"
)

BOX_FOLDER = os.path.join(
    PLOT_FOLDER,
    "Descriptor_Boxplots"
)

os.makedirs(PLOT_FOLDER, exist_ok=True)
os.makedirs(SCATTER_FOLDER, exist_ok=True)
os.makedirs(BOX_FOLDER, exist_ok=True)


# ============================================================
# 7. xTB CHECK / INSTALL
# ============================================================

print("=" * 80)
print("CHECKING xTB")
print("=" * 80)

xtb_path = shutil.which("xtb")

if xtb_path is None:

    print("\n✗ xTB executable not found.")
    print("Installing xTB binary...\n")

    xtb_archive = "/content/xtb.tar.xz"

    subprocess.run(
        [
            "wget",
            "-q",
            "--show-progress",
            "https://github.com/grimme-lab/xtb/releases/latest/download/xtb-6.7.1-linux-x86_64.tar.xz",
            "-O",
            xtb_archive
        ],
        check=True
    )

    subprocess.run(
        [
            "tar",
            "-xf",
            xtb_archive,
            "-C",
            "/content"
        ],
        check=True
    )

    xtb_dirs = sorted([
        d
        for d in os.listdir("/content")
        if d.startswith("xtb-")
        and os.path.isdir(
            os.path.join("/content", d)
        )
    ])

    if not xtb_dirs:

        raise RuntimeError(
            "xTB archive extracted but "
            "xTB directory was not found."
        )

    xtb_dir = os.path.join(
        "/content",
        xtb_dirs[-1]
    )

    xtb_bin = os.path.join(
        xtb_dir,
        "bin"
    )

    xtb_executable = os.path.join(
        xtb_bin,
        "xtb"
    )

    if not os.path.exists(
        xtb_executable
    ):

        raise RuntimeError(
            f"xTB executable not found:\n"
            f"{xtb_executable}"
        )

    os.environ["PATH"] = (
        xtb_bin
        + ":"
        + os.environ["PATH"]
    )

    os.environ["XTBHOME"] = xtb_dir

    print("\n✓ xTB installed")
    print(
        "Executable:",
        xtb_executable
    )


xtb_path = shutil.which("xtb")

if xtb_path is None:

    raise RuntimeError(
        "xTB is still unavailable."
    )


print("\n" + "=" * 80)
print("xTB VERIFICATION")
print("=" * 80)

print(
    "xTB executable:",
    xtb_path
)

version = subprocess.run(
    ["xtb", "--version"],
    capture_output=True,
    text=True
)

print(version.stdout)

print("=" * 80)
print("✓ xTB READY")
print("=" * 80)


# ============================================================
# 7b. IMPORT MORFEUS
#
# Only now, with xtb confirmed reachable via shutil.which()
# (checked implicitly by everything above), is it safe to
# import morfeus. This is what makes morfeus.XTB's internal
# @requires_executable("xtb") check pass instead of getting
# permanently cached as "missing" for this kernel session.
# ============================================================

from morfeus import (
    SASA,
    SolidAngle,
    BuriedVolume,
    Sterimol,
    BiteAngle,
    Pyramidalization,
    XTB
)

print("✓ morfeus imported (xtb was already on PATH)")


# ============================================================
# 8. HELPER FUNCTIONS
# ============================================================

def normalize_element(element):
    """Return a normalized chemical element symbol.

    XYZ files normally contain symbols such as C, N, O, S, P, H, and Pd.
    This helper also tolerates mixed-case/whitespace input so donor
    classification does not fail on element-label formatting.
    """
    if element is None:
        return ""

    text = str(element).strip()
    if not text:
        return ""

    # Common two-letter element symbols used in this project.
    if text.lower() == "pd":
        return "Pd"

    return text[0].upper() + text[1:].lower()


def finite(value):

    try:

        value = float(value)

        if np.isfinite(value):
            return value

    except Exception:
        pass

    return np.nan


def distance(a, b):

    return float(
        np.linalg.norm(
            np.asarray(a)
            -
            np.asarray(b)
        )
    )


def angle_deg(a, b, c):

    v1 = (
        np.asarray(a)
        -
        np.asarray(b)
    )

    v2 = (
        np.asarray(c)
        -
        np.asarray(b)
    )

    n1 = np.linalg.norm(v1)
    n2 = np.linalg.norm(v2)

    if n1 == 0 or n2 == 0:
        return np.nan

    cosine = np.dot(v1, v2) / (
        n1 * n2
    )

    cosine = np.clip(
        cosine,
        -1.0,
        1.0
    )

    return float(
        np.degrees(
            np.arccos(cosine)
        )
    )


def clean_ligand_name(filename):

    base = os.path.basename(filename)

    base = re.sub(
        r"\.xyz$",
        "",
        base,
        flags=re.IGNORECASE
    )

    base = re.sub(
        r"[_-](new|opt|optimized)$",
        "",
        base,
        flags=re.IGNORECASE
    )

    match = re.search(
        r"L\d+",
        base,
        flags=re.IGNORECASE
    )

    if match:
        return match.group(0).upper()

    return base


def safe_filename(text):

    return re.sub(
        r"[^A-Za-z0-9_.-]+",
        "_",
        str(text)
    )


# ============================================================
# 9. XYZ READER
# ============================================================

def read_xyz_file(filepath):

    with open(
        filepath,
        "r",
        encoding="utf-8"
    ) as f:

        lines = [
            x.strip()
            for x in f
            if x.strip()
        ]

    if len(lines) < 3:

        raise ValueError(
            "XYZ file too short."
        )

    n_atoms = int(
        lines[0].split()[0]
    )

    atom_lines = lines[2:]

    if len(atom_lines) < n_atoms:

        raise ValueError(
            "Number of coordinate lines "
            "is smaller than atom count."
        )

    elements = []
    coordinates = []

    for line in atom_lines[:n_atoms]:

        parts = line.split()

        elements.append(
            parts[0]
        )

        coordinates.append([
            float(parts[1]),
            float(parts[2]),
            float(parts[3])
        ])

    coordinates = np.asarray(
        coordinates,
        dtype=float
    )

    return elements, coordinates


# ============================================================
# 10. FIND XYZ FILES
# ============================================================

xyz_files = sorted(
    glob.glob(
        os.path.join(
            XYZ_FOLDER,
            "*.xyz"
        )
    )
)

print("\n" + "=" * 80)
print("XYZ FILES")
print("=" * 80)

print(
    f"Found {len(xyz_files)} XYZ files."
)

if len(xyz_files) == 0:

    raise FileNotFoundError(
        f"No XYZ files found in:\n"
        f"{XYZ_FOLDER}"
    )

for f in xyz_files:
    print(
        " ",
        os.path.basename(f)
    )


# ============================================================
# 11. COVALENT RADII
# ============================================================

COVALENT_RADII = {

    "H": 0.31,
    "C": 0.76,
    "N": 0.71,
    "O": 0.66,
    "F": 0.57,

    "P": 1.07,
    "S": 1.05,

    "Cl": 1.02,
    "Br": 1.20,
    "I": 1.39,

    "Pd": 1.39
}


# ============================================================
# 12. COVALENT NEIGHBORS
# ============================================================

def covalent_neighbors(
    elements,
    coordinates,
    atom_id,
    allowed_indices=None,
    scale=1.25
):

    if allowed_indices is None:

        allowed_indices = range(
            1,
            len(elements) + 1
        )

    atom_element = elements[
        atom_id - 1
    ]

    if atom_element not in COVALENT_RADII:

        return []

    r1 = COVALENT_RADII[
        atom_element
    ]

    result = []

    for j in allowed_indices:

        if j == atom_id:
            continue

        element_j = elements[
            j - 1
        ]

        if element_j not in COVALENT_RADII:
            continue

        r2 = COVALENT_RADII[
            element_j
        ]

        d = distance(
            coordinates[atom_id - 1],
            coordinates[j - 1]
        )

        cutoff = scale * (
            r1 + r2
        )

        if d <= cutoff:

            result.append(
                (j, d)
            )

    return sorted(
        result,
        key=lambda x: x[1]
    )


# ============================================================
# 13. Pd DONOR IDENTIFICATION + LIGAND CLASSIFICATION
#
# CLASSIFICATION IS STRUCTURE-BASED, NOT L-NUMBER-BASED.
#
# Pd = atom 1
# Ligand = atom 17 -> last atom
# H is NEVER considered a Pd donor.
#
# Ordinary donor windows used here:
#   N/O : 2.00 - 2.40 Å
#   S   : 2.40 - 2.65 Å
#   P   : 2.00 - 2.40 Å (retained from the original script)
#
# Special pi coordination:
#   aromatic C(pi) is considered separately, only as a Pd-pi
#   contact representative; it is NOT treated as an ordinary
#   sigma Pd-C donor.
# ============================================================

PD_DONOR_WINDOWS = {
    "N": (2.00, 2.40),
    "O": (2.00, 2.40),
    "S": (2.40, 2.65),
    "P": (2.00, 2.40),
}

PI_C_MIN_DISTANCE = 2.00
PI_C_MAX_DISTANCE = 3.00


def find_pd_donors(elements, coordinates):
    """
    Find Pd-coordination candidates from the actual geometry.

    Important rules:
      1. Pd is atom 1.
      2. Only ligand atoms (17 -> last) are examined.
      3. H is never considered.
      4. N/O/S/P are ordinary donor candidates.
      5. Candidate selection is based on Pd distance.
      6. S uses its longer Pd-S coordination window.

    Returns:
      candidates  = all N/O/S/P candidates with distances and
                    element-specific windows
      qualifying  = candidates inside their element-specific window,
                    sorted by Pd distance
    """

    pd_xyz = coordinates[PD_ATOM - 1]
    candidates = []

    for atom_id in range(LIGAND_START, len(elements) + 1):

        element = normalize_element(elements[atom_id - 1])

        # Hydrogen is explicitly excluded.
        if element == "H":
            continue

        if element not in PD_DONOR_WINDOWS:
            continue

        d = distance(pd_xyz, coordinates[atom_id - 1])
        dmin, dmax = PD_DONOR_WINDOWS[element]

        candidates.append({
            "Atom": atom_id,
            "Element": element,
            "Pd_Distance_A": d,
            "Window_Min_A": dmin,
            "Window_Max_A": dmax,
            "Qualifies": dmin <= d <= dmax,
        })

    candidates = sorted(
        candidates,
        key=lambda x: x["Pd_Distance_A"]
    )

    qualifying = [
        x for x in candidates
        if x["Qualifies"]
    ]

    return candidates, qualifying


def find_pi_carbon_candidates(elements, coordinates):
    """
    Find aromatic C atoms that can serve as geometric C(pi)
    representatives for Pd-pi coordination.

    This is intentionally separate from ordinary donor selection.
    Carbon is NOT treated as a normal sigma donor here.
    """

    pd_xyz = coordinates[PD_ATOM - 1]
    candidates = []

    ligand_indices = list(
        range(LIGAND_START, len(elements) + 1)
    )

    for atom_id in ligand_indices:

        element = normalize_element(elements[atom_id - 1])

        if element != "C":
            continue

        d = distance(
            pd_xyz,
            coordinates[atom_id - 1]
        )

        if PI_C_MIN_DISTANCE <= d <= PI_C_MAX_DISTANCE:
            candidates.append({
                "Atom": atom_id,
                "Element": "C(pi)",
                "Pd_Distance_A": d,
            })

    return sorted(
        candidates,
        key=lambda x: x["Pd_Distance_A"]
    )


def classify_ligand_from_coordination(
    donor1_element,
    donor2_element,
    donor1_atom=None,
    donor2_atom=None,
    pi_coordination=False,
):
    """
    Assign LigandClass from the actual Pd-coordinating atoms.

    NN = N + N
    NO = N + O
    NS = N + S
    NP = N + P
    OP = O + P
    OO = O + O
    PP = P + P
    NPI = N + C(pi)

    The function does not use ligand number.
    """

    if donor1_element is None or donor2_element is None:
        return "NOT_AVAILABLE"

    pair = [str(donor1_element), str(donor2_element)]

    if pi_coordination and "C(pi)" in pair and "N" in pair:
        return "NPI"

    pair_sorted = tuple(sorted(pair))

    pair_map = {
        ("N", "N"): "NN",
        ("N", "O"): "NO",
        ("N", "P"): "NP",
        ("N", "S"): "NS",
        ("O", "O"): "OO",
        ("O", "P"): "OP",
        ("O", "S"): "OS",
        ("P", "P"): "PP",
        ("P", "S"): "PS",
        ("S", "S"): "SS",
    }

    return pair_map.get(
        pair_sorted,
        "OTHER_" + "-".join(pair_sorted)
    )


def select_pd_donors(elements, coordinates, qualifying):
    """
    Select the two Pd-coordinating partners from actual geometry.

    The primary rule is nearest Pd distance, with one important
    chemical safeguard: when a chelated pair such as N+N, N+O,
    N+S, or N+P is available, the two closest qualifying atoms are
    used. No ligand-number lookup is performed.

    If more than two qualifying ordinary donors exist, the two
    shortest Pd distances are retained and the QC column records
    that multiple candidates were present.
    """

    qualifying = sorted(
        qualifying,
        key=lambda x: x["Pd_Distance_A"]
    )

    result = {
        "LigandClass": "NOT_AVAILABLE",
        "ExpectedDonorPair": None,
        "Donor1_Atom": np.nan,
        "Donor1_Element": None,
        "Pd_Donor1_Distance_A": np.nan,
        "Donor2_Atom": np.nan,
        "Donor2_Element": None,
        "Pd_Donor2_Distance_A": np.nan,
        "Donor_Selection_Check": None,
        "PiCoordination": False,
        "PiCoordinationAtom": np.nan,
        "PiCoordinationPd_C_Distance_A": np.nan,
        "PiCoordinationDescription": None,
    }

    if len(qualifying) == 0:
        result["Donor_Selection_Check"] = "NO_QUALIFYING_DONORS"
        return result, None, None

    selected = qualifying[:2]

    donor1 = selected[0]["Atom"]
    donor2 = selected[1]["Atom"] if len(selected) >= 2 else None

    result["Donor1_Atom"] = donor1
    result["Donor1_Element"] = selected[0]["Element"]
    result["Pd_Donor1_Distance_A"] = selected[0]["Pd_Distance_A"]

    if donor2 is not None:
        result["Donor2_Atom"] = donor2
        result["Donor2_Element"] = selected[1]["Element"]
        result["Pd_Donor2_Distance_A"] = selected[1]["Pd_Distance_A"]

    if donor2 is None:
        result["LigandClass"] = selected[0]["Element"]
        result["Donor_Selection_Check"] = "ONLY_ONE_QUALIFYING_DONOR"
        return result, donor1, None

    ligand_class = classify_ligand_from_coordination(
        result["Donor1_Element"],
        result["Donor2_Element"],
        donor1,
        donor2,
        pi_coordination=False,
    )

    result["LigandClass"] = ligand_class
    result["ExpectedDonorPair"] = (
        f"{result['Donor1_Element']}-{result['Donor2_Element']}"
    )

    if len(qualifying) == 2:
        result["Donor_Selection_Check"] = "PASS"
    else:
        result["Donor_Selection_Check"] = (
            "PASS_MULTIPLE_CANDIDATES_NEAREST_TWO"
        )

    # --------------------------------------------------------
    # Detect possible N + pi coordination separately.
    # This does NOT replace a valid ordinary donor pair.
    # It is primarily useful for structures with only one
    # ordinary donor plus an aromatic Pd-pi contact.
    # --------------------------------------------------------
    if len(qualifying) == 1 and result["Donor1_Element"] == "N":

        pi_candidates = find_pi_carbon_candidates(
            elements,
            coordinates
        )

        if pi_candidates:
            pi_c = pi_candidates[0]

            result["Donor2_Atom"] = pi_c["Atom"]
            result["Donor2_Element"] = "C(pi)"
            result["Pd_Donor2_Distance_A"] = pi_c["Pd_Distance_A"]
            result["LigandClass"] = "NPI"
            result["ExpectedDonorPair"] = "N-C(pi)"
            result["PiCoordination"] = True
            result["PiCoordinationAtom"] = pi_c["Atom"]
            result["PiCoordinationPd_C_Distance_A"] = pi_c["Pd_Distance_A"]
            result["PiCoordinationDescription"] = (
                "Pd-N plus Pd···C(aromatic) pi contact; "
                "C(pi) is not treated as a sigma donor."
            )
            result["Donor_Selection_Check"] = "PASS_N_PI"
            return result, donor1, pi_c["Atom"]

    return result, donor1, donor2


# ============================================================
# 14. BUILD LIGAND GRAPH
# ============================================================

def build_ligand_graph(
    elements,
    coordinates
):

    ligand_indices = list(
        range(
            LIGAND_START,
            len(elements) + 1
        )
    )

    graph = {
        i: set()
        for i in ligand_indices
    }

    for i in ligand_indices:

        neighbors = covalent_neighbors(
            elements,
            coordinates,
            i,
            allowed_indices=ligand_indices
        )

        for j, d in neighbors:

            graph[i].add(j)
            graph[j].add(i)

    return graph


# ============================================================
# 15. SHORTEST DONOR-DONOR LIGAND PATH
# ============================================================

def shortest_ligand_path(
    graph,
    start,
    end
):

    if start is None or end is None:
        return None

    queue = [
        (
            start,
            [start]
        )
    ]

    visited = {
        start
    }

    while queue:

        current, path = queue.pop(0)

        if current == end:
            return path

        for neighbor in sorted(
            graph.get(current, [])
        ):

            if neighbor not in visited:

                visited.add(
                    neighbor
                )

                queue.append(
                    (
                        neighbor,
                        path + [neighbor]
                    )
                )

    return None


# ============================================================
# 16. PALLADACYCLE CLASSIFICATION
#
# NN is subdivided by the conventional Pd-containing ring size:
#
#   NN_5 = Pd-N-C-C-N-Pd  -> 5-membered ring
#   NN_6 = Pd-N-C-C-C-N-Pd -> 6-membered ring
#
# NO is retained as NO and is not forced into NN_5/NN_6.
#
# Ring size includes Pd.
# ============================================================

def determine_palladacycle(
    elements,
    coordinates,
    donor1,
    donor2,
    ligand_class=None,
):

    result = {
        "Palladacycle_Type": "NOT_AVAILABLE",
        "Palladacycle_Ring_Size": np.nan,
        "Palladacycle_Ligand_Path": None,
        "Palladacycle_Path_Bond_Mean_A": np.nan,
        "Palladacycle_Path_Bond_SD_A": np.nan,
        "Palladacycle_Path_Bond_CV": np.nan,
        "Expected_Palladacycle_Type": None,
        "Palladacycle_Type_Check": "NO_EXPECTATION",
    }

    if ligand_class is None:
        ligand_class = classify_ligand_from_coordination(
            elements[donor1 - 1] if donor1 else None,
            elements[donor2 - 1] if donor2 else None,
            donor1,
            donor2,
        )

    if donor1 is None or donor2 is None:
        return result

    # C(pi) is not an ordinary covalent ligand-backbone endpoint.
    if ligand_class == "NPI":
        result["Palladacycle_Type"] = "NPI"
        result["Palladacycle_Type_Check"] = "NOT_APPLICABLE"
        return result

    graph = build_ligand_graph(
        elements,
        coordinates
    )

    path = shortest_ligand_path(
        graph,
        donor1,
        donor2
    )

    if path is None:
        result["Palladacycle_Type"] = ligand_class
        result["Palladacycle_Type_Check"] = "PATH_NOT_FOUND"
        return result

    result["Palladacycle_Ligand_Path"] = "-".join(
        map(str, path)
    )

    # --------------------------------------------------------
    # Ring size includes Pd.
    # Pd-N-(ligand path)-N-Pd has:
    #     len(path) ligand atoms + 1 Pd atom
    # --------------------------------------------------------
    ring_size = len(path) + 1
    result["Palladacycle_Ring_Size"] = ring_size

    if ligand_class == "NN":

        if len(path) == 4:
            # Pd-N-C-C-N-Pd
            result["Palladacycle_Type"] = "NN_5"

        elif len(path) == 5:
            # Pd-N-C-C-C-N-Pd
            result["Palladacycle_Type"] = "NN_6"

        else:
            result["Palladacycle_Type"] = f"NN_RING_{ring_size}"

    elif ligand_class == "NO":
        result["Palladacycle_Type"] = "NO"

    else:
        result["Palladacycle_Type"] = ligand_class

    # --------------------------------------------------------
    # Path bond geometry
    # --------------------------------------------------------
    bond_lengths = []

    for i in range(len(path) - 1):
        a = path[i]
        b = path[i + 1]

        bond_lengths.append(
            distance(
                coordinates[a - 1],
                coordinates[b - 1]
            )
        )

    if bond_lengths:
        mean_bond = float(np.mean(bond_lengths))
        sd_bond = float(np.std(bond_lengths, ddof=0))

        result["Palladacycle_Path_Bond_Mean_A"] = mean_bond
        result["Palladacycle_Path_Bond_SD_A"] = sd_bond

        if mean_bond > 0:
            result["Palladacycle_Path_Bond_CV"] = (
                sd_bond / mean_bond
            )

    # No ligand-number expectation is used anymore.
    result["Expected_Palladacycle_Type"] = result["Palladacycle_Type"]
    result["Palladacycle_Type_Check"] = "PASS"

    return result


# ============================================================
# 17. STERIC DESCRIPTORS
# ============================================================

def calculate_steric(
    elements,
    coordinates,
    donor1
):

    result = {}

    ligand_indices = list(
        range(
            LIGAND_START,
            len(elements) + 1
        )
    )

    ligand_elements = [
        elements[i - 1]
        for i in ligand_indices
    ]

    ligand_coordinates = coordinates[
        np.array(ligand_indices) - 1
    ]

    # --------------------------------------------------------
    # Maximum ligand extension
    # --------------------------------------------------------

    max_extension = 0.0

    for i in range(
        len(ligand_coordinates)
    ):

        for j in range(
            i + 1,
            len(ligand_coordinates)
        ):

            d = distance(
                ligand_coordinates[i],
                ligand_coordinates[j]
            )

            max_extension = max(
                max_extension,
                d
            )

    result[
        "Ligand_Maximum_Extension_A"
    ] = finite(
        max_extension
    )


    # --------------------------------------------------------
    # SASA
    # --------------------------------------------------------

    try:

        sasa = SASA(
            ligand_elements,
            ligand_coordinates,
            probe_radius=SASA_PROBE_RADIUS
        )

        result[
            "Ligand_SASA_A2"
        ] = finite(
            sasa.area
        )

        result[
            "Ligand_SASA_Volume_A3"
        ] = finite(
            sasa.volume
        )

    except Exception as e:

        result[
            "SASA_Warning"
        ] = str(e)


    # --------------------------------------------------------
    # Solid angle
    #
    # ligand-only:
    # reduced system = Pd + ligand
    # --------------------------------------------------------

    try:

        reduced_elements = (
            ["Pd"]
            +
            ligand_elements
        )

        reduced_coordinates = np.vstack([
            coordinates[
                PD_ATOM - 1
            ],
            ligand_coordinates
        ])

        sa = SolidAngle(
            reduced_elements,
            reduced_coordinates,
            metal_index=1,
            radii_type=SOLID_ANGLE_RADII_TYPE
        )

        result[
            "Ligand_Solid_Angle_sr"
        ] = finite(
            sa.solid_angle
        )

        result[
            "Ligand_Solid_Cone_Angle_deg"
        ] = finite(
            sa.cone_angle
        )

        result[
            "Ligand_Solid_G_percent"
        ] = finite(
            sa.G
        )

    except Exception as e:

        result[
            "SolidAngle_Warning"
        ] = str(e)


    # --------------------------------------------------------
    # Buried volume and nonredundant distal sterics
    # --------------------------------------------------------

    excluded_atoms = [

        i
        for i in range(
            1,
            len(elements) + 1
        )

        if (
            i != PD_ATOM
            and
            i not in ligand_indices
        )
    ]


    for radius in VBUR_RADII:

        key = f"Ligand_%Vbur_{radius:.1f}A"

        try:

            bv = BuriedVolume(
                elements,
                coordinates,
                metal_index=PD_ATOM,
                excluded_atoms=excluded_atoms,
                radius=radius,
                include_hs=False,
                radii_type=VBUR_RADII_TYPE,
                density=VBUR_DENSITY
            )

            result[key] = finite(
                bv.fraction_buried_volume * 100.0
            )

        except Exception as e:

            result[key] = np.nan
            result[f"{key}_Warning"] = str(e)


    # --------------------------------------------------------
    # Distal volume outside a single proximal sphere
    # --------------------------------------------------------

    try:

        bv_distal = BuriedVolume(
            elements,
            coordinates,
            metal_index=PD_ATOM,
            excluded_atoms=excluded_atoms,
            radius=DISTAL_VOLUME_RADIUS,
            include_hs=False,
            radii_type=VBUR_RADII_TYPE,
            density=VBUR_DENSITY
        )

        bv_distal.compute_distal_volume(method="sasa")

        distal = finite(bv_distal.distal_volume)
        molecular = finite(bv_distal.molecular_volume)

        result[
            f"Ligand_Distal_Volume_Outside_{DISTAL_VOLUME_RADIUS:.1f}A_A3"
        ] = distal

        result["Ligand_Molecular_Volume_A3"] = molecular

        result[
            f"Ligand_Distal_Volume_Fraction_Outside_{DISTAL_VOLUME_RADIUS:.1f}A"
        ] = finite(distal / molecular) if (
            pd.notna(distal)
            and pd.notna(molecular)
            and molecular > 0
        ) else np.nan

    except Exception as e:

        result["Distal_Volume_Warning"] = str(e)
        result[
            f"Ligand_Distal_Volume_Outside_{DISTAL_VOLUME_RADIUS:.1f}A_A3"
        ] = np.nan
        result["Ligand_Molecular_Volume_A3"] = np.nan
        result[
            f"Ligand_Distal_Volume_Fraction_Outside_{DISTAL_VOLUME_RADIUS:.1f}A"
        ] = np.nan


    # --------------------------------------------------------
    # Optional directional sterics
    #
    # Disabled by default because quadrant/octet labels require
    # identical chemical orientation across every XYZ structure.
    # --------------------------------------------------------

    if CALCULATE_DIRECTIONAL_VBUR:

        try:

            bv_directional = BuriedVolume(
                elements,
                coordinates,
                metal_index=PD_ATOM,
                excluded_atoms=excluded_atoms,
                radius=DISTAL_VOLUME_RADIUS,
                include_hs=False,
                radii_type=VBUR_RADII_TYPE,
                density=VBUR_DENSITY
            )

            bv_directional.octant_analysis()

            q_values = [
                finite(value * 100.0)
                for value in bv_directional.quadrants.values()
            ]

            o_values = [
                finite(value * 100.0)
                for value in bv_directional.octants.values()
            ]

            q_values = [x for x in q_values if pd.notna(x)]
            o_values = [x for x in o_values if pd.notna(x)]

            if q_values:

                result["Vbur_3.5A_Quadrant_Range_percent"] = finite(
                    max(q_values) - min(q_values)
                )

                result["Vbur_3.5A_Quadrant_SD_percent"] = finite(
                    np.std(q_values, ddof=0)
                )

            if o_values:

                result["Vbur_3.5A_Octant_Range_percent"] = finite(
                    max(o_values) - min(o_values)
                )

                result["Vbur_3.5A_Octant_SD_percent"] = finite(
                    np.std(o_values, ddof=0)
                )

        except Exception as e:

            result["Directional_Vbur_Warning"] = str(e)


    # --------------------------------------------------------
    # Dispersion descriptor, if supported by this Morfeus build
    # --------------------------------------------------------

    if CALCULATE_DISPERSION:

        try:

            from morfeus import Dispersion

            disp = Dispersion(
                ligand_elements,
                ligand_coordinates
            )

            pint = finite(disp.p_int)
            sasa_value = result.get("Ligand_SASA_A2", np.nan)
            volume_value = result.get("Ligand_Molecular_Volume_A3", np.nan)

            result["Ligand_Pint"] = pint
            result["Ligand_Pint_per_SASA"] = finite(
                pint / sasa_value
            ) if pd.notna(sasa_value) and sasa_value > 0 else np.nan
            result["Ligand_Pint_per_Volume"] = finite(
                pint / volume_value
            ) if pd.notna(volume_value) and volume_value > 0 else np.nan

        except Exception as e:

            result["Dispersion_Warning"] = str(e)
            result["Ligand_Pint"] = np.nan
            result["Ligand_Pint_per_SASA"] = np.nan
            result["Ligand_Pint_per_Volume"] = np.nan


    # --------------------------------------------------------
    # Pd-oriented custom Sterimol
    #
    # IMPORTANT:
    # This is NOT standard H-dummy Sterimol.
    # It is a Pd-oriented custom descriptor.
    # --------------------------------------------------------

    if donor1 is not None:

        try:

            reduced_elements = (
                ["Pd"]
                +
                ligand_elements
            )

            reduced_coordinates = np.vstack([
                coordinates[
                    PD_ATOM - 1
                ],
                ligand_coordinates
            ])

            original_to_reduced = {
                original: i + 2
                for i, original
                in enumerate(
                    ligand_indices
                )
            }

            attached_index = (
                original_to_reduced[
                    donor1
                ]
            )

            ster = Sterimol(
                reduced_elements,
                reduced_coordinates,
                dummy_index=1,
                attached_index=attached_index,
                radii_type=STERIMOL_RADII_TYPE
            )

            result[
                "Pd_Oriented_Sterimol_B1_A"
            ] = finite(
                ster.B_1_value
            )

            result[
                "Pd_Oriented_Sterimol_B5_A"
            ] = finite(
                ster.B_5_value
            )

            result[
                "Pd_Oriented_Sterimol_L_A"
            ] = finite(
                ster.L_value
            )

        except Exception as e:

            result[
                "Sterimol_Warning"
            ] = str(e)


    return result


# ============================================================
# 18. COORDINATION DESCRIPTORS
# ============================================================

def calculate_coordination(
    elements,
    coordinates,
    donor1,
    donor2
):

    result = {}

    if donor1 is None:

        return {
            "Pd_Donor1_Distance_A": np.nan,
            "Pd_Donor2_Distance_A": np.nan,
            "Average_Pd_Donor_Distance_A": np.nan,
            "Delta_Pd_Donor_Distance_A": np.nan,
            "Pd_Donor_Distance_Ratio": np.nan,
            "Bite_Angle_deg": np.nan,
            "Donor1_Donor2_Distance_A": np.nan,
            "Pd_Coordination_Distortion_A": np.nan
        }


    pd_xyz = coordinates[
        PD_ATOM - 1
    ]


    d1 = distance(
        pd_xyz,
        coordinates[
            donor1 - 1
        ]
    )


    result[
        "Pd_Donor1_Distance_A"
    ] = d1


    if donor2 is not None:

        d2 = distance(
            pd_xyz,
            coordinates[
                donor2 - 1
            ]
        )

        result[
            "Pd_Donor2_Distance_A"
        ] = d2

        avg = (
            d1 + d2
        ) / 2.0

        delta = abs(
            d1 - d2
        )

        result[
            "Average_Pd_Donor_Distance_A"
        ] = avg

        result[
            "Delta_Pd_Donor_Distance_A"
        ] = delta

        if min(d1, d2) > 0:

            result[
                "Pd_Donor_Distance_Ratio"
            ] = (
                max(d1, d2)
                /
                min(d1, d2)
            )


        result[
            "Pd_Coordination_Distortion_A"
        ] = np.sqrt(
            (
                (d1 - avg) ** 2
                +
                (d2 - avg) ** 2
            ) / 2.0
        )


        result[
            "Donor1_Donor2_Distance_A"
        ] = distance(
            coordinates[
                donor1 - 1
            ],
            coordinates[
                donor2 - 1
            ]
        )

        try:

            bite = BiteAngle(
                coordinates,
                PD_ATOM,
                donor1,
                donor2
            )

            result[
                "Bite_Angle_deg"
            ] = finite(
                bite.angle
            )

        except Exception as e:

            result[
                "Bite_Angle_deg"
            ] = np.nan

            result[
                "Bite_Angle_Warning"
            ] = str(e)

    else:

        result[
            "Pd_Donor2_Distance_A"
        ] = np.nan

        result[
            "Average_Pd_Donor_Distance_A"
        ] = np.nan

        result[
            "Delta_Pd_Donor_Distance_A"
        ] = np.nan

        result[
            "Pd_Donor_Distance_Ratio"
        ] = np.nan

        result[
            "Pd_Coordination_Distortion_A"
        ] = np.nan

        result[
            "Donor1_Donor2_Distance_A"
        ] = np.nan

        result[
            "Bite_Angle_deg"
        ] = np.nan


    return result


# ============================================================
# 19. PYRAMIDALIZATION
# ============================================================

def calculate_pyramidalization(
    elements,
    coordinates,
    donor1,
    donor2
):

    result = {}

    ligand_indices = list(
        range(
            LIGAND_START,
            len(elements) + 1
        )
    )


    def calculate_one(
        donor,
        prefix
    ):

        if donor is None:

            return {

                f"{prefix}_Neighbor_Count":
                    np.nan,

                f"{prefix}_P":
                    np.nan,

                f"{prefix}_P_Angle_deg":
                    np.nan,

                f"{prefix}_P_Warning":
                    "Donor unavailable.",

                f"{prefix}_Mean_Bond_Angle_deg":
                    np.nan,

                f"{prefix}_Mean_Bond_Angle_N":
                    np.nan
            }


        neighbors = covalent_neighbors(
            elements,
            coordinates,
            donor,
            allowed_indices=ligand_indices
        )

        neighbor_ids = [
            x[0]
            for x in neighbors
        ]

        out = {

            f"{prefix}_Neighbor_Count":
                len(neighbor_ids)
        }


        # ----------------------------------------------------
        # Generalized local bond-angle descriptor.
        #
        # Unlike Pyramidalization.P (which strictly requires 3
        # real bonded neighbors and is undefined for genuinely
        # 2-coordinate sp2 donors like pyridine/imine N), this
        # works for ANY donor with >=2 bonded neighbors:
        #   - 2 neighbors -> the single neighbor-donor-neighbor
        #     angle (e.g. the internal ring angle at a pyridine
        #     N)
        #   - 3+ neighbors -> the average of every pairwise
        #     neighbor-donor-neighbor angle
        # This is always populated for the aromatic/2-coordinate
        # donors that make Pyramidalization_P come back NaN,
        # so it survives the "no missing values" filter used
        # before correlation analysis where Pyramidalization_P
        # does not.
        # ----------------------------------------------------

        if len(neighbor_ids) >= 2:

            donor_xyz = coordinates[donor - 1]

            pairwise_angles = []

            for a_id, b_id in itertools.combinations(neighbor_ids, 2):

                a_xyz = coordinates[a_id - 1]
                b_xyz = coordinates[b_id - 1]

                ang = angle_deg(a_xyz, donor_xyz, b_xyz)

                if pd.notna(ang):
                    pairwise_angles.append(ang)

            if pairwise_angles:

                out[f"{prefix}_Mean_Bond_Angle_deg"] = finite(
                    np.mean(pairwise_angles)
                )

                out[f"{prefix}_Mean_Bond_Angle_N"] = len(pairwise_angles)

            else:

                out[f"{prefix}_Mean_Bond_Angle_deg"] = np.nan
                out[f"{prefix}_Mean_Bond_Angle_N"] = np.nan

        else:

            out[f"{prefix}_Mean_Bond_Angle_deg"] = np.nan
            out[f"{prefix}_Mean_Bond_Angle_N"] = np.nan


        # ----------------------------------------------------
        # Need at least 3 neighbors
        # ----------------------------------------------------

        if len(neighbor_ids) < 3:

            out[
                f"{prefix}_P"
            ] = np.nan

            out[
                f"{prefix}_P_Angle_deg"
            ] = np.nan

            out[
                f"{prefix}_P_Warning"
            ] = (
                f"{len(neighbor_ids)} "
                "covalent neighbors; "
                "pyramidalization not safely defined."
            )

            return out


        selected = neighbor_ids[:3]


        try:

            pyr = Pyramidalization(
                coordinates,
                donor,
                neighbor_indices=selected
            )

            out[
                f"{prefix}_P"
            ] = finite(
                pyr.P
            )

            out[
                f"{prefix}_P_Angle_deg"
            ] = finite(
                pyr.P_angle
            )

        except Exception as e:

            out[
                f"{prefix}_P"
            ] = np.nan

            out[
                f"{prefix}_P_Angle_deg"
            ] = np.nan

            out[
                f"{prefix}_P_Warning"
            ] = str(e)


        return out


    d1 = calculate_one(
        donor1,
        "Donor1_Pyramidalization"
    )

    d2 = calculate_one(
        donor2,
        "Donor2_Pyramidalization"
    )

    result.update(d1)
    result.update(d2)


    # --------------------------------------------------------
    # Average / delta P
    # --------------------------------------------------------

    p1 = result.get(
        "Donor1_Pyramidalization_P",
        np.nan
    )

    p2 = result.get(
        "Donor2_Pyramidalization_P",
        np.nan
    )

    if (
        pd.notna(p1)
        and
        pd.notna(p2)
    ):

        result[
            "Average_Pyramidalization_P"
        ] = (
            p1 + p2
        ) / 2.0

        result[
            "Delta_Pyramidalization_P"
        ] = abs(
            p1 - p2
        )

    else:

        result[
            "Average_Pyramidalization_P"
        ] = np.nan

        result[
            "Delta_Pyramidalization_P"
        ] = np.nan


    # --------------------------------------------------------
    # Average / delta P angle
    # --------------------------------------------------------

    a1 = result.get(
        "Donor1_Pyramidalization_P_Angle_deg",
        np.nan
    )

    a2 = result.get(
        "Donor2_Pyramidalization_P_Angle_deg",
        np.nan
    )

    if (
        pd.notna(a1)
        and
        pd.notna(a2)
    ):

        result[
            "Average_Pyramidalization_Angle_deg"
        ] = (
            a1 + a2
        ) / 2.0

        result[
            "Delta_Pyramidalization_Angle_deg"
        ] = abs(
            a1 - a2
        )

    else:

        result[
            "Average_Pyramidalization_Angle_deg"
        ] = np.nan

        result[
            "Delta_Pyramidalization_Angle_deg"
        ] = np.nan


    # --------------------------------------------------------
    # Average / delta of the generalized mean bond angle.
    #
    # This pair is populated for nearly every ligand (any donor
    # with >=2 bonded neighbors), unlike Average/Delta_
    # Pyramidalization_P above which needs BOTH donors to have
    # exactly 3 real neighbors and is NaN for almost this whole
    # dataset.
    # --------------------------------------------------------

    m1 = result.get(
        "Donor1_Pyramidalization_Mean_Bond_Angle_deg",
        np.nan
    )

    m2 = result.get(
        "Donor2_Pyramidalization_Mean_Bond_Angle_deg",
        np.nan
    )

    if (
        pd.notna(m1)
        and
        pd.notna(m2)
    ):

        result[
            "Average_Donor_Mean_Bond_Angle_deg"
        ] = (
            m1 + m2
        ) / 2.0

        result[
            "Delta_Donor_Mean_Bond_Angle_deg"
        ] = abs(
            m1 - m2
        )

    else:

        result[
            "Average_Donor_Mean_Bond_Angle_deg"
        ] = np.nan

        result[
            "Delta_Donor_Mean_Bond_Angle_deg"
        ] = np.nan


    return result


# ============================================================
# 20. xTB ELECTRONIC DESCRIPTORS
#
# IMPORTANT:
# ONLY ligand atoms 17 -> LAST
#
# Pd is NOT included in isolated ligand xTB.
# ============================================================

def calculate_xtb(
    elements,
    coordinates,
    donor1,
    donor2
):

    result = {}

    ligand_indices = list(
        range(
            LIGAND_START,
            len(elements) + 1
        )
    )

    ligand_elements = [
        elements[i - 1]
        for i in ligand_indices
    ]

    ligand_coordinates = coordinates[
        np.array(ligand_indices) - 1
    ]


    try:

        xtb = XTB(
            ligand_elements,
            ligand_coordinates,
            method=XTB_METHOD,
            charge=XTB_CHARGE,
            n_unpaired=XTB_N_UNPAIRED,
            n_processes=XTB_PROCESSES
        )


        # ----------------------------------------------------
        # Charges
        #
        # VERIFIED against morfeus source: XTB.get_charges()
        # returns a dict keyed 1..N (1-indexed), e.g.
        # {1: -0.35, 2: 0.12, ...} -- NOT a list, NOT 0-indexed.
        # np.sum(charges) on a dict does not sum the values at
        # all -- it silently returns the dict unchanged, which
        # finite() then quietly turns into NaN with no error.
        # ----------------------------------------------------

        charges = xtb.get_charges(
            model="Mulliken"
        )

        result[
            "XTB_Ligand_Charge_Sum"
        ] = finite(
            sum(charges.values())
        )


        # ----------------------------------------------------
        # HOMO / LUMO
        # ----------------------------------------------------

        result[
            "XTB_HOMO_eV"
        ] = finite(
            xtb.get_homo(
                unit="eV"
            )
        )

        result[
            "XTB_LUMO_eV"
        ] = finite(
            xtb.get_lumo(
                unit="eV"
            )
        )

        result[
            "XTB_HOMO_LUMO_Gap_eV"
        ] = finite(
            xtb.get_homo_lumo_gap()
        )


        # ----------------------------------------------------
        # Fermi
        # ----------------------------------------------------

        result[
            "XTB_Fermi_Level_eV"
        ] = finite(
            xtb.get_fermi_level()
        )


        # ----------------------------------------------------
        # IP / EA
        # ----------------------------------------------------

        result[
            "XTB_IP_eV"
        ] = finite(
            xtb.get_ip(
                corrected=True
            )
        )

        result[
            "XTB_EA_eV"
        ] = finite(
            xtb.get_ea(
                corrected=True
            )
        )


        # ----------------------------------------------------
        # Chemical potential
        # ----------------------------------------------------

        result[
            "XTB_Chemical_Potential_eV"
        ] = finite(
            xtb.get_chemical_potential(
                corrected=True
            )
        )


        # ----------------------------------------------------
        # Electronegativity
        # ----------------------------------------------------

        result[
            "XTB_Electronegativity_eV"
        ] = finite(
            xtb.get_electronegativity(
                corrected=True
            )
        )


        # ----------------------------------------------------
        # Hardness / softness
        # ----------------------------------------------------

        result[
            "XTB_Hardness_eV"
        ] = finite(
            xtb.get_hardness()
        )

        result[
            "XTB_Softness"
        ] = finite(
            xtb.get_softness()
        )


        # ----------------------------------------------------
        # Global electrophilicity / nucleophilicity
        # ----------------------------------------------------

        try:

            result[
                "XTB_Electrophilicity"
            ] = finite(
                xtb.get_global_descriptor(
                    "electrophilicity",
                    corrected=True
                )
            )

        except Exception:

            result[
                "XTB_Electrophilicity"
            ] = np.nan


        try:

            result[
                "XTB_Nucleophilicity"
            ] = finite(
                xtb.get_global_descriptor(
                    "nucleophilicity",
                    corrected=True
                )
            )

        except Exception:

            result[
                "XTB_Nucleophilicity"
            ] = np.nan


        # ----------------------------------------------------
        # Dipole
        # ----------------------------------------------------

        result[
            "XTB_Dipole_Debye"
        ] = finite(
            xtb.get_dipole_moment(
                unit="debye"
            )
        )


        # ----------------------------------------------------
        # Polarizability
        # ----------------------------------------------------

        try:

            result[
                "XTB_Polarizability"
            ] = finite(
                xtb.get_molecular_polarizability()
            )

        except Exception:

            result[
                "XTB_Polarizability"
            ] = np.nan


        # ----------------------------------------------------
        # NFOD
        # ----------------------------------------------------

        try:

            result[
                "XTB_NFOD"
            ] = finite(
                xtb.get_nfod()
            )

        except Exception:

            result[
                "XTB_NFOD"
            ] = np.nan


        # ----------------------------------------------------
        # Donor charges
        #
        # Original atom ID -> isolated ligand index.
        # FIXED: get_charges() keys start at 1 (see note above),
        # so enumerate must also start at 1, not the default 0.
        # With start=0, idx1 for the very first ligand atom was
        # 0, which is not a key in the 1-indexed charges dict --
        # this raised a KeyError caught by the outer except
        # block, silently discarding every donor-charge value.
        # ----------------------------------------------------

        original_to_xtb = {
            original: i
            for i, original
            in enumerate(
                ligand_indices,
                start=1
            )
        }

        if donor1 is not None:

            idx1 = original_to_xtb[
                donor1
            ]

            result[
                "Donor1_Mulliken_Charge"
            ] = finite(
                charges[idx1]
            )

        else:

            result[
                "Donor1_Mulliken_Charge"
            ] = np.nan


        if donor2 is not None:

            idx2 = original_to_xtb[
                donor2
            ]

            result[
                "Donor2_Mulliken_Charge"
            ] = finite(
                charges[idx2]
            )

        else:

            result[
                "Donor2_Mulliken_Charge"
            ] = np.nan


        # ----------------------------------------------------
        # Donor charge summary
        # ----------------------------------------------------

        q1 = result.get(
            "Donor1_Mulliken_Charge",
            np.nan
        )

        q2 = result.get(
            "Donor2_Mulliken_Charge",
            np.nan
        )

        if (
            pd.notna(q1)
            and
            pd.notna(q2)
        ):

            result[
                "Average_Donor_Charge"
            ] = (
                q1 + q2
            ) / 2.0

            result[
                "Delta_Donor_Charge"
            ] = abs(
                q1 - q2
            )

            result[
                "Donor_Charge_Sum"
            ] = q1 + q2

        else:

            result[
                "Average_Donor_Charge"
            ] = np.nan

            result[
                "Delta_Donor_Charge"
            ] = np.nan

            result[
                "Donor_Charge_Sum"
            ] = np.nan


    except Exception as e:

        result[
            "XTB_Warning"
        ] = str(e)


    return result


# ============================================================
# 22. PROCESS ONE XYZ
# ============================================================

def process_xyz(filepath):

    ligand_name = clean_ligand_name(
        filepath
    )

    elements, coordinates = (
        read_xyz_file(filepath)
    )


    row = {

        "Ligand": ligand_name,

        "XYZ_File":
            os.path.basename(filepath),

        "N_Atoms":
            len(elements),

        "Pd_Atom":
            PD_ATOM,

        "Ligand_Start_Atom":
            LIGAND_START,

        "Pd_Donor_Min_A":
            PD_DONOR_MIN,

        "Pd_Donor_Max_A":
            PD_DONOR_MAX
    }


    # ========================================================
    # DONORS
    # ========================================================

    candidates, qualifying = find_pd_donors(
        elements,
        coordinates
    )

    donor_selection, donor1, donor2 = select_pd_donors(
        elements,
        coordinates,
        qualifying
    )

    row["Number_of_Donor_Candidates"] = len(candidates)
    row["Number_of_Qualifying_Pd_Donors"] = len(qualifying)

    row.update(donor_selection)

    # ========================================================
    # PALLADACYCLE
    # ========================================================

    palladacycle = determine_palladacycle(
        elements,
        coordinates,
        donor1,
        donor2,
        ligand_class=row.get("LigandClass")
    )

    row.update(palladacycle)


    # ========================================================
    # STERIC
    # ========================================================

    steric = calculate_steric(
        elements,
        coordinates,
        donor1
    )

    row.update(
        steric
    )


    # ========================================================
    # COORDINATION
    # ========================================================

    coordination = (
        calculate_coordination(
            elements,
            coordinates,
            donor1,
            donor2
        )
    )

    row.update(
        coordination
    )


    # ========================================================
    # PYRAMIDALIZATION
    # ========================================================

    pyr = calculate_pyramidalization(
        elements,
        coordinates,
        donor1,
        donor2
    )

    row.update(
        pyr
    )


    # ========================================================
    # xTB
    # ========================================================

    xtb = calculate_xtb(
        elements,
        coordinates,
        donor1,
        donor2
    )

    row.update(
        xtb
    )


    # ========================================================
    # YIELD
    # ========================================================

    row[
        "Yield"
    ] = YIELD_DATA.get(
        ligand_name,
        np.nan
    )


    # ========================================================
    # WARNINGS
    # ========================================================

    warnings_list = []


    if row[
        "Donor_Selection_Check"
    ] != "PASS":

        warnings_list.append(
            row[
                "Donor_Selection_Check"
            ]
        )


    if (
        row[
            "Palladacycle_Type_Check"
        ]
        ==
        "CHECK"
    ):

        warnings_list.append(
            "Palladacycle type mismatch."
        )


    if pd.isna(
        row.get(
            "Bite_Angle_deg",
            np.nan
        )
    ):

        warnings_list.append(
            "Bite angle unavailable."
        )


    if pd.isna(
        row.get(
            "XTB_HOMO_eV",
            np.nan
        )
    ):

        warnings_list.append(
            "xTB descriptors unavailable."
        )


    row[
        "Warnings"
    ] = " || ".join(
        warnings_list
    )


    row[
        "Overall_Status"
    ] = (
        "PASS"
        if len(warnings_list) == 0
        else "WARNING"
    )


    return row


# ============================================================
# 23. PROCESS ALL XYZ
# ============================================================

print("\n")
print("=" * 80)
print("PROCESSING XYZ FILES")
print("=" * 80)


all_rows = []


for i, filepath in enumerate(
    xyz_files,
    start=1
):

    print(
        f"\n[{i}/{len(xyz_files)}] "
        f"{os.path.basename(filepath)}"
    )

    try:

        row = process_xyz(
            filepath
        )

        all_rows.append(
            row
        )

        print(
            "  Ligand:",
            row["Ligand"]
        )

        print(
            "  Donor 1:",
            row.get(
                "Donor1_Atom"
            ),
            row.get(
                "Donor1_Element"
            ),
            row.get(
                "Pd_Donor1_Distance_A"
            )
        )

        print(
            "  Donor 2:",
            row.get(
                "Donor2_Atom"
            ),
            row.get(
                "Donor2_Element"
            ),
            row.get(
                "Pd_Donor2_Distance_A"
            )
        )

        print(
            "  Palladacycle:",
            row.get(
                "Palladacycle_Type"
            )
        )

        print(
            "  Yield:",
            row.get(
                "Yield"
            )
        )

    except Exception as e:

        print(
            "  ERROR:",
            e
        )

        all_rows.append({

            "Ligand":
                clean_ligand_name(filepath),

            "XYZ_File":
                os.path.basename(filepath),

            "Overall_Status":
                "ERROR",

            "Warnings":
                str(e),

            "Yield":
                YIELD_DATA.get(
                    clean_ligand_name(filepath),
                    np.nan
                )
        })


# ============================================================
# 24. MASTER DATAFRAME
# ============================================================

master_df = pd.DataFrame(
    all_rows
)


# ============================================================
# 25. SORT L7 -> L25
# ============================================================

def ligand_sort_number(x):

    match = re.search(
        r"L(\d+)",
        str(x)
    )

    if match:
        return int(
            match.group(1)
        )

    return 9999


master_df[
    "_Sort"
] = master_df[
    "Ligand"
].apply(
    ligand_sort_number
)

master_df = (
    master_df
    .sort_values("_Sort")
    .drop(
        columns="_Sort"
    )
    .reset_index(
        drop=True
    )
)


# ============================================================
# 26. VERIFY YIELD MATCHING
# ============================================================

print("\n")
print("=" * 80)
print("YIELD MATCHING")
print("=" * 80)

yield_check = master_df[
    [
        "Ligand",
        "Yield"
    ]
]

print(
    yield_check.to_string(
        index=False
    )
)

print(
    "\nTotal ligands:",
    len(master_df)
)

print(
    "Yield available:",
    master_df[
        "Yield"
    ].notna().sum()
)

print(
    "Yield missing:",
    master_df[
        "Yield"
    ].isna().sum()
)


# ============================================================
# 27. DESCRIPTOR CATEGORIES
# ============================================================

steric_cols = [
    c for c in master_df.columns
    if (
        c.startswith("Ligand_SASA")
        or
        c.startswith("Ligand_%Vbur")
        or
        c.startswith("Ligand_Solid")
        or
        c.startswith("Ligand_Maximum")
        or
        c.startswith("Pd_Oriented_Sterimol")
        or
        c.startswith("Ligand_Distal")
        or
        c.startswith("Ligand_Molecular_Volume")
        or
        c.startswith("Ligand_Pint")
        or
        c.startswith("Vbur_")
    )
]


coord_cols = [
    c for c in master_df.columns
    if (
        c.startswith("Pd_Donor")
        or
        c.startswith("Average_Pd")
        or
        c.startswith("Delta_Pd")
        or
        c.startswith("Donor1_Donor2")
        or
        c.startswith("Bite_Angle")
        or
        c.startswith("Pd_Coordination")
        or
        c.startswith("Palladacycle_")
    )
]


donor_cols = [
    c for c in master_df.columns
    if (
        "Donor1_Mulliken" in c
        or
        "Donor2_Mulliken" in c
        or
        "Average_Donor_Charge" in c
        or
        "Delta_Donor_Charge" in c
        or
        "Donor_Charge_Sum" in c
    )
]


pyr_cols = [
    c for c in master_df.columns
    if (
        "Pyramidalization" in c
        or
        c.startswith("Average_Donor_Mean_Bond_Angle")
        or
        c.startswith("Delta_Donor_Mean_Bond_Angle")
    )
]


xtb_cols = [
    c for c in master_df.columns
    if c.startswith("XTB_")
]


print("\nDescriptor counts:")

print(
    "A Steric:",
    len(steric_cols)
)

print(
    "B Coordination:",
    len(coord_cols)
)

print(
    "C Donor electronic:",
    len(donor_cols)
)

print(
    "D Pyramidalization:",
    len(pyr_cols)
)

print(
    "E xTB:",
    len(xtb_cols)
)

# ============================================================
# 28. NUMERIC DATASET
# ============================================================

numeric_df = master_df.select_dtypes(
    include=[np.number]
).copy()


# ============================================================
# 29. REMOVE NON-PREDICTOR / QC VARIABLES
#
# These stay in ALL_DESCRIPTORS but are NOT predictors.
# ============================================================

EXCLUDE_FROM_CORRELATION = {

    "Pd_Atom",

    "Ligand_Start_Atom",

    "Pd_Donor_Min_A",

    "Pd_Donor_Max_A",

    "Donor1_Atom",

    "Donor2_Atom",

    # Redundant duplicate of Palladacycle_Ring_Size
    "Palladacycle_Path_Atom_Count",

    "N_Atoms",

    "Number_of_Donor_Candidates",

    "Number_of_Qualifying_Pd_Donors",

    # Pyramidalization QC
    "Donor1_Pyramidalization_Neighbor_Count",

    "Donor2_Pyramidalization_Neighbor_Count",

    "Donor1_Pyramidalization_Mean_Bond_Angle_N",

    "Donor2_Pyramidalization_Mean_Bond_Angle_N",

    # Metadata / settings
    "XTB_Charge",
    "XTB_N_Unpaired",
    "XTB_Method"
}


numeric_df = numeric_df.drop(
    columns=[
        c
        for c in EXCLUDE_FROM_CORRELATION
        if c in numeric_df.columns
    ],
    errors="ignore"
)


# ============================================================
# 30. REMOVE CONSTANT DESCRIPTORS
# ============================================================

constant_columns = []

for col in numeric_df.columns:

    if col == "Yield":
        continue

    values = numeric_df[
        col
    ].dropna()

    if values.nunique() <= 1:

        constant_columns.append(
            col
        )


numeric_df = numeric_df.drop(
    columns=constant_columns,
    errors="ignore"
)


print("\nConstant descriptors removed:")

if constant_columns:

    for c in constant_columns:
        print(" ",
              c)

else:

    print("  None")


# ============================================================
# 31. REQUIRE COMPLETE DESCRIPTORS
#
# We do NOT impute missing descriptor values.
# ============================================================

complete_columns = []

missing_columns = []

for col in numeric_df.columns:

    if numeric_df[
        col
    ].notna().all():

        complete_columns.append(
            col
        )

    else:

        missing_columns.append(
            col
        )


# Yield must remain
if "Yield" not in complete_columns:

    raise ValueError(
        "Yield contains missing values."
    )


correlation_base = numeric_df[
    complete_columns
].copy()


# ============================================================
# 32. OUTLIER SCREENING
#
# Descriptor removed if >10% observations are outliers
# by either IQR or MAD criterion.
#
# With 19 ligands:
# 2 or more outliers => >10%
# ============================================================

outlier_records = []

excluded_outliers = []

retained = []


descriptor_columns = [
    c
    for c in correlation_base.columns
    if c != "Yield"
]


for col in descriptor_columns:

    x = correlation_base[
        col
    ].dropna().astype(float)

    n = len(x)


    # --------------------------------------------------------
    # IQR
    # --------------------------------------------------------

    q1 = x.quantile(
        0.25
    )

    q3 = x.quantile(
        0.75
    )

    iqr = q3 - q1

    if iqr > 0:

        lower = (
            q1
            -
            IQR_MULTIPLIER * iqr
        )

        upper = (
            q3
            +
            IQR_MULTIPLIER * iqr
        )

        iqr_mask = (
            (x < lower)
            |
            (x > upper)
        )

        n_iqr = int(
            iqr_mask.sum()
        )

    else:

        n_iqr = 0


    # --------------------------------------------------------
    # MAD
    # --------------------------------------------------------

    median = x.median()

    mad = np.median(
        np.abs(
            x - median
        )
    )

    if mad > 0:

        modified_z = (
            0.6745
            *
            (x - median)
            /
            mad
        )

        mad_mask = (
            np.abs(
                modified_z
            )
            >
            MAD_Z_LIMIT
        )

        n_mad = int(
            mad_mask.sum()
        )

    else:

        n_mad = 0


    iqr_fraction = (
        n_iqr / n
        if n > 0
        else np.nan
    )

    mad_fraction = (
        n_mad / n
        if n > 0
        else np.nan
    )


    remove = (
        iqr_fraction
        >
        OUTLIER_FRACTION_LIMIT
        or
        mad_fraction
        >
        OUTLIER_FRACTION_LIMIT
    )


    outlier_records.append({

        "Descriptor":
            col,

        "N":
            n,

        "IQR_Outliers":
            n_iqr,

        "IQR_Outlier_Fraction":
            iqr_fraction,

        "MAD_Outliers":
            n_mad,

        "MAD_Outlier_Fraction":
            mad_fraction,

        "Removed":
            remove
    })


    if remove:

        excluded_outliers.append(
            col
        )

    else:

        retained.append(
            col
        )


outlier_df = pd.DataFrame(
    outlier_records
)


# ============================================================
# 33. FINAL FILTERED CORRELATION DATA
#
# Remove exact duplicate descriptors before correlation analysis.
# Ring size and path atom count encode the same project variable;
# ring size is retained as the chemically clearer descriptor.
# ============================================================

if "Palladacycle_Path_Atom_Count" in retained:
    retained.remove("Palladacycle_Path_Atom_Count")



correlation_columns = [
    "Yield"
] + retained


correlation_df = correlation_base[
    correlation_columns
].copy()


print("\n" + "=" * 80)
print("CORRELATION DATASET")
print("=" * 80)

print(
    "Yield + retained descriptors:",
    correlation_df.shape
)

print(
    "\nRetained descriptors:",
    len(retained)
)

print(
    "Removed due to excessive outliers:",
    len(excluded_outliers)
)


if excluded_outliers:

    print("\nRemoved:")

    for c in excluded_outliers:
        print(
            " ",
            c
        )


# ============================================================
# 34. FULL SQUARE PEARSON
# ============================================================

pearson_matrix_full = (
    correlation_df.corr(
        method="pearson"
    )
)


# ============================================================
# 35. FULL SQUARE SPEARMAN
# ============================================================

spearman_matrix_full = (
    correlation_df.corr(
        method="spearman"
    )
)


print(
    "\nFull Pearson matrix:",
    pearson_matrix_full.shape
)

print(
    "Full Spearman matrix:",
    spearman_matrix_full.shape
)


# ============================================================
# 36. ALSO CREATE RAW COMPLETE-DESCRIPTOR MATRICES
#
# Before outlier filtering
# ============================================================

raw_correlation_columns = [
    "Yield"
] + [
    c
    for c in correlation_base.columns
    if c != "Yield"
]

raw_correlation_df = correlation_base[
    raw_correlation_columns
].copy()


pearson_matrix_raw = (
    raw_correlation_df.corr(
        method="pearson"
    )
)


spearman_matrix_raw = (
    raw_correlation_df.corr(
        method="spearman"
    )
)


# ============================================================
# 37–38. PUBLICATION-QUALITY CORRELATION MATRICES
# ============================================================

MATRIX_ANNOTATION_MAX = 30
MATRIX_DPI = 600
MATRIX_BASE_FONT = 10
MATRIX_TICK_FONT = 9
MATRIX_TITLE_FONT = 16


def plot_correlation_matrix(matrix, title, colorbar_label, outfile):

    n = len(matrix)
    side = max(16, min(32, n * 0.68))
    annotation_font = max(8, min(14, 260 / max(n, 1)))

    fig, ax = plt.subplots(figsize=(side, side), constrained_layout=True)

    sns.heatmap(
        matrix,
        ax=ax,
        cmap="RdBu_r",
        center=0,
        vmin=-1,
        vmax=1,
        square=True,
        linewidths=0.35,
        linecolor="white",
        annot=(n <= MATRIX_ANNOTATION_MAX),
        fmt=".2f",
        annot_kws={
            "fontsize": annotation_font,
            "fontweight": "normal"
        },
        cbar_kws={
            "label": colorbar_label,
            "shrink": 0.78,
            "pad": 0.02
        }
    )

    ax.set_title(title, fontsize=MATRIX_TITLE_FONT, pad=16, weight="bold")
    ax.set_xticklabels(ax.get_xticklabels(), rotation=90, ha="center", fontsize=MATRIX_TICK_FONT)
    ax.set_yticklabels(ax.get_yticklabels(), rotation=0, fontsize=MATRIX_TICK_FONT)
    ax.tick_params(length=0)

    cbar = ax.collections[0].colorbar
    cbar.ax.tick_params(labelsize=9)
    cbar.set_label(colorbar_label, fontsize=10)

    fig.savefig(outfile, dpi=MATRIX_DPI, bbox_inches="tight", facecolor="white")
    plt.show()
    plt.close(fig)


plot_correlation_matrix(
    pearson_matrix_full,
    "Pearson Correlation Matrix\nYield + Retained Ligand Descriptors",
    "Pearson r",
    os.path.join(PLOT_FOLDER, "Pearson_FULL_SQUARE_Correlation_Matrix.png")
)

plot_correlation_matrix(
    spearman_matrix_full,
    "Spearman Correlation Matrix\nYield + Retained Ligand Descriptors",
    "Spearman ρ",
    os.path.join(PLOT_FOLDER, "Spearman_FULL_SQUARE_Correlation_Matrix.png")
)


# ============================================================
# 39. YIELD CORRELATIONS
# ============================================================

correlation_results = []


for descriptor in retained:

    temp = correlation_df[
        [
            "Yield",
            descriptor
        ]
    ].dropna()

    n = len(temp)


    if n >= 3:

        try:

            r, p_r = pearsonr(
                temp["Yield"],
                temp[descriptor]
            )

        except Exception:

            r = np.nan
            p_r = np.nan

        try:

            rho, p_s = spearmanr(
                temp["Yield"],
                temp[descriptor]
            )

        except Exception:

            rho = np.nan
            p_s = np.nan

    else:

        r = np.nan
        p_r = np.nan

        rho = np.nan
        p_s = np.nan


    correlation_results.append({

        "Descriptor":
            descriptor,

        "N":
            n,

        "Pearson_r":
            r,

        "Pearson_p":
            p_r,

        "Spearman_rho":
            rho,

        "Spearman_p":
            p_s,

        "Abs_Pearson_r":
            abs(r)
            if pd.notna(r)
            else np.nan,

        "Abs_Spearman_rho":
            abs(rho)
            if pd.notna(rho)
            else np.nan
    })


yield_correlation_df = pd.DataFrame(
    correlation_results
)

# Multiple-testing correction for the yield correlation screen.
# These adjusted p-values are the publication-facing significance values.
try:

    from statsmodels.stats.multitest import multipletests

    if len(yield_correlation_df) > 0:

        yield_correlation_df["Pearson_FDR_p"] = multipletests(
            yield_correlation_df["Pearson_p"].fillna(1.0),
            method="fdr_bh"
        )[1]

        yield_correlation_df["Spearman_FDR_p"] = multipletests(
            yield_correlation_df["Spearman_p"].fillna(1.0),
            method="fdr_bh"
        )[1]

except Exception as e:

    yield_correlation_df["FDR_Warning"] = str(e)


yield_correlation_df[
    "Mean_Abs_Correlation"
] = (
    yield_correlation_df[
        "Abs_Pearson_r"
    ]
    +
    yield_correlation_df[
        "Abs_Spearman_rho"
    ]
) / 2.0


yield_correlation_df = (
    yield_correlation_df
    .sort_values(
        "Mean_Abs_Correlation",
        ascending=False
    )
    .reset_index(
        drop=True
    )
)


# ============================================================
# 40. PRINT YIELD CORRELATIONS
# ============================================================

print("\n")
print("=" * 100)
print("YIELD vs RETAINED DESCRIPTORS")
print("=" * 100)

display(
    yield_correlation_df
)


# ============================================================
# 41. PEARSON YIELD RANKING
# ============================================================

pearson_rank = (
    yield_correlation_df
    .sort_values(
        "Pearson_r",
        ascending=True
    )
)


plt.figure(
    figsize=(
        10,
        max(
            6,
            len(pearson_rank)
            * 0.32
        )
    )
)


plt.barh(
    pearson_rank[
        "Descriptor"
    ],
    pearson_rank[
        "Pearson_r"
    ]
)


plt.axvline(
    0,
    linewidth=1
)

plt.xlabel(
    "Pearson r"
)

plt.ylabel(
    "Descriptor"
)

plt.title(
    "Pearson Correlation with Yield"
)

plt.tight_layout()


plt.savefig(
    os.path.join(
        PLOT_FOLDER,
        "Pearson_Yield_Correlation_Ranking.png"
    ),
    dpi=400,
    bbox_inches="tight"
)

plt.show()

plt.close()


# ============================================================
# 42. SPEARMAN YIELD RANKING
# ============================================================

spearman_rank = (
    yield_correlation_df
    .sort_values(
        "Spearman_rho",
        ascending=True
    )
)


plt.figure(
    figsize=(
        10,
        max(
            6,
            len(spearman_rank)
            * 0.32
        )
    )
)


plt.barh(
    spearman_rank[
        "Descriptor"
    ],
    spearman_rank[
        "Spearman_rho"
    ]
)


plt.axvline(
    0,
    linewidth=1
)

plt.xlabel(
    "Spearman ρ"
)

plt.ylabel(
    "Descriptor"
)

plt.title(
    "Spearman Correlation with Yield"
)

plt.tight_layout()


plt.savefig(
    os.path.join(
        PLOT_FOLDER,
        "Spearman_Yield_Correlation_Ranking.png"
    ),
    dpi=400,
    bbox_inches="tight"
)

plt.show()

plt.close()


# ============================================================
# 43. PEARSON vs SPEARMAN
# ============================================================

plt.figure(
    figsize=(8, 8)
)


plt.scatter(
    yield_correlation_df[
        "Pearson_r"
    ],
    yield_correlation_df[
        "Spearman_rho"
    ],
    s=65,
    alpha=0.85
)


for _, row in (
    yield_correlation_df.iterrows()
):

    if (
        pd.notna(
            row["Pearson_r"]
        )
        and
        pd.notna(
            row["Spearman_rho"]
        )
    ):

        plt.annotate(
            row["Descriptor"],
            (
                row["Pearson_r"],
                row["Spearman_rho"]
            ),
            xytext=(4, 4),
            textcoords="offset points",
            fontsize=7
        )


plt.axhline(
    0,
    linewidth=1
)

plt.axvline(
    0,
    linewidth=1
)

plt.xlabel(
    "Pearson r"
)

plt.ylabel(
    "Spearman ρ"
)

plt.title(
    "Pearson vs Spearman Correlation with Yield"
)

plt.grid(
    alpha=0.25
)

plt.tight_layout()


plt.savefig(
    os.path.join(
        PLOT_FOLDER,
        "Pearson_vs_Spearman_Yield.png"
    ),
    dpi=400,
    bbox_inches="tight"
)

plt.show()

plt.close()


# ============================================================
# 44. YIELD vs EVERY DESCRIPTOR
#
# IMPORTANT:
# NO REGRESSION LINE
# ============================================================

for descriptor in retained:

    temp = correlation_df[
        [
            "Yield",
            descriptor
        ]
    ].dropna()

    if len(temp) < 3:
        continue


    try:

        r, p_r = pearsonr(
            temp[descriptor],
            temp["Yield"]
        )

    except Exception:

        r = np.nan
        p_r = np.nan


    try:

        rho, p_s = spearmanr(
            temp[descriptor],
            temp["Yield"]
        )

    except Exception:

        rho = np.nan
        p_s = np.nan


    plt.figure(
        figsize=(7, 5)
    )


    plt.scatter(
        temp[descriptor],
        temp["Yield"],
        s=55,
        alpha=0.85
    )


    # --------------------------------------------------------
    # Ligand labels
    # --------------------------------------------------------

    for idx in temp.index:

        label = master_df.loc[
            idx,
            "Ligand"
        ]

        if pd.notna(label):

            plt.annotate(
                str(label),
                (
                    temp.loc[
                        idx,
                        descriptor
                    ],
                    temp.loc[
                        idx,
                        "Yield"
                    ]
                ),
                xytext=(4, 4),
                textcoords="offset points",
                fontsize=7
            )


    plt.xlabel(
        descriptor
    )

    plt.ylabel(
        "Yield (%)"
    )

    plt.title(
        f"Yield vs {descriptor}"
    )


    stats_text = (
        f"Pearson r = {r:.3f}\n"
        f"Pearson p = {p_r:.3g}\n"
        f"Spearman ρ = {rho:.3f}\n"
        f"Spearman p = {p_s:.3g}\n"
        f"n = {len(temp)}"
    )


    plt.text(
        0.03,
        0.97,
        stats_text,
        transform=plt.gca().transAxes,
        va="top",
        fontsize=9,
        bbox=dict(
            boxstyle="round",
            alpha=0.15
        )
    )


    plt.grid(
        alpha=0.25
    )

    plt.tight_layout()


    outfile = os.path.join(
        SCATTER_FOLDER,
        "Yield_vs_"
        +
        safe_filename(
            descriptor
        )
        +
        ".png"
    )


    plt.savefig(
        outfile,
        dpi=400,
        bbox_inches="tight"
    )

    plt.close()


# ============================================================
# 45. DESCRIPTOR BOXPLOTS
# ============================================================

for descriptor in retained:

    values = correlation_df[
        descriptor
    ].dropna()

    if len(values) < 3:
        continue


    plt.figure(
        figsize=(6, 4)
    )


    plt.boxplot(
        values
    )

    plt.ylabel(
        descriptor
    )

    plt.title(
        f"Distribution: {descriptor}"
    )

    plt.grid(
        axis="y",
        alpha=0.25
    )

    plt.tight_layout()


    outfile = os.path.join(
        BOX_FOLDER,
        "Boxplot_"
        +
        safe_filename(
            descriptor
        )
        +
        ".png"
    )


    plt.savefig(
        outfile,
        dpi=350,
        bbox_inches="tight"
    )

    plt.close()


# ============================================================
# 46. HIGH CORRELATION BETWEEN DESCRIPTORS
# ============================================================

high_pairs = []


for i in range(
    len(retained)
):

    for j in range(
        i + 1,
        len(retained)
    ):

        d1 = retained[i]
        d2 = retained[j]


        r = pearson_matrix_full.loc[
            d1,
            d2
        ]

        rho = spearman_matrix_full.loc[
            d1,
            d2
        ]

        if (
            abs(r)
            >=
            HIGH_CORR_THRESHOLD
            or
            abs(rho)
            >=
            HIGH_CORR_THRESHOLD
        ):

            high_pairs.append({

                "Descriptor_1":
                    d1,

                "Descriptor_2":
                    d2,

                "Pearson_r":
                    r,

                "Spearman_rho":
                    rho,

                "Max_Abs_Correlation":
                    max(
                        abs(r),
                        abs(rho)
                    )
            })


high_corr_df = pd.DataFrame(
    high_pairs,
    columns=[
        "Descriptor_1",
        "Descriptor_2",
        "Pearson_r",
        "Spearman_rho",
        "Max_Abs_Correlation"
    ]
)


if len(high_corr_df) > 0:

    high_corr_df = (
        high_corr_df
        .sort_values(
            "Max_Abs_Correlation",
            ascending=False
        )
        .reset_index(
            drop=True
        )
    )


# ============================================================
# 47. YIELD BY PALLADACYCLE TYPE
# ============================================================

class_df = master_df[
    [
        "Palladacycle_Type",
        "Yield"
    ]
].dropna()


if len(class_df) > 0:

    plt.figure(
        figsize=(7, 5)
    )


    sns.boxplot(
        data=class_df,
        x="Palladacycle_Type",
        y="Yield"
    )


    sns.stripplot(
        data=class_df,
        x="Palladacycle_Type",
        y="Yield",
        jitter=True,
        alpha=0.7
    )


    plt.xlabel(
        "Palladacycle type"
    )

    plt.ylabel(
        "Yield (%)"
    )

    plt.title(
        "Yield by Palladacycle Type"
    )

    plt.tight_layout()


    plt.savefig(
        os.path.join(
            PLOT_FOLDER,
            "Yield_by_Palladacycle_Type.png"
        ),
        dpi=400,
        bbox_inches="tight"
    )

    plt.show()

    plt.close()


# ============================================================
# 48. SAVE VALIDATION TABLE
# ============================================================

validation_columns = [

    "Ligand",

    "LigandClass",
    "ExpectedDonorPair",
    "Donor1_Atom",
    "Donor1_Element",
    "Pd_Donor1_Distance_A",
    "Donor2_Atom",
    "Donor2_Element",
    "Pd_Donor2_Distance_A",
    "Donor_Selection_Check",
    "PiCoordination",
    "PiCoordinationAtom",
    "PiCoordinationPd_C_Distance_A",

    "XYZ_File",

    "N_Atoms",

    "Number_of_Qualifying_Pd_Donors",

    "Palladacycle_Type",

    "Expected_Palladacycle_Type",

    "Palladacycle_Type_Check",

    "Palladacycle_Ligand_Path",

    "Overall_Status",

    "Warnings"
]


validation_columns = [
    c
    for c in validation_columns
    if c in master_df.columns
]


validation_df = master_df[
    validation_columns
].copy()


# ============================================================
# 49. FINAL EXCEL WORKBOOK
# ============================================================

print("\n")
print("=" * 80)
print("WRITING FINAL EXCEL WORKBOOK")
print("=" * 80)


with pd.ExcelWriter(
    FINAL_EXCEL,
    engine="openpyxl"
) as writer:


    # --------------------------------------------------------
    # COMPLETE RAW DATA
    # --------------------------------------------------------

    master_df.to_excel(
        writer,
        sheet_name="ALL_DESCRIPTORS",
        index=False
    )


    # --------------------------------------------------------
    # VALIDATION
    # --------------------------------------------------------

    validation_df.to_excel(
        writer,
        sheet_name="VALIDATION",
        index=False
    )


    # --------------------------------------------------------
    # YIELD CORRELATION DATA
    # --------------------------------------------------------

    correlation_df.to_excel(
        writer,
        sheet_name="Correlation_Data",
        index=False
    )


    # --------------------------------------------------------
    # FULL RAW PEARSON
    # --------------------------------------------------------

    pearson_matrix_raw.to_excel(
        writer,
        sheet_name="Pearson_RAW_FULL"
    )


    # --------------------------------------------------------
    # FULL RAW SPEARMAN
    # --------------------------------------------------------

    spearman_matrix_raw.to_excel(
        writer,
        sheet_name="Spearman_RAW_FULL"
    )


    # --------------------------------------------------------
    # FILTERED PEARSON
    # --------------------------------------------------------

    pearson_matrix_full.to_excel(
        writer,
        sheet_name="Pearson_FILTERED_FULL"
    )


    # --------------------------------------------------------
    # FILTERED SPEARMAN
    # --------------------------------------------------------

    spearman_matrix_full.to_excel(
        writer,
        sheet_name="Spearman_FILTERED_FULL"
    )


    # --------------------------------------------------------
    # YIELD CORRELATIONS
    # --------------------------------------------------------

    yield_correlation_df.to_excel(
        writer,
        sheet_name="Yield_Correlations",
        index=False
    )


    # --------------------------------------------------------
    # OUTLIER SCREENING
    # --------------------------------------------------------

    outlier_df.to_excel(
        writer,
        sheet_name="Outlier_Screening",
        index=False
    )


    # --------------------------------------------------------
    # HIGH CORRELATION PAIRS
    # --------------------------------------------------------

    high_corr_df.to_excel(
        writer,
        sheet_name="High_Correlation_Pairs",
        index=False
    )


    # --------------------------------------------------------
    # A - STERIC
    # --------------------------------------------------------

    cols = [
        "Ligand"
    ] + [
        c for c in steric_cols
        if c in master_df.columns
    ] + ["Yield"]

    master_df[
        list(dict.fromkeys(cols))
    ].to_excel(
        writer,
        sheet_name="A_STERIC",
        index=False
    )


    # --------------------------------------------------------
    # B - COORDINATION
    # --------------------------------------------------------

    cols = [
        "Ligand"
    ] + [
        c for c in coord_cols
        if c in master_df.columns
    ] + ["Yield"]

    master_df[
        list(dict.fromkeys(cols))
    ].to_excel(
        writer,
        sheet_name="B_COORDINATION",
        index=False
    )


    # --------------------------------------------------------
    # C - DONOR ELECTRONIC
    # --------------------------------------------------------

    cols = [
        "Ligand"
    ] + [
        c for c in donor_cols
        if c in master_df.columns
    ] + ["Yield"]

    master_df[
        list(dict.fromkeys(cols))
    ].to_excel(
        writer,
        sheet_name="C_DONOR_ELECTRONIC",
        index=False
    )


    # --------------------------------------------------------
    # D - PYRAMIDALIZATION
    # --------------------------------------------------------

    cols = [
        "Ligand"
    ] + [
        c for c in pyr_cols
        if c in master_df.columns
    ] + ["Yield"]

    master_df[
        list(dict.fromkeys(cols))
    ].to_excel(
        writer,
        sheet_name="D_PYRAMIDALIZATION",
        index=False
    )


    # --------------------------------------------------------
    # E - xTB
    # --------------------------------------------------------

    cols = [
        "Ligand"
    ] + [
        c for c in xtb_cols
        if c in master_df.columns
    ] + ["Yield"]

    master_df[
        list(dict.fromkeys(cols))
    ].to_excel(
        writer,
        sheet_name="E_XTB",
        index=False
    )


    # --------------------------------------------------------
    # YIELD DATA
    # --------------------------------------------------------

    yield_table = pd.DataFrame({

        "Ligand":
            list(YIELD_DATA.keys()),

        "Yield":
            list(YIELD_DATA.values())
    })


    yield_table.to_excel(
        writer,
        sheet_name="YIELD_DATA",
        index=False
    )


    # --------------------------------------------------------
    # REMOVED DESCRIPTORS
    # --------------------------------------------------------

    removed_records = []

    for c in constant_columns:

        removed_records.append({

            "Descriptor":
                c,

            "Reason":
                "Constant descriptor"
        })

    for c in missing_columns:

        if c != "Yield":

            removed_records.append({

                "Descriptor":
                    c,

                "Reason":
                    "Contains missing value(s)"
            })

    for c in excluded_outliers:

        removed_records.append({

            "Descriptor":
                c,

            "Reason":
                "More than 10% strong outliers"
        })

    removed_df = pd.DataFrame(
        removed_records
    )

    removed_df.to_excel(
        writer,
        sheet_name="REMOVED_DESCRIPTORS",
        index=False
    )


# ============================================================
# 50. FINAL SUMMARY
# ============================================================

print("\n")
print("=" * 100)
print("FINAL ANALYSIS COMPLETE")
print("=" * 100)

print(
    "\nExcel:"
)

print(
    FINAL_EXCEL
)

print(
    "\nPlots:"
)

print(
    PLOT_FOLDER
)

print(
    "\nNumber of ligands:",
    len(master_df)
)

print(
    "Total descriptors in raw table:",
    len(
        [
            c for c in master_df.columns
            if c not in [
                "Ligand",
                "XYZ_File",
                "Warnings",
                "Overall_Status"
            ]
        ]
    )
)

print(
    "Retained correlation descriptors:",
    len(retained)
)

print(
    "Constant removed:",
    len(constant_columns)
)

print(
    "Missing-value removed:",
    len(
        [
            c
            for c in missing_columns
            if c != "Yield"
        ]
    )
)

print(
    "Outlier removed:",
    len(excluded_outliers)
)

print(
    "\nPearson matrix:",
    pearson_matrix_full.shape
)

print(
    "Spearman matrix:",
    spearman_matrix_full.shape
)

print(
    "\nTop 10 Yield correlations:"
)

display(
    yield_correlation_df.head(10)
)

print(
    "\n" + "=" * 100
)
print(
    "✓ ALL PROCESSING FINISHED"
)
print(
    "=" * 100
)
```
