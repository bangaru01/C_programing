```python
# ============================================================
# Nat_Syn-DM: Pd γ-ARYLATION
# COMPLETE LIGAND + S-SUBSTITUENT DESCRIPTOR ANALYSIS
#
# Required XYZ atom numbering:
#   Atom 1       = Pd
#   Atoms 2-19   = substrate
#   Atoms 20-end = ligand
#
# Ligand classes:
#   L10-L19 = expected NN/NO ligands
#   L20-L27 = expected NS ligands
#
# Pd coordinating donor rules:
#   N/O: 2.00-2.30 Å
#   S:   2.40-2.65 Å
#   H is NEVER considered a Pd donor.
# ============================================================


# ============================================================
# 1. INSTALL REQUIRED PACKAGES
# ============================================================

!pip -q install morfeus-ml pandas numpy scipy openpyxl matplotlib seaborn statsmodels


# ============================================================
# 2. IMPORTS
# ============================================================

import os
import re
import glob
import math
import shutil
import warnings
import itertools
from collections import deque

import numpy as np
import pandas as pd

import matplotlib.pyplot as plt
import seaborn as sns

from scipy.stats import pearsonr, spearmanr

warnings.filterwarnings("ignore")

pd.set_option("display.max_columns", 200)
pd.set_option("display.width", 200)


# ============================================================
# 3. MOUNT GOOGLE DRIVE
# ============================================================

from google.colab import drive
drive.mount("/content/drive")


# ============================================================
# 4. USER SETTINGS
# ============================================================

# ------------------------------------------------------------
# Input and output folders
# ------------------------------------------------------------

XYZ_FOLDER = "/content/drive/MyDrive/Nat_Syn-DM"

OUTPUT_FOLDER = (
    "/content/drive/MyDrive/"
    "Nat_Syn_DM_Gamma_Arylation_Results"
)

os.makedirs(OUTPUT_FOLDER, exist_ok=True)

PLOT_FOLDER = os.path.join(OUTPUT_FOLDER, "Plots")
SCATTER_FOLDER = os.path.join(PLOT_FOLDER, "Yield_vs_Descriptors")
BOXPLOT_FOLDER = os.path.join(PLOT_FOLDER, "Descriptor_Boxplots")

os.makedirs(PLOT_FOLDER, exist_ok=True)
os.makedirs(SCATTER_FOLDER, exist_ok=True)
os.makedirs(BOXPLOT_FOLDER, exist_ok=True)

FINAL_EXCEL = os.path.join(
    OUTPUT_FOLDER,
    "Nat_Syn_DM_Gamma_Arylation_Descriptor_Analysis.xlsx"
)

MASTER_CSV = os.path.join(
    OUTPUT_FOLDER,
    "Nat_Syn_DM_All_Descriptors.csv"
)

YIELD_CORRELATION_CSV = os.path.join(
    OUTPUT_FOLDER,
    "Nat_Syn_DM_Yield_Correlations.csv"
)


# ------------------------------------------------------------
# Required chemical atom numbering
# ------------------------------------------------------------

PD_ATOM = 1

SUBSTRATE_START = 2
SUBSTRATE_END = 19

LIGAND_START = 20


# ------------------------------------------------------------
# Reaction information
# ------------------------------------------------------------

REACTION_TYPE = "Nat_Syn-Gamma_Arylation"


# ------------------------------------------------------------
# Pd donor identification
#
# IMPORTANT:
# Hydrogen is completely excluded.
# Only N, O, and S may be Pd donor candidates.
# ------------------------------------------------------------

# ============================================================
# Pd-donor distance windows in Å
#
# STRICT:
# Primary coordination-distance criteria.
#
# EXTENDED:
# Used to retain chemically plausible elongated donor contacts
# for chelated NN, NO, and NS ligand structures.
# ============================================================

DONOR_ELEMENTS = {"N", "O", "S"}

PD_DONOR_WINDOWS_STRICT = {
    "N": (2.00, 2.30),
    "O": (2.00, 2.30),
    "S": (2.40, 2.65),
}

PD_DONOR_WINDOWS_EXTENDED = {
    "N": (1.90, 2.45),
    "O": (1.90, 2.50),
    "S": (2.20, 2.80),
}


# ------------------------------------------------------------
# Ligand class assignment
# ------------------------------------------------------------

# ============================================================
# LIGAND CLASS ASSIGNMENT
# ============================================================

LIGAND_CLASSES = {
    # NN ligands: L10, L12, L13, L14, L18, L19
    **{
        f"L{i}": "NN"
        for i in [10, 12, 13, 14, 18, 19]
    },

    # NO ligands: L11, L15, L16, L17
    **{
        f"L{i}": "NO"
        for i in [11, 15, 16, 17]
    },

    # NS ligands: L20-L25
    **{
        f"L{i}": "NS"
        for i in range(20, 26)
    },

    # L26: N + aromatic pi-cloud coordination
    "L26": "NPI",
}

EXPECTED_DONOR_PAIRS = {
    "NN": ["N", "N"],
    "NO": ["N", "O"],
    "NS": ["N", "S"],
    "NPI": ["N", "C(pi)"],
}

# L26 pi-coordination geometry: the aromatic carbon is used as a
# geometric representative of the Pd interaction with the pi cloud.
L26_PI_C_MIN_DISTANCE_A = 2.00
L26_PI_C_MAX_DISTANCE_A = 3.00


# ------------------------------------------------------------
# S-substituent assignment for NS ligands
#
# USER ACTION REQUIRED:
# Update this mapping based on your ligand structures.
#
# Available labels suggested:
#   "S-Me"
#   "S-Et"
#   "S-Pr"
#   "S-iPr"
#   "S-cHex"
#   "S-Aryl"
#   "Other"
#
# This column enables direct comparison of S-substituent identity
# versus reaction yield.
# ------------------------------------------------------------

S_SUBSTITUENT_CLASS = {
    "L20": "UPDATE_ME",
    "L21": "UPDATE_ME",
    "L22": "UPDATE_ME",
    "L23": "UPDATE_ME",
    "L24": "UPDATE_ME",
    "L25": "UPDATE_ME",
    "L26": "UPDATE_ME",
}


# ------------------------------------------------------------
# Yield data
#
# USER ACTION REQUIRED:
# Update values here.
#
# Use numerical yields, normally percentages.
# Examples:
#   "L10": 42,
#   "L11": 71.5,
#   "L20": 88,
#
# Leave np.nan if yield is not available.
# ------------------------------------------------------------

YIELD_DATA = {
    "L10": 0,
    "L11": 1,
    "L12": 7,
    "L13": 0,
    "L14": 0,
    "L15": 0,
    "L16": 0,
    "L17": 0,
    "L18": 0,
    "L19": 25,
    "L20": 70,
    "L21": 64,
    "L22": 57,
    "L23": 55,
    "L24": 50,
    "L25": 9,
    "L26": 0,
}


# ------------------------------------------------------------
# Morfeus settings
# ------------------------------------------------------------

SASA_PROBE_RADIUS = 1.4

VBUR_RADII = [3.0, 4.0, 5.5]
DISTAL_VOLUME_RADIUS = 3.5

# Morfeus BuriedVolume uses density, not npoints.
# Smaller density = more grid points / higher precision.
VBUR_DENSITY = 0.0008

VBUR_RADII_TYPE = "bondi"
SOLID_ANGLE_RADII_TYPE = "bondi"
STERIMOL_RADII_TYPE = "crc"

CALCULATE_DISPERSION = True


# ------------------------------------------------------------
# xTB settings
#
# Set RUN_XTB = False if xTB is unavailable or if you want
# only geometry and steric descriptors.
# ------------------------------------------------------------

RUN_XTB = True

XTB_METHOD = 2
XTB_CHARGE = 0
XTB_NUNPAIRED = 0
XTB_PROCESSES = 2


# ------------------------------------------------------------
# Statistical settings
# ------------------------------------------------------------

MINIMUM_CORRELATION_N = 5
HIGH_CORRELATION_THRESHOLD = 0.80

OUTLIER_FRACTION_LIMIT = 0.10
IQR_MULTIPLIER = 1.5
MAD_Z_LIMIT = 3.5


# ============================================================
# 5. xTB INSTALLATION AND MORFEUS IMPORT
# ============================================================

# Morfeus XTB must only be imported after xTB is available.
# If xTB is missing, this cell installs a standard Linux xTB binary.

if RUN_XTB:
    xtb_path = shutil.which("xtb")

    if xtb_path is None:
        print("xTB not found. Installing xTB...")

        xtb_archive = "/content/xtb.tar.xz"

        !wget -q --show-progress \
            https://github.com/grimme-lab/xtb/releases/latest/download/xtb-6.7.1-linux-x86_64.tar.xz \
            -O /content/xtb.tar.xz

        !tar -xf /content/xtb.tar.xz -C /content

        xtb_dirs = sorted(
            [
                os.path.join("/content", folder)
                for folder in os.listdir("/content")
                if folder.startswith("xtb-")
                and os.path.isdir(os.path.join("/content", folder))
            ]
        )

        if len(xtb_dirs) == 0:
            raise RuntimeError(
                "xTB archive was extracted, but xTB directory was not found."
            )

        xtb_dir = xtb_dirs[-1]
        xtb_bin = os.path.join(xtb_dir, "bin")
        xtb_executable = os.path.join(xtb_bin, "xtb")

        if not os.path.exists(xtb_executable):
            raise RuntimeError(
                f"xTB executable not found: {xtb_executable}"
            )

        os.environ["PATH"] = xtb_bin + ":" + os.environ["PATH"]
        os.environ["XTBHOME"] = xtb_dir

    xtb_path = shutil.which("xtb")

    if xtb_path is None:
        raise RuntimeError("xTB is still unavailable after installation.")

    print("xTB executable:", xtb_path)

    !xtb --version


# Import Morfeus only after xTB path is available.
from morfeus import (
    SASA,
    SolidAngle,
    BuriedVolume,
    Sterimol,
    Pyramidalization,
)

if RUN_XTB:
    from morfeus import XTB

if CALCULATE_DISPERSION:
    try:
        from morfeus import Dispersion
        DISPERSION_AVAILABLE = True
    except Exception:
        DISPERSION_AVAILABLE = False
else:
    DISPERSION_AVAILABLE = False


# ============================================================
# 6. GENERAL HELPER FUNCTIONS
# ============================================================

def finite_value(value):
    """Convert valid finite values to float; otherwise return NaN."""
    try:
        numeric_value = float(value)
        if np.isfinite(numeric_value):
            return numeric_value
    except Exception:
        pass

    return np.nan


def normalize_element(element):
    """Normalize atomic symbols read from XYZ files."""
    return str(element).strip().capitalize()


def distance_a(coord_a, coord_b):
    """Return Cartesian distance in Å."""
    return float(
        np.linalg.norm(
            np.asarray(coord_a, dtype=float)
            - np.asarray(coord_b, dtype=float)
        )
    )


def angle_deg(coord_a, coord_b, coord_c):
    """
    Return angle A-B-C in degrees.
    The central atom is coord_b.
    """
    vector_1 = np.asarray(coord_a, dtype=float) - np.asarray(coord_b, dtype=float)
    vector_2 = np.asarray(coord_c, dtype=float) - np.asarray(coord_b, dtype=float)

    norm_1 = np.linalg.norm(vector_1)
    norm_2 = np.linalg.norm(vector_2)

    if norm_1 == 0 or norm_2 == 0:
        return np.nan

    cosine = np.dot(vector_1, vector_2) / (norm_1 * norm_2)
    cosine = np.clip(cosine, -1.0, 1.0)

    return float(np.degrees(np.arccos(cosine)))


def safe_filename(text):
    """Create safe filenames for plot output."""
    return re.sub(r"[^A-Za-z0-9_.-]+", "_", str(text))


def clean_ligand_name(filepath):
    """
    Extract ligand identity such as L10 from XYZ filename.

    Works for filenames such as:
        L10.xyz
        L10_optimized.xyz
        ligand-L10.xyz
        complex_L20_final.xyz
    """
    basename = os.path.basename(filepath)
    basename = re.sub(r"\.xyz$", "", basename, flags=re.IGNORECASE)

    match = re.search(r"(L\d+)", basename, flags=re.IGNORECASE)

    if match:
        return match.group(1).upper()

    return basename


def ligand_sort_number(ligand_name):
    """Sort L10, L11, ..., L27 numerically."""
    match = re.search(r"L(\d+)", str(ligand_name), flags=re.IGNORECASE)

    if match:
        return int(match.group(1))

    return 999999


def classify_ligand(ligand_name):
    """
    Return manually assigned ligand class:
    NN, NO, NS, or UNKNOWN.
    """
    ligand_name = str(ligand_name).upper()

    return LIGAND_CLASSES.get(
        ligand_name,
        "UNKNOWN",
    )


def expected_donor_pair(ligand_name):
    """Return expected donor elements as a list, for example ['N', 'O' 'S']."""
    ligand_class = classify_ligand(ligand_name)
    return EXPECTED_DONOR_PAIRS.get(ligand_class)


def ligand_indices(elements):
    """
    Return original 1-based XYZ indices for ligand atoms:
    atoms 20 through the final atom.
    """
    return list(range(LIGAND_START, len(elements) + 1))


def substrate_indices(elements):
    """Return original 1-based XYZ indices for substrate atoms 2-19."""
    maximum_atom = min(SUBSTRATE_END, len(elements))
    return list(range(SUBSTRATE_START, maximum_atom + 1))


# ============================================================
# 7. XYZ READER
# ============================================================

def read_xyz_file(filepath):
    """Read a standard XYZ file."""
    with open(filepath, "r", encoding="utf-8") as handle:
        lines = [line.strip() for line in handle if line.strip()]

    if len(lines) < 3:
        raise ValueError(f"XYZ file is too short: {filepath}")

    number_of_atoms = int(lines[0].split()[0])
    atom_lines = lines[2:2 + number_of_atoms]

    if len(atom_lines) < number_of_atoms:
        raise ValueError(
            f"XYZ coordinate lines are fewer than specified atom count in {filepath}"
        )

    elements = []
    coordinates = []

    for line in atom_lines:
        parts = line.split()

        if len(parts) < 4:
            raise ValueError(f"Malformed XYZ atom line: {line}")

        elements.append(normalize_element(parts[0]))
        coordinates.append(
            [
                float(parts[1]),
                float(parts[2]),
                float(parts[3]),
            ]
        )

    return elements, np.asarray(coordinates, dtype=float)


# ============================================================
# 8. STRUCTURE VALIDATION
# ============================================================

def validate_structure(ligand_name, elements, coordinates):
    """
    Validate required Nat_Syn-DM numbering and expected ligand content.
    """
    warnings_list = []

    if len(elements) != len(coordinates):
        warnings_list.append(
            "Element count and coordinate count are not identical."
        )

    if len(elements) < LIGAND_START:
        warnings_list.append(
            f"Structure has fewer than {LIGAND_START} atoms."
        )

    if len(elements) >= 1:
        if normalize_element(elements[PD_ATOM - 1]) != "Pd":
            warnings_list.append("Atom 1 is not Pd.")

    if len(elements) >= SUBSTRATE_END:
        substrate_atom_count = len(
            range(SUBSTRATE_START, SUBSTRATE_END + 1)
        )

        if substrate_atom_count != 18:
            warnings_list.append(
                "Substrate atom range 2-19 is not 18 atoms."
            )

    if not np.isfinite(coordinates).all():
        warnings_list.append("XYZ contains non-finite coordinates.")

    ligand_atom_ids = ligand_indices(elements)

    if len(ligand_atom_ids) == 0:
        warnings_list.append("No ligand atoms found from atom 20 onward.")
        return warnings_list

    ligand_elements = [
        normalize_element(elements[atom_id - 1])
        for atom_id in ligand_atom_ids
    ]

    ligand_class = classify_ligand(ligand_name)

    # --------------------------------------------------------
    # FIX: this block previously mixed tabs and spaces and had
    # the if/elif statements at inconsistent indentation levels,
    # which raised a TabError / IndentationError. Rewritten here
    # with consistent 4-space indentation.
    # --------------------------------------------------------

    if ligand_class == "NN":
        if ligand_elements.count("N") < 2:
            warnings_list.append(
                "Expected NN ligand but fewer than two ligand N atoms were found."
            )

    elif ligand_class == "NO":
        if "N" not in ligand_elements:
            warnings_list.append(
                "Expected NO ligand but no ligand N atom was found."
            )

        if "O" not in ligand_elements:
            warnings_list.append(
                "Expected NO ligand but no ligand O atom was found."
            )

    elif ligand_class == "NS":
        if "N" not in ligand_elements:
            warnings_list.append(
                "Expected NS ligand but no ligand N atom was found."
            )

        if "S" not in ligand_elements:
            warnings_list.append(
                "Expected NS ligand but no ligand S atom was found."
            )

    return warnings_list


# ============================================================
# 9. COVALENT CONNECTIVITY
# ============================================================

COVALENT_RADII = {
    "H": 0.31,
    "C": 0.76,
    "N": 0.71,
    "O": 0.66,
    "F": 0.57,
    "P": 1.07,
    "S": 1.05,
    "Cl": 1.02,
    "Br": 1.20,
    "I": 1.39,
    "Pd": 1.39,
}


def covalent_neighbors(
    elements,
    coordinates,
    atom_id,
    allowed_indices=None,
    scale=1.25,
):
    """
    Identify covalent neighbors from XYZ geometry and covalent radii.

    atom_id is 1-based.
    """
    if allowed_indices is None:
        allowed_indices = list(range(1, len(elements) + 1))

    atom_element = normalize_element(elements[atom_id - 1])

    if atom_element not in COVALENT_RADII:
        return []

    atom_radius = COVALENT_RADII[atom_element]
    atom_coord = coordinates[atom_id - 1]

    neighbors = []

    for other_id in allowed_indices:
        if other_id == atom_id:
            continue

        other_element = normalize_element(elements[other_id - 1])

        if other_element not in COVALENT_RADII:
            continue

        other_radius = COVALENT_RADII[other_element]
        other_coord = coordinates[other_id - 1]

        current_distance = distance_a(atom_coord, other_coord)
        cutoff_distance = scale * (atom_radius + other_radius)

        if current_distance <= cutoff_distance:
            neighbors.append((other_id, current_distance))

    return sorted(neighbors, key=lambda item: item[1])


def build_ligand_graph(elements, coordinates):
    """
    Build an undirected covalent graph only for ligand atoms 20-end.
    Pd and substrate atoms are excluded.
    """
    ligand_atom_ids = ligand_indices(elements)

    graph = {
        atom_id: set()
        for atom_id in ligand_atom_ids
    }

    for atom_id in ligand_atom_ids:
        neighbors = covalent_neighbors(
            elements=elements,
            coordinates=coordinates,
            atom_id=atom_id,
            allowed_indices=ligand_atom_ids,
        )

        for neighbor_id, _ in neighbors:
            graph[atom_id].add(neighbor_id)
            graph[neighbor_id].add(atom_id)

    return graph


def shortest_graph_path(graph, start_atom, end_atom):
    """Return shortest atom path in an undirected graph."""
    if start_atom is None or end_atom is None:
        return None

    if start_atom not in graph or end_atom not in graph:
        return None

    queue = deque([(start_atom, [start_atom])])
    visited = {start_atom}

    while queue:
        current_atom, path = queue.popleft()

        if current_atom == end_atom:
            return path

        for neighbor in sorted(graph.get(current_atom, [])):
            if neighbor not in visited:
                visited.add(neighbor)
                queue.append((neighbor, path + [neighbor]))

    return None


def connected_component(graph, start_atom):
    """Return all atoms in the graph component containing start_atom."""
    if start_atom is None or start_atom not in graph:
        return set()

    visited = {start_atom}
    queue = deque([start_atom])

    while queue:
        current_atom = queue.popleft()

        for neighbor in graph.get(current_atom, set()):
            if neighbor not in visited:
                visited.add(neighbor)
                queue.append(neighbor)

    return visited


# ============================================================
# 10. Pd DONOR IDENTIFICATION
# ============================================================

def find_pd_donor_candidates(elements, coordinates):
    """
    Find ligand N/O/S atoms close to Pd.

    H is never considered.

    Each candidate receives:
    - StrictQualifies: within the preferred Pd-X range.
    - ExtendedQualifies: within an extended plausible Pd-X range.

    Donor selection should use ExtendedQualifies so that an
    elongated but structurally intended chelating contact is not
    discarded. The StrictQualifies column remains as QC information.
    """
    pd_coordinate = coordinates[PD_ATOM - 1]
    all_candidates = []

    for atom_id in ligand_indices(elements):
        element = normalize_element(elements[atom_id - 1])

        if element not in {"N", "O", "S"}:
            continue

        strict_min, strict_max = PD_DONOR_WINDOWS_STRICT[element]
        extended_min, extended_max = PD_DONOR_WINDOWS_EXTENDED[element]

        pd_distance = distance_a(
            pd_coordinate,
            coordinates[atom_id - 1],
        )

        all_candidates.append({
            "Atom": atom_id,
            "Element": element,
            "PdDistance_A": pd_distance,

            "StrictWindowMin_A": strict_min,
            "StrictWindowMax_A": strict_max,

            "ExtendedWindowMin_A": extended_min,
            "ExtendedWindowMax_A": extended_max,

            "StrictQualifies": (
                strict_min <= pd_distance <= strict_max
            ),

            "ExtendedQualifies": (
                extended_min <= pd_distance <= extended_max
            ),
        })

    all_candidates = sorted(
        all_candidates,
        key=lambda item: item["PdDistance_A"],
    )

    qualifying_candidates = [
        item
        for item in all_candidates
        if item["ExtendedQualifies"]
    ]

    return all_candidates, qualifying_candidates


def find_l26_pi_carbon_candidate(elements, coordinates):
    """
    Identify ligand carbon atoms that can represent Pd interaction
    with the aromatic pi cloud in L26.

    Candidate carbons are ligand C atoms with at least two ligand-C
    covalent neighbours and Pd-C distances of 2.00-3.00 A. The
    nearest qualifying carbon is selected.
    """
    pd_coordinate = coordinates[PD_ATOM - 1]
    ligand_atom_ids = ligand_indices(elements)
    candidates = []

    for atom_id in ligand_atom_ids:
        if normalize_element(elements[atom_id - 1]) != "C":
            continue

        carbon_neighbors = [
            neighbour_id
            for neighbour_id, _ in covalent_neighbors(
                elements, coordinates, atom_id, ligand_atom_ids
            )
            if normalize_element(elements[neighbour_id - 1]) == "C"
        ]

        if len(carbon_neighbors) < 2:
            continue

        pd_c_distance = distance_a(
            pd_coordinate, coordinates[atom_id - 1]
        )

        if L26_PI_C_MIN_DISTANCE_A <= pd_c_distance <= L26_PI_C_MAX_DISTANCE_A:
            candidates.append({
                "Atom": atom_id,
                "Element": "C(pi)",
                "PdDistance_A": pd_c_distance,
                "CarbonNeighborCount": len(carbon_neighbors),
                "CarbonNeighbors": carbon_neighbors,
            })

    return sorted(candidates, key=lambda item: item["PdDistance_A"])


def select_pd_donors(ligand_name, qualifying_candidates, elements=None, coordinates=None):
    """Select the two Pd-coordination representatives for each ligand.

    Standard ligands use the expected N/N, N/O, or N/S donor pair.
    L26 is treated separately as N + aromatic C(pi): N is the direct
    donor and the nearest qualifying aromatic carbon represents the
    Pd interaction with the aromatic pi cloud. Sulfur is not treated
    as a coordinating donor for L26.
    """
    ligand_class = classify_ligand(ligand_name)
    expected_pair = expected_donor_pair(ligand_name)

    qualifying_candidates = sorted(
        qualifying_candidates, key=lambda item: item["PdDistance_A"]
    )

    result = {
        "LigandClass": ligand_class,
        "ExpectedDonorPair": (
            "-".join(expected_pair) if expected_pair is not None else "UNKNOWN"
        ),
        "Donor1Atom": np.nan,
        "Donor1Element": None,
        "Pd_Donor1_Distance_A": np.nan,
        "Donor2Atom": np.nan,
        "Donor2Element": None,
        "Pd_Donor2_Distance_A": np.nan,
        "DonorSelectionCheck": None,
        "PiCoordination": False,
        "PiCoordinationAtom": np.nan,
        "PiCoordinationPd_C_Distance_A": np.nan,
        "PiCoordinationDescription": None,
        "L26_PiCarbonCandidates": None,
    }

    # --------------------------------------------------------
    # Special L26: N + aromatic C(pi)
    # --------------------------------------------------------
    if ligand_class == "NPI":
        if elements is None or coordinates is None:
            result["DonorSelectionCheck"] = "L26_COORDINATES_REQUIRED"
            return result, None, None

        n_candidates = [
            item for item in qualifying_candidates
            if item["Element"] == "N"
        ]
        pi_candidates = find_l26_pi_carbon_candidate(elements, coordinates)
        result["L26_PiCarbonCandidates"] = pi_candidates

        if not n_candidates:
            result["DonorSelectionCheck"] = "L26_NO_QUALIFYING_N_DONOR"
            return result, None, None
        if not pi_candidates:
            result["DonorSelectionCheck"] = "L26_NO_PI_CARBON_CANDIDATE"
            return result, None, None

        donor1_item = n_candidates[0]
        donor2_item = pi_candidates[0]

        result["Donor1Atom"] = donor1_item["Atom"]
        result["Donor1Element"] = "N"
        result["Pd_Donor1_Distance_A"] = donor1_item["PdDistance_A"]
        result["Donor2Atom"] = donor2_item["Atom"]
        result["Donor2Element"] = "C(pi)"
        result["Pd_Donor2_Distance_A"] = donor2_item["PdDistance_A"]
        result["PiCoordination"] = True
        result["PiCoordinationAtom"] = donor2_item["Atom"]
        result["PiCoordinationPd_C_Distance_A"] = donor2_item["PdDistance_A"]
        result["PiCoordinationDescription"] = (
            "Pd interacts with the aromatic pi cloud; the nearest "
            "qualifying aromatic C is used as the geometric representative. "
            "Sulfur is distal/non-coordinating in L26."
        )
        result["DonorSelectionCheck"] = "PASS_N_PI"
        return result, donor1_item["Atom"], donor2_item["Atom"]

    # --------------------------------------------------------
    # Standard N/N, N/O, or N/S selection
    # --------------------------------------------------------
    result["L26_PiCarbonCandidates"] = None

    if len(qualifying_candidates) == 0:
        result["DonorSelectionCheck"] = "NO_QUALIFYING_NOS_DONORS"
        return result, None, None

    selected_pair = None
    if expected_pair is not None and len(qualifying_candidates) >= 2:
        expected_sorted = sorted(expected_pair)
        for i in range(len(qualifying_candidates)):
            for j in range(i + 1, len(qualifying_candidates)):
                pair = [qualifying_candidates[i], qualifying_candidates[j]]
                if sorted([item["Element"] for item in pair]) == expected_sorted:
                    selected_pair = pair
                    break
            if selected_pair is not None:
                break

    if selected_pair is None and len(qualifying_candidates) >= 2:
        selected_pair = qualifying_candidates[:2]
    if selected_pair is None:
        selected_pair = qualifying_candidates[:1]

    selected_pair = sorted(selected_pair, key=lambda item: item["PdDistance_A"])
    donor1 = selected_pair[0]["Atom"]
    donor2 = selected_pair[1]["Atom"] if len(selected_pair) == 2 else None

    result["Donor1Atom"] = donor1
    result["Donor1Element"] = selected_pair[0]["Element"]
    result["Pd_Donor1_Distance_A"] = selected_pair[0]["PdDistance_A"]

    if donor2 is not None:
        result["Donor2Atom"] = donor2
        result["Donor2Element"] = selected_pair[1]["Element"]
        result["Pd_Donor2_Distance_A"] = selected_pair[1]["PdDistance_A"]

    selected_elements = sorted([item["Element"] for item in selected_pair])
    if donor2 is None:
        result["DonorSelectionCheck"] = "ONLY_ONE_QUALIFYING_DONOR"
    elif expected_pair is None:
        result["DonorSelectionCheck"] = "AUTO_DETECTED"
    elif selected_elements == sorted(expected_pair):
        result["DonorSelectionCheck"] = "PASS"
    else:
        result["DonorSelectionCheck"] = "CHECK_EXPECTED_DONOR_MISMATCH"

    return result, donor1, donor2

# ============================================================
# 11. LIGAND DONOR-PATH DESCRIPTORS
# ============================================================

def calculate_ligand_donor_path(
    elements,
    coordinates,
    donor1,
    donor2,
):
    """
    Calculate covalent shortest-path descriptors between donor atoms.

    These are ligand backbone descriptors, not beta-C-H
    palladacycle descriptors.
    """
    result = {
        "LigandDonorPath": None,
        "LigandDonorPathAtomCount": np.nan,
        "LigandDonorPathBondCount": np.nan,
        "LigandDonorPathMeanBond_A": np.nan,
        "LigandDonorPathSD_A": np.nan,
        "LigandDonorPathCV": np.nan,
    }

    if donor1 is None or donor2 is None:
        return result

    graph = build_ligand_graph(elements, coordinates)
    path = shortest_graph_path(graph, donor1, donor2)

    if path is None:
        return result

    result["LigandDonorPath"] = "-".join(str(atom) for atom in path)
    result["LigandDonorPathAtomCount"] = len(path)
    result["LigandDonorPathBondCount"] = len(path) - 1

    bond_lengths = []

    for index in range(len(path) - 1):
        atom_i = path[index]
        atom_j = path[index + 1]

        bond_lengths.append(
            distance_a(
                coordinates[atom_i - 1],
                coordinates[atom_j - 1],
            )
        )

    if len(bond_lengths) > 0:
        mean_bond_length = float(np.mean(bond_lengths))
        sd_bond_length = float(np.std(bond_lengths, ddof=0))

        result["LigandDonorPathMeanBond_A"] = mean_bond_length
        result["LigandDonorPathSD_A"] = sd_bond_length

        if mean_bond_length > 0:
            result["LigandDonorPathCV"] = (
                sd_bond_length / mean_bond_length
            )

    return result


# ============================================================
# 12. S-SUBSTITUENT STRUCTURAL / STERIC DESCRIPTORS
# ============================================================

def identify_s_substituent_atoms(
    elements,
    coordinates,
    sulfur_atom,
    donor_partner_atom=None,
):
    """
    Identify the covalent substituent attached to a coordinating sulfur.

    Strategy:
    1. Build a ligand-only connectivity graph.
    2. Locate ligand atoms directly bonded to S.
    3. Exclude Pd because Pd is outside the ligand graph.
    4. If another coordinating donor is directly bonded to S,
       exclude it from the substituent search.
    5. For every S neighbor, identify the graph component reached
       after removing sulfur.
    6. Select the largest carbon-containing component as the
       likely S-alkyl or S-aryl substituent.

    This is designed for comparing S-Me, S-Et, S-Pr, S-cHex, etc.
    """
    if sulfur_atom is None:
        return {
            "S_SubstituentRootAtom": None,
            "S_SubstituentAtoms": [],
            "S_SubstituentNeighborCandidates": [],
            "S_SubstituentWarning": "No coordinating sulfur atom available.",
        }

    if normalize_element(elements[sulfur_atom - 1]) != "S":
        return {
            "S_SubstituentRootAtom": None,
            "S_SubstituentAtoms": [],
            "S_SubstituentNeighborCandidates": [],
            "S_SubstituentWarning": "Selected sulfur atom is not S.",
        }

    graph = build_ligand_graph(elements, coordinates)

    sulfur_neighbors = sorted(graph.get(sulfur_atom, set()))

    if donor_partner_atom is not None:
        sulfur_neighbors = [
            atom_id
            for atom_id in sulfur_neighbors
            if atom_id != donor_partner_atom
        ]

    if len(sulfur_neighbors) == 0:
        return {
            "S_SubstituentRootAtom": None,
            "S_SubstituentAtoms": [],
            "S_SubstituentNeighborCandidates": [],
            "S_SubstituentWarning": (
                "No ligand covalent neighbors found for coordinating S."
            ),
        }

    # Create a graph with S removed, so each direct neighbor leads
    # to a separate branch/component.
    graph_without_s = {
        atom_id: set(neighbors)
        for atom_id, neighbors in graph.items()
        if atom_id != sulfur_atom
    }

    for atom_id in graph_without_s:
        graph_without_s[atom_id].discard(sulfur_atom)

    branch_records = []

    for neighbor_atom in sulfur_neighbors:
        component = connected_component(
            graph_without_s,
            neighbor_atom,
        )

        component_elements = [
            normalize_element(elements[atom_id - 1])
            for atom_id in component
        ]

        carbon_count = component_elements.count("C")
        heavy_atom_count = sum(
            element != "H"
            for element in component_elements
        )

        branch_records.append({
            "RootAtom": neighbor_atom,
            "RootElement": normalize_element(
                elements[neighbor_atom - 1]
            ),
            "Atoms": sorted(component),
            "CarbonCount": carbon_count,
            "HeavyAtomCount": heavy_atom_count,
            "AtomCount": len(component),
        })

    # Prefer a carbon-containing branch, then choose:
    # largest carbon count -> largest heavy atom count -> largest atom count.
    carbon_branches = [
        branch
        for branch in branch_records
        if branch["CarbonCount"] > 0
    ]

    if len(carbon_branches) > 0:
        selected_branch = sorted(
            carbon_branches,
            key=lambda branch: (
                branch["CarbonCount"],
                branch["HeavyAtomCount"],
                branch["AtomCount"],
            ),
            reverse=True,
        )[0]
    else:
        selected_branch = sorted(
            branch_records,
            key=lambda branch: (
                branch["HeavyAtomCount"],
                branch["AtomCount"],
            ),
            reverse=True,
        )[0]

    return {
        "S_SubstituentRootAtom": selected_branch["RootAtom"],
        "S_SubstituentAtoms": selected_branch["Atoms"],
        "S_SubstituentNeighborCandidates": branch_records,
        "S_SubstituentWarning": None,
    }


def calculate_s_substituent_descriptors(
    ligand_name,
    elements,
    coordinates,
    donor1,
    donor2,
):
    """
    Calculate local steric and structural descriptors for the
    substituent attached to coordinating sulfur.

    Applies to NS ligands or any ligand where one selected Pd donor is S.
    """
    result = {
        "S_Substituent_Class": S_SUBSTITUENT_CLASS.get(
            ligand_name,
            np.nan,
        ),
        "CoordinatingSAtom": np.nan,
        "CoordinatingS_Pd_Distance_A": np.nan,
        "S_SubstituentRootAtom": np.nan,
        "S_SubstituentRootElement": None,
        "S_SubstituentAtomIDs": None,
        "S_SubstituentAtomCount": np.nan,
        "S_SubstituentHeavyAtomCount": np.nan,
        "S_SubstituentCarbonCount": np.nan,
        "S_SubstituentHydrogenCount": np.nan,
        "S_Substituent_HeteroatomCount": np.nan,
        "S_C_SubstituentBond_A": np.nan,
        "S_Substituent_MaxDistanceFromS_A": np.nan,
        "S_Substituent_MeanDistanceFromS_A": np.nan,
        "S_Substituent_MaxInternalDistance_A": np.nan,
        "S_Substituent_SASA_A2": np.nan,
        "S_Substituent_SASA_Volume_A3": np.nan,
        "S_Substituent_SolidAngle_sr": np.nan,
        "S_Substituent_ConeAngle_deg": np.nan,
        "S_Substituent_G_percent": np.nan,
        "S_Substituent_%Vbur_3.0A": np.nan,
        "S_Substituent_%Vbur_4.0A": np.nan,
        "S_Substituent_%Vbur_5.5A": np.nan,
        "S_SubstituentWarning": None,
    }

    # Identify which selected donor is sulfur.
    sulfur_atom = None
    partner_donor = None

    if donor1 is not None:
        if normalize_element(elements[donor1 - 1]) == "S":
            sulfur_atom = donor1
            partner_donor = donor2

    if sulfur_atom is None and donor2 is not None:
        if normalize_element(elements[donor2 - 1]) == "S":
            sulfur_atom = donor2
            partner_donor = donor1

    if sulfur_atom is None:
        result["S_SubstituentWarning"] = (
            "No coordinating sulfur donor was selected."
        )
        return result

    result["CoordinatingSAtom"] = sulfur_atom
    result["CoordinatingS_Pd_Distance_A"] = distance_a(
        coordinates[PD_ATOM - 1],
        coordinates[sulfur_atom - 1],
    )

    substituent_info = identify_s_substituent_atoms(
        elements=elements,
        coordinates=coordinates,
        sulfur_atom=sulfur_atom,
        donor_partner_atom=partner_donor,
    )

    substituent_atom_ids = substituent_info["S_SubstituentAtoms"]
    root_atom = substituent_info["S_SubstituentRootAtom"]

    if substituent_info["S_SubstituentWarning"] is not None:
        result["S_SubstituentWarning"] = (
            substituent_info["S_SubstituentWarning"]
        )

    if root_atom is None or len(substituent_atom_ids) == 0:
        return result

    result["S_SubstituentRootAtom"] = root_atom
    result["S_SubstituentRootElement"] = normalize_element(
        elements[root_atom - 1]
    )

    result["S_SubstituentAtomIDs"] = "-".join(
        str(atom_id)
        for atom_id in substituent_atom_ids
    )

    substituent_elements = [
        normalize_element(elements[atom_id - 1])
        for atom_id in substituent_atom_ids
    ]

    substituent_coordinates = coordinates[
        np.asarray(substituent_atom_ids, dtype=int) - 1
    ]

    result["S_SubstituentAtomCount"] = len(substituent_atom_ids)
    result["S_SubstituentHeavyAtomCount"] = sum(
        element != "H"
        for element in substituent_elements
    )
    result["S_SubstituentCarbonCount"] = substituent_elements.count("C")
    result["S_SubstituentHydrogenCount"] = substituent_elements.count("H")
    result["S_Substituent_HeteroatomCount"] = sum(
        element not in {"H", "C"}
        for element in substituent_elements
    )

    # S-root bond length. Normally this should correspond to S-C.
    result["S_C_SubstituentBond_A"] = distance_a(
        coordinates[sulfur_atom - 1],
        coordinates[root_atom - 1],
    )

    sulfur_coordinate = coordinates[sulfur_atom - 1]

    sulfur_distances = [
        distance_a(
            sulfur_coordinate,
            coordinates[atom_id - 1],
        )
        for atom_id in substituent_atom_ids
    ]

    if len(sulfur_distances) > 0:
        result["S_Substituent_MaxDistanceFromS_A"] = float(
            np.max(sulfur_distances)
        )
        result["S_Substituent_MeanDistanceFromS_A"] = float(
            np.mean(sulfur_distances)
        )

    max_internal_distance = 0.0

    for index_i in range(len(substituent_atom_ids)):
        for index_j in range(index_i + 1, len(substituent_atom_ids)):
            current_distance = distance_a(
                substituent_coordinates[index_i],
                substituent_coordinates[index_j],
            )

            max_internal_distance = max(
                max_internal_distance,
                current_distance,
            )

    result["S_Substituent_MaxInternalDistance_A"] = (
        max_internal_distance
    )

    # Local substituent SASA and volume
    try:
        substituent_sasa = SASA(
            substituent_elements,
            substituent_coordinates,
            probe_radius=SASA_PROBE_RADIUS,
        )

        result["S_Substituent_SASA_A2"] = finite_value(
            substituent_sasa.area
        )
        result["S_Substituent_SASA_Volume_A3"] = finite_value(
            substituent_sasa.volume
        )

    except Exception as error:
        result["S_SubstituentWarning"] = (
            f"{result['S_SubstituentWarning'] or ''} "
            f"S-substituent SASA failed: {error}"
        ).strip()

    # S-substituent solid angle viewed from Pd.
    try:
        reduced_elements = ["Pd"] + substituent_elements
        reduced_coordinates = np.vstack(
            [
                coordinates[PD_ATOM - 1],
                substituent_coordinates,
            ]
        )

        solid_angle = SolidAngle(
            reduced_elements,
            reduced_coordinates,
            metal_index=1,
            radii_type=SOLID_ANGLE_RADII_TYPE,
        )

        result["S_Substituent_SolidAngle_sr"] = finite_value(
            solid_angle.solid_angle
        )
        result["S_Substituent_ConeAngle_deg"] = finite_value(
            solid_angle.cone_angle
        )
        result["S_Substituent_G_percent"] = finite_value(
            solid_angle.G
        )

    except Exception as error:
        result["S_SubstituentWarning"] = (
            f"{result['S_SubstituentWarning'] or ''} "
            f"S-substituent solid-angle failed: {error}"
        ).strip()

    # Pd-centred buried volume due only to S-substituent atoms.
    #
    # Pd remains atom 1.
    # All atoms except Pd and S-substituent atoms are excluded.
    try:
        atoms_to_include = set(substituent_atom_ids)

        excluded_atoms = [
            atom_id
            for atom_id in range(1, len(elements) + 1)
            if atom_id != PD_ATOM
            and atom_id not in atoms_to_include
        ]

        for radius in VBUR_RADII:
            descriptor_name = (
                f"S_Substituent_%Vbur_{radius:.1f}A"
            )

            buried_volume = BuriedVolume(
                elements,
                coordinates,
                metal_index=PD_ATOM,
                excluded_atoms=excluded_atoms,
                radius=radius,
                include_hs=False,
                radii_type=VBUR_RADII_TYPE,
                density=VBUR_DENSITY,
            )

            result[descriptor_name] = finite_value(
                100.0 * buried_volume.fraction_buried_volume
            )

    except Exception as error:
        result["S_SubstituentWarning"] = (
            f"{result['S_SubstituentWarning'] or ''} "
            f"S-substituent Vbur failed: {error}"
        ).strip()

    return result


# ============================================================
# 13. WHOLE-LIGAND STERIC DESCRIPTORS
# ============================================================

def calculate_ligand_steric_descriptors(
    elements,
    coordinates,
    donor1,
):
    """
    Calculate descriptors using ligand atoms 20-last only.

    The substrate atoms 2-19 are excluded from ligand-specific
    buried-volume calculations.
    """
    result = {
        "LigandMaximumExtension_A": np.nan,
        "LigandSASA_A2": np.nan,
        "LigandSASA_Volume_A3": np.nan,
        "LigandSolidAngle_sr": np.nan,
        "LigandSolidConeAngle_deg": np.nan,
        "LigandSolidG_percent": np.nan,
        "Ligand_%Vbur_3.0A": np.nan,
        "Ligand_%Vbur_4.0A": np.nan,
        "Ligand_%Vbur_5.5A": np.nan,
        "LigandDistalVolumeOutside_3.5A_A3": np.nan,
        "LigandMolecularVolume_A3": np.nan,
        "LigandDistalVolumeFractionOutside_3.5A": np.nan,
        "LigandPint": np.nan,
        "LigandPint_per_SASA": np.nan,
        "LigandPint_per_Volume": np.nan,
        "PdOrientedSterimol_B1_A": np.nan,
        "PdOrientedSterimol_B5_A": np.nan,
        "PdOrientedSterimol_L_A": np.nan,
        "LigandStericWarning": None,
    }

    ligand_atom_ids = ligand_indices(elements)

    ligand_elements = [
        normalize_element(elements[atom_id - 1])
        for atom_id in ligand_atom_ids
    ]

    ligand_coordinates = coordinates[
        np.asarray(ligand_atom_ids, dtype=int) - 1
    ]

    # Maximum ligand extension
    maximum_extension = 0.0

    for index_i in range(len(ligand_atom_ids)):
        for index_j in range(index_i + 1, len(ligand_atom_ids)):
            current_distance = distance_a(
                ligand_coordinates[index_i],
                ligand_coordinates[index_j],
            )

            maximum_extension = max(
                maximum_extension,
                current_distance,
            )

    result["LigandMaximumExtension_A"] = finite_value(
        maximum_extension
    )

    # Ligand SASA
    try:
        ligand_sasa = SASA(
            ligand_elements,
            ligand_coordinates,
            probe_radius=SASA_PROBE_RADIUS,
        )

        result["LigandSASA_A2"] = finite_value(ligand_sasa.area)
        result["LigandSASA_Volume_A3"] = finite_value(
            ligand_sasa.volume
        )

    except Exception as error:
        result["LigandStericWarning"] = (
            f"Ligand SASA failed: {error}"
        )

    # Ligand solid angle around Pd
    try:
        reduced_elements = ["Pd"] + ligand_elements
        reduced_coordinates = np.vstack(
            [
                coordinates[PD_ATOM - 1],
                ligand_coordinates,
            ]
        )

        solid_angle = SolidAngle(
            reduced_elements,
            reduced_coordinates,
            metal_index=1,
            radii_type=SOLID_ANGLE_RADII_TYPE,
        )

        result["LigandSolidAngle_sr"] = finite_value(
            solid_angle.solid_angle
        )
        result["LigandSolidConeAngle_deg"] = finite_value(
            solid_angle.cone_angle
        )
        result["LigandSolidG_percent"] = finite_value(
            solid_angle.G
        )

    except Exception as error:
        result["LigandStericWarning"] = (
            f"{result['LigandStericWarning'] or ''} "
            f"Ligand solid angle failed: {error}"
        ).strip()

    # Ligand-only buried volume around Pd.
    # Exclude every non-Pd atom that is not a ligand atom.
    try:
        ligand_atom_set = set(ligand_atom_ids)

        excluded_atoms = [
            atom_id
            for atom_id in range(1, len(elements) + 1)
            if atom_id != PD_ATOM
            and atom_id not in ligand_atom_set
        ]

        for radius in VBUR_RADII:
            descriptor_name = f"Ligand_%Vbur_{radius:.1f}A"

            buried_volume = BuriedVolume(
                elements,
                coordinates,
                metal_index=PD_ATOM,
                excluded_atoms=excluded_atoms,
                radius=radius,
                include_hs=False,
                radii_type=VBUR_RADII_TYPE,
                density=VBUR_DENSITY,
            )

            result[descriptor_name] = finite_value(
                buried_volume.fraction_buried_volume * 100.0
            )

        distal_volume_calculation = BuriedVolume(
            elements,
            coordinates,
            metal_index=PD_ATOM,
            excluded_atoms=excluded_atoms,
            radius=DISTAL_VOLUME_RADIUS,
            include_hs=False,
            radii_type=VBUR_RADII_TYPE,
            density=VBUR_DENSITY,
        )

        distal_volume_calculation.compute_distal_volume(
            method="sasa"
        )

        distal_volume = finite_value(
            distal_volume_calculation.distal_volume
        )
        molecular_volume = finite_value(
            distal_volume_calculation.molecular_volume
        )

        result[
            "LigandDistalVolumeOutside_3.5A_A3"
        ] = distal_volume

        result["LigandMolecularVolume_A3"] = molecular_volume

        if (
            pd.notna(distal_volume)
            and pd.notna(molecular_volume)
            and molecular_volume > 0
        ):
            result[
                "LigandDistalVolumeFractionOutside_3.5A"
            ] = distal_volume / molecular_volume

    except Exception as error:
        result["LigandStericWarning"] = (
            f"{result['LigandStericWarning'] or ''} "
            f"Ligand buried volume failed: {error}"
        ).strip()

    # Ligand dispersion descriptor
    if CALCULATE_DISPERSION and DISPERSION_AVAILABLE:
        try:
            dispersion = Dispersion(
                ligand_elements,
                ligand_coordinates,
            )

            pint = finite_value(dispersion.p_int)

            result["LigandPint"] = pint

            sasa_area = result.get("LigandSASA_A2", np.nan)
            ligand_volume = result.get(
                "LigandMolecularVolume_A3",
                np.nan,
            )

            if pd.notna(pint) and pd.notna(sasa_area) and sasa_area > 0:
                result["LigandPint_per_SASA"] = pint / sasa_area

            if (
                pd.notna(pint)
                and pd.notna(ligand_volume)
                and ligand_volume > 0
            ):
                result["LigandPint_per_Volume"] = pint / ligand_volume

        except Exception as error:
            result["LigandStericWarning"] = (
                f"{result['LigandStericWarning'] or ''} "
                f"Ligand dispersion failed: {error}"
            ).strip()

    # Pd-oriented Sterimol.
    #
    # This describes ligand dimensions with Pd as a dummy atom and
    # the selected donor atom as the attached atom.
    if donor1 is not None:
        try:
            reduced_elements = ["Pd"] + ligand_elements
            reduced_coordinates = np.vstack(
                [
                    coordinates[PD_ATOM - 1],
                    ligand_coordinates,
                ]
            )

            original_to_reduced = {
                original_atom: reduced_atom
                for reduced_atom, original_atom in enumerate(
                    ligand_atom_ids,
                    start=2,
                )
            }

            donor1_reduced = original_to_reduced[donor1]

            sterimol = Sterimol(
                reduced_elements,
                reduced_coordinates,
                dummy_index=1,
                attached_index=donor1_reduced,
                radii_type=STERIMOL_RADII_TYPE,
            )

            result["PdOrientedSterimol_B1_A"] = finite_value(
                sterimol.B_1_value
            )
            result["PdOrientedSterimol_B5_A"] = finite_value(
                sterimol.B_5_value
            )
            result["PdOrientedSterimol_L_A"] = finite_value(
                sterimol.L_value
            )

        except Exception as error:
            result["LigandStericWarning"] = (
                f"{result['LigandStericWarning'] or ''} "
                f"Pd-oriented Sterimol failed: {error}"
            ).strip()

    return result


# ============================================================
# 14. Pd COORDINATION DESCRIPTORS
# ============================================================

def calculate_coordination_descriptors(
    elements,
    coordinates,
    donor1,
    donor2,
):
    """Calculate Pd-donor geometry descriptors."""
    result = {
        "Pd_Donor1_Distance_A": np.nan,
        "Pd_Donor2_Distance_A": np.nan,
        "Average_Pd_Donor_Distance_A": np.nan,
        "Delta_Pd_Donor_Distance_A": np.nan,
        "Pd_Donor_Distance_Ratio": np.nan,
        "Donor1_Donor2_Distance_A": np.nan,
        "Bite_Angle_deg": np.nan,
        "Pd_Coordination_Distortion_A": np.nan,
        "Pd_N_PiC_DistanceDifference_A": np.nan,
        "Pd_N_PiC_DistanceRatio": np.nan,
    }

    if donor1 is None:
        return result

    pd_coordinate = coordinates[PD_ATOM - 1]
    donor1_coordinate = coordinates[donor1 - 1]

    donor1_distance = distance_a(
        pd_coordinate,
        donor1_coordinate,
    )

    result["Pd_Donor1_Distance_A"] = donor1_distance

    if donor2 is None:
        return result

    donor2_coordinate = coordinates[donor2 - 1]

    donor2_distance = distance_a(
        pd_coordinate,
        donor2_coordinate,
    )

    result["Pd_Donor2_Distance_A"] = donor2_distance

    if (
        normalize_element(elements[donor1 - 1]) == "N"
        and normalize_element(elements[donor2 - 1]) == "C"
    ):
        result["Pd_N_PiC_DistanceDifference_A"] = abs(
            donor1_distance - donor2_distance
        )
        if donor2_distance > 0:
            result["Pd_N_PiC_DistanceRatio"] = (
                donor1_distance / donor2_distance
            )

    average_distance = (
        donor1_distance + donor2_distance
    ) / 2.0

    result["Average_Pd_Donor_Distance_A"] = average_distance
    result["Delta_Pd_Donor_Distance_A"] = abs(
        donor1_distance - donor2_distance
    )

    if min(donor1_distance, donor2_distance) > 0:
        result["Pd_Donor_Distance_Ratio"] = (
            max(donor1_distance, donor2_distance)
            / min(donor1_distance, donor2_distance)
        )

    result["Donor1_Donor2_Distance_A"] = distance_a(
        donor1_coordinate,
        donor2_coordinate,
    )

    result["Bite_Angle_deg"] = angle_deg(
        donor1_coordinate,
        pd_coordinate,
        donor2_coordinate,
    )

    result["Pd_Coordination_Distortion_A"] = float(
        np.sqrt(
            (
                (donor1_distance - average_distance) ** 2
                + (donor2_distance - average_distance) ** 2
            ) / 2.0
        )
    )

    return result


# ============================================================
# 15. DONOR LOCAL-GEOMETRY / PYRAMIDALIZATION DESCRIPTORS
# ============================================================

def calculate_one_donor_geometry(
    elements,
    coordinates,
    donor_atom,
    prefix,
):
    """
    Calculate general local donor geometry.

    The mean bond angle works for donors with at least two covalent
    ligand neighbors. Morfeus pyramidalization is calculated only
    when the donor has exactly three neighbors.
    """
    output = {
        f"{prefix}_NeighborCount": np.nan,
        f"{prefix}_MeanBondAngle_deg": np.nan,
        f"{prefix}_MeanBondAngle_N": np.nan,
        f"{prefix}_Pyramidalization_P": np.nan,
        f"{prefix}_Pyramidalization_PAngle_deg": np.nan,
        f"{prefix}_GeometryWarning": None,
    }

    if donor_atom is None:
        output[f"{prefix}_GeometryWarning"] = (
            "Donor atom unavailable."
        )
        return output

    ligand_atom_ids = ligand_indices(elements)

    neighbors = covalent_neighbors(
        elements=elements,
        coordinates=coordinates,
        atom_id=donor_atom,
        allowed_indices=ligand_atom_ids,
    )

    neighbor_ids = [neighbor_id for neighbor_id, _ in neighbors]

    output[f"{prefix}_NeighborCount"] = len(neighbor_ids)

    if len(neighbor_ids) >= 2:
        donor_coordinate = coordinates[donor_atom - 1]

        all_angles = []

        for atom_i, atom_j in itertools.combinations(
            neighbor_ids,
            2,
        ):
            current_angle = angle_deg(
                coordinates[atom_i - 1],
                donor_coordinate,
                coordinates[atom_j - 1],
            )

            if pd.notna(current_angle):
                all_angles.append(current_angle)

        if len(all_angles) > 0:
            output[f"{prefix}_MeanBondAngle_deg"] = float(
                np.mean(all_angles)
            )
            output[f"{prefix}_MeanBondAngle_N"] = len(all_angles)

    if len(neighbor_ids) != 3:
        output[f"{prefix}_GeometryWarning"] = (
            f"{len(neighbor_ids)} covalent ligand neighbors; "
            "Pyramidalization requires exactly 3."
        )
        return output

    try:
        pyramidalization = Pyramidalization(
            coordinates,
            donor_atom,
            neighbor_indices=neighbor_ids,
        )

        output[f"{prefix}_Pyramidalization_P"] = finite_value(
            pyramidalization.P
        )
        output[f"{prefix}_Pyramidalization_PAngle_deg"] = finite_value(
            pyramidalization.P_angle
        )

    except Exception as error:
        output[f"{prefix}_GeometryWarning"] = (
            f"Pyramidalization failed: {error}"
        )

    return output


def calculate_donor_geometry_descriptors(
    elements,
    coordinates,
    donor1,
    donor2,
):
    """Calculate donor geometry and donor-pair summary descriptors."""
    result = {}

    donor1_result = calculate_one_donor_geometry(
        elements,
        coordinates,
        donor1,
        prefix="Donor1",
    )

    donor2_result = calculate_one_donor_geometry(
        elements,
        coordinates,
        donor2,
        prefix="Donor2",
    )

    result.update(donor1_result)
    result.update(donor2_result)

    summary_fields = {
        "AverageDonor_Pyramidalization_P": np.nan,
        "DeltaDonor_Pyramidalization_P": np.nan,
        "AverageDonor_Pyramidalization_PAngle_deg": np.nan,
        "DeltaDonor_Pyramidalization_PAngle_deg": np.nan,
        "AverageDonor_MeanBondAngle_deg": np.nan,
        "DeltaDonor_MeanBondAngle_deg": np.nan,
    }

    result.update(summary_fields)

    p1 = result.get("Donor1_Pyramidalization_P", np.nan)
    p2 = result.get("Donor2_Pyramidalization_P", np.nan)

    if pd.notna(p1) and pd.notna(p2):
        result["AverageDonor_Pyramidalization_P"] = (
            p1 + p2
        ) / 2.0

        result["DeltaDonor_Pyramidalization_P"] = abs(p1 - p2)

    angle1 = result.get(
        "Donor1_Pyramidalization_PAngle_deg",
        np.nan,
    )
    angle2 = result.get(
        "Donor2_Pyramidalization_PAngle_deg",
        np.nan,
    )

    if pd.notna(angle1) and pd.notna(angle2):
        result[
            "AverageDonor_Pyramidalization_PAngle_deg"
        ] = (angle1 + angle2) / 2.0

        result[
            "DeltaDonor_Pyramidalization_PAngle_deg"
        ] = abs(angle1 - angle2)

    mean_angle1 = result.get(
        "Donor1_MeanBondAngle_deg",
        np.nan,
    )
    mean_angle2 = result.get(
        "Donor2_MeanBondAngle_deg",
        np.nan,
    )

    if pd.notna(mean_angle1) and pd.notna(mean_angle2):
        result["AverageDonor_MeanBondAngle_deg"] = (
            mean_angle1 + mean_angle2
        ) / 2.0

        result["DeltaDonor_MeanBondAngle_deg"] = abs(
            mean_angle1 - mean_angle2
        )

    return result


# ============================================================
# 16. xTB ELECTRONIC DESCRIPTORS
# ============================================================

def calculate_xtb_descriptors(
    elements,
    coordinates,
    donor1,
    donor2,
):
    """
    Calculate isolated ligand xTB descriptors using ligand atoms 20-end.

    Pd and substrate atoms are excluded from xTB calculation.
    """
    result = {
        "XTB_LigandChargeSum": np.nan,
        "XTB_HOMO_eV": np.nan,
        "XTB_LUMO_eV": np.nan,
        "XTB_HOMO_LUMO_Gap_eV": np.nan,
        "XTB_FermiLevel_eV": np.nan,
        "XTB_IP_eV": np.nan,
        "XTB_EA_eV": np.nan,
        "XTB_ChemicalPotential_eV": np.nan,
        "XTB_Electronegativity_eV": np.nan,
        "XTB_Hardness_eV": np.nan,
        "XTB_Softness": np.nan,
        "XTB_Electrophilicity": np.nan,
        "XTB_Nucleophilicity": np.nan,
        "XTB_Dipole_Debye": np.nan,
        "XTB_Polarizability": np.nan,
        "XTB_NFOD": np.nan,
        "Donor1_MullikenCharge": np.nan,
        "Donor2_MullikenCharge": np.nan,
        "AverageDonorCharge": np.nan,
        "DeltaDonorCharge": np.nan,
        "DonorChargeSum": np.nan,
        "XTBWarning": None,
    }

    if not RUN_XTB:
        result["XTBWarning"] = "RUN_XTB is False."
        return result

    ligand_atom_ids = ligand_indices(elements)

    ligand_elements = [
        normalize_element(elements[atom_id - 1])
        for atom_id in ligand_atom_ids
    ]

    ligand_coordinates = coordinates[
        np.asarray(ligand_atom_ids, dtype=int) - 1
    ]

    try:
        xtb_calculation = XTB(
            ligand_elements,
            ligand_coordinates,
            method=XTB_METHOD,
            charge=XTB_CHARGE,
            n_unpaired=XTB_NUNPAIRED,
            n_processes=XTB_PROCESSES,
        )

        charges = xtb_calculation.get_charges(model="Mulliken")

        result["XTB_LigandChargeSum"] = finite_value(
            sum(charges.values())
        )

        result["XTB_HOMO_eV"] = finite_value(
            xtb_calculation.get_homo(unit="eV")
        )

        result["XTB_LUMO_eV"] = finite_value(
            xtb_calculation.get_lumo(unit="eV")
        )

        result["XTB_HOMO_LUMO_Gap_eV"] = finite_value(
            xtb_calculation.get_homo_lumo_gap()
        )

        result["XTB_FermiLevel_eV"] = finite_value(
            xtb_calculation.get_fermi_level()
        )

        result["XTB_IP_eV"] = finite_value(
            xtb_calculation.get_ip(corrected=True)
        )

        result["XTB_EA_eV"] = finite_value(
            xtb_calculation.get_ea(corrected=True)
        )

        result["XTB_ChemicalPotential_eV"] = finite_value(
            xtb_calculation.get_chemical_potential(
                corrected=True
            )
        )

        result["XTB_Electronegativity_eV"] = finite_value(
            xtb_calculation.get_electronegativity(
                corrected=True
            )
        )

        result["XTB_Hardness_eV"] = finite_value(
            xtb_calculation.get_hardness()
        )

        result["XTB_Softness"] = finite_value(
            xtb_calculation.get_softness()
        )

        try:
            result["XTB_Electrophilicity"] = finite_value(
                xtb_calculation.get_global_descriptor(
                    "electrophilicity",
                    corrected=True,
                )
            )
        except Exception:
            pass

        try:
            result["XTB_Nucleophilicity"] = finite_value(
                xtb_calculation.get_global_descriptor(
                    "nucleophilicity",
                    corrected=True,
                )
            )
        except Exception:
            pass

        result["XTB_Dipole_Debye"] = finite_value(
            xtb_calculation.get_dipole_moment(unit="debye")
        )

        try:
            result["XTB_Polarizability"] = finite_value(
                xtb_calculation.get_molecular_polarizability()
            )
        except Exception:
            pass

        try:
            result["XTB_NFOD"] = finite_value(
                xtb_calculation.get_nfod()
            )
        except Exception:
            pass

        # Map original XYZ indices (20-last) to isolated ligand xTB
        # indices, which are 1-based in the charge dictionary.
        original_to_xtb_index = {
            original_atom: xtb_atom
            for xtb_atom, original_atom in enumerate(
                ligand_atom_ids,
                start=1,
            )
        }

        if donor1 is not None:
            donor1_xtb_index = original_to_xtb_index.get(donor1)

            if donor1_xtb_index is not None:
                result["Donor1_MullikenCharge"] = finite_value(
                    charges[donor1_xtb_index]
                )

        if donor2 is not None:
            donor2_xtb_index = original_to_xtb_index.get(donor2)

            if donor2_xtb_index is not None:
                result["Donor2_MullikenCharge"] = finite_value(
                    charges[donor2_xtb_index]
                )

        donor_charge_1 = result["Donor1_MullikenCharge"]
        donor_charge_2 = result["Donor2_MullikenCharge"]

        if pd.notna(donor_charge_1) and pd.notna(donor_charge_2):
            result["AverageDonorCharge"] = (
                donor_charge_1 + donor_charge_2
            ) / 2.0

            result["DeltaDonorCharge"] = abs(
                donor_charge_1 - donor_charge_2
            )

            result["DonorChargeSum"] = (
                donor_charge_1 + donor_charge_2
            )

    except Exception as error:
        result["XTBWarning"] = str(error)

    return result


# ============================================================
# 17. PROCESS ONE XYZ FILE
# ============================================================

def process_xyz_file(filepath):
    """Process one Nat_Syn-DM Pd complex XYZ file."""
    ligand_name = clean_ligand_name(filepath)

    elements, coordinates = read_xyz_file(filepath)

    validation_warnings = validate_structure(
        ligand_name,
        elements,
        coordinates,
    )

    row = {
        "Ligand": ligand_name,
        "XYZFile": os.path.basename(filepath),
        "ReactionType": REACTION_TYPE,
        "NAtoms": len(elements),
        "PdAtom": PD_ATOM,
        "SubstrateStartAtom": SUBSTRATE_START,
        "SubstrateEndAtom": SUBSTRATE_END,
        "LigandStartAtom": LIGAND_START,
        "LigandEndAtom": len(elements),
        "Yield": YIELD_DATA.get(ligand_name, np.nan),
        "StructureValidation": (
            "PASS"
            if len(validation_warnings) == 0
            else " | ".join(validation_warnings)
        ),
    }

    # --------------------------------------------------------
    # Pd donor assignment
    # --------------------------------------------------------

    all_candidates, qualifying_candidates = (
        find_pd_donor_candidates(
            elements,
            coordinates,
        )
    )

    donor_information, donor1, donor2 = select_pd_donors(
        ligand_name,
        qualifying_candidates,
        elements=elements,
        coordinates=coordinates,
    )

    row.update(donor_information)

    row["NumberOfNOSLigandCandidates"] = len(all_candidates)
    row["NumberOfQualifyingPdDonors"] = len(
        qualifying_candidates
    )

    # --------------------------------------------------------
    # FIX: find_pd_donor_candidates() now returns Strict/Extended
    # window and qualification keys (StrictWindowMin_A,
    # StrictWindowMax_A, ExtendedWindowMin_A, ExtendedWindowMax_A,
    # StrictQualifies, ExtendedQualifies) instead of the old
    # WindowMin_A / WindowMax_A / Qualifies keys. This block was
    # still reading the old key names, which raised a KeyError
    # inside process_xyz_file() for every ligand. That exception
    # was silently caught by the try/except in Section 18, so every
    # row fell back to the bare stub (Ligand/XYZFile/ReactionType/
    # Yield/ProcessingError only) - which is why master_df never
    # had a "CoordinatingSAtom" column and Section 20 raised
    # KeyError: 'CoordinatingSAtom'. Updated to the current keys,
    # and "qualifying" now refers to ExtendedQualifies (the same
    # criterion find_pd_donor_candidates uses to build
    # qualifying_candidates).
    # --------------------------------------------------------

    row["AllNOSLigandAtoms"] = "; ".join(
        (
            f"{item['Atom']}:{item['Element']}:"
            f"{item['PdDistance_A']:.3f}A "
            f"[strict {item['StrictWindowMin_A']:.2f}-"
            f"{item['StrictWindowMax_A']:.2f}A; "
            f"extended {item['ExtendedWindowMin_A']:.2f}-"
            f"{item['ExtendedWindowMax_A']:.2f}A] "
            f"{'STRICT_PASS' if item['StrictQualifies'] else ('EXTENDED_PASS' if item['ExtendedQualifies'] else 'FAIL')}"
        )
        for item in all_candidates
    )

    row["QualifyingPdDonors"] = "; ".join(
        (
            f"{item['Atom']}:{item['Element']}:"
            f"{item['PdDistance_A']:.3f}A"
        )
        for item in qualifying_candidates
    )

    row["RejectedNOSAtoms"] = "; ".join(
        (
            f"{item['Atom']}:{item['Element']}:"
            f"{item['PdDistance_A']:.3f}A"
        )
        for item in all_candidates
        if not item["ExtendedQualifies"]
    )

    # --------------------------------------------------------
    # Ligand backbone / donor-path descriptors
    # --------------------------------------------------------

    row.update(
        calculate_ligand_donor_path(
            elements,
            coordinates,
            donor1,
            donor2,
        )
    )

    # --------------------------------------------------------
    # Whole-ligand sterics
    # --------------------------------------------------------

    row.update(
        calculate_ligand_steric_descriptors(
            elements,
            coordinates,
            donor1,
        )
    )

    # --------------------------------------------------------
    # Coordination geometry
    # --------------------------------------------------------

    row.update(
        calculate_coordination_descriptors(
            elements,
            coordinates,
            donor1,
            donor2,
        )
    )

    # --------------------------------------------------------
    # Local donor geometry
    # --------------------------------------------------------

    row.update(
        calculate_donor_geometry_descriptors(
            elements,
            coordinates,
            donor1,
            donor2,
        )
    )

    # --------------------------------------------------------
    # S-substituent descriptors
    # --------------------------------------------------------

    row.update(
        calculate_s_substituent_descriptors(
            ligand_name,
            elements,
            coordinates,
            donor1,
            donor2,
        )
    )

    # --------------------------------------------------------
    # Isolated ligand xTB descriptors
    # --------------------------------------------------------

    row.update(
        calculate_xtb_descriptors(
            elements,
            coordinates,
            donor1,
            donor2,
        )
    )

    # --------------------------------------------------------
    # Warnings
    # --------------------------------------------------------

    warning_list = []

    if row["StructureValidation"] != "PASS":
        warning_list.append(row["StructureValidation"])

    if row["DonorSelectionCheck"] != "PASS":
        warning_list.append(
            f"Donor QC: {row['DonorSelectionCheck']}"
        )

    if pd.isna(row.get("Bite_Angle_deg", np.nan)):
        warning_list.append("Bite angle unavailable.")

    if row.get("S_SubstituentWarning"):
        warning_list.append(
            f"S-substituent: {row['S_SubstituentWarning']}"
        )

    if row.get("LigandStericWarning"):
        warning_list.append(
            f"Sterics: {row['LigandStericWarning']}"
        )

    if row.get("XTBWarning"):
        warning_list.append(
            f"xTB: {row['XTBWarning']}"
        )

    row["Warnings"] = " | ".join(warning_list)

    return row


# ============================================================
# 18. PROCESS ALL XYZ FILES
# ============================================================

xyz_files = sorted(
    glob.glob(
        os.path.join(XYZ_FOLDER, "*.xyz")
    )
)

print("=" * 100)
print("XYZ FILE DISCOVERY")
print("=" * 100)
print("XYZ folder:", XYZ_FOLDER)
print("Number of XYZ files found:", len(xyz_files))

if len(xyz_files) == 0:
    raise FileNotFoundError(
        f"No .xyz files found in: {XYZ_FOLDER}"
    )

for xyz_file in xyz_files:
    print(" -", os.path.basename(xyz_file))

all_rows = []

for xyz_file in xyz_files:
    print("\n" + "=" * 100)
    print("PROCESSING:", os.path.basename(xyz_file))
    print("=" * 100)

    try:
        row = process_xyz_file(xyz_file)
        all_rows.append(row)

        print(
            "Ligand:",
            row["Ligand"],
            "| Class:",
            row["LigandClass"],
            "| Donors:",
            row["Donor1Element"],
            row["Donor2Element"],
            "| QC:",
            row["DonorSelectionCheck"],
        )

    except Exception as error:
        ligand_name = clean_ligand_name(xyz_file)

        print(
            f"FAILED: {os.path.basename(xyz_file)}"
        )
        print("Error:", error)

        all_rows.append({
            "Ligand": ligand_name,
            "XYZFile": os.path.basename(xyz_file),
            "ReactionType": REACTION_TYPE,
            "Yield": YIELD_DATA.get(ligand_name, np.nan),
            "ProcessingError": str(error),
        })

master_df = pd.DataFrame(all_rows)

master_df["SortNumber"] = master_df["Ligand"].apply(
    ligand_sort_number
)

master_df = (
    master_df
    .sort_values("SortNumber")
    .drop(columns="SortNumber")
    .reset_index(drop=True)
)

master_df.to_csv(
    MASTER_CSV,
    index=False,
)

print("\n" + "=" * 100)
print("MASTER DESCRIPTOR TABLE")
print("=" * 100)

display(master_df)


# ============================================================
# 19. DONOR ASSIGNMENT QC TABLE
# ============================================================

donor_qc_columns = [
    "Ligand",
    "Yield",
    "LigandClass",
    "ExpectedDonorPair",
    "Donor1Atom",
    "Donor1Element",
    "Pd_Donor1_Distance_A",
    "Donor2Atom",
    "Donor2Element",
    "Pd_Donor2_Distance_A",
    "DonorSelectionCheck",
    "PiCoordination",
    "PiCoordinationAtom",
    "PiCoordinationPd_C_Distance_A",
    "PiCoordinationDescription",
    "NumberOfNOSLigandCandidates",
    "NumberOfQualifyingPdDonors",
    "AllNOSLigandAtoms",
    "QualifyingPdDonors",
    "RejectedNOSAtoms",
    "StructureValidation",
    "Warnings",
]

donor_qc_columns = [
    column
    for column in donor_qc_columns
    if column in master_df.columns
]

donor_qc_df = master_df[donor_qc_columns].copy()

print("\n" + "=" * 100)
print("DONOR ASSIGNMENT QUALITY CONTROL")
print("=" * 100)

display(donor_qc_df)


# ============================================================
# 20. S-SUBSTITUENT ANALYSIS TABLE
# ============================================================

s_substituent_columns = [
    "Ligand",
    "Yield",
    "LigandClass",
    "S_Substituent_Class",
    "CoordinatingSAtom",
    "CoordinatingS_Pd_Distance_A",
    "S_SubstituentRootAtom",
    "S_SubstituentRootElement",
    "S_SubstituentAtomIDs",
    "S_SubstituentAtomCount",
    "S_SubstituentHeavyAtomCount",
    "S_SubstituentCarbonCount",
    "S_SubstituentHydrogenCount",
    "S_Substituent_HeteroatomCount",
    "S_C_SubstituentBond_A",
    "S_Substituent_MaxDistanceFromS_A",
    "S_Substituent_MeanDistanceFromS_A",
    "S_Substituent_MaxInternalDistance_A",
    "S_Substituent_SASA_A2",
    "S_Substituent_SASA_Volume_A3",
    "S_Substituent_SolidAngle_sr",
    "S_Substituent_ConeAngle_deg",
    "S_Substituent_G_percent",
    "S_Substituent_%Vbur_3.0A",
    "S_Substituent_%Vbur_4.0A",
    "S_Substituent_%Vbur_5.5A",
    "S_SubstituentWarning",
]

s_substituent_columns = [
    column
    for column in s_substituent_columns
    if column in master_df.columns
]

s_substituent_df = master_df[
    master_df["CoordinatingSAtom"].notna()
][s_substituent_columns].copy()

print("\n" + "=" * 100)
print("S-SUBSTITUENT STERIC / STRUCTURAL DESCRIPTORS")
print("=" * 100)

display(s_substituent_df)


# ============================================================
# 21. YIELD CORRELATION ANALYSIS
# ============================================================

# Exclude metadata, atom-index labels, and QC columns from
# numerical descriptor correlation screening.

non_predictor_columns = {
    "Yield",
    "NAtoms",
    "PdAtom",
    "SubstrateStartAtom",
    "SubstrateEndAtom",
    "LigandStartAtom",
    "LigandEndAtom",
    "Donor1Atom",
    "Donor2Atom",
    "NumberOfNOSLigandCandidates",
    "NumberOfQualifyingPdDonors",
    "CoordinatingSAtom",
    "S_SubstituentRootAtom",
    "S_SubstituentAtomCount",
    "S_SubstituentHeavyAtomCount",
    "S_SubstituentCarbonCount",
    "S_SubstituentHydrogenCount",
    "S_Substituent_HeteroatomCount",
    "LigandDonorPathAtomCount",
    "LigandDonorPathBondCount",
    "Donor1_NeighborCount",
    "Donor2_NeighborCount",
    "Donor1_MeanBondAngle_N",
    "Donor2_MeanBondAngle_N",
}

numeric_columns = master_df.select_dtypes(
    include=[np.number]
).columns.tolist()

descriptor_columns = [
    column
    for column in numeric_columns
    if column not in non_predictor_columns
    and column != "Yield"
]

correlation_records = []

for descriptor in descriptor_columns:
    temporary_df = master_df[
        ["Yield", descriptor]
    ].dropna()

    n_observations = len(temporary_df)

    if n_observations < MINIMUM_CORRELATION_N:
        correlation_records.append({
            "Descriptor": descriptor,
            "N": n_observations,
            "Pearson_r": np.nan,
            "Pearson_p": np.nan,
            "Spearman_rho": np.nan,
            "Spearman_p": np.nan,
            "Abs_Pearson_r": np.nan,
            "Abs_Spearman_rho": np.nan,
            "Mean_Abs_Correlation": np.nan,
        })
        continue

    try:
        pearson_r, pearson_p = pearsonr(
            temporary_df["Yield"],
            temporary_df[descriptor],
        )
    except Exception:
        pearson_r, pearson_p = np.nan, np.nan

    try:
        spearman_rho, spearman_p = spearmanr(
            temporary_df["Yield"],
            temporary_df[descriptor],
        )
    except Exception:
        spearman_rho, spearman_p = np.nan, np.nan

    correlation_records.append({
        "Descriptor": descriptor,
        "N": n_observations,
        "Pearson_r": pearson_r,
        "Pearson_p": pearson_p,
        "Spearman_rho": spearman_rho,
        "Spearman_p": spearman_p,
        "Abs_Pearson_r": (
            abs(pearson_r)
            if pd.notna(pearson_r)
            else np.nan
        ),
        "Abs_Spearman_rho": (
            abs(spearman_rho)
            if pd.notna(spearman_rho)
            else np.nan
        ),
        "Mean_Abs_Correlation": (
            np.nanmean([
                abs(pearson_r)
                if pd.notna(pearson_r)
                else np.nan,
                abs(spearman_rho)
                if pd.notna(spearman_rho)
                else np.nan,
            ])
        ),
    })

yield_correlation_df = pd.DataFrame(
    correlation_records
)

# Benjamini-Hochberg FDR correction, if statsmodels is available.
try:
    from statsmodels.stats.multitest import multipletests

    valid_pearson = yield_correlation_df["Pearson_p"].notna()

    yield_correlation_df["Pearson_FDR_p"] = np.nan

    if valid_pearson.sum() > 0:
        yield_correlation_df.loc[
            valid_pearson,
            "Pearson_FDR_p",
        ] = multipletests(
            yield_correlation_df.loc[
                valid_pearson,
                "Pearson_p",
            ],
            method="fdr_bh",
        )[1]

    valid_spearman = yield_correlation_df["Spearman_p"].notna()

    yield_correlation_df["Spearman_FDR_p"] = np.nan

    if valid_spearman.sum() > 0:
        yield_correlation_df.loc[
            valid_spearman,
            "Spearman_FDR_p",
        ] = multipletests(
            yield_correlation_df.loc[
                valid_spearman,
                "Spearman_p",
            ],
            method="fdr_bh",
        )[1]

except Exception as error:
    yield_correlation_df["CorrelationFDRWarning"] = str(error)

yield_correlation_df = (
    yield_correlation_df
    .sort_values(
        "Mean_Abs_Correlation",
        ascending=False,
        na_position="last",
    )
    .reset_index(drop=True)
)

yield_correlation_df.to_csv(
    YIELD_CORRELATION_CSV,
    index=False,
)

print("\n" + "=" * 100)
print("YIELD CORRELATION RANKING")
print("=" * 100)

display(yield_correlation_df)


# ============================================================
# 22. OUTLIER SCREENING
# ============================================================

def calculate_outlier_statistics(
    values,
    iqr_multiplier=IQR_MULTIPLIER,
    mad_z_limit=MAD_Z_LIMIT,
):
    """
    Calculate IQR and MAD outlier masks.

    This is an audit tool. It flags unusual descriptors but does not
    automatically remove chemically meaningful ligands or descriptors.
    """
    values = pd.Series(values).dropna().astype(float)

    result = {
        "N_Valid": len(values),
        "N_Missing": np.nan,
        "N_Unique": values.nunique(),
        "IQR_Outliers": np.nan,
        "MAD_Outliers": np.nan,
        "Fraction_by_IQR": np.nan,
        "Fraction_by_MAD": np.nan,
        "IQR_Q1": np.nan,
        "IQR_Q3": np.nan,
        "IQR": np.nan,
        "IQR_Lower_Limit": np.nan,
        "IQR_Upper_Limit": np.nan,
        "Median": np.nan,
        "MAD": np.nan,
    }

    if len(values) < 3:
        return result, pd.Series(dtype=bool), pd.Series(dtype=bool)

    q1 = values.quantile(0.25)
    q3 = values.quantile(0.75)
    iqr = q3 - q1

    lower_limit = q1 - iqr_multiplier * iqr
    upper_limit = q3 + iqr_multiplier * iqr

    iqr_mask = (
        (values < lower_limit)
        | (values > upper_limit)
    )

    median_value = values.median()
    mad_value = np.median(
        np.abs(values - median_value)
    )

    if mad_value > 0:
        modified_z_scores = (
            0.6745 * (values - median_value) / mad_value
        )

        mad_mask = (
            np.abs(modified_z_scores) > mad_z_limit
        )
    else:
        mad_mask = pd.Series(
            False,
            index=values.index,
        )

    result.update({
        "IQR_Outliers": int(iqr_mask.sum()),
        "MAD_Outliers": int(mad_mask.sum()),
        "Fraction_by_IQR": float(iqr_mask.mean()),
        "Fraction_by_MAD": float(mad_mask.mean()),
        "IQR_Q1": q1,
        "IQR_Q3": q3,
        "IQR": iqr,
        "IQR_Lower_Limit": lower_limit,
        "IQR_Upper_Limit": upper_limit,
        "Median": median_value,
        "MAD": mad_value,
    })

    return result, iqr_mask, mad_mask


outlier_summary_records = []
outlier_detail_records = []

for descriptor in descriptor_columns:
    numeric_values = master_df[descriptor].dropna()

    if len(numeric_values) < 3:
        outlier_summary_records.append({
            "Descriptor": descriptor,
            "N_Valid": len(numeric_values),
            "N_Missing": master_df[descriptor].isna().sum(),
            "N_Unique": numeric_values.nunique(),
            "Automatic_Decision": "NOT_EVALUABLE_N_LT_3",
            "Publication_Recommendation": (
                "Exclude from outlier screening: "
                "fewer than three valid values."
            ),
        })
        continue

    statistics, iqr_mask, mad_mask = calculate_outlier_statistics(
        numeric_values
    )

    iqr_fraction = statistics["Fraction_by_IQR"]
    mad_fraction = statistics["Fraction_by_MAD"]

    if (
        iqr_fraction >= OUTLIER_FRACTION_LIMIT
        or mad_fraction >= OUTLIER_FRACTION_LIMIT
    ):
        automatic_decision = "FLAGGED_EXCESSIVE_OUTLIERS"
        recommendation = (
            "Manual chemical inspection required; "
            "do not delete descriptor solely from statistical flag."
        )

    elif (
        statistics["IQR_Outliers"] > 0
        or statistics["MAD_Outliers"] > 0
    ):
        automatic_decision = "OUTLIERS_PRESENT_REVIEW"
        recommendation = (
            "Retain provisionally; inspect flagged ligands and "
            "perform sensitivity analysis if necessary."
        )

    else:
        automatic_decision = "NO_OUTLIERS_FLAGGED"
        recommendation = (
            "Retain if chemically valid and complete."
        )

    outlier_summary_records.append({
        "Descriptor": descriptor,
        "N_Valid": len(numeric_values),
        "N_Missing": master_df[descriptor].isna().sum(),
        "N_Unique": numeric_values.nunique(),
        **statistics,
        "Automatic_Decision": automatic_decision,
        "Publication_Recommendation": recommendation,
    })

    q1 = statistics["IQR_Q1"]
    q3 = statistics["IQR_Q3"]
    iqr = statistics["IQR"]
    lower_limit = statistics["IQR_Lower_Limit"]
    upper_limit = statistics["IQR_Upper_Limit"]
    median_value = statistics["Median"]
    mad_value = statistics["MAD"]

    for original_index, value in numeric_values.items():
        current_iqr_flag = bool(
            value < lower_limit or value > upper_limit
        )

        if mad_value > 0:
            modified_z_score = (
                0.6745 * (value - median_value) / mad_value
            )
            current_mad_flag = (
                abs(modified_z_score) > MAD_Z_LIMIT
            )
        else:
            modified_z_score = np.nan
            current_mad_flag = False

        if current_iqr_flag or current_mad_flag:
            if current_iqr_flag and current_mad_flag:
                flag_method = "IQR_AND_MAD"
            elif current_iqr_flag:
                flag_method = "IQR_ONLY"
            else:
                flag_method = "MAD_ONLY"

            outlier_detail_records.append({
                "Descriptor": descriptor,
                "Ligand": master_df.loc[
                    original_index,
                    "Ligand",
                ],
                "Value": value,
                "Flag_Method": flag_method,
                "IQR_Flag": current_iqr_flag,
                "MAD_Flag": current_mad_flag,
                "Modified_Z_Score": modified_z_score,
                "IQR_Lower_Limit": lower_limit,
                "IQR_Upper_Limit": upper_limit,
                "Median": median_value,
                "MAD": mad_value,
            })

outlier_summary_df = pd.DataFrame(
    outlier_summary_records
)

outlier_detail_df = pd.DataFrame(
    outlier_detail_records
)

print("\n" + "=" * 100)
print("OUTLIER SUMMARY")
print("=" * 100)

display(outlier_summary_df)

print("\n" + "=" * 100)
print("OUTLIER DETAILS")
print("=" * 100)

display(outlier_detail_df)


# ============================================================
# 23. PLOTS
# ============================================================

# ------------------------------------------------------------
# S-substituent class vs yield
# ------------------------------------------------------------

s_class_yield_df = master_df[
    master_df["S_Substituent_Class"].notna()
    & master_df["Yield"].notna()
].copy()

s_class_yield_df = s_class_yield_df[
    ~s_class_yield_df["S_Substituent_Class"].astype(str).str.contains(
        "UPDATE",
        case=False,
        na=False,
    )
]

if len(s_class_yield_df) >= 2:
    plt.figure(figsize=(10, 6))

    sns.stripplot(
        data=s_class_yield_df,
        x="S_Substituent_Class",
        y="Yield",
        hue="S_Substituent_Class",
        legend=False,
        size=10,
        jitter=0.10,
    )

    plt.title(
        "γ-Arylation Yield by Coordinating S Substituent"
    )
    plt.xlabel("S substituent class")
    plt.ylabel("Yield")
    plt.grid(axis="y", alpha=0.25)
    plt.tight_layout()

    plt.savefig(
        os.path.join(
            PLOT_FOLDER,
            "Yield_by_S_Substituent_Class.png",
        ),
        dpi=400,
        bbox_inches="tight",
    )

    plt.show()
    plt.close()


# ------------------------------------------------------------
# Yield vs S-substituent descriptors
# ------------------------------------------------------------

s_descriptor_columns = [
    column
    for column in [
        "S_SubstituentCarbonCount",
        "S_SubstituentHeavyAtomCount",
        "S_Substituent_MaxDistanceFromS_A",
        "S_Substituent_MaxInternalDistance_A",
        "S_Substituent_SASA_A2",
        "S_Substituent_SASA_Volume_A3",
        "S_Substituent_SolidAngle_sr",
        "S_Substituent_ConeAngle_deg",
        "S_Substituent_%Vbur_3.0A",
        "S_Substituent_%Vbur_4.0A",
        "S_Substituent_%Vbur_5.5A",
        "CoordinatingS_Pd_Distance_A",
    ]
    if column in master_df.columns
]

for descriptor in s_descriptor_columns:
    plot_df = master_df[
        ["Ligand", "Yield", descriptor]
    ].dropna()

    if len(plot_df) < MINIMUM_CORRELATION_N:
        continue

    plt.figure(figsize=(7, 6))

    sns.regplot(
        data=plot_df,
        x=descriptor,
        y="Yield",
        scatter_kws={"s": 65, "alpha": 0.85},
        line_kws={"color": "black"},
    )

    for _, plot_row in plot_df.iterrows():
        plt.annotate(
            plot_row["Ligand"],
            (
                plot_row[descriptor],
                plot_row["Yield"],
            ),
            xytext=(4, 4),
            textcoords="offset points",
            fontsize=8,
        )

    plt.title(f"Yield vs {descriptor}")
    plt.xlabel(descriptor)
    plt.ylabel("Yield")
    plt.grid(alpha=0.25)
    plt.tight_layout()

    plt.savefig(
        os.path.join(
            SCATTER_FOLDER,
            f"Yield_vs_{safe_filename(descriptor)}.png",
        ),
        dpi=400,
        bbox_inches="tight",
    )

    plt.show()
    plt.close()


# ------------------------------------------------------------
# Top yield correlations
# ------------------------------------------------------------

top_correlation_plot_df = yield_correlation_df.dropna(
    subset=["Mean_Abs_Correlation"]
).head(20)

if len(top_correlation_plot_df) > 0:
    plt.figure(
        figsize=(
            10,
            max(6, 0.35 * len(top_correlation_plot_df)),
        )
    )

    sns.barplot(
        data=top_correlation_plot_df,
        y="Descriptor",
        x="Mean_Abs_Correlation",
        color="steelblue",
    )

    plt.title(
        "Top Descriptor Correlations with γ-Arylation Yield"
    )
    plt.xlabel(
        "Mean of |Pearson r| and |Spearman ρ|"
    )
    plt.ylabel("Descriptor")
    plt.xlim(0, 1)
    plt.grid(axis="x", alpha=0.25)
    plt.tight_layout()

    plt.savefig(
        os.path.join(
            PLOT_FOLDER,
            "Top_Yield_Correlations.png",
        ),
        dpi=400,
        bbox_inches="tight",
    )

    plt.show()
    plt.close()


# ============================================================
# 24. EXPORT EXCEL WORKBOOK
# ============================================================

with pd.ExcelWriter(
    FINAL_EXCEL,
    engine="openpyxl",
) as writer:

    master_df.to_excel(
        writer,
        sheet_name="All_Descriptors",
        index=False,
    )

    donor_qc_df.to_excel(
        writer,
        sheet_name="Donor_QC",
        index=False,
    )

    s_substituent_df.to_excel(
        writer,
        sheet_name="S_Substituent_Descriptors",
        index=False,
    )

    yield_correlation_df.to_excel(
        writer,
        sheet_name="Yield_Correlations",
        index=False,
    )

    outlier_summary_df.to_excel(
        writer,
        sheet_name="Outlier_Summary",
        index=False,
    )

    outlier_detail_df.to_excel(
        writer,
        sheet_name="Outlier_Details",
        index=False,
    )

    # Ligand class / S substituent category table
    category_df = master_df[
        [
            column
            for column in [
                "Ligand",
                "Yield",
                "LigandClass",
                "ExpectedDonorPair",
                "Donor1Element",
                "Donor2Element",
                "S_Substituent_Class",
                "DonorSelectionCheck",
            ]
            if column in master_df.columns
        ]
    ].copy()

    category_df.to_excel(
        writer,
        sheet_name="Ligand_Categories",
        index=False,
    )


# ============================================================
# 25. FINAL SUMMARY
# ============================================================

print("\n" + "=" * 100)
print("ANALYSIS COMPLETE")
print("=" * 100)

print("Input XYZ folder:")
print(XYZ_FOLDER)

print("\nOutput folder:")
print(OUTPUT_FOLDER)

print("\nExcel workbook:")
print(FINAL_EXCEL)

print("\nMaster descriptor CSV:")
print(MASTER_CSV)

print("\nYield correlation CSV:")
print(YIELD_CORRELATION_CSV)

print("\nNumber of processed structures:")
print(len(master_df))

print("\nNumber of structures with yields:")
print(master_df["Yield"].notna().sum())

print("\nDonor assignment QC:")
if "DonorSelectionCheck" in master_df.columns:
    display(
        master_df[
            [
                "Ligand",
                "LigandClass",
                "Donor1Element",
                "Pd_Donor1_Distance_A",
                "Donor2Element",
                "Pd_Donor2_Distance_A",
                "DonorSelectionCheck",
                "Yield",
            ]
        ]
    )

print("\nS-substituent QC:")
if len(s_substituent_df) > 0:
    display(s_substituent_df)

print("\nDone.")


# ============================================================
# ============================================================
# 26. FINAL DESCRIPTOR REDUCTION + PUBLICATION CORRELATION MATRICES
# ============================================================
# PRINCIPLE:
#   1. Remove exact mathematical duplicates first.
#   2. Do NOT remove descriptors merely because they belong to the same
#      chemical/physical family.
#   3. For descriptors in the same family, remove one only when the
#      observed Pearson OR Spearman correlation reaches the redundancy
#      threshold. The descriptor with the stronger Yield association is
#      retained as the representative; valid-N and original order are
#      secondary tie-breakers.
#   4. Apply the same statistical redundancy test to every remaining
#      descriptor pair, irrespective of family.
#   5. Recalculate Yield correlations after the final reduction.
#   6. Export a complete audit so every deletion is traceable.
# ============================================================


# ============================================================
# 26. FINAL REDUCTION / PUBLICATION OUTPUTS
# ============================================================

DESCRIPTOR_REDUNDANCY_THRESHOLD = 0.90
TOP_N_DESCRIPTORS = 10

# Output directory
OUTPUT_DIR = OUTPUT_FOLDER
os.makedirs(OUTPUT_DIR, exist_ok=True)

# Main output files
YIELDCORRELATIONCSV = os.path.join(
    OUTPUT_DIR,
    "Yield_Correlation_Final_Reduced.csv"
)

REDUCTION_AUDIT_XLSX = os.path.join(
    OUTPUT_DIR,
    "Descriptor_Reduction_Audit.xlsx"
)

PEARSON_TOP10_PNG = os.path.join(
    OUTPUT_DIR,
    "Pearson_TOP_10_Reduced_Descriptors.png"
)

SPEARMAN_TOP10_PNG = os.path.join(
    OUTPUT_DIR,
    "Spearman_TOP_10_Reduced_Descriptors.png"
)

PEARSON_ALL_PNG = os.path.join(
    OUTPUT_DIR,
    "Pearson_All_Reduced_Descriptors.png"
)

SPEARMAN_ALL_PNG = os.path.join(
    OUTPUT_DIR,
    "Spearman_All_Reduced_Descriptors.png"
)


# ------------------------------------------------------------
# Descriptor families are CHEMICAL/PHYSICAL GROUPINGS only.
# Membership alone NEVER causes deletion.
# ------------------------------------------------------------
DESCRIPTOR_FAMILIES = {
    "Pd_donor_distance_asymmetry": [
        "Delta_Pd_Donor_Distance_A",
        "Pd_Coordination_Distortion_A",
        "Pd_Donor_Distance_Ratio",
    ],
    "donor_path_bond_variability": [
        "LigandDonorPathSD_A",
        "LigandDonorPathCV",
    ],
    "donor_local_angle_geometry": [
        "Donor1_MeanBondAngle_deg",
        "Donor2_MeanBondAngle_deg",
        "AverageDonor_MeanBondAngle_deg",
        "DeltaDonor_MeanBondAngle_deg",
    ],
    "donor_pocket_separation": [
        "Donor1_Donor2_Distance_A",
        "LigandDonorPathMeanBond_A",
    ],
    "absolute_pd_donor_distances": [
        "Pd_Donor1_Distance_A",
        "Pd_Donor2_Distance_A",
        "Average_Pd_Donor_Distance_A",
    ],
}

# Exact mathematical equivalence.
# For two donor distances d1 and d2:
#   Delta = |d1-d2|
#   CoordinationDistortion = |d1-d2|/2
# Therefore they contain identical information.
EXACT_MATHEMATICAL_EQUIVALENCE = [
    (
        "Delta_Pd_Donor_Distance_A",
        "Pd_Coordination_Distortion_A",
        "Delta_Pd_Donor_Distance_A = 2 * Pd_Coordination_Distortion_A",
    ),
]


def _finite_abs(value):
    try:
        value = float(value)
        return abs(value) if np.isfinite(value) else np.nan
    except Exception:
        return np.nan


def _pairwise_correlations(dataframe, descriptor_1, descriptor_2):
    pair = dataframe[[descriptor_1, descriptor_2]].dropna()
    if len(pair) < MINIMUM_CORRELATION_N:
        return np.nan, np.nan, len(pair)

    try:
        pearson_value = pearsonr(
            pair[descriptor_1], pair[descriptor_2]
        )[0]
    except Exception:
        pearson_value = np.nan

    try:
        spearman_value = spearmanr(
            pair[descriptor_1], pair[descriptor_2]
        )[0]
    except Exception:
        spearman_value = np.nan

    return pearson_value, spearman_value, len(pair)


def _yield_score(dataframe, descriptor):
    if descriptor not in dataframe.columns:
        return np.nan
    pair = dataframe[["Yield", descriptor]].dropna()
    if len(pair) < MINIMUM_CORRELATION_N:
        return np.nan

    values = []
    try:
        values.append(abs(pearsonr(pair["Yield"], pair[descriptor])[0]))
    except Exception:
        pass
    try:
        values.append(abs(spearmanr(pair["Yield"], pair[descriptor])[0]))
    except Exception:
        pass

    values = [x for x in values if pd.notna(x)]
    return float(np.mean(values)) if values else np.nan


def _family_for_pair(descriptor_1, descriptor_2):
    families = []
    for family_name, members in DESCRIPTOR_FAMILIES.items():
        if descriptor_1 in members and descriptor_2 in members:
            families.append(family_name)
    return families[0] if families else ""


def _choose_representative(descriptors, dataframe, yield_scores, original_order):
    """Yield score -> valid N -> original order."""
    ranking = []
    for descriptor in descriptors:
        score = yield_scores.get(descriptor, np.nan)
        ranking.append({
            "Descriptor": descriptor,
            "YieldScore": score if pd.notna(score) else -np.inf,
            "ValidN": int(dataframe[descriptor].notna().sum()),
            "OriginalOrder": original_order.get(descriptor, 10**9),
        })

    ranking_df = pd.DataFrame(ranking).sort_values(
        ["YieldScore", "ValidN", "OriginalOrder"],
        ascending=[False, False, True],
    )
    return ranking_df.iloc[0]["Descriptor"]


def reduce_descriptor_redundancy(
    dataframe,
    candidate_descriptors,
    threshold=0.90,
):
    """
    Four-stage reduction:
      A. Exact mathematical equivalence.
      B. Chemical/physical family-aware statistical reduction.
      C. Global statistical redundancy reduction for all remaining pairs.
      D. Final verification that no remaining pair exceeds threshold.

    Chemical/physical family membership is descriptive only; it does not
    automatically delete a descriptor.
    """
    candidate_descriptors = [
        d for d in candidate_descriptors
        if d in dataframe.columns and d != "Yield"
    ]
    original_order = {
        descriptor: i for i, descriptor in enumerate(candidate_descriptors)
    }
    yield_scores = {
        descriptor: _yield_score(dataframe, descriptor)
        for descriptor in candidate_descriptors
    }

    active = list(candidate_descriptors)
    audit_records = []
    high_corr_records = []

    # --------------------------------------------------------
    # A. EXACT MATHEMATICAL EQUIVALENCE
    # --------------------------------------------------------
    for d1, d2, relationship in EXACT_MATHEMATICAL_EQUIVALENCE:
        if d1 not in active or d2 not in active:
            continue

        retained = _choose_representative(
            [d1, d2], dataframe, yield_scores, original_order
        )
        removed = d2 if retained == d1 else d1
        active.remove(removed)

        audit_records.append({
            "DescriptorRemoved": removed,
            "DescriptorRetained": retained,
            "ReductionLevel": "A_Mathematical",
            "Family": "Exact_mathematical_equivalence",
            "PearsonCorrelation": 1.0,
            "SpearmanCorrelation": 1.0,
            "MaximumAbsoluteCorrelation": 1.0,
            "PairwiseN": int(
                dataframe[[d1, d2]].dropna().shape[0]
            ),
            "Reason": "Exact mathematical equivalence",
            "Relationship": relationship,
            "RetainedYieldScore": yield_scores.get(retained, np.nan),
            "RemovedYieldScore": yield_scores.get(removed, np.nan),
        })

    # --------------------------------------------------------
    # B. FAMILY-AWARE STATISTICAL REDUCTION
    # --------------------------------------------------------
    # Only highly correlated members of the same chemical/physical family
    # are reduced here. A family by itself is NOT a deletion criterion.
    family_changed = True
    while family_changed:
        family_changed = False

        for family_name, members in DESCRIPTOR_FAMILIES.items():
            available = [d for d in members if d in active]
            if len(available) < 2:
                continue

            best_pair = None
            best_abs_corr = -np.inf

            for i in range(len(available)):
                for j in range(i + 1, len(available)):
                    d1, d2 = available[i], available[j]
                    pr, sr, n_pair = _pairwise_correlations(
                        dataframe, d1, d2
                    )
                    max_abs = np.nanmax([
                        _finite_abs(pr),
                        _finite_abs(sr),
                    ])

                    if pd.notna(max_abs):
                        high_corr_records.append({
                            "Descriptor1": d1,
                            "Descriptor2": d2,
                            "Family": family_name,
                            "PearsonCorrelation": pr,
                            "SpearmanCorrelation": sr,
                            "MaximumAbsoluteCorrelation": max_abs,
                            "PairwiseN": n_pair,
                            "Stage": "B_FamilyAware",
                        })

                    if (
                        pd.notna(max_abs)
                        and max_abs >= threshold
                        and max_abs > best_abs_corr
                    ):
                        best_abs_corr = max_abs
                        best_pair = (d1, d2, pr, sr, n_pair)

            if best_pair is None:
                continue

            d1, d2, pr, sr, n_pair = best_pair
            retained = _choose_representative(
                [d1, d2], dataframe, yield_scores, original_order
            )
            removed = d2 if retained == d1 else d1
            active.remove(removed)

            audit_records.append({
                "DescriptorRemoved": removed,
                "DescriptorRetained": retained,
                "ReductionLevel": "B_Chemical_Physical",
                "Family": family_name,
                "PearsonCorrelation": pr,
                "SpearmanCorrelation": sr,
                "MaximumAbsoluteCorrelation": best_abs_corr,
                "PairwiseN": n_pair,
                "Reason": (
                    "Same chemical/physical descriptor family AND "
                    f"|r| >= {threshold:.2f}"
                ),
                "Relationship": "Family similarity; not assumed mathematically identical",
                "RetainedYieldScore": yield_scores.get(retained, np.nan),
                "RemovedYieldScore": yield_scores.get(removed, np.nan),
            })
            family_changed = True
            break

        # restart after each removal so correlations are always recalculated

    # --------------------------------------------------------
    # C. GLOBAL HIGH-CORRELATION REDUCTION
    # --------------------------------------------------------
    changed = True
    while changed and len(active) >= 2:
        changed = False
        best_pair = None
        best_abs_corr = -np.inf

        for i in range(len(active)):
            for j in range(i + 1, len(active)):
                d1, d2 = active[i], active[j]
                pr, sr, n_pair = _pairwise_correlations(
                    dataframe, d1, d2
                )
                max_abs = np.nanmax([
                    _finite_abs(pr),
                    _finite_abs(sr),
                ])

                if pd.notna(max_abs):
                    high_corr_records.append({
                        "Descriptor1": d1,
                        "Descriptor2": d2,
                        "Family": _family_for_pair(d1, d2) or "Cross-family",
                        "PearsonCorrelation": pr,
                        "SpearmanCorrelation": sr,
                        "MaximumAbsoluteCorrelation": max_abs,
                        "PairwiseN": n_pair,
                        "Stage": "C_Global",
                    })

                if (
                    pd.notna(max_abs)
                    and max_abs >= threshold
                    and max_abs > best_abs_corr
                ):
                    best_abs_corr = max_abs
                    best_pair = (d1, d2, pr, sr, n_pair)

        if best_pair is None:
            break

        d1, d2, pr, sr, n_pair = best_pair
        retained = _choose_representative(
            [d1, d2], dataframe, yield_scores, original_order
        )
        removed = d2 if retained == d1 else d1
        active.remove(removed)

        audit_records.append({
            "DescriptorRemoved": removed,
            "DescriptorRetained": retained,
            "ReductionLevel": "C_Global_Statistical",
            "Family": _family_for_pair(d1, d2) or "Cross-family",
            "PearsonCorrelation": pr,
            "SpearmanCorrelation": sr,
            "MaximumAbsoluteCorrelation": best_abs_corr,
            "PairwiseN": n_pair,
            "Reason": f"Remaining descriptor pair has |r| >= {threshold:.2f}",
            "Relationship": "Statistical redundancy",
            "RetainedYieldScore": yield_scores.get(retained, np.nan),
            "RemovedYieldScore": yield_scores.get(removed, np.nan),
        })
        changed = True

    # --------------------------------------------------------
    # D. FINAL VERIFICATION
    # --------------------------------------------------------
    final_high_corr = []
    if len(active) >= 2:
        for i in range(len(active)):
            for j in range(i + 1, len(active)):
                d1, d2 = active[i], active[j]
                pr, sr, n_pair = _pairwise_correlations(
                    dataframe, d1, d2
                )
                max_abs = np.nanmax([
                    _finite_abs(pr),
                    _finite_abs(sr),
                ])
                if pd.notna(max_abs) and max_abs >= threshold:
                    final_high_corr.append({
                        "Descriptor1": d1,
                        "Descriptor2": d2,
                        "Family": _family_for_pair(d1, d2) or "Cross-family",
                        "PearsonCorrelation": pr,
                        "SpearmanCorrelation": sr,
                        "MaximumAbsoluteCorrelation": max_abs,
                        "PairwiseN": n_pair,
                    })

    active = sorted(active, key=lambda d: original_order[d])
    audit_df = pd.DataFrame(audit_records)
    high_corr_df = pd.DataFrame(high_corr_records)
    final_high_corr_df = pd.DataFrame(final_high_corr)

    return active, audit_df, high_corr_df, final_high_corr_df


# ------------------------------------------------------------
# Establish the descriptor pool from the ORIGINAL filtering stage.
# ------------------------------------------------------------
# This is deliberately constructed here because the original notebook
# does not create a universal `retaineddescriptors` variable before this
# section.
_evaluable = set()
if "yield_correlation_df" in globals():
    _evaluable = set(
        yield_correlation_df.loc[
            yield_correlation_df["Mean_Abs_Correlation"].notna(),
            "Descriptor",
        ]
    )

_excessive_outlier = set()
if "outlier_summary_df" in globals() and "Automatic_Decision" in outlier_summary_df.columns:
    _excessive_outlier = set(
        outlier_summary_df.loc[
            outlier_summary_df["Automatic_Decision"] == "FLAGGED_EXCESSIVE_OUTLIERS",
            "Descriptor",
        ]
    )

_zero_variance = {
    d for d in _evaluable
    if d in master_df.columns
    and (
        pd.isna(master_df[d].std(skipna=True))
        or master_df[d].std(skipna=True) == 0
    )
}

descriptors_before_reduction = [
    d for d in descriptor_columns
    if d in _evaluable
    and d not in _excessive_outlier
    and d not in _zero_variance
    and d in master_df.columns
]

if len(descriptors_before_reduction) < 2:
    raise ValueError(
        "Fewer than two descriptors are available for redundancy reduction."
    )

reduction_base_df = master_df[["Yield"] + descriptors_before_reduction].copy()

(
    retained_reduced_descriptors,
    descriptor_reduction_audit_df,
    high_correlation_pairs_reduction_df,
    final_high_correlation_pairs_df,
) = reduce_descriptor_redundancy(
    dataframe=reduction_base_df,
    candidate_descriptors=descriptors_before_reduction,
    threshold=DESCRIPTOR_REDUNDANCY_THRESHOLD,
)

retaineddescriptors = list(retained_reduced_descriptors)
retained = list(retained_reduced_descriptors)
correlationcolumns = ["Yield"] + retaineddescriptors
correlationdf = master_df[correlationcolumns].copy()

print("=" * 100)
print("FINAL DESCRIPTOR REDUCTION")
print("=" * 100)
print("Descriptors before reduction:", len(descriptors_before_reduction))
print("Descriptors after reduction: ", len(retaineddescriptors))
print("Redundancy threshold:          |Pearson| or |Spearman| >=", DESCRIPTOR_REDUNDANCY_THRESHOLD)
print("\nFinal retained descriptors:")
for descriptor in retaineddescriptors:
    print(" -", descriptor)

print("\nReduction audit:")
display(descriptor_reduction_audit_df)

print("\nFINAL VERIFICATION: remaining high-correlation pairs")
if len(final_high_correlation_pairs_df) == 0:
    print("PASS: No remaining descriptor pair has |Pearson| or |Spearman| >= threshold.")
else:
    display(final_high_correlation_pairs_df)

# ------------------------------------------------------------
# Recalculate Yield correlations AFTER reduction.
# ------------------------------------------------------------
correlation_records = []

for descriptor in retaineddescriptors:
    pair = master_df[["Yield", descriptor]].dropna()
    n = len(pair)

    if n < MINIMUM_CORRELATION_N:
        correlation_records.append({
            "Descriptor": descriptor,
            "N": n,
            "Pearson_r": np.nan,
            "Pearson_p": np.nan,
            "Spearman_rho": np.nan,
            "Spearman_p": np.nan,
            "Abs_Pearson_r": np.nan,
            "Abs_Spearman_rho": np.nan,
            "Mean_Abs_Correlation": np.nan,
        })
        continue

    try:
        pearson_r, pearson_p = pearsonr(pair["Yield"], pair[descriptor])
    except Exception:
        pearson_r, pearson_p = np.nan, np.nan

    try:
        spearman_rho, spearman_p = spearmanr(pair["Yield"], pair[descriptor])
    except Exception:
        spearman_rho, spearman_p = np.nan, np.nan

    abs_p = abs(pearson_r) if pd.notna(pearson_r) else np.nan
    abs_s = abs(spearman_rho) if pd.notna(spearman_rho) else np.nan
    scores = [x for x in [abs_p, abs_s] if pd.notna(x)]

    correlation_records.append({
        "Descriptor": descriptor,
        "N": n,
        "Pearson_r": pearson_r,
        "Pearson_p": pearson_p,
        "Spearman_rho": spearman_rho,
        "Spearman_p": spearman_p,
        "Abs_Pearson_r": abs_p,
        "Abs_Spearman_rho": abs_s,
        "Mean_Abs_Correlation": float(np.mean(scores)) if scores else np.nan,
    })

yieldcorrelationdf = pd.DataFrame(correlation_records)

# Benjamini-Hochberg FDR correction.
yieldcorrelationdf["Pearson_FDR_p"] = np.nan
yieldcorrelationdf["Spearman_FDR_p"] = np.nan

try:
    from statsmodels.stats.multitest import multipletests

    valid = yieldcorrelationdf["Pearson_p"].notna()
    if valid.any():
        yieldcorrelationdf.loc[valid, "Pearson_FDR_p"] = multipletests(
            yieldcorrelationdf.loc[valid, "Pearson_p"],
            method="fdr_bh",
        )[1]

    valid = yieldcorrelationdf["Spearman_p"].notna()
    if valid.any():
        yieldcorrelationdf.loc[valid, "Spearman_FDR_p"] = multipletests(
            yieldcorrelationdf.loc[valid, "Spearman_p"],
            method="fdr_bh",
        )[1]
except Exception as error:
    yieldcorrelationdf["FDR_Warning"] = str(error)

yieldcorrelationdf = yieldcorrelationdf.sort_values(
    "Mean_Abs_Correlation", ascending=False, na_position="last"
).reset_index(drop=True)

yieldcorrelationdf.to_csv(YIELDCORRELATIONCSV, index=False)

print("\n" + "=" * 100)
print("YIELD CORRELATIONS AFTER FINAL REDUCTION")
print("=" * 100)
display(yieldcorrelationdf)

# ------------------------------------------------------------
# Correlation matrices: TOP 10 and ALL FINAL REDUCED.
# ------------------------------------------------------------
reduced_correlation_df = master_df[["Yield"] + retaineddescriptors].copy()

topnsummary = yieldcorrelationdf.dropna(
    subset=["Mean_Abs_Correlation"]
).head(TOP_N_DESCRIPTORS).copy()

topn_descriptors = topnsummary["Descriptor"].tolist()
topn_df = reduced_correlation_df[["Yield"] + topn_descriptors].copy()

pearson_matrix_topn = topn_df.corr(method="pearson")
spearman_matrix_topn = topn_df.corr(method="spearman")
pearson_matrix_reduced = reduced_correlation_df.corr(method="pearson")
spearman_matrix_reduced = reduced_correlation_df.corr(method="spearman")


def _publication_label(label, width=22):
    text = str(label).replace("_", " ")
    words = text.split()
    lines, current = [], ""
    for word in words:
        if not current:
            current = word
        elif len(current) + len(word) + 1 <= width:
            current += " " + word
        else:
            lines.append(current)
            current = word
    if current:
        lines.append(current)
    return "\n".join(lines)


def plot_publication_correlation_matrix(matrix, title, cbar_label, filepath):
    n = matrix.shape[0]
    if n <= 11:
        fig_w, fig_h, label_fs, annot_fs, lw = 15, 13, 11, 10, 0.6
    elif n <= 20:
        fig_w, fig_h, label_fs, annot_fs, lw = 19, 17, 9, 8, 0.5
    elif n <= 30:
        fig_w, fig_h, label_fs, annot_fs, lw = 24, 22, 8, 7, 0.45
    else:
        fig_w = max(28, 0.82 * n)
        fig_h = max(26, 0.76 * n)
        label_fs, annot_fs, lw = 7, 6, 0.35

    display_matrix = matrix.copy()
    display_matrix.index = [_publication_label(x, 24) for x in display_matrix.index]
    display_matrix.columns = [_publication_label(x, 24) for x in display_matrix.columns]

    plt.figure(figsize=(fig_w, fig_h))
    ax = sns.heatmap(
        display_matrix,
        annot=True,
        fmt=".2f",
        annot_kws={"fontsize": annot_fs, "ha": "center", "va": "center"},
        cmap="RdBu_r",
        vmin=-1,
        vmax=1,
        center=0,
        square=True,
        linewidths=lw,
        linecolor="white",
        cbar_kws={"label": cbar_label, "shrink": 0.72, "pad": 0.02},
    )
    ax.set_title(title, fontsize=16 if n <= 20 else 15, fontweight="bold", pad=20)
    ax.set_xlabel("")
    ax.set_ylabel("")
    plt.xticks(rotation=45, ha="right", rotation_mode="anchor", fontsize=label_fs)
    plt.yticks(rotation=0, fontsize=label_fs)
    plt.subplots_adjust(left=0.24, right=0.93, bottom=0.26, top=0.92)
    plt.savefig(filepath, dpi=600, bbox_inches="tight", pad_inches=0.25)
    plt.show()
    plt.close()


PEARSON_TOPN_PNG = os.path.join(
    PLOT_FOLDER, f"Pearson_TOP_{TOP_N_DESCRIPTORS}_Reduced_Descriptors.png"
)
SPEARMAN_TOPN_PNG = os.path.join(
    PLOT_FOLDER, f"Spearman_TOP_{TOP_N_DESCRIPTORS}_Reduced_Descriptors.png"
)
PEARSON_ALL_REDUCED_PNG = os.path.join(
    PLOT_FOLDER, "Pearson_All_Reduced_Descriptors.png"
)
SPEARMAN_ALL_REDUCED_PNG = os.path.join(
    PLOT_FOLDER, "Spearman_All_Reduced_Descriptors.png"
)

plot_publication_correlation_matrix(
    pearson_matrix_topn,
    f"Pearson Correlation: Yield + Top {TOP_N_DESCRIPTORS} Reduced Descriptors",
    "Pearson r",
    PEARSON_TOPN_PNG,
)
plot_publication_correlation_matrix(
    spearman_matrix_topn,
    f"Spearman Correlation: Yield + Top {TOP_N_DESCRIPTORS} Reduced Descriptors",
    "Spearman ρ",
    SPEARMAN_TOPN_PNG,
)
plot_publication_correlation_matrix(
    pearson_matrix_reduced,
    "Pearson Correlation: Yield + All Final Reduced Descriptors",
    "Pearson r",
    PEARSON_ALL_REDUCED_PNG,
)
plot_publication_correlation_matrix(
    spearman_matrix_reduced,
    "Spearman Correlation: Yield + All Final Reduced Descriptors",
    "Spearman ρ",
    SPEARMAN_ALL_REDUCED_PNG,
)

# ------------------------------------------------------------
# Final Excel export.
# ------------------------------------------------------------
with pd.ExcelWriter(
    FINAL_EXCEL,
    engine="openpyxl",
    mode="a",
    if_sheet_exists="replace",
) as writer:
    descriptor_reduction_audit_df.to_excel(
        writer, sheet_name="Descriptor_Reduction_Audit", index=False
    )
    high_correlation_pairs_reduction_df.to_excel(
        writer, sheet_name="High_Correlation_Pairs_Reduced", index=False
    )
    final_high_correlation_pairs_df.to_excel(
        writer, sheet_name="Final_High_Correlation_Check", index=False
    )
    pd.DataFrame({
        "RetainedReducedDescriptor": retaineddescriptors
    }).to_excel(
        writer, sheet_name="Final_Reduced_Descriptors", index=False
    )
    topnsummary.to_excel(
        writer, sheet_name="TopN_Reduced_Correlations", index=False
    )
    pearson_matrix_topn.to_excel(writer, sheet_name="Pearson_TopN_Reduced")
    spearman_matrix_topn.to_excel(writer, sheet_name="Spearman_TopN_Reduced")
    pearson_matrix_reduced.to_excel(writer, sheet_name="Pearson_All_Reduced")
    spearman_matrix_reduced.to_excel(writer, sheet_name="Spearman_All_Reduced")

print("\nFINAL OUTPUTS")
print("Excel:", FINAL_EXCEL)
print("Top-10 Pearson:", PEARSON_TOPN_PNG)
print("Top-10 Spearman:", SPEARMAN_TOPN_PNG)
print("All-reduced Pearson:", PEARSON_ALL_REDUCED_PNG)
print("All-reduced Spearman:", SPEARMAN_ALL_REDUCED_PNG)
print("\nDone: mathematical -> chemical/physical family-aware -> statistical -> final verification.")
```

```python
# ============================================================
# CATEGORY YIELD BOXPLOTS
# Uses exact variable names from the attached full code
# ============================================================

# ------------------------------------------------------------
# 1. Create actual Pd donor-pair category
# ------------------------------------------------------------

master_df = master_df.copy()

master_df["Actual_Donor_Pair"] = np.nan

if {
    "Donor1Element",
    "Donor2Element",
}.issubset(master_df.columns):

    for index, row in master_df.iterrows():

        # L26 special N + aromatic pi-cloud coordination
        if bool(row.get("PiCoordination", False)):
            master_df.loc[
                index,
                "Actual_Donor_Pair",
            ] = "N-Cpi"

            continue

        donor1 = row.get(
            "Donor1Element",
            np.nan,
        )

        donor2 = row.get(
            "Donor2Element",
            np.nan,
        )

        if pd.isna(donor1) or pd.isna(donor2):
            continue

        donor_set = {
            str(donor1),
            str(donor2),
        }

        if donor_set == {"N"}:
            pair = "N-N"

        elif donor_set == {"N", "O"}:
            pair = "N-O"

        elif donor_set == {"N", "S"}:
            pair = "N-S"

        elif "C(pi)" in donor_set or "Cpi" in donor_set:
            pair = "N-Cpi"

        else:
            pair = "-".join(sorted(donor_set))

        master_df.loc[
            index,
            "Actual_Donor_Pair",
        ] = pair


# ------------------------------------------------------------
# 2. Helper function: prepare plot data
# ------------------------------------------------------------

def prepare_category_yield_data(
    dataframe,
    category_column,
):
    required_columns = {
        "Ligand",
        "Yield",
        category_column,
    }

    missing_columns = (
        required_columns.difference(
            dataframe.columns
        )
    )

    if missing_columns:
        print(
            f"Skipping {category_column}; missing: "
            f"{sorted(missing_columns)}"
        )

        return pd.DataFrame()

    plot_df = dataframe[
        [
            "Ligand",
            category_column,
            "Yield",
        ]
    ].copy()

    plot_df["Yield"] = pd.to_numeric(
        plot_df["Yield"],
        errors="coerce",
    )

    plot_df = plot_df.dropna(
        subset=[
            category_column,
            "Yield",
        ]
    )

    plot_df[category_column] = (
        plot_df[category_column]
        .astype(str)
        .str.strip()
    )

    invalid_values = {
        "",
        "NAN",
        "NONE",
        "UNKNOWN",
        "UPDATE_ME",
        "UPDATEME",
    }

    invalid_mask = (
        plot_df[category_column]
        .str.upper()
        .isin(invalid_values)
    )

    update_mask = (
        plot_df[category_column]
        .str.upper()
        .str.contains(
            "UPDATE",
            na=False,
        )
    )

    plot_df = plot_df[
        ~(invalid_mask | update_mask)
    ].reset_index(drop=True)

    return plot_df


# ------------------------------------------------------------
# 3. Helper function: boxplot with ligand labels
# ------------------------------------------------------------

def create_category_yield_boxplot(
    dataframe,
    category_column,
    filename,
    title,
    xlabel,
    category_order,
):
    plot_df = prepare_category_yield_data(
        dataframe,
        category_column,
    )

    if len(plot_df) == 0:
        print(
            f"No valid data for {category_column}."
        )

        return plot_df

    populated_categories = (
        plot_df[category_column]
        .unique()
        .tolist()
    )

    if len(populated_categories) < 2:
        print(
            f"Skipping {category_column}: "
            "fewer than two populated categories."
        )

        return plot_df

    category_order = [
        category
        for category in category_order
        if category in populated_categories
    ]

    additional_categories = [
        category
        for category in sorted(populated_categories)
        if category not in category_order
    ]

    category_order.extend(
        additional_categories
    )

    plt.figure(figsize=(9, 6))

    sns.boxplot(
        data=plot_df,
        x=category_column,
        y="Yield",
        order=category_order,
        color="white",
        width=0.55,
        linewidth=1.3,
        fliersize=0,
    )

    sns.stripplot(
        data=plot_df,
        x=category_column,
        y="Yield",
        order=category_order,
        hue=category_column,
        palette="deep",
        jitter=0.12,
        dodge=False,
        size=8,
        alpha=0.85,
        linewidth=0.7,
        edgecolor="black",
        legend=False,
    )

    x_positions = {
        category: position
        for position, category in enumerate(
            category_order
        )
    }

    yield_min = float(
        plot_df["Yield"].min()
    )

    yield_max = float(
        plot_df["Yield"].max()
    )

    yield_range = yield_max - yield_min

    if yield_range == 0:
        yield_range = 10.0

    label_offset = 0.035 * yield_range

    for _, row in plot_df.iterrows():
        category = row[category_column]

        if category not in x_positions:
            continue

        ligand_number = ligand_sort_number(
            row["Ligand"]
        )

        x_offset = (
            (ligand_number % 5) - 2
        ) * 0.025

        plt.text(
            x_positions[category] + x_offset,
            row["Yield"] + label_offset,
            str(row["Ligand"]),
            ha="center",
            va="bottom",
            fontsize=8,
        )

    plt.xlabel(xlabel)
    plt.ylabel("γ-arylated lactonization yield")
    plt.title(
        title,
        fontsize=14,
        fontweight="bold",
    )

    plt.grid(
        axis="y",
        alpha=0.25,
    )

    plt.tight_layout()

    output_path = os.path.join(
        BOXPLOT_FOLDER,
        filename,
    )

    plt.savefig(
        output_path,
        dpi=600,
        bbox_inches="tight",
        facecolor="white",
    )

    plt.show()
    plt.close()

    print("Saved:", output_path)

    return plot_df


# ------------------------------------------------------------
# 4. Helper function: summary statistics
# ------------------------------------------------------------

def make_category_summary(
    plot_df,
    category_column,
):
    if plot_df is None or len(plot_df) == 0:
        return pd.DataFrame()

    return (
        plot_df
        .groupby(category_column)["Yield"]
        .agg(
            N="count",
            Mean="mean",
            Median="median",
            SD="std",
            Minimum="min",
            Maximum="max",
        )
        .reset_index()
        .sort_values(
            "Mean",
            ascending=False,
        )
        .reset_index(drop=True)
    )


# ============================================================
# 5. Yield by ligand class
# ============================================================

ligand_class_plot_df = (
    create_category_yield_boxplot(
        dataframe=master_df,
        category_column="LigandClass",
        filename="Yield_by_LigandClass.png",
        title="γ-arylated lactonization lactonization Yield by Ligand Class",
        xlabel="Ligand class",
        category_order=[
            "NN",
            "NO",
            "NS",
            "NPI",
        ],
    )
)

ligand_class_summary_df = make_category_summary(
    ligand_class_plot_df,
    "LigandClass",
)


# ============================================================
# 6. Yield by S-substituent class
# ============================================================

ns_master_df = master_df[
    master_df["LigandClass"].eq("NS")
].copy()

s_substituent_plot_df = (
    create_category_yield_boxplot(
        dataframe=ns_master_df,
        category_column="S_Substituent_Class",
        filename="Yield_by_S_Substituent_Informative.png",
        title=(
            "γ-arylated lactonization Yield by Coordinating "
            "S-Substituent"
        ),
        xlabel="S-substituent class",
        category_order=[
            "S-Me",
            "S-Et",
            "S-Pr",
            "S-iPr",
            "S-cHex",
            "S-Aryl",
            "Other",
        ],
    )
)

s_substituent_summary_df = make_category_summary(
    s_substituent_plot_df,
    "S_Substituent_Class",
)


# ============================================================
# 7. Yield by actual Pd donor pair
# ============================================================

donor_pair_plot_df = (
    create_category_yield_boxplot(
        dataframe=master_df,
        category_column="Actual_Donor_Pair",
        filename="Yield_by_Actual_Pd_Donor_Pair.png",
        title=(
            "γ-arylated lactonization  Yield by Actual Pd Donor Pair"
        ),
        xlabel="Actual Pd donor pair",
        category_order=[
            "N-N",
            "N-O",
            "N-S",
            "N-Cpi",
        ],
    )
)

donor_pair_summary_df = make_category_summary(
    donor_pair_plot_df,
    "Actual_Donor_Pair",
)


# ============================================================
# 8. Yield by donor-selection QC
# ============================================================

donor_qc_plot_df = (
    create_category_yield_boxplot(
        dataframe=master_df,
        category_column="DonorSelectionCheck",
        filename="Yield_by_Donor_Selection_QC.png",
        title=(
            "γ-arylated lactonization  Yield by Donor-Selection QC"
        ),
        xlabel="Donor-selection QC",
        category_order=[
            "PASS",
            "PASS_N_PI",
            "AUTO_DETECTED",
            "CHECK_EXPECTED_DONOR_MISMATCH",
            "ONLY_ONE_QUALIFYING_DONOR",
            "NO_QUALIFYING_NOS_DONORS",
            "L26_NO_PI_CARBON_CANDIDATE",
        ],
    )
)

donor_qc_summary_df = make_category_summary(
    donor_qc_plot_df,
    "DonorSelectionCheck",
)


# ============================================================
# 9. Display summaries
# ============================================================

print("=" * 100)
print("YIELD BY LIGAND CLASS")
print("=" * 100)
display(ligand_class_summary_df)

print("=" * 100)
print("YIELD BY S-SUBSTITUENT")
print("=" * 100)
display(s_substituent_summary_df)

print("=" * 100)
print("YIELD BY ACTUAL Pd DONOR PAIR")
print("=" * 100)
display(donor_pair_summary_df)

print("=" * 100)
print("YIELD BY DONOR-SELECTION QC")
print("=" * 100)
display(donor_qc_summary_df)
```

```python
# ============================================================
# Nat_Syn-DM: Pd γ-ARYLATION
# COMPLETE LIGAND + S-SUBSTITUENT DESCRIPTOR ANALYSIS
#
# Required XYZ atom numbering:
#   Atom 1       = Pd
#   Atoms 2-19   = substrate
#   Atoms 20-end = ligand
#
# Ligand classes:
#   L10-L19 = expected NN/NO ligands
#   L20-L27 = expected NS ligands
#
# Pd coordinating donor rules:
#   N/O: 2.00-2.30 Å
#   S:   2.40-2.65 Å
#   H is NEVER considered a Pd donor.
# ============================================================


# ============================================================
# 1. INSTALL REQUIRED PACKAGES
# ============================================================

!pip -q install morfeus-ml pandas numpy scipy openpyxl matplotlib seaborn statsmodels


# ============================================================
# 2. IMPORTS
# ============================================================

import os
import re
import glob
import math
import shutil
import warnings
import itertools
from collections import deque

import numpy as np
import pandas as pd

import matplotlib.pyplot as plt
import seaborn as sns

from scipy.stats import pearsonr, spearmanr

warnings.filterwarnings("ignore")

pd.set_option("display.max_columns", 200)
pd.set_option("display.width", 200)


# ============================================================
# 3. MOUNT GOOGLE DRIVE
# ============================================================

from google.colab import drive
drive.mount("/content/drive")


# ============================================================
# 4. USER SETTINGS
# ============================================================

# ------------------------------------------------------------
# Input and output folders
# ------------------------------------------------------------

XYZ_FOLDER = "/content/drive/MyDrive/Nat_Syn-DM"

OUTPUT_FOLDER = (
    "/content/drive/MyDrive/"
    "Nat_Syn_DM_Gamma_Arylation_Results"
)

os.makedirs(OUTPUT_FOLDER, exist_ok=True)

PLOT_FOLDER = os.path.join(OUTPUT_FOLDER, "Plots")
SCATTER_FOLDER = os.path.join(PLOT_FOLDER, "Yield_vs_Descriptors")
BOXPLOT_FOLDER = os.path.join(PLOT_FOLDER, "Descriptor_Boxplots")

os.makedirs(PLOT_FOLDER, exist_ok=True)
os.makedirs(SCATTER_FOLDER, exist_ok=True)
os.makedirs(BOXPLOT_FOLDER, exist_ok=True)

FINAL_EXCEL = os.path.join(
    OUTPUT_FOLDER,
    "Nat_Syn_DM_Gamma_Arylation_Descriptor_Analysis.xlsx"
)

MASTER_CSV = os.path.join(
    OUTPUT_FOLDER,
    "Nat_Syn_DM_All_Descriptors.csv"
)

YIELD_CORRELATION_CSV = os.path.join(
    OUTPUT_FOLDER,
    "Nat_Syn_DM_Yield_Correlations.csv"
)


# ------------------------------------------------------------
# Required chemical atom numbering
# ------------------------------------------------------------

PD_ATOM = 1

SUBSTRATE_START = 2
SUBSTRATE_END = 19

LIGAND_START = 20


# ------------------------------------------------------------
# Reaction information
# ------------------------------------------------------------

REACTION_TYPE = "Nat_Syn-Gamma_Arylation"


# ------------------------------------------------------------
# Pd donor identification
#
# IMPORTANT:
# Hydrogen is completely excluded.
# Only N, O, and S may be Pd donor candidates.
# ------------------------------------------------------------

# ============================================================
# Pd-donor distance windows in Å
#
# STRICT:
# Primary coordination-distance criteria.
#
# EXTENDED:
# Used to retain chemically plausible elongated donor contacts
# for chelated NN, NO, and NS ligand structures.
# ============================================================

DONOR_ELEMENTS = {"N", "O", "S"}

PD_DONOR_WINDOWS_STRICT = {
    "N": (2.00, 2.30),
    "O": (2.00, 2.30),
    "S": (2.40, 2.65),
}

PD_DONOR_WINDOWS_EXTENDED = {
    "N": (1.90, 2.45),
    "O": (1.90, 2.50),
    "S": (2.20, 2.80),
}


# ------------------------------------------------------------
# Ligand class assignment
# ------------------------------------------------------------

# ============================================================
# LIGAND CLASS ASSIGNMENT
# ============================================================

LIGAND_CLASSES = {
    # NN ligands: L10, L12, L13, L14, L18, L19
    **{
        f"L{i}": "NN"
        for i in [10, 12, 13, 14, 18, 19]
    },

    # NO ligands: L11, L15, L16, L17
    **{
        f"L{i}": "NO"
        for i in [11, 15, 16, 17]
    },

    # NS ligands: L20-L25
    **{
        f"L{i}": "NS"
        for i in range(20, 26)
    },

    # L26: N + aromatic pi-cloud coordination
    "L26": "NPI",
}

EXPECTED_DONOR_PAIRS = {
    "NN": ["N", "N"],
    "NO": ["N", "O"],
    "NS": ["N", "S"],
    "NPI": ["N", "C(pi)"],
}

# L26 pi-coordination geometry: the aromatic carbon is used as a
# geometric representative of the Pd interaction with the pi cloud.
L26_PI_C_MIN_DISTANCE_A = 2.00
L26_PI_C_MAX_DISTANCE_A = 3.00


# ------------------------------------------------------------
# S-substituent assignment for NS ligands
#
# USER ACTION REQUIRED:
# Update this mapping based on your ligand structures.
#
# Available labels suggested:
#   "S-Me"
#   "S-Et"
#   "S-Pr"
#   "S-iPr"
#   "S-cHex"
#   "S-Aryl"
#   "Other"
#
# This column enables direct comparison of S-substituent identity
# versus reaction yield.
# ------------------------------------------------------------

S_SUBSTITUENT_CLASS = {
    "L20": "UPDATE_ME",
    "L21": "UPDATE_ME",
    "L22": "UPDATE_ME",
    "L23": "UPDATE_ME",
    "L24": "UPDATE_ME",
    "L25": "UPDATE_ME",
    "L26": "UPDATE_ME",
}


# ------------------------------------------------------------
# Yield data
#
# USER ACTION REQUIRED:
# Update values here.
#
# Use numerical yields, normally percentages.
# Examples:
#   "L10": 42,
#   "L11": 71.5,
#   "L20": 88,
#
# Leave np.nan if yield is not available.
# ------------------------------------------------------------

YIELD_DATA = {
    "L10": 0,
    "L11": 1,
    "L12": 7,
    "L13": 0,
    "L14": 0,
    "L15": 0,
    "L16": 0,
    "L17": 0,
    "L18": 0,
    "L19": 25,
    "L20": 70,
    "L21": 64,
    "L22": 57,
    "L23": 55,
    "L24": 50,
    "L25": 9,
    "L26": 0,
}


# ------------------------------------------------------------
# Morfeus settings
# ------------------------------------------------------------

SASA_PROBE_RADIUS = 1.4

VBUR_RADII = [3.0, 4.0, 5.5]
DISTAL_VOLUME_RADIUS = 3.5

# Morfeus BuriedVolume uses density, not npoints.
# Smaller density = more grid points / higher precision.
VBUR_DENSITY = 0.0008

VBUR_RADII_TYPE = "bondi"
SOLID_ANGLE_RADII_TYPE = "bondi"
STERIMOL_RADII_TYPE = "crc"

CALCULATE_DISPERSION = True


# ------------------------------------------------------------
# xTB settings
#
# Set RUN_XTB = False if xTB is unavailable or if you want
# only geometry and steric descriptors.
# ------------------------------------------------------------

RUN_XTB = True

XTB_METHOD = 2
XTB_CHARGE = 0
XTB_NUNPAIRED = 0
XTB_PROCESSES = 2


# ------------------------------------------------------------
# Statistical settings
# ------------------------------------------------------------

MINIMUM_CORRELATION_N = 5
HIGH_CORRELATION_THRESHOLD = 0.80

OUTLIER_FRACTION_LIMIT = 0.10
IQR_MULTIPLIER = 1.5
MAD_Z_LIMIT = 3.5


# ============================================================
# 5. xTB INSTALLATION AND MORFEUS IMPORT
# ============================================================

# Morfeus XTB must only be imported after xTB is available.
# If xTB is missing, this cell installs a standard Linux xTB binary.

if RUN_XTB:
    xtb_path = shutil.which("xtb")

    if xtb_path is None:
        print("xTB not found. Installing xTB...")

        xtb_archive = "/content/xtb.tar.xz"

        !wget -q --show-progress \
            https://github.com/grimme-lab/xtb/releases/latest/download/xtb-6.7.1-linux-x86_64.tar.xz \
            -O /content/xtb.tar.xz

        !tar -xf /content/xtb.tar.xz -C /content

        xtb_dirs = sorted(
            [
                os.path.join("/content", folder)
                for folder in os.listdir("/content")
                if folder.startswith("xtb-")
                and os.path.isdir(os.path.join("/content", folder))
            ]
        )

        if len(xtb_dirs) == 0:
            raise RuntimeError(
                "xTB archive was extracted, but xTB directory was not found."
            )

        xtb_dir = xtb_dirs[-1]
        xtb_bin = os.path.join(xtb_dir, "bin")
        xtb_executable = os.path.join(xtb_bin, "xtb")

        if not os.path.exists(xtb_executable):
            raise RuntimeError(
                f"xTB executable not found: {xtb_executable}"
            )

        os.environ["PATH"] = xtb_bin + ":" + os.environ["PATH"]
        os.environ["XTBHOME"] = xtb_dir

    xtb_path = shutil.which("xtb")

    if xtb_path is None:
        raise RuntimeError("xTB is still unavailable after installation.")

    print("xTB executable:", xtb_path)

    !xtb --version


# Import Morfeus only after xTB path is available.
from morfeus import (
    SASA,
    SolidAngle,
    BuriedVolume,
    Sterimol,
    Pyramidalization,
)

if RUN_XTB:
    from morfeus import XTB

if CALCULATE_DISPERSION:
    try:
        from morfeus import Dispersion
        DISPERSION_AVAILABLE = True
    except Exception:
        DISPERSION_AVAILABLE = False
else:
    DISPERSION_AVAILABLE = False


# ============================================================
# 6. GENERAL HELPER FUNCTIONS
# ============================================================

def finite_value(value):
    """Convert valid finite values to float; otherwise return NaN."""
    try:
        numeric_value = float(value)
        if np.isfinite(numeric_value):
            return numeric_value
    except Exception:
        pass

    return np.nan


def normalize_element(element):
    """Normalize atomic symbols read from XYZ files."""
    return str(element).strip().capitalize()


def distance_a(coord_a, coord_b):
    """Return Cartesian distance in Å."""
    return float(
        np.linalg.norm(
            np.asarray(coord_a, dtype=float)
            - np.asarray(coord_b, dtype=float)
        )
    )


def angle_deg(coord_a, coord_b, coord_c):
    """
    Return angle A-B-C in degrees.
    The central atom is coord_b.
    """
    vector_1 = np.asarray(coord_a, dtype=float) - np.asarray(coord_b, dtype=float)
    vector_2 = np.asarray(coord_c, dtype=float) - np.asarray(coord_b, dtype=float)

    norm_1 = np.linalg.norm(vector_1)
    norm_2 = np.linalg.norm(vector_2)

    if norm_1 == 0 or norm_2 == 0:
        return np.nan

    cosine = np.dot(vector_1, vector_2) / (norm_1 * norm_2)
    cosine = np.clip(cosine, -1.0, 1.0)

    return float(np.degrees(np.arccos(cosine)))


def safe_filename(text):
    """Create safe filenames for plot output."""
    return re.sub(r"[^A-Za-z0-9_.-]+", "_", str(text))


def clean_ligand_name(filepath):
    """
    Extract ligand identity such as L10 from XYZ filename.

    Works for filenames such as:
        L10.xyz
        L10_optimized.xyz
        ligand-L10.xyz
        complex_L20_final.xyz
    """
    basename = os.path.basename(filepath)
    basename = re.sub(r"\.xyz$", "", basename, flags=re.IGNORECASE)

    match = re.search(r"(L\d+)", basename, flags=re.IGNORECASE)

    if match:
        return match.group(1).upper()

    return basename


def ligand_sort_number(ligand_name):
    """Sort L10, L11, ..., L27 numerically."""
    match = re.search(r"L(\d+)", str(ligand_name), flags=re.IGNORECASE)

    if match:
        return int(match.group(1))

    return 999999


def classify_ligand(ligand_name):
    """
    Return manually assigned ligand class:
    NN, NO, NS, or UNKNOWN.
    """
    ligand_name = str(ligand_name).upper()

    return LIGAND_CLASSES.get(
        ligand_name,
        "UNKNOWN",
    )


def expected_donor_pair(ligand_name):
    """Return expected donor elements as a list, for example ['N', 'O' 'S']."""
    ligand_class = classify_ligand(ligand_name)
    return EXPECTED_DONOR_PAIRS.get(ligand_class)


def ligand_indices(elements):
    """
    Return original 1-based XYZ indices for ligand atoms:
    atoms 20 through the final atom.
    """
    return list(range(LIGAND_START, len(elements) + 1))


def substrate_indices(elements):
    """Return original 1-based XYZ indices for substrate atoms 2-19."""
    maximum_atom = min(SUBSTRATE_END, len(elements))
    return list(range(SUBSTRATE_START, maximum_atom + 1))


# ============================================================
# 7. XYZ READER
# ============================================================

def read_xyz_file(filepath):
    """Read a standard XYZ file."""
    with open(filepath, "r", encoding="utf-8") as handle:
        lines = [line.strip() for line in handle if line.strip()]

    if len(lines) < 3:
        raise ValueError(f"XYZ file is too short: {filepath}")

    number_of_atoms = int(lines[0].split()[0])
    atom_lines = lines[2:2 + number_of_atoms]

    if len(atom_lines) < number_of_atoms:
        raise ValueError(
            f"XYZ coordinate lines are fewer than specified atom count in {filepath}"
        )

    elements = []
    coordinates = []

    for line in atom_lines:
        parts = line.split()

        if len(parts) < 4:
            raise ValueError(f"Malformed XYZ atom line: {line}")

        elements.append(normalize_element(parts[0]))
        coordinates.append(
            [
                float(parts[1]),
                float(parts[2]),
                float(parts[3]),
            ]
        )

    return elements, np.asarray(coordinates, dtype=float)


# ============================================================
# 8. STRUCTURE VALIDATION
# ============================================================

def validate_structure(ligand_name, elements, coordinates):
    """
    Validate required Nat_Syn-DM numbering and expected ligand content.
    """
    warnings_list = []

    if len(elements) != len(coordinates):
        warnings_list.append(
            "Element count and coordinate count are not identical."
        )

    if len(elements) < LIGAND_START:
        warnings_list.append(
            f"Structure has fewer than {LIGAND_START} atoms."
        )

    if len(elements) >= 1:
        if normalize_element(elements[PD_ATOM - 1]) != "Pd":
            warnings_list.append("Atom 1 is not Pd.")

    if len(elements) >= SUBSTRATE_END:
        substrate_atom_count = len(
            range(SUBSTRATE_START, SUBSTRATE_END + 1)
        )

        if substrate_atom_count != 18:
            warnings_list.append(
                "Substrate atom range 2-19 is not 18 atoms."
            )

    if not np.isfinite(coordinates).all():
        warnings_list.append("XYZ contains non-finite coordinates.")

    ligand_atom_ids = ligand_indices(elements)

    if len(ligand_atom_ids) == 0:
        warnings_list.append("No ligand atoms found from atom 20 onward.")
        return warnings_list

    ligand_elements = [
        normalize_element(elements[atom_id - 1])
        for atom_id in ligand_atom_ids
    ]

    ligand_class = classify_ligand(ligand_name)

    # --------------------------------------------------------
    # FIX: this block previously mixed tabs and spaces and had
    # the if/elif statements at inconsistent indentation levels,
    # which raised a TabError / IndentationError. Rewritten here
    # with consistent 4-space indentation.
    # --------------------------------------------------------

    if ligand_class == "NN":
        if ligand_elements.count("N") < 2:
            warnings_list.append(
                "Expected NN ligand but fewer than two ligand N atoms were found."
            )

    elif ligand_class == "NO":
        if "N" not in ligand_elements:
            warnings_list.append(
                "Expected NO ligand but no ligand N atom was found."
            )

        if "O" not in ligand_elements:
            warnings_list.append(
                "Expected NO ligand but no ligand O atom was found."
            )

    elif ligand_class == "NS":
        if "N" not in ligand_elements:
            warnings_list.append(
                "Expected NS ligand but no ligand N atom was found."
            )

        if "S" not in ligand_elements:
            warnings_list.append(
                "Expected NS ligand but no ligand S atom was found."
            )

    return warnings_list


# ============================================================
# 9. COVALENT CONNECTIVITY
# ============================================================

COVALENT_RADII = {
    "H": 0.31,
    "C": 0.76,
    "N": 0.71,
    "O": 0.66,
    "F": 0.57,
    "P": 1.07,
    "S": 1.05,
    "Cl": 1.02,
    "Br": 1.20,
    "I": 1.39,
    "Pd": 1.39,
}


def covalent_neighbors(
    elements,
    coordinates,
    atom_id,
    allowed_indices=None,
    scale=1.25,
):
    """
    Identify covalent neighbors from XYZ geometry and covalent radii.

    atom_id is 1-based.
    """
    if allowed_indices is None:
        allowed_indices = list(range(1, len(elements) + 1))

    atom_element = normalize_element(elements[atom_id - 1])

    if atom_element not in COVALENT_RADII:
        return []

    atom_radius = COVALENT_RADII[atom_element]
    atom_coord = coordinates[atom_id - 1]

    neighbors = []

    for other_id in allowed_indices:
        if other_id == atom_id:
            continue

        other_element = normalize_element(elements[other_id - 1])

        if other_element not in COVALENT_RADII:
            continue

        other_radius = COVALENT_RADII[other_element]
        other_coord = coordinates[other_id - 1]

        current_distance = distance_a(atom_coord, other_coord)
        cutoff_distance = scale * (atom_radius + other_radius)

        if current_distance <= cutoff_distance:
            neighbors.append((other_id, current_distance))

    return sorted(neighbors, key=lambda item: item[1])


def build_ligand_graph(elements, coordinates):
    """
    Build an undirected covalent graph only for ligand atoms 20-end.
    Pd and substrate atoms are excluded.
    """
    ligand_atom_ids = ligand_indices(elements)

    graph = {
        atom_id: set()
        for atom_id in ligand_atom_ids
    }

    for atom_id in ligand_atom_ids:
        neighbors = covalent_neighbors(
            elements=elements,
            coordinates=coordinates,
            atom_id=atom_id,
            allowed_indices=ligand_atom_ids,
        )

        for neighbor_id, _ in neighbors:
            graph[atom_id].add(neighbor_id)
            graph[neighbor_id].add(atom_id)

    return graph


def shortest_graph_path(graph, start_atom, end_atom):
    """Return shortest atom path in an undirected graph."""
    if start_atom is None or end_atom is None:
        return None

    if start_atom not in graph or end_atom not in graph:
        return None

    queue = deque([(start_atom, [start_atom])])
    visited = {start_atom}

    while queue:
        current_atom, path = queue.popleft()

        if current_atom == end_atom:
            return path

        for neighbor in sorted(graph.get(current_atom, [])):
            if neighbor not in visited:
                visited.add(neighbor)
                queue.append((neighbor, path + [neighbor]))

    return None


def connected_component(graph, start_atom):
    """Return all atoms in the graph component containing start_atom."""
    if start_atom is None or start_atom not in graph:
        return set()

    visited = {start_atom}
    queue = deque([start_atom])

    while queue:
        current_atom = queue.popleft()

        for neighbor in graph.get(current_atom, set()):
            if neighbor not in visited:
                visited.add(neighbor)
                queue.append(neighbor)

    return visited


# ============================================================
# 10. Pd DONOR IDENTIFICATION
# ============================================================

def find_pd_donor_candidates(elements, coordinates):
    """
    Find ligand N/O/S atoms close to Pd.

    H is never considered.

    Each candidate receives:
    - StrictQualifies: within the preferred Pd-X range.
    - ExtendedQualifies: within an extended plausible Pd-X range.

    Donor selection should use ExtendedQualifies so that an
    elongated but structurally intended chelating contact is not
    discarded. The StrictQualifies column remains as QC information.
    """
    pd_coordinate = coordinates[PD_ATOM - 1]
    all_candidates = []

    for atom_id in ligand_indices(elements):
        element = normalize_element(elements[atom_id - 1])

        if element not in {"N", "O", "S"}:
            continue

        strict_min, strict_max = PD_DONOR_WINDOWS_STRICT[element]
        extended_min, extended_max = PD_DONOR_WINDOWS_EXTENDED[element]

        pd_distance = distance_a(
            pd_coordinate,
            coordinates[atom_id - 1],
        )

        all_candidates.append({
            "Atom": atom_id,
            "Element": element,
            "PdDistance_A": pd_distance,

            "StrictWindowMin_A": strict_min,
            "StrictWindowMax_A": strict_max,

            "ExtendedWindowMin_A": extended_min,
            "ExtendedWindowMax_A": extended_max,

            "StrictQualifies": (
                strict_min <= pd_distance <= strict_max
            ),

            "ExtendedQualifies": (
                extended_min <= pd_distance <= extended_max
            ),
        })

    all_candidates = sorted(
        all_candidates,
        key=lambda item: item["PdDistance_A"],
    )

    qualifying_candidates = [
        item
        for item in all_candidates
        if item["ExtendedQualifies"]
    ]

    return all_candidates, qualifying_candidates


def find_l26_pi_carbon_candidate(elements, coordinates):
    """
    Identify ligand carbon atoms that can represent Pd interaction
    with the aromatic pi cloud in L26.

    Candidate carbons are ligand C atoms with at least two ligand-C
    covalent neighbours and Pd-C distances of 2.00-3.00 A. The
    nearest qualifying carbon is selected.
    """
    pd_coordinate = coordinates[PD_ATOM - 1]
    ligand_atom_ids = ligand_indices(elements)
    candidates = []

    for atom_id in ligand_atom_ids:
        if normalize_element(elements[atom_id - 1]) != "C":
            continue

        carbon_neighbors = [
            neighbour_id
            for neighbour_id, _ in covalent_neighbors(
                elements, coordinates, atom_id, ligand_atom_ids
            )
            if normalize_element(elements[neighbour_id - 1]) == "C"
        ]

        if len(carbon_neighbors) < 2:
            continue

        pd_c_distance = distance_a(
            pd_coordinate, coordinates[atom_id - 1]
        )

        if L26_PI_C_MIN_DISTANCE_A <= pd_c_distance <= L26_PI_C_MAX_DISTANCE_A:
            candidates.append({
                "Atom": atom_id,
                "Element": "C(pi)",
                "PdDistance_A": pd_c_distance,
                "CarbonNeighborCount": len(carbon_neighbors),
                "CarbonNeighbors": carbon_neighbors,
            })

    return sorted(candidates, key=lambda item: item["PdDistance_A"])


def select_pd_donors(ligand_name, qualifying_candidates, elements=None, coordinates=None):
    """Select the two Pd-coordination representatives for each ligand.

    Standard ligands use the expected N/N, N/O, or N/S donor pair.
    L26 is treated separately as N + aromatic C(pi): N is the direct
    donor and the nearest qualifying aromatic carbon represents the
    Pd interaction with the aromatic pi cloud. Sulfur is not treated
    as a coordinating donor for L26.
    """
    ligand_class = classify_ligand(ligand_name)
    expected_pair = expected_donor_pair(ligand_name)

    qualifying_candidates = sorted(
        qualifying_candidates, key=lambda item: item["PdDistance_A"]
    )

    result = {
        "LigandClass": ligand_class,
        "ExpectedDonorPair": (
            "-".join(expected_pair) if expected_pair is not None else "UNKNOWN"
        ),
        "Donor1Atom": np.nan,
        "Donor1Element": None,
        "Pd_Donor1_Distance_A": np.nan,
        "Donor2Atom": np.nan,
        "Donor2Element": None,
        "Pd_Donor2_Distance_A": np.nan,
        "DonorSelectionCheck": None,
        "PiCoordination": False,
        "PiCoordinationAtom": np.nan,
        "PiCoordinationPd_C_Distance_A": np.nan,
        "PiCoordinationDescription": None,
        "L26_PiCarbonCandidates": None,
    }

    # --------------------------------------------------------
    # Special L26: N + aromatic C(pi)
    # --------------------------------------------------------
    if ligand_class == "NPI":
        if elements is None or coordinates is None:
            result["DonorSelectionCheck"] = "L26_COORDINATES_REQUIRED"
            return result, None, None

        n_candidates = [
            item for item in qualifying_candidates
            if item["Element"] == "N"
        ]
        pi_candidates = find_l26_pi_carbon_candidate(elements, coordinates)
        result["L26_PiCarbonCandidates"] = pi_candidates

        if not n_candidates:
            result["DonorSelectionCheck"] = "L26_NO_QUALIFYING_N_DONOR"
            return result, None, None
        if not pi_candidates:
            result["DonorSelectionCheck"] = "L26_NO_PI_CARBON_CANDIDATE"
            return result, None, None

        donor1_item = n_candidates[0]
        donor2_item = pi_candidates[0]

        result["Donor1Atom"] = donor1_item["Atom"]
        result["Donor1Element"] = "N"
        result["Pd_Donor1_Distance_A"] = donor1_item["PdDistance_A"]
        result["Donor2Atom"] = donor2_item["Atom"]
        result["Donor2Element"] = "C(pi)"
        result["Pd_Donor2_Distance_A"] = donor2_item["PdDistance_A"]
        result["PiCoordination"] = True
        result["PiCoordinationAtom"] = donor2_item["Atom"]
        result["PiCoordinationPd_C_Distance_A"] = donor2_item["PdDistance_A"]
        result["PiCoordinationDescription"] = (
            "Pd interacts with the aromatic pi cloud; the nearest "
            "qualifying aromatic C is used as the geometric representative. "
            "Sulfur is distal/non-coordinating in L26."
        )
        result["DonorSelectionCheck"] = "PASS_N_PI"
        return result, donor1_item["Atom"], donor2_item["Atom"]

    # --------------------------------------------------------
    # Standard N/N, N/O, or N/S selection
    # --------------------------------------------------------
    result["L26_PiCarbonCandidates"] = None

    if len(qualifying_candidates) == 0:
        result["DonorSelectionCheck"] = "NO_QUALIFYING_NOS_DONORS"
        return result, None, None

    selected_pair = None
    if expected_pair is not None and len(qualifying_candidates) >= 2:
        expected_sorted = sorted(expected_pair)
        for i in range(len(qualifying_candidates)):
            for j in range(i + 1, len(qualifying_candidates)):
                pair = [qualifying_candidates[i], qualifying_candidates[j]]
                if sorted([item["Element"] for item in pair]) == expected_sorted:
                    selected_pair = pair
                    break
            if selected_pair is not None:
                break

    if selected_pair is None and len(qualifying_candidates) >= 2:
        selected_pair = qualifying_candidates[:2]
    if selected_pair is None:
        selected_pair = qualifying_candidates[:1]

    selected_pair = sorted(selected_pair, key=lambda item: item["PdDistance_A"])
    donor1 = selected_pair[0]["Atom"]
    donor2 = selected_pair[1]["Atom"] if len(selected_pair) == 2 else None

    result["Donor1Atom"] = donor1
    result["Donor1Element"] = selected_pair[0]["Element"]
    result["Pd_Donor1_Distance_A"] = selected_pair[0]["PdDistance_A"]

    if donor2 is not None:
        result["Donor2Atom"] = donor2
        result["Donor2Element"] = selected_pair[1]["Element"]
        result["Pd_Donor2_Distance_A"] = selected_pair[1]["PdDistance_A"]

    selected_elements = sorted([item["Element"] for item in selected_pair])
    if donor2 is None:
        result["DonorSelectionCheck"] = "ONLY_ONE_QUALIFYING_DONOR"
    elif expected_pair is None:
        result["DonorSelectionCheck"] = "AUTO_DETECTED"
    elif selected_elements == sorted(expected_pair):
        result["DonorSelectionCheck"] = "PASS"
    else:
        result["DonorSelectionCheck"] = "CHECK_EXPECTED_DONOR_MISMATCH"

    return result, donor1, donor2

# ============================================================
# 11. LIGAND DONOR-PATH DESCRIPTORS
# ============================================================

def calculate_ligand_donor_path(
    elements,
    coordinates,
    donor1,
    donor2,
):
    """
    Calculate covalent shortest-path descriptors between donor atoms.

    These are ligand backbone descriptors, not beta-C-H
    palladacycle descriptors.
    """
    result = {
        "LigandDonorPath": None,
        "LigandDonorPathAtomCount": np.nan,
        "LigandDonorPathBondCount": np.nan,
        "LigandDonorPathMeanBond_A": np.nan,
        "LigandDonorPathSD_A": np.nan,
        "LigandDonorPathCV": np.nan,
    }

    if donor1 is None or donor2 is None:
        return result

    graph = build_ligand_graph(elements, coordinates)
    path = shortest_graph_path(graph, donor1, donor2)

    if path is None:
        return result

    result["LigandDonorPath"] = "-".join(str(atom) for atom in path)
    result["LigandDonorPathAtomCount"] = len(path)
    result["LigandDonorPathBondCount"] = len(path) - 1

    bond_lengths = []

    for index in range(len(path) - 1):
        atom_i = path[index]
        atom_j = path[index + 1]

        bond_lengths.append(
            distance_a(
                coordinates[atom_i - 1],
                coordinates[atom_j - 1],
            )
        )

    if len(bond_lengths) > 0:
        mean_bond_length = float(np.mean(bond_lengths))
        sd_bond_length = float(np.std(bond_lengths, ddof=0))

        result["LigandDonorPathMeanBond_A"] = mean_bond_length
        result["LigandDonorPathSD_A"] = sd_bond_length

        if mean_bond_length > 0:
            result["LigandDonorPathCV"] = (
                sd_bond_length / mean_bond_length
            )

    return result


# ============================================================
# 12. S-SUBSTITUENT STRUCTURAL / STERIC DESCRIPTORS
# ============================================================

def identify_s_substituent_atoms(
    elements,
    coordinates,
    sulfur_atom,
    donor_partner_atom=None,
):
    """
    Identify the covalent substituent attached to a coordinating sulfur.

    Strategy:
    1. Build a ligand-only connectivity graph.
    2. Locate ligand atoms directly bonded to S.
    3. Exclude Pd because Pd is outside the ligand graph.
    4. If another coordinating donor is directly bonded to S,
       exclude it from the substituent search.
    5. For every S neighbor, identify the graph component reached
       after removing sulfur.
    6. Select the largest carbon-containing component as the
       likely S-alkyl or S-aryl substituent.

    This is designed for comparing S-Me, S-Et, S-Pr, S-cHex, etc.
    """
    if sulfur_atom is None:
        return {
            "S_SubstituentRootAtom": None,
            "S_SubstituentAtoms": [],
            "S_SubstituentNeighborCandidates": [],
            "S_SubstituentWarning": "No coordinating sulfur atom available.",
        }

    if normalize_element(elements[sulfur_atom - 1]) != "S":
        return {
            "S_SubstituentRootAtom": None,
            "S_SubstituentAtoms": [],
            "S_SubstituentNeighborCandidates": [],
            "S_SubstituentWarning": "Selected sulfur atom is not S.",
        }

    graph = build_ligand_graph(elements, coordinates)

    sulfur_neighbors = sorted(graph.get(sulfur_atom, set()))

    if donor_partner_atom is not None:
        sulfur_neighbors = [
            atom_id
            for atom_id in sulfur_neighbors
            if atom_id != donor_partner_atom
        ]

    if len(sulfur_neighbors) == 0:
        return {
            "S_SubstituentRootAtom": None,
            "S_SubstituentAtoms": [],
            "S_SubstituentNeighborCandidates": [],
            "S_SubstituentWarning": (
                "No ligand covalent neighbors found for coordinating S."
            ),
        }

    # Create a graph with S removed, so each direct neighbor leads
    # to a separate branch/component.
    graph_without_s = {
        atom_id: set(neighbors)
        for atom_id, neighbors in graph.items()
        if atom_id != sulfur_atom
    }

    for atom_id in graph_without_s:
        graph_without_s[atom_id].discard(sulfur_atom)

    branch_records = []

    for neighbor_atom in sulfur_neighbors:
        component = connected_component(
            graph_without_s,
            neighbor_atom,
        )

        component_elements = [
            normalize_element(elements[atom_id - 1])
            for atom_id in component
        ]

        carbon_count = component_elements.count("C")
        heavy_atom_count = sum(
            element != "H"
            for element in component_elements
        )

        branch_records.append({
            "RootAtom": neighbor_atom,
            "RootElement": normalize_element(
                elements[neighbor_atom - 1]
            ),
            "Atoms": sorted(component),
            "CarbonCount": carbon_count,
            "HeavyAtomCount": heavy_atom_count,
            "AtomCount": len(component),
        })

    # Prefer a carbon-containing branch, then choose:
    # largest carbon count -> largest heavy atom count -> largest atom count.
    carbon_branches = [
        branch
        for branch in branch_records
        if branch["CarbonCount"] > 0
    ]

    if len(carbon_branches) > 0:
        selected_branch = sorted(
            carbon_branches,
            key=lambda branch: (
                branch["CarbonCount"],
                branch["HeavyAtomCount"],
                branch["AtomCount"],
            ),
            reverse=True,
        )[0]
    else:
        selected_branch = sorted(
            branch_records,
            key=lambda branch: (
                branch["HeavyAtomCount"],
                branch["AtomCount"],
            ),
            reverse=True,
        )[0]

    return {
        "S_SubstituentRootAtom": selected_branch["RootAtom"],
        "S_SubstituentAtoms": selected_branch["Atoms"],
        "S_SubstituentNeighborCandidates": branch_records,
        "S_SubstituentWarning": None,
    }


def calculate_s_substituent_descriptors(
    ligand_name,
    elements,
    coordinates,
    donor1,
    donor2,
):
    """
    Calculate local steric and structural descriptors for the
    substituent attached to coordinating sulfur.

    Applies to NS ligands or any ligand where one selected Pd donor is S.
    """
    result = {
        "S_Substituent_Class": S_SUBSTITUENT_CLASS.get(
            ligand_name,
            np.nan,
        ),
        "CoordinatingSAtom": np.nan,
        "CoordinatingS_Pd_Distance_A": np.nan,
        "S_SubstituentRootAtom": np.nan,
        "S_SubstituentRootElement": None,
        "S_SubstituentAtomIDs": None,
        "S_SubstituentAtomCount": np.nan,
        "S_SubstituentHeavyAtomCount": np.nan,
        "S_SubstituentCarbonCount": np.nan,
        "S_SubstituentHydrogenCount": np.nan,
        "S_Substituent_HeteroatomCount": np.nan,
        "S_C_SubstituentBond_A": np.nan,
        "S_Substituent_MaxDistanceFromS_A": np.nan,
        "S_Substituent_MeanDistanceFromS_A": np.nan,
        "S_Substituent_MaxInternalDistance_A": np.nan,
        "S_Substituent_SASA_A2": np.nan,
        "S_Substituent_SASA_Volume_A3": np.nan,
        "S_Substituent_SolidAngle_sr": np.nan,
        "S_Substituent_ConeAngle_deg": np.nan,
        "S_Substituent_G_percent": np.nan,
        "S_Substituent_%Vbur_3.0A": np.nan,
        "S_Substituent_%Vbur_4.0A": np.nan,
        "S_Substituent_%Vbur_5.5A": np.nan,
        "S_SubstituentWarning": None,
    }

    # Identify which selected donor is sulfur.
    sulfur_atom = None
    partner_donor = None

    if donor1 is not None:
        if normalize_element(elements[donor1 - 1]) == "S":
            sulfur_atom = donor1
            partner_donor = donor2

    if sulfur_atom is None and donor2 is not None:
        if normalize_element(elements[donor2 - 1]) == "S":
            sulfur_atom = donor2
            partner_donor = donor1

    if sulfur_atom is None:
        result["S_SubstituentWarning"] = (
            "No coordinating sulfur donor was selected."
        )
        return result

    result["CoordinatingSAtom"] = sulfur_atom
    result["CoordinatingS_Pd_Distance_A"] = distance_a(
        coordinates[PD_ATOM - 1],
        coordinates[sulfur_atom - 1],
    )

    substituent_info = identify_s_substituent_atoms(
        elements=elements,
        coordinates=coordinates,
        sulfur_atom=sulfur_atom,
        donor_partner_atom=partner_donor,
    )

    substituent_atom_ids = substituent_info["S_SubstituentAtoms"]
    root_atom = substituent_info["S_SubstituentRootAtom"]

    if substituent_info["S_SubstituentWarning"] is not None:
        result["S_SubstituentWarning"] = (
            substituent_info["S_SubstituentWarning"]
        )

    if root_atom is None or len(substituent_atom_ids) == 0:
        return result

    result["S_SubstituentRootAtom"] = root_atom
    result["S_SubstituentRootElement"] = normalize_element(
        elements[root_atom - 1]
    )

    result["S_SubstituentAtomIDs"] = "-".join(
        str(atom_id)
        for atom_id in substituent_atom_ids
    )

    substituent_elements = [
        normalize_element(elements[atom_id - 1])
        for atom_id in substituent_atom_ids
    ]

    substituent_coordinates = coordinates[
        np.asarray(substituent_atom_ids, dtype=int) - 1
    ]

    result["S_SubstituentAtomCount"] = len(substituent_atom_ids)
    result["S_SubstituentHeavyAtomCount"] = sum(
        element != "H"
        for element in substituent_elements
    )
    result["S_SubstituentCarbonCount"] = substituent_elements.count("C")
    result["S_SubstituentHydrogenCount"] = substituent_elements.count("H")
    result["S_Substituent_HeteroatomCount"] = sum(
        element not in {"H", "C"}
        for element in substituent_elements
    )

    # S-root bond length. Normally this should correspond to S-C.
    result["S_C_SubstituentBond_A"] = distance_a(
        coordinates[sulfur_atom - 1],
        coordinates[root_atom - 1],
    )

    sulfur_coordinate = coordinates[sulfur_atom - 1]

    sulfur_distances = [
        distance_a(
            sulfur_coordinate,
            coordinates[atom_id - 1],
        )
        for atom_id in substituent_atom_ids
    ]

    if len(sulfur_distances) > 0:
        result["S_Substituent_MaxDistanceFromS_A"] = float(
            np.max(sulfur_distances)
        )
        result["S_Substituent_MeanDistanceFromS_A"] = float(
            np.mean(sulfur_distances)
        )

    max_internal_distance = 0.0

    for index_i in range(len(substituent_atom_ids)):
        for index_j in range(index_i + 1, len(substituent_atom_ids)):
            current_distance = distance_a(
                substituent_coordinates[index_i],
                substituent_coordinates[index_j],
            )

            max_internal_distance = max(
                max_internal_distance,
                current_distance,
            )

    result["S_Substituent_MaxInternalDistance_A"] = (
        max_internal_distance
    )

    # Local substituent SASA and volume
    try:
        substituent_sasa = SASA(
            substituent_elements,
            substituent_coordinates,
            probe_radius=SASA_PROBE_RADIUS,
        )

        result["S_Substituent_SASA_A2"] = finite_value(
            substituent_sasa.area
        )
        result["S_Substituent_SASA_Volume_A3"] = finite_value(
            substituent_sasa.volume
        )

    except Exception as error:
        result["S_SubstituentWarning"] = (
            f"{result['S_SubstituentWarning'] or ''} "
            f"S-substituent SASA failed: {error}"
        ).strip()

    # S-substituent solid angle viewed from Pd.
    try:
        reduced_elements = ["Pd"] + substituent_elements
        reduced_coordinates = np.vstack(
            [
                coordinates[PD_ATOM - 1],
                substituent_coordinates,
            ]
        )

        solid_angle = SolidAngle(
            reduced_elements,
            reduced_coordinates,
            metal_index=1,
            radii_type=SOLID_ANGLE_RADII_TYPE,
        )

        result["S_Substituent_SolidAngle_sr"] = finite_value(
            solid_angle.solid_angle
        )
        result["S_Substituent_ConeAngle_deg"] = finite_value(
            solid_angle.cone_angle
        )
        result["S_Substituent_G_percent"] = finite_value(
            solid_angle.G
        )

    except Exception as error:
        result["S_SubstituentWarning"] = (
            f"{result['S_SubstituentWarning'] or ''} "
            f"S-substituent solid-angle failed: {error}"
        ).strip()

    # Pd-centred buried volume due only to S-substituent atoms.
    #
    # Pd remains atom 1.
    # All atoms except Pd and S-substituent atoms are excluded.
    try:
        atoms_to_include = set(substituent_atom_ids)

        excluded_atoms = [
            atom_id
            for atom_id in range(1, len(elements) + 1)
            if atom_id != PD_ATOM
            and atom_id not in atoms_to_include
        ]

        for radius in VBUR_RADII:
            descriptor_name = (
                f"S_Substituent_%Vbur_{radius:.1f}A"
            )

            buried_volume = BuriedVolume(
                elements,
                coordinates,
                metal_index=PD_ATOM,
                excluded_atoms=excluded_atoms,
                radius=radius,
                include_hs=False,
                radii_type=VBUR_RADII_TYPE,
                density=VBUR_DENSITY,
            )

            result[descriptor_name] = finite_value(
                100.0 * buried_volume.fraction_buried_volume
            )

    except Exception as error:
        result["S_SubstituentWarning"] = (
            f"{result['S_SubstituentWarning'] or ''} "
            f"S-substituent Vbur failed: {error}"
        ).strip()

    return result


# ============================================================
# 13. WHOLE-LIGAND STERIC DESCRIPTORS
# ============================================================

def calculate_ligand_steric_descriptors(
    elements,
    coordinates,
    donor1,
):
    """
    Calculate descriptors using ligand atoms 20-last only.

    The substrate atoms 2-19 are excluded from ligand-specific
    buried-volume calculations.
    """
    result = {
        "LigandMaximumExtension_A": np.nan,
        "LigandSASA_A2": np.nan,
        "LigandSASA_Volume_A3": np.nan,
        "LigandSolidAngle_sr": np.nan,
        "LigandSolidConeAngle_deg": np.nan,
        "LigandSolidG_percent": np.nan,
        "Ligand_%Vbur_3.0A": np.nan,
        "Ligand_%Vbur_4.0A": np.nan,
        "Ligand_%Vbur_5.5A": np.nan,
        "LigandDistalVolumeOutside_3.5A_A3": np.nan,
        "LigandMolecularVolume_A3": np.nan,
        "LigandDistalVolumeFractionOutside_3.5A": np.nan,
        "LigandPint": np.nan,
        "LigandPint_per_SASA": np.nan,
        "LigandPint_per_Volume": np.nan,
        "PdOrientedSterimol_B1_A": np.nan,
        "PdOrientedSterimol_B5_A": np.nan,
        "PdOrientedSterimol_L_A": np.nan,
        "LigandStericWarning": None,
    }

    ligand_atom_ids = ligand_indices(elements)

    ligand_elements = [
        normalize_element(elements[atom_id - 1])
        for atom_id in ligand_atom_ids
    ]

    ligand_coordinates = coordinates[
        np.asarray(ligand_atom_ids, dtype=int) - 1
    ]

    # Maximum ligand extension
    maximum_extension = 0.0

    for index_i in range(len(ligand_atom_ids)):
        for index_j in range(index_i + 1, len(ligand_atom_ids)):
            current_distance = distance_a(
                ligand_coordinates[index_i],
                ligand_coordinates[index_j],
            )

            maximum_extension = max(
                maximum_extension,
                current_distance,
            )

    result["LigandMaximumExtension_A"] = finite_value(
        maximum_extension
    )

    # Ligand SASA
    try:
        ligand_sasa = SASA(
            ligand_elements,
            ligand_coordinates,
            probe_radius=SASA_PROBE_RADIUS,
        )

        result["LigandSASA_A2"] = finite_value(ligand_sasa.area)
        result["LigandSASA_Volume_A3"] = finite_value(
            ligand_sasa.volume
        )

    except Exception as error:
        result["LigandStericWarning"] = (
            f"Ligand SASA failed: {error}"
        )

    # Ligand solid angle around Pd
    try:
        reduced_elements = ["Pd"] + ligand_elements
        reduced_coordinates = np.vstack(
            [
                coordinates[PD_ATOM - 1],
                ligand_coordinates,
            ]
        )

        solid_angle = SolidAngle(
            reduced_elements,
            reduced_coordinates,
            metal_index=1,
            radii_type=SOLID_ANGLE_RADII_TYPE,
        )

        result["LigandSolidAngle_sr"] = finite_value(
            solid_angle.solid_angle
        )
        result["LigandSolidConeAngle_deg"] = finite_value(
            solid_angle.cone_angle
        )
        result["LigandSolidG_percent"] = finite_value(
            solid_angle.G
        )

    except Exception as error:
        result["LigandStericWarning"] = (
            f"{result['LigandStericWarning'] or ''} "
            f"Ligand solid angle failed: {error}"
        ).strip()

    # Ligand-only buried volume around Pd.
    # Exclude every non-Pd atom that is not a ligand atom.
    try:
        ligand_atom_set = set(ligand_atom_ids)

        excluded_atoms = [
            atom_id
            for atom_id in range(1, len(elements) + 1)
            if atom_id != PD_ATOM
            and atom_id not in ligand_atom_set
        ]

        for radius in VBUR_RADII:
            descriptor_name = f"Ligand_%Vbur_{radius:.1f}A"

            buried_volume = BuriedVolume(
                elements,
                coordinates,
                metal_index=PD_ATOM,
                excluded_atoms=excluded_atoms,
                radius=radius,
                include_hs=False,
                radii_type=VBUR_RADII_TYPE,
                density=VBUR_DENSITY,
            )

            result[descriptor_name] = finite_value(
                buried_volume.fraction_buried_volume * 100.0
            )

        distal_volume_calculation = BuriedVolume(
            elements,
            coordinates,
            metal_index=PD_ATOM,
            excluded_atoms=excluded_atoms,
            radius=DISTAL_VOLUME_RADIUS,
            include_hs=False,
            radii_type=VBUR_RADII_TYPE,
            density=VBUR_DENSITY,
        )

        distal_volume_calculation.compute_distal_volume(
            method="sasa"
        )

        distal_volume = finite_value(
            distal_volume_calculation.distal_volume
        )
        molecular_volume = finite_value(
            distal_volume_calculation.molecular_volume
        )

        result[
            "LigandDistalVolumeOutside_3.5A_A3"
        ] = distal_volume

        result["LigandMolecularVolume_A3"] = molecular_volume

        if (
            pd.notna(distal_volume)
            and pd.notna(molecular_volume)
            and molecular_volume > 0
        ):
            result[
                "LigandDistalVolumeFractionOutside_3.5A"
            ] = distal_volume / molecular_volume

    except Exception as error:
        result["LigandStericWarning"] = (
            f"{result['LigandStericWarning'] or ''} "
            f"Ligand buried volume failed: {error}"
        ).strip()

    # Ligand dispersion descriptor
    if CALCULATE_DISPERSION and DISPERSION_AVAILABLE:
        try:
            dispersion = Dispersion(
                ligand_elements,
                ligand_coordinates,
            )

            pint = finite_value(dispersion.p_int)

            result["LigandPint"] = pint

            sasa_area = result.get("LigandSASA_A2", np.nan)
            ligand_volume = result.get(
                "LigandMolecularVolume_A3",
                np.nan,
            )

            if pd.notna(pint) and pd.notna(sasa_area) and sasa_area > 0:
                result["LigandPint_per_SASA"] = pint / sasa_area

            if (
                pd.notna(pint)
                and pd.notna(ligand_volume)
                and ligand_volume > 0
            ):
                result["LigandPint_per_Volume"] = pint / ligand_volume

        except Exception as error:
            result["LigandStericWarning"] = (
                f"{result['LigandStericWarning'] or ''} "
                f"Ligand dispersion failed: {error}"
            ).strip()

    # Pd-oriented Sterimol.
    #
    # This describes ligand dimensions with Pd as a dummy atom and
    # the selected donor atom as the attached atom.
    if donor1 is not None:
        try:
            reduced_elements = ["Pd"] + ligand_elements
            reduced_coordinates = np.vstack(
                [
                    coordinates[PD_ATOM - 1],
                    ligand_coordinates,
                ]
            )

            original_to_reduced = {
                original_atom: reduced_atom
                for reduced_atom, original_atom in enumerate(
                    ligand_atom_ids,
                    start=2,
                )
            }

            donor1_reduced = original_to_reduced[donor1]

            sterimol = Sterimol(
                reduced_elements,
                reduced_coordinates,
                dummy_index=1,
                attached_index=donor1_reduced,
                radii_type=STERIMOL_RADII_TYPE,
            )

            result["PdOrientedSterimol_B1_A"] = finite_value(
                sterimol.B_1_value
            )
            result["PdOrientedSterimol_B5_A"] = finite_value(
                sterimol.B_5_value
            )
            result["PdOrientedSterimol_L_A"] = finite_value(
                sterimol.L_value
            )

        except Exception as error:
            result["LigandStericWarning"] = (
                f"{result['LigandStericWarning'] or ''} "
                f"Pd-oriented Sterimol failed: {error}"
            ).strip()

    return result


# ============================================================
# 14. Pd COORDINATION DESCRIPTORS
# ============================================================

def calculate_coordination_descriptors(
    elements,
    coordinates,
    donor1,
    donor2,
):
    """Calculate Pd-donor geometry descriptors."""
    result = {
        "Pd_Donor1_Distance_A": np.nan,
        "Pd_Donor2_Distance_A": np.nan,
        "Average_Pd_Donor_Distance_A": np.nan,
        "Delta_Pd_Donor_Distance_A": np.nan,
        "Pd_Donor_Distance_Ratio": np.nan,
        "Donor1_Donor2_Distance_A": np.nan,
        "Bite_Angle_deg": np.nan,
        "Pd_Coordination_Distortion_A": np.nan,
        "Pd_N_PiC_DistanceDifference_A": np.nan,
        "Pd_N_PiC_DistanceRatio": np.nan,
    }

    if donor1 is None:
        return result

    pd_coordinate = coordinates[PD_ATOM - 1]
    donor1_coordinate = coordinates[donor1 - 1]

    donor1_distance = distance_a(
        pd_coordinate,
        donor1_coordinate,
    )

    result["Pd_Donor1_Distance_A"] = donor1_distance

    if donor2 is None:
        return result

    donor2_coordinate = coordinates[donor2 - 1]

    donor2_distance = distance_a(
        pd_coordinate,
        donor2_coordinate,
    )

    result["Pd_Donor2_Distance_A"] = donor2_distance

    if (
        normalize_element(elements[donor1 - 1]) == "N"
        and normalize_element(elements[donor2 - 1]) == "C"
    ):
        result["Pd_N_PiC_DistanceDifference_A"] = abs(
            donor1_distance - donor2_distance
        )
        if donor2_distance > 0:
            result["Pd_N_PiC_DistanceRatio"] = (
                donor1_distance / donor2_distance
            )

    average_distance = (
        donor1_distance + donor2_distance
    ) / 2.0

    result["Average_Pd_Donor_Distance_A"] = average_distance
    result["Delta_Pd_Donor_Distance_A"] = abs(
        donor1_distance - donor2_distance
    )

    if min(donor1_distance, donor2_distance) > 0:
        result["Pd_Donor_Distance_Ratio"] = (
            max(donor1_distance, donor2_distance)
            / min(donor1_distance, donor2_distance)
        )

    result["Donor1_Donor2_Distance_A"] = distance_a(
        donor1_coordinate,
        donor2_coordinate,
    )

    result["Bite_Angle_deg"] = angle_deg(
        donor1_coordinate,
        pd_coordinate,
        donor2_coordinate,
    )

    result["Pd_Coordination_Distortion_A"] = float(
        np.sqrt(
            (
                (donor1_distance - average_distance) ** 2
                + (donor2_distance - average_distance) ** 2
            ) / 2.0
        )
    )

    return result


# ============================================================
# 15. DONOR LOCAL-GEOMETRY / PYRAMIDALIZATION DESCRIPTORS
# ============================================================

def calculate_one_donor_geometry(
    elements,
    coordinates,
    donor_atom,
    prefix,
):
    """
    Calculate general local donor geometry.

    The mean bond angle works for donors with at least two covalent
    ligand neighbors. Morfeus pyramidalization is calculated only
    when the donor has exactly three neighbors.
    """
    output = {
        f"{prefix}_NeighborCount": np.nan,
        f"{prefix}_MeanBondAngle_deg": np.nan,
        f"{prefix}_MeanBondAngle_N": np.nan,
        f"{prefix}_Pyramidalization_P": np.nan,
        f"{prefix}_Pyramidalization_PAngle_deg": np.nan,
        f"{prefix}_GeometryWarning": None,
    }

    if donor_atom is None:
        output[f"{prefix}_GeometryWarning"] = (
            "Donor atom unavailable."
        )
        return output

    ligand_atom_ids = ligand_indices(elements)

    neighbors = covalent_neighbors(
        elements=elements,
        coordinates=coordinates,
        atom_id=donor_atom,
        allowed_indices=ligand_atom_ids,
    )

    neighbor_ids = [neighbor_id for neighbor_id, _ in neighbors]

    output[f"{prefix}_NeighborCount"] = len(neighbor_ids)

    if len(neighbor_ids) >= 2:
        donor_coordinate = coordinates[donor_atom - 1]

        all_angles = []

        for atom_i, atom_j in itertools.combinations(
            neighbor_ids,
            2,
        ):
            current_angle = angle_deg(
                coordinates[atom_i - 1],
                donor_coordinate,
                coordinates[atom_j - 1],
            )

            if pd.notna(current_angle):
                all_angles.append(current_angle)

        if len(all_angles) > 0:
            output[f"{prefix}_MeanBondAngle_deg"] = float(
                np.mean(all_angles)
            )
            output[f"{prefix}_MeanBondAngle_N"] = len(all_angles)

    if len(neighbor_ids) != 3:
        output[f"{prefix}_GeometryWarning"] = (
            f"{len(neighbor_ids)} covalent ligand neighbors; "
            "Pyramidalization requires exactly 3."
        )
        return output

    try:
        pyramidalization = Pyramidalization(
            coordinates,
            donor_atom,
            neighbor_indices=neighbor_ids,
        )

        output[f"{prefix}_Pyramidalization_P"] = finite_value(
            pyramidalization.P
        )
        output[f"{prefix}_Pyramidalization_PAngle_deg"] = finite_value(
            pyramidalization.P_angle
        )

    except Exception as error:
        output[f"{prefix}_GeometryWarning"] = (
            f"Pyramidalization failed: {error}"
        )

    return output


def calculate_donor_geometry_descriptors(
    elements,
    coordinates,
    donor1,
    donor2,
):
    """Calculate donor geometry and donor-pair summary descriptors."""
    result = {}

    donor1_result = calculate_one_donor_geometry(
        elements,
        coordinates,
        donor1,
        prefix="Donor1",
    )

    donor2_result = calculate_one_donor_geometry(
        elements,
        coordinates,
        donor2,
        prefix="Donor2",
    )

    result.update(donor1_result)
    result.update(donor2_result)

    summary_fields = {
        "AverageDonor_Pyramidalization_P": np.nan,
        "DeltaDonor_Pyramidalization_P": np.nan,
        "AverageDonor_Pyramidalization_PAngle_deg": np.nan,
        "DeltaDonor_Pyramidalization_PAngle_deg": np.nan,
        "AverageDonor_MeanBondAngle_deg": np.nan,
        "DeltaDonor_MeanBondAngle_deg": np.nan,
    }

    result.update(summary_fields)

    p1 = result.get("Donor1_Pyramidalization_P", np.nan)
    p2 = result.get("Donor2_Pyramidalization_P", np.nan)

    if pd.notna(p1) and pd.notna(p2):
        result["AverageDonor_Pyramidalization_P"] = (
            p1 + p2
        ) / 2.0

        result["DeltaDonor_Pyramidalization_P"] = abs(p1 - p2)

    angle1 = result.get(
        "Donor1_Pyramidalization_PAngle_deg",
        np.nan,
    )
    angle2 = result.get(
        "Donor2_Pyramidalization_PAngle_deg",
        np.nan,
    )

    if pd.notna(angle1) and pd.notna(angle2):
        result[
            "AverageDonor_Pyramidalization_PAngle_deg"
        ] = (angle1 + angle2) / 2.0

        result[
            "DeltaDonor_Pyramidalization_PAngle_deg"
        ] = abs(angle1 - angle2)

    mean_angle1 = result.get(
        "Donor1_MeanBondAngle_deg",
        np.nan,
    )
    mean_angle2 = result.get(
        "Donor2_MeanBondAngle_deg",
        np.nan,
    )

    if pd.notna(mean_angle1) and pd.notna(mean_angle2):
        result["AverageDonor_MeanBondAngle_deg"] = (
            mean_angle1 + mean_angle2
        ) / 2.0

        result["DeltaDonor_MeanBondAngle_deg"] = abs(
            mean_angle1 - mean_angle2
        )

    return result


# ============================================================
# 16. xTB ELECTRONIC DESCRIPTORS
# ============================================================

def calculate_xtb_descriptors(
    elements,
    coordinates,
    donor1,
    donor2,
):
    """
    Calculate isolated ligand xTB descriptors using ligand atoms 20-end.

    Pd and substrate atoms are excluded from xTB calculation.
    """
    result = {
        "XTB_LigandChargeSum": np.nan,
        "XTB_HOMO_eV": np.nan,
        "XTB_LUMO_eV": np.nan,
        "XTB_HOMO_LUMO_Gap_eV": np.nan,
        "XTB_FermiLevel_eV": np.nan,
        "XTB_IP_eV": np.nan,
        "XTB_EA_eV": np.nan,
        "XTB_ChemicalPotential_eV": np.nan,
        "XTB_Electronegativity_eV": np.nan,
        "XTB_Hardness_eV": np.nan,
        "XTB_Softness": np.nan,
        "XTB_Electrophilicity": np.nan,
        "XTB_Nucleophilicity": np.nan,
        "XTB_Dipole_Debye": np.nan,
        "XTB_Polarizability": np.nan,
        "XTB_NFOD": np.nan,
        "Donor1_MullikenCharge": np.nan,
        "Donor2_MullikenCharge": np.nan,
        "AverageDonorCharge": np.nan,
        "DeltaDonorCharge": np.nan,
        "DonorChargeSum": np.nan,
        "XTBWarning": None,
    }

    if not RUN_XTB:
        result["XTBWarning"] = "RUN_XTB is False."
        return result

    ligand_atom_ids = ligand_indices(elements)

    ligand_elements = [
        normalize_element(elements[atom_id - 1])
        for atom_id in ligand_atom_ids
    ]

    ligand_coordinates = coordinates[
        np.asarray(ligand_atom_ids, dtype=int) - 1
    ]

    try:
        xtb_calculation = XTB(
            ligand_elements,
            ligand_coordinates,
            method=XTB_METHOD,
            charge=XTB_CHARGE,
            n_unpaired=XTB_NUNPAIRED,
            n_processes=XTB_PROCESSES,
        )

        charges = xtb_calculation.get_charges(model="Mulliken")

        result["XTB_LigandChargeSum"] = finite_value(
            sum(charges.values())
        )

        result["XTB_HOMO_eV"] = finite_value(
            xtb_calculation.get_homo(unit="eV")
        )

        result["XTB_LUMO_eV"] = finite_value(
            xtb_calculation.get_lumo(unit="eV")
        )

        result["XTB_HOMO_LUMO_Gap_eV"] = finite_value(
            xtb_calculation.get_homo_lumo_gap()
        )

        result["XTB_FermiLevel_eV"] = finite_value(
            xtb_calculation.get_fermi_level()
        )

        result["XTB_IP_eV"] = finite_value(
            xtb_calculation.get_ip(corrected=True)
        )

        result["XTB_EA_eV"] = finite_value(
            xtb_calculation.get_ea(corrected=True)
        )

        result["XTB_ChemicalPotential_eV"] = finite_value(
            xtb_calculation.get_chemical_potential(
                corrected=True
            )
        )

        result["XTB_Electronegativity_eV"] = finite_value(
            xtb_calculation.get_electronegativity(
                corrected=True
            )
        )

        result["XTB_Hardness_eV"] = finite_value(
            xtb_calculation.get_hardness()
        )

        result["XTB_Softness"] = finite_value(
            xtb_calculation.get_softness()
        )

        try:
            result["XTB_Electrophilicity"] = finite_value(
                xtb_calculation.get_global_descriptor(
                    "electrophilicity",
                    corrected=True,
                )
            )
        except Exception:
            pass

        try:
            result["XTB_Nucleophilicity"] = finite_value(
                xtb_calculation.get_global_descriptor(
                    "nucleophilicity",
                    corrected=True,
                )
            )
        except Exception:
            pass

        result["XTB_Dipole_Debye"] = finite_value(
            xtb_calculation.get_dipole_moment(unit="debye")
        )

        try:
            result["XTB_Polarizability"] = finite_value(
                xtb_calculation.get_molecular_polarizability()
            )
        except Exception:
            pass

        try:
            result["XTB_NFOD"] = finite_value(
                xtb_calculation.get_nfod()
            )
        except Exception:
            pass

        # Map original XYZ indices (20-last) to isolated ligand xTB
        # indices, which are 1-based in the charge dictionary.
        original_to_xtb_index = {
            original_atom: xtb_atom
            for xtb_atom, original_atom in enumerate(
                ligand_atom_ids,
                start=1,
            )
        }

        if donor1 is not None:
            donor1_xtb_index = original_to_xtb_index.get(donor1)

            if donor1_xtb_index is not None:
                result["Donor1_MullikenCharge"] = finite_value(
                    charges[donor1_xtb_index]
                )

        if donor2 is not None:
            donor2_xtb_index = original_to_xtb_index.get(donor2)

            if donor2_xtb_index is not None:
                result["Donor2_MullikenCharge"] = finite_value(
                    charges[donor2_xtb_index]
                )

        donor_charge_1 = result["Donor1_MullikenCharge"]
        donor_charge_2 = result["Donor2_MullikenCharge"]

        if pd.notna(donor_charge_1) and pd.notna(donor_charge_2):
            result["AverageDonorCharge"] = (
                donor_charge_1 + donor_charge_2
            ) / 2.0

            result["DeltaDonorCharge"] = abs(
                donor_charge_1 - donor_charge_2
            )

            result["DonorChargeSum"] = (
                donor_charge_1 + donor_charge_2
            )

    except Exception as error:
        result["XTBWarning"] = str(error)

    return result


# ============================================================
# 17. PROCESS ONE XYZ FILE
# ============================================================

def process_xyz_file(filepath):
    """Process one Nat_Syn-DM Pd complex XYZ file."""
    ligand_name = clean_ligand_name(filepath)

    elements, coordinates = read_xyz_file(filepath)

    validation_warnings = validate_structure(
        ligand_name,
        elements,
        coordinates,
    )

    row = {
        "Ligand": ligand_name,
        "XYZFile": os.path.basename(filepath),
        "ReactionType": REACTION_TYPE,
        "NAtoms": len(elements),
        "PdAtom": PD_ATOM,
        "SubstrateStartAtom": SUBSTRATE_START,
        "SubstrateEndAtom": SUBSTRATE_END,
        "LigandStartAtom": LIGAND_START,
        "LigandEndAtom": len(elements),
        "Yield": YIELD_DATA.get(ligand_name, np.nan),
        "StructureValidation": (
            "PASS"
            if len(validation_warnings) == 0
            else " | ".join(validation_warnings)
        ),
    }

    # --------------------------------------------------------
    # Pd donor assignment
    # --------------------------------------------------------

    all_candidates, qualifying_candidates = (
        find_pd_donor_candidates(
            elements,
            coordinates,
        )
    )

    donor_information, donor1, donor2 = select_pd_donors(
        ligand_name,
        qualifying_candidates,
        elements=elements,
        coordinates=coordinates,
    )

    row.update(donor_information)

    row["NumberOfNOSLigandCandidates"] = len(all_candidates)
    row["NumberOfQualifyingPdDonors"] = len(
        qualifying_candidates
    )

    # --------------------------------------------------------
    # FIX: find_pd_donor_candidates() now returns Strict/Extended
    # window and qualification keys (StrictWindowMin_A,
    # StrictWindowMax_A, ExtendedWindowMin_A, ExtendedWindowMax_A,
    # StrictQualifies, ExtendedQualifies) instead of the old
    # WindowMin_A / WindowMax_A / Qualifies keys. This block was
    # still reading the old key names, which raised a KeyError
    # inside process_xyz_file() for every ligand. That exception
    # was silently caught by the try/except in Section 18, so every
    # row fell back to the bare stub (Ligand/XYZFile/ReactionType/
    # Yield/ProcessingError only) - which is why master_df never
    # had a "CoordinatingSAtom" column and Section 20 raised
    # KeyError: 'CoordinatingSAtom'. Updated to the current keys,
    # and "qualifying" now refers to ExtendedQualifies (the same
    # criterion find_pd_donor_candidates uses to build
    # qualifying_candidates).
    # --------------------------------------------------------

    row["AllNOSLigandAtoms"] = "; ".join(
        (
            f"{item['Atom']}:{item['Element']}:"
            f"{item['PdDistance_A']:.3f}A "
            f"[strict {item['StrictWindowMin_A']:.2f}-"
            f"{item['StrictWindowMax_A']:.2f}A; "
            f"extended {item['ExtendedWindowMin_A']:.2f}-"
            f"{item['ExtendedWindowMax_A']:.2f}A] "
            f"{'STRICT_PASS' if item['StrictQualifies'] else ('EXTENDED_PASS' if item['ExtendedQualifies'] else 'FAIL')}"
        )
        for item in all_candidates
    )

    row["QualifyingPdDonors"] = "; ".join(
        (
            f"{item['Atom']}:{item['Element']}:"
            f"{item['PdDistance_A']:.3f}A"
        )
        for item in qualifying_candidates
    )

    row["RejectedNOSAtoms"] = "; ".join(
        (
            f"{item['Atom']}:{item['Element']}:"
            f"{item['PdDistance_A']:.3f}A"
        )
        for item in all_candidates
        if not item["ExtendedQualifies"]
    )

    # --------------------------------------------------------
    # Ligand backbone / donor-path descriptors
    # --------------------------------------------------------

    row.update(
        calculate_ligand_donor_path(
            elements,
            coordinates,
            donor1,
            donor2,
        )
    )

    # --------------------------------------------------------
    # Whole-ligand sterics
    # --------------------------------------------------------

    row.update(
        calculate_ligand_steric_descriptors(
            elements,
            coordinates,
            donor1,
        )
    )

    # --------------------------------------------------------
    # Coordination geometry
    # --------------------------------------------------------

    row.update(
        calculate_coordination_descriptors(
            elements,
            coordinates,
            donor1,
            donor2,
        )
    )

    # --------------------------------------------------------
    # Local donor geometry
    # --------------------------------------------------------

    row.update(
        calculate_donor_geometry_descriptors(
            elements,
            coordinates,
            donor1,
            donor2,
        )
    )

    # --------------------------------------------------------
    # S-substituent descriptors
    # --------------------------------------------------------

    row.update(
        calculate_s_substituent_descriptors(
            ligand_name,
            elements,
            coordinates,
            donor1,
            donor2,
        )
    )

    # --------------------------------------------------------
    # Isolated ligand xTB descriptors
    # --------------------------------------------------------

    row.update(
        calculate_xtb_descriptors(
            elements,
            coordinates,
            donor1,
            donor2,
        )
    )

    # --------------------------------------------------------
    # Warnings
    # --------------------------------------------------------

    warning_list = []

    if row["StructureValidation"] != "PASS":
        warning_list.append(row["StructureValidation"])

    if row["DonorSelectionCheck"] != "PASS":
        warning_list.append(
            f"Donor QC: {row['DonorSelectionCheck']}"
        )

    if pd.isna(row.get("Bite_Angle_deg", np.nan)):
        warning_list.append("Bite angle unavailable.")

    if row.get("S_SubstituentWarning"):
        warning_list.append(
            f"S-substituent: {row['S_SubstituentWarning']}"
        )

    if row.get("LigandStericWarning"):
        warning_list.append(
            f"Sterics: {row['LigandStericWarning']}"
        )

    if row.get("XTBWarning"):
        warning_list.append(
            f"xTB: {row['XTBWarning']}"
        )

    row["Warnings"] = " | ".join(warning_list)

    return row


# ============================================================
# 18. PROCESS ALL XYZ FILES
# ============================================================

xyz_files = sorted(
    glob.glob(
        os.path.join(XYZ_FOLDER, "*.xyz")
    )
)

print("=" * 100)
print("XYZ FILE DISCOVERY")
print("=" * 100)
print("XYZ folder:", XYZ_FOLDER)
print("Number of XYZ files found:", len(xyz_files))

if len(xyz_files) == 0:
    raise FileNotFoundError(
        f"No .xyz files found in: {XYZ_FOLDER}"
    )

for xyz_file in xyz_files:
    print(" -", os.path.basename(xyz_file))

all_rows = []

for xyz_file in xyz_files:
    print("\n" + "=" * 100)
    print("PROCESSING:", os.path.basename(xyz_file))
    print("=" * 100)

    try:
        row = process_xyz_file(xyz_file)
        all_rows.append(row)

        print(
            "Ligand:",
            row["Ligand"],
            "| Class:",
            row["LigandClass"],
            "| Donors:",
            row["Donor1Element"],
            row["Donor2Element"],
            "| QC:",
            row["DonorSelectionCheck"],
        )

    except Exception as error:
        ligand_name = clean_ligand_name(xyz_file)

        print(
            f"FAILED: {os.path.basename(xyz_file)}"
        )
        print("Error:", error)

        all_rows.append({
            "Ligand": ligand_name,
            "XYZFile": os.path.basename(xyz_file),
            "ReactionType": REACTION_TYPE,
            "Yield": YIELD_DATA.get(ligand_name, np.nan),
            "ProcessingError": str(error),
        })

master_df = pd.DataFrame(all_rows)

master_df["SortNumber"] = master_df["Ligand"].apply(
    ligand_sort_number
)

master_df = (
    master_df
    .sort_values("SortNumber")
    .drop(columns="SortNumber")
    .reset_index(drop=True)
)

master_df.to_csv(
    MASTER_CSV,
    index=False,
)

print("\n" + "=" * 100)
print("MASTER DESCRIPTOR TABLE")
print("=" * 100)

display(master_df)


# ============================================================
# 19. DONOR ASSIGNMENT QC TABLE
# ============================================================

donor_qc_columns = [
    "Ligand",
    "Yield",
    "LigandClass",
    "ExpectedDonorPair",
    "Donor1Atom",
    "Donor1Element",
    "Pd_Donor1_Distance_A",
    "Donor2Atom",
    "Donor2Element",
    "Pd_Donor2_Distance_A",
    "DonorSelectionCheck",
    "PiCoordination",
    "PiCoordinationAtom",
    "PiCoordinationPd_C_Distance_A",
    "PiCoordinationDescription",
    "NumberOfNOSLigandCandidates",
    "NumberOfQualifyingPdDonors",
    "AllNOSLigandAtoms",
    "QualifyingPdDonors",
    "RejectedNOSAtoms",
    "StructureValidation",
    "Warnings",
]

donor_qc_columns = [
    column
    for column in donor_qc_columns
    if column in master_df.columns
]

donor_qc_df = master_df[donor_qc_columns].copy()

print("\n" + "=" * 100)
print("DONOR ASSIGNMENT QUALITY CONTROL")
print("=" * 100)

display(donor_qc_df)


# ============================================================
# 20. S-SUBSTITUENT ANALYSIS TABLE
# ============================================================

s_substituent_columns = [
    "Ligand",
    "Yield",
    "LigandClass",
    "S_Substituent_Class",
    "CoordinatingSAtom",
    "CoordinatingS_Pd_Distance_A",
    "S_SubstituentRootAtom",
    "S_SubstituentRootElement",
    "S_SubstituentAtomIDs",
    "S_SubstituentAtomCount",
    "S_SubstituentHeavyAtomCount",
    "S_SubstituentCarbonCount",
    "S_SubstituentHydrogenCount",
    "S_Substituent_HeteroatomCount",
    "S_C_SubstituentBond_A",
    "S_Substituent_MaxDistanceFromS_A",
    "S_Substituent_MeanDistanceFromS_A",
    "S_Substituent_MaxInternalDistance_A",
    "S_Substituent_SASA_A2",
    "S_Substituent_SASA_Volume_A3",
    "S_Substituent_SolidAngle_sr",
    "S_Substituent_ConeAngle_deg",
    "S_Substituent_G_percent",
    "S_Substituent_%Vbur_3.0A",
    "S_Substituent_%Vbur_4.0A",
    "S_Substituent_%Vbur_5.5A",
    "S_SubstituentWarning",
]

s_substituent_columns = [
    column
    for column in s_substituent_columns
    if column in master_df.columns
]

s_substituent_df = master_df[
    master_df["CoordinatingSAtom"].notna()
][s_substituent_columns].copy()

print("\n" + "=" * 100)
print("S-SUBSTITUENT STERIC / STRUCTURAL DESCRIPTORS")
print("=" * 100)

display(s_substituent_df)


# ============================================================
# 21. YIELD CORRELATION ANALYSIS
# ============================================================

# Exclude metadata, atom-index labels, and QC columns from
# numerical descriptor correlation screening.

non_predictor_columns = {
    "Yield",
    "NAtoms",
    "PdAtom",
    "SubstrateStartAtom",
    "SubstrateEndAtom",
    "LigandStartAtom",
    "LigandEndAtom",
    "Donor1Atom",
    "Donor2Atom",
    "NumberOfNOSLigandCandidates",
    "NumberOfQualifyingPdDonors",
    "CoordinatingSAtom",
    "S_SubstituentRootAtom",
    "S_SubstituentAtomCount",
    "S_SubstituentHeavyAtomCount",
    "S_SubstituentCarbonCount",
    "S_SubstituentHydrogenCount",
    "S_Substituent_HeteroatomCount",
    "LigandDonorPathAtomCount",
    "LigandDonorPathBondCount",
    "Donor1_NeighborCount",
    "Donor2_NeighborCount",
    "Donor1_MeanBondAngle_N",
    "Donor2_MeanBondAngle_N",
}

numeric_columns = master_df.select_dtypes(
    include=[np.number]
).columns.tolist()

descriptor_columns = [
    column
    for column in numeric_columns
    if column not in non_predictor_columns
    and column != "Yield"
]

correlation_records = []

for descriptor in descriptor_columns:
    temporary_df = master_df[
        ["Yield", descriptor]
    ].dropna()

    n_observations = len(temporary_df)

    if n_observations < MINIMUM_CORRELATION_N:
        correlation_records.append({
            "Descriptor": descriptor,
            "N": n_observations,
            "Pearson_r": np.nan,
            "Pearson_p": np.nan,
            "Spearman_rho": np.nan,
            "Spearman_p": np.nan,
            "Abs_Pearson_r": np.nan,
            "Abs_Spearman_rho": np.nan,
            "Mean_Abs_Correlation": np.nan,
        })
        continue

    try:
        pearson_r, pearson_p = pearsonr(
            temporary_df["Yield"],
            temporary_df[descriptor],
        )
    except Exception:
        pearson_r, pearson_p = np.nan, np.nan

    try:
        spearman_rho, spearman_p = spearmanr(
            temporary_df["Yield"],
            temporary_df[descriptor],
        )
    except Exception:
        spearman_rho, spearman_p = np.nan, np.nan

    correlation_records.append({
        "Descriptor": descriptor,
        "N": n_observations,
        "Pearson_r": pearson_r,
        "Pearson_p": pearson_p,
        "Spearman_rho": spearman_rho,
        "Spearman_p": spearman_p,
        "Abs_Pearson_r": (
            abs(pearson_r)
            if pd.notna(pearson_r)
            else np.nan
        ),
        "Abs_Spearman_rho": (
            abs(spearman_rho)
            if pd.notna(spearman_rho)
            else np.nan
        ),
        "Mean_Abs_Correlation": (
            np.nanmean([
                abs(pearson_r)
                if pd.notna(pearson_r)
                else np.nan,
                abs(spearman_rho)
                if pd.notna(spearman_rho)
                else np.nan,
            ])
        ),
    })

yield_correlation_df = pd.DataFrame(
    correlation_records
)

# Benjamini-Hochberg FDR correction, if statsmodels is available.
try:
    from statsmodels.stats.multitest import multipletests

    valid_pearson = yield_correlation_df["Pearson_p"].notna()

    yield_correlation_df["Pearson_FDR_p"] = np.nan

    if valid_pearson.sum() > 0:
        yield_correlation_df.loc[
            valid_pearson,
            "Pearson_FDR_p",
        ] = multipletests(
            yield_correlation_df.loc[
                valid_pearson,
                "Pearson_p",
            ],
            method="fdr_bh",
        )[1]

    valid_spearman = yield_correlation_df["Spearman_p"].notna()

    yield_correlation_df["Spearman_FDR_p"] = np.nan

    if valid_spearman.sum() > 0:
        yield_correlation_df.loc[
            valid_spearman,
            "Spearman_FDR_p",
        ] = multipletests(
            yield_correlation_df.loc[
                valid_spearman,
                "Spearman_p",
            ],
            method="fdr_bh",
        )[1]

except Exception as error:
    yield_correlation_df["CorrelationFDRWarning"] = str(error)

yield_correlation_df = (
    yield_correlation_df
    .sort_values(
        "Mean_Abs_Correlation",
        ascending=False,
        na_position="last",
    )
    .reset_index(drop=True)
)

yield_correlation_df.to_csv(
    YIELD_CORRELATION_CSV,
    index=False,
)

print("\n" + "=" * 100)
print("YIELD CORRELATION RANKING")
print("=" * 100)

display(yield_correlation_df)


# ============================================================
# 22. OUTLIER SCREENING
# ============================================================

def calculate_outlier_statistics(
    values,
    iqr_multiplier=IQR_MULTIPLIER,
    mad_z_limit=MAD_Z_LIMIT,
):
    """
    Calculate IQR and MAD outlier masks.

    This is an audit tool. It flags unusual descriptors but does not
    automatically remove chemically meaningful ligands or descriptors.
    """
    values = pd.Series(values).dropna().astype(float)

    result = {
        "N_Valid": len(values),
        "N_Missing": np.nan,
        "N_Unique": values.nunique(),
        "IQR_Outliers": np.nan,
        "MAD_Outliers": np.nan,
        "Fraction_by_IQR": np.nan,
        "Fraction_by_MAD": np.nan,
        "IQR_Q1": np.nan,
        "IQR_Q3": np.nan,
        "IQR": np.nan,
        "IQR_Lower_Limit": np.nan,
        "IQR_Upper_Limit": np.nan,
        "Median": np.nan,
        "MAD": np.nan,
    }

    if len(values) < 3:
        return result, pd.Series(dtype=bool), pd.Series(dtype=bool)

    q1 = values.quantile(0.25)
    q3 = values.quantile(0.75)
    iqr = q3 - q1

    lower_limit = q1 - iqr_multiplier * iqr
    upper_limit = q3 + iqr_multiplier * iqr

    iqr_mask = (
        (values < lower_limit)
        | (values > upper_limit)
    )

    median_value = values.median()
    mad_value = np.median(
        np.abs(values - median_value)
    )

    if mad_value > 0:
        modified_z_scores = (
            0.6745 * (values - median_value) / mad_value
        )

        mad_mask = (
            np.abs(modified_z_scores) > mad_z_limit
        )
    else:
        mad_mask = pd.Series(
            False,
            index=values.index,
        )

    result.update({
        "IQR_Outliers": int(iqr_mask.sum()),
        "MAD_Outliers": int(mad_mask.sum()),
        "Fraction_by_IQR": float(iqr_mask.mean()),
        "Fraction_by_MAD": float(mad_mask.mean()),
        "IQR_Q1": q1,
        "IQR_Q3": q3,
        "IQR": iqr,
        "IQR_Lower_Limit": lower_limit,
        "IQR_Upper_Limit": upper_limit,
        "Median": median_value,
        "MAD": mad_value,
    })

    return result, iqr_mask, mad_mask


outlier_summary_records = []
outlier_detail_records = []

for descriptor in descriptor_columns:
    numeric_values = master_df[descriptor].dropna()

    if len(numeric_values) < 3:
        outlier_summary_records.append({
            "Descriptor": descriptor,
            "N_Valid": len(numeric_values),
            "N_Missing": master_df[descriptor].isna().sum(),
            "N_Unique": numeric_values.nunique(),
            "Automatic_Decision": "NOT_EVALUABLE_N_LT_3",
            "Publication_Recommendation": (
                "Exclude from outlier screening: "
                "fewer than three valid values."
            ),
        })
        continue

    statistics, iqr_mask, mad_mask = calculate_outlier_statistics(
        numeric_values
    )

    iqr_fraction = statistics["Fraction_by_IQR"]
    mad_fraction = statistics["Fraction_by_MAD"]

    if (
        iqr_fraction >= OUTLIER_FRACTION_LIMIT
        or mad_fraction >= OUTLIER_FRACTION_LIMIT
    ):
        automatic_decision = "FLAGGED_EXCESSIVE_OUTLIERS"
        recommendation = (
            "Manual chemical inspection required; "
            "do not delete descriptor solely from statistical flag."
        )

    elif (
        statistics["IQR_Outliers"] > 0
        or statistics["MAD_Outliers"] > 0
    ):
        automatic_decision = "OUTLIERS_PRESENT_REVIEW"
        recommendation = (
            "Retain provisionally; inspect flagged ligands and "
            "perform sensitivity analysis if necessary."
        )

    else:
        automatic_decision = "NO_OUTLIERS_FLAGGED"
        recommendation = (
            "Retain if chemically valid and complete."
        )

    outlier_summary_records.append({
        "Descriptor": descriptor,
        "N_Valid": len(numeric_values),
        "N_Missing": master_df[descriptor].isna().sum(),
        "N_Unique": numeric_values.nunique(),
        **statistics,
        "Automatic_Decision": automatic_decision,
        "Publication_Recommendation": recommendation,
    })

    q1 = statistics["IQR_Q1"]
    q3 = statistics["IQR_Q3"]
    iqr = statistics["IQR"]
    lower_limit = statistics["IQR_Lower_Limit"]
    upper_limit = statistics["IQR_Upper_Limit"]
    median_value = statistics["Median"]
    mad_value = statistics["MAD"]

    for original_index, value in numeric_values.items():
        current_iqr_flag = bool(
            value < lower_limit or value > upper_limit
        )

        if mad_value > 0:
            modified_z_score = (
                0.6745 * (value - median_value) / mad_value
            )
            current_mad_flag = (
                abs(modified_z_score) > MAD_Z_LIMIT
            )
        else:
            modified_z_score = np.nan
            current_mad_flag = False

        if current_iqr_flag or current_mad_flag:
            if current_iqr_flag and current_mad_flag:
                flag_method = "IQR_AND_MAD"
            elif current_iqr_flag:
                flag_method = "IQR_ONLY"
            else:
                flag_method = "MAD_ONLY"

            outlier_detail_records.append({
                "Descriptor": descriptor,
                "Ligand": master_df.loc[
                    original_index,
                    "Ligand",
                ],
                "Value": value,
                "Flag_Method": flag_method,
                "IQR_Flag": current_iqr_flag,
                "MAD_Flag": current_mad_flag,
                "Modified_Z_Score": modified_z_score,
                "IQR_Lower_Limit": lower_limit,
                "IQR_Upper_Limit": upper_limit,
                "Median": median_value,
                "MAD": mad_value,
            })

outlier_summary_df = pd.DataFrame(
    outlier_summary_records
)

outlier_detail_df = pd.DataFrame(
    outlier_detail_records
)

print("\n" + "=" * 100)
print("OUTLIER SUMMARY")
print("=" * 100)

display(outlier_summary_df)

print("\n" + "=" * 100)
print("OUTLIER DETAILS")
print("=" * 100)

display(outlier_detail_df)


# ============================================================
# 23. PLOTS
# ============================================================

# ------------------------------------------------------------
# S-substituent class vs yield
# ------------------------------------------------------------

s_class_yield_df = master_df[
    master_df["S_Substituent_Class"].notna()
    & master_df["Yield"].notna()
].copy()

s_class_yield_df = s_class_yield_df[
    ~s_class_yield_df["S_Substituent_Class"].astype(str).str.contains(
        "UPDATE",
        case=False,
        na=False,
    )
]

if len(s_class_yield_df) >= 2:
    plt.figure(figsize=(10, 6))

    sns.stripplot(
        data=s_class_yield_df,
        x="S_Substituent_Class",
        y="Yield",
        hue="S_Substituent_Class",
        legend=False,
        size=10,
        jitter=0.10,
    )

    plt.title(
        "γ-Arylation Yield by Coordinating S Substituent"
    )
    plt.xlabel("S substituent class")
    plt.ylabel("Yield")
    plt.grid(axis="y", alpha=0.25)
    plt.tight_layout()

    plt.savefig(
        os.path.join(
            PLOT_FOLDER,
            "Yield_by_S_Substituent_Class.png",
        ),
        dpi=400,
        bbox_inches="tight",
    )

    plt.show()
    plt.close()


# ------------------------------------------------------------
# Yield vs S-substituent descriptors
# ------------------------------------------------------------

s_descriptor_columns = [
    column
    for column in [
        "S_SubstituentCarbonCount",
        "S_SubstituentHeavyAtomCount",
        "S_Substituent_MaxDistanceFromS_A",
        "S_Substituent_MaxInternalDistance_A",
        "S_Substituent_SASA_A2",
        "S_Substituent_SASA_Volume_A3",
        "S_Substituent_SolidAngle_sr",
        "S_Substituent_ConeAngle_deg",
        "S_Substituent_%Vbur_3.0A",
        "S_Substituent_%Vbur_4.0A",
        "S_Substituent_%Vbur_5.5A",
        "CoordinatingS_Pd_Distance_A",
    ]
    if column in master_df.columns
]

for descriptor in s_descriptor_columns:
    plot_df = master_df[
        ["Ligand", "Yield", descriptor]
    ].dropna()

    if len(plot_df) < MINIMUM_CORRELATION_N:
        continue

    plt.figure(figsize=(7, 6))

    sns.regplot(
        data=plot_df,
        x=descriptor,
        y="Yield",
        scatter_kws={"s": 65, "alpha": 0.85},
        line_kws={"color": "black"},
    )

    for _, plot_row in plot_df.iterrows():
        plt.annotate(
            plot_row["Ligand"],
            (
                plot_row[descriptor],
                plot_row["Yield"],
            ),
            xytext=(4, 4),
            textcoords="offset points",
            fontsize=8,
        )

    plt.title(f"Yield vs {descriptor}")
    plt.xlabel(descriptor)
    plt.ylabel("Yield")
    plt.grid(alpha=0.25)
    plt.tight_layout()

    plt.savefig(
        os.path.join(
            SCATTER_FOLDER,
            f"Yield_vs_{safe_filename(descriptor)}.png",
        ),
        dpi=400,
        bbox_inches="tight",
    )

    plt.show()
    plt.close()


# ------------------------------------------------------------
# Top yield correlations
# ------------------------------------------------------------

top_correlation_plot_df = yield_correlation_df.dropna(
    subset=["Mean_Abs_Correlation"]
).head(20)

if len(top_correlation_plot_df) > 0:
    plt.figure(
        figsize=(
            10,
            max(6, 0.35 * len(top_correlation_plot_df)),
        )
    )

    sns.barplot(
        data=top_correlation_plot_df,
        y="Descriptor",
        x="Mean_Abs_Correlation",
        color="steelblue",
    )

    plt.title(
        "Top Descriptor Correlations with γ-Arylation Yield"
    )
    plt.xlabel(
        "Mean of |Pearson r| and |Spearman ρ|"
    )
    plt.ylabel("Descriptor")
    plt.xlim(0, 1)
    plt.grid(axis="x", alpha=0.25)
    plt.tight_layout()

    plt.savefig(
        os.path.join(
            PLOT_FOLDER,
            "Top_Yield_Correlations.png",
        ),
        dpi=400,
        bbox_inches="tight",
    )

    plt.show()
    plt.close()


# ============================================================
# 24. EXPORT EXCEL WORKBOOK
# ============================================================

with pd.ExcelWriter(
    FINAL_EXCEL,
    engine="openpyxl",
) as writer:

    master_df.to_excel(
        writer,
        sheet_name="All_Descriptors",
        index=False,
    )

    donor_qc_df.to_excel(
        writer,
        sheet_name="Donor_QC",
        index=False,
    )

    s_substituent_df.to_excel(
        writer,
        sheet_name="S_Substituent_Descriptors",
        index=False,
    )

    yield_correlation_df.to_excel(
        writer,
        sheet_name="Yield_Correlations",
        index=False,
    )

    outlier_summary_df.to_excel(
        writer,
        sheet_name="Outlier_Summary",
        index=False,
    )

    outlier_detail_df.to_excel(
        writer,
        sheet_name="Outlier_Details",
        index=False,
    )

    # Ligand class / S substituent category table
    category_df = master_df[
        [
            column
            for column in [
                "Ligand",
                "Yield",
                "LigandClass",
                "ExpectedDonorPair",
                "Donor1Element",
                "Donor2Element",
                "S_Substituent_Class",
                "DonorSelectionCheck",
            ]
            if column in master_df.columns
        ]
    ].copy()

    category_df.to_excel(
        writer,
        sheet_name="Ligand_Categories",
        index=False,
    )


# ============================================================
# 25. FINAL SUMMARY
# ============================================================

print("\n" + "=" * 100)
print("ANALYSIS COMPLETE")
print("=" * 100)

print("Input XYZ folder:")
print(XYZ_FOLDER)

print("\nOutput folder:")
print(OUTPUT_FOLDER)

print("\nExcel workbook:")
print(FINAL_EXCEL)

print("\nMaster descriptor CSV:")
print(MASTER_CSV)

print("\nYield correlation CSV:")
print(YIELD_CORRELATION_CSV)

print("\nNumber of processed structures:")
print(len(master_df))

print("\nNumber of structures with yields:")
print(master_df["Yield"].notna().sum())

print("\nDonor assignment QC:")
if "DonorSelectionCheck" in master_df.columns:
    display(
        master_df[
            [
                "Ligand",
                "LigandClass",
                "Donor1Element",
                "Pd_Donor1_Distance_A",
                "Donor2Element",
                "Pd_Donor2_Distance_A",
                "DonorSelectionCheck",
                "Yield",
            ]
        ]
    )

print("\nS-substituent QC:")
if len(s_substituent_df) > 0:
    display(s_substituent_df)

print("\nDone.")


# ============================================================
# ============================================================
# 26. FINAL DESCRIPTOR REDUCTION + PUBLICATION CORRELATION MATRICES
# ============================================================
# PRINCIPLE:
#   1. Remove exact mathematical duplicates first.
#   2. Do NOT remove descriptors merely because they belong to the same
#      chemical/physical family.
#   3. For descriptors in the same family, remove one only when the
#      observed Pearson OR Spearman correlation reaches the redundancy
#      threshold. The descriptor with the stronger Yield association is
#      retained as the representative; valid-N and original order are
#      secondary tie-breakers.
#   4. Apply the same statistical redundancy test to every remaining
#      descriptor pair, irrespective of family.
#   5. Recalculate Yield correlations after the final reduction.
#   6. Export a complete audit so every deletion is traceable.
# ============================================================


# ============================================================
# 26. FINAL REDUCTION / PUBLICATION OUTPUTS
# ============================================================

DESCRIPTOR_REDUNDANCY_THRESHOLD = 0.90
TOP_N_DESCRIPTORS = 10

# Output directory
OUTPUT_DIR = OUTPUT_FOLDER
os.makedirs(OUTPUT_DIR, exist_ok=True)

# Main output files
YIELDCORRELATIONCSV = os.path.join(
    OUTPUT_DIR,
    "Yield_Correlation_Final_Reduced.csv"
)

REDUCTION_AUDIT_XLSX = os.path.join(
    OUTPUT_DIR,
    "Descriptor_Reduction_Audit.xlsx"
)

PEARSON_TOP10_PNG = os.path.join(
    OUTPUT_DIR,
    "Pearson_TOP_10_Reduced_Descriptors.png"
)

SPEARMAN_TOP10_PNG = os.path.join(
    OUTPUT_DIR,
    "Spearman_TOP_10_Reduced_Descriptors.png"
)

PEARSON_ALL_PNG = os.path.join(
    OUTPUT_DIR,
    "Pearson_All_Reduced_Descriptors.png"
)

SPEARMAN_ALL_PNG = os.path.join(
    OUTPUT_DIR,
    "Spearman_All_Reduced_Descriptors.png"
)


# ------------------------------------------------------------
# Descriptor families are CHEMICAL/PHYSICAL GROUPINGS only.
# Membership alone NEVER causes deletion.
# ------------------------------------------------------------
DESCRIPTOR_FAMILIES = {
    "Pd_donor_distance_asymmetry": [
        "Delta_Pd_Donor_Distance_A",
        "Pd_Coordination_Distortion_A",
        "Pd_Donor_Distance_Ratio",
    ],
    "donor_path_bond_variability": [
        "LigandDonorPathSD_A",
        "LigandDonorPathCV",
    ],
    "donor_local_angle_geometry": [
        "Donor1_MeanBondAngle_deg",
        "Donor2_MeanBondAngle_deg",
        "AverageDonor_MeanBondAngle_deg",
        "DeltaDonor_MeanBondAngle_deg",
    ],
    "donor_pocket_separation": [
        "Donor1_Donor2_Distance_A",
        "LigandDonorPathMeanBond_A",
    ],
    "absolute_pd_donor_distances": [
        "Pd_Donor1_Distance_A",
        "Pd_Donor2_Distance_A",
        "Average_Pd_Donor_Distance_A",
    ],
}

# Exact mathematical equivalence.
# For two donor distances d1 and d2:
#   Delta = |d1-d2|
#   CoordinationDistortion = |d1-d2|/2
# Therefore they contain identical information.
EXACT_MATHEMATICAL_EQUIVALENCE = [
    (
        "Delta_Pd_Donor_Distance_A",
        "Pd_Coordination_Distortion_A",
        "Delta_Pd_Donor_Distance_A = 2 * Pd_Coordination_Distortion_A",
    ),
]


def _finite_abs(value):
    try:
        value = float(value)
        return abs(value) if np.isfinite(value) else np.nan
    except Exception:
        return np.nan


def _pairwise_correlations(dataframe, descriptor_1, descriptor_2):
    pair = dataframe[[descriptor_1, descriptor_2]].dropna()
    if len(pair) < MINIMUM_CORRELATION_N:
        return np.nan, np.nan, len(pair)

    try:
        pearson_value = pearsonr(
            pair[descriptor_1], pair[descriptor_2]
        )[0]
    except Exception:
        pearson_value = np.nan

    try:
        spearman_value = spearmanr(
            pair[descriptor_1], pair[descriptor_2]
        )[0]
    except Exception:
        spearman_value = np.nan

    return pearson_value, spearman_value, len(pair)


def _yield_score(dataframe, descriptor):
    if descriptor not in dataframe.columns:
        return np.nan
    pair = dataframe[["Yield", descriptor]].dropna()
    if len(pair) < MINIMUM_CORRELATION_N:
        return np.nan

    values = []
    try:
        values.append(abs(pearsonr(pair["Yield"], pair[descriptor])[0]))
    except Exception:
        pass
    try:
        values.append(abs(spearmanr(pair["Yield"], pair[descriptor])[0]))
    except Exception:
        pass

    values = [x for x in values if pd.notna(x)]
    return float(np.mean(values)) if values else np.nan


def _family_for_pair(descriptor_1, descriptor_2):
    families = []
    for family_name, members in DESCRIPTOR_FAMILIES.items():
        if descriptor_1 in members and descriptor_2 in members:
            families.append(family_name)
    return families[0] if families else ""


def _choose_representative(descriptors, dataframe, yield_scores, original_order):
    """Yield score -> valid N -> original order."""
    ranking = []
    for descriptor in descriptors:
        score = yield_scores.get(descriptor, np.nan)
        ranking.append({
            "Descriptor": descriptor,
            "YieldScore": score if pd.notna(score) else -np.inf,
            "ValidN": int(dataframe[descriptor].notna().sum()),
            "OriginalOrder": original_order.get(descriptor, 10**9),
        })

    ranking_df = pd.DataFrame(ranking).sort_values(
        ["YieldScore", "ValidN", "OriginalOrder"],
        ascending=[False, False, True],
    )
    return ranking_df.iloc[0]["Descriptor"]


def reduce_descriptor_redundancy(
    dataframe,
    candidate_descriptors,
    threshold=0.90,
):
    """
    Four-stage reduction:
      A. Exact mathematical equivalence.
      B. Chemical/physical family-aware statistical reduction.
      C. Global statistical redundancy reduction for all remaining pairs.
      D. Final verification that no remaining pair exceeds threshold.

    Chemical/physical family membership is descriptive only; it does not
    automatically delete a descriptor.
    """
    candidate_descriptors = [
        d for d in candidate_descriptors
        if d in dataframe.columns and d != "Yield"
    ]
    original_order = {
        descriptor: i for i, descriptor in enumerate(candidate_descriptors)
    }
    yield_scores = {
        descriptor: _yield_score(dataframe, descriptor)
        for descriptor in candidate_descriptors
    }

    active = list(candidate_descriptors)
    audit_records = []
    high_corr_records = []

    # --------------------------------------------------------
    # A. EXACT MATHEMATICAL EQUIVALENCE
    # --------------------------------------------------------
    for d1, d2, relationship in EXACT_MATHEMATICAL_EQUIVALENCE:
        if d1 not in active or d2 not in active:
            continue

        retained = _choose_representative(
            [d1, d2], dataframe, yield_scores, original_order
        )
        removed = d2 if retained == d1 else d1
        active.remove(removed)

        audit_records.append({
            "DescriptorRemoved": removed,
            "DescriptorRetained": retained,
            "ReductionLevel": "A_Mathematical",
            "Family": "Exact_mathematical_equivalence",
            "PearsonCorrelation": 1.0,
            "SpearmanCorrelation": 1.0,
            "MaximumAbsoluteCorrelation": 1.0,
            "PairwiseN": int(
                dataframe[[d1, d2]].dropna().shape[0]
            ),
            "Reason": "Exact mathematical equivalence",
            "Relationship": relationship,
            "RetainedYieldScore": yield_scores.get(retained, np.nan),
            "RemovedYieldScore": yield_scores.get(removed, np.nan),
        })

    # --------------------------------------------------------
    # B. FAMILY-AWARE STATISTICAL REDUCTION
    # --------------------------------------------------------
    # Only highly correlated members of the same chemical/physical family
    # are reduced here. A family by itself is NOT a deletion criterion.
    family_changed = True
    while family_changed:
        family_changed = False

        for family_name, members in DESCRIPTOR_FAMILIES.items():
            available = [d for d in members if d in active]
            if len(available) < 2:
                continue

            best_pair = None
            best_abs_corr = -np.inf

            for i in range(len(available)):
                for j in range(i + 1, len(available)):
                    d1, d2 = available[i], available[j]
                    pr, sr, n_pair = _pairwise_correlations(
                        dataframe, d1, d2
                    )
                    max_abs = np.nanmax([
                        _finite_abs(pr),
                        _finite_abs(sr),
                    ])

                    if pd.notna(max_abs):
                        high_corr_records.append({
                            "Descriptor1": d1,
                            "Descriptor2": d2,
                            "Family": family_name,
                            "PearsonCorrelation": pr,
                            "SpearmanCorrelation": sr,
                            "MaximumAbsoluteCorrelation": max_abs,
                            "PairwiseN": n_pair,
                            "Stage": "B_FamilyAware",
                        })

                    if (
                        pd.notna(max_abs)
                        and max_abs >= threshold
                        and max_abs > best_abs_corr
                    ):
                        best_abs_corr = max_abs
                        best_pair = (d1, d2, pr, sr, n_pair)

            if best_pair is None:
                continue

            d1, d2, pr, sr, n_pair = best_pair
            retained = _choose_representative(
                [d1, d2], dataframe, yield_scores, original_order
            )
            removed = d2 if retained == d1 else d1
            active.remove(removed)

            audit_records.append({
                "DescriptorRemoved": removed,
                "DescriptorRetained": retained,
                "ReductionLevel": "B_Chemical_Physical",
                "Family": family_name,
                "PearsonCorrelation": pr,
                "SpearmanCorrelation": sr,
                "MaximumAbsoluteCorrelation": best_abs_corr,
                "PairwiseN": n_pair,
                "Reason": (
                    "Same chemical/physical descriptor family AND "
                    f"|r| >= {threshold:.2f}"
                ),
                "Relationship": "Family similarity; not assumed mathematically identical",
                "RetainedYieldScore": yield_scores.get(retained, np.nan),
                "RemovedYieldScore": yield_scores.get(removed, np.nan),
            })
            family_changed = True
            break

        # restart after each removal so correlations are always recalculated

    # --------------------------------------------------------
    # C. GLOBAL HIGH-CORRELATION REDUCTION
    # --------------------------------------------------------
    changed = True
    while changed and len(active) >= 2:
        changed = False
        best_pair = None
        best_abs_corr = -np.inf

        for i in range(len(active)):
            for j in range(i + 1, len(active)):
                d1, d2 = active[i], active[j]
                pr, sr, n_pair = _pairwise_correlations(
                    dataframe, d1, d2
                )
                max_abs = np.nanmax([
                    _finite_abs(pr),
                    _finite_abs(sr),
                ])

                if pd.notna(max_abs):
                    high_corr_records.append({
                        "Descriptor1": d1,
                        "Descriptor2": d2,
                        "Family": _family_for_pair(d1, d2) or "Cross-family",
                        "PearsonCorrelation": pr,
                        "SpearmanCorrelation": sr,
                        "MaximumAbsoluteCorrelation": max_abs,
                        "PairwiseN": n_pair,
                        "Stage": "C_Global",
                    })

                if (
                    pd.notna(max_abs)
                    and max_abs >= threshold
                    and max_abs > best_abs_corr
                ):
                    best_abs_corr = max_abs
                    best_pair = (d1, d2, pr, sr, n_pair)

        if best_pair is None:
            break

        d1, d2, pr, sr, n_pair = best_pair
        retained = _choose_representative(
            [d1, d2], dataframe, yield_scores, original_order
        )
        removed = d2 if retained == d1 else d1
        active.remove(removed)

        audit_records.append({
            "DescriptorRemoved": removed,
            "DescriptorRetained": retained,
            "ReductionLevel": "C_Global_Statistical",
            "Family": _family_for_pair(d1, d2) or "Cross-family",
            "PearsonCorrelation": pr,
            "SpearmanCorrelation": sr,
            "MaximumAbsoluteCorrelation": best_abs_corr,
            "PairwiseN": n_pair,
            "Reason": f"Remaining descriptor pair has |r| >= {threshold:.2f}",
            "Relationship": "Statistical redundancy",
            "RetainedYieldScore": yield_scores.get(retained, np.nan),
            "RemovedYieldScore": yield_scores.get(removed, np.nan),
        })
        changed = True

    # --------------------------------------------------------
    # D. FINAL VERIFICATION
    # --------------------------------------------------------
    final_high_corr = []
    if len(active) >= 2:
        for i in range(len(active)):
            for j in range(i + 1, len(active)):
                d1, d2 = active[i], active[j]
                pr, sr, n_pair = _pairwise_correlations(
                    dataframe, d1, d2
                )
                max_abs = np.nanmax([
                    _finite_abs(pr),
                    _finite_abs(sr),
                ])
                if pd.notna(max_abs) and max_abs >= threshold:
                    final_high_corr.append({
                        "Descriptor1": d1,
                        "Descriptor2": d2,
                        "Family": _family_for_pair(d1, d2) or "Cross-family",
                        "PearsonCorrelation": pr,
                        "SpearmanCorrelation": sr,
                        "MaximumAbsoluteCorrelation": max_abs,
                        "PairwiseN": n_pair,
                    })

    active = sorted(active, key=lambda d: original_order[d])
    audit_df = pd.DataFrame(audit_records)
    high_corr_df = pd.DataFrame(high_corr_records)
    final_high_corr_df = pd.DataFrame(final_high_corr)

    return active, audit_df, high_corr_df, final_high_corr_df


# ------------------------------------------------------------
# Establish the descriptor pool from the ORIGINAL filtering stage.
# ------------------------------------------------------------
# This is deliberately constructed here because the original notebook
# does not create a universal `retaineddescriptors` variable before this
# section.
_evaluable = set()
if "yield_correlation_df" in globals():
    _evaluable = set(
        yield_correlation_df.loc[
            yield_correlation_df["Mean_Abs_Correlation"].notna(),
            "Descriptor",
        ]
    )

_excessive_outlier = set()
if "outlier_summary_df" in globals() and "Automatic_Decision" in outlier_summary_df.columns:
    _excessive_outlier = set(
        outlier_summary_df.loc[
            outlier_summary_df["Automatic_Decision"] == "FLAGGED_EXCESSIVE_OUTLIERS",
            "Descriptor",
        ]
    )

_zero_variance = {
    d for d in _evaluable
    if d in master_df.columns
    and (
        pd.isna(master_df[d].std(skipna=True))
        or master_df[d].std(skipna=True) == 0
    )
}

descriptors_before_reduction = [
    d for d in descriptor_columns
    if d in _evaluable
    and d not in _excessive_outlier
    and d not in _zero_variance
    and d in master_df.columns
]

if len(descriptors_before_reduction) < 2:
    raise ValueError(
        "Fewer than two descriptors are available for redundancy reduction."
    )

reduction_base_df = master_df[["Yield"] + descriptors_before_reduction].copy()

(
    retained_reduced_descriptors,
    descriptor_reduction_audit_df,
    high_correlation_pairs_reduction_df,
    final_high_correlation_pairs_df,
) = reduce_descriptor_redundancy(
    dataframe=reduction_base_df,
    candidate_descriptors=descriptors_before_reduction,
    threshold=DESCRIPTOR_REDUNDANCY_THRESHOLD,
)

retaineddescriptors = list(retained_reduced_descriptors)
retained = list(retained_reduced_descriptors)
correlationcolumns = ["Yield"] + retaineddescriptors
correlationdf = master_df[correlationcolumns].copy()

print("=" * 100)
print("FINAL DESCRIPTOR REDUCTION")
print("=" * 100)
print("Descriptors before reduction:", len(descriptors_before_reduction))
print("Descriptors after reduction: ", len(retaineddescriptors))
print("Redundancy threshold:          |Pearson| or |Spearman| >=", DESCRIPTOR_REDUNDANCY_THRESHOLD)
print("\nFinal retained descriptors:")
for descriptor in retaineddescriptors:
    print(" -", descriptor)

print("\nReduction audit:")
display(descriptor_reduction_audit_df)

print("\nFINAL VERIFICATION: remaining high-correlation pairs")
if len(final_high_correlation_pairs_df) == 0:
    print("PASS: No remaining descriptor pair has |Pearson| or |Spearman| >= threshold.")
else:
    display(final_high_correlation_pairs_df)

# ------------------------------------------------------------
# Recalculate Yield correlations AFTER reduction.
# ------------------------------------------------------------
correlation_records = []

for descriptor in retaineddescriptors:
    pair = master_df[["Yield", descriptor]].dropna()
    n = len(pair)

    if n < MINIMUM_CORRELATION_N:
        correlation_records.append({
            "Descriptor": descriptor,
            "N": n,
            "Pearson_r": np.nan,
            "Pearson_p": np.nan,
            "Spearman_rho": np.nan,
            "Spearman_p": np.nan,
            "Abs_Pearson_r": np.nan,
            "Abs_Spearman_rho": np.nan,
            "Mean_Abs_Correlation": np.nan,
        })
        continue

    try:
        pearson_r, pearson_p = pearsonr(pair["Yield"], pair[descriptor])
    except Exception:
        pearson_r, pearson_p = np.nan, np.nan

    try:
        spearman_rho, spearman_p = spearmanr(pair["Yield"], pair[descriptor])
    except Exception:
        spearman_rho, spearman_p = np.nan, np.nan

    abs_p = abs(pearson_r) if pd.notna(pearson_r) else np.nan
    abs_s = abs(spearman_rho) if pd.notna(spearman_rho) else np.nan
    scores = [x for x in [abs_p, abs_s] if pd.notna(x)]

    correlation_records.append({
        "Descriptor": descriptor,
        "N": n,
        "Pearson_r": pearson_r,
        "Pearson_p": pearson_p,
        "Spearman_rho": spearman_rho,
        "Spearman_p": spearman_p,
        "Abs_Pearson_r": abs_p,
        "Abs_Spearman_rho": abs_s,
        "Mean_Abs_Correlation": float(np.mean(scores)) if scores else np.nan,
    })

yieldcorrelationdf = pd.DataFrame(correlation_records)

# Benjamini-Hochberg FDR correction.
yieldcorrelationdf["Pearson_FDR_p"] = np.nan
yieldcorrelationdf["Spearman_FDR_p"] = np.nan

try:
    from statsmodels.stats.multitest import multipletests

    valid = yieldcorrelationdf["Pearson_p"].notna()
    if valid.any():
        yieldcorrelationdf.loc[valid, "Pearson_FDR_p"] = multipletests(
            yieldcorrelationdf.loc[valid, "Pearson_p"],
            method="fdr_bh",
        )[1]

    valid = yieldcorrelationdf["Spearman_p"].notna()
    if valid.any():
        yieldcorrelationdf.loc[valid, "Spearman_FDR_p"] = multipletests(
            yieldcorrelationdf.loc[valid, "Spearman_p"],
            method="fdr_bh",
        )[1]
except Exception as error:
    yieldcorrelationdf["FDR_Warning"] = str(error)

yieldcorrelationdf = yieldcorrelationdf.sort_values(
    "Mean_Abs_Correlation", ascending=False, na_position="last"
).reset_index(drop=True)

yieldcorrelationdf.to_csv(YIELDCORRELATIONCSV, index=False)

print("\n" + "=" * 100)
print("YIELD CORRELATIONS AFTER FINAL REDUCTION")
print("=" * 100)
display(yieldcorrelationdf)

# ------------------------------------------------------------
# Correlation matrices: TOP 10 and ALL FINAL REDUCED.
# ------------------------------------------------------------
reduced_correlation_df = master_df[["Yield"] + retaineddescriptors].copy()

topnsummary = yieldcorrelationdf.dropna(
    subset=["Mean_Abs_Correlation"]
).head(TOP_N_DESCRIPTORS).copy()

topn_descriptors = topnsummary["Descriptor"].tolist()
topn_df = reduced_correlation_df[["Yield"] + topn_descriptors].copy()

pearson_matrix_topn = topn_df.corr(method="pearson")
spearman_matrix_topn = topn_df.corr(method="spearman")
pearson_matrix_reduced = reduced_correlation_df.corr(method="pearson")
spearman_matrix_reduced = reduced_correlation_df.corr(method="spearman")


def _publication_label(label, width=22):
    text = str(label).replace("_", " ")
    words = text.split()
    lines, current = [], ""
    for word in words:
        if not current:
            current = word
        elif len(current) + len(word) + 1 <= width:
            current += " " + word
        else:
            lines.append(current)
            current = word
    if current:
        lines.append(current)
    return "\n".join(lines)


def plot_publication_correlation_matrix(matrix, title, cbar_label, filepath):
    n = matrix.shape[0]
    if n <= 11:
        fig_w, fig_h, label_fs, annot_fs, lw = 15, 13, 11, 10, 0.6
    elif n <= 20:
        fig_w, fig_h, label_fs, annot_fs, lw = 19, 17, 9, 8, 0.5
    elif n <= 30:
        fig_w, fig_h, label_fs, annot_fs, lw = 24, 22, 8, 7, 0.45
    else:
        fig_w = max(28, 0.82 * n)
        fig_h = max(26, 0.76 * n)
        label_fs, annot_fs, lw = 7, 6, 0.35

    display_matrix = matrix.copy()
    display_matrix.index = [_publication_label(x, 24) for x in display_matrix.index]
    display_matrix.columns = [_publication_label(x, 24) for x in display_matrix.columns]

    plt.figure(figsize=(fig_w, fig_h))
    ax = sns.heatmap(
        display_matrix,
        annot=True,
        fmt=".2f",
        annot_kws={"fontsize": annot_fs, "ha": "center", "va": "center"},
        cmap="RdBu_r",
        vmin=-1,
        vmax=1,
        center=0,
        square=True,
        linewidths=lw,
        linecolor="white",
        cbar_kws={"label": cbar_label, "shrink": 0.72, "pad": 0.02},
    )
    ax.set_title(title, fontsize=16 if n <= 20 else 15, fontweight="bold", pad=20)
    ax.set_xlabel("")
    ax.set_ylabel("")
    plt.xticks(rotation=45, ha="right", rotation_mode="anchor", fontsize=label_fs)
    plt.yticks(rotation=0, fontsize=label_fs)
    plt.subplots_adjust(left=0.24, right=0.93, bottom=0.26, top=0.92)
    plt.savefig(filepath, dpi=600, bbox_inches="tight", pad_inches=0.25)
    plt.show()
    plt.close()


PEARSON_TOPN_PNG = os.path.join(
    PLOT_FOLDER, f"Pearson_TOP_{TOP_N_DESCRIPTORS}_Reduced_Descriptors.png"
)
SPEARMAN_TOPN_PNG = os.path.join(
    PLOT_FOLDER, f"Spearman_TOP_{TOP_N_DESCRIPTORS}_Reduced_Descriptors.png"
)
PEARSON_ALL_REDUCED_PNG = os.path.join(
    PLOT_FOLDER, "Pearson_All_Reduced_Descriptors.png"
)
SPEARMAN_ALL_REDUCED_PNG = os.path.join(
    PLOT_FOLDER, "Spearman_All_Reduced_Descriptors.png"
)

plot_publication_correlation_matrix(
    pearson_matrix_topn,
    f"Pearson Correlation: Yield + Top {TOP_N_DESCRIPTORS} Reduced Descriptors",
    "Pearson r",
    PEARSON_TOPN_PNG,
)
plot_publication_correlation_matrix(
    spearman_matrix_topn,
    f"Spearman Correlation: Yield + Top {TOP_N_DESCRIPTORS} Reduced Descriptors",
    "Spearman ρ",
    SPEARMAN_TOPN_PNG,
)
plot_publication_correlation_matrix(
    pearson_matrix_reduced,
    "Pearson Correlation: Yield + All Final Reduced Descriptors",
    "Pearson r",
    PEARSON_ALL_REDUCED_PNG,
)
plot_publication_correlation_matrix(
    spearman_matrix_reduced,
    "Spearman Correlation: Yield + All Final Reduced Descriptors",
    "Spearman ρ",
    SPEARMAN_ALL_REDUCED_PNG,
)

# ------------------------------------------------------------
# Final Excel export.
# ------------------------------------------------------------
with pd.ExcelWriter(
    FINAL_EXCEL,
    engine="openpyxl",
    mode="a",
    if_sheet_exists="replace",
) as writer:
    descriptor_reduction_audit_df.to_excel(
        writer, sheet_name="Descriptor_Reduction_Audit", index=False
    )
    high_correlation_pairs_reduction_df.to_excel(
        writer, sheet_name="High_Correlation_Pairs_Reduced", index=False
    )
    final_high_correlation_pairs_df.to_excel(
        writer, sheet_name="Final_High_Correlation_Check", index=False
    )
    pd.DataFrame({
        "RetainedReducedDescriptor": retaineddescriptors
    }).to_excel(
        writer, sheet_name="Final_Reduced_Descriptors", index=False
    )
    topnsummary.to_excel(
        writer, sheet_name="TopN_Reduced_Correlations", index=False
    )
    pearson_matrix_topn.to_excel(writer, sheet_name="Pearson_TopN_Reduced")
    spearman_matrix_topn.to_excel(writer, sheet_name="Spearman_TopN_Reduced")
    pearson_matrix_reduced.to_excel(writer, sheet_name="Pearson_All_Reduced")
    spearman_matrix_reduced.to_excel(writer, sheet_name="Spearman_All_Reduced")

print("\nFINAL OUTPUTS")
print("Excel:", FINAL_EXCEL)
print("Top-10 Pearson:", PEARSON_TOPN_PNG)
print("Top-10 Spearman:", SPEARMAN_TOPN_PNG)
print("All-reduced Pearson:", PEARSON_ALL_REDUCED_PNG)
print("All-reduced Spearman:", SPEARMAN_ALL_REDUCED_PNG)
print("\nDone: mathematical -> chemical/physical family-aware -> statistical -> final verification.")
```

```python
# ============================================================
# NN vs NS LIGAND COMPARISON
# Reviewer-response analysis:
# Structural, geometric, steric, and electronic differences
# ============================================================

import os
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns

from scipy.stats import (
    mannwhitneyu,
    ttest_ind,
)


# ============================================================
# 1. SELECT NN AND NS LIGANDS
# ============================================================

if "master_df" not in globals():
    raise NameError(
        "master_df is not defined. "
        "Run the complete descriptor workflow first."
    )

if "LigandClass" not in master_df.columns:
    raise KeyError(
        "LigandClass column is missing from master_df."
    )

nn_ns_df = master_df[
    master_df["LigandClass"].isin(
        ["NN", "NS"]
    )
].copy()

nn_ns_df = nn_ns_df[
    nn_ns_df["Yield"].notna()
].copy()

nn_ns_df["LigandClass"] = pd.Categorical(
    nn_ns_df["LigandClass"],
    categories=["NN", "NS"],
    ordered=True,
)

print("=" * 100)
print("NN vs NS DATASET")
print("=" * 100)

display(
    nn_ns_df[
        [
            "Ligand",
            "LigandClass",
            "Yield",
        ]
    ].sort_values(
        ["LigandClass", "Ligand"]
    )
)

print(
    "NN observations:",
    (nn_ns_df["LigandClass"] == "NN").sum(),
)

print(
    "NS observations:",
    (nn_ns_df["LigandClass"] == "NS").sum(),
)


# ============================================================
# 2. DESCRIPTOR GROUPS
# ============================================================

DESCRIPTOR_GROUPS = {
    "Coordination_geometry": [
        "Pd_Donor1_Distance_A",
        "Pd_Donor2_Distance_A",
        "Average_Pd_Donor_Distance_A",
        "Delta_Pd_Donor_Distance_A",
        "Pd_Donor_Distance_Ratio",
        "Donor1_Donor2_Distance_A",
        "Bite_Angle_deg",
        "Pd_Coordination_Distortion_A",
    ],

    "Donor_path_and_internal_geometry": [
        "LigandDonorPathAtomCount",
        "LigandDonorPathBondCount",
        "LigandDonorPathMeanBond_A",
        "LigandDonorPathSD_A",
        "LigandDonorPathCV",
        "Donor1_MeanBondAngle_deg",
        "Donor2_MeanBondAngle_deg",
        "AverageDonor_MeanBondAngle_deg",
        "DeltaDonor_MeanBondAngle_deg",
    ],

    "Global_and_distal_sterics": [
        "LigandMaximumExtension_A",
        "LigandSASA_A2",
        "LigandSASA_Volume_A3",
        "LigandMolecularVolume_A3",
        "LigandDistalVolumeOutside_3.5A_A3",
        "LigandDistalVolumeFractionOutside_3.5A",
        "LigandSolidAngle_sr",
        "LigandSolidConeAngle_deg",
        "LigandSolidG_percent",
        "Ligand_%Vbur_3.0A",
        "Ligand_%Vbur_4.0A",
        "Ligand_%Vbur_5.5A",
        "PdOrientedSterimol_B1_A",
        "PdOrientedSterimol_B5_A",
        "PdOrientedSterimol_L_A",
    ],

    "Ligand_contact_and_dispersion": [
        "LigandPint",
        "LigandPint_per_SASA",
        "LigandPint_per_Volume",
    ],

    "Electronic_xTB": [
        "XTB_LigandChargeSum",
        "XTB_HOMO_eV",
        "XTB_LUMO_eV",
        "XTB_HOMO_LUMO_Gap_eV",
        "XTB_FermiLevel_eV",
        "XTB_IP_eV",
        "XTB_EA_eV",
        "XTB_ChemicalPotential_eV",
        "XTB_Electronegativity_eV",
        "XTB_Hardness_eV",
        "XTB_Softness",
        "XTB_Electrophilicity",
        "XTB_Nucleophilicity",
        "XTB_Dipole_Debye",
        "XTB_Polarizability",
        "XTB_NFOD",
        "Donor1_MullikenCharge",
        "Donor2_MullikenCharge",
        "AverageDonorCharge",
        "DeltaDonorCharge",
        "DonorChargeSum",
    ],

    "S_substituent_specific": [
        "CoordinatingS_Pd_Distance_A",
        "S_C_SubstituentBond_A",
        "S_SubstituentAtomCount",
        "S_SubstituentHeavyAtomCount",
        "S_SubstituentCarbonCount",
        "S_SubstituentHydrogenCount",
        "S_Substituent_HeteroatomCount",
        "S_Substituent_MaxDistanceFromS_A",
        "S_Substituent_MeanDistanceFromS_A",
        "S_Substituent_MaxInternalDistance_A",
        "S_Substituent_SASA_A2",
        "S_Substituent_SASA_Volume_A3",
        "S_Substituent_SolidAngle_sr",
        "S_Substituent_ConeAngle_deg",
        "S_Substituent_G_percent",
        "S_Substituent_%Vbur_3.0A",
        "S_Substituent_%Vbur_4.0A",
        "S_Substituent_%Vbur_5.5A",
    ],
}


# Keep only descriptors that exist in master_df.
available_descriptor_groups = {
    group_name: [
        descriptor
        for descriptor in descriptors
        if descriptor in nn_ns_df.columns
    ]
    for group_name, descriptors in DESCRIPTOR_GROUPS.items()
}

available_descriptor_groups = {
    group_name: descriptors
    for group_name, descriptors in available_descriptor_groups.items()
    if descriptors
}


# ============================================================
# 3. GROUP-WISE NN VS NS STATISTICS
# ============================================================

comparison_records = []

for group_name, descriptors in (
    available_descriptor_groups.items()
):

    for descriptor in descriptors:

        nn_values = pd.to_numeric(
            nn_ns_df.loc[
                nn_ns_df["LigandClass"] == "NN",
                descriptor,
            ],
            errors="coerce",
        ).dropna()

        ns_values = pd.to_numeric(
            nn_ns_df.loc[
                nn_ns_df["LigandClass"] == "NS",
                descriptor,
            ],
            errors="coerce",
        ).dropna()

        if len(nn_values) < 2 or len(ns_values) < 2:
            comparison_records.append({
                "Group": group_name,
                "Descriptor": descriptor,
                "NN_N": len(nn_values),
                "NS_N": len(ns_values),
                "NN_Mean": np.nan,
                "NS_Mean": np.nan,
                "NN_Median": np.nan,
                "NS_Median": np.nan,
                "Difference_NS_minus_NN": np.nan,
                "FoldChange_NS_over_NN": np.nan,
                "MannWhitney_U": np.nan,
                "MannWhitney_p": np.nan,
                "Welch_t": np.nan,
                "Welch_p": np.nan,
                "EffectSize_r": np.nan,
            })
            continue

        nn_mean = float(nn_values.mean())
        ns_mean = float(ns_values.mean())

        nn_median = float(nn_values.median())
        ns_median = float(ns_values.median())

        difference = ns_mean - nn_mean

        if abs(nn_mean) > 1e-12:
            fold_change = ns_mean / nn_mean
        else:
            fold_change = np.nan

        try:
            mannwhitney_u, mannwhitney_p = (
                mannwhitneyu(
                    nn_values,
                    ns_values,
                    alternative="two-sided",
                )
            )
        except Exception:
            mannwhitney_u = np.nan
            mannwhitney_p = np.nan

        try:
            welch_t, welch_p = ttest_ind(
                nn_values,
                ns_values,
                equal_var=False,
                nan_policy="omit",
            )
        except Exception:
            welch_t = np.nan
            welch_p = np.nan

        # Rank-biserial-style effect-size estimate based on
        # the Mann-Whitney U statistic.
        if pd.notna(mannwhitney_u):
            effect_size_r = (
                1.0
                - (
                    2.0 * mannwhitney_u
                    / (len(nn_values) * len(ns_values))
                )
            )
        else:
            effect_size_r = np.nan

        comparison_records.append({
            "Group": group_name,
            "Descriptor": descriptor,
            "NN_N": len(nn_values),
            "NS_N": len(ns_values),
            "NN_Mean": nn_mean,
            "NS_Mean": ns_mean,
            "NN_Median": nn_median,
            "NS_Median": ns_median,
            "Difference_NS_minus_NN": difference,
            "FoldChange_NS_over_NN": fold_change,
            "MannWhitney_U": mannwhitney_u,
            "MannWhitney_p": mannwhitney_p,
            "Welch_t": welch_t,
            "Welch_p": welch_p,
            "EffectSize_r": effect_size_r,
        })


nn_ns_comparison_df = pd.DataFrame(
    comparison_records
)


# ============================================================
# 4. MULTIPLE-TESTING CORRECTION
# ============================================================

try:
    from statsmodels.stats.multitest import multipletests

    valid_mannwhitney = (
        nn_ns_comparison_df["MannWhitney_p"].notna()
    )

    nn_ns_comparison_df["MannWhitney_FDR_p"] = np.nan

    if valid_mannwhitney.any():
        nn_ns_comparison_df.loc[
            valid_mannwhitney,
            "MannWhitney_FDR_p",
        ] = multipletests(
            nn_ns_comparison_df.loc[
                valid_mannwhitney,
                "MannWhitney_p",
            ],
            method="fdr_bh",
        )[1]

    valid_welch = (
        nn_ns_comparison_df["Welch_p"].notna()
    )

    nn_ns_comparison_df["Welch_FDR_p"] = np.nan

    if valid_welch.any():
        nn_ns_comparison_df.loc[
            valid_welch,
            "Welch_FDR_p",
        ] = multipletests(
            nn_ns_comparison_df.loc[
                valid_welch,
                "Welch_p",
            ],
            method="fdr_bh",
        )[1]

except Exception as error:
    nn_ns_comparison_df["FDR_Warning"] = str(error)


# Rank descriptors by absolute NN/NS separation.
nn_ns_comparison_df["Abs_Mean_Difference"] = (
    nn_ns_comparison_df[
        "Difference_NS_minus_NN"
    ].abs()
)

nn_ns_comparison_df = (
    nn_ns_comparison_df
    .sort_values(
        [
            "MannWhitney_FDR_p",
            "Abs_Mean_Difference",
        ],
        ascending=[
            True,
            False,
        ],
        na_position="last",
    )
    .reset_index(drop=True)
)

print("=" * 100)
print("NN VS NS DESCRIPTOR COMPARISON")
print("=" * 100)

display(nn_ns_comparison_df)

# ============================================================
# 5. TOP FEATURES DISTINGUISHING NN FROM NS
# ============================================================

top_nn_ns_features_df = nn_ns_comparison_df[
    [
        "Group",
        "Descriptor",
        "NN_N",
        "NS_N",
        "NN_Mean",
        "NS_Mean",
        "Difference_NS_minus_NN",
        "Abs_Mean_Difference",
        "MannWhitney_p",
        "MannWhitney_FDR_p",
        "Welch_p",
        "Welch_FDR_p",
        "EffectSize_r",
    ]
].copy()

top_nn_ns_features_df = (
    top_nn_ns_features_df
    .sort_values(
        [
            "MannWhitney_FDR_p",
            "Abs_Mean_Difference",
        ],
        ascending=[
            True,
            False,
        ],
        na_position="last",
    )
    .head(20)
    .reset_index(drop=True)
)

print("=" * 100)
print("TOP FEATURES DISTINGUISHING NN AND NS")
print("=" * 100)

display(top_nn_ns_features_df)

# ============================================================
# 6. STANDARDIZED NN VS NS EFFECT-SIZE RANKING
# ============================================================

standardized_records = []

for group_name, descriptors in (
    available_descriptor_groups.items()
):

    for descriptor in descriptors:

        values = pd.to_numeric(
            nn_ns_df[descriptor],
            errors="coerce",
        )

        nn_values = values[
            nn_ns_df["LigandClass"] == "NN"
        ].dropna()

        ns_values = values[
            nn_ns_df["LigandClass"] == "NS"
        ].dropna()

        if len(nn_values) < 2 or len(ns_values) < 2:
            continue

        pooled_sd = np.sqrt(
            (
                (len(nn_values) - 1) * nn_values.var(ddof=1)
                + (len(ns_values) - 1) * ns_values.var(ddof=1)
            )
            / (
                len(nn_values) + len(ns_values) - 2
            )
        )

        if pooled_sd > 0:
            standardized_difference = (
                ns_values.mean() - nn_values.mean()
            ) / pooled_sd
        else:
            standardized_difference = np.nan

        standardized_records.append({
            "Group": group_name,
            "Descriptor": descriptor,
            "NN_Mean": nn_values.mean(),
            "NS_Mean": ns_values.mean(),
            "Pooled_SD": pooled_sd,
            "Standardized_NS_minus_NN": (
                standardized_difference
            ),
            "Absolute_Standardized_Difference": (
                abs(standardized_difference)
                if pd.notna(standardized_difference)
                else np.nan
            ),
        })

standardized_nn_ns_df = pd.DataFrame(
    standardized_records
).sort_values(
    "Absolute_Standardized_Difference",
    ascending=False,
)

print("=" * 100)
print("STANDARDIZED NN VS NS DIFFERENCES")
print("=" * 100)

display(
    standardized_nn_ns_df.head(20)
)


# ============================================================
# 7. BOXPLOTS FOR THE MOST IMPORTANT FEATURES
# ============================================================

# Select top descriptors by standardized NN/NS separation.
top_plot_descriptors = (
    standardized_nn_ns_df
    .dropna(
        subset=[
            "Absolute_Standardized_Difference"
        ]
    )
    .head(8)["Descriptor"]
    .tolist()
)

for descriptor in top_plot_descriptors:

    plot_df = nn_ns_df[
        [
            "Ligand",
            "LigandClass",
            descriptor,
        ]
    ].copy()

    plot_df[descriptor] = pd.to_numeric(
        plot_df[descriptor],
        errors="coerce",
    )

    plot_df = plot_df.dropna(
        subset=[descriptor]
    )

    if len(plot_df) == 0:
        continue

    plt.figure(figsize=(7, 5))

    sns.boxplot(
        data=plot_df,
        x="LigandClass",
        y=descriptor,
        order=["NN", "NS"],
        color="white",
        width=0.55,
        linewidth=1.4,
        fliersize=0,
    )

    sns.stripplot(
        data=plot_df,
        x="LigandClass",
        y=descriptor,
        order=["NN", "NS"],
        hue="LigandClass",
        palette={
            "NN": "#4472C4",
            "NS": "#C00000",
        },
        dodge=False,
        jitter=0.12,
        size=8,
        edgecolor="black",
        linewidth=0.7,
        legend=False,
    )

    for _, row in plot_df.iterrows():

        x_position = (
            0
            if row["LigandClass"] == "NN"
            else 1
        )

        ligand_number = ligand_sort_number(
            row["Ligand"]
        )

        x_offset = (
            (ligand_number % 5) - 2
        ) * 0.025

        plt.text(
            x_position + x_offset,
            row[descriptor],
            row["Ligand"],
            fontsize=8,
            ha="center",
            va="bottom",
        )

    plt.xlabel("Ligand donor class")
    plt.ylabel(descriptor)
    plt.title(
        f"NN versus NS: {descriptor}",
        fontweight="bold",
    )
    plt.grid(
        axis="y",
        alpha=0.25,
    )
    plt.tight_layout()

    output_path = os.path.join(
        PLOT_FOLDER,
        f"NN_vs_NS_{safe_filename(descriptor)}.png",
    )

    plt.savefig(
        output_path,
        dpi=600,
        bbox_inches="tight",
        facecolor="white",
    )

    plt.show()
    plt.close()


# ============================================================
# 8. YIELD COMPARISON
# ============================================================

yield_summary_df = (
    nn_ns_df
    .groupby("LigandClass")["Yield"]
    .agg(
        N="count",
        Mean="mean",
        Median="median",
        SD="std",
        Minimum="min",
        Maximum="max",
    )
    .reset_index()
)

print("=" * 100)
print("NN VS NS YIELD SUMMARY")
print("=" * 100)

display(yield_summary_df)


plt.figure(figsize=(6, 5))

sns.boxplot(
    data=nn_ns_df,
    x="LigandClass",
    y="Yield",
    order=["NN", "NS"],
    color="white",
    width=0.55,
    linewidth=1.4,
    fliersize=0,
)

sns.stripplot(
    data=nn_ns_df,
    x="LigandClass",
    y="Yield",
    order=["NN", "NS"],
    hue="LigandClass",
    palette={
        "NN": "#4472C4",
        "NS": "#C00000",
    },
    dodge=False,
    jitter=0.12,
    size=8,
    edgecolor="black",
    linewidth=0.7,
    legend=False,
)

for _, row in nn_ns_df.iterrows():

    x_position = (
        0
        if row["LigandClass"] == "NN"
        else 1
    )

    ligand_number = ligand_sort_number(
        row["Ligand"]
    )

    x_offset = (
        (ligand_number % 5) - 2
    ) * 0.025

    plt.text(
        x_position + x_offset,
        row["Yield"] + 1.5,
        row["Ligand"],
        fontsize=8,
        ha="center",
        va="bottom",
    )

plt.xlabel("Ligand donor class")
plt.ylabel("γ-Arylation yield")
plt.title(
    "γ-Arylation Yield: NN versus NS",
    fontweight="bold",
)
plt.grid(
    axis="y",
    alpha=0.25,
)
plt.tight_layout()

yield_plot_path = os.path.join(
    PLOT_FOLDER,
    "NN_vs_NS_Yield.png",
)

plt.savefig(
    yield_plot_path,
    dpi=600,
    bbox_inches="tight",
    facecolor="white",
)

plt.show()
plt.close()


# ============================================================
# 9. EXPORT REVIEWER-FOCUSED RESULTS
# ============================================================

reviewer_output_excel = os.path.join(
    OUTPUT_FOLDER,
    "NN_vs_NS_Reviewer_Analysis.xlsx",
)

with pd.ExcelWriter(
    reviewer_output_excel,
    engine="openpyxl",
) as writer:

    nn_ns_df.to_excel(
        writer,
        sheet_name="NN_NS_Raw_Data",
        index=False,
    )

    nn_ns_comparison_df.to_excel(
        writer,
        sheet_name="NN_NS_Statistics",
        index=False,
    )

    standardized_nn_ns_df.to_excel(
        writer,
        sheet_name="Standardized_Differences",
        index=False,
    )

    top_nn_ns_features_df.to_excel(
        writer,
        sheet_name="Top_NN_NS_Features",
        index=False,
    )

    yield_summary_df.to_excel(
        writer,
        sheet_name="NN_NS_Yield_Summary",
        index=False,
    )

print("=" * 100)
print("NN VS NS REVIEWER ANALYSIS COMPLETE")
print("=" * 100)

print(
    "Reviewer analysis workbook:",
    reviewer_output_excel,
)

print(
    "Feature plots saved in:",
    PLOT_FOLDER,
)
```

```python
# ============================================================
# NN VS NS REVIEWER FIGURE
# Standardized descriptor differences and yield comparison
# ============================================================

import matplotlib.pyplot as plt
import seaborn as sns
import numpy as np
import pandas as pd
import os


# ------------------------------------------------------------
# 1. Select chemically relevant descriptors
# ------------------------------------------------------------

REVIEWER_COORDINATION_DESCRIPTORS = [
    "Delta_Pd_Donor_Distance_A",
    "Pd_Donor_Distance_Ratio",
    "Donor1_Donor2_Distance_A",
    "DeltaDonor_MeanBondAngle_deg",
    "LigandDonorPathMeanBond_A",
]

REVIEWER_STERIC_ELECTRONIC_DESCRIPTORS = [
    "Ligand_%Vbur_5.5A",
    "LigandMolecularVolume_A3",
    "LigandSolidAngle_sr",
    "XTB_Polarizability",
]

REVIEWER_DESCRIPTORS = (
    REVIEWER_COORDINATION_DESCRIPTORS
    + REVIEWER_STERIC_ELECTRONIC_DESCRIPTORS
)

REVIEWER_DESCRIPTORS = [
    descriptor
    for descriptor in REVIEWER_DESCRIPTORS
    if descriptor in nn_ns_df.columns
]


# ------------------------------------------------------------
# 2. Create standardized difference table
# ------------------------------------------------------------

reviewer_effect_records = []

for descriptor in REVIEWER_DESCRIPTORS:

    nn_values = pd.to_numeric(
        nn_ns_df.loc[
            nn_ns_df["LigandClass"] == "NN",
            descriptor,
        ],
        errors="coerce",
    ).dropna()

    ns_values = pd.to_numeric(
        nn_ns_df.loc[
            nn_ns_df["LigandClass"] == "NS",
            descriptor,
        ],
        errors="coerce",
    ).dropna()

    if len(nn_values) < 2 or len(ns_values) < 2:
        continue

    pooled_variance = (
        (
            (len(nn_values) - 1)
            * nn_values.var(ddof=1)
        )
        + (
            (len(ns_values) - 1)
            * ns_values.var(ddof=1)
        )
    ) / (
        len(nn_values)
        + len(ns_values)
        - 2
    )

    pooled_sd = np.sqrt(pooled_variance)

    if pooled_sd > 0:
        standardized_difference = (
            ns_values.mean()
            - nn_values.mean()
        ) / pooled_sd
    else:
        standardized_difference = np.nan

    reviewer_effect_records.append({
        "Descriptor": descriptor,
        "NN_Mean": nn_values.mean(),
        "NS_Mean": ns_values.mean(),
        "Pooled_SD": pooled_sd,
        "Standardized_NS_minus_NN": (
            standardized_difference
        ),
        "Absolute_Effect_Size": (
            abs(standardized_difference)
            if pd.notna(standardized_difference)
            else np.nan
        ),
    })

reviewer_effect_df = pd.DataFrame(
    reviewer_effect_records
)

reviewer_effect_df = (
    reviewer_effect_df
    .sort_values(
        "Standardized_NS_minus_NN",
        ascending=True,
    )
    .reset_index(drop=True)
)

display(reviewer_effect_df)


# ------------------------------------------------------------
# 3. Publication-style three-panel figure
# ------------------------------------------------------------

fig, axes = plt.subplots(
    1,
    3,
    figsize=(20, 6),
    gridspec_kw={
        "width_ratios": [0.85, 1.25, 1.25],
    },
)

# ------------------------------------------------------------
# Panel A: Yield
# ------------------------------------------------------------

yield_plot_df = nn_ns_df[
    [
        "Ligand",
        "LigandClass",
        "Yield",
    ]
].copy()

sns.boxplot(
    data=yield_plot_df,
    x="LigandClass",
    y="Yield",
    order=["NN", "NS"],
    color="white",
    width=0.55,
    linewidth=1.5,
    fliersize=0,
    ax=axes[0],
)

sns.stripplot(
    data=yield_plot_df,
    x="LigandClass",
    y="Yield",
    order=["NN", "NS"],
    hue="LigandClass",
    palette={
        "NN": "#4472C4",
        "NS": "#C00000",
    },
    dodge=False,
    jitter=0.12,
    size=8,
    edgecolor="black",
    linewidth=0.7,
    legend=False,
    ax=axes[0],
)

for _, row in yield_plot_df.iterrows():

    x_position = (
        0
        if row["LigandClass"] == "NN"
        else 1
    )

    ligand_number = ligand_sort_number(
        row["Ligand"]
    )

    x_offset = (
        (ligand_number % 5) - 2
    ) * 0.025

    axes[0].text(
        x_position + x_offset,
        row["Yield"] + 1.5,
        row["Ligand"],
        fontsize=8,
        ha="center",
        va="bottom",
    )

axes[0].set_title(
    "A. γ-Arylation yield",
    fontweight="bold",
)

axes[0].set_xlabel(
    "Ligand class"
)

axes[0].set_ylabel(
    "Yield"
)

axes[0].grid(
    axis="y",
    alpha=0.25,
)


# ------------------------------------------------------------
# Panel B: Coordination geometry
# ------------------------------------------------------------

coordination_effect_df = reviewer_effect_df[
    reviewer_effect_df["Descriptor"].isin(
        REVIEWER_COORDINATION_DESCRIPTORS
    )
].copy()

coordination_effect_df["Label"] = (
    coordination_effect_df["Descriptor"]
    .str.replace(
        "_A",
        "",
        regex=False,
    )
    .str.replace(
        "_deg",
        "",
        regex=False,
    )
    .str.replace(
        "Delta_",
        "Δ ",
        regex=False,
    )
    .str.replace(
        "Donor",
        "Donor ",
        regex=False,
    )
    .str.replace(
        "_",
        " ",
        regex=False,
    )
)

axes[1].barh(
    coordination_effect_df["Label"],
    coordination_effect_df[
        "Standardized_NS_minus_NN"
    ],
    color="#C00000",
    alpha=0.85,
)

axes[1].axvline(
    0,
    color="black",
    linewidth=1,
)

axes[1].set_title(
    "B. Coordination geometry",
    fontweight="bold",
)

axes[1].set_xlabel(
    "Standardized difference: NS − NN"
)

axes[1].set_ylabel("")


# ------------------------------------------------------------
# Panel C: Steric/electronic descriptors
# ------------------------------------------------------------

steric_effect_df = reviewer_effect_df[
    reviewer_effect_df["Descriptor"].isin(
        REVIEWER_STERIC_ELECTRONIC_DESCRIPTORS
    )
].copy()

steric_effect_df["Label"] = (
    steric_effect_df["Descriptor"]
    .str.replace(
        "Ligand_",
        "",
        regex=False,
    )
    .str.replace(
        "_A3",
        "",
        regex=False,
    )
    .str.replace(
        "_sr",
        "",
        regex=False,
    )
    .str.replace(
        "_5.5A",
        " 5.5 Å",
        regex=False,
    )
    .str.replace(
        "_",
        " ",
        regex=False,
    )
)

axes[2].barh(
    steric_effect_df["Label"],
    steric_effect_df[
        "Standardized_NS_minus_NN"
    ],
    color="#ED7D31",
    alpha=0.85,
)

axes[2].axvline(
    0,
    color="black",
    linewidth=1,
)

axes[2].set_title(
    "C. Steric and electronic properties",
    fontweight="bold",
)

axes[2].set_xlabel(
    "Standardized difference: NS − NN"
)

axes[2].set_ylabel("")


plt.suptitle(
    "Structural Features Distinguishing NN and NS Ligands",
    fontsize=16,
    fontweight="bold",
)

plt.tight_layout()

reviewer_figure_path = os.path.join(
    PLOT_FOLDER,
    "NN_vs_NS_Reviewer_Structure_Comparison.png",
)

plt.savefig(
    reviewer_figure_path,
    dpi=600,
    bbox_inches="tight",
    facecolor="white",
)

plt.show()
plt.close()

print(
    "Reviewer figure saved:",
    reviewer_figure_path,
)


# ------------------------------------------------------------
# 4. Export the figure data
# ------------------------------------------------------------

reviewer_effect_csv = os.path.join(
    OUTPUT_FOLDER,
    "NN_vs_NS_Reviewer_Effect_Sizes.csv",
)

reviewer_effect_df.to_csv(
    reviewer_effect_csv,
    index=False,
)

print(
    "Effect-size table saved:",
    reviewer_effect_csv,
)
```

```python
# ============================================================
# CLEAR NN VS NS COMPARISON FIGURE
# Two colored distributions for every descriptor
# NN = blue, NS = red
# ============================================================

import os
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns


NN_COLOR = "#4472C4"
NS_COLOR = "#C00000"


# ------------------------------------------------------------
# Descriptor groups for the figure
# ------------------------------------------------------------

COORDINATION_PLOT_DESCRIPTORS = [
    "Delta_Pd_Donor_Distance_A",
    "Pd_Donor_Distance_Ratio",
    "Donor1_Donor2_Distance_A",
    "DeltaDonor_MeanBondAngle_deg",
    "LigandDonorPathMeanBond_A",
]

STERIC_ELECTRONIC_PLOT_DESCRIPTORS = [
    "Ligand_%Vbur_5.5A",
    "LigandMolecularVolume_A3",
    "LigandSolidAngle_sr",
    "XTB_Polarizability",
]


def prepare_standardized_long_data(
    dataframe,
    descriptors,
):
    """
    Convert descriptor values to standardized long-format data.

    Each descriptor is standardized independently using the
    combined NN + NS mean and standard deviation:

        z = (value - combined_mean) / combined_SD

    This preserves the direction of NN versus NS differences while
    allowing descriptors with different units to be plotted together.
    """
    records = []

    for descriptor in descriptors:

        if descriptor not in dataframe.columns:
            continue

        values = pd.to_numeric(
            dataframe[descriptor],
            errors="coerce",
        )

        combined_mean = values.mean()
        combined_sd = values.std(ddof=1)

        if pd.isna(combined_sd) or combined_sd == 0:
            continue

        for index, value in values.items():

            if pd.isna(value):
                continue

            ligand_class = dataframe.loc[
                index,
                "LigandClass",
            ]

            if ligand_class not in {
                "NN",
                "NS",
            }:
                continue

            records.append({
                "Descriptor": descriptor,
                "Ligand": dataframe.loc[
                    index,
                    "Ligand",
                ],
                "LigandClass": ligand_class,
                "RawValue": value,
                "StandardizedValue": (
                    value - combined_mean
                ) / combined_sd,
            })

    return pd.DataFrame(records)


def readable_descriptor_label(descriptor):
    """
    Convert descriptor names into readable plot labels.
    """
    label = str(descriptor)

    replacements = {
        "Delta_": "Δ ",
        "LigandDonorPathMeanBond_A": (
            "Donor-path mean bond length (Å)"
        ),
        "DeltaDonor_MeanBondAngle_deg": (
            "Donor-angle difference (°)"
        ),
        "Delta_Pd_Donor_Distance_A": (
            "Pd–donor distance difference (Å)"
        ),
        "Pd_Donor_Distance_Ratio": (
            "Pd–donor distance ratio"
        ),
        "Donor1_Donor2_Distance_A": (
            "Donor–donor distance (Å)"
        ),
        "Ligand_%Vbur_5.5A": (
            "Ligand %Vbur (5.5 Å)"
        ),
        "LigandMolecularVolume_A3": (
            "Ligand molecular volume (Å³)"
        ),
        "LigandSolidAngle_sr": (
            "Ligand solid angle (sr)"
        ),
        "XTB_Polarizability": (
            "XTB polarizability"
        ),
    }

    if label in replacements:
        return replacements[label]

    return (
        label
        .replace("_A3", " (Å³)")
        .replace("_A2", " (Å²)")
        .replace("_A", " (Å)")
        .replace("_deg", " (°)")
        .replace("_", " ")
    )


def plot_nn_ns_descriptor_panel(
    dataframe,
    descriptors,
    title,
    filename,
):
    """
    Create horizontal NN/NS boxplots for every descriptor.

    Blue = NN
    Red = NS
    Individual ligands are overlaid and labeled.
    """
    long_df = prepare_standardized_long_data(
        dataframe,
        descriptors,
    )

    if len(long_df) == 0:
        print(
            f"No valid data available for {title}."
        )
        return None

    descriptor_order = [
        descriptor
        for descriptor in descriptors
        if descriptor in long_df["Descriptor"].unique()
    ]

    label_map = {
        descriptor: readable_descriptor_label(
            descriptor
        )
        for descriptor in descriptor_order
    }

    long_df["DescriptorLabel"] = (
        long_df["Descriptor"].map(label_map)
    )

    # Reverse order so the first descriptor appears at the top.
    display_order = [
        label_map[descriptor]
        for descriptor in descriptor_order
    ][::-1]

    plt.figure(
        figsize=(
            11,
            max(
                5,
                0.85 * len(display_order),
            ),
        )
    )

    ax = sns.boxplot(
        data=long_df,
        y="DescriptorLabel",
        x="StandardizedValue",
        hue="LigandClass",
        order=display_order,
        hue_order=["NN", "NS"],
        palette={
            "NN": NN_COLOR,
            "NS": NS_COLOR,
        },
        width=0.65,
        linewidth=1.2,
        fliersize=0,
        dodge=True,
    )

    # Add individual observations as points.
    sns.stripplot(
        data=long_df,
        y="DescriptorLabel",
        x="StandardizedValue",
        hue="LigandClass",
        order=display_order,
        hue_order=["NN", "NS"],
        palette={
            "NN": NN_COLOR,
            "NS": NS_COLOR,
        },
        dodge=True,
        jitter=0.08,
        size=6,
        alpha=0.9,
        edgecolor="black",
        linewidth=0.5,
        legend=False,
        ax=ax,
    )

    ax.axvline(
        0,
        color="black",
        linewidth=0.9,
        linestyle="--",
    )

    ax.set_title(
        title,
        fontsize=14,
        fontweight="bold",
    )

    ax.set_xlabel(
        "Standardized descriptor value "
        "(combined NN + NS scale)"
    )

    ax.set_ylabel("")

    ax.legend(
        title="Ligand class",
        loc="lower right",
        frameon=True,
    )

    ax.grid(
        axis="x",
        alpha=0.25,
    )

    plt.tight_layout()

    output_path = os.path.join(
        PLOT_FOLDER,
        filename,
    )

    plt.savefig(
        output_path,
        dpi=600,
        bbox_inches="tight",
        facecolor="white",
    )

    plt.show()
    plt.close()

    print(
        "Saved:",
        output_path,
    )

    return long_df


# ============================================================
# PANEL A: COORDINATION GEOMETRY
# ============================================================

coordination_long_df = (
    plot_nn_ns_descriptor_panel(
        dataframe=nn_ns_df,
        descriptors=COORDINATION_PLOT_DESCRIPTORS,
        title=(
            "NN versus NS: Pd Coordination "
            "and Donor-Path Geometry"
        ),
        filename=(
            "NN_vs_NS_Coordination_Geometry.png"
        ),
    )
)


# ============================================================
# PANEL B: STERIC AND ELECTRONIC PROPERTIES
# ============================================================

steric_electronic_long_df = (
    plot_nn_ns_descriptor_panel(
        dataframe=nn_ns_df,
        descriptors=STERIC_ELECTRONIC_PLOT_DESCRIPTORS,
        title=(
            "NN versus NS: Steric and Electronic "
            "Properties"
        ),
        filename=(
            "NN_vs_NS_Steric_Electronic.png"
        ),
    )
)
```

```python
# ============================================================
# NN VS NS BOXPLOTS WITH CLEAR LIGAND LABELS
# ============================================================

NN_COLOR = "#4472C4"
NS_COLOR = "#C00000"


def plot_nn_ns_descriptor_panel_with_labels(
    dataframe,
    descriptors,
    title,
    filename,
):
    """
    Plot separate NN and NS distributions for each descriptor.

    Each ligand label is shown beside its individual point.
    Descriptor values are standardized independently so that
    different units can be compared in one figure.
    """
    long_df = prepare_standardized_long_data(
        dataframe,
        descriptors,
    )

    if len(long_df) == 0:
        print(
            f"No valid data available for {title}."
        )
        return None

    descriptor_order = [
        descriptor
        for descriptor in descriptors
        if descriptor in long_df["Descriptor"].unique()
    ]

    label_map = {
        descriptor: readable_descriptor_label(
            descriptor
        )
        for descriptor in descriptor_order
    }

    long_df["DescriptorLabel"] = (
        long_df["Descriptor"].map(label_map)
    )

    # Display first descriptor at the top.
    display_order = [
        label_map[descriptor]
        for descriptor in descriptor_order
    ][::-1]

    plt.figure(
        figsize=(
            13,
            max(
                6,
                0.95 * len(display_order),
            ),
        )
    )

    ax = sns.boxplot(
        data=long_df,
        y="DescriptorLabel",
        x="StandardizedValue",
        hue="LigandClass",
        order=display_order,
        hue_order=["NN", "NS"],
        palette={
            "NN": NN_COLOR,
            "NS": NS_COLOR,
        },
        width=0.65,
        linewidth=1.3,
        fliersize=0,
        dodge=True,
    )

    sns.stripplot(
        data=long_df,
        y="DescriptorLabel",
        x="StandardizedValue",
        hue="LigandClass",
        order=display_order,
        hue_order=["NN", "NS"],
        palette={
            "NN": NN_COLOR,
            "NS": NS_COLOR,
        },
        dodge=True,
        jitter=0.08,
        size=7,
        alpha=0.95,
        edgecolor="black",
        linewidth=0.6,
        legend=False,
        ax=ax,
    )

    # Zero line represents the combined NN + NS mean.
    ax.axvline(
        0,
        color="black",
        linewidth=0.9,
        linestyle="--",
    )

    # Map displayed descriptor label to y-coordinate.
    y_coordinate_map = {
        label: coordinate
        for coordinate, label in enumerate(
            display_order
        )
    }

    # Add ligand labels and black mean diamonds.
    for descriptor_label in display_order:

        descriptor_data = long_df[
            long_df["DescriptorLabel"]
            == descriptor_label
        ]

        y_position = y_coordinate_map[
            descriptor_label
        ]

        for ligand_class, class_offset in [
            ("NN", -0.20),
            ("NS", 0.20),
        ]:

            class_data = descriptor_data[
                descriptor_data["LigandClass"]
                == ligand_class
            ].copy()

            if len(class_data) == 0:
                continue

            # Add the group mean as a black diamond.
            mean_value = class_data[
                "StandardizedValue"
            ].mean()

            ax.scatter(
                mean_value,
                y_position + class_offset,
                marker="D",
                s=42,
                color="black",
                edgecolor="white",
                linewidth=0.5,
                zorder=10,
            )

            # Add each ligand label.
            for _, row in class_data.iterrows():

                x_value = row[
                    "StandardizedValue"
                ]

                ligand_name = str(
                    row["Ligand"]
                )

                # Slight horizontal offset based on ligand
                # number prevents labels from sitting exactly
                # on top of the marker.
                ligand_number = ligand_sort_number(
                    ligand_name
                )

                text_offset = (
                    (ligand_number % 3) - 1
                ) * 0.025

                ax.text(
                    x_value + text_offset,
                    y_position + class_offset + 0.075,
                    ligand_name,
                    fontsize=8,
                    color=(
                        NN_COLOR
                        if ligand_class == "NN"
                        else NS_COLOR
                    ),
                    fontweight="bold",
                    ha="center",
                    va="bottom",
                    zorder=11,
                )

    ax.set_title(
        title,
        fontsize=14,
        fontweight="bold",
    )

    ax.set_xlabel(
        "Standardized descriptor value "
        "(combined NN + NS scale)"
    )

    ax.set_ylabel("")

    ax.legend(
        title="Ligand class",
        loc="lower right",
        frameon=True,
    )

    ax.grid(
        axis="x",
        alpha=0.25,
    )

    plt.tight_layout()

    output_path = os.path.join(
        PLOT_FOLDER,
        filename,
    )

    plt.savefig(
        output_path,
        dpi=600,
        bbox_inches="tight",
        facecolor="white",
    )

    plt.show()
    plt.close()

    print(
        "Saved:",
        output_path,
    )

    return long_df
```

```python
# ============================================================
# NN VS NS PLOTS WITH NEUTRAL BOXES
# Colored points, black ligand labels
# ============================================================

NN_COLOR = "#4472C4"
NS_COLOR = "#C00000"


def plot_nn_ns_descriptor_panel_clear_labels(
    dataframe,
    descriptors,
    title,
    filename,
):
    """
    Neutral boxplots:
        - boxes are light gray/white
        - NN points are blue
        - NS points are red
        - ligand labels are black with white background
        - black diamonds indicate group means
    """
    long_df = prepare_standardized_long_data(
        dataframe,
        descriptors,
    )

    if len(long_df) == 0:
        print(
            f"No valid data available for {title}."
        )
        return None

    descriptor_order = [
        descriptor
        for descriptor in descriptors
        if descriptor in long_df["Descriptor"].unique()
    ]

    label_map = {
        descriptor: readable_descriptor_label(
            descriptor
        )
        for descriptor in descriptor_order
    }

    long_df["DescriptorLabel"] = (
        long_df["Descriptor"].map(label_map)
    )

    display_order = [
        label_map[descriptor]
        for descriptor in descriptor_order
    ][::-1]

    plt.figure(
        figsize=(
            14,
            max(
                6,
                1.0 * len(display_order),
            ),
        )
    )

    ax = sns.boxplot(
        data=long_df,
        y="DescriptorLabel",
        x="StandardizedValue",
        order=display_order,
        color="#E6E6E6",
        width=0.60,
        linewidth=1.4,
        fliersize=0,
        boxprops={
            "facecolor": "#E6E6E6",
            "edgecolor": "black",
        },
        whiskerprops={
            "color": "black",
            "linewidth": 1.2,
        },
        capprops={
            "color": "black",
            "linewidth": 1.2,
        },
        medianprops={
            "color": "black",
            "linewidth": 1.5,
        },
    )

    # Manually plot NN and NS points at separate vertical offsets.
    y_coordinate_map = {
        label: coordinate
        for coordinate, label in enumerate(
            display_order
        )
    }

    for descriptor_label in display_order:

        descriptor_data = long_df[
            long_df["DescriptorLabel"]
            == descriptor_label
        ]

        y_position = y_coordinate_map[
            descriptor_label
        ]

        for ligand_class, color, offset in [
            ("NN", NN_COLOR, -0.18),
            ("NS", NS_COLOR, 0.18),
        ]:

            class_data = descriptor_data[
                descriptor_data["LigandClass"]
                == ligand_class
            ].copy()

            if len(class_data) == 0:
                continue

            # Deterministic horizontal jitter.
            for _, row in class_data.iterrows():

                ligand_number = ligand_sort_number(
                    row["Ligand"]
                )

                horizontal_jitter = (
                    (ligand_number % 5) - 2
                ) * 0.025

                x_value = (
                    row["StandardizedValue"]
                    + horizontal_jitter
                )

                ax.scatter(
                    x_value,
                    y_position + offset,
                    s=70,
                    color=color,
                    edgecolor="black",
                    linewidth=0.7,
                    zorder=5,
                )

                # Black label with white background.
                ax.annotate(
                    str(row["Ligand"]),
                    (
                        x_value,
                        y_position + offset,
                    ),
                    xytext=(
                        0,
                        9,
                    ),
                    textcoords="offset points",
                    ha="center",
                    va="bottom",
                    fontsize=8,
                    color="black",
                    fontweight="bold",
                    bbox={
                        "boxstyle": "round,pad=0.15",
                        "facecolor": "white",
                        "edgecolor": color,
                        "linewidth": 0.8,
                        "alpha": 0.95,
                    },
                    zorder=10,
                )

            # Black diamond for the class mean.
            class_mean = class_data[
                "StandardizedValue"
            ].mean()

            ax.scatter(
                class_mean,
                y_position + offset,
                marker="D",
                s=55,
                color="black",
                edgecolor="white",
                linewidth=0.6,
                zorder=8,
            )

    # Legend for points.
    ax.scatter(
        [],
        [],
        s=70,
        color=NN_COLOR,
        edgecolor="black",
        label="NN ligands",
    )

    ax.scatter(
        [],
        [],
        s=70,
        color=NS_COLOR,
        edgecolor="black",
        label="NS ligands",
    )

    ax.scatter(
        [],
        [],
        marker="D",
        s=55,
        color="black",
        label="Class mean",
    )

    ax.axvline(
        0,
        color="black",
        linewidth=0.9,
        linestyle="--",
    )

    ax.set_title(
        title,
        fontsize=14,
        fontweight="bold",
    )

    ax.set_xlabel(
        "Standardized descriptor value "
        "(combined NN + NS scale)"
    )

    ax.set_ylabel("")

    ax.legend(
        loc="lower right",
        frameon=True,
    )

    ax.grid(
        axis="x",
        alpha=0.25,
    )

    plt.tight_layout()

    output_path = os.path.join(
        PLOT_FOLDER,
        filename,
    )

    plt.savefig(
        output_path,
        dpi=600,
        bbox_inches="tight",
        facecolor="white",
    )

    plt.show()
    plt.close()

    print(
        "Saved:",
        output_path,
    )

    return long_df
```

```python
coordination_long_df = (
    plot_nn_ns_descriptor_panel_with_labels(
        dataframe=nn_ns_df,
        descriptors=[
            "Delta_Pd_Donor_Distance_A",
            "Pd_Donor_Distance_Ratio",
            "Donor1_Donor2_Distance_A",
            "DeltaDonor_MeanBondAngle_deg",
            "LigandDonorPathMeanBond_A",
        ],
        title=(
            "NN versus NS: Coordination Geometry "
            "with Individual Ligands"
        ),
        filename=(
            "NN_vs_NS_Coordination_With_Ligand_Labels.png"
        ),
    )
)
```

```python
# ============================================================
# COMPLETE NN vs NS REVIEWER-FOCUSED ANALYSIS
#
# NN points  = blue
# NS points  = red
# Boxplots   = neutral light gray
# Labels     = black with white background
# Means      = black diamonds
# ============================================================

import os
import re
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns

from scipy.stats import mannwhitneyu, ttest_ind


# ============================================================
# 1. LOAD MASTER DATAFRAME
# ============================================================

if "master_df" not in globals():

    if "MASTER_CSV" in globals():
        master_csv_path = MASTER_CSV
    else:
        master_csv_path = (
            "/content/drive/MyDrive/"
            "Nat_Syn_DM_Gamma_Arylation_Results/"
            "Nat_Syn_DM_All_Descriptors.csv"
        )

    if not os.path.exists(master_csv_path):
        raise FileNotFoundError(
            "master_df is not available and the master CSV "
            f"was not found:\n{master_csv_path}"
        )

    master_df = pd.read_csv(master_csv_path)


if "OUTPUT_FOLDER" not in globals():
    OUTPUT_FOLDER = (
        "/content/drive/MyDrive/"
        "Nat_Syn_DM_Gamma_Arylation_Results"
    )

if "PLOT_FOLDER" not in globals():
    PLOT_FOLDER = os.path.join(
        OUTPUT_FOLDER,
        "Plots",
    )

os.makedirs(
    OUTPUT_FOLDER,
    exist_ok=True,
)

os.makedirs(
    PLOT_FOLDER,
    exist_ok=True,
)


# ============================================================
# 2. REQUIRED COLUMNS
# ============================================================

required_columns = [
    "Ligand",
    "LigandClass",
    "Yield",
]

missing_columns = [
    column
    for column in required_columns
    if column not in master_df.columns
]

if missing_columns:
    raise KeyError(
        f"Missing required columns: {missing_columns}"
    )


# ============================================================
# 3. LIGAND SORTING FUNCTION
# ============================================================

def ligand_sort_number(ligand_name):
    match = re.search(
        r"L(\d+)",
        str(ligand_name),
        flags=re.IGNORECASE,
    )

    if match:
        return int(match.group(1))

    return 999999


# ============================================================
# 4. SELECT NN AND NS LIGANDS
# ============================================================

nn_ns_df = master_df[
    master_df["LigandClass"].isin(
        ["NN", "NS"]
    )
].copy()

nn_ns_df["Yield"] = pd.to_numeric(
    nn_ns_df["Yield"],
    errors="coerce",
)

nn_ns_df = nn_ns_df[
    nn_ns_df["Yield"].notna()
].copy()

nn_ns_df["LigandClass"] = pd.Categorical(
    nn_ns_df["LigandClass"],
    categories=["NN", "NS"],
    ordered=True,
)

nn_ns_df = nn_ns_df.sort_values(
    "Ligand",
    key=lambda series: series.map(
        ligand_sort_number
    ),
).reset_index(drop=True)

print("=" * 100)
print("NN VS NS DATASET")
print("=" * 100)

display(
    nn_ns_df[
        [
            "Ligand",
            "LigandClass",
            "Yield",
        ]
    ]
)

print(
    "NN observations:",
    int(
        (nn_ns_df["LigandClass"] == "NN").sum()
    ),
)

print(
    "NS observations:",
    int(
        (nn_ns_df["LigandClass"] == "NS").sum()
    ),
)


# ============================================================
# 5. DESCRIPTOR GROUPS
# ============================================================

DESCRIPTOR_GROUPS = {
    "Coordination_geometry": [
        "Pd_Donor1_Distance_A",
        "Pd_Donor2_Distance_A",
        "Average_Pd_Donor_Distance_A",
        "Delta_Pd_Donor_Distance_A",
        "Pd_Donor_Distance_Ratio",
        "Donor1_Donor2_Distance_A",
        "Bite_Angle_deg",
        "Pd_Coordination_Distortion_A",
    ],

    "Donor_path_and_internal_geometry": [
        "LigandDonorPathAtomCount",
        "LigandDonorPathBondCount",
        "LigandDonorPathMeanBond_A",
        "LigandDonorPathSD_A",
        "LigandDonorPathCV",
        "Donor1_MeanBondAngle_deg",
        "Donor2_MeanBondAngle_deg",
        "AverageDonor_MeanBondAngle_deg",
        "DeltaDonor_MeanBondAngle_deg",
    ],

    "Global_and_distal_sterics": [
        "LigandMaximumExtension_A",
        "LigandSASA_A2",
        "LigandSASA_Volume_A3",
        "LigandMolecularVolume_A3",
        "LigandDistalVolumeOutside_3.5A_A3",
        "LigandDistalVolumeFractionOutside_3.5A",
        "LigandSolidAngle_sr",
        "LigandSolidConeAngle_deg",
        "LigandSolidG_percent",
        "Ligand_%Vbur_3.0A",
        "Ligand_%Vbur_4.0A",
        "Ligand_%Vbur_5.5A",
        "PdOrientedSterimol_B1_A",
        "PdOrientedSterimol_B5_A",
        "PdOrientedSterimol_L_A",
    ],

    "Ligand_contact_and_dispersion": [
        "LigandPint",
        "LigandPint_per_SASA",
        "LigandPint_per_Volume",
    ],

    "Electronic_xTB": [
        "XTB_LigandChargeSum",
        "XTB_HOMO_eV",
        "XTB_LUMO_eV",
        "XTB_HOMO_LUMO_Gap_eV",
        "XTB_FermiLevel_eV",
        "XTB_IP_eV",
        "XTB_EA_eV",
        "XTB_ChemicalPotential_eV",
        "XTB_Electronegativity_eV",
        "XTB_Hardness_eV",
        "XTB_Softness",
        "XTB_Electrophilicity",
        "XTB_Nucleophilicity",
        "XTB_Dipole_Debye",
        "XTB_Polarizability",
        "XTB_NFOD",
        "Donor1_MullikenCharge",
        "Donor2_MullikenCharge",
        "AverageDonorCharge",
        "DeltaDonorCharge",
        "DonorChargeSum",
    ],
}


available_descriptor_groups = {
    group_name: [
        descriptor
        for descriptor in descriptors
        if descriptor in nn_ns_df.columns
    ]
    for group_name, descriptors in DESCRIPTOR_GROUPS.items()
}

available_descriptor_groups = {
    group_name: descriptors
    for group_name, descriptors in available_descriptor_groups.items()
    if descriptors
}


# ============================================================
# 6. NN VS NS STATISTICS
# ============================================================

comparison_records = []

for group_name, descriptors in (
    available_descriptor_groups.items()
):

    for descriptor in descriptors:

        nn_values = pd.to_numeric(
            nn_ns_df.loc[
                nn_ns_df["LigandClass"] == "NN",
                descriptor,
            ],
            errors="coerce",
        ).dropna()

        ns_values = pd.to_numeric(
            nn_ns_df.loc[
                nn_ns_df["LigandClass"] == "NS",
                descriptor,
            ],
            errors="coerce",
        ).dropna()

        if len(nn_values) < 2 or len(ns_values) < 2:
            continue

        nn_mean = float(nn_values.mean())
        ns_mean = float(ns_values.mean())

        nn_median = float(nn_values.median())
        ns_median = float(ns_values.median())

        mean_difference = ns_mean - nn_mean

        try:
            mannwhitney_u, mannwhitney_p = mannwhitneyu(
                nn_values,
                ns_values,
                alternative="two-sided",
            )
        except Exception:
            mannwhitney_u = np.nan
            mannwhitney_p = np.nan

        try:
            welch_t, welch_p = ttest_ind(
                nn_values,
                ns_values,
                equal_var=False,
                nan_policy="omit",
            )
        except Exception:
            welch_t = np.nan
            welch_p = np.nan

        if pd.notna(mannwhitney_u):
            effect_size_r = (
                1.0
                - (
                    2.0
                    * mannwhitney_u
                    / (
                        len(nn_values)
                        * len(ns_values)
                    )
                )
            )
        else:
            effect_size_r = np.nan

        comparison_records.append({
            "Group": group_name,
            "Descriptor": descriptor,
            "NN_N": len(nn_values),
            "NS_N": len(ns_values),
            "NN_Mean": nn_mean,
            "NS_Mean": ns_mean,
            "NN_Median": nn_median,
            "NS_Median": ns_median,
            "Difference_NS_minus_NN": mean_difference,
            "Abs_Mean_Difference": abs(
                mean_difference
            ),
            "MannWhitney_U": mannwhitney_u,
            "MannWhitney_p": mannwhitney_p,
            "Welch_t": welch_t,
            "Welch_p": welch_p,
            "EffectSize_r": effect_size_r,
        })

nn_ns_comparison_df = pd.DataFrame(
    comparison_records
)


# ============================================================
# 7. FDR CORRECTION
# ============================================================

try:
    from statsmodels.stats.multitest import (
        multipletests
    )

    nn_ns_comparison_df[
        "MannWhitney_FDR_p"
    ] = np.nan

    nn_ns_comparison_df[
        "Welch_FDR_p"
    ] = np.nan

    valid_mannwhitney = (
        nn_ns_comparison_df[
            "MannWhitney_p"
        ].notna()
    )

    if valid_mannwhitney.any():
        nn_ns_comparison_df.loc[
            valid_mannwhitney,
            "MannWhitney_FDR_p",
        ] = multipletests(
            nn_ns_comparison_df.loc[
                valid_mannwhitney,
                "MannWhitney_p",
            ],
            method="fdr_bh",
        )[1]

    valid_welch = (
        nn_ns_comparison_df[
            "Welch_p"
        ].notna()
    )

    if valid_welch.any():
        nn_ns_comparison_df.loc[
            valid_welch,
            "Welch_FDR_p",
        ] = multipletests(
            nn_ns_comparison_df.loc[
                valid_welch,
                "Welch_p",
            ],
            method="fdr_bh",
        )[1]

except Exception as error:
    nn_ns_comparison_df[
        "FDR_Warning"
    ] = str(error)


nn_ns_comparison_df = (
    nn_ns_comparison_df
    .sort_values(
        [
            "MannWhitney_FDR_p",
            "Abs_Mean_Difference",
        ],
        ascending=[
            True,
            False,
        ],
        na_position="last",
    )
    .reset_index(drop=True)
)

print("=" * 100)
print("NN VS NS DESCRIPTOR COMPARISON")
print("=" * 100)

display(nn_ns_comparison_df)


# ============================================================
# 8. STANDARDIZED EFFECT SIZES
# ============================================================

standardized_records = []

for group_name, descriptors in (
    available_descriptor_groups.items()
):

    for descriptor in descriptors:

        values = pd.to_numeric(
            nn_ns_df[descriptor],
            errors="coerce",
        )

        nn_values = values[
            nn_ns_df["LigandClass"] == "NN"
        ].dropna()

        ns_values = values[
            nn_ns_df["LigandClass"] == "NS"
        ].dropna()

        if len(nn_values) < 2 or len(ns_values) < 2:
            continue

        pooled_variance = (
            (
                (len(nn_values) - 1)
                * nn_values.var(ddof=1)
            )
            + (
                (len(ns_values) - 1)
                * ns_values.var(ddof=1)
            )
        ) / (
            len(nn_values)
            + len(ns_values)
            - 2
        )

        pooled_sd = np.sqrt(
            pooled_variance
        )

        if pooled_sd > 0:
            standardized_difference = (
                ns_values.mean()
                - nn_values.mean()
            ) / pooled_sd
        else:
            standardized_difference = np.nan

        standardized_records.append({
            "Group": group_name,
            "Descriptor": descriptor,
            "NN_Mean": nn_values.mean(),
            "NS_Mean": ns_values.mean(),
            "Pooled_SD": pooled_sd,
            "Standardized_NS_minus_NN": (
                standardized_difference
            ),
            "Absolute_Standardized_Difference": (
                abs(standardized_difference)
                if pd.notna(
                    standardized_difference
                )
                else np.nan
            ),
        })

standardized_nn_ns_df = (
    pd.DataFrame(standardized_records)
    .sort_values(
        "Absolute_Standardized_Difference",
        ascending=False,
    )
    .reset_index(drop=True)
)

print("=" * 100)
print("STANDARDIZED NN VS NS DIFFERENCES")
print("=" * 100)

display(
    standardized_nn_ns_df.head(25)
)


# ============================================================
# 9. PLOT SETTINGS
# ============================================================

NN_COLOR = "#4472C4"
NS_COLOR = "#C00000"


def readable_descriptor_label(
    descriptor
):
    labels = {
        "Delta_Pd_Donor_Distance_A":
            "Δ Pd–donor distance (Å)",

        "Pd_Donor_Distance_Ratio":
            "Pd–donor distance ratio",

        "Donor1_Donor2_Distance_A":
            "Donor–donor distance (Å)",

        "DeltaDonor_MeanBondAngle_deg":
            "Donor-angle difference (°)",

        "LigandDonorPathMeanBond_A":
            "Donor-path mean bond length (Å)",

        "Ligand_%Vbur_5.5A":
            "Ligand %Vbur (5.5 Å)",

        "LigandMolecularVolume_A3":
            "Ligand molecular volume (Å³)",

        "LigandSolidAngle_sr":
            "Ligand solid angle (sr)",

        "XTB_Polarizability":
            "XTB polarizability",
    }

    if descriptor in labels:
        return labels[descriptor]

    return str(descriptor).replace(
        "_",
        " ",
    )


def prepare_standardized_long_data(
    dataframe,
    descriptors,
):
    records = []

    for descriptor in descriptors:

        if descriptor not in dataframe.columns:
            continue

        values = pd.to_numeric(
            dataframe[descriptor],
            errors="coerce",
        )

        combined_mean = values.mean()
        combined_sd = values.std(ddof=1)

        if (
            pd.isna(combined_sd)
            or combined_sd == 0
        ):
            continue

        for index, value in values.items():

            if pd.isna(value):
                continue

            ligand_class = dataframe.loc[
                index,
                "LigandClass",
            ]

            if ligand_class not in [
                "NN",
                "NS",
            ]:
                continue

            records.append({
                "Descriptor": descriptor,
                "DescriptorLabel": (
                    readable_descriptor_label(
                        descriptor
                    )
                ),
                "Ligand": dataframe.loc[
                    index,
                    "Ligand",
                ],
                "LigandClass": ligand_class,
                "StandardizedValue": (
                    value - combined_mean
                ) / combined_sd,
            })

    return pd.DataFrame(records)


# ============================================================
# 10. CLEAR NN VS NS PLOT
# ============================================================

def plot_nn_ns_descriptor_panel(
    dataframe,
    descriptors,
    title,
    filename,
):
    """
    Neutral gray boxes.
    Blue points = NN.
    Red points = NS.
    Black labels with white boxes.
    Black diamonds = class means.
    """
    long_df = prepare_standardized_long_data(
        dataframe,
        descriptors,
    )

    if len(long_df) == 0:
        print(
            f"No valid data for {title}"
        )
        return long_df

    descriptor_order = [
        descriptor
        for descriptor in descriptors
        if descriptor in (
            long_df["Descriptor"].unique()
        )
    ]

    display_order = [
        readable_descriptor_label(
            descriptor
        )
        for descriptor in descriptor_order
    ][::-1]

    plt.figure(
        figsize=(
            14,
            max(
                6,
                1.0 * len(display_order),
            ),
        )
    )

    ax = sns.boxplot(
        data=long_df,
        y="DescriptorLabel",
        x="StandardizedValue",
        order=display_order,
        color="#E6E6E6",
        width=0.60,
        linewidth=1.4,
        fliersize=0,
        boxprops={
            "facecolor": "#E6E6E6",
            "edgecolor": "black",
        },
        whiskerprops={
            "color": "black",
            "linewidth": 1.2,
        },
        capprops={
            "color": "black",
            "linewidth": 1.2,
        },
        medianprops={
            "color": "black",
            "linewidth": 1.5,
        },
    )

    y_position_map = {
        label: position
        for position, label in enumerate(
            display_order
        )
    }

    for descriptor_label in display_order:

        descriptor_data = long_df[
            long_df["DescriptorLabel"]
            == descriptor_label
        ]

        y_position = y_position_map[
            descriptor_label
        ]

        for (
            ligand_class,
            color,
            vertical_offset,
        ) in [
            ("NN", NN_COLOR, -0.18),
            ("NS", NS_COLOR, 0.18),
        ]:

            class_data = descriptor_data[
                descriptor_data["LigandClass"]
                == ligand_class
            ].copy()

            if len(class_data) == 0:
                continue

            class_mean = class_data[
                "StandardizedValue"
            ].mean()

            # Black diamond = class mean.
            ax.scatter(
                class_mean,
                y_position + vertical_offset,
                marker="D",
                s=55,
                color="black",
                edgecolor="white",
                linewidth=0.6,
                zorder=9,
            )

            for _, row in class_data.iterrows():

                ligand_number = ligand_sort_number(
                    row["Ligand"]
                )

                horizontal_jitter = (
                    (ligand_number % 5) - 2
                ) * 0.025

                x_value = (
                    row["StandardizedValue"]
                    + horizontal_jitter
                )

                # Colored individual point.
                ax.scatter(
                    x_value,
                    y_position + vertical_offset,
                    s=70,
                    color=color,
                    edgecolor="black",
                    linewidth=0.7,
                    zorder=6,
                )

                # Black label with white background.
                ax.annotate(
                    str(row["Ligand"]),
                    (
                        x_value,
                        y_position + vertical_offset,
                    ),
                    xytext=(0, 9),
                    textcoords="offset points",
                    ha="center",
                    va="bottom",
                    fontsize=8,
                    color="black",
                    fontweight="bold",
                    bbox={
                        "boxstyle": "round,pad=0.15",
                        "facecolor": "white",
                        "edgecolor": color,
                        "linewidth": 0.8,
                        "alpha": 0.95,
                    },
                    zorder=10,
                )

    # Legend for point colors and means.
    ax.scatter(
        [],
        [],
        s=70,
        color=NN_COLOR,
        edgecolor="black",
        label="NN ligands",
    )

    ax.scatter(
        [],
        [],
        s=70,
        color=NS_COLOR,
        edgecolor="black",
        label="NS ligands",
    )

    ax.scatter(
        [],
        [],
        marker="D",
        s=55,
        color="black",
        label="Class mean",
    )

    ax.axvline(
        0,
        color="black",
        linewidth=0.9,
        linestyle="--",
    )

    ax.set_title(
        title,
        fontsize=14,
        fontweight="bold",
    )

    ax.set_xlabel(
        "Standardized descriptor value "
        "(combined NN + NS scale)"
    )

    ax.set_ylabel("")

    ax.legend(
        loc="lower right",
        frameon=True,
    )

    ax.grid(
        axis="x",
        alpha=0.25,
    )

    plt.tight_layout()

    output_path = os.path.join(
        PLOT_FOLDER,
        filename,
    )

    plt.savefig(
        output_path,
        dpi=600,
        bbox_inches="tight",
        facecolor="white",
    )

    plt.show()
    plt.close()

    print(
        "Saved:",
        output_path,
    )

    return long_df


# ============================================================
# 11. COORDINATION PLOT
# ============================================================

coordination_plot_df = (
    plot_nn_ns_descriptor_panel(
        dataframe=nn_ns_df,
        descriptors=[
            "Delta_Pd_Donor_Distance_A",
            "Pd_Donor_Distance_Ratio",
            "Donor1_Donor2_Distance_A",
            "DeltaDonor_MeanBondAngle_deg",
            "LigandDonorPathMeanBond_A",
        ],
        title=(
            "NN versus NS: Pd Coordination "
            "and Donor-Path Geometry"
        ),
        filename=(
            "NN_vs_NS_Coordination_Clear_Labels.png"
        ),
    )
)


# ============================================================
# 12. STERIC / ELECTRONIC PLOT
# ============================================================

steric_electronic_plot_df = (
    plot_nn_ns_descriptor_panel(
        dataframe=nn_ns_df,
        descriptors=[
            "Ligand_%Vbur_5.5A",
            "LigandMolecularVolume_A3",
            "LigandSolidAngle_sr",
            "XTB_Polarizability",
        ],
        title=(
            "NN versus NS: Steric and "
            "Electronic Properties"
        ),
        filename=(
            "NN_vs_NS_Steric_Electronic_Clear_Labels.png"
        ),
    )
)


# ============================================================
# 13. YIELD PLOT
# ============================================================

yield_plot_df = nn_ns_df[
    [
        "Ligand",
        "LigandClass",
        "Yield",
    ]
].copy()

plt.figure(figsize=(7, 6))

sns.boxplot(
    data=yield_plot_df,
    x="LigandClass",
    y="Yield",
    order=["NN", "NS"],
    color="#E6E6E6",
    width=0.55,
    linewidth=1.4,
    fliersize=0,
    boxprops={
        "facecolor": "#E6E6E6",
        "edgecolor": "black",
    },
    medianprops={
        "color": "black",
        "linewidth": 1.5,
    },
)

for ligand_class, color, x_position in [
    ("NN", NN_COLOR, 0),
    ("NS", NS_COLOR, 1),
]:

    class_data = yield_plot_df[
        yield_plot_df["LigandClass"]
        == ligand_class
    ]

    for _, row in class_data.iterrows():

        ligand_number = ligand_sort_number(
            row["Ligand"]
        )

        x_jitter = (
            (ligand_number % 5) - 2
        ) * 0.025

        x_value = x_position + x_jitter

        plt.scatter(
            x_value,
            row["Yield"],
            s=80,
            color=color,
            edgecolor="black",
            linewidth=0.7,
            zorder=6,
        )

        plt.annotate(
            str(row["Ligand"]),
            (
                x_value,
                row["Yield"],
            ),
            xytext=(0, 9),
            textcoords="offset points",
            ha="center",
            va="bottom",
            fontsize=8,
            color="black",
            fontweight="bold",
            bbox={
                "boxstyle": "round,pad=0.15",
                "facecolor": "white",
                "edgecolor": color,
                "linewidth": 0.8,
                "alpha": 0.95,
            },
        )

plt.xlabel("Ligand class")
plt.ylabel("unstaurated γ-arylated lactonization yield")
plt.title(
    "NN versus NS: unstaurated γ-arylated lactonization yield",
    fontweight="bold",
)
plt.grid(
    axis="y",
    alpha=0.25,
)
plt.tight_layout()

yield_plot_path = os.path.join(
    PLOT_FOLDER,
    "NN_vs_NS_Yield_Clear_Labels.png",
)

plt.savefig(
    yield_plot_path,
    dpi=600,
    bbox_inches="tight",
    facecolor="white",
)

plt.show()
plt.close()

print(
    "Saved:",
    yield_plot_path,
)


# ============================================================
# 14. YIELD SUMMARY
# ============================================================

yield_summary_df = (
    nn_ns_df
    .groupby("LigandClass")["Yield"]
    .agg(
        N="count",
        Mean="mean",
        Median="median",
        SD="std",
        Minimum="min",
        Maximum="max",
    )
    .reset_index()
)

print("=" * 100)
print("NN VS NS YIELD SUMMARY")
print("=" * 100)

display(yield_summary_df)


# ============================================================
# 15. EXPORT REVIEWER ANALYSIS
# ============================================================

reviewer_output_excel = os.path.join(
    OUTPUT_FOLDER,
    "NN_vs_NS_Reviewer_Analysis.xlsx",
)

with pd.ExcelWriter(
    reviewer_output_excel,
    engine="openpyxl",
) as writer:

    nn_ns_df.to_excel(
        writer,
        sheet_name="NN_NS_Raw_Data",
        index=False,
    )

    nn_ns_comparison_df.to_excel(
        writer,
        sheet_name="NN_NS_Statistics",
        index=False,
    )

    standardized_nn_ns_df.to_excel(
        writer,
        sheet_name="Standardized_Differences",
        index=False,
    )

    yield_summary_df.to_excel(
        writer,
        sheet_name="NN_NS_Yield_Summary",
        index=False,
    )

    if isinstance(
        coordination_plot_df,
        pd.DataFrame,
    ):
        coordination_plot_df.to_excel(
            writer,
            sheet_name="Coordination_Plot_Data",
            index=False,
        )

    if isinstance(
        steric_electronic_plot_df,
        pd.DataFrame,
    ):
        steric_electronic_plot_df.to_excel(
            writer,
            sheet_name="Steric_Electronic_Plot_Data",
            index=False,
        )

print("=" * 100)
print("NN VS NS REVIEWER ANALYSIS COMPLETE")
print("=" * 100)

print(
    "Reviewer workbook:",
    reviewer_output_excel,
)

print(
    "Plots folder:",
    PLOT_FOLDER,
)
```

```python
# ============================================================
# NN / NO / NS COMPARATIVE SUMMARY TABLE
# ============================================================

comparison_classes = ["NN", "NO", "NS"]

class_comparison_df = master_df[
    master_df["LigandClass"].isin(
        comparison_classes
    )
].copy()

class_comparison_df["Yield"] = pd.to_numeric(
    class_comparison_df["Yield"],
    errors="coerce",
)

class_descriptor_list = [
    "Yield",
    "Pd_Donor2_Distance_A",
    "Delta_Pd_Donor_Distance_A",
    "Pd_Coordination_Distortion_A",
    "Donor1_Donor2_Distance_A",
    "Bite_Angle_deg",
    "LigandDonorPathMeanBond_A",
    "DeltaDonor_MeanBondAngle_deg",
    "Ligand_%Vbur_5.5A",
    "LigandMolecularVolume_A3",
    "XTB_Polarizability",
]

class_descriptor_list = [
    descriptor
    for descriptor in class_descriptor_list
    if descriptor in class_comparison_df.columns
]

summary_records = []

for descriptor in class_descriptor_list:

    for ligand_class in comparison_classes:

        values = pd.to_numeric(
            class_comparison_df.loc[
                class_comparison_df["LigandClass"]
                == ligand_class,
                descriptor,
            ],
            errors="coerce",
        ).dropna()

        if len(values) == 0:
            summary_records.append({
                "Descriptor": descriptor,
                "LigandClass": ligand_class,
                "N": 0,
                "Mean": np.nan,
                "SD": np.nan,
                "Median": np.nan,
                "Minimum": np.nan,
                "Maximum": np.nan,
                "Mean_SD": "NA",
            })

        elif len(values) == 1:
            summary_records.append({
                "Descriptor": descriptor,
                "LigandClass": ligand_class,
                "N": 1,
                "Mean": values.iloc[0],
                "SD": np.nan,
                "Median": values.iloc[0],
                "Minimum": values.iloc[0],
                "Maximum": values.iloc[0],
                "Mean_SD": (
                    f"{values.iloc[0]:.3f} ± NA"
                ),
            })

        else:
            mean_value = values.mean()
            sd_value = values.std(ddof=1)

            summary_records.append({
                "Descriptor": descriptor,
                "LigandClass": ligand_class,
                "N": len(values),
                "Mean": mean_value,
                "SD": sd_value,
                "Median": values.median(),
                "Minimum": values.min(),
                "Maximum": values.max(),
                "Mean_SD": (
                    f"{mean_value:.3f} ± "
                    f"{sd_value:.3f}"
                ),
            })

class_summary_long_df = pd.DataFrame(
    summary_records
)

class_summary_table_df = (
    class_summary_long_df
    .pivot(
        index="Descriptor",
        columns="LigandClass",
        values="Mean_SD",
    )
    .reset_index()
)

print("=" * 100)
print("NN / NO / NS COMPARATIVE SUMMARY")
print("=" * 100)

display(class_summary_table_df)
```

```python
# ============================================================
# COMPLETE NN vs NO vs NS REVIEWER-FOCUSED ANALYSIS
#
# NN points  = blue
# NO points  = green
# NS points  = red
# Boxplots   = neutral light gray
# Labels     = black with white background
# Means      = black diamonds
# ============================================================

import os
import re
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns

from scipy.stats import mannwhitneyu, ttest_ind


# ============================================================
# 1. LOAD MASTER DATAFRAME
# ============================================================

if "master_df" not in globals():

    if "MASTER_CSV" in globals():
        master_csv_path = MASTER_CSV

    else:
        master_csv_path = (
            "/content/drive/MyDrive/"
            "Nat_Syn_DM_Gamma_Arylation_Results/"
            "Nat_Syn_DM_All_Descriptors.csv"
        )

    if not os.path.exists(master_csv_path):
        raise FileNotFoundError(
            "master_df is not available and the master CSV "
            f"was not found:\n{master_csv_path}"
        )

    master_df = pd.read_csv(
        master_csv_path
    )


if "OUTPUT_FOLDER" not in globals():
    OUTPUT_FOLDER = (
        "/content/drive/MyDrive/"
        "Nat_Syn_DM_Gamma_Arylation_Results"
    )


if "PLOT_FOLDER" not in globals():
    PLOT_FOLDER = os.path.join(
        OUTPUT_FOLDER,
        "Plots",
    )


os.makedirs(
    OUTPUT_FOLDER,
    exist_ok=True,
)

os.makedirs(
    PLOT_FOLDER,
    exist_ok=True,
)


# ============================================================
# 2. REQUIRED COLUMNS
# ============================================================

required_columns = [
    "Ligand",
    "LigandClass",
    "Yield",
]

missing_columns = [
    column
    for column in required_columns
    if column not in master_df.columns
]

if missing_columns:
    raise KeyError(
        f"Missing required columns: {missing_columns}"
    )


# ============================================================
# 3. LIGAND SORTING FUNCTION
# ============================================================

def ligand_sort_number(ligand_name):
    match = re.search(
        r"L(\d+)",
        str(ligand_name),
        flags=re.IGNORECASE,
    )

    if match:
        return int(match.group(1))

    return 999999


# ============================================================
# 4. SELECT NN, NO, AND NS LIGANDS
# ============================================================

COMPARISON_CLASSES = [
    "NN",
    "NO",
    "NS",
]

nn_no_ns_df = master_df[
    master_df["LigandClass"].isin(
        COMPARISON_CLASSES
    )
].copy()

nn_no_ns_df["Yield"] = pd.to_numeric(
    nn_no_ns_df["Yield"],
    errors="coerce",
)

nn_no_ns_df = nn_no_ns_df[
    nn_no_ns_df["Yield"].notna()
].copy()

nn_no_ns_df["LigandClass"] = pd.Categorical(
    nn_no_ns_df["LigandClass"],
    categories=COMPARISON_CLASSES,
    ordered=True,
)

nn_no_ns_df = nn_no_ns_df.sort_values(
    "Ligand",
    key=lambda series: series.map(
        ligand_sort_number
    ),
).reset_index(drop=True)

CLASS_COLORS = {
    "NN": "#4472C4",
    "NO": "#70AD47",
    "NS": "#C00000",
}

print("=" * 100)
print("NN / NO / NS DATASET")
print("=" * 100)

display(
    nn_no_ns_df[
        [
            "Ligand",
            "LigandClass",
            "Yield",
        ]
    ]
)

for ligand_class in COMPARISON_CLASSES:
    print(
        f"{ligand_class} observations:",
        int(
            (
                nn_no_ns_df["LigandClass"]
                == ligand_class
            ).sum()
        ),
    )


# ============================================================
# 5. DESCRIPTOR GROUPS
# ============================================================

DESCRIPTOR_GROUPS = {
    "Coordination_geometry": [
        "Pd_Donor1_Distance_A",
        "Pd_Donor2_Distance_A",
        "Average_Pd_Donor_Distance_A",
        "Delta_Pd_Donor_Distance_A",
        "Pd_Donor_Distance_Ratio",
        "Donor1_Donor2_Distance_A",
        "Bite_Angle_deg",
        "Pd_Coordination_Distortion_A",
    ],

    "Donor_path_and_internal_geometry": [
        "LigandDonorPathAtomCount",
        "LigandDonorPathBondCount",
        "LigandDonorPathMeanBond_A",
        "LigandDonorPathSD_A",
        "LigandDonorPathCV",
        "Donor1_MeanBondAngle_deg",
        "Donor2_MeanBondAngle_deg",
        "AverageDonor_MeanBondAngle_deg",
        "DeltaDonor_MeanBondAngle_deg",
    ],

    "Global_and_distal_sterics": [
        "LigandMaximumExtension_A",
        "LigandSASA_A2",
        "LigandSASA_Volume_A3",
        "LigandMolecularVolume_A3",
        "LigandDistalVolumeOutside_3.5A_A3",
        "LigandDistalVolumeFractionOutside_3.5A",
        "LigandSolidAngle_sr",
        "LigandSolidConeAngle_deg",
        "LigandSolidG_percent",
        "Ligand_%Vbur_3.0A",
        "Ligand_%Vbur_4.0A",
        "Ligand_%Vbur_5.5A",
        "PdOrientedSterimol_B1_A",
        "PdOrientedSterimol_B5_A",
        "PdOrientedSterimol_L_A",
    ],

    "Ligand_contact_and_dispersion": [
        "LigandPint",
        "LigandPint_per_SASA",
        "LigandPint_per_Volume",
    ],

    "Electronic_xTB": [
        "XTB_LigandChargeSum",
        "XTB_HOMO_eV",
        "XTB_LUMO_eV",
        "XTB_HOMO_LUMO_Gap_eV",
        "XTB_FermiLevel_eV",
        "XTB_IP_eV",
        "XTB_EA_eV",
        "XTB_ChemicalPotential_eV",
        "XTB_Electronegativity_eV",
        "XTB_Hardness_eV",
        "XTB_Softness",
        "XTB_Electrophilicity",
        "XTB_Nucleophilicity",
        "XTB_Dipole_Debye",
        "XTB_Polarizability",
        "XTB_NFOD",
        "Donor1_MullikenCharge",
        "Donor2_MullikenCharge",
        "AverageDonorCharge",
        "DeltaDonorCharge",
        "DonorChargeSum",
    ],
}


available_descriptor_groups = {
    group_name: [
        descriptor
        for descriptor in descriptors
        if descriptor in nn_no_ns_df.columns
    ]
    for group_name, descriptors in (
        DESCRIPTOR_GROUPS.items()
    )
}

available_descriptor_groups = {
    group_name: descriptors
    for group_name, descriptors in (
        available_descriptor_groups.items()
    )
    if descriptors
}


# ============================================================
# 6. PAIRWISE CLASS COMPARISONS
# ============================================================

PAIRWISE_CLASS_COMPARISONS = [
    ("NN", "NO"),
    ("NN", "NS"),
    ("NO", "NS"),
]


def calculate_pairwise_comparison(
    dataframe,
    class_a,
    class_b,
    group_name,
    descriptor,
):
    values_a = pd.to_numeric(
        dataframe.loc[
            dataframe["LigandClass"] == class_a,
            descriptor,
        ],
        errors="coerce",
    ).dropna()

    values_b = pd.to_numeric(
        dataframe.loc[
            dataframe["LigandClass"] == class_b,
            descriptor,
        ],
        errors="coerce",
    ).dropna()

    if len(values_a) < 2 or len(values_b) < 2:
        return None

    mean_a = float(values_a.mean())
    mean_b = float(values_b.mean())

    median_a = float(values_a.median())
    median_b = float(values_b.median())

    difference_b_minus_a = mean_b - mean_a

    try:
        mannwhitney_u, mannwhitney_p = (
            mannwhitneyu(
                values_a,
                values_b,
                alternative="two-sided",
            )
        )
    except Exception:
        mannwhitney_u = np.nan
        mannwhitney_p = np.nan

    try:
        welch_t, welch_p = ttest_ind(
            values_a,
            values_b,
            equal_var=False,
            nan_policy="omit",
        )
    except Exception:
        welch_t = np.nan
        welch_p = np.nan

    if pd.notna(mannwhitney_u):
        effect_size_r = (
            1.0
            - (
                2.0
                * mannwhitney_u
                / (
                    len(values_a)
                    * len(values_b)
                )
            )
        )
    else:
        effect_size_r = np.nan

    return {
        "Group": group_name,
        "Descriptor": descriptor,
        "Class_A": class_a,
        "Class_B": class_b,
        "Class_A_N": len(values_a),
        "Class_B_N": len(values_b),
        "Class_A_Mean": mean_a,
        "Class_B_Mean": mean_b,
        "Class_A_Median": median_a,
        "Class_B_Median": median_b,
        "Difference_B_minus_A": difference_b_minus_a,
        "Abs_Mean_Difference": abs(
            difference_b_minus_a
        ),
        "MannWhitney_U": mannwhitney_u,
        "MannWhitney_p": mannwhitney_p,
        "Welch_t": welch_t,
        "Welch_p": welch_p,
        "EffectSize_r": effect_size_r,
    }


comparison_records = []

for group_name, descriptors in (
    available_descriptor_groups.items()
):

    for descriptor in descriptors:

        for class_a, class_b in (
            PAIRWISE_CLASS_COMPARISONS
        ):

            record = calculate_pairwise_comparison(
                dataframe=nn_no_ns_df,
                class_a=class_a,
                class_b=class_b,
                group_name=group_name,
                descriptor=descriptor,
            )

            if record is not None:
                comparison_records.append(
                    record
                )

nn_no_ns_comparison_df = pd.DataFrame(
    comparison_records
)


# ============================================================
# 7. FDR CORRECTION
# ============================================================

try:
    from statsmodels.stats.multitest import (
        multipletests
    )

    nn_no_ns_comparison_df[
        "MannWhitney_FDR_p"
    ] = np.nan

    nn_no_ns_comparison_df[
        "Welch_FDR_p"
    ] = np.nan

    valid_mannwhitney = (
        nn_no_ns_comparison_df[
            "MannWhitney_p"
        ].notna()
    )

    if valid_mannwhitney.any():
        nn_no_ns_comparison_df.loc[
            valid_mannwhitney,
            "MannWhitney_FDR_p",
        ] = multipletests(
            nn_no_ns_comparison_df.loc[
                valid_mannwhitney,
                "MannWhitney_p",
            ],
            method="fdr_bh",
        )[1]

    valid_welch = (
        nn_no_ns_comparison_df[
            "Welch_p"
        ].notna()
    )

    if valid_welch.any():
        nn_no_ns_comparison_df.loc[
            valid_welch,
            "Welch_FDR_p",
        ] = multipletests(
            nn_no_ns_comparison_df.loc[
                valid_welch,
                "Welch_p",
            ],
            method="fdr_bh",
        )[1]

except Exception as error:
    nn_no_ns_comparison_df[
        "FDR_Warning"
    ] = str(error)


nn_no_ns_comparison_df = (
    nn_no_ns_comparison_df
    .sort_values(
        [
            "MannWhitney_FDR_p",
            "Abs_Mean_Difference",
        ],
        ascending=[
            True,
            False,
        ],
        na_position="last",
    )
    .reset_index(drop=True)
)

print("=" * 100)
print("NN / NO / NS PAIRWISE DESCRIPTOR COMPARISON")
print("=" * 100)

display(nn_no_ns_comparison_df)


# ============================================================
# 8. CLASS-WISE SUMMARY TABLE
# ============================================================

class_summary_records = []

for group_name, descriptors in (
    available_descriptor_groups.items()
):

    for descriptor in descriptors:

        for ligand_class in COMPARISON_CLASSES:

            values = pd.to_numeric(
                nn_no_ns_df.loc[
                    nn_no_ns_df["LigandClass"]
                    == ligand_class,
                    descriptor,
                ],
                errors="coerce",
            ).dropna()

            if len(values) == 0:
                continue

            class_summary_records.append({
                "Group": group_name,
                "Descriptor": descriptor,
                "LigandClass": ligand_class,
                "N": len(values),
                "Mean": values.mean(),
                "SD": (
                    values.std(ddof=1)
                    if len(values) > 1
                    else np.nan
                ),
                "Median": values.median(),
                "Minimum": values.min(),
                "Maximum": values.max(),
            })

class_summary_df = pd.DataFrame(
    class_summary_records
)

print("=" * 100)
print("NN / NO / NS CLASS SUMMARY")
print("=" * 100)

display(class_summary_df)


# ============================================================
# 9. STANDARDIZED EFFECT SIZES
# ============================================================

# Standardize each descriptor using all NN + NO + NS values.
# Values are plotted as class distributions, not as a single
# class-to-class comparison.

standardized_records = []

for group_name, descriptors in (
    available_descriptor_groups.items()
):

    for descriptor in descriptors:

        values = pd.to_numeric(
            nn_no_ns_df[descriptor],
            errors="coerce",
        )

        combined_mean = values.mean()
        combined_sd = values.std(ddof=1)

        if (
            pd.isna(combined_sd)
            or combined_sd == 0
        ):
            continue

        for index, value in values.items():

            if pd.isna(value):
                continue

            standardized_records.append({
                "Group": group_name,
                "Descriptor": descriptor,
                "DescriptorLabel": descriptor,
                "Ligand": nn_no_ns_df.loc[
                    index,
                    "Ligand",
                ],
                "LigandClass": nn_no_ns_df.loc[
                    index,
                    "LigandClass",
                ],
                "StandardizedValue": (
                    value - combined_mean
                ) / combined_sd,
            })

standardized_long_df = pd.DataFrame(
    standardized_records
)


# ============================================================
# 10. READABLE DESCRIPTOR LABELS
# ============================================================

def readable_descriptor_label(
    descriptor
):
    labels = {
        "Delta_Pd_Donor_Distance_A":
            "Δ Pd–donor distance (Å)",

        "Pd_Donor_Distance_Ratio":
            "Pd–donor distance ratio",

        "Donor1_Donor2_Distance_A":
            "Donor–donor distance (Å)",

        "DeltaDonor_MeanBondAngle_deg":
            "Donor-angle difference (°)",

        "LigandDonorPathMeanBond_A":
            "Donor-path mean bond length (Å)",

        "Ligand_%Vbur_5.5A":
            "Ligand %Vbur (5.5 Å)",

        "LigandMolecularVolume_A3":
            "Ligand molecular volume (Å³)",

        "LigandSolidAngle_sr":
            "Ligand solid angle (sr)",

        "XTB_Polarizability":
            "XTB polarizability",
    }

    if descriptor in labels:
        return labels[descriptor]

    return str(descriptor).replace(
        "_",
        " ",
    )


standardized_long_df[
    "DescriptorLabel"
] = standardized_long_df[
    "Descriptor"
].apply(
    readable_descriptor_label
)


# ============================================================
# 11. PLOT FUNCTION WITH NN / NO / NS
# ============================================================

def plot_three_class_descriptor_panel(
    dataframe,
    descriptors,
    title,
    filename,
):
    plot_df = dataframe[
        dataframe["Descriptor"].isin(
            descriptors
        )
    ].copy()

    if len(plot_df) == 0:
        print(
            f"No valid data for {title}"
        )
        return plot_df

    descriptor_order = [
        descriptor
        for descriptor in descriptors
        if descriptor in plot_df[
            "Descriptor"
        ].unique()
    ]

    display_order = [
        readable_descriptor_label(
            descriptor
        )
        for descriptor in descriptor_order
    ][::-1]

    plot_df[
        "DescriptorLabel"
    ] = plot_df[
        "Descriptor"
    ].apply(
        readable_descriptor_label
    )

    plt.figure(
        figsize=(
            15,
            max(
                6,
                1.0 * len(display_order),
            ),
        )
    )

    # Neutral combined boxplots.
    ax = sns.boxplot(
        data=plot_df,
        y="DescriptorLabel",
        x="StandardizedValue",
        order=display_order,
        color="#E6E6E6",
        width=0.62,
        linewidth=1.4,
        fliersize=0,
        boxprops={
            "facecolor": "#E6E6E6",
            "edgecolor": "black",
        },
        whiskerprops={
            "color": "black",
            "linewidth": 1.2,
        },
        capprops={
            "color": "black",
            "linewidth": 1.2,
        },
        medianprops={
            "color": "black",
            "linewidth": 1.5,
        },
    )

    y_position_map = {
        label: position
        for position, label in enumerate(
            display_order
        )
    }

    class_offsets = {
        "NN": -0.22,
        "NO": 0.00,
        "NS": 0.22,
    }

    for descriptor_label in display_order:

        descriptor_data = plot_df[
            plot_df["DescriptorLabel"]
            == descriptor_label
        ]

        y_position = y_position_map[
            descriptor_label
        ]

        for ligand_class in COMPARISON_CLASSES:

            class_data = descriptor_data[
                descriptor_data["LigandClass"]
                == ligand_class
            ].copy()

            if len(class_data) == 0:
                continue

            color = CLASS_COLORS[
                ligand_class
            ]

            vertical_offset = class_offsets[
                ligand_class
            ]

            class_mean = class_data[
                "StandardizedValue"
            ].mean()

            # Black diamond = class mean.
            ax.scatter(
                class_mean,
                y_position + vertical_offset,
                marker="D",
                s=55,
                color="black",
                edgecolor="white",
                linewidth=0.6,
                zorder=9,
            )

            for _, row in class_data.iterrows():

                ligand_number = ligand_sort_number(
                    row["Ligand"]
                )

                horizontal_jitter = (
                    (ligand_number % 5) - 2
                ) * 0.025

                x_value = (
                    row["StandardizedValue"]
                    + horizontal_jitter
                )

                # Colored point.
                ax.scatter(
                    x_value,
                    y_position + vertical_offset,
                    s=70,
                    color=color,
                    edgecolor="black",
                    linewidth=0.7,
                    zorder=6,
                )

                # Black label with white background.
                ax.annotate(
                    str(row["Ligand"]),
                    (
                        x_value,
                        y_position + vertical_offset,
                    ),
                    xytext=(0, 9),
                    textcoords="offset points",
                    ha="center",
                    va="bottom",
                    fontsize=8,
                    color="black",
                    fontweight="bold",
                    bbox={
                        "boxstyle": "round,pad=0.15",
                        "facecolor": "white",
                        "edgecolor": color,
                        "linewidth": 0.8,
                        "alpha": 0.95,
                    },
                    zorder=10,
                )

    # Legend.
    for ligand_class in COMPARISON_CLASSES:
        ax.scatter(
            [],
            [],
            s=70,
            color=CLASS_COLORS[
                ligand_class
            ],
            edgecolor="black",
            label=f"{ligand_class} ligands",
        )

    ax.scatter(
        [],
        [],
        marker="D",
        s=55,
        color="black",
        label="Class mean",
    )

    ax.axvline(
        0,
        color="black",
        linewidth=0.9,
        linestyle="--",
    )

    ax.set_title(
        title,
        fontsize=14,
        fontweight="bold",
    )

    ax.set_xlabel(
        "Standardized descriptor value "
        "(NN + NO + NS scale)"
    )

    ax.set_ylabel("")

    ax.legend(
        loc="lower right",
        frameon=True,
    )

    ax.grid(
        axis="x",
        alpha=0.25,
    )

    plt.tight_layout()

    output_path = os.path.join(
        PLOT_FOLDER,
        filename,
    )

    plt.savefig(
        output_path,
        dpi=600,
        bbox_inches="tight",
        facecolor="white",
    )

    plt.show()
    plt.close()

    print(
        "Saved:",
        output_path,
    )

    return plot_df


# ============================================================
# 12. COORDINATION PLOT
# ============================================================

coordination_plot_df = (
    plot_three_class_descriptor_panel(
        dataframe=standardized_long_df,
        descriptors=[
            "Delta_Pd_Donor_Distance_A",
            "Pd_Donor_Distance_Ratio",
            "Donor1_Donor2_Distance_A",
            "DeltaDonor_MeanBondAngle_deg",
            "LigandDonorPathMeanBond_A",
        ],
        title=(
            "NN vs NO vs NS: "
            "Pd Coordination and Donor-Path Geometry"
        ),
        filename=(
            "NN_NO_NS_Coordination_Clear_Labels.png"
        ),
    )
)


# ============================================================
# 13. STERIC / ELECTRONIC PLOT
# ============================================================

steric_electronic_plot_df = (
    plot_three_class_descriptor_panel(
        dataframe=standardized_long_df,
        descriptors=[
            "Ligand_%Vbur_5.5A",
            "LigandMolecularVolume_A3",
            "LigandSolidAngle_sr",
            "XTB_Polarizability",
        ],
        title=(
            "NN vs NO vs NS: "
            "Steric and Electronic Properties"
        ),
        filename=(
            "NN_NO_NS_Steric_Electronic_Clear_Labels.png"
        ),
    )
)


# ============================================================
# 14. YIELD PLOT
# ============================================================

plt.figure(figsize=(8, 6))

sns.boxplot(
    data=nn_no_ns_df,
    x="LigandClass",
    y="Yield",
    order=COMPARISON_CLASSES,
    color="#E6E6E6",
    width=0.55,
    linewidth=1.4,
    fliersize=0,
    boxprops={
        "facecolor": "#E6E6E6",
        "edgecolor": "black",
    },
    medianprops={
        "color": "black",
        "linewidth": 1.5,
    },
)

for ligand_class in COMPARISON_CLASSES:

    class_data = nn_no_ns_df[
        nn_no_ns_df["LigandClass"]
        == ligand_class
    ]

    x_position = (
        COMPARISON_CLASSES.index(
            ligand_class
        )
    )

    color = CLASS_COLORS[
        ligand_class
    ]

    for _, row in class_data.iterrows():

        ligand_number = ligand_sort_number(
            row["Ligand"]
        )

        x_jitter = (
            (ligand_number % 5) - 2
        ) * 0.025

        x_value = x_position + x_jitter

        plt.scatter(
            x_value,
            row["Yield"],
            s=80,
            color=color,
            edgecolor="black",
            linewidth=0.7,
            zorder=6,
        )

        plt.annotate(
            str(row["Ligand"]),
            (
                x_value,
                row["Yield"],
            ),
            xytext=(0, 9),
            textcoords="offset points",
            ha="center",
            va="bottom",
            fontsize=8,
            color="black",
            fontweight="bold",
            bbox={
                "boxstyle": "round,pad=0.15",
                "facecolor": "white",
                "edgecolor": color,
                "linewidth": 0.8,
                "alpha": 0.95,
            },
        )

plt.xlabel("Ligand class")
plt.ylabel(
    "Unsaturated γ-arylated lactonization yield"
)
plt.title(
    "NN versus NO versus NS: "
    "Unsaturated γ-Arylated Lactonization Yield",
    fontweight="bold",
)

plt.grid(
    axis="y",
    alpha=0.25,
)

plt.tight_layout()

yield_plot_path = os.path.join(
    PLOT_FOLDER,
    "NN_NO_NS_Yield_Clear_Labels.png",
)

plt.savefig(
    yield_plot_path,
    dpi=600,
    bbox_inches="tight",
    facecolor="white",
)

plt.show()
plt.close()

print(
    "Saved:",
    yield_plot_path,
)


# ============================================================
# 15. CLASS-WISE YIELD SUMMARY
# ============================================================

yield_summary_df = (
    nn_no_ns_df
    .groupby("LigandClass")["Yield"]
    .agg(
        N="count",
        Mean="mean",
        Median="median",
        SD="std",
        Minimum="min",
        Maximum="max",
    )
    .reset_index()
)

print("=" * 100)
print("NN / NO / NS YIELD SUMMARY")
print("=" * 100)

display(yield_summary_df)


# ============================================================
# 16. EXPORT RESULTS
# ============================================================

reviewer_output_excel = os.path.join(
    OUTPUT_FOLDER,
    "NN_NO_NS_Reviewer_Analysis.xlsx",
)

with pd.ExcelWriter(
    reviewer_output_excel,
    engine="openpyxl",
) as writer:

    nn_no_ns_df.to_excel(
        writer,
        sheet_name="NN_NO_NS_Raw_Data",
        index=False,
    )

    nn_no_ns_comparison_df.to_excel(
        writer,
        sheet_name="Pairwise_Statistics",
        index=False,
    )

    class_summary_df.to_excel(
        writer,
        sheet_name="Class_Summaries",
        index=False,
    )

    yield_summary_df.to_excel(
        writer,
        sheet_name="Yield_Summary",
        index=False,
    )

    standardized_long_df.to_excel(
        writer,
        sheet_name="Standardized_Plot_Data",
        index=False,
    )

    if isinstance(
        coordination_plot_df,
        pd.DataFrame,
    ):
        coordination_plot_df.to_excel(
            writer,
            sheet_name="Coordination_Plot_Data",
            index=False,
        )

    if isinstance(
        steric_electronic_plot_df,
        pd.DataFrame,
    ):
        steric_electronic_plot_df.to_excel(
            writer,
            sheet_name="Steric_Electronic_Plot_Data",
            index=False,
        )

print("=" * 100)
print("NN / NO / NS REVIEWER ANALYSIS COMPLETE")
print("=" * 100)

print(
    "Reviewer workbook:",
    reviewer_output_excel,
)

print(
    "Plots folder:",
    PLOT_FOLDER,
)
```

```python
# ============================================================
# PUBLICATION-STYLE NN / NO / NS SUMMARY TABLE
# ============================================================

comparison_classes = [
    "NN",
    "NO",
    "NS",
]

comparison_df = master_df[
    master_df["LigandClass"].isin(
        comparison_classes
    )
].copy()

comparison_df["Yield"] = pd.to_numeric(
    comparison_df["Yield"],
    errors="coerce",
)

DESCRIPTOR_LABELS = {
    "Yield": "Mean yield (%)",
    "Pd_Donor2_Distance_A": (
        "Pd–donor 2 distance (Å)"
    ),
    "Delta_Pd_Donor_Distance_A": (
        "ΔPd–donor distance (Å)"
    ),
    "Pd_Coordination_Distortion_A": (
        "Pd coordination distortion (Å)"
    ),
    "Donor1_Donor2_Distance_A": (
        "Donor–donor distance (Å)"
    ),
    "Bite_Angle_deg": (
        "Bite angle (°)"
    ),
    "LigandDonorPathMeanBond_A": (
        "Donor-path mean bond length (Å)"
    ),
    "DeltaDonor_MeanBondAngle_deg": (
        "Donor-angle difference (°)"
    ),
    "Ligand_%Vbur_5.5A": (
        "Ligand %Vbur (5.5 Å)"
    ),
    "LigandMolecularVolume_A3": (
        "Molecular volume (Å³)"
    ),
    "XTB_Polarizability": (
        "XTB polarizability"
    ),
}

STRUCTURAL_INTERPRETATIONS = {
    "Yield": (
        "Class-level catalytic performance"
    ),
    "Pd_Donor2_Distance_A": (
        "Longer secondary donor interaction in NS"
    ),
    "Delta_Pd_Donor_Distance_A": (
        "Greater donor-distance asymmetry in NS"
    ),
    "Pd_Coordination_Distortion_A": (
        "More asymmetric coordination environment"
    ),
    "Donor1_Donor2_Distance_A": (
        "More open chelating pocket"
    ),
    "Bite_Angle_deg": (
        "Modified chelate geometry"
    ),
    "LigandDonorPathMeanBond_A": (
        "Distinct donor-framework geometry"
    ),
    "DeltaDonor_MeanBondAngle_deg": (
        "Greater donor asymmetry"
    ),
    "Ligand_%Vbur_5.5A": (
        "Greater distal steric occupation"
    ),
    "LigandMolecularVolume_A3": (
        "Larger ligand environment"
    ),
    "XTB_Polarizability": (
        "Greater electronic deformability"
    ),
}

DESCRIPTORS_FOR_TABLE = [
    descriptor
    for descriptor in DESCRIPTOR_LABELS
    if descriptor in comparison_df.columns
]


def format_mean_sd(values):
    """
    Return mean ± SD for a class.

    For a single valid observation, SD is reported as NA.
    """
    values = pd.to_numeric(
        values,
        errors="coerce",
    ).dropna()

    if len(values) == 0:
        return "NA"

    mean_value = values.mean()

    if len(values) == 1:
        return f"{mean_value:.3f} ± NA"

    standard_deviation = values.std(
        ddof=1
    )

    return (
        f"{mean_value:.3f} ± "
        f"{standard_deviation:.3f}"
    )


publication_records = []

for descriptor in DESCRIPTORS_FOR_TABLE:

    class_values = {}

    for ligand_class in comparison_classes:
        class_values[ligand_class] = pd.to_numeric(
            comparison_df.loc[
                comparison_df["LigandClass"]
                == ligand_class,
                descriptor,
            ],
            errors="coerce",
        ).dropna()

    class_means = {}

    for ligand_class, values in class_values.items():
        class_means[ligand_class] = (
            values.mean()
            if len(values) > 0
            else np.nan
        )

    nn_mean = class_means["NN"]
    no_mean = class_means["NO"]
    ns_mean = class_means["NS"]

    publication_records.append({
        "Descriptor": DESCRIPTOR_LABELS[
            descriptor
        ],
        "NN ligands, mean ± SD": format_mean_sd(
            class_values["NN"]
        ),
        "NO ligands, mean ± SD": format_mean_sd(
            class_values["NO"]
        ),
        "NS ligands, mean ± SD": format_mean_sd(
            class_values["NS"]
        ),
        "NO − NN": (
            no_mean - nn_mean
            if pd.notna(no_mean)
            and pd.notna(nn_mean)
            else np.nan
        ),
        "NS − NN": (
            ns_mean - nn_mean
            if pd.notna(ns_mean)
            and pd.notna(nn_mean)
            else np.nan
        ),
        "NS − NO": (
            ns_mean - no_mean
            if pd.notna(ns_mean)
            and pd.notna(no_mean)
            else np.nan
        ),
        "Structural implication": (
            STRUCTURAL_INTERPRETATIONS[
                descriptor
            ]
        ),
    })

nn_no_ns_publication_table = pd.DataFrame(
    publication_records
)

print("=" * 100)
print("PUBLICATION-STYLE NN / NO / NS TABLE")
print("=" * 100)

display(nn_no_ns_publication_table)
```
