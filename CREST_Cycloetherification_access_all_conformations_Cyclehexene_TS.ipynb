```python
# ============================================================
# FAST TARGETED CYCLOHEXENE-TYPE TS PUCKERING SCAN
#
# Ring:
#       O17 - C18 - C19 - C20 = C21 - C22
#
# Forming bond:
#       O17 - C22
#
# Goal:
#   Start from the lowest-energy existing BOAT conformer
#   and rapidly probe:
#
#       Boat
#       Twist-boat
#       Half-chair
#       Half-boat
#       Chair-like
#       intermediate/distorted forms
#
# IMPORTANT:
#   This is NOT a full CREST search.
#   It is a targeted geometric PES scan.
# ============================================================


# ============================================================
# 1. INSTALL xTB
# ============================================================

!apt-get -qq update
!apt-get -qq install -y xtb

!pip -q install numpy pandas matplotlib scipy


# ============================================================
# 2. IMPORTS
# ============================================================

import os
import re
import glob
import shutil
import subprocess
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt

from pathlib import Path


# ============================================================
# 3. GOOGLE DRIVE
# ============================================================

from google.colab import drive

drive.mount("/content/drive")


BASE = Path(
    "/content/drive/MyDrive/Cycloetherification_xyz_files"
)

CREST_ROOT = BASE / "CREST_results"

OUTPUT_ROOT = (
    CREST_ROOT /
    "FAST_Cyclohexene_PES"
)

OUTPUT_ROOT.mkdir(
    parents=True,
    exist_ok=True
)


# ============================================================
# 4. USER SETTINGS
# ============================================================

# ------------------------------------------------------------
# Process one mother TS or all mothers
# ------------------------------------------------------------

MOTHER_TS = None

# Example:
#
# MOTHER_TS = "RR_CH2-Ph"
#
# or:
#
# MOTHER_TS = "RR_tBu"
#
# None = all mothers


# ------------------------------------------------------------
# SIX-MEMBERED RING
#
# O17-C18-C19-C20-C21-C22
# ------------------------------------------------------------

RING_ATOMS = [
    17,
    18,
    19,
    20,
    21,
    22
]


# ------------------------------------------------------------
# DOUBLE BOND
#
# C20=C21
# ------------------------------------------------------------

DOUBLE_BOND = (
    20,
    21
)


# ------------------------------------------------------------
# FORMING BOND
#
# O17-C22
# ------------------------------------------------------------

FORMING_BOND = (
    17,
    22
)


# ------------------------------------------------------------
# Core atoms
#
# Useful if alignment is required later.
# ------------------------------------------------------------

CORE_ATOMS = list(range(1, 44))


# ------------------------------------------------------------
# Puckering amplitude scan
#
# Smaller = less distortion
# Larger  = stronger ring puckering
#
# These are geometric perturbations, NOT barriers.
# ------------------------------------------------------------

Q_SCALE_FACTORS = [
    0.60,
    0.80,
    1.00,
    1.20,
    1.40
]


# ------------------------------------------------------------
# Number of phase points
#
# 12 phase points cover the puckering cycle.
# ------------------------------------------------------------

N_PHASES = 12


# ------------------------------------------------------------
# xTB settings
# ------------------------------------------------------------

XTB_TIMEOUT = 300


print("BASE =", BASE)
print("CREST_ROOT =", CREST_ROOT)
print("OUTPUT_ROOT =", OUTPUT_ROOT)

print("\nRing =", RING_ATOMS)
print("Double bond =", DOUBLE_BOND)
print("Forming bond =", FORMING_BOND)


# ============================================================
# 5. CHECK xTB
# ============================================================

XTB = shutil.which("xtb")

if XTB is None:
    raise RuntimeError(
        "xTB was not found."
    )

print("\nxTB executable:", XTB)

subprocess.run(
    [XTB, "--version"],
    check=False
)


# ============================================================
# 6. XYZ READER
# ============================================================

def read_xyz(path):

    path = Path(path)

    with open(path) as f:

        lines = [
            line.rstrip()
            for line in f
        ]

    n = int(lines[0])

    comment = lines[1]

    atoms = []
    coords = []

    for line in lines[2:2+n]:

        p = line.split()

        atoms.append(p[0])

        coords.append([
            float(p[1]),
            float(p[2]),
            float(p[3])
        ])

    return (
        atoms,
        np.asarray(coords, dtype=float),
        comment
    )


# ============================================================
# 7. XYZ WRITER
# ============================================================

def write_xyz(
    path,
    atoms,
    coords,
    comment=""
):

    path = Path(path)

    with open(path, "w") as f:

        f.write(
            f"{len(atoms)}\n"
        )

        f.write(
            comment + "\n"
        )

        for atom, xyz in zip(
            atoms,
            coords
        ):

            f.write(
                f"{atom:2s} "
                f"{xyz[0]: .8f} "
                f"{xyz[1]: .8f} "
                f"{xyz[2]: .8f}\n"
            )


# ============================================================
# 8. FIND MOTHER TS FILES
# ============================================================

def is_generated_file(path):

    name = Path(path).name

    bad = [
        "_retained_",
        "_raw_crest",
        "_crest",
        "_aligned",
        "_SUPERIMPOSED"
    ]

    return any(
        x in name
        for x in bad
    )


def find_mothers():

    files = []

    for p in sorted(
        BASE.glob("*.xyz")
    ):

        if not is_generated_file(p):

            files.append(p)

    return files


MOTHER_FILES = find_mothers()

print(
    "\nMother TS files:",
    len(MOTHER_FILES)
)

for p in MOTHER_FILES:

    print(
        " ",
        p.name
    )


# ============================================================
# 9. FIND ONLY INDIVIDUAL RETAINED STRUCTURES
# ============================================================

def find_retained_files(
    mother_name
):

    # IMPORTANT:
    #
    # This deliberately matches:
    #
    # retained_001.xyz
    # retained_002.xyz
    #
    # but NOT:
    #
    # retained_ensemble.xyz

    pattern = (
        CREST_ROOT /
        "**" /
        f"{mother_name}_retained_[0-9][0-9][0-9].xyz"
    )

    files = sorted(
        glob.glob(
            str(pattern),
            recursive=True
        )
    )

    return [
        Path(x)
        for x in files
    ]


# ============================================================
# 10. GEOMETRY FUNCTIONS
# ============================================================

def distance(
    coords,
    a,
    b
):

    return float(
        np.linalg.norm(
            coords[a-1]
            -
            coords[b-1]
        )
    )


def ring_centroid(
    coords
):

    idx = np.asarray(
        RING_ATOMS
    ) - 1

    return coords[idx].mean(
        axis=0
    )


# ============================================================
# 11. RING PLANE / NORMAL
# ============================================================

def ring_normal(
    coords
):

    idx = np.asarray(
        RING_ATOMS
    ) - 1

    ring = coords[idx]

    centered = (
        ring -
        ring.mean(axis=0)
    )

    _, _, vh = np.linalg.svd(
        centered
    )

    normal = vh[-1]

    normal /= np.linalg.norm(
        normal
    )

    return normal


# ============================================================
# 12. CREMER–POPLE
# ============================================================

def cremer_pople(
    coords
):

    idx = np.asarray(
        RING_ATOMS
    ) - 1

    ring = coords[idx].copy()

    ring -= ring.mean(
        axis=0
    )

    _, _, vh = np.linalg.svd(
        ring,
        full_matrices=False
    )

    normal = vh[-1]

    if normal[2] < 0:

        normal = -normal

    z = ring @ normal

    j = np.arange(6)

    q2c = (
        np.sqrt(1/3)
        *
        np.sum(
            z *
            np.cos(
                2*np.pi*j/3
            )
        )
    )

    q2s = (
        np.sqrt(1/3)
        *
        np.sum(
            z *
            np.sin(
                2*np.pi*j/3
            )
        )
    )

    q3 = (
        np.sqrt(1/6)
        *
        np.sum(
            z *
            ((-1)**j)
        )
    )

    q2 = np.sqrt(
        q2c**2 +
        q2s**2
    )

    Q = np.sqrt(
        q2**2 +
        q3**2
    )

    if Q < 1e-10:

        return (
            Q,
            np.nan,
            np.nan
        )

    theta = np.degrees(
        np.arccos(
            np.clip(
                q3/Q,
                -1,
                1
            )
        )
    )

    phi = np.degrees(
        np.arctan2(
            q2s,
            q2c
        )
    )

    phi %= 360

    return (
        float(Q),
        float(theta),
        float(phi)
    )


# ============================================================
# 13. SIMPLE PUCKERING REGION
# ============================================================

def pucker_region(
    theta,
    phi
):

    if not np.isfinite(theta):

        return "Undefined"

    # Chair-like
    if (
        theta < 20
        or
        theta > 160
    ):

        return "Chair-like"


    # Boat / twist-boat region
    #
    # For a six-membered ring these lie around
    # the equatorial part of CP space.

    if 70 <= theta <= 110:

        # Canonical phase sectors
        # are approximate because atom ordering
        # defines the phase origin.

        phase = (
            round(
                phi / 60
            )
            * 60
        ) % 360

        phase_distance = min(
            abs(phi - phase),
            360 - abs(phi - phase)
        )

        if phase_distance <= 15:

            return "Boat-region"

        return "Twist-boat-region"


    # Half-chair / half-boat region
    if (
        40 <= theta <= 70
        or
        110 <= theta <= 140
    ):

        return "Half-chair / half-boat region"


    return "Intermediate"


# ============================================================
# 14. XTB SINGLE-POINT
# ============================================================

def xtb_energy(
    xyz_file,
    workdir
):

    workdir = Path(workdir)

    workdir.mkdir(
        parents=True,
        exist_ok=True
    )

    local_xyz = (
        workdir /
        "structure.xyz"
    )

    shutil.copy2(
        xyz_file,
        local_xyz
    )

    cmd = [
        XTB,
        "structure.xyz",
        "--gfn2",
        "--sp"
    ]

    result = subprocess.run(
        cmd,
        cwd=workdir,
        stdout=subprocess.PIPE,
        stderr=subprocess.STDOUT,
        text=True,
        timeout=XTB_TIMEOUT
    )

    output = result.stdout

    patterns = [
        r"TOTAL ENERGY\s+"
        r"([-+]?\d+\.\d+(?:[Ee][-+]?\d+)?)",

        r"::\s*total energy\s+"
        r"([-+]?\d+\.\d+(?:[Ee][-+]?\d+)?)"
    ]

    energy_Eh = None

    for pattern in patterns:

        m = re.findall(
            pattern,
            output,
            flags=re.I
        )

        if m:

            energy_Eh = float(
                m[-1]
            )

            break

    if energy_Eh is None:

        raise RuntimeError(
            "Could not extract xTB energy."
        )

    energy_kcal = (
        energy_Eh *
        627.509474
    )

    return (
        energy_Eh,
        energy_kcal
    )


# ============================================================
# 15. IMPORTANT:
#     GENERATE PUCKERING DISTORTIONS
# ============================================================

def generate_pucker(
    coords,
    amplitude,
    phase_deg
):

    """
    Generate a controlled ring puckering
    from the selected Boat structure.

    The entire molecule is NOT rotated independently.

    Only the six ring atoms are displaced along
    the instantaneous ring normal.

    This is a targeted geometric scan.

    It does NOT claim that the resulting structure
    is an optimized minimum.
    """

    new = coords.copy()

    idx = np.asarray(
        RING_ATOMS
    ) - 1

    normal = ring_normal(
        coords
    )

    phase = np.radians(
        phase_deg
    )

    j = np.arange(6)

    # Two six-membered-ring puckering components
    displacement = (
        np.cos(
            2*np.pi*j/3
            + phase
        )
        *
        amplitude
    )

    # Add alternating component
    displacement += (
        0.50 *
        amplitude *
        ((-1)**j)
    )

    for atom_index, dz in zip(
        idx,
        displacement
    ):

        new[atom_index] += (
            dz *
            normal
        )

    return new


# ============================================================
# 16. CREATE TARGETED CANDIDATES
# ============================================================

def generate_candidates(
    seed_coords
):

    candidates = []

    # --------------------------------------------
    # Determine natural puckering amplitude
    # from seed CP Q
    # --------------------------------------------

    Q0, theta0, phi0 = \
        cremer_pople(
            seed_coords
        )

    # A modest amplitude based on seed geometry.
    #
    # We deliberately avoid huge distortions.

    base_amp = max(
        0.10,
        min(
            0.40,
            0.50 * Q0
        )
    )

    phases = np.linspace(
        0,
        360,
        N_PHASES,
        endpoint=False
    )

    for scale in Q_SCALE_FACTORS:

        amp = (
            base_amp *
            scale
        )

        for phase in phases:

            coords = generate_pucker(
                seed_coords,
                amp,
                phase
            )

            Q, theta, phi = \
                cremer_pople(
                    coords
                )

            candidates.append({

                "amplitude":
                    amp,

                "phase_deg":
                    phase,

                "coords":
                    coords,

                "Q_A":
                    Q,

                "theta_deg":
                    theta,

                "phi_deg":
                    phi,

                "pucker_region":
                    pucker_region(
                        theta,
                        phi
                    )

            })

    return candidates


# ============================================================
# 17. REMOVE GEOMETRIC DUPLICATES
# ============================================================

def geometry_rmsd(
    A,
    B
):

    idx = (
        np.asarray(
            RING_ATOMS
        ) - 1
    )

    A = A[idx]
    B = B[idx]

    A -= A.mean(
        axis=0
    )

    B -= B.mean(
        axis=0
    )

    H = A.T @ B

    U, S, Vt = np.linalg.svd(
        H
    )

    d = np.linalg.det(
        Vt.T @ U.T
    )

    if d < 0:

        Vt[-1] *= -1

    R = Vt.T @ U.T

    Afit = A @ R

    return np.sqrt(
        np.mean(
            np.sum(
                (Afit-B)**2,
                axis=1
            )
        )
    )


def deduplicate(
    candidates,
    cutoff=0.15
):

    kept = []

    for candidate in candidates:

        duplicate = False

        for old in kept:

            if geometry_rmsd(
                candidate["coords"],
                old["coords"]
            ) < cutoff:

                duplicate = True
                break

        if not duplicate:

            kept.append(
                candidate
            )

    return kept


# ============================================================
# 18. ANALYZE EXISTING RETAINED STRUCTURES
# ============================================================

def analyze_existing(
    mother_name,
    mother_file
):

    files = find_retained_files(
        mother_name
    )

    print(
        f"\n{mother_name}: "
        f"{len(files)} individual retained structures"
    )

    records = []

    for xyz in files:

        atoms, coords, comment = \
            read_xyz(xyz)

        Q, theta, phi = \
            cremer_pople(
                coords
            )

        records.append({

            "filename":
                xyz.name,

            "path":
                str(xyz),

            "Q_A":
                Q,

            "theta_deg":
                theta,

            "phi_deg":
                phi,

            "pucker_region":
                pucker_region(
                    theta,
                    phi
                ),

            "coords":
                coords,

            "atoms":
                atoms

        })

    return records


# ============================================================
# 19. CALCULATE EXISTING xTB ENERGIES
# ============================================================

def energy_existing(
    records,
    mother_output
):

    energy_dir = (
        mother_output /
        "Existing_xTB_energy"
    )

    energy_dir.mkdir(
        parents=True,
        exist_ok=True
    )

    output = []

    for i, rec in enumerate(
        records,
        start=1
    ):

        xyz = (
            energy_dir /
            f"existing_{i:03d}.xyz"
        )

        write_xyz(
            xyz,
            rec["atoms"],
            rec["coords"],
            "Fresh_GFN2_xTB"
        )

        E_Eh, E_kcal = \
            xtb_energy(
                xyz,
                energy_dir /
                f"run_{i:03d}"
            )

        row = {
            key: value
            for key, value in rec.items()
            if key not in [
                "coords",
                "atoms"
            ]
        }

        row[
            "energy_Eh"
        ] = E_Eh

        row[
            "energy_kcal_mol"
        ] = E_kcal

        output.append(row)

    df = pd.DataFrame(
        output
    )

    Emin = df[
        "energy_kcal_mol"
    ].min()

    df[
        "DeltaE_kcal_mol"
    ] = (
        df["energy_kcal_mol"]
        -
        Emin
    )

    return df


# ============================================================
# 20. FIND LOWEST-ENERGY BOAT
# ============================================================

def choose_boat(
    df
):

    boat = df[
        df[
            "pucker_region"
        ].str.contains(
            "Boat",
            na=False
        )
    ].copy()

    if len(boat) == 0:

        return None

    boat = boat.sort_values(
        "energy_kcal_mol"
    )

    return boat.iloc[0]


# ============================================================
# 21. PROCESS ONE MOTHER TS
# ============================================================

def process_mother(
    mother_file
):

    mother_name = (
        mother_file.stem
    )

    print("\n")
    print("=" * 75)
    print(
        "PROCESSING:",
        mother_name
    )
    print("=" * 75)

    mother_output = (
        OUTPUT_ROOT /
        mother_name
    )

    mother_output.mkdir(
        parents=True,
        exist_ok=True
    )


    # --------------------------------------------------------
    # Existing structures
    # --------------------------------------------------------

    records = analyze_existing(
        mother_name,
        mother_file
    )

    if len(records) == 0:

        print(
            "No retained structures."
        )

        return


    # --------------------------------------------------------
    # Fresh xTB energies
    # --------------------------------------------------------

    existing_df = energy_existing(
        records,
        mother_output
    )


    existing_df.to_csv(
        mother_output /
        "EXISTING_RETAINED_FRESH_XTB.csv",
        index=False
    )


    print("\nExisting structures:")
    display(
        existing_df[
            [
                "filename",
                "energy_kcal_mol",
                "DeltaE_kcal_mol",
                "Q_A",
                "theta_deg",
                "phi_deg",
                "pucker_region"
            ]
        ].sort_values(
            "DeltaE_kcal_mol"
        )
    )


    # --------------------------------------------------------
    # Lowest-energy Boat
    # --------------------------------------------------------

    boat = choose_boat(
        existing_df
    )

    if boat is None:

        print(
            "\nNo Boat-region structure "
            "was identified."
        )

        return


    print("\n")
    print("=" * 75)
    print("LOWEST-ENERGY BOAT")
    print("=" * 75)

    print(
        "File:",
        boat["filename"]
    )

    print(
        "Energy:",
        boat["energy_kcal_mol"],
        "kcal/mol"
    )

    print(
        "ΔE:",
        boat["DeltaE_kcal_mol"],
        "kcal/mol"
    )

    print(
        "Q:",
        boat["Q_A"]
    )

    print(
        "theta:",
        boat["theta_deg"]
    )

    print(
        "phi:",
        boat["phi_deg"]
    )


    # --------------------------------------------------------
    # Save Boat seed
    # --------------------------------------------------------

    seed_dir = (
        mother_output /
        "LOWEST_BOAT_SEED"
    )

    seed_dir.mkdir(
        parents=True,
        exist_ok=True
    )

    seed_path = (
        seed_dir /
        "lowest_energy_Boat.xyz"
    )

    # Locate original
    original_seed = Path(
        boat["path"]
    )

    atoms, seed_coords, comment = \
        read_xyz(
            original_seed
        )

    write_xyz(
        seed_path,
        atoms,
        seed_coords,
        "LOWEST_ENERGY_BOAT_SEED"
    )


    # --------------------------------------------------------
    # Generate targeted puckering candidates
    # --------------------------------------------------------

    candidates = generate_candidates(
        seed_coords
    )

    print(
        "\nGenerated candidates:",
        len(candidates)
    )


    # --------------------------------------------------------
    # Deduplicate
    # --------------------------------------------------------

    candidates = deduplicate(
        candidates,
        cutoff=0.15
    )

    print(
        "After geometric deduplication:",
        len(candidates)
    )


    # --------------------------------------------------------
    # Candidate directory
    # --------------------------------------------------------

    candidate_dir = (
        mother_output /
        "TARGETED_PUCKERING_STRUCTURES"
    )

    candidate_dir.mkdir(
        parents=True,
        exist_ok=True
    )


    # --------------------------------------------------------
    # xTB energy for every candidate
    # --------------------------------------------------------

    energy_dir = (
        mother_output /
        "TARGETED_xTB"
    )

    energy_dir.mkdir(
        parents=True,
        exist_ok=True
    )


    results = []


    for i, candidate in enumerate(
        candidates,
        start=1
    ):

        xyz = (
            candidate_dir /
            f"candidate_{i:04d}.xyz"
        )

        comment = (
            f"targeted_pucker "
            f"amplitude={candidate['amplitude']:.5f} "
            f"phase={candidate['phase_deg']:.2f}"
        )

        write_xyz(
            xyz,
            atoms,
            candidate["coords"],
            comment
        )


        # xTB energy

        try:

            E_Eh, E_kcal = \
                xtb_energy(
                    xyz,
                    energy_dir /
                    f"candidate_{i:04d}"
                )

        except Exception as e:

            print(
                "xTB FAILED:",
                i,
                e
            )

            E_Eh = np.nan
            E_kcal = np.nan


        results.append({

            "candidate":
                i,

            "amplitude_A":
                candidate["amplitude"],

            "phase_deg":
                candidate["phase_deg"],

            "energy_Eh":
                E_Eh,

            "energy_kcal_mol":
                E_kcal,

            "Q_A":
                candidate["Q_A"],

            "theta_deg":
                candidate["theta_deg"],

            "phi_deg":
                candidate["phi_deg"],

            "pucker_region":
                candidate["pucker_region"],

            "xyz_file":
                str(xyz)

        })


    candidate_df = pd.DataFrame(
        results
    )


    # --------------------------------------------------------
    # Relative energy
    # --------------------------------------------------------

    valid = candidate_df[
        "energy_kcal_mol"
    ].notna()

    Emin = candidate_df.loc[
        valid,
        "energy_kcal_mol"
    ].min()

    candidate_df[
        "DeltaE_kcal_mol"
    ] = (
        candidate_df[
            "energy_kcal_mol"
        ]
        -
        Emin
    )


    # --------------------------------------------------------
    # Energy relative to LOWEST BOAT
    # --------------------------------------------------------

    boat_energy = boat[
        "energy_kcal_mol"
    ]

    candidate_df[
        "DeltaE_relative_to_existing_Boat"
    ] = (
        candidate_df[
            "energy_kcal_mol"
        ]
        -
        boat_energy
    )


    # --------------------------------------------------------
    # Save
    # --------------------------------------------------------

    candidate_df.to_csv(
        mother_output /
        "TARGETED_PUCKERING_PES.csv",
        index=False
    )


    # --------------------------------------------------------
    # Lowest candidate in each region
    # --------------------------------------------------------

    minima = []

    for region, sub in candidate_df.groupby(
        "pucker_region"
    ):

        sub = sub.dropna(
            subset=[
                "energy_kcal_mol"
            ]
        )

        if len(sub) == 0:
            continue

        row = sub.loc[
            sub[
                "energy_kcal_mol"
            ].idxmin()
        ]

        minima.append(row)


    minima_df = pd.DataFrame(
        minima
    )

    if len(minima_df):

        minima_df = minima_df.sort_values(
            "energy_kcal_mol"
        )

        minima_df[
            "DeltaE_from_lowest_candidate"
        ] = (
            minima_df[
                "energy_kcal_mol"
            ]
            -
            minima_df[
                "energy_kcal_mol"
            ].min()
        )

        minima_df[
            "DeltaE_relative_to_Boat"
        ] = (
            minima_df[
                "energy_kcal_mol"
            ]
            -
            boat_energy
        )


    minima_df.to_csv(
        mother_output /
        "LOWEST_ENERGY_BY_PUCKERING_REGION.csv",
        index=False
    )


    # ========================================================
    # PRINT MAIN RESULT
    # ========================================================

    print("\n")
    print("=" * 75)
    print(
        "LOWEST ENERGY BY PUCKERING REGION"
    )
    print("=" * 75)

    display(
        minima_df[
            [
                "pucker_region",
                "energy_kcal_mol",
                "DeltaE_from_lowest_candidate",
                "DeltaE_relative_to_Boat",
                "Q_A",
                "theta_deg",
                "phi_deg",
                "xyz_file"
            ]
        ]
    )


    # ========================================================
    # PLOT 1 — ENERGY VS THETA
    # ========================================================

    plotdf = candidate_df.dropna(
        subset=[
            "energy_kcal_mol",
            "theta_deg"
        ]
    )

    if len(plotdf):

        plt.figure(
            figsize=(8,6)
        )

        for region, sub in plotdf.groupby(
            "pucker_region"
        ):

            plt.scatter(
                sub["theta_deg"],
                sub["DeltaE_kcal_mol"],
                label=region,
                alpha=0.75
            )

        plt.xlabel(
            "Cremer–Pople θ (degrees)"
        )

        plt.ylabel(
            "ΔE (kcal mol$^{-1}$)"
        )

        plt.title(
            f"{mother_name}: "
            "Targeted cyclohexene TS puckering"
        )

        plt.legend(
            fontsize=8
        )

        plt.tight_layout()

        plt.savefig(
            mother_output /
            "PES_theta_vs_energy.png",
            dpi=300
        )

        plt.show()


    # ========================================================
    # PLOT 2 — ENERGY VS Q
    # ========================================================

    if len(plotdf):

        plt.figure(
            figsize=(8,6)
        )

        for region, sub in plotdf.groupby(
            "pucker_region"
        ):

            plt.scatter(
                sub["Q_A"],
                sub["DeltaE_kcal_mol"],
                label=region,
                alpha=0.75
            )

        plt.xlabel(
            "Cremer–Pople Q (Å)"
        )

        plt.ylabel(
            "ΔE (kcal mol$^{-1}$)"
        )

        plt.title(
            f"{mother_name}: "
            "Puckering amplitude vs energy"
        )

        plt.legend(
            fontsize=8
        )

        plt.tight_layout()

        plt.savefig(
            mother_output /
            "PES_Q_vs_energy.png",
            dpi=300
        )

        plt.show()


    print("\n")
    print("=" * 75)
    print(
        "FINISHED:",
        mother_name
    )
    print("=" * 75)

    print(
        "\nOutput:",
        mother_output
    )

    return {
        "existing":
            existing_df,

        "candidates":
            candidate_df,

        "minima":
            minima_df
    }


# ============================================================
# 22. SELECT MOTHERS
# ============================================================

if MOTHER_TS is None:

    selected = MOTHER_FILES

else:

    selected = [
        p for p in MOTHER_FILES
        if p.stem == MOTHER_TS
    ]


print(
    "\nSelected:",
    len(selected)
)

for p in selected:

    print(
        " ",
        p.name
    )


# ============================================================
# 23. RUN
# ============================================================

ALL_RESULTS = {}

for mother in selected:

    try:

        result = process_mother(
            mother
        )

        ALL_RESULTS[
            mother.stem
        ] = result

    except Exception as e:

        print(
            "\nFAILED:",
            mother.name
        )

        print(
            repr(e)
        )


print("\n")
print("=" * 75)
print("ALL DONE")
print("=" * 75)

print(
    "Results saved in:",
    OUTPUT_ROOT
)
```

```python
# ============================================================
# POST-PROCESSING:
# Cremer–Pople based puckering classification
#
# Input:
#   TARGETED_PUCKERING_PES.csv
#
# Existing coordinates:
#   Q, theta, phi
#
# Output:
#   1. TARGETED_PUCKERING_PES_CLASSIFIED.csv
#   2. LOWEST_ENERGY_BY_PUCKERING_CLASS.csv
#   3. Classified_XYZ/
#
# No CREST or xTB calculations are performed here.
# ============================================================

import os
import re
import shutil
import numpy as np
import pandas as pd
from pathlib import Path

# ------------------------------------------------------------
# 1. Google Drive
# ------------------------------------------------------------

from google.colab import drive
drive.mount("/content/drive")

BASE = Path("/content/drive/MyDrive/Cycloetherification_xyz_files")
CREST_ROOT = BASE / "CREST_results"
PES_ROOT = CREST_ROOT / "FAST_Cyclohexene_PES"

print("PES root:")
print(PES_ROOT)

# ------------------------------------------------------------
# 2. Find all mother-TS result folders
# ------------------------------------------------------------

mother_folders = [
    p for p in PES_ROOT.iterdir()
    if p.is_dir()
]

print(f"\nMother TS folders found: {len(mother_folders)}")

for p in mother_folders:
    print("  ", p.name)


# ============================================================
# 3. Cremer–Pople classification
# ============================================================

def classify_puckering(Q, theta, phi):
    """
    Approximate qualitative classification of six-membered
    ring puckering using Cremer–Pople Q, theta and phi.

    Important:
    These are 'like' classifications for the constrained
    cyclohexene-type TS ring, not claims of ideal canonical
    cyclohexane conformations.
    """

    # Missing values
    if pd.isna(Q) or pd.isna(theta) or pd.isna(phi):
        return "Not calculated"

    Q = float(Q)
    theta = float(theta)
    phi = float(phi) % 360.0

    # Essentially planar
    if Q < 0.05:
        return "Near-planar"

    # --------------------------------------------------------
    # Six-membered ring:
    #
    # theta ~ 0 / 180:
    #     chair regions
    #
    # theta ~ 90:
    #     boat/twist-boat region
    #
    # intermediate theta:
    #     half-chair / half-boat / intermediate regions
    #
    # Because this is a cyclohexene-type TS, we deliberately
    # use "like" terminology.
    # --------------------------------------------------------

    # Chair-like regions
    if theta <= 20 or theta >= 160:
        return "Chair-like"

    # Boat / twist-boat region
    if 70 <= theta <= 110:

        # Normalize phi to 0–180 for easier comparison
        phi180 = phi % 180.0

        # Approximate boat vs twist-boat division.
        #
        # The exact phase relationship depends on the atom
        # numbering / Cremer–Pople convention. Therefore this
        # is intentionally labeled as approximate.
        if phi180 <= 25 or phi180 >= 155:
            return "Boat-like"

        elif 55 <= phi180 <= 125:
            return "Twist-boat-like"

        else:
            return "Boat/Twist-boat intermediate"

    # --------------------------------------------------------
    # Half-chair / half-boat regions
    # --------------------------------------------------------

    # Near lower intermediate theta
    if 20 < theta < 70:

        if theta < 45:
            return "Half-chair-like"

        else:
            return "Half-boat-like / intermediate"

    # Near upper intermediate theta
    if 110 < theta < 160:

        if theta > 135:
            return "Half-chair-like"

        else:
            return "Half-boat-like / intermediate"

    return "Intermediate"


# ============================================================
# 4. Locate candidate XYZ files
# ============================================================

def find_candidate_xyz(mother_folder, filename):
    """
    Locate an XYZ file whether the CSV contains:
      1. just the filename, or
      2. an absolute path.
    """

    if filename is None:
        return None

    filename = str(filename).strip()

    if not filename or filename.lower() == "nan":
        return None

    # --------------------------------------------------------
    # Case 1: CSV contains an absolute path
    # --------------------------------------------------------
    possible_path = Path(filename)

    if possible_path.is_absolute() and possible_path.exists():
        return possible_path

    # --------------------------------------------------------
    # Case 2: CSV contains just the filename
    # --------------------------------------------------------
    direct_path = mother_folder / filename

    if direct_path.exists():
        return direct_path

    # --------------------------------------------------------
    # Case 3: Search recursively using ONLY the basename
    # This avoids the pathlib absolute-pattern error.
    # --------------------------------------------------------
    basename = possible_path.name

    matches = list(
        mother_folder.rglob(basename)
    )

    if matches:
        return matches[0]

    return None

# ============================================================
# 5. Process every mother TS independently
# ============================================================

all_classified = []
all_summary = []

for mother_folder in sorted(mother_folders):

    mother = mother_folder.name

    csv_file = mother_folder / "TARGETED_PUCKERING_PES.csv"

    if not csv_file.exists():
        print(f"\nSKIP: {mother}")
        print("  TARGETED_PUCKERING_PES.csv not found.")
        continue

    print("\n" + "=" * 70)
    print(f"PROCESSING: {mother}")
    print("=" * 70)

    df = pd.read_csv(csv_file)

    print(f"Rows found: {len(df)}")

    # --------------------------------------------------------
    # Find actual column names
    # --------------------------------------------------------

    def find_column(possible_names):

        lower_map = {
            str(c).strip().lower(): c
            for c in df.columns
        }

        for name in possible_names:
            if name.lower() in lower_map:
                return lower_map[name.lower()]

        # Partial matching
        for c in df.columns:
            cl = str(c).lower()
            for name in possible_names:
                if name.lower() in cl:
                    return c

        return None

    q_col = find_column([
        "Q",
        "Q_Angstrom",
        "Q (Angstrom)",
        "puckering_Q"
    ])

    theta_col = find_column([
        "theta",
        "theta_deg",
        "theta (deg)",
        "Cremer_Pople_theta"
    ])

    phi_col = find_column([
        "phi",
        "phi_deg",
        "phi (deg)",
        "Cremer_Pople_phi"
    ])

    energy_col = find_column([
        "DeltaE_kcal_mol",
        "DeltaE",
        "deltaE",
        "Energy",
        "energy_kcal_mol"
    ])

    file_col = find_column([
        "filename",
        "file",
        "xyz",
        "structure",
        "structure_file"
    ])

    print("\nDetected columns:")
    print("  Q     :", q_col)
    print("  theta :", theta_col)
    print("  phi   :", phi_col)
    print("  energy:", energy_col)
    print("  file  :", file_col)

    if q_col is None or theta_col is None or phi_col is None:
        print("\nWARNING:")
        print("Q/theta/phi columns were not found.")
        print("Classification cannot be performed for this mother.")
        continue

    # --------------------------------------------------------
    # Convert numeric columns
    # --------------------------------------------------------

    df["Q_numeric"] = pd.to_numeric(df[q_col], errors="coerce")
    df["theta_numeric"] = pd.to_numeric(df[theta_col], errors="coerce")
    df["phi_numeric"] = pd.to_numeric(df[phi_col], errors="coerce")

    if energy_col is not None:
        df["Energy_numeric"] = pd.to_numeric(
            df[energy_col], errors="coerce"
        )
    else:
        df["Energy_numeric"] = np.nan

    # --------------------------------------------------------
    # Assign puckering class
    # --------------------------------------------------------

    df["Puckering_class"] = [
        classify_puckering(Q, theta, phi)
        for Q, theta, phi in zip(
            df["Q_numeric"],
            df["theta_numeric"],
            df["phi_numeric"]
        )
    ]

    # --------------------------------------------------------
    # Add mother name
    # --------------------------------------------------------

    df["Mother_TS"] = mother

    # --------------------------------------------------------
    # Save classified CSV
    # --------------------------------------------------------

    classified_csv = (
        mother_folder /
        "TARGETED_PUCKERING_PES_CLASSIFIED.csv"
    )

    df.to_csv(classified_csv, index=False)

    print("\nSaved:")
    print(" ", classified_csv)

    # --------------------------------------------------------
    # Create classified XYZ directory
    # --------------------------------------------------------

    xyz_out = mother_folder / "Classified_XYZ"
    xyz_out.mkdir(exist_ok=True)

    # --------------------------------------------------------
    # Copy XYZ files with puckering class
    # --------------------------------------------------------

    if file_col is not None:

        for idx, row in df.iterrows():

            filename = str(row[file_col])

            if filename.lower() == "nan":
                continue

            source = find_candidate_xyz(
                mother_folder,
                filename
            )

            if source is None:

                print(
                    f"  XYZ not found: {filename}"
                )

                continue

            puckering = str(
                row["Puckering_class"]
            )

            # Safe filename
            safe_class = re.sub(
                r"[^A-Za-z0-9_-]+",
                "_",
                puckering
            )

            candidate_number = idx + 1

            new_name = (
                f"candidate_{candidate_number:04d}"
                f"_{safe_class}.xyz"
            )

            destination = xyz_out / new_name

            shutil.copy2(
                source,
                destination
            )

    # --------------------------------------------------------
    # Lowest-energy structure of each class
    # --------------------------------------------------------

    valid_energy = df[
        np.isfinite(df["Energy_numeric"])
    ].copy()

    if len(valid_energy) > 0:

        summary_rows = []

        for puckering_class, group in valid_energy.groupby(
            "Puckering_class"
        ):

            group = group.sort_values(
                "Energy_numeric"
            )

            lowest = group.iloc[0]

            summary_rows.append({

                "Mother_TS": mother,

                "Puckering_class":
                    puckering_class,

                "Lowest_Energy":
                    lowest["Energy_numeric"],

                "Q_Angstrom":
                    lowest["Q_numeric"],

                "theta_deg":
                    lowest["theta_numeric"],

                "phi_deg":
                    lowest["phi_numeric"],

                "Filename":
                    lowest[file_col]
                    if file_col is not None
                    else "",

                "Number_of_structures":
                    len(group)
            })

        summary_df = pd.DataFrame(summary_rows)

        summary_file = (
            mother_folder /
            "LOWEST_ENERGY_BY_PUCKERING_CLASS.csv"
        )

        summary_df.to_csv(
            summary_file,
            index=False
        )

        all_summary.append(summary_df)

        print("\nLowest energy by puckering class:")

        display(
            summary_df.sort_values(
                "Lowest_Energy"
            )
        )

    all_classified.append(df)


# ============================================================
# 6. Combined summary for ALL mother TS
# ============================================================

if all_classified:

    combined_df = pd.concat(
        all_classified,
        ignore_index=True
    )

    combined_file = (
        PES_ROOT /
        "ALL_TARGETED_PUCKERING_PES_CLASSIFIED.csv"
    )

    combined_df.to_csv(
        combined_file,
        index=False
    )

    print("\n" + "=" * 70)
    print("COMBINED CLASSIFIED DATASET")
    print("=" * 70)

    print(
        f"Total structures: {len(combined_df)}"
    )

    print(
        f"Saved:\n{combined_file}"
    )


# ============================================================
# 7. Combined lowest-energy summary
# ============================================================

if all_summary:

    combined_summary = pd.concat(
        all_summary,
        ignore_index=True
    )

    combined_summary_file = (
        PES_ROOT /
        "ALL_LOWEST_ENERGY_BY_PUCKERING_CLASS.csv"
    )

    combined_summary.to_csv(
        combined_summary_file,
        index=False
    )

    print("\n" + "=" * 70)
    print("FINAL SUMMARY")
    print("=" * 70)

    display(
        combined_summary.sort_values(
            ["Mother_TS", "Lowest_Energy"]
        )
    )

    print(
        "\nSaved:\n"
        f"{combined_summary_file}"
    )


print("\n" + "=" * 70)
print("DONE")
print("=" * 70)
print("\nNo CREST calculation was performed.")
print("No xTB calculation was performed.")
print("Only post-processing/classification was performed.")
```
