```python
# ============================================================
# FAST TARGETED CYCLOHEXENE-TYPE TS PUCKERING SCAN
#
# Ring:
#       O17 - C18 - C19 - C20 = C21 - C22
#
# Forming bond:
#       O17 - C22
#
# Goal:
#   Start from the lowest-energy existing BOAT conformer
#   and rapidly probe:
#
#       Boat
#       Twist-boat
#       Half-chair
#       Half-boat
#       Chair-like
#       intermediate/distorted forms
#
# IMPORTANT:
#   This is NOT a full CREST search.
#   It is a targeted geometric PES scan.
# ============================================================


# ============================================================
# 1. INSTALL xTB
# ============================================================

!apt-get -qq update
!apt-get -qq install -y xtb

!pip -q install numpy pandas matplotlib scipy


# ============================================================
# 2. IMPORTS
# ============================================================

import os
import re
import glob
import shutil
import subprocess
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt

from pathlib import Path


# ============================================================
# 3. GOOGLE DRIVE
# ============================================================

from google.colab import drive

drive.mount("/content/drive")


BASE = Path(
    "/content/drive/MyDrive/Cycloetherification_xyz_files"
)

CREST_ROOT = BASE / "CREST_results"

OUTPUT_ROOT = (
    CREST_ROOT /
    "FAST_Cyclohexene_PES"
)

OUTPUT_ROOT.mkdir(
    parents=True,
    exist_ok=True
)


# ============================================================
# 4. USER SETTINGS
# ============================================================

# ------------------------------------------------------------
# Process one mother TS or all mothers
# ------------------------------------------------------------

MOTHER_TS = None

# Example:
#
# MOTHER_TS = "RR_CH2-Ph"
#
# or:
#
# MOTHER_TS = "RR_tBu"
#
# None = all mothers


# ------------------------------------------------------------
# SIX-MEMBERED RING
#
# O17-C18-C19-C20-C21-C22
# ------------------------------------------------------------

RING_ATOMS = [
    17,
    18,
    19,
    20,
    21,
    22
]


# ------------------------------------------------------------
# DOUBLE BOND
#
# C20=C21
# ------------------------------------------------------------

DOUBLE_BOND = (
    20,
    21
)


# ------------------------------------------------------------
# FORMING BOND
#
# O17-C22
# ------------------------------------------------------------

FORMING_BOND = (
    17,
    22
)


# ------------------------------------------------------------
# Core atoms
#
# Useful if alignment is required later.
# ------------------------------------------------------------

CORE_ATOMS = list(range(1, 44))


# ------------------------------------------------------------
# Puckering amplitude scan
#
# Smaller = less distortion
# Larger  = stronger ring puckering
#
# These are geometric perturbations, NOT barriers.
# ------------------------------------------------------------

Q_SCALE_FACTORS = [
    0.60,
    0.80,
    1.00,
    1.20,
    1.40
]


# ------------------------------------------------------------
# Number of phase points
#
# 12 phase points cover the puckering cycle.
# ------------------------------------------------------------

N_PHASES = 12


# ------------------------------------------------------------
# xTB settings
# ------------------------------------------------------------

XTB_TIMEOUT = 300


print("BASE =", BASE)
print("CREST_ROOT =", CREST_ROOT)
print("OUTPUT_ROOT =", OUTPUT_ROOT)

print("\nRing =", RING_ATOMS)
print("Double bond =", DOUBLE_BOND)
print("Forming bond =", FORMING_BOND)


# ============================================================
# 5. CHECK xTB
# ============================================================

XTB = shutil.which("xtb")

if XTB is None:
    raise RuntimeError(
        "xTB was not found."
    )

print("\nxTB executable:", XTB)

subprocess.run(
    [XTB, "--version"],
    check=False
)


# ============================================================
# 6. XYZ READER
# ============================================================

def read_xyz(path):

    path = Path(path)

    with open(path) as f:

        lines = [
            line.rstrip()
            for line in f
        ]

    n = int(lines[0])

    comment = lines[1]

    atoms = []
    coords = []

    for line in lines[2:2+n]:

        p = line.split()

        atoms.append(p[0])

        coords.append([
            float(p[1]),
            float(p[2]),
            float(p[3])
        ])

    return (
        atoms,
        np.asarray(coords, dtype=float),
        comment
    )


# ============================================================
# 7. XYZ WRITER
# ============================================================

def write_xyz(
    path,
    atoms,
    coords,
    comment=""
):

    path = Path(path)

    with open(path, "w") as f:

        f.write(
            f"{len(atoms)}\n"
        )

        f.write(
            comment + "\n"
        )

        for atom, xyz in zip(
            atoms,
            coords
        ):

            f.write(
                f"{atom:2s} "
                f"{xyz[0]: .8f} "
                f"{xyz[1]: .8f} "
                f"{xyz[2]: .8f}\n"
            )


# ============================================================
# 8. FIND MOTHER TS FILES
# ============================================================

def is_generated_file(path):

    name = Path(path).name

    bad = [
        "_retained_",
        "_raw_crest",
        "_crest",
        "_aligned",
        "_SUPERIMPOSED"
    ]

    return any(
        x in name
        for x in bad
    )


def find_mothers():

    files = []

    for p in sorted(
        BASE.glob("*.xyz")
    ):

        if not is_generated_file(p):

            files.append(p)

    return files


MOTHER_FILES = find_mothers()

print(
    "\nMother TS files:",
    len(MOTHER_FILES)
)

for p in MOTHER_FILES:

    print(
        " ",
        p.name
    )


# ============================================================
# 9. FIND ONLY INDIVIDUAL RETAINED STRUCTURES
# ============================================================

def find_retained_files(
    mother_name
):

    # IMPORTANT:
    #
    # This deliberately matches:
    #
    # retained_001.xyz
    # retained_002.xyz
    #
    # but NOT:
    #
    # retained_ensemble.xyz

    pattern = (
        CREST_ROOT /
        "**" /
        f"{mother_name}_retained_[0-9][0-9][0-9].xyz"
    )

    files = sorted(
        glob.glob(
            str(pattern),
            recursive=True
        )
    )

    return [
        Path(x)
        for x in files
    ]


# ============================================================
# 10. GEOMETRY FUNCTIONS
# ============================================================

def distance(
    coords,
    a,
    b
):

    return float(
        np.linalg.norm(
            coords[a-1]
            -
            coords[b-1]
        )
    )


def ring_centroid(
    coords
):

    idx = np.asarray(
        RING_ATOMS
    ) - 1

    return coords[idx].mean(
        axis=0
    )


# ============================================================
# 11. RING PLANE / NORMAL
# ============================================================

def ring_normal(
    coords
):

    idx = np.asarray(
        RING_ATOMS
    ) - 1

    ring = coords[idx]

    centered = (
        ring -
        ring.mean(axis=0)
    )

    _, _, vh = np.linalg.svd(
        centered
    )

    normal = vh[-1]

    normal /= np.linalg.norm(
        normal
    )

    return normal


# ============================================================
# 12. CREMER–POPLE
# ============================================================

def cremer_pople(
    coords
):

    idx = np.asarray(
        RING_ATOMS
    ) - 1

    ring = coords[idx].copy()

    ring -= ring.mean(
        axis=0
    )

    _, _, vh = np.linalg.svd(
        ring,
        full_matrices=False
    )

    normal = vh[-1]

    if normal[2] < 0:

        normal = -normal

    z = ring @ normal

    j = np.arange(6)

    q2c = (
        np.sqrt(1/3)
        *
        np.sum(
            z *
            np.cos(
                2*np.pi*j/3
            )
        )
    )

    q2s = (
        np.sqrt(1/3)
        *
        np.sum(
            z *
            np.sin(
                2*np.pi*j/3
            )
        )
    )

    q3 = (
        np.sqrt(1/6)
        *
        np.sum(
            z *
            ((-1)**j)
        )
    )

    q2 = np.sqrt(
        q2c**2 +
        q2s**2
    )

    Q = np.sqrt(
        q2**2 +
        q3**2
    )

    if Q < 1e-10:

        return (
            Q,
            np.nan,
            np.nan
        )

    theta = np.degrees(
        np.arccos(
            np.clip(
                q3/Q,
                -1,
                1
            )
        )
    )

    phi = np.degrees(
        np.arctan2(
            q2s,
            q2c
        )
    )

    phi %= 360

    return (
        float(Q),
        float(theta),
        float(phi)
    )


# ============================================================
# 13. SIMPLE PUCKERING REGION
# ============================================================

def pucker_region(
    theta,
    phi
):

    if not np.isfinite(theta):

        return "Undefined"

    # Chair-like
    if (
        theta < 20
        or
        theta > 160
    ):

        return "Chair-like"


    # Boat / twist-boat region
    #
    # For a six-membered ring these lie around
    # the equatorial part of CP space.

    if 70 <= theta <= 110:

        # Canonical phase sectors
        # are approximate because atom ordering
        # defines the phase origin.

        phase = (
            round(
                phi / 60
            )
            * 60
        ) % 360

        phase_distance = min(
            abs(phi - phase),
            360 - abs(phi - phase)
        )

        if phase_distance <= 15:

            return "Boat-region"

        return "Twist-boat-region"


    # Half-chair / half-boat region
    if (
        40 <= theta <= 70
        or
        110 <= theta <= 140
    ):

        return "Half-chair / half-boat region"


    return "Intermediate"


# ============================================================
# 14. XTB SINGLE-POINT
# ============================================================

def xtb_energy(
    xyz_file,
    workdir
):

    workdir = Path(workdir)

    workdir.mkdir(
        parents=True,
        exist_ok=True
    )

    local_xyz = (
        workdir /
        "structure.xyz"
    )

    shutil.copy2(
        xyz_file,
        local_xyz
    )

    cmd = [
        XTB,
        "structure.xyz",
        "--gfn2",
        "--sp"
    ]

    result = subprocess.run(
        cmd,
        cwd=workdir,
        stdout=subprocess.PIPE,
        stderr=subprocess.STDOUT,
        text=True,
        timeout=XTB_TIMEOUT
    )

    output = result.stdout

    patterns = [
        r"TOTAL ENERGY\s+"
        r"([-+]?\d+\.\d+(?:[Ee][-+]?\d+)?)",

        r"::\s*total energy\s+"
        r"([-+]?\d+\.\d+(?:[Ee][-+]?\d+)?)"
    ]

    energy_Eh = None

    for pattern in patterns:

        m = re.findall(
            pattern,
            output,
            flags=re.I
        )

        if m:

            energy_Eh = float(
                m[-1]
            )

            break

    if energy_Eh is None:

        raise RuntimeError(
            "Could not extract xTB energy."
        )

    energy_kcal = (
        energy_Eh *
        627.509474
    )

    return (
        energy_Eh,
        energy_kcal
    )


# ============================================================
# 15. IMPORTANT:
#     GENERATE PUCKERING DISTORTIONS
# ============================================================

def generate_pucker(
    coords,
    amplitude,
    phase_deg
):

    """
    Generate a controlled ring puckering
    from the selected Boat structure.

    The entire molecule is NOT rotated independently.

    Only the six ring atoms are displaced along
    the instantaneous ring normal.

    This is a targeted geometric scan.

    It does NOT claim that the resulting structure
    is an optimized minimum.
    """

    new = coords.copy()

    idx = np.asarray(
        RING_ATOMS
    ) - 1

    normal = ring_normal(
        coords
    )

    phase = np.radians(
        phase_deg
    )

    j = np.arange(6)

    # Two six-membered-ring puckering components
    displacement = (
        np.cos(
            2*np.pi*j/3
            + phase
        )
        *
        amplitude
    )

    # Add alternating component
    displacement += (
        0.50 *
        amplitude *
        ((-1)**j)
    )

    for atom_index, dz in zip(
        idx,
        displacement
    ):

        new[atom_index] += (
            dz *
            normal
        )

    return new


# ============================================================
# 16. CREATE TARGETED CANDIDATES
# ============================================================

def generate_candidates(
    seed_coords
):

    candidates = []

    # --------------------------------------------
    # Determine natural puckering amplitude
    # from seed CP Q
    # --------------------------------------------

    Q0, theta0, phi0 = \
        cremer_pople(
            seed_coords
        )

    # A modest amplitude based on seed geometry.
    #
    # We deliberately avoid huge distortions.

    base_amp = max(
        0.10,
        min(
            0.40,
            0.50 * Q0
        )
    )

    phases = np.linspace(
        0,
        360,
        N_PHASES,
        endpoint=False
    )

    for scale in Q_SCALE_FACTORS:

        amp = (
            base_amp *
            scale
        )

        for phase in phases:

            coords = generate_pucker(
                seed_coords,
                amp,
                phase
            )

            Q, theta, phi = \
                cremer_pople(
                    coords
                )

            candidates.append({

                "amplitude":
                    amp,

                "phase_deg":
                    phase,

                "coords":
                    coords,

                "Q_A":
                    Q,

                "theta_deg":
                    theta,

                "phi_deg":
                    phi,

                "pucker_region":
                    pucker_region(
                        theta,
                        phi
                    )

            })

    return candidates


# ============================================================
# 17. REMOVE GEOMETRIC DUPLICATES
# ============================================================

def geometry_rmsd(
    A,
    B
):

    idx = (
        np.asarray(
            RING_ATOMS
        ) - 1
    )

    A = A[idx]
    B = B[idx]

    A -= A.mean(
        axis=0
    )

    B -= B.mean(
        axis=0
    )

    H = A.T @ B

    U, S, Vt = np.linalg.svd(
        H
    )

    d = np.linalg.det(
        Vt.T @ U.T
    )

    if d < 0:

        Vt[-1] *= -1

    R = Vt.T @ U.T

    Afit = A @ R

    return np.sqrt(
        np.mean(
            np.sum(
                (Afit-B)**2,
                axis=1
            )
        )
    )


def deduplicate(
    candidates,
    cutoff=0.15
):

    kept = []

    for candidate in candidates:

        duplicate = False

        for old in kept:

            if geometry_rmsd(
                candidate["coords"],
                old["coords"]
            ) < cutoff:

                duplicate = True
                break

        if not duplicate:

            kept.append(
                candidate
            )

    return kept


# ============================================================
# 18. ANALYZE EXISTING RETAINED STRUCTURES
# ============================================================

def analyze_existing(
    mother_name,
    mother_file
):

    files = find_retained_files(
        mother_name
    )

    print(
        f"\n{mother_name}: "
        f"{len(files)} individual retained structures"
    )

    records = []

    for xyz in files:

        atoms, coords, comment = \
            read_xyz(xyz)

        Q, theta, phi = \
            cremer_pople(
                coords
            )

        records.append({

            "filename":
                xyz.name,

            "path":
                str(xyz),

            "Q_A":
                Q,

            "theta_deg":
                theta,

            "phi_deg":
                phi,

            "pucker_region":
                pucker_region(
                    theta,
                    phi
                ),

            "coords":
                coords,

            "atoms":
                atoms

        })

    return records


# ============================================================
# 19. CALCULATE EXISTING xTB ENERGIES
# ============================================================

def energy_existing(
    records,
    mother_output
):

    energy_dir = (
        mother_output /
        "Existing_xTB_energy"
    )

    energy_dir.mkdir(
        parents=True,
        exist_ok=True
    )

    output = []

    for i, rec in enumerate(
        records,
        start=1
    ):

        xyz = (
            energy_dir /
            f"existing_{i:03d}.xyz"
        )

        write_xyz(
            xyz,
            rec["atoms"],
            rec["coords"],
            "Fresh_GFN2_xTB"
        )

        E_Eh, E_kcal = \
            xtb_energy(
                xyz,
                energy_dir /
                f"run_{i:03d}"
            )

        row = {
            key: value
            for key, value in rec.items()
            if key not in [
                "coords",
                "atoms"
            ]
        }

        row[
            "energy_Eh"
        ] = E_Eh

        row[
            "energy_kcal_mol"
        ] = E_kcal

        output.append(row)

    df = pd.DataFrame(
        output
    )

    Emin = df[
        "energy_kcal_mol"
    ].min()

    df[
        "DeltaE_kcal_mol"
    ] = (
        df["energy_kcal_mol"]
        -
        Emin
    )

    return df


# ============================================================
# 20. FIND LOWEST-ENERGY BOAT
# ============================================================

def choose_boat(
    df
):

    boat = df[
        df[
            "pucker_region"
        ].str.contains(
            "Boat",
            na=False
        )
    ].copy()

    if len(boat) == 0:

        return None

    boat = boat.sort_values(
        "energy_kcal_mol"
    )

    return boat.iloc[0]


# ============================================================
# 21. PROCESS ONE MOTHER TS
# ============================================================

def process_mother(
    mother_file
):

    mother_name = (
        mother_file.stem
    )

    print("\n")
    print("=" * 75)
    print(
        "PROCESSING:",
        mother_name
    )
    print("=" * 75)

    mother_output = (
        OUTPUT_ROOT /
        mother_name
    )

    mother_output.mkdir(
        parents=True,
        exist_ok=True
    )


    # --------------------------------------------------------
    # Existing structures
    # --------------------------------------------------------

    records = analyze_existing(
        mother_name,
        mother_file
    )

    if len(records) == 0:

        print(
            "No retained structures."
        )

        return


    # --------------------------------------------------------
    # Fresh xTB energies
    # --------------------------------------------------------

    existing_df = energy_existing(
        records,
        mother_output
    )


    existing_df.to_csv(
        mother_output /
        "EXISTING_RETAINED_FRESH_XTB.csv",
        index=False
    )


    print("\nExisting structures:")
    display(
        existing_df[
            [
                "filename",
                "energy_kcal_mol",
                "DeltaE_kcal_mol",
                "Q_A",
                "theta_deg",
                "phi_deg",
                "pucker_region"
            ]
        ].sort_values(
            "DeltaE_kcal_mol"
        )
    )


    # --------------------------------------------------------
    # Lowest-energy Boat
    # --------------------------------------------------------

    boat = choose_boat(
        existing_df
    )

    if boat is None:

        print(
            "\nNo Boat-region structure "
            "was identified."
        )

        return


    print("\n")
    print("=" * 75)
    print("LOWEST-ENERGY BOAT")
    print("=" * 75)

    print(
        "File:",
        boat["filename"]
    )

    print(
        "Energy:",
        boat["energy_kcal_mol"],
        "kcal/mol"
    )

    print(
        "ΔE:",
        boat["DeltaE_kcal_mol"],
        "kcal/mol"
    )

    print(
        "Q:",
        boat["Q_A"]
    )

    print(
        "theta:",
        boat["theta_deg"]
    )

    print(
        "phi:",
        boat["phi_deg"]
    )


    # --------------------------------------------------------
    # Save Boat seed
    # --------------------------------------------------------

    seed_dir = (
        mother_output /
        "LOWEST_BOAT_SEED"
    )

    seed_dir.mkdir(
        parents=True,
        exist_ok=True
    )

    seed_path = (
        seed_dir /
        "lowest_energy_Boat.xyz"
    )

    # Locate original
    original_seed = Path(
        boat["path"]
    )

    atoms, seed_coords, comment = \
        read_xyz(
            original_seed
        )

    write_xyz(
        seed_path,
        atoms,
        seed_coords,
        "LOWEST_ENERGY_BOAT_SEED"
    )


    # --------------------------------------------------------
    # Generate targeted puckering candidates
    # --------------------------------------------------------

    candidates = generate_candidates(
        seed_coords
    )

    print(
        "\nGenerated candidates:",
        len(candidates)
    )


    # --------------------------------------------------------
    # Deduplicate
    # --------------------------------------------------------

    candidates = deduplicate(
        candidates,
        cutoff=0.15
    )

    print(
        "After geometric deduplication:",
        len(candidates)
    )


    # --------------------------------------------------------
    # Candidate directory
    # --------------------------------------------------------

    candidate_dir = (
        mother_output /
        "TARGETED_PUCKERING_STRUCTURES"
    )

    candidate_dir.mkdir(
        parents=True,
        exist_ok=True
    )


    # --------------------------------------------------------
    # xTB energy for every candidate
    # --------------------------------------------------------

    energy_dir = (
        mother_output /
        "TARGETED_xTB"
    )

    energy_dir.mkdir(
        parents=True,
        exist_ok=True
    )


    results = []


    for i, candidate in enumerate(
        candidates,
        start=1
    ):

        xyz = (
            candidate_dir /
            f"candidate_{i:04d}.xyz"
        )

        comment = (
            f"targeted_pucker "
            f"amplitude={candidate['amplitude']:.5f} "
            f"phase={candidate['phase_deg']:.2f}"
        )

        write_xyz(
            xyz,
            atoms,
            candidate["coords"],
            comment
        )


        # xTB energy

        try:

            E_Eh, E_kcal = \
                xtb_energy(
                    xyz,
                    energy_dir /
                    f"candidate_{i:04d}"
                )

        except Exception as e:

            print(
                "xTB FAILED:",
                i,
                e
            )

            E_Eh = np.nan
            E_kcal = np.nan


        results.append({

            "candidate":
                i,

            "amplitude_A":
                candidate["amplitude"],

            "phase_deg":
                candidate["phase_deg"],

            "energy_Eh":
                E_Eh,

            "energy_kcal_mol":
                E_kcal,

            "Q_A":
                candidate["Q_A"],

            "theta_deg":
                candidate["theta_deg"],

            "phi_deg":
                candidate["phi_deg"],

            "pucker_region":
                candidate["pucker_region"],

            "xyz_file":
                str(xyz)

        })


    candidate_df = pd.DataFrame(
        results
    )


    # --------------------------------------------------------
    # Relative energy
    # --------------------------------------------------------

    valid = candidate_df[
        "energy_kcal_mol"
    ].notna()

    Emin = candidate_df.loc[
        valid,
        "energy_kcal_mol"
    ].min()

    candidate_df[
        "DeltaE_kcal_mol"
    ] = (
        candidate_df[
            "energy_kcal_mol"
        ]
        -
        Emin
    )


    # --------------------------------------------------------
    # Energy relative to LOWEST BOAT
    # --------------------------------------------------------

    boat_energy = boat[
        "energy_kcal_mol"
    ]

    candidate_df[
        "DeltaE_relative_to_existing_Boat"
    ] = (
        candidate_df[
            "energy_kcal_mol"
        ]
        -
        boat_energy
    )


    # --------------------------------------------------------
    # Save
    # --------------------------------------------------------

    candidate_df.to_csv(
        mother_output /
        "TARGETED_PUCKERING_PES.csv",
        index=False
    )


    # --------------------------------------------------------
    # Lowest candidate in each region
    # --------------------------------------------------------

    minima = []

    for region, sub in candidate_df.groupby(
        "pucker_region"
    ):

        sub = sub.dropna(
            subset=[
                "energy_kcal_mol"
            ]
        )

        if len(sub) == 0:
            continue

        row = sub.loc[
            sub[
                "energy_kcal_mol"
            ].idxmin()
        ]

        minima.append(row)


    minima_df = pd.DataFrame(
        minima
    )

    if len(minima_df):

        minima_df = minima_df.sort_values(
            "energy_kcal_mol"
        )

        minima_df[
            "DeltaE_from_lowest_candidate"
        ] = (
            minima_df[
                "energy_kcal_mol"
            ]
            -
            minima_df[
                "energy_kcal_mol"
            ].min()
        )

        minima_df[
            "DeltaE_relative_to_Boat"
        ] = (
            minima_df[
                "energy_kcal_mol"
            ]
            -
            boat_energy
        )


    minima_df.to_csv(
        mother_output /
        "LOWEST_ENERGY_BY_PUCKERING_REGION.csv",
        index=False
    )


    # ========================================================
    # PRINT MAIN RESULT
    # ========================================================

    print("\n")
    print("=" * 75)
    print(
        "LOWEST ENERGY BY PUCKERING REGION"
    )
    print("=" * 75)

    display(
        minima_df[
            [
                "pucker_region",
                "energy_kcal_mol",
                "DeltaE_from_lowest_candidate",
                "DeltaE_relative_to_Boat",
                "Q_A",
                "theta_deg",
                "phi_deg",
                "xyz_file"
            ]
        ]
    )


    # ========================================================
    # PLOT 1 — ENERGY VS THETA
    # ========================================================

    plotdf = candidate_df.dropna(
        subset=[
            "energy_kcal_mol",
            "theta_deg"
        ]
    )

    if len(plotdf):

        plt.figure(
            figsize=(8,6)
        )

        for region, sub in plotdf.groupby(
            "pucker_region"
        ):

            plt.scatter(
                sub["theta_deg"],
                sub["DeltaE_kcal_mol"],
                label=region,
                alpha=0.75
            )

        plt.xlabel(
            "Cremer–Pople θ (degrees)"
        )

        plt.ylabel(
            "ΔE (kcal mol$^{-1}$)"
        )

        plt.title(
            f"{mother_name}: "
            "Targeted cyclohexene TS puckering"
        )

        plt.legend(
            fontsize=8
        )

        plt.tight_layout()

        plt.savefig(
            mother_output /
            "PES_theta_vs_energy.png",
            dpi=300
        )

        plt.show()


    # ========================================================
    # PLOT 2 — ENERGY VS Q
    # ========================================================

    if len(plotdf):

        plt.figure(
            figsize=(8,6)
        )

        for region, sub in plotdf.groupby(
            "pucker_region"
        ):

            plt.scatter(
                sub["Q_A"],
                sub["DeltaE_kcal_mol"],
                label=region,
                alpha=0.75
            )

        plt.xlabel(
            "Cremer–Pople Q (Å)"
        )

        plt.ylabel(
            "ΔE (kcal mol$^{-1}$)"
        )

        plt.title(
            f"{mother_name}: "
            "Puckering amplitude vs energy"
        )

        plt.legend(
            fontsize=8
        )

        plt.tight_layout()

        plt.savefig(
            mother_output /
            "PES_Q_vs_energy.png",
            dpi=300
        )

        plt.show()


    print("\n")
    print("=" * 75)
    print(
        "FINISHED:",
        mother_name
    )
    print("=" * 75)

    print(
        "\nOutput:",
        mother_output
    )

    return {
        "existing":
            existing_df,

        "candidates":
            candidate_df,

        "minima":
            minima_df
    }


# ============================================================
# 22. SELECT MOTHERS
# ============================================================

if MOTHER_TS is None:

    selected = MOTHER_FILES

else:

    selected = [
        p for p in MOTHER_FILES
        if p.stem == MOTHER_TS
    ]


print(
    "\nSelected:",
    len(selected)
)

for p in selected:

    print(
        " ",
        p.name
    )


# ============================================================
# 23. RUN
# ============================================================

ALL_RESULTS = {}

for mother in selected:

    try:

        result = process_mother(
            mother
        )

        ALL_RESULTS[
            mother.stem
        ] = result

    except Exception as e:

        print(
            "\nFAILED:",
            mother.name
        )

        print(
            repr(e)
        )


print("\n")
print("=" * 75)
print("ALL DONE")
print("=" * 75)

print(
    "Results saved in:",
    OUTPUT_ROOT
)
```

```python
# ============================================================
# POST-PROCESSING:
# Cremer–Pople based puckering classification
#
# Input:
#   TARGETED_PUCKERING_PES.csv
#
# Existing coordinates:
#   Q, theta, phi
#
# Output:
#   1. TARGETED_PUCKERING_PES_CLASSIFIED.csv
#   2. LOWEST_ENERGY_BY_PUCKERING_CLASS.csv
#   3. Classified_XYZ/
#
# No CREST or xTB calculations are performed here.
# ============================================================

import os
import re
import shutil
import numpy as np
import pandas as pd
from pathlib import Path

# ------------------------------------------------------------
# 1. Google Drive
# ------------------------------------------------------------

from google.colab import drive
drive.mount("/content/drive")

BASE = Path("/content/drive/MyDrive/Cycloetherification_xyz_files")
CREST_ROOT = BASE / "CREST_results"
PES_ROOT = CREST_ROOT / "FAST_Cyclohexene_PES"

print("PES root:")
print(PES_ROOT)

# ------------------------------------------------------------
# 2. Find all mother-TS result folders
# ------------------------------------------------------------

mother_folders = [
    p for p in PES_ROOT.iterdir()
    if p.is_dir()
]

print(f"\nMother TS folders found: {len(mother_folders)}")

for p in mother_folders:
    print("  ", p.name)


# ============================================================
# 3. Cremer–Pople classification
# ============================================================

def classify_puckering(Q, theta, phi):
    """
    Approximate qualitative classification of six-membered
    ring puckering using Cremer–Pople Q, theta and phi.

    Important:
    These are 'like' classifications for the constrained
    cyclohexene-type TS ring, not claims of ideal canonical
    cyclohexane conformations.
    """

    # Missing values
    if pd.isna(Q) or pd.isna(theta) or pd.isna(phi):
        return "Not calculated"

    Q = float(Q)
    theta = float(theta)
    phi = float(phi) % 360.0

    # Essentially planar
    if Q < 0.05:
        return "Near-planar"

    # --------------------------------------------------------
    # Six-membered ring:
    #
    # theta ~ 0 / 180:
    #     chair regions
    #
    # theta ~ 90:
    #     boat/twist-boat region
    #
    # intermediate theta:
    #     half-chair / half-boat / intermediate regions
    #
    # Because this is a cyclohexene-type TS, we deliberately
    # use "like" terminology.
    # --------------------------------------------------------

    # Chair-like regions
    if theta <= 20 or theta >= 160:
        return "Chair-like"

    # Boat / twist-boat region
    if 70 <= theta <= 110:

        # Normalize phi to 0–180 for easier comparison
        phi180 = phi % 180.0

        # Approximate boat vs twist-boat division.
        #
        # The exact phase relationship depends on the atom
        # numbering / Cremer–Pople convention. Therefore this
        # is intentionally labeled as approximate.
        if phi180 <= 25 or phi180 >= 155:
            return "Boat-like"

        elif 55 <= phi180 <= 125:
            return "Twist-boat-like"

        else:
            return "Boat/Twist-boat intermediate"

    # --------------------------------------------------------
    # Half-chair / half-boat regions
    # --------------------------------------------------------

    # Near lower intermediate theta
    if 20 < theta < 70:

        if theta < 45:
            return "Half-chair-like"

        else:
            return "Half-boat-like / intermediate"

    # Near upper intermediate theta
    if 110 < theta < 160:

        if theta > 135:
            return "Half-chair-like"

        else:
            return "Half-boat-like / intermediate"

    return "Intermediate"


# ============================================================
# 4. Locate candidate XYZ files
# ============================================================

def find_candidate_xyz(mother_folder, filename):
    """
    Locate an XYZ file whether the CSV contains:
      1. just the filename, or
      2. an absolute path.
    """

    if filename is None:
        return None

    filename = str(filename).strip()

    if not filename or filename.lower() == "nan":
        return None

    # --------------------------------------------------------
    # Case 1: CSV contains an absolute path
    # --------------------------------------------------------
    possible_path = Path(filename)

    if possible_path.is_absolute() and possible_path.exists():
        return possible_path

    # --------------------------------------------------------
    # Case 2: CSV contains just the filename
    # --------------------------------------------------------
    direct_path = mother_folder / filename

    if direct_path.exists():
        return direct_path

    # --------------------------------------------------------
    # Case 3: Search recursively using ONLY the basename
    # This avoids the pathlib absolute-pattern error.
    # --------------------------------------------------------
    basename = possible_path.name

    matches = list(
        mother_folder.rglob(basename)
    )

    if matches:
        return matches[0]

    return None

# ============================================================
# 5. Process every mother TS independently
# ============================================================

all_classified = []
all_summary = []

for mother_folder in sorted(mother_folders):

    mother = mother_folder.name

    csv_file = mother_folder / "TARGETED_PUCKERING_PES.csv"

    if not csv_file.exists():
        print(f"\nSKIP: {mother}")
        print("  TARGETED_PUCKERING_PES.csv not found.")
        continue

    print("\n" + "=" * 70)
    print(f"PROCESSING: {mother}")
    print("=" * 70)

    df = pd.read_csv(csv_file)

    print(f"Rows found: {len(df)}")

    # --------------------------------------------------------
    # Find actual column names
    # --------------------------------------------------------

    def find_column(possible_names):

        lower_map = {
            str(c).strip().lower(): c
            for c in df.columns
        }

        for name in possible_names:
            if name.lower() in lower_map:
                return lower_map[name.lower()]

        # Partial matching
        for c in df.columns:
            cl = str(c).lower()
            for name in possible_names:
                if name.lower() in cl:
                    return c

        return None

    q_col = find_column([
        "Q",
        "Q_Angstrom",
        "Q (Angstrom)",
        "puckering_Q"
    ])

    theta_col = find_column([
        "theta",
        "theta_deg",
        "theta (deg)",
        "Cremer_Pople_theta"
    ])

    phi_col = find_column([
        "phi",
        "phi_deg",
        "phi (deg)",
        "Cremer_Pople_phi"
    ])

    energy_col = find_column([
        "DeltaE_kcal_mol",
        "DeltaE",
        "deltaE",
        "Energy",
        "energy_kcal_mol"
    ])

    file_col = find_column([
        "filename",
        "file",
        "xyz",
        "structure",
        "structure_file"
    ])

    print("\nDetected columns:")
    print("  Q     :", q_col)
    print("  theta :", theta_col)
    print("  phi   :", phi_col)
    print("  energy:", energy_col)
    print("  file  :", file_col)

    if q_col is None or theta_col is None or phi_col is None:
        print("\nWARNING:")
        print("Q/theta/phi columns were not found.")
        print("Classification cannot be performed for this mother.")
        continue

    # --------------------------------------------------------
    # Convert numeric columns
    # --------------------------------------------------------

    df["Q_numeric"] = pd.to_numeric(df[q_col], errors="coerce")
    df["theta_numeric"] = pd.to_numeric(df[theta_col], errors="coerce")
    df["phi_numeric"] = pd.to_numeric(df[phi_col], errors="coerce")

    if energy_col is not None:
        df["Energy_numeric"] = pd.to_numeric(
            df[energy_col], errors="coerce"
        )
    else:
        df["Energy_numeric"] = np.nan

    # --------------------------------------------------------
    # Assign puckering class
    # --------------------------------------------------------

    df["Puckering_class"] = [
        classify_puckering(Q, theta, phi)
        for Q, theta, phi in zip(
            df["Q_numeric"],
            df["theta_numeric"],
            df["phi_numeric"]
        )
    ]

    # --------------------------------------------------------
    # Add mother name
    # --------------------------------------------------------

    df["Mother_TS"] = mother

    # --------------------------------------------------------
    # Save classified CSV
    # --------------------------------------------------------

    classified_csv = (
        mother_folder /
        "TARGETED_PUCKERING_PES_CLASSIFIED.csv"
    )

    df.to_csv(classified_csv, index=False)

    print("\nSaved:")
    print(" ", classified_csv)

    # --------------------------------------------------------
    # Create classified XYZ directory
    # --------------------------------------------------------

    xyz_out = mother_folder / "Classified_XYZ"
    xyz_out.mkdir(exist_ok=True)

    # --------------------------------------------------------
    # Copy XYZ files with puckering class
    # --------------------------------------------------------

    if file_col is not None:

        for idx, row in df.iterrows():

            filename = str(row[file_col])

            if filename.lower() == "nan":
                continue

            source = find_candidate_xyz(
                mother_folder,
                filename
            )

            if source is None:

                print(
                    f"  XYZ not found: {filename}"
                )

                continue

            puckering = str(
                row["Puckering_class"]
            )

            # Safe filename
            safe_class = re.sub(
                r"[^A-Za-z0-9_-]+",
                "_",
                puckering
            )

            candidate_number = idx + 1

            new_name = (
                f"candidate_{candidate_number:04d}"
                f"_{safe_class}.xyz"
            )

            destination = xyz_out / new_name

            shutil.copy2(
                source,
                destination
            )

    # --------------------------------------------------------
    # Lowest-energy structure of each class
    # --------------------------------------------------------

    valid_energy = df[
        np.isfinite(df["Energy_numeric"])
    ].copy()

    if len(valid_energy) > 0:

        summary_rows = []

        for puckering_class, group in valid_energy.groupby(
            "Puckering_class"
        ):

            group = group.sort_values(
                "Energy_numeric"
            )

            lowest = group.iloc[0]

            summary_rows.append({

                "Mother_TS": mother,

                "Puckering_class":
                    puckering_class,

                "Lowest_Energy":
                    lowest["Energy_numeric"],

                "Q_Angstrom":
                    lowest["Q_numeric"],

                "theta_deg":
                    lowest["theta_numeric"],

                "phi_deg":
                    lowest["phi_numeric"],

                "Filename":
                    lowest[file_col]
                    if file_col is not None
                    else "",

                "Number_of_structures":
                    len(group)
            })

        summary_df = pd.DataFrame(summary_rows)

        summary_file = (
            mother_folder /
            "LOWEST_ENERGY_BY_PUCKERING_CLASS.csv"
        )

        summary_df.to_csv(
            summary_file,
            index=False
        )

        all_summary.append(summary_df)

        print("\nLowest energy by puckering class:")

        display(
            summary_df.sort_values(
                "Lowest_Energy"
            )
        )

    all_classified.append(df)


# ============================================================
# 6. Combined summary for ALL mother TS
# ============================================================

if all_classified:

    combined_df = pd.concat(
        all_classified,
        ignore_index=True
    )

    combined_file = (
        PES_ROOT /
        "ALL_TARGETED_PUCKERING_PES_CLASSIFIED.csv"
    )

    combined_df.to_csv(
        combined_file,
        index=False
    )

    print("\n" + "=" * 70)
    print("COMBINED CLASSIFIED DATASET")
    print("=" * 70)

    print(
        f"Total structures: {len(combined_df)}"
    )

    print(
        f"Saved:\n{combined_file}"
    )


# ============================================================
# 7. Combined lowest-energy summary
# ============================================================

if all_summary:

    combined_summary = pd.concat(
        all_summary,
        ignore_index=True
    )

    combined_summary_file = (
        PES_ROOT /
        "ALL_LOWEST_ENERGY_BY_PUCKERING_CLASS.csv"
    )

    combined_summary.to_csv(
        combined_summary_file,
        index=False
    )

    print("\n" + "=" * 70)
    print("FINAL SUMMARY")
    print("=" * 70)

    display(
        combined_summary.sort_values(
            ["Mother_TS", "Lowest_Energy"]
        )
    )

    print(
        "\nSaved:\n"
        f"{combined_summary_file}"
    )


print("\n" + "=" * 70)
print("DONE")
print("=" * 70)
print("\nNo CREST calculation was performed.")
print("No xTB calculation was performed.")
print("Only post-processing/classification was performed.")
```

```python
# ============================================================
# MASTER INTRA-R3 + INTER-R3 PUCKERING ANALYSIS
# CORRECTED VERSION — RECURSIVE SEARCH
#
# This version does NOT assume that R3 folders are directly
# inside FAST_Cyclohexene_PES.
#
# It searches recursively for:
#
#   TARGETED_PUCKERING_PES_CLASSIFIED.csv
#
# and processes every mother/R3 independently.
#
# NO CREST
# NO xTB
# NO new quantum calculations
# ============================================================

import os
import re
import shutil
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
from pathlib import Path

# ------------------------------------------------------------
# 1. Google Drive
# ------------------------------------------------------------

# Drive is already mounted in this Colab session.
# Do NOT call drive.mount() again.

import os

DRIVE_ROOT = "/content/drive/MyDrive"

if not os.path.isdir(DRIVE_ROOT):
    raise RuntimeError(
        "Google Drive is not mounted. "
        "Please mount Drive once before running this cell."
    )

print("Google Drive is already mounted.")
print("MyDrive exists:", os.path.isdir(DRIVE_ROOT))

# ------------------------------------------------------------
# 2. Main paths
# ------------------------------------------------------------

BASE = Path(
    "/content/drive/MyDrive/"
    "Cycloetherification_xyz_files"
)

PES_ROOT = (
    BASE /
    "CREST_results" /
    "FAST_Cyclohexene_PES"
)

OUTPUT_ROOT = (
    PES_ROOT /
    "MASTER_INTRA_INTER_ANALYSIS"
)

OUTPUT_ROOT.mkdir(
    parents=True,
    exist_ok=True
)

BEST_ROOT = (
    OUTPUT_ROOT /
    "BEST_CONFORMERS"
)

BEST_ROOT.mkdir(
    parents=True,
    exist_ok=True
)

print("=" * 75)
print("MASTER INTRA-R3 + INTER-R3 PUCKERING ANALYSIS")
print("=" * 75)

print("\nInput:")
print(PES_ROOT)

print("\nOutput:")
print(OUTPUT_ROOT)


# ============================================================
# 3. FIND ALL CLASSIFIED CSV FILES RECURSIVELY
# ============================================================

classified_files = sorted(
    PES_ROOT.rglob(
        "TARGETED_PUCKERING_PES_CLASSIFIED.csv"
    )
)

print("\n" + "=" * 75)
print("SEARCH RESULTS")
print("=" * 75)

print(
    f"\nClassified CSV files found: "
    f"{len(classified_files)}"
)

if not classified_files:

    print(
        "\nERROR: No "
        "TARGETED_PUCKERING_PES_CLASSIFIED.csv "
        "files were found."
    )

    print(
        "\nLet's inspect the actual directory structure:"
    )

    # Show directories containing CSV files
    csv_files = sorted(
        PES_ROOT.rglob("*.csv")
    )

    print(
        f"\nTotal CSV files found: "
        f"{len(csv_files)}"
    )

    for f in csv_files[:100]:
        print("   ", f)

    raise RuntimeError(
        "\nNo TARGETED_PUCKERING_PES_CLASSIFIED.csv "
        "files found under FAST_Cyclohexene_PES."
    )


for f in classified_files:

    print(
        "\n   ",
        f
    )


# ============================================================
# 4. HELPER FUNCTIONS
# ============================================================

def find_column(df, candidates):

    # Exact matching
    lower_map = {
        str(c).strip().lower(): c
        for c in df.columns
    }

    for candidate in candidates:

        key = candidate.lower()

        if key in lower_map:
            return lower_map[key]

    # Partial matching
    for c in df.columns:

        cl = str(c).strip().lower()

        for candidate in candidates:

            if candidate.lower() in cl:
                return c

    return None


def clean_class_name(x):

    if pd.isna(x):
        return "Unknown"

    x = str(x).strip()

    replacements = {

        "Boat/Twist-boat intermediate":
            "Boat/Twist intermediate",

        "Half-boat-like / intermediate":
            "Half-boat/intermediate"
    }

    return replacements.get(
        x,
        x
    )


def safe_filename(x):

    return re.sub(
        r"[^A-Za-z0-9_.-]+",
        "_",
        str(x)
    )


# ============================================================
# 5. READ EVERY R3 / MOTHER CSV
# ============================================================

all_data = []

print("\n" + "=" * 75)
print("READING R3 RESULTS")
print("=" * 75)


for csv_file in classified_files:

    # --------------------------------------------------------
    # The mother/R3 folder is the directory immediately
    # containing the classified CSV.
    # --------------------------------------------------------

    mother_folder = csv_file.parent

    mother = mother_folder.name

    print(
        f"\nProcessing: {mother}"
    )

    print(
        f"CSV: {csv_file}"
    )

    try:

        df = pd.read_csv(
            csv_file
        )

    except Exception as e:

        print(
            f"  ERROR reading file: {e}"
        )

        continue

    print(
        f"  Structures: {len(df)}"
    )

    # --------------------------------------------------------
    # Detect columns
    # --------------------------------------------------------

    q_col = find_column(
        df,
        [
            "Q_numeric",
            "Q",
            "puckering_Q"
        ]
    )

    theta_col = find_column(
        df,
        [
            "theta_numeric",
            "theta",
            "theta_deg"
        ]
    )

    phi_col = find_column(
        df,
        [
            "phi_numeric",
            "phi",
            "phi_deg"
        ]
    )

    energy_col = find_column(
        df,
        [
            "Energy_numeric",
            "DeltaE_kcal_mol",
            "DeltaE",
            "energy"
        ]
    )

    class_col = find_column(
        df,
        [
            "Puckering_class",
            "puckering_class"
        ]
    )

    file_col = find_column(
        df,
        [
            "filename",
            "file",
            "xyz",
            "structure",
            "structure_file"
        ]
    )

    print("  Columns:")
    print("     Q      :", q_col)
    print("     theta  :", theta_col)
    print("     phi    :", phi_col)
    print("     Energy :", energy_col)
    print("     Class  :", class_col)
    print("     XYZ    :", file_col)

    if energy_col is None:

        print(
            "  WARNING: Energy column not found."
        )

        continue

    # --------------------------------------------------------
    # Standardized columns
    # --------------------------------------------------------

    df["Mother_TS"] = mother

    df["Source_CSV"] = str(
        csv_file
    )

    df["Source_folder"] = str(
        mother_folder
    )

    if q_col is not None:

        df["Q_Angstrom"] = pd.to_numeric(
            df[q_col],
            errors="coerce"
        )

    else:

        df["Q_Angstrom"] = np.nan

    if theta_col is not None:

        df["theta_deg"] = pd.to_numeric(
            df[theta_col],
            errors="coerce"
        )

    else:

        df["theta_deg"] = np.nan

    if phi_col is not None:

        df["phi_deg"] = pd.to_numeric(
            df[phi_col],
            errors="coerce"
        )

    else:

        df["phi_deg"] = np.nan

    df["Energy"] = pd.to_numeric(
        df[energy_col],
        errors="coerce"
    )

    if class_col is not None:

        df["Puckering_class_clean"] = (
            df[class_col]
            .apply(clean_class_name)
        )

    else:

        df["Puckering_class_clean"] = (
            "Unknown"
        )

    if file_col is not None:

        df["XYZ_filename"] = (
            df[file_col]
            .astype(str)
        )

    else:

        df["XYZ_filename"] = ""

    # --------------------------------------------------------
    # IMPORTANT:
    #
    # Calculate ΔE relative to the LOWEST conformer
    # WITHIN THIS R3 ONLY.
    # --------------------------------------------------------

    valid = np.isfinite(
        df["Energy"]
    )

    if valid.any():

        minimum_energy = df.loc[
            valid,
            "Energy"
        ].min()

        df["DeltaE_within_R3"] = (
            df["Energy"] -
            minimum_energy
        )

    else:

        df["DeltaE_within_R3"] = np.nan

    all_data.append(df)


# ============================================================
# 6. COMBINE EVERYTHING
# ============================================================

if not all_data:

    raise RuntimeError(
        "No usable classified R3 datasets were found."
    )

all_df = pd.concat(
    all_data,
    ignore_index=True
)

print("\n" + "=" * 75)
print("COMBINED DATA")
print("=" * 75)

print(
    "\nTotal structures:",
    len(all_df)
)

print(
    "Total R3 groups:",
    all_df["Mother_TS"].nunique()
)

print(
    "\nR3 groups:"
)

for x in sorted(
    all_df["Mother_TS"].unique()
):

    print("   ", x)


# ============================================================
# 7. TABLE 1 — INTRA-R3
#
# Lowest-energy structure of EACH puckering class
# within each R3.
# ============================================================

print("\n" + "=" * 75)
print("TABLE 1 — INTRA-R3")
print("=" * 75)

intra_rows = []

for mother, group in all_df.groupby(
    "Mother_TS"
):

    group = group[
        np.isfinite(group["Energy"])
    ].copy()

    if group.empty:
        continue

    # Lowest overall conformer for this R3
    global_min = group.loc[
        group["Energy"].idxmin()
    ]

    global_energy = (
        global_min["Energy"]
    )

    # One lowest conformer per class
    for puckering_class, class_group in (
        group.groupby(
            "Puckering_class_clean"
        )
    ):

        lowest = class_group.loc[
            class_group["Energy"].idxmin()
        ]

        intra_rows.append({

            "R3":
                mother,

            "Puckering_class":
                puckering_class,

            "Lowest_Energy":
                lowest["Energy"],

            "DeltaE_from_R3_global_min":
                (
                    lowest["Energy"] -
                    global_energy
                ),

            "Q_Angstrom":
                lowest["Q_Angstrom"],

            "theta_deg":
                lowest["theta_deg"],

            "phi_deg":
                lowest["phi_deg"],

            "XYZ_filename":
                lowest["XYZ_filename"],

            "Number_of_structures":
                len(class_group)
        })


intra_df = pd.DataFrame(
    intra_rows
)

intra_df = intra_df.sort_values(
    [
        "R3",
        "DeltaE_from_R3_global_min"
    ]
)

intra_file = (
    OUTPUT_ROOT /
    "INTRA_R3_CONFORMATION_TABLE.csv"
)

intra_df.to_csv(
    intra_file,
    index=False
)

print(
    "\nSaved:"
)

print(intra_file)

display(
    intra_df
)


# ============================================================
# 8. TABLE 2 — INTER-R3
#
# Lowest-energy conformer for each R3.
# ============================================================

print("\n" + "=" * 75)
print("TABLE 2 — INTER-R3")
print("=" * 75)

inter_rows = []

for mother, group in all_df.groupby(
    "Mother_TS"
):

    group = group[
        np.isfinite(group["Energy"])
    ].copy()

    if group.empty:
        continue

    lowest = group.loc[
        group["Energy"].idxmin()
    ]

    inter_rows.append({

        "R3":
            mother,

        "Best_puckering_class":
            lowest["Puckering_class_clean"],

        "Lowest_Energy":
            lowest["Energy"],

        "Q_Angstrom":
            lowest["Q_Angstrom"],

        "theta_deg":
            lowest["theta_deg"],

        "phi_deg":
            lowest["phi_deg"],

        "XYZ_filename":
            lowest["XYZ_filename"],

        "Number_of_structures":
            len(group)
    })


inter_df = pd.DataFrame(
    inter_rows
)

inter_df = inter_df.sort_values(
    "R3"
)

inter_file = (
    OUTPUT_ROOT /
    "INTER_R3_LOWEST_CONFORMERS.csv"
)

inter_df.to_csv(
    inter_file,
    index=False
)

print(
    "\nSaved:"
)

print(inter_file)

display(
    inter_df
)


# ============================================================
# 9. COMPACT INTRA-R3 MATRIX
# ============================================================

pivot_intra = (
    intra_df
    .pivot_table(
        index="R3",
        columns="Puckering_class",
        values="DeltaE_from_R3_global_min",
        aggfunc="min"
    )
    .reset_index()
)

pivot_file = (
    OUTPUT_ROOT /
    "INTRA_R3_CONFORMATION_ENERGY_MATRIX.csv"
)

pivot_intra.to_csv(
    pivot_file,
    index=False
)

print(
    "\nCompact INTRA-R3 matrix:"
)

display(
    pivot_intra
)


# ============================================================
# 10. PES MAP 1
#
# INTRA-R3:
# ΔE within each R3 vs theta
# ============================================================

print(
    "\nGenerating INTRA-R3 PES map..."
)

plot_df = all_df[
    np.isfinite(
        all_df["Energy"]
    )
    &
    np.isfinite(
        all_df["theta_deg"]
    )
].copy()

fig, ax = plt.subplots(
    figsize=(12, 8)
)

for mother, group in plot_df.groupby(
    "Mother_TS"
):

    ax.scatter(
        group["theta_deg"],
        group["DeltaE_within_R3"],
        s=40,
        alpha=0.70,
        label=mother
    )

ax.set_xlabel(
    "Cremer–Pople θ (degree)"
)

ax.set_ylabel(
    "ΔE within R3 (kcal mol$^{-1}$)"
)

ax.set_title(
    "INTRA-R3 Cyclohexene TS Puckering PES"
)

ax.axhline(
    0,
    linewidth=1
)

ax.grid(
    alpha=0.25
)

ax.legend(
    bbox_to_anchor=(1.02, 1),
    loc="upper left",
    fontsize=8
)

plt.tight_layout()

intra_pes_file = (
    OUTPUT_ROOT /
    "INTRA_R3_PES_MAP_theta_vs_DeltaE.png"
)

plt.savefig(
    intra_pes_file,
    dpi=300,
    bbox_inches="tight"
)

plt.show()

print(
    "\nSaved:"
)

print(intra_pes_file)


# ============================================================
# 11. PES MAP 2
#
# INTER-R3:
# R3 × puckering class
#
# Values are ΔE relative to the LOWEST conformer
# within each R3.
# ============================================================

print(
    "\nGenerating INTER-R3 puckering heatmap..."
)

heat_df = (
    intra_df
    .pivot_table(
        index="R3",
        columns="Puckering_class",
        values="DeltaE_from_R3_global_min",
        aggfunc="min"
    )
)

# Consistent column order
preferred_order = [
    "Chair-like",
    "Half-chair-like",
    "Half-boat/intermediate",
    "Boat-like",
    "Twist-boat-like",
    "Boat/Twist intermediate",
    "Intermediate",
    "Near-planar",
    "Unknown"
]

existing_columns = [
    x for x in preferred_order
    if x in heat_df.columns
]

remaining_columns = [
    x for x in heat_df.columns
    if x not in existing_columns
]

heat_df = heat_df[
    existing_columns +
    remaining_columns
]

fig, ax = plt.subplots(
    figsize=(
        max(9, 1.4 * len(heat_df.columns)),
        max(6, 0.45 * len(heat_df))
    )
)

matrix = heat_df.values

im = ax.imshow(
    matrix,
    aspect="auto",
    interpolation="nearest"
)

ax.set_xticks(
    np.arange(
        len(heat_df.columns)
    )
)

ax.set_xticklabels(
    heat_df.columns,
    rotation=45,
    ha="right"
)

ax.set_yticks(
    np.arange(
        len(heat_df.index)
    )
)

ax.set_yticklabels(
    heat_df.index
)

ax.set_xlabel(
    "Puckering class"
)

ax.set_ylabel(
    "R3"
)

ax.set_title(
    "INTER-R3 Puckering Energy Landscape\n"
    "ΔE relative to the lowest conformer of each R3"
)

# Add values
for i in range(
    matrix.shape[0]
):

    for j in range(
        matrix.shape[1]
    ):

        value = matrix[i, j]

        if np.isfinite(value):

            ax.text(
                j,
                i,
                f"{value:.2f}",
                ha="center",
                va="center",
                fontsize=8
            )

cbar = plt.colorbar(
    im,
    ax=ax
)

cbar.set_label(
    "ΔE within R3 (kcal mol$^{-1}$)"
)

plt.tight_layout()

inter_pes_file = (
    OUTPUT_ROOT /
    "INTER_R3_PUCKERING_HEATMAP.png"
)

plt.savefig(
    inter_pes_file,
    dpi=300,
    bbox_inches="tight"
)

plt.show()

print(
    "\nSaved:"
)

print(inter_pes_file)


# ============================================================
# 12. COPY BEST CONFORMER FOR EVERY R3
# ============================================================

print("\n" + "=" * 75)
print("COPYING BEST CONFORMERS")
print("=" * 75)

best_rows = []

for number, (_, row) in enumerate(
    inter_df.iterrows(),
    start=1
):

    mother = row["R3"]

    filename = str(
        row["XYZ_filename"]
    ).strip()

    if (
        not filename
        or filename.lower() == "nan"
    ):

        print(
            f"WARNING: no XYZ filename for {mother}"
        )

        continue

    mother_folder = (
        PES_ROOT /
        mother
    )

    basename = Path(
        filename
    ).name

    # --------------------------------------------------------
    # Possible locations
    # --------------------------------------------------------

    possible_locations = [

        mother_folder /
        "TARGETED_PUCKERING_STRUCTURES" /
        basename,

        mother_folder /
        "Classified_XYZ" /
        basename,

        mother_folder /
        basename
    ]

    source = None

    for candidate in possible_locations:

        if candidate.exists():

            source = candidate
            break

    # --------------------------------------------------------
    # Recursive fallback
    # --------------------------------------------------------

    if source is None:

        matches = list(
            mother_folder.rglob(
                basename
            )
        )

        if matches:

            source = matches[0]

    # --------------------------------------------------------
    # If still missing
    # --------------------------------------------------------

    if source is None:

        print(
            f"WARNING: XYZ not found:"
            f"\n  R3 = {mother}"
            f"\n  file = {basename}"
        )

        continue

    # --------------------------------------------------------
    # New filename
    # --------------------------------------------------------

    class_name = safe_filename(
        row["Best_puckering_class"]
    )

    new_name = (
        f"best_conformer{number}_"
        f"{safe_filename(mother)}_"
        f"{class_name}.xyz"
    )

    destination = (
        BEST_ROOT /
        new_name
    )

    shutil.copy2(
        source,
        destination
    )

    print(
        f"{number:02d}. "
        f"{mother}"
        f"  →  "
        f"{new_name}"
    )

    best_rows.append({

        "Best_number":
            number,

        "R3":
            mother,

        "Puckering_class":
            row["Best_puckering_class"],

        "Energy":
            row["Lowest_Energy"],

        "Q_Angstrom":
            row["Q_Angstrom"],

        "theta_deg":
            row["theta_deg"],

        "phi_deg":
            row["phi_deg"],

        "Original_XYZ":
            str(source),

        "Copied_XYZ":
            str(destination)
    })


best_df = pd.DataFrame(
    best_rows
)

best_file = (
    OUTPUT_ROOT /
    "BEST_CONFORMERS_MASTER_TABLE.csv"
)

best_df.to_csv(
    best_file,
    index=False
)


# ============================================================
# 13. GLOBAL MINIMUM PUCKERING CLASS DISTRIBUTION
# ============================================================

class_distribution = (
    inter_df[
        "Best_puckering_class"
    ]
    .value_counts()
    .rename_axis(
        "Puckering_class"
    )
    .reset_index(
        name="Number_of_R3_groups"
    )
)

class_file = (
    OUTPUT_ROOT /
    "GLOBAL_MINIMUM_PUCKERING_CLASS_DISTRIBUTION.csv"
)

class_distribution.to_csv(
    class_file,
    index=False
)

print(
    "\nGlobal minimum puckering classes:"
)

display(
    class_distribution
)


# ============================================================
# 14. SAVE COMPLETE DATASET
# ============================================================

all_file = (
    OUTPUT_ROOT /
    "ALL_INTRA_R3_ANALYSIS.csv"
)

all_df.to_csv(
    all_file,
    index=False
)


# ============================================================
# 15. FINAL REPORT
# ============================================================

print("\n")
print("=" * 75)
print("MASTER ANALYSIS COMPLETED")
print("=" * 75)

print(
    "\nTotal R3 groups:",
    all_df["Mother_TS"].nunique()
)

print(
    "Total structures:",
    len(all_df)
)

print(
    "\nOUTPUT DIRECTORY:"
)

print(
    OUTPUT_ROOT
)

print("\nMAIN FILES:")

print(
    "\n1. INTRA-R3:"
)

print(
    intra_file
)

print(
    "\n2. INTER-R3:"
)

print(
    inter_file
)

print(
    "\n3. INTRA-R3 PES:"
)

print(
    intra_pes_file
)

print(
    "\n4. INTER-R3 PES:"
)

print(
    inter_pes_file
)

print(
    "\n5. BEST CONFORMER TABLE:"
)

print(
    best_file
)

print(
    "\n6. BEST XYZ STRUCTURES:"
)

print(
    BEST_ROOT
)

print("\n" + "=" * 75)
print("DONE — NO NEW CREST/xTB CALCULATIONS")
print("=" * 75)
```

```python
# ================================================================
# FINAL CREST CONFORMATIONAL ANALYSIS
# DDQ / R3 CYCLOETHERIFICATION
#
# SOURCE:
#   ALL_retained_pucker_results.csv
#   = actual retained CREST conformers
#
# NO CREST / xTB calculations are performed.
#
# ANALYSES:
#   1. Lowest conformer of every R3
#   2. Boltzmann populations of actual CREST conformers
#   3. Puckering-class populations
#   4. Conformational preference map
#   5. RR vs RS comparison
#   6. Conformational flexibility
#   7. Torsional flexibility
#   8. Experimental %de comparison
#   9. Pearson + Spearman correlations
#  10. Publication-oriented figures
#  11. Best retained XYZ structures
#  12. Master Excel workbook
#
# IMPORTANT:
#   Absolute energies are NOT compared between different R3s.
#   All conformational ΔE values are calculated within the same R3.
# ================================================================


# ================================================================
# 1. IMPORTS
# ================================================================

from pathlib import Path
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import shutil
import re
import math

from scipy.stats import pearsonr, spearmanr


# ================================================================
# 2. GOOGLE DRIVE CHECK
# ================================================================
# Do NOT mount Drive here.
# Drive should already be mounted by your first Colab cell.

DRIVE_ROOT = Path("/content/drive/MyDrive")

if not DRIVE_ROOT.exists():
    raise RuntimeError(
        "Google Drive is not mounted.\n"
        "Run your Drive-mount cell ONCE, then run this cell."
    )

print("Google Drive is already mounted.")


# ================================================================
# 3. PATHS
# ================================================================

BASE = (
    DRIVE_ROOT
    / "Cycloetherification_xyz_files"
)

CREST_ROOT = (
    BASE
    / "CREST_results"
)

INPUT_CSV = (
    CREST_ROOT
    / "ALL_retained_pucker_results.csv"
)

OUTPUT_ROOT = (
    CREST_ROOT
    / "FINAL_CRE​ST_CONFORMATIONAL_ANALYSIS"
)

# Remove accidental hidden Unicode possibility:
OUTPUT_ROOT = (
    CREST_ROOT
    / "FINAL_CREST_CONFORMATIONAL_ANALYSIS"
)

TABLE_ROOT = OUTPUT_ROOT / "TABLES"
FIGURE_ROOT = OUTPUT_ROOT / "FIGURES"
BEST_ROOT = OUTPUT_ROOT / "BEST_RETAINED_CONFORMERS"

TABLE_ROOT.mkdir(
    parents=True,
    exist_ok=True
)

FIGURE_ROOT.mkdir(
    parents=True,
    exist_ok=True
)

BEST_ROOT.mkdir(
    parents=True,
    exist_ok=True
)


print("\n================================================")
print("INPUT")
print("================================================")
print(INPUT_CSV)

if not INPUT_CSV.exists():
    raise FileNotFoundError(
        f"\nInput file not found:\n{INPUT_CSV}"
    )

print("\nInput file exists: YES")


# ================================================================
# 4. THERMODYNAMIC CONSTANTS
# ================================================================

T = 298.15

R = 0.001987204258

RT = R * T

print("\nTemperature:", T, "K")
print("RT:", RT, "kcal mol^-1")


# ================================================================
# 5. EXPERIMENTAL %de DATA
# ================================================================
#
# RR is the experimental major product.
#
# Folder/R3 names are matched to your CREST naming.
# ================================================================

EXPERIMENTAL_DE = {

    "Me": 56,
    "Et": 56,
    "iPr": 72,
    "nBu": 56,
    "tBu": 98,
    "Me-Propane": 60,
    "Cy": 72,
    "CH2CO2Me": 42,
    "CMe2CO2Me": 92,
    "allyl": 34,
    "dimethyl-allyl": 90,
    "CH2-dioxane": 34,
    "CH2-Ph": 56,
    "CH2CN": 60,
    "Ph": 34,
    "alkene": 34,
    "alkyne": 0,
}


R3_DISPLAY = {

    "Me": "Me",

    "Et": "Et",

    "iPr": "iPr",

    "nBu": "n-Bu",

    "tBu": "t-Bu",

    "Me-Propane":
        "CH2-CH(Me)-CH3",

    "Cy": "Cy",

    "CH2CO2Me":
        "CH2CO2Me",

    "CMe2CO2Me":
        "CMe2CO2Me",

    "allyl":
        "CH2CH=CH2",

    "dimethyl-allyl":
        "CMe2CH=CH2",

    "CH2-dioxane":
        "CH2-1,3-dioxolane",

    "CH2-Ph":
        "CH2Ph",

    "CH2CN":
        "CH2CN",

    "Ph":
        "Phenyl",

    "alkene":
        "CH=CH2",

    "alkyne":
        "C#CH",
}


# ================================================================
# 6. READ ACTUAL CREST DATA
# ================================================================

print("\n================================================")
print("READING ACTUAL CREST RETAINED CONFORMERS")
print("================================================")

df = pd.read_csv(INPUT_CSV)

print("Rows in source CSV:", len(df))

print("\nColumns detected:")

for i, col in enumerate(df.columns, 1):
    print(f"{i:2d}. {col}")


# ================================================================
# 7. CHECK REQUIRED COLUMNS
# ================================================================

required_columns = [

    "structure",
    "original_conformer",
    "energy_Eh",
    "forming_bond_A",
    "Q_A",
    "theta_deg",
    "phi_deg",
    "classification",

    "torsion_1_deg",
    "torsion_2_deg",
    "torsion_3_deg",
    "torsion_4_deg",
    "torsion_5_deg",
    "torsion_6_deg",

    "retained_conformer",
    "status",
    "rel_kcal",
]

missing = [
    c for c in required_columns
    if c not in df.columns
]

if missing:

    raise RuntimeError(
        "Required columns missing:\n"
        + "\n".join(missing)
    )

print("\nAll required columns are present.")


# ================================================================
# 8. KEEP ONLY ACTUAL RETAINED STRUCTURES
# ================================================================

df["status"] = (
    df["status"]
    .astype(str)
    .str.strip()
    .str.lower()
)

df = df[
    df["status"] == "retained"
].copy()

print(
    "\nRows with status='retained':",
    len(df)
)


# ================================================================
# 9. IDENTIFY RR / RS
# ================================================================

df["structure"] = (
    df["structure"]
    .astype(str)
    .str.strip()
)

df["Configuration"] = np.where(
    df["structure"].str.startswith("RR_"),
    "RR",
    np.where(
        df["structure"].str.startswith("RS_"),
        "RS",
        "Unknown"
    )
)

unknown = df[
    df["Configuration"] == "Unknown"
]

if len(unknown) > 0:

    print(
        "\nWARNING:",
        len(unknown),
        "rows have unknown RR/RS assignment."
    )

    print(
        unknown["structure"].tolist()
    )

    df = df[
        df["Configuration"] != "Unknown"
    ].copy()


# ================================================================
# 10. EXTRACT R3 NAME
# ================================================================

df["R3"] = (
    df["structure"]
    .str.replace(
        r"^(RR_|RS_)",
        "",
        regex=True
    )
)

print("\nR3 systems found:")

for r3 in sorted(df["R3"].unique()):

    print(
        f"  {r3}: "
        f"{len(df[df['R3'] == r3])} retained"
    )


# ================================================================
# 11. NUMERIC CONVERSION
# ================================================================

numeric_columns = [

    "energy_Eh",
    "forming_bond_A",
    "Q_A",
    "theta_deg",
    "phi_deg",

    "torsion_1_deg",
    "torsion_2_deg",
    "torsion_3_deg",
    "torsion_4_deg",
    "torsion_5_deg",
    "torsion_6_deg",

    "rel_kcal",
]

for col in numeric_columns:

    df[col] = pd.to_numeric(
        df[col],
        errors="coerce"
    )


# ================================================================
# 12. WITHIN-R3 MINIMUM ENERGY
# ================================================================
#
# We use rel_kcal from the CREST analysis where available.
#
# To guarantee consistency, we also calculate ΔE directly
# from energy_Eh within each individual R3/configuration.
# ================================================================

df["E_min_Eh"] = (
    df
    .groupby(
        [
            "R3",
            "Configuration"
        ]
    )["energy_Eh"]
    .transform("min")
)

df["DeltaE_kcal_mol"] = (
    df["energy_Eh"]
    - df["E_min_Eh"]
) * 627.509474


# ================================================================
# 13. BOLTZMANN POPULATIONS
# ================================================================

df["Boltzmann_weight"] = np.exp(
    -df["DeltaE_kcal_mol"] / RT
)

df["Population_fraction"] = (
    df["Boltzmann_weight"]
    /
    df.groupby(
        [
            "R3",
            "Configuration"
        ]
    )["Boltzmann_weight"]
    .transform("sum")
)

df["Population_percent"] = (
    df["Population_fraction"] * 100
)


# ================================================================
# 14. ACTUAL CREST POPULATION TABLE
# ================================================================

population_table = df[
    [
        "R3",
        "Configuration",
        "retained_conformer",
        "structure",
        "energy_Eh",
        "DeltaE_kcal_mol",
        "Q_A",
        "theta_deg",
        "phi_deg",
        "classification",
        "Population_percent",
    ]
].copy()

population_table = population_table.sort_values(
    [
        "R3",
        "Configuration",
        "DeltaE_kcal_mol"
    ]
)

population_table.to_csv(
    TABLE_ROOT
    / "01_ACTUAL_CREST_CONFORMER_POPULATIONS.csv",
    index=False
)


# ================================================================
# 15. LOWEST CONFORMER OF EVERY R3 / RR / RS
# ================================================================

lowest_rows = []

for (
    r3,
    config
), group in df.groupby(
    [
        "R3",
        "Configuration"
    ]
):

    row = group.loc[
        group["DeltaE_kcal_mol"].idxmin()
    ]

    lowest_rows.append({

        "R3":
            r3,

        "Configuration":
            config,

        "Lowest_conformer":
            row["retained_conformer"],

        "Structure":
            row["structure"],

        "Energy_Eh":
            row["energy_Eh"],

        "DeltaE_kcal_mol":
            row["DeltaE_kcal_mol"],

        "Q_A":
            row["Q_A"],

        "theta_deg":
            row["theta_deg"],

        "phi_deg":
            row["phi_deg"],

        "Puckering_class":
            row["classification"],

        "Population_percent":
            row["Population_percent"],

        "N_retained":
            len(group),
    })


lowest = pd.DataFrame(
    lowest_rows
)

lowest = lowest.sort_values(
    [
        "R3",
        "Configuration"
    ]
)

lowest.to_csv(
    TABLE_ROOT
    / "02_LOWEST_CONFORMER_EVERY_R3.csv",
    index=False
)


# ================================================================
# 16. PUCKERING-CLASS POPULATIONS
# ================================================================

class_population = (
    df
    .groupby(
        [
            "R3",
            "Configuration",
            "classification"
        ],
        as_index=False
    )
    .agg(

        N_conformers=(
            "structure",
            "count"
        ),

        Lowest_DeltaE_kcal_mol=(
            "DeltaE_kcal_mol",
            "min"
        ),

        Total_population_percent=(
            "Population_percent",
            "sum"
        ),
    )
)

class_population.to_csv(
    TABLE_ROOT
    / "03_PUCKERING_CLASS_POPULATIONS.csv",
    index=False
)


# ================================================================
# 17. PUCKERING PREFERENCE MAP
# ================================================================

preference_map = (
    class_population
    .pivot_table(
        index=[
            "R3",
            "Configuration"
        ],
        columns="classification",
        values="Lowest_DeltaE_kcal_mol",
        aggfunc="min"
    )
)

preference_map.to_csv(
    TABLE_ROOT
    / "04_CONFORMATIONAL_PREFERENCE_MAP.csv"
)


# ================================================================
# 18. PREFERRED PUCKERING CLASS
# ================================================================

preferred_rows = []

for (
    r3,
    config
), group in class_population.groupby(
    [
        "R3",
        "Configuration"
    ]
):

    row = group.loc[
        group["Lowest_DeltaE_kcal_mol"].idxmin()
    ]

    preferred_rows.append({

        "R3":
            r3,

        "Configuration":
            config,

        "Preferred_puckering_class":
            row["classification"],

        "Lowest_class_DeltaE_kcal_mol":
            row["Lowest_DeltaE_kcal_mol"],

        "Population_percent_of_class":
            row["Total_population_percent"],

        "N_conformers_in_class":
            row["N_conformers"],
    })


preferred = pd.DataFrame(
    preferred_rows
)

preferred.to_csv(
    TABLE_ROOT
    / "05_PREFERRED_PUCKERING_CLASS.csv",
    index=False
)


# ================================================================
# 19. CONFORMATIONAL FLEXIBILITY
# ================================================================

flexibility_rows = []

for (
    r3,
    config
), group in df.groupby(
    [
        "R3",
        "Configuration"
    ]
):

    populations = (
        group["Population_fraction"]
        .values
    )

    N_eff = (
        1.0
        /
        np.sum(
            populations ** 2
        )
    )

    # ------------------------------------------------------------
    # Six torsional coordinates
    # ------------------------------------------------------------

    torsion_columns = [
        "torsion_1_deg",
        "torsion_2_deg",
        "torsion_3_deg",
        "torsion_4_deg",
        "torsion_5_deg",
        "torsion_6_deg",
    ]

    torsion_std = (
        group[
            torsion_columns
        ]
        .std(
            skipna=True
        )
    )

    # RMS torsional spread
    torsional_flexibility = np.sqrt(
        np.nanmean(
            torsion_std.values ** 2
        )
    )

    # ------------------------------------------------------------
    # Energy range
    # ------------------------------------------------------------

    energy_span = (
        group["DeltaE_kcal_mol"].max()
    )

    flexibility_rows.append({

        "R3":
            r3,

        "Configuration":
            config,

        "N_retained":
            len(group),

        "N_puckering_classes":
            group["classification"].nunique(),

        "N_within_0.5_kcal":
            int(
                (
                    group["DeltaE_kcal_mol"]
                    <= 0.5
                ).sum()
            ),

        "N_within_1.0_kcal":
            int(
                (
                    group["DeltaE_kcal_mol"]
                    <= 1.0
                ).sum()
            ),

        "N_within_2.0_kcal":
            int(
                (
                    group["DeltaE_kcal_mol"]
                    <= 2.0
                ).sum()
            ),

        "N_within_3.0_kcal":
            int(
                (
                    group["DeltaE_kcal_mol"]
                    <= 3.0
                ).sum()
            ),

        "N_within_5.0_kcal":
            int(
                (
                    group["DeltaE_kcal_mol"]
                    <= 5.0
                ).sum()
            ),

        "Energy_span_kcal_mol":
            energy_span,

        "Lowest_conformer_population_percent":
            group["Population_percent"].max(),

        "Effective_N_conformers":
            N_eff,

        "Torsional_flexibility_deg":
            torsional_flexibility,

        "Mean_Q_A":
            group["Q_A"].mean(),

        "Std_Q_A":
            group["Q_A"].std(),

        "Mean_theta_deg":
            group["theta_deg"].mean(),

        "Std_theta_deg":
            group["theta_deg"].std(),

    })


flexibility = pd.DataFrame(
    flexibility_rows
)

flexibility.to_csv(
    TABLE_ROOT
    / "06_CONFORMATIONAL_FLEXIBILITY.csv",
    index=False
)


# ================================================================
# 20. RR vs RS COMPARISON
# ================================================================

rr = lowest[
    lowest["Configuration"] == "RR"
].copy()

rs = lowest[
    lowest["Configuration"] == "RS"
].copy()


rr = rr.rename(
    columns={
        c: f"{c}_RR"
        for c in rr.columns
        if c != "R3"
    }
)

rs = rs.rename(
    columns={
        c: f"{c}_RS"
        for c in rs.columns
        if c != "R3"
    }
)


rr_rs = pd.merge(
    rr,
    rs,
    on="R3",
    how="inner"
)


# ---------------------------------------------------------------
# E_RS - E_RR
#
# Positive = RS minimum is higher than RR minimum
# Negative = RS minimum is lower than RR minimum
# ---------------------------------------------------------------

rr_rs["DeltaDeltaE_RS_minus_RR"] = (
    rr_rs["Energy_Eh_RS"]
    -
    rr_rs["Energy_Eh_RR"]
) * 627.509474


rr_rs["Abs_DeltaDeltaE"] = (
    np.abs(
        rr_rs[
            "DeltaDeltaE_RS_minus_RR"
        ]
    )
)


rr_rs["Same_puckering_class"] = (
    rr_rs[
        "Puckering_class_RR"
    ]
    ==
    rr_rs[
        "Puckering_class_RS"
    ]
)


# ---------------------------------------------------------------
# Population ratio
# ---------------------------------------------------------------

rr_rs["RR_lowest_population_percent"] = (
    rr_rs[
        "Population_percent_RR"
    ]
)

rr_rs["RS_lowest_population_percent"] = (
    rr_rs[
        "Population_percent_RS"
    ]
)

rr_rs["Population_difference_RR_minus_RS"] = (
    rr_rs[
        "RR_lowest_population_percent"
    ]
    -
    rr_rs[
        "RS_lowest_population_percent"
    ]
)


rr_rs.to_csv(
    TABLE_ROOT
    / "07_RR_vs_RS_COMPARISON.csv",
    index=False
)


# ================================================================
# 21. EXPERIMENTAL DATAFRAME
# ================================================================

experimental = pd.DataFrame(
    [
        {
            "R3": r3,
            "Experimental_de_percent": de,
        }
        for r3, de in EXPERIMENTAL_DE.items()
    ]
)


# ---------------------------------------------------------------
# RR/RS ratio from %de
# ---------------------------------------------------------------

experimental["RR_RS_ratio"] = np.nan

valid = (
    experimental[
        "Experimental_de_percent"
    ] < 100
)

experimental.loc[
    valid,
    "RR_RS_ratio"
] = (
    100
    + experimental.loc[
        valid,
        "Experimental_de_percent"
    ]
) / (
    100
    - experimental.loc[
        valid,
        "Experimental_de_percent"
    ]
)


# ---------------------------------------------------------------
# Apparent selectivity-derived energy
#
# This is NOT a computed activation free energy.
# ---------------------------------------------------------------

experimental[
    "Apparent_DeltaDeltaG_kcal_mol"
] = np.nan

experimental.loc[
    valid,
    "Apparent_DeltaDeltaG_kcal_mol"
] = (
    -RT
    *
    np.log(
        experimental.loc[
            valid,
            "RR_RS_ratio"
        ]
    )
)


# ================================================================
# 22. MERGE EXPERIMENT WITH RR/RS
# ================================================================

rr_rs_exp = pd.merge(
    rr_rs,
    experimental,
    on="R3",
    how="inner"
)


# Add display name
rr_rs_exp["R3_display"] = (
    rr_rs_exp["R3"]
    .map(R3_DISPLAY)
    .fillna(
        rr_rs_exp["R3"]
    )
)


rr_rs_exp.to_csv(
    TABLE_ROOT
    / "08_RR_RS_PLUS_EXPERIMENTAL_SELECTIVITY.csv",
    index=False
)


# ================================================================
# 23. MERGE EXPERIMENT WITH FLEXIBILITY
# ================================================================

flex_exp = pd.merge(
    flexibility,
    experimental,
    on="R3",
    how="inner"
)

flex_exp["R3_display"] = (
    flex_exp["R3"]
    .map(R3_DISPLAY)
    .fillna(
        flex_exp["R3"]
    )
)

flex_exp.to_csv(
    TABLE_ROOT
    / "09_FLEXIBILITY_PLUS_EXPERIMENTAL_DE.csv",
    index=False
)


# ================================================================
# 24. CORRELATION FUNCTION
# ================================================================

def calculate_correlation(
    data,
    x,
    y,
    label
):

    sub = data[
        [
            x,
            y
        ]
    ].dropna()

    n = len(sub)

    if n < 3:

        return {

            "Analysis": label,

            "N": n,

            "Pearson_r": np.nan,

            "Pearson_p": np.nan,

            "Spearman_rho": np.nan,

            "Spearman_p": np.nan,
        }

    pr, pp = pearsonr(
        sub[x],
        sub[y]
    )

    sr, sp = spearmanr(
        sub[x],
        sub[y]
    )

    return {

        "Analysis": label,

        "N": n,

        "Pearson_r": pr,

        "Pearson_p": pp,

        "Spearman_rho": sr,

        "Spearman_p": sp,
    }


# ================================================================
# 25. CORRELATIONS
# ================================================================

correlation_results = []


# %de vs RR-RS energy difference
correlation_results.append(
    calculate_correlation(
        rr_rs_exp,
        "DeltaDeltaE_RS_minus_RR",
        "Experimental_de_percent",
        "%de vs E(RS)-E(RR)"
    )
)


# %de vs absolute energy difference
correlation_results.append(
    calculate_correlation(
        rr_rs_exp,
        "Abs_DeltaDeltaE",
        "Experimental_de_percent",
        "%de vs |E(RS)-E(RR)|"
    )
)


# %de vs RR lowest-conformer population
correlation_results.append(
    calculate_correlation(
        rr_rs_exp,
        "RR_lowest_population_percent",
        "Experimental_de_percent",
        "%de vs RR lowest-conformer population"
    )
)


# %de vs effective conformer number
correlation_results.append(
    calculate_correlation(
        flex_exp,
        "Effective_N_conformers",
        "Experimental_de_percent",
        "%de vs effective conformational number"
    )
)


# %de vs number within 1 kcal
correlation_results.append(
    calculate_correlation(
        flex_exp,
        "N_within_1.0_kcal",
        "Experimental_de_percent",
        "%de vs N within 1 kcal mol-1"
    )
)


# %de vs number within 2 kcal
correlation_results.append(
    calculate_correlation(
        flex_exp,
        "N_within_2.0_kcal",
        "Experimental_de_percent",
        "%de vs N within 2 kcal mol-1"
    )
)


# %de vs torsional flexibility
correlation_results.append(
    calculate_correlation(
        flex_exp,
        "Torsional_flexibility_deg",
        "Experimental_de_percent",
        "%de vs torsional flexibility"
    )
)


correlations = pd.DataFrame(
    correlation_results
)

correlations.to_csv(
    TABLE_ROOT
    / "10_EXPERIMENTAL_DE_CORRELATIONS.csv",
    index=False
)


# ================================================================
# 26. FIGURE 1
#
# ACTUAL CREST PUCKERING-CLASS POPULATIONS
#
# RR and RS separately
# ================================================================

for config in ["RR", "RS"]:

    sub = class_population[
        class_population[
            "Configuration"
        ] == config
    ].copy()

    if sub.empty:
        continue

    matrix = (
        sub
        .pivot_table(
            index="R3",
            columns="classification",
            values="Total_population_percent",
            aggfunc="sum",
            fill_value=0
        )
    )

    # Keep R3 order according to experimental data
    ordered_r3 = [
        r3
        for r3 in EXPERIMENTAL_DE
        if r3 in matrix.index
    ]

    matrix = matrix.reindex(
        ordered_r3
    )

    fig, ax = plt.subplots(
        figsize=(16, 8)
    )

    bottom = np.zeros(
        len(matrix)
    )

    for pclass in matrix.columns:

        values = (
            matrix[pclass]
            .values
        )

        ax.bar(
            matrix.index,
            values,
            bottom=bottom,
            label=pclass
        )

        bottom += values

    ax.set_xlabel("R3")

    ax.set_ylabel(
        "Population (%)"
    )

    ax.set_title(
        f"Actual CREST Conformer Populations — {config}\n"
        f"298.15 K"
    )

    ax.set_ylim(
        0,
        100
    )

    ax.tick_params(
        axis="x",
        rotation=65
    )

    ax.legend(
        bbox_to_anchor=(1.02, 1),
        loc="upper left"
    )

    ax.grid(
        axis="y",
        alpha=0.25
    )

    plt.tight_layout()

    plt.savefig(
        FIGURE_ROOT
        / f"FIGURE_1_{config}_ACTUAL_CREST_POPULATIONS.png",
        dpi=600,
        bbox_inches="tight"
    )

    plt.show()


# ================================================================
# 27. FIGURE 2
#
# RR vs RS POPULATION DIFFERENCE
# ================================================================

fig, ax = plt.subplots(
    figsize=(15, 7)
)

plot = rr_rs_exp.copy()

x = np.arange(
    len(plot)
)

ax.bar(
    x,
    plot[
        "Population_difference_RR_minus_RS"
    ]
)

ax.axhline(
    0,
    linewidth=1
)

ax.set_xticks(x)

ax.set_xticklabels(
    plot["R3_display"],
    rotation=65,
    ha="right"
)

ax.set_ylabel(
    "Lowest-conformer population: RR − RS (%)"
)

ax.set_xlabel("R3")

ax.set_title(
    "RR vs RS Lowest-Conformer Population Difference"
)

ax.grid(
    axis="y",
    alpha=0.25
)

plt.tight_layout()

plt.savefig(
    FIGURE_ROOT
    / "FIGURE_2_RR_vs_RS_POPULATION_DIFFERENCE.png",
    dpi=600,
    bbox_inches="tight"
)

plt.show()


# ================================================================
# 28. FIGURE 3
#
# RR-RS MINIMUM CONFORMER ENERGY DIFFERENCE
# ================================================================

fig, ax = plt.subplots(
    figsize=(15, 7)
)

x = np.arange(
    len(rr_rs_exp)
)

ax.bar(
    x,
    rr_rs_exp[
        "DeltaDeltaE_RS_minus_RR"
    ]
)

ax.axhline(
    0,
    linewidth=1
)

ax.set_xticks(x)

ax.set_xticklabels(
    rr_rs_exp["R3_display"],
    rotation=65,
    ha="right"
)

ax.set_xlabel("R3")

ax.set_ylabel(
    "E(RS) − E(RR)\n(kcal mol$^{-1}$)"
)

ax.set_title(
    "RR vs RS Lowest Retained-CREST Conformer"
)

ax.grid(
    axis="y",
    alpha=0.25
)

plt.tight_layout()

plt.savefig(
    FIGURE_ROOT
    / "FIGURE_3_RR_vs_RS_ENERGY_DIFFERENCE.png",
    dpi=600,
    bbox_inches="tight"
)

plt.show()


# ================================================================
# 29. FIGURE 4
#
# EXPERIMENTAL %de vs RR-RS ENERGY DIFFERENCE
# ================================================================

plot = rr_rs_exp.dropna(
    subset=[
        "DeltaDeltaE_RS_minus_RR",
        "Experimental_de_percent"
    ]
).copy()

fig, ax = plt.subplots(
    figsize=(9, 7)
)

ax.scatter(
    plot[
        "DeltaDeltaE_RS_minus_RR"
    ],
    plot[
        "Experimental_de_percent"
    ],
    s=80
)

for _, row in plot.iterrows():

    ax.annotate(
        row["R3_display"],
        (
            row[
                "DeltaDeltaE_RS_minus_RR"
            ],
            row[
                "Experimental_de_percent"
            ]
        ),
        xytext=(5, 5),
        textcoords="offset points",
        fontsize=8
    )


if len(plot) >= 2:

    slope, intercept = np.polyfit(
        plot[
            "DeltaDeltaE_RS_minus_RR"
        ],
        plot[
            "Experimental_de_percent"
        ],
        1
    )

    xx = np.linspace(
        plot[
            "DeltaDeltaE_RS_minus_RR"
        ].min(),
        plot[
            "DeltaDeltaE_RS_minus_RR"
        ].max(),
        200
    )

    yy = (
        slope * xx
        + intercept
    )

    ax.plot(
        xx,
        yy
    )


ax.axvline(
    0,
    linewidth=0.8
)

ax.set_xlabel(
    "E(RS) − E(RR) of lowest retained conformer\n"
    "(kcal mol$^{-1}$)"
)

ax.set_ylabel(
    "Experimental %de (RR major)"
)

ax.set_title(
    "Experimental Selectivity vs RR–RS Conformational Energy Difference"
)

ax.grid(
    alpha=0.25
)

plt.tight_layout()

plt.savefig(
    FIGURE_ROOT
    / "FIGURE_4_EXPERIMENTAL_DE_vs_RR_RS_ENERGY.png",
    dpi=600,
    bbox_inches="tight"
)

plt.show()


# ================================================================
# 30. FIGURE 5
#
# EXPERIMENTAL %de vs CONFORMATIONAL FLEXIBILITY
# ================================================================

plot = flex_exp.dropna(
    subset=[
        "Effective_N_conformers",
        "Experimental_de_percent"
    ]
).copy()

fig, ax = plt.subplots(
    figsize=(9, 7)
)

ax.scatter(
    plot[
        "Effective_N_conformers"
    ],
    plot[
        "Experimental_de_percent"
    ],
    s=80
)

for _, row in plot.iterrows():

    ax.annotate(
        row["R3_display"],
        (
            row[
                "Effective_N_conformers"
            ],
            row[
                "Experimental_de_percent"
            ]
        ),
        xytext=(5, 5),
        textcoords="offset points",
        fontsize=8
    )


ax.set_xlabel(
    "Effective number of conformers ($N_{eff}$)"
)

ax.set_ylabel(
    "Experimental %de (RR major)"
)

ax.set_title(
    "Experimental Selectivity vs Conformational Flexibility"
)

ax.grid(
    alpha=0.25
)

plt.tight_layout()

plt.savefig(
    FIGURE_ROOT
    / "FIGURE_5_EXPERIMENTAL_DE_vs_FLEXIBILITY.png",
    dpi=600,
    bbox_inches="tight"
)

plt.show()


# ================================================================
# 31. FIGURE 6
#
# EXPERIMENTAL %de vs RR LOWEST-CONFORMER POPULATION
# ================================================================

plot = rr_rs_exp.dropna(
    subset=[
        "RR_lowest_population_percent",
        "Experimental_de_percent"
    ]
).copy()

fig, ax = plt.subplots(
    figsize=(9, 7)
)

ax.scatter(
    plot[
        "RR_lowest_population_percent"
    ],
    plot[
        "Experimental_de_percent"
    ],
    s=80
)

for _, row in plot.iterrows():

    ax.annotate(
        row["R3_display"],
        (
            row[
                "RR_lowest_population_percent"
            ],
            row[
                "Experimental_de_percent"
            ]
        ),
        xytext=(5, 5),
        textcoords="offset points",
        fontsize=8
    )


ax.set_xlabel(
    "Population of lowest RR conformer (%)"
)

ax.set_ylabel(
    "Experimental %de (RR major)"
)

ax.set_title(
    "Experimental Selectivity vs Lowest-RR-Conformer Population"
)

ax.grid(
    alpha=0.25
)

plt.tight_layout()

plt.savefig(
    FIGURE_ROOT
    / "FIGURE_6_EXPERIMENTAL_DE_vs_RR_POPULATION.png",
    dpi=600,
    bbox_inches="tight"
)

plt.show()


# ================================================================
# 32. FIGURE 7
#
# CONFORMATIONAL FLEXIBILITY MAP
#
# N_eff for RR and RS
# ================================================================

flex_matrix = (
    flexibility
    .pivot(
        index="R3",
        columns="Configuration",
        values="Effective_N_conformers"
    )
)

ordered_r3 = [
    r3
    for r3 in EXPERIMENTAL_DE
    if r3 in flex_matrix.index
]

flex_matrix = flex_matrix.reindex(
    ordered_r3
)

fig, ax = plt.subplots(
    figsize=(15, 7)
)

x = np.arange(
    len(flex_matrix)
)

if "RR" in flex_matrix.columns:

    ax.plot(
        x,
        flex_matrix["RR"],
        marker="o",
        label="RR"
    )

if "RS" in flex_matrix.columns:

    ax.plot(
        x,
        flex_matrix["RS"],
        marker="s",
        label="RS"
    )

ax.set_xticks(x)

ax.set_xticklabels(
    [
        R3_DISPLAY.get(
            r3,
            r3
        )
        for r3 in flex_matrix.index
    ],
    rotation=65,
    ha="right"
)

ax.set_xlabel("R3")

ax.set_ylabel(
    "Effective number of conformers ($N_{eff}$)"
)

ax.set_title(
    "Conformational Flexibility of Actual CREST Ensembles"
)

ax.legend()

ax.grid(
    alpha=0.25
)

plt.tight_layout()

plt.savefig(
    FIGURE_ROOT
    / "FIGURE_7_CONFORMATIONAL_FLEXIBILITY_RR_RS.png",
    dpi=600,
    bbox_inches="tight"
)

plt.show()


# ================================================================
# 33. COPY BEST ACTUAL RETAINED XYZ STRUCTURES
# ================================================================

print("\n================================================")
print("COPYING BEST ACTUAL RETAINED XYZ STRUCTURES")
print("================================================")

best_copy_rows = []


for _, row in lowest.iterrows():

    r3 = row["R3"]

    config = row["Configuration"]

    conformer = int(
        row["Lowest_conformer"]
    )

    # ------------------------------------------------------------
    # Expected CREST folder
    # ------------------------------------------------------------

    mother_folder = (
        CREST_ROOT
        / f"{config}_{r3}_CREST_conformations"
    )

    # Expected filename
    expected_name = (
        f"{config}_{r3}"
        f"_retained_{conformer:03d}.xyz"
    )

    source = (
        mother_folder
        / expected_name
    )

    # ------------------------------------------------------------
    # Search fallback
    # ------------------------------------------------------------

    if not source.exists():

        search_root = (
            CREST_ROOT
        )

        candidates = list(
            search_root.rglob(
                expected_name
            )
        )

        if candidates:

            source = candidates[0]

    # ------------------------------------------------------------
    # Copy
    # ------------------------------------------------------------

    if source.exists():

        destination_name = (
            f"BEST_{config}_{r3}"
            f"_retained_{conformer:03d}.xyz"
        )

        destination = (
            BEST_ROOT
            / destination_name
        )

        shutil.copy2(
            source,
            destination
        )

        status = "Copied"

    else:

        destination = ""

        status = "NOT FOUND"

        print(
            "NOT FOUND:",
            expected_name
        )

    best_copy_rows.append({

        "R3":
            r3,

        "Configuration":
            config,

        "Conformer":
            conformer,

        "Puckering_class":
            row["Puckering_class"],

        "Source":
            str(source),

        "Destination":
            str(destination),

        "Status":
            status,
    })


best_copy = pd.DataFrame(
    best_copy_rows
)

best_copy.to_csv(
    TABLE_ROOT
    / "11_BEST_RETAINED_XYZ_FILES.csv",
    index=False
)


# ================================================================
# 34. MASTER DATASET SAVE
# ================================================================

df.to_csv(
    TABLE_ROOT
    / "00_MASTER_ACTUAL_CREST_DATASET_WITH_POPULATIONS.csv",
    index=False
)


# ================================================================
# 35. MASTER EXCEL WORKBOOK
# ================================================================

excel_file = (
    OUTPUT_ROOT
    / "MASTER_CREST_CONFORMATIONAL_ANALYSIS.xlsx"
)

with pd.ExcelWriter(
    excel_file,
    engine="openpyxl"
) as writer:

    df.to_excel(
        writer,
        sheet_name="All_CRE​ST_Retained",
        index=False
    )

    # Correct sheet name without hidden character
    # The previous sheet is replaced below by using a fresh writer
    pass


# Recreate workbook cleanly with normal ASCII sheet names

with pd.ExcelWriter(
    excel_file,
    engine="openpyxl"
) as writer:

    df.to_excel(
        writer,
        sheet_name="All_CRE​ST_Retained".replace(
            "\u200b",
            ""
        ),
        index=False
    )

    lowest.to_excel(
        writer,
        sheet_name="Lowest_Conformer",
        index=False
    )

    class_population.to_excel(
        writer,
        sheet_name="Class_Populations",
        index=False
    )

    preference_map.to_excel(
        writer,
        sheet_name="Preference_Map"
    )

    preferred.to_excel(
        writer,
        sheet_name="Preferred_Class",
        index=False
    )

    population_table.to_excel(
        writer,
        sheet_name="Conformer_Populations",
        index=False
    )

    flexibility.to_excel(
        writer,
        sheet_name="Flexibility",
        index=False
    )

    rr_rs.to_excel(
        writer,
        sheet_name="RR_vs_RS",
        index=False
    )

    rr_rs_exp.to_excel(
        writer,
        sheet_name="RR_RS_Experimental",
        index=False
    )

    correlations.to_excel(
        writer,
        sheet_name="Correlations",
        index=False
    )

    best_copy.to_excel(
        writer,
        sheet_name="Best_XYZ",
        index=False
    )


# ================================================================
# 36. FINAL SUMMARY TABLE
# ================================================================

summary = pd.merge(
    lowest[
        [
            "R3",
            "Configuration",
            "Puckering_class",
            "Q_A",
            "theta_deg",
            "phi_deg",
            "Population_percent",
            "N_retained"
        ]
    ],
    flexibility[
        [
            "R3",
            "Configuration",
            "N_puckering_classes",
            "N_within_0.5_kcal",
            "N_within_1.0_kcal",
            "N_within_2.0_kcal",
            "Effective_N_conformers",
            "Torsional_flexibility_deg"
        ]
    ],
    on=[
        "R3",
        "Configuration"
    ],
    how="left"
)

summary = pd.merge(
    summary,
    experimental,
    on="R3",
    how="left"
)

summary.to_csv(
    TABLE_ROOT
    / "12_FINAL_R3_CONFORMATIONAL_SUMMARY.csv",
    index=False
)


# ================================================================
# 37. FINAL PRINTED SUMMARY
# ================================================================

print("\n\n")
print("=" * 70)
print("FINAL CREST CONFORMATIONAL ANALYSIS COMPLETE")
print("=" * 70)

print(
    "\nActual retained CREST conformers:",
    len(df)
)

print(
    "R3 systems:",
    df["R3"].nunique()
)

print(
    "RR retained conformers:",
    len(
        df[
            df["Configuration"] == "RR"
        ]
    )
)

print(
    "RS retained conformers:",
    len(
        df[
            df["Configuration"] == "RS"
        ]
    )
)


print("\nOUTPUT DIRECTORY:")
print(OUTPUT_ROOT)


print("\nTABLES:")

for file in sorted(
    TABLE_ROOT.glob("*.csv")
):

    print(
        "  ",
        file.name
    )


print("\nFIGURES:")

for file in sorted(
    FIGURE_ROOT.glob("*.png")
):

    print(
        "  ",
        file.name
    )


print("\nBEST RETAINED XYZ:")
print(BEST_ROOT)


print("\nMASTER EXCEL:")
print(excel_file)


print("\n" + "=" * 70)
print("SCIENTIFIC NOTES")
print("=" * 70)

print(
    "\n1. Population analysis uses ONLY actual retained CREST "
    "conformers from ALL_retained_pucker_results.csv."
)

print(
    "\n2. TARGETED_PUCKERING_PES_CLASSIFIED.csv files are NOT "
    "used for the population calculation."
)

print(
    "\n3. No CREST or xTB calculations were performed."
)

print(
    "\n4. ΔE values are calculated within each individual "
    "R3/configuration."
)

print(
    "\n5. Absolute energies between different R3 molecules "
    "are not compared."
)

print(
    "\n6. RR vs RS energy differences are compared only for "
    "the same R3."
)

print(
    "\n7. Boltzmann populations represent the sampled retained "
    "CREST ensemble and are therefore approximate."
)

print(
    "\n8. Puckering classifications are taken from the existing "
    "CREST-analysis classification column."
)

print(
    "\n9. Selectivity-derived apparent ΔΔG is an energy-equivalent "
    "from experimental product ratio, NOT a computed activation "
    "free energy."
)

print(
    "\n10. The targeted puckering PES remains a separate analysis "
    "and should not be interpreted as a population distribution."
)

print("\n" + "=" * 70)
print("DONE")
print("=" * 70)
```
