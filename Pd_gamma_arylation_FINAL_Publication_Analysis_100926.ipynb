```python
# ============================================================
# Pd γ-ARYLATION
# FINAL PUBLICATION-ORIENTED DESCRIPTOR + YIELD ANALYSIS
#
# INPUT:
#     Optimized XYZ files for the β-C-H-activated intermediate 3
#
# REQUIRED ATOM NUMBERING:
#     Pd atom       = 1
#     Ligand atoms  = 17 to final atom
#
# MORFEUS VERSION:
#     Tested syntax for Morfeus 0.8.0:
#     - BuriedVolume(..., density=0.001)
#     - compute_distal_volume(method="sasa", ...)
#
# OUTPUT:
#     One Excel workbook and publication-quality PNG/PDF plots.
# ============================================================


# ============================================================
# 1. SAFE PACKAGE INSTALLATION
#
# Do NOT install unpinned numpy, pandas, scipy, matplotlib,
# or seaborn in Google Colab.
# ============================================================

!pip -q install --no-deps morfeus-ml
!pip -q install openpyxl statsmodels


# ============================================================
# 2. IMPORTS
# ============================================================

import os
import re
import glob
import shutil
import subprocess
import warnings
import inspect

import numpy as np
import pandas as pd

import matplotlib.pyplot as plt
import seaborn as sns

from scipy.stats import pearsonr, spearmanr
from statsmodels.stats.multitest import multipletests

import morfeus

from morfeus import (
    SASA,
    SolidAngle,
    BuriedVolume,
    Sterimol,
    BiteAngle,
    XTB
)

warnings.filterwarnings("default")


# ============================================================
# 3. GOOGLE DRIVE
# ============================================================

from google.colab import drive

drive.mount("/content/drive")


# ============================================================
# 4. USER SETTINGS
# ============================================================

XYZ_FOLDER = (
    "/content/drive/MyDrive/"
    "Suman_files"
)

OUTPUT_FOLDER = (
    "/content/drive/MyDrive/"
    "Pd_gamma_arylation_FINAL_RESULTS_100926"
)

os.makedirs(
    OUTPUT_FOLDER,
    exist_ok=True
)

PLOT_FOLDER = os.path.join(
    OUTPUT_FOLDER,
    "PLOTS"
)

os.makedirs(
    PLOT_FOLDER,
    exist_ok=True
)

FINAL_EXCEL = os.path.join(
    OUTPUT_FOLDER,
    "Pd_gamma_arylation_FINAL_Publication_Analysis_100926.xlsx"
)


# ============================================================
# 5. CHEMICAL SETTINGS
#
# Confirm these atom indices are correct for every XYZ file.
# ============================================================

PD_ATOM = 1

LIGAND_START = 17

PD_DONOR_MIN = 2.00
PD_DONOR_MAX = 2.40

DONOR_ELEMENTS = {
    "N",
    "P",
    "S",
    "O"
}


# ============================================================
# 6. MORFEUS 0.8.0 SETTINGS
# ============================================================

SASA_PROBE_RADIUS = 1.4

# Nonredundant Pd-centered buried-volume radii.
#
# 3.0 Å = immediate coordination pocket.
# 4.0 Å = near-Pd steric region.
# 5.5 Å = broader ligand envelope.
VBUR_RADII = [
    3.0,
    4.0,
    5.5
]

# Defines the metal-centered sphere used to separate
# proximal and distal ligand volume.
DISTAL_VOLUME_RADIUS = 3.5

# Morfeus 0.8.0 uses density, NOT n_points.
# Smaller density = higher numerical precision and slower run.
VBUR_DENSITY = 0.001

# Used by compute_distal_volume(method="sasa").
DISTAL_SASA_DENSITY = 0.01

VBUR_RADII_TYPE = "bondi"

SOLID_ANGLE_RADII_TYPE = "bondi"

STERIMOL_RADII_TYPE = "crc"


# ============================================================
# 7. xTB SETTINGS
#
# xTB descriptors below are isolated-ligand quantities:
# ligand atoms only; Pd/substrate/acetate/Na are excluded.
# ============================================================

XTB_METHOD = 2

XTB_CHARGE = 0

XTB_N_UNPAIRED = 0

XTB_PROCESSES = 2


# ============================================================
# 8. STATISTICAL / PLOT SETTINGS
# ============================================================

HIGH_CORR_THRESHOLD = 0.80

REDUNDANCY_SPEARMAN_THRESHOLD = 0.90

MATRIX_ANNOTATION_MAX = 30

MATRIX_DPI = 600


# ============================================================
# 9. YIELD DATA
# ============================================================

YIELD_DATA = {

    "L7": 0,
    "L8": 0,
    "L9": 0,
    "L10": 0,
    "L11": 0,
    "L12": 0,
    "L13": 0,
    "L14": 0,
    "L15": 0,

    "L16": 12,
    "L17": 53,
    "L18": 46,
    "L19": 67,
    "L20": 40,
    "L21": 31,

    "L22": 0,
    "L23": 0,
    "L24": 0,
    "L25": 0
}


# ============================================================
# 10. VERSION / API CHECK
# ============================================================

print("=" * 90)
print("SOFTWARE CHECK")
print("=" * 90)

print("Morfeus version:")

try:
    print(morfeus.__version__)
except Exception:
    print("Version unavailable")

print("\nBuriedVolume signature:")
print(
    inspect.signature(
        BuriedVolume
    )
)

print("\ncompute_distal_volume signature:")
print(
    inspect.signature(
        BuriedVolume.compute_distal_volume
    )
)


# ============================================================
# 11. xTB CHECK / INSTALL
# ============================================================

print("\n" + "=" * 90)
print("xTB CHECK")
print("=" * 90)

xtb_path = shutil.which("xtb")

if xtb_path is None:

    print("xTB executable not found. Installing xTB 6.7.1...")

    xtb_archive = "/content/xtb.tar.xz"

    subprocess.run(
        [
            "wget",
            "-q",
            "--show-progress",
            "https://github.com/grimme-lab/xtb/releases/download/v6.7.1/xtb-6.7.1-linux-x86_64.tar.xz",
            "-O",
            xtb_archive
        ],
        check=True
    )

    subprocess.run(
        [
            "tar",
            "-xf",
            xtb_archive,
            "-C",
            "/content"
        ],
        check=True
    )

    xtb_dirs = sorted([
        directory
        for directory in os.listdir("/content")
        if (
            directory.startswith("xtb-")
            and os.path.isdir(
                os.path.join(
                    "/content",
                    directory
                )
            )
        )
    ])

    if not xtb_dirs:

        raise RuntimeError(
            "xTB directory not found after extraction."
        )

    xtb_dir = os.path.join(
        "/content",
        xtb_dirs[-1]
    )

    xtb_bin = os.path.join(
        xtb_dir,
        "bin"
    )

    xtb_executable = os.path.join(
        xtb_bin,
        "xtb"
    )

    if not os.path.exists(
        xtb_executable
    ):

        raise RuntimeError(
            f"xTB executable missing: {xtb_executable}"
        )

    os.environ["PATH"] = (
        xtb_bin
        + ":"
        + os.environ["PATH"]
    )

    os.environ["XTBHOME"] = xtb_dir


xtb_path = shutil.which("xtb")

if xtb_path is None:

    raise RuntimeError(
        "xTB is unavailable."
    )

print("xTB executable:")
print(xtb_path)

xtb_version = subprocess.run(
    ["xtb", "--version"],
    capture_output=True,
    text=True
)

print(xtb_version.stdout)


# ============================================================
# 12. HELPER FUNCTIONS
# ============================================================

def finite(value):

    try:

        value = float(value)

        if np.isfinite(value):
            return value

    except Exception:
        pass

    return np.nan


def distance(a, b):

    return float(
        np.linalg.norm(
            np.asarray(a)
            -
            np.asarray(b)
        )
    )


def clean_ligand_name(filename):

    base = os.path.basename(filename)

    base = re.sub(
        r"\.xyz$",
        "",
        base,
        flags=re.IGNORECASE
    )

    base = re.sub(
        r"[_-](new|opt|optimized)$",
        "",
        base,
        flags=re.IGNORECASE
    )

    match = re.search(
        r"L\d+",
        base,
        flags=re.IGNORECASE
    )

    if match:
        return match.group(0).upper()

    return base


def safe_filename(text):

    return re.sub(
        r"[^A-Za-z0-9_.-]+",
        "_",
        str(text)
    )


def ligand_sort_number(value):

    match = re.search(
        r"L(\d+)",
        str(value),
        flags=re.IGNORECASE
    )

    if match:
        return int(match.group(1))

    return 9999


# ============================================================
# 13. XYZ READER
# ============================================================

def read_xyz_file(filepath):

    with open(
        filepath,
        "r",
        encoding="utf-8"
    ) as f:

        lines = [
            line.strip()
            for line in f
            if line.strip()
        ]

    if len(lines) < 3:

        raise ValueError(
            "XYZ file too short."
        )

    n_atoms = int(
        lines[0].split()[0]
    )

    atom_lines = lines[2:]

    if len(atom_lines) < n_atoms:

        raise ValueError(
            "XYZ has fewer coordinate lines than declared atoms."
        )

    elements = []
    coordinates = []

    for line in atom_lines[:n_atoms]:

        parts = line.split()

        if len(parts) < 4:

            raise ValueError(
                f"Invalid coordinate line: {line}"
            )

        elements.append(
            parts[0]
        )

        coordinates.append([
            float(parts[1]),
            float(parts[2]),
            float(parts[3])
        ])

    return (
        elements,
        np.asarray(
            coordinates,
            dtype=float
        )
    )


# ============================================================
# 14. COVALENT CONNECTIVITY HELPERS
# ============================================================

COVALENT_RADII = {

    "H": 0.31,
    "C": 0.76,
    "N": 0.71,
    "O": 0.66,
    "F": 0.57,

    "P": 1.07,
    "S": 1.05,

    "Cl": 1.02,
    "Br": 1.20,
    "I": 1.39,

    "Pd": 1.39
}


def covalent_neighbors(
    elements,
    coordinates,
    atom_id,
    allowed_indices=None,
    scale=1.25
):

    if allowed_indices is None:

        allowed_indices = range(
            1,
            len(elements) + 1
        )

    element_i = elements[
        atom_id - 1
    ]

    if element_i not in COVALENT_RADII:
        return []

    radius_i = COVALENT_RADII[
        element_i
    ]

    neighbors = []

    for atom_j in allowed_indices:

        if atom_j == atom_id:
            continue

        element_j = elements[
            atom_j - 1
        ]

        if element_j not in COVALENT_RADII:
            continue

        radius_j = COVALENT_RADII[
            element_j
        ]

        d = distance(
            coordinates[
                atom_id - 1
            ],
            coordinates[
                atom_j - 1
            ]
        )

        cutoff = scale * (
            radius_i + radius_j
        )

        if d <= cutoff:

            neighbors.append(
                (atom_j, d)
            )

    return sorted(
        neighbors,
        key=lambda item: item[1]
    )


def build_ligand_graph(
    elements,
    coordinates
):

    ligand_indices = list(
        range(
            LIGAND_START,
            len(elements) + 1
        )
    )

    graph = {
        atom: set()
        for atom in ligand_indices
    }

    for atom_i in ligand_indices:

        neighbors = covalent_neighbors(
            elements,
            coordinates,
            atom_i,
            allowed_indices=ligand_indices
        )

        for atom_j, _ in neighbors:

            graph[atom_i].add(atom_j)
            graph[atom_j].add(atom_i)

    return graph


def shortest_ligand_path(
    graph,
    start,
    end
):

    if start is None or end is None:
        return None

    queue = [
        (
            start,
            [start]
        )
    ]

    visited = {
        start
    }

    while queue:

        current, path = queue.pop(0)

        if current == end:
            return path

        for neighbor in sorted(
            graph.get(current, [])
        ):

            if neighbor not in visited:

                visited.add(neighbor)

                queue.append(
                    (
                        neighbor,
                        path + [neighbor]
                    )
                )

    return None


# ============================================================
# 15. Pd DONOR IDENTIFICATION
# ============================================================

def find_pd_donors(
    elements,
    coordinates
):

    pd_coordinates = coordinates[
        PD_ATOM - 1
    ]

    candidates = []

    for atom_id in range(
        LIGAND_START,
        len(elements) + 1
    ):

        element = elements[
            atom_id - 1
        ]

        if element not in DONOR_ELEMENTS:
            continue

        d = distance(
            pd_coordinates,
            coordinates[
                atom_id - 1
            ]
        )

        candidates.append({

            "Atom": atom_id,

            "Element": element,

            "Pd_Distance_A": d
        })

    qualifying = [

        item
        for item in candidates

        if (
            PD_DONOR_MIN
            <=
            item["Pd_Distance_A"]
            <=
            PD_DONOR_MAX
        )
    ]

    qualifying = sorted(
        qualifying,
        key=lambda item: item[
            "Pd_Distance_A"
        ]
    )

    return candidates, qualifying


# ============================================================
# 16. PALLADACYCLE DESCRIPTORS
# ============================================================

def determine_palladacycle(
    ligand_name,
    elements,
    coordinates,
    donor1,
    donor2
):

    result = {

        "Palladacycle_Type":
            None,

        "Palladacycle_Ring_Size":
            np.nan,

        "Palladacycle_Ligand_Path":
            None,

        "Palladacycle_Path_Bond_Mean_A":
            np.nan,

        "Palladacycle_Path_Bond_SD_A":
            np.nan,

        "Palladacycle_Path_Bond_CV":
            np.nan,

        "Expected_Palladacycle_Type":
            None,

        "Palladacycle_Type_Check":
            None
    }


    ligand_match = re.search(
        r"L(\d+)",
        ligand_name,
        flags=re.IGNORECASE
    )

    if ligand_match:

        ligand_number = int(
            ligand_match.group(1)
        )

        if 7 <= ligand_number <= 15:

            result[
                "Expected_Palladacycle_Type"
            ] = "4-membered"

        elif 16 <= ligand_number <= 25:

            result[
                "Expected_Palladacycle_Type"
            ] = "5-membered"


    if donor1 is None or donor2 is None:

        result[
            "Palladacycle_Type"
        ] = "NOT_AVAILABLE"

        return result


    graph = build_ligand_graph(
        elements,
        coordinates
    )

    path = shortest_ligand_path(
        graph,
        donor1,
        donor2
    )

    if path is None:

        result[
            "Palladacycle_Type"
        ] = "NOT_FOUND"

        return result


    result[
        "Palladacycle_Ligand_Path"
    ] = "-".join(
        map(str, path)
    )

    path_count = len(path)

    if path_count == 4:

        result[
            "Palladacycle_Ring_Size"
        ] = 4

        result[
            "Palladacycle_Type"
        ] = "4-membered"

    elif path_count == 5:

        result[
            "Palladacycle_Ring_Size"
        ] = 5

        result[
            "Palladacycle_Type"
        ] = "5-membered"

    else:

        result[
            "Palladacycle_Type"
        ] = f"OTHER_{path_count}"


    bond_lengths = []

    for i in range(
        len(path) - 1
    ):

        atom_a = path[i]
        atom_b = path[i + 1]

        bond_lengths.append(
            distance(
                coordinates[
                    atom_a - 1
                ],
                coordinates[
                    atom_b - 1
                ]
            )
        )


    if bond_lengths:

        mean_bond = np.mean(
            bond_lengths
        )

        sd_bond = np.std(
            bond_lengths,
            ddof=0
        )

        result[
            "Palladacycle_Path_Bond_Mean_A"
        ] = finite(
            mean_bond
        )

        result[
            "Palladacycle_Path_Bond_SD_A"
        ] = finite(
            sd_bond
        )

        result[
            "Palladacycle_Path_Bond_CV"
        ] = finite(
            sd_bond / mean_bond
        ) if mean_bond > 0 else np.nan


    expected = result[
        "Expected_Palladacycle_Type"
    ]

    actual = result[
        "Palladacycle_Type"
    ]

    if expected is None:

        result[
            "Palladacycle_Type_Check"
        ] = "NO_EXPECTATION"

    elif expected == actual:

        result[
            "Palladacycle_Type_Check"
        ] = "PASS"

    else:

        result[
            "Palladacycle_Type_Check"
        ] = "CHECK"

    return result


# ============================================================
# 17. STERIC DESCRIPTORS
#
# Includes:
# - SASA
# - Solid angle
# - %Vbur at 3.0, 4.0, and 5.5 Å
# - True distal volume outside a 3.5 Å Pd-centered sphere
# - Pd-oriented Sterimol
# - Ligand P_int if available
# ============================================================

def calculate_steric(
    elements,
    coordinates,
    donor1
):

    result = {}

    ligand_indices = list(
        range(
            LIGAND_START,
            len(elements) + 1
        )
    )

    ligand_elements = [
        elements[
            atom_id - 1
        ]
        for atom_id in ligand_indices
    ]

    ligand_coordinates = coordinates[
        np.array(ligand_indices) - 1
    ]


    # --------------------------------------------------------
    # Maximum ligand extension
    # --------------------------------------------------------

    maximum_extension = 0.0

    for i in range(
        len(ligand_coordinates)
    ):

        for j in range(
            i + 1,
            len(ligand_coordinates)
        ):

            maximum_extension = max(
                maximum_extension,
                distance(
                    ligand_coordinates[i],
                    ligand_coordinates[j]
                )
            )

    result[
        "Ligand_Maximum_Extension_A"
    ] = finite(
        maximum_extension
    )


    # --------------------------------------------------------
    # SASA
    # --------------------------------------------------------

    try:

        sasa = SASA(
            ligand_elements,
            ligand_coordinates,
            probe_radius=SASA_PROBE_RADIUS
        )

        result[
            "Ligand_SASA_A2"
        ] = finite(
            sasa.area
        )

        result[
            "Ligand_SASA_Volume_A3"
        ] = finite(
            sasa.volume
        )

    except Exception as error:

        result[
            "SASA_Warning"
        ] = str(error)

        result[
            "Ligand_SASA_A2"
        ] = np.nan

        result[
            "Ligand_SASA_Volume_A3"
        ] = np.nan


    # --------------------------------------------------------
    # Solid angle
    # --------------------------------------------------------

    try:

        reduced_elements = (
            ["Pd"]
            +
            ligand_elements
        )

        reduced_coordinates = np.vstack([
            coordinates[
                PD_ATOM - 1
            ],
            ligand_coordinates
        ])

        solid_angle = SolidAngle(
            reduced_elements,
            reduced_coordinates,
            metal_index=1,
            radii_type=SOLID_ANGLE_RADII_TYPE
        )

        result[
            "Ligand_Solid_Angle_sr"
        ] = finite(
            solid_angle.solid_angle
        )

        result[
            "Ligand_Solid_Cone_Angle_deg"
        ] = finite(
            solid_angle.cone_angle
        )

        result[
            "Ligand_Solid_G_percent"
        ] = finite(
            solid_angle.G
        )

    except Exception as error:

        result[
            "SolidAngle_Warning"
        ] = str(error)

        result[
            "Ligand_Solid_Angle_sr"
        ] = np.nan

        result[
            "Ligand_Solid_Cone_Angle_deg"
        ] = np.nan

        result[
            "Ligand_Solid_G_percent"
        ] = np.nan


    # --------------------------------------------------------
    # Exclude all nonligand atoms except Pd.
    # --------------------------------------------------------

    excluded_atoms = [

        atom_id
        for atom_id in range(
            1,
            len(elements) + 1
        )

        if (
            atom_id != PD_ATOM
            and
            atom_id not in ligand_indices
        )
    ]


    # --------------------------------------------------------
    # Pd-centered %Vbur
    #
    # Morfeus 0.8.0 compatible:
    # uses density rather than n_points.
    # --------------------------------------------------------

    for radius in VBUR_RADII:

        descriptor_name = (
            f"Ligand_%Vbur_{radius:.1f}A"
        )

        try:

            buried_volume = BuriedVolume(
                elements,
                coordinates,
                metal_index=PD_ATOM,
                excluded_atoms=excluded_atoms,
                radius=radius,
                include_hs=False,
                radii_type=VBUR_RADII_TYPE,
                density=VBUR_DENSITY
            )

            result[
                descriptor_name
            ] = finite(
                buried_volume.fraction_buried_volume
                * 100.0
            )

        except Exception as error:

            result[
                descriptor_name
            ] = np.nan

            result[
                f"{descriptor_name}_Warning"
            ] = str(error)


    # --------------------------------------------------------
    # TRUE DISTAL VOLUME
    #
    # Volume outside a Pd-centered sphere of radius 3.5 Å.
    # Verified Morfeus 0.8.0 API:
    #
    # compute_distal_volume(
    #     method="sasa",
    #     octants=False,
    #     sasa_density=0.01
    # )
    # --------------------------------------------------------

    distal_volume_name = (
        f"Ligand_Distal_Volume_Outside_"
        f"{DISTAL_VOLUME_RADIUS:.1f}A_A3"
    )

    distal_fraction_name = (
        f"Ligand_Distal_Volume_Fraction_Outside_"
        f"{DISTAL_VOLUME_RADIUS:.1f}A"
    )

    try:

        distal_bv = BuriedVolume(
            elements,
            coordinates,
            metal_index=PD_ATOM,
            excluded_atoms=excluded_atoms,
            radius=DISTAL_VOLUME_RADIUS,
            include_hs=False,
            radii_type=VBUR_RADII_TYPE,
            density=VBUR_DENSITY
        )

        distal_bv.compute_distal_volume(
            method="sasa",
            octants=False,
            sasa_density=DISTAL_SASA_DENSITY
        )

        distal_volume = finite(
            distal_bv.distal_volume
        )

        molecular_volume = finite(
            distal_bv.molecular_volume
        )

        result[
            distal_volume_name
        ] = distal_volume

        result[
            "Ligand_Molecular_Volume_A3"
        ] = molecular_volume

        result[
            distal_fraction_name
        ] = finite(
            distal_volume
            /
            molecular_volume
        ) if (
            pd.notna(distal_volume)
            and
            pd.notna(molecular_volume)
            and
            molecular_volume > 0
        ) else np.nan

    except Exception as error:

        result[
            "Distal_Volume_Warning"
        ] = str(error)

        result[
            distal_volume_name
        ] = np.nan

        result[
            "Ligand_Molecular_Volume_A3"
        ] = np.nan

        result[
            distal_fraction_name
        ] = np.nan


    # --------------------------------------------------------
    # Pd-oriented Sterimol
    # --------------------------------------------------------

    if donor1 is not None:

        try:

            reduced_elements = (
                ["Pd"]
                +
                ligand_elements
            )

            reduced_coordinates = np.vstack([
                coordinates[
                    PD_ATOM - 1
                ],
                ligand_coordinates
            ])

            original_to_reduced = {
                original_atom: (
                    reduced_index + 2
                )
                for reduced_index, original_atom
                in enumerate(
                    ligand_indices
                )
            }

            attached_index = original_to_reduced[
                donor1
            ]

            sterimol = Sterimol(
                reduced_elements,
                reduced_coordinates,
                dummy_index=1,
                attached_index=attached_index,
                radii_type=STERIMOL_RADII_TYPE
            )

            result[
                "Pd_Oriented_Sterimol_B1_A"
            ] = finite(
                sterimol.B_1_value
            )

            result[
                "Pd_Oriented_Sterimol_B5_A"
            ] = finite(
                sterimol.B_5_value
            )

            result[
                "Pd_Oriented_Sterimol_L_A"
            ] = finite(
                sterimol.L_value
            )

        except Exception as error:

            result[
                "Sterimol_Warning"
            ] = str(error)

            result[
                "Pd_Oriented_Sterimol_B1_A"
            ] = np.nan

            result[
                "Pd_Oriented_Sterimol_B5_A"
            ] = np.nan

            result[
                "Pd_Oriented_Sterimol_L_A"
            ] = np.nan


    # --------------------------------------------------------
    # Ligand dispersion potential, P_int
    #
    # Kept as optional: if the installed Morfeus build does not
    # support Dispersion/P_int, it is excluded automatically.
    # --------------------------------------------------------

    try:

        from morfeus import Dispersion

        dispersion = Dispersion(
            ligand_elements,
            ligand_coordinates
        )

        p_int = finite(
            dispersion.p_int
        )

        result[
            "Ligand_Pint"
        ] = p_int

        sasa_area = result.get(
            "Ligand_SASA_A2",
            np.nan
        )

        result[
            "Ligand_Pint_per_SASA"
        ] = finite(
            p_int / sasa_area
        ) if (
            pd.notna(p_int)
            and
            pd.notna(sasa_area)
            and
            sasa_area > 0
        ) else np.nan

    except Exception as error:

        result[
            "Dispersion_Warning"
        ] = str(error)

        result[
            "Ligand_Pint"
        ] = np.nan

        result[
            "Ligand_Pint_per_SASA"
        ] = np.nan

    return result


# ============================================================
# 18. COORDINATION DESCRIPTORS
# ============================================================

def calculate_coordination(
    elements,
    coordinates,
    donor1,
    donor2
):

    result = {

        "Pd_Donor1_Distance_A":
            np.nan,

        "Pd_Donor2_Distance_A":
            np.nan,

        "Average_Pd_Donor_Distance_A":
            np.nan,

        "Delta_Pd_Donor_Distance_A":
            np.nan,

        "Pd_Donor_Distance_Ratio":
            np.nan,

        "Donor1_Donor2_Distance_A":
            np.nan,

        "Bite_Angle_deg":
            np.nan
    }

    if donor1 is None:
        return result


    pd_coordinate = coordinates[
        PD_ATOM - 1
    ]

    donor1_distance = distance(
        pd_coordinate,
        coordinates[
            donor1 - 1
        ]
    )

    result[
        "Pd_Donor1_Distance_A"
    ] = finite(
        donor1_distance
    )

    if donor2 is None:
        return result


    donor2_distance = distance(
        pd_coordinate,
        coordinates[
            donor2 - 1
        ]
    )

    result[
        "Pd_Donor2_Distance_A"
    ] = finite(
        donor2_distance
    )

    result[
        "Average_Pd_Donor_Distance_A"
    ] = finite(
        (
            donor1_distance
            +
            donor2_distance
        )
        /
        2.0
    )

    result[
        "Delta_Pd_Donor_Distance_A"
    ] = finite(
        abs(
            donor1_distance
            -
            donor2_distance
        )
    )

    result[
        "Pd_Donor_Distance_Ratio"
    ] = finite(
        max(
            donor1_distance,
            donor2_distance
        )
        /
        min(
            donor1_distance,
            donor2_distance
        )
    ) if min(
        donor1_distance,
        donor2_distance
    ) > 0 else np.nan

    result[
        "Donor1_Donor2_Distance_A"
    ] = finite(
        distance(
            coordinates[
                donor1 - 1
            ],
            coordinates[
                donor2 - 1
            ]
        )
    )

    try:

        bite = BiteAngle(
            coordinates,
            PD_ATOM,
            donor1,
            donor2
        )

        result[
            "Bite_Angle_deg"
        ] = finite(
            bite.angle
        )

    except Exception as error:

        result[
            "Bite_Angle_Warning"
        ] = str(error)

    return result


# ============================================================
# 19. ISOLATED-LIGAND xTB DESCRIPTORS
#
# Limited to a compact, interpretable set. This avoids many
# redundant global electronic descriptors.
# ============================================================

def calculate_xtb(
    elements,
    coordinates,
    donor1,
    donor2
):

    result = {}

    ligand_indices = list(
        range(
            LIGAND_START,
            len(elements) + 1
        )
    )

    ligand_elements = [
        elements[
            atom_id - 1
        ]
        for atom_id in ligand_indices
    ]

    ligand_coordinates = coordinates[
        np.array(ligand_indices) - 1
    ]

    try:

        xtb = XTB(
            ligand_elements,
            ligand_coordinates,
            method=XTB_METHOD,
            charge=XTB_CHARGE,
            n_unpaired=XTB_N_UNPAIRED,
            n_processes=XTB_PROCESSES
        )

        charges = xtb.get_charges(
            model="Mulliken"
        )

        result[
            "XTB_IsolatedLigand_HOMO_eV"
        ] = finite(
            xtb.get_homo(
                unit="eV"
            )
        )

        result[
            "XTB_IsolatedLigand_LUMO_eV"
        ] = finite(
            xtb.get_lumo(
                unit="eV"
            )
        )

        result[
            "XTB_IsolatedLigand_HOMO_LUMO_Gap_eV"
        ] = finite(
            xtb.get_homo_lumo_gap()
        )

        result[
            "XTB_IsolatedLigand_Dipole_Debye"
        ] = finite(
            xtb.get_dipole_moment(
                unit="debye"
            )
        )

        original_to_xtb_index = {

            original_atom:
                ligand_index

            for ligand_index, original_atom
            in enumerate(
                ligand_indices
            )
        }

        donor1_charge = np.nan
        donor2_charge = np.nan

        if donor1 is not None:

            donor1_charge = finite(
                charges[
                    original_to_xtb_index[
                        donor1
                    ]
                ]
            )

        if donor2 is not None:

            donor2_charge = finite(
                charges[
                    original_to_xtb_index[
                        donor2
                    ]
                ]
            )

        result[
            "XTB_IsolatedLigand_Donor1_Mulliken_Charge"
        ] = donor1_charge

        result[
            "XTB_IsolatedLigand_Donor2_Mulliken_Charge"
        ] = donor2_charge

        result[
            "XTB_IsolatedLigand_Average_Donor_Charge"
        ] = finite(
            (
                donor1_charge
                +
                donor2_charge
            )
            /
            2.0
        ) if (
            pd.notna(donor1_charge)
            and
            pd.notna(donor2_charge)
        ) else np.nan

    except Exception as error:

        result[
            "XTB_Warning"
        ] = str(error)

    return result


# ============================================================
# 20. PROCESS ONE XYZ FILE
# ============================================================

def process_xyz(
    filepath
):

    ligand_name = clean_ligand_name(
        filepath
    )

    elements, coordinates = read_xyz_file(
        filepath
    )

    if PD_ATOM > len(elements):

        raise ValueError(
            f"{ligand_name}: Pd atom index "
            f"{PD_ATOM} is outside the XYZ file."
        )

    if LIGAND_START > len(elements):

        raise ValueError(
            f"{ligand_name}: ligand start atom "
            f"{LIGAND_START} is outside the XYZ file."
        )


    row = {

        "Ligand":
            ligand_name,

        "XYZ_File":
            os.path.basename(filepath),

        "Yield":
            YIELD_DATA.get(
                ligand_name,
                np.nan
            ),

        "N_Atoms":
            len(elements),

        "Pd_Atom":
            PD_ATOM,

        "Ligand_Start_Atom":
            LIGAND_START
    }


    candidates, qualifying = find_pd_donors(
        elements,
        coordinates
    )

    row[
        "Number_of_Donor_Candidates"
    ] = len(
        candidates
    )

    row[
        "Number_of_Qualifying_Pd_Donors"
    ] = len(
        qualifying
    )


    donor1 = (
        qualifying[0]["Atom"]
        if len(qualifying) >= 1
        else None
    )

    donor2 = (
        qualifying[1]["Atom"]
        if len(qualifying) >= 2
        else None
    )


    row[
        "Donor1_Atom"
    ] = donor1

    row[
        "Donor2_Atom"
    ] = donor2

    row[
        "Donor1_Element"
    ] = (
        qualifying[0]["Element"]
        if len(qualifying) >= 1
        else None
    )

    row[
        "Donor2_Element"
    ] = (
        qualifying[1]["Element"]
        if len(qualifying) >= 2
        else None
    )


    if len(qualifying) == 2:

        row[
            "Donor_Selection_Check"
        ] = "PASS"

    elif len(qualifying) > 2:

        row[
            "Donor_Selection_Check"
        ] = "CHECK_MULTIPLE_QUALIFYING_DONORS"

    elif len(qualifying) == 1:

        row[
            "Donor_Selection_Check"
        ] = "ONLY_ONE_QUALIFYING_DONOR"

    else:

        row[
            "Donor_Selection_Check"
        ] = "NO_QUALIFYING_DONORS"


    row.update(
        determine_palladacycle(
            ligand_name,
            elements,
            coordinates,
            donor1,
            donor2
        )
    )

    row.update(
        calculate_steric(
            elements,
            coordinates,
            donor1
        )
    )

    row.update(
        calculate_coordination(
            elements,
            coordinates,
            donor1,
            donor2
        )
    )

    row.update(
        calculate_xtb(
            elements,
            coordinates,
            donor1,
            donor2
        )
    )


    warning_list = []

    if row[
        "Donor_Selection_Check"
    ] != "PASS":

        warning_list.append(
            row[
                "Donor_Selection_Check"
            ]
        )

    if row.get(
        "Palladacycle_Type_Check",
        None
    ) == "CHECK":

        warning_list.append(
            "Palladacycle type mismatch."
        )

    if pd.isna(
        row[
            "Yield"
        ]
    ):

        warning_list.append(
            "Yield unavailable."
        )

    row[
        "Warnings"
    ] = " || ".join(
        warning_list
    )

    row[
        "Overall_Status"
    ] = (
        "PASS"
        if len(warning_list) == 0
        else "WARNING"
    )

    return row


# ============================================================
# 21. FIND AND PROCESS XYZ FILES
# ============================================================

xyz_files = sorted(
    glob.glob(
        os.path.join(
            XYZ_FOLDER,
            "*.xyz"
        )
    )
)

if len(xyz_files) == 0:

    raise FileNotFoundError(
        f"No .xyz files found in: {XYZ_FOLDER}"
    )


print("\n" + "=" * 90)
print("PROCESSING XYZ FILES")
print("=" * 90)

all_rows = []

for number, filepath in enumerate(
    xyz_files,
    start=1
):

    print(
        f"[{number}/{len(xyz_files)}] "
        f"{os.path.basename(filepath)}"
    )

    try:

        row = process_xyz(
            filepath
        )

        all_rows.append(
            row
        )

        print(
            "  Ligand:",
            row["Ligand"]
        )

        print(
            "  Yield:",
            row["Yield"]
        )

        print(
            "  Donors:",
            row["Donor1_Atom"],
            row["Donor2_Atom"]
        )

    except Exception as error:

        ligand_name = clean_ligand_name(
            filepath
        )

        print(
            "  ERROR:",
            error
        )

        all_rows.append({

            "Ligand":
                ligand_name,

            "XYZ_File":
                os.path.basename(filepath),

            "Yield":
                YIELD_DATA.get(
                    ligand_name,
                    np.nan
                ),

            "Overall_Status":
                "ERROR",

            "Warnings":
                str(error)
        })


master_df = pd.DataFrame(
    all_rows
)

master_df["_Sort"] = master_df[
    "Ligand"
].apply(
    ligand_sort_number
)

master_df = (
    master_df
    .sort_values("_Sort")
    .drop(
        columns="_Sort"
    )
    .reset_index(
        drop=True
    )
)


# ============================================================
# 22. VALIDATION
# ============================================================

if master_df[
    "Yield"
].isna().any():

    missing_yield_ligands = master_df.loc[
        master_df[
            "Yield"
        ].isna(),
        "Ligand"
    ].tolist()

    raise ValueError(
        "Missing yield data for: "
        + ", ".join(
            missing_yield_ligands
        )
    )


validation_columns = [

    "Ligand",

    "XYZ_File",

    "Yield",

    "N_Atoms",

    "Donor1_Atom",

    "Donor1_Element",

    "Donor2_Atom",

    "Donor2_Element",

    "Number_of_Qualifying_Pd_Donors",

    "Donor_Selection_Check",

    "Palladacycle_Type",

    "Expected_Palladacycle_Type",

    "Palladacycle_Type_Check",

    "Palladacycle_Ligand_Path",

    "Overall_Status",

    "Warnings"
]

validation_columns = [

    column
    for column in validation_columns
    if column in master_df.columns
]

validation_df = master_df[
    validation_columns
].copy()


# ============================================================
# 23. COMPLETE DESCRIPTORS ONLY
#
# Any descriptor with even one missing value is excluded from:
# - ALL_VALID_DESCRIPTORS
# - correlation analysis
# ============================================================

numeric_df = master_df.select_dtypes(
    include=[np.number]
).copy()


NON_DESCRIPTOR_COLUMNS = {

    "Pd_Atom",

    "Ligand_Start_Atom",

    "N_Atoms",

    "Donor1_Atom",

    "Donor2_Atom",

    "Number_of_Donor_Candidates",

    "Number_of_Qualifying_Pd_Donors"
}


numeric_descriptor_df = numeric_df.drop(
    columns=[
        column
        for column in NON_DESCRIPTOR_COLUMNS
        if column in numeric_df.columns
    ],
    errors="ignore"
)


missing_columns = []

constant_columns = []

for descriptor in numeric_descriptor_df.columns:

    if descriptor == "Yield":
        continue

    values = numeric_descriptor_df[
        descriptor
    ]

    if not values.notna().all():

        missing_columns.append(
            descriptor
        )

    elif values.nunique() <= 1:

        constant_columns.append(
            descriptor
        )


complete_descriptor_columns = [

    descriptor
    for descriptor in numeric_descriptor_df.columns

    if (
        descriptor not in missing_columns
        and
        descriptor not in constant_columns
    )
]


if "Yield" not in complete_descriptor_columns:

    raise RuntimeError(
        "Yield must be complete."
    )


all_valid_df = pd.concat(
    [

        master_df[
            [
                "Ligand",
                "XYZ_File"
            ]
        ].reset_index(
            drop=True
        ),

        numeric_descriptor_df[
            complete_descriptor_columns
        ].reset_index(
            drop=True
        )
    ],
    axis=1
)


# ============================================================
# 24. NONREDUNDANT CORRELATION POLICY
#
# These values stay in ALL_VALID_DESCRIPTORS but are removed
# from final correlation ranking for scientific reasons.
# ============================================================

EXCLUDE_FROM_CORRELATION = {

    # Same project classification as Ring_Size:
    "Palladacycle_Path_Atom_Count",

    # Exact relationship for a two-donor system:
    "Pd_Coordination_Distortion_A",

    # Closely related Pd-donor asymmetry metric:
    "Pd_Donor_Distance_Ratio",

    # Donor1/Donor2 charges are summarized by average:
    "XTB_IsolatedLigand_Donor1_Mulliken_Charge",

    "XTB_IsolatedLigand_Donor2_Mulliken_Charge",

    # Raw volume values are size-dependent:
    f"Ligand_Distal_Volume_Outside_{DISTAL_VOLUME_RADIUS:.1f}A_A3",

    "Ligand_Molecular_Volume_A3",

    # SASA area and volume are related:
    "Ligand_SASA_A2",

    # Solid-angle family:
    "Ligand_Solid_Angle_sr",

    "Ligand_Solid_G_percent",

    # Retain B5 as primary Sterimol width:
    "Pd_Oriented_Sterimol_B1_A",

    "Pd_Oriented_Sterimol_L_A",

    # Retain HOMO rather than an overlapping orbital set:
    "XTB_IsolatedLigand_LUMO_eV",

    "XTB_IsolatedLigand_HOMO_LUMO_Gap_eV",

    "XTB_IsolatedLigand_Dipole_Debye"
}


correlation_candidates = [

    column
    for column in complete_descriptor_columns

    if (
        column != "Yield"
        and
        column not in EXCLUDE_FROM_CORRELATION
    )
]


# Scientifically preferred order for resolving high correlations.
DESCRIPTOR_PRIORITY = [

    "Pd_Donor1_Distance_A",

    "Palladacycle_Ring_Size",

    "Bite_Angle_deg",

    "Ligand_%Vbur_3.0A",

    f"Ligand_Distal_Volume_Fraction_Outside_{DISTAL_VOLUME_RADIUS:.1f}A",

    "Pd_Oriented_Sterimol_B5_A",

    "Ligand_Pint",

    "XTB_IsolatedLigand_HOMO_eV",

    "XTB_IsolatedLigand_Average_Donor_Charge",

    "Palladacycle_Path_Bond_Mean_A",

    "Palladacycle_Path_Bond_SD_A",

    "Palladacycle_Path_Bond_CV",

    "Ligand_%Vbur_4.0A",

    "Ligand_%Vbur_5.5A",

    "Donor1_Donor2_Distance_A",

    "Delta_Pd_Donor_Distance_A",

    "Average_Pd_Donor_Distance_A",

    "Pd_Donor2_Distance_A",

    "Ligand_Solid_Cone_Angle_deg",

    "Ligand_SASA_Volume_A3",

    "Ligand_Maximum_Extension_A",

    "Ligand_Pint_per_SASA"
]


priority_lookup = {

    descriptor:
        index

    for index, descriptor
    in enumerate(
        DESCRIPTOR_PRIORITY
    )
}


correlation_candidates = sorted(
    correlation_candidates,
    key=lambda descriptor: (

        priority_lookup.get(
            descriptor,
            9999
        ),

        descriptor
    )
)


candidate_df = numeric_descriptor_df[
    correlation_candidates
].copy()


candidate_spearman_matrix = candidate_df.corr(
    method="spearman"
)


retained_descriptors = []

redundant_records = []

for descriptor in correlation_candidates:

    redundant_with = None

    for already_kept in retained_descriptors:

        rho = candidate_spearman_matrix.loc[
            descriptor,
            already_kept
        ]

        if (
            pd.notna(rho)
            and
            abs(rho)
            >=
            REDUNDANCY_SPEARMAN_THRESHOLD
        ):

            redundant_with = already_kept
            break


    if redundant_with is None:

        retained_descriptors.append(
            descriptor
        )

    else:

        redundant_records.append({

            "Descriptor":
                descriptor,

            "Reason":
                "High descriptor-descriptor Spearman correlation",

            "Redundant_With":
                redundant_with,

            "Spearman_rho":
                candidate_spearman_matrix.loc[
                    descriptor,
                    redundant_with
                ]
        })


correlation_df = pd.concat(
    [

        master_df[
            [
                "Ligand"
            ]
        ].reset_index(
            drop=True
        ),

        numeric_descriptor_df[
            ["Yield"]
            +
            retained_descriptors
        ].reset_index(
            drop=True
        )
    ],
    axis=1
)


# ============================================================
# 25. CORRELATION MATRICES
# ============================================================

matrix_df = correlation_df.drop(
    columns="Ligand"
)

pearson_matrix = matrix_df.corr(
    method="pearson"
)

spearman_matrix = matrix_df.corr(
    method="spearman"
)


# ============================================================
# 26. YIELD CORRELATIONS
# ============================================================

correlation_records = []

for descriptor in retained_descriptors:

    x = correlation_df[
        descriptor
    ]

    y = correlation_df[
        "Yield"
    ]

    try:

        pearson_r, pearson_p = pearsonr(
            x,
            y
        )

    except Exception:

        pearson_r = np.nan
        pearson_p = np.nan


    try:

        spearman_rho, spearman_p = spearmanr(
            x,
            y
        )

    except Exception:

        spearman_rho = np.nan
        spearman_p = np.nan


    correlation_records.append({

        "Descriptor":
            descriptor,

        "N":
            len(correlation_df),

        "Pearson_r":
            pearson_r,

        "Pearson_p":
            pearson_p,

        "Spearman_rho":
            spearman_rho,

        "Spearman_p":
            spearman_p,

        "Abs_Pearson_r":
            abs(pearson_r)
            if pd.notna(pearson_r)
            else np.nan,

        "Abs_Spearman_rho":
            abs(spearman_rho)
            if pd.notna(spearman_rho)
            else np.nan
    })


yield_correlation_df = pd.DataFrame(
    correlation_records
)


yield_correlation_df[
    "Pearson_FDR_q"
] = multipletests(
    yield_correlation_df[
        "Pearson_p"
    ].fillna(1.0),
    method="fdr_bh"
)[1]


yield_correlation_df[
    "Spearman_FDR_q"
] = multipletests(
    yield_correlation_df[
        "Spearman_p"
    ].fillna(1.0),
    method="fdr_bh"
)[1]


yield_correlation_df[
    "Mean_Abs_Correlation"
] = (

    yield_correlation_df[
        "Abs_Pearson_r"
    ]

    +

    yield_correlation_df[
        "Abs_Spearman_rho"
    ]
) / 2.0


yield_correlation_df = (
    yield_correlation_df
    .sort_values(
        "Mean_Abs_Correlation",
        ascending=False
    )
    .reset_index(
        drop=True
    )
)

yield_correlation_df.insert(
    0,
    "Rank",
    np.arange(
        1,
        len(yield_correlation_df) + 1
    )
)


# ============================================================
# 27. LEAVE-ONE-OUT ROBUSTNESS
# ============================================================

loo_records = []

for descriptor in retained_descriptors:

    loo_pearson = []

    loo_spearman = []

    for omitted_index in correlation_df.index:

        temporary_df = correlation_df.drop(
            index=omitted_index
        )

        try:

            r_value, _ = pearsonr(
                temporary_df[
                    descriptor
                ],
                temporary_df[
                    "Yield"
                ]
            )

            loo_pearson.append(
                r_value
            )

        except Exception:
            pass


        try:

            rho_value, _ = spearmanr(
                temporary_df[
                    descriptor
                ],
                temporary_df[
                    "Yield"
                ]
            )

            loo_spearman.append(
                rho_value
            )

        except Exception:
            pass


    full_result = yield_correlation_df.loc[
        yield_correlation_df[
            "Descriptor"
        ] == descriptor
    ].iloc[0]


    loo_records.append({

        "Descriptor":
            descriptor,

        "Full_Pearson_r":
            full_result[
                "Pearson_r"
            ],

        "LOO_Pearson_Min":
            np.min(loo_pearson)
            if loo_pearson
            else np.nan,

        "LOO_Pearson_Max":
            np.max(loo_pearson)
            if loo_pearson
            else np.nan,

        "LOO_Pearson_Mean":
            np.mean(loo_pearson)
            if loo_pearson
            else np.nan,

        "LOO_Pearson_SD":
            np.std(
                loo_pearson,
                ddof=0
            )
            if loo_pearson
            else np.nan,

        "Pearson_Sign_Changes":
            sum(
                np.sign(value)
                !=
                np.sign(
                    full_result[
                        "Pearson_r"
                    ]
                )
                for value in loo_pearson
            ),

        "Full_Spearman_rho":
            full_result[
                "Spearman_rho"
            ],

        "LOO_Spearman_Min":
            np.min(loo_spearman)
            if loo_spearman
            else np.nan,

        "LOO_Spearman_Max":
            np.max(loo_spearman)
            if loo_spearman
            else np.nan,

        "LOO_Spearman_Mean":
            np.mean(loo_spearman)
            if loo_spearman
            else np.nan,

        "LOO_Spearman_SD":
            np.std(
                loo_spearman,
                ddof=0
            )
            if loo_spearman
            else np.nan,

        "Spearman_Sign_Changes":
            sum(
                np.sign(value)
                !=
                np.sign(
                    full_result[
                        "Spearman_rho"
                    ]
                )
                for value in loo_spearman
            )
    })


loo_df = pd.DataFrame(
    loo_records
)


# ============================================================
# 28. HIGH CORRELATION PAIRS
# ============================================================

high_pair_records = []

for i, descriptor_1 in enumerate(
    retained_descriptors
):

    for descriptor_2 in retained_descriptors[
        i + 1:
    ]:

        pearson_value = pearson_matrix.loc[
            descriptor_1,
            descriptor_2
        ]

        spearman_value = spearman_matrix.loc[
            descriptor_1,
            descriptor_2
        ]

        if (
            abs(pearson_value)
            >=
            HIGH_CORR_THRESHOLD
            or
            abs(spearman_value)
            >=
            HIGH_CORR_THRESHOLD
        ):

            high_pair_records.append({

                "Descriptor_1":
                    descriptor_1,

                "Descriptor_2":
                    descriptor_2,

                "Pearson_r":
                    pearson_value,

                "Spearman_rho":
                    spearman_value,

                "Max_Abs_Correlation":
                    max(
                        abs(pearson_value),
                        abs(spearman_value)
                    )
            })


high_corr_df = pd.DataFrame(
    high_pair_records,
    columns=[

        "Descriptor_1",

        "Descriptor_2",

        "Pearson_r",

        "Spearman_rho",

        "Max_Abs_Correlation"
    ]
)


if len(high_corr_df) > 0:

    high_corr_df = (
        high_corr_df
        .sort_values(
            "Max_Abs_Correlation",
            ascending=False
        )
        .reset_index(
            drop=True
        )
    )


# ============================================================
# 29. ALL-DESCRIPTOR AUDIT
# ============================================================

all_numeric_columns = numeric_df.columns.tolist()

rank_lookup = {

    row["Descriptor"]:
        row

    for _, row
    in yield_correlation_df.iterrows()
}


excluded_reason_lookup = {

    descriptor:
        "Excluded by nonredundant correlation policy."

    for descriptor
    in EXCLUDE_FROM_CORRELATION
}


for record in redundant_records:

    excluded_reason_lookup[
        record["Descriptor"]
    ] = (
        f"Redundant with {record['Redundant_With']}; "
        f"Spearman rho = {record['Spearman_rho']:.3f}"
    )


audit_records = []

for descriptor in all_numeric_columns:

    n_valid = int(
        numeric_df[
            descriptor
        ].notna().sum()
    )

    n_missing = int(
        numeric_df[
            descriptor
        ].isna().sum()
    )

    n_unique = int(
        numeric_df[
            descriptor
        ].dropna().nunique()
    )

    record = {

        "Descriptor":
            descriptor,

        "N_Valid":
            n_valid,

        "N_Missing":
            n_missing,

        "N_Unique":
            n_unique,

        "Status":
            "",

        "Final_Rank":
            np.nan,

        "Pearson_r":
            np.nan,

        "Pearson_p":
            np.nan,

        "Pearson_FDR_q":
            np.nan,

        "Spearman_rho":
            np.nan,

        "Spearman_p":
            np.nan,

        "Spearman_FDR_q":
            np.nan,

        "Mean_Abs_Correlation":
            np.nan,

        "Reason":
            ""
    }


    if descriptor == "Yield":

        record[
            "Status"
        ] = "RESPONSE_VARIABLE"

        record[
            "Reason"
        ] = "Experimental response variable."

    elif descriptor in NON_DESCRIPTOR_COLUMNS:

        record[
            "Status"
        ] = "NOT_A_CHEMICAL_DESCRIPTOR"

        record[
            "Reason"
        ] = (
            "Metadata, atom index, or QC count."
        )

    elif descriptor in missing_columns:

        record[
            "Status"
        ] = "EXCLUDED_MISSING_VALUES"

        record[
            "Reason"
        ] = (
            "One or more ligand values were missing."
        )

    elif descriptor in constant_columns:

        record[
            "Status"
        ] = "EXCLUDED_CONSTANT"

        record[
            "Reason"
        ] = (
            "No variation across ligands."
        )

    elif descriptor in rank_lookup:

        row = rank_lookup[
            descriptor
        ]

        record[
            "Status"
        ] = "RANKED"

        record[
            "Final_Rank"
        ] = row[
            "Rank"
        ]

        record[
            "Pearson_r"
        ] = row[
            "Pearson_r"
        ]

        record[
            "Pearson_p"
        ] = row[
            "Pearson_p"
        ]

        record[
            "Pearson_FDR_q"
        ] = row[
            "Pearson_FDR_q"
        ]

        record[
            "Spearman_rho"
        ] = row[
            "Spearman_rho"
        ]

        record[
            "Spearman_p"
        ] = row[
            "Spearman_p"
        ]

        record[
            "Spearman_FDR_q"
        ] = row[
            "Spearman_FDR_q"
        ]

        record[
            "Mean_Abs_Correlation"
        ] = row[
            "Mean_Abs_Correlation"
        ]

        record[
            "Reason"
        ] = (
            "Complete descriptor retained for "
            "correlation analysis."
        )

    elif descriptor in excluded_reason_lookup:

        record[
            "Status"
        ] = "EXCLUDED_BY_POLICY"

        record[
            "Reason"
        ] = excluded_reason_lookup[
            descriptor
        ]

    else:

        record[
            "Status"
        ] = "NOT_RANKED"

        record[
            "Reason"
        ] = (
            "Complete numerical descriptor not selected "
            "for final correlation analysis."
        )


    audit_records.append(
        record
    )


descriptor_audit_df = pd.DataFrame(
    audit_records
)


status_priority = {

    "RANKED": 1,

    "EXCLUDED_BY_POLICY": 2,

    "EXCLUDED_MISSING_VALUES": 3,

    "EXCLUDED_CONSTANT": 4,

    "NOT_RANKED": 5,

    "NOT_A_CHEMICAL_DESCRIPTOR": 6,

    "RESPONSE_VARIABLE": 7
}


descriptor_audit_df[
    "_Status_Order"
] = descriptor_audit_df[
    "Status"
].map(
    status_priority
).fillna(99)


descriptor_audit_df = (
    descriptor_audit_df
    .sort_values(
        by=[

            "_Status_Order",

            "Final_Rank",

            "Descriptor"
        ]
    )
    .drop(
        columns="_Status_Order"
    )
    .reset_index(
        drop=True
    )
)


# ============================================================
# 30. REMOVED-DESCRIPTOR TABLE
# ============================================================

removed_records = []

for descriptor in missing_columns:

    removed_records.append({

        "Descriptor":
            descriptor,

        "Reason":
            "Contains missing value(s)",

        "Details":
            "Excluded from all complete-descriptor analyses."
    })


for descriptor in constant_columns:

    removed_records.append({

        "Descriptor":
            descriptor,

        "Reason":
            "Constant descriptor",

        "Details":
            "No variation among ligands."
    })


for descriptor in EXCLUDE_FROM_CORRELATION:

    if descriptor in all_valid_df.columns:

        removed_records.append({

            "Descriptor":
                descriptor,

            "Reason":
                "Excluded from final correlation analysis",

            "Details":
                "Intentional nonredundancy/QC policy."
        })


for record in redundant_records:

    removed_records.append({

        "Descriptor":
            record["Descriptor"],

        "Reason":
            "High descriptor-descriptor correlation",

        "Details":
            (
                f"Redundant with "
                f"{record['Redundant_With']}; "
                f"Spearman rho = "
                f"{record['Spearman_rho']:.3f}"
            )
    })


removed_df = pd.DataFrame(
    removed_records,
    columns=[

        "Descriptor",

        "Reason",

        "Details"
    ]
)


# ============================================================
# 31. PUBLICATION-QUALITY MATRIX PLOTS
# ============================================================

def plot_correlation_matrix(
    matrix,
    title,
    colorbar_label,
    filename_prefix
):

    n = len(matrix)

    figure_size = max(
        15,
        min(
            30,
            n * 0.72
        )
    )

    annotation_font_size = max(
        8,
        min(
            14,
            260 / max(n, 1)
        )
    )

    fig, ax = plt.subplots(
        figsize=(
            figure_size,
            figure_size
        ),
        constrained_layout=True
    )

    heatmap = sns.heatmap(
        matrix,
        ax=ax,
        cmap="RdBu_r",
        center=0,
        vmin=-1,
        vmax=1,
        square=True,
        linewidths=0.35,
        linecolor="white",
        annot=(
            n <= MATRIX_ANNOTATION_MAX
        ),
        fmt=".2f",
        annot_kws={
            "fontsize":
                annotation_font_size
        },
        cbar_kws={
            "label":
                colorbar_label,

            "shrink":
                0.78,

            "pad":
                0.02
        }
    )

    ax.set_title(
        title,
        fontsize=16,
        fontweight="bold",
        pad=16
    )

    ax.set_xticklabels(
        ax.get_xticklabels(),
        rotation=90,
        ha="center",
        fontsize=9
    )

    ax.set_yticklabels(
        ax.get_yticklabels(),
        rotation=0,
        fontsize=9
    )

    ax.tick_params(
        length=0
    )

    colorbar = heatmap.collections[0].colorbar

    colorbar.ax.tick_params(
        labelsize=9
    )

    colorbar.set_label(
        colorbar_label,
        fontsize=10
    )

    png_file = os.path.join(
        PLOT_FOLDER,
        f"{filename_prefix}.png"
    )

    pdf_file = os.path.join(
        PLOT_FOLDER,
        f"{filename_prefix}.pdf"
    )

    fig.savefig(
        png_file,
        dpi=MATRIX_DPI,
        bbox_inches="tight",
        facecolor="white"
    )

    fig.savefig(
        pdf_file,
        dpi=MATRIX_DPI,
        bbox_inches="tight",
        facecolor="white"
    )

    plt.show()

    plt.close(fig)


plot_correlation_matrix(
    pearson_matrix,
    "Pearson Correlation Matrix\nYield + Complete Nonredundant Descriptors",
    "Pearson r",
    "Pearson_Correlation_Matrix"
)

plot_correlation_matrix(
    spearman_matrix,
    "Spearman Correlation Matrix\nYield + Complete Nonredundant Descriptors",
    "Spearman ρ",
    "Spearman_Correlation_Matrix"
)


# ============================================================
# 32. YIELD-RANKING PLOT
# ============================================================

ranking_plot_df = (
    yield_correlation_df
    .sort_values(
        "Pearson_r",
        ascending=True
    )
)

plt.figure(
    figsize=(
        10,
        max(
            6,
            len(ranking_plot_df)
            * 0.42
        )
    )
)

bar_colors = [

    "#2166ac"
    if value < 0
    else "#b2182b"

    for value in ranking_plot_df[
        "Pearson_r"
    ]
]

plt.barh(
    ranking_plot_df[
        "Descriptor"
    ],
    ranking_plot_df[
        "Pearson_r"
    ],
    color=bar_colors
)

plt.axvline(
    0,
    color="black",
    linewidth=1
)

plt.xlabel(
    "Pearson r with Yield",
    fontsize=11
)

plt.ylabel(
    "Descriptor",
    fontsize=11
)

plt.title(
    "Yield Correlation Ranking",
    fontsize=14,
    fontweight="bold"
)

plt.tight_layout()

plt.savefig(
    os.path.join(
        PLOT_FOLDER,
        "Yield_Pearson_Correlation_Ranking.png"
    ),
    dpi=600,
    bbox_inches="tight",
    facecolor="white"
)

plt.savefig(
    os.path.join(
        PLOT_FOLDER,
        "Yield_Pearson_Correlation_Ranking.pdf"
    ),
    dpi=600,
    bbox_inches="tight",
    facecolor="white"
)

plt.show()

plt.close()


# ============================================================
# 33. WRITE ONE EXCEL WORKBOOK
# ============================================================

with pd.ExcelWriter(
    FINAL_EXCEL,
    engine="openpyxl"
) as writer:

    all_valid_df.to_excel(
        writer,
        sheet_name="ALL_VALID_DESCRIPTORS",
        index=False
    )

    validation_df.to_excel(
        writer,
        sheet_name="VALIDATION",
        index=False
    )

    correlation_df.to_excel(
        writer,
        sheet_name="CORRELATION_DATA",
        index=False
    )

    yield_correlation_df.to_excel(
        writer,
        sheet_name="YIELD_CORRELATIONS",
        index=False
    )

    pearson_matrix.to_excel(
        writer,
        sheet_name="PEARSON_MATRIX"
    )

    spearman_matrix.to_excel(
        writer,
        sheet_name="SPEARMAN_MATRIX"
    )

    loo_df.to_excel(
        writer,
        sheet_name="LOO_ROBUSTNESS",
        index=False
    )

    high_corr_df.to_excel(
        writer,
        sheet_name="HIGH_CORR_PAIRS",
        index=False
    )

    removed_df.to_excel(
        writer,
        sheet_name="REMOVED_DESCRIPTORS",
        index=False
    )

    descriptor_audit_df.to_excel(
        writer,
        sheet_name="ALL_DESCRIPTOR_AUDIT",
        index=False
    )

    pd.DataFrame({

        "Ligand":
            list(
                YIELD_DATA.keys()
            ),

        "Yield":
            list(
                YIELD_DATA.values()
            )

    }).to_excel(
        writer,
        sheet_name="YIELD_DATA",
        index=False
    )


# ============================================================
# 34. FINAL REPORT
# ============================================================

print("\n" + "=" * 100)
print("FINAL PUBLICATION ANALYSIS COMPLETE")
print("=" * 100)

print(
    "\nProcessed XYZ files:",
    len(master_df)
)

print(
    "Complete descriptor columns in ALL_VALID_DESCRIPTORS:",
    len(all_valid_df.columns) - 3
)

print(
    "Final nonredundant correlation descriptors:",
    len(retained_descriptors)
)

print(
    "Descriptors excluded due to missing values:",
    len(missing_columns)
)

print(
    "Descriptors excluded due to no variation:",
    len(constant_columns)
)

print(
    "\nExcel workbook:"
)

print(
    FINAL_EXCEL
)

print(
    "\nPlots folder:"
)

print(
    PLOT_FOLDER
)

print(
    "\nTOP 10 YIELD CORRELATIONS:"
)

display(
    yield_correlation_df.head(10)
)

print("\nDone.")
```

```python
# ============================================================
# TOP-10 YIELD-CORRELATED DESCRIPTORS
# PUBLICATION-QUALITY PEARSON + SPEARMAN MATRICES
#
# Run this cell AFTER the full analysis has created:
#   correlation_df
#   yield_correlation_df
#   PLOT_FOLDER
#   OUTPUT_FOLDER
# ============================================================

import os
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns


# ============================================================
# 1. SETTINGS
# ============================================================

TOP_N = 10

TOP10_OUTPUT_EXCEL = os.path.join(
    OUTPUT_FOLDER,
    "Top10_Descriptor_Correlation_Matrices.xlsx"
)

TOP10_PEARSON_PNG = os.path.join(
    PLOT_FOLDER,
    "Top10_Pearson_Correlation_Matrix.png"
)

TOP10_SPEARMAN_PNG = os.path.join(
    PLOT_FOLDER,
    "Top10_Spearman_Correlation_Matrix.png"
)

TOP10_PEARSON_PDF = os.path.join(
    PLOT_FOLDER,
    "Top10_Pearson_Correlation_Matrix.pdf"
)

TOP10_SPEARMAN_PDF = os.path.join(
    PLOT_FOLDER,
    "Top10_Spearman_Correlation_Matrix.pdf"
)


# ============================================================
# 2. SELECT TOP 10 DESCRIPTORS
#
# Uses Mean_Abs_Correlation ranking from your output table.
# ============================================================

top10_table = (
    yield_correlation_df
    .dropna(
        subset=[
            "Descriptor",
            "Mean_Abs_Correlation"
        ]
    )
    .sort_values(
        "Mean_Abs_Correlation",
        ascending=False
    )
    .head(TOP_N)
    .copy()
    .reset_index(drop=True)
)

top10_table.insert(
    0,
    "Manuscript_Rank",
    np.arange(
        1,
        len(top10_table) + 1
    )
)

top10_descriptors = top10_table[
    "Descriptor"
].tolist()


# ============================================================
# 3. VERIFY DESCRIPTORS EXIST IN CORRELATION DATA
# ============================================================

missing_descriptors = [

    descriptor
    for descriptor in top10_descriptors

    if descriptor not in correlation_df.columns
]

if missing_descriptors:

    raise ValueError(
        "These top-10 descriptors are not present in "
        "correlation_df:\n"
        + "\n".join(missing_descriptors)
    )


# ============================================================
# 4. CREATE TOP-10 DATASET
#
# Yield is kept as the first row/column.
# ============================================================

top10_columns = [
    "Yield"
] + top10_descriptors

top10_df = correlation_df[
    top10_columns
].copy()


# ============================================================
# 5. CREATE CORRELATION MATRICES
# ============================================================

pearson_top10 = top10_df.corr(
    method="pearson"
)

spearman_top10 = top10_df.corr(
    method="spearman"
)


# ============================================================
# 6. SHORTER DISPLAY LABELS
#
# These improve manuscript figure readability without changing
# the actual Excel descriptor names.
# ============================================================

DISPLAY_LABELS = {

    "Yield":
        "Yield",

    "Pd_Donor1_Distance_A":
        "Pd–Donor1 distance",

    "Palladacycle_Ring_Size":
        "Palladacycle ring size",

    "Donor1_Donor2_Distance_A":
        "Donor1–Donor2 distance",

    "Bite_Angle_deg":
        "Bite angle",

    "Pd_Oriented_Sterimol_B5_A":
        "Pd-oriented Sterimol B5",

    "Ligand_Pint":
        "Ligand P_int",

    "Ligand_SASA_Volume_A3":
        "Ligand SASA volume",

    "Ligand_Solid_Cone_Angle_deg":
        "Solid cone angle",

    "Delta_Pd_Donor_Distance_A":
        "Δ(Pd–donor distance)",

    "Palladacycle_Path_Bond_Mean_A":
        "Mean palladacycle bond"
}


def replace_labels(matrix):

    display_matrix = matrix.copy()

    display_matrix.index = [
        DISPLAY_LABELS.get(
            item,
            item
        )
        for item in display_matrix.index
    ]

    display_matrix.columns = [
        DISPLAY_LABELS.get(
            item,
            item
        )
        for item in display_matrix.columns
    ]

    return display_matrix


pearson_plot_matrix = replace_labels(
    pearson_top10
)

spearman_plot_matrix = replace_labels(
    spearman_top10
)


# ============================================================
# 7. PLOT FUNCTION
#
# Produces full square annotated correlation matrix.
# ============================================================

def plot_top10_matrix(
    matrix,
    title,
    colorbar_label,
    png_file,
    pdf_file
):

    sns.set_theme(
        style="white",
        font_scale=1.0
    )

    fig, ax = plt.subplots(
        figsize=(14, 12),
        constrained_layout=True
    )

    heatmap = sns.heatmap(
        matrix,
        ax=ax,
        cmap="RdBu_r",
        center=0,
        vmin=-1,
        vmax=1,
        square=True,
        annot=True,
        fmt=".2f",
        linewidths=0.70,
        linecolor="white",
        cbar_kws={
            "label": colorbar_label,
            "shrink": 0.82,
            "pad": 0.02
        },
        annot_kws={
            "fontsize": 11,
            "fontweight": "normal"
        }
    )

    ax.set_title(
        title,
        fontsize=18,
        fontweight="bold",
        pad=18
    )

    ax.set_xticklabels(
        ax.get_xticklabels(),
        rotation=45,
        ha="right",
        rotation_mode="anchor",
        fontsize=11
    )

    ax.set_yticklabels(
        ax.get_yticklabels(),
        rotation=0,
        fontsize=11
    )

    ax.tick_params(
        length=0
    )

    colorbar = heatmap.collections[0].colorbar

    colorbar.ax.tick_params(
        labelsize=11
    )

    colorbar.set_label(
        colorbar_label,
        fontsize=12
    )

    fig.savefig(
        png_file,
        dpi=600,
        bbox_inches="tight",
        facecolor="white"
    )

    fig.savefig(
        pdf_file,
        dpi=600,
        bbox_inches="tight",
        facecolor="white"
    )

    plt.show()

    plt.close(fig)


# ============================================================
# 8. PLOT PEARSON MATRIX
# ============================================================

plot_top10_matrix(
    pearson_plot_matrix,
    "Top 10 Ligand Descriptors: Pearson Correlation Matrix",
    "Pearson r",
    TOP10_PEARSON_PNG,
    TOP10_PEARSON_PDF
)


# ============================================================
# 9. PLOT SPEARMAN MATRIX
# ============================================================

plot_top10_matrix(
    spearman_plot_matrix,
    "Top 10 Ligand Descriptors: Spearman Correlation Matrix",
    "Spearman ρ",
    TOP10_SPEARMAN_PNG,
    TOP10_SPEARMAN_PDF
)


# ============================================================
# 10. WRITE TOP-10 RESULTS TO EXCEL
# ============================================================

with pd.ExcelWriter(
    TOP10_OUTPUT_EXCEL,
    engine="openpyxl"
) as writer:

    top10_table.to_excel(
        writer,
        sheet_name="Top10_Yield_Ranking",
        index=False
    )

    top10_df.to_excel(
        writer,
        sheet_name="Top10_Descriptor_Data",
        index=False
    )

    pearson_top10.to_excel(
        writer,
        sheet_name="Top10_Pearson_Matrix"
    )

    spearman_top10.to_excel(
        writer,
        sheet_name="Top10_Spearman_Matrix"
    )


# ============================================================
# 11. DISPLAY FINAL TOP-10 TABLE
# ============================================================

print("=" * 100)
print("TOP 10 DESCRIPTORS USED FOR MANUSCRIPT CORRELATION MATRICES")
print("=" * 100)

display(
    top10_table[
        [
            "Manuscript_Rank",
            "Descriptor",
            "N",
            "Pearson_r",
            "Pearson_p",
            "Pearson_FDR_q",
            "Spearman_rho",
            "Spearman_p",
            "Spearman_FDR_q",
            "Mean_Abs_Correlation"
        ]
    ]
)

print("\nTop-10 Excel workbook:")
print(TOP10_OUTPUT_EXCEL)

print("\nPearson figure:")
print(TOP10_PEARSON_PNG)

print("\nSpearman figure:")
print(TOP10_SPEARMAN_PNG)
```

```python
# ============================================================
# AUDIT TABLE:
# STATUS AND FINAL RANK OF EVERY GENERATED NUMERIC DESCRIPTOR
#
# Run after master_df, correlation_df, yield_correlation_df,
# constant_columns, missing_columns, and removed_df exist.
# ============================================================

import numpy as np
import pandas as pd


# ------------------------------------------------------------
# Identify every numeric column in the raw descriptor table.
# ------------------------------------------------------------

all_numeric_columns = master_df.select_dtypes(
    include=[np.number]
).columns.tolist()


# ------------------------------------------------------------
# Define metadata / atom-index / QC columns.
# These are not chemical descriptors.
# ------------------------------------------------------------

non_descriptor_columns = {

    "Yield",

    "Pd_Atom",

    "Ligand_Start_Atom",

    "N_Atoms",

    "Donor1_Atom",

    "Donor2_Atom",

    "Number_of_Donor_Candidates",

    "Number_of_Qualifying_Pd_Donors",

    "Donor1_Pyramidalization_Neighbor_Count",

    "Donor2_Pyramidalization_Neighbor_Count"
}


# ------------------------------------------------------------
# Create a lookup from ranked correlation table.
# ------------------------------------------------------------

rank_lookup = {}

if "Rank" in yield_correlation_df.columns:

    for _, row in yield_correlation_df.iterrows():

        rank_lookup[
            row["Descriptor"]
        ] = {

            "Final_Rank":
                row["Rank"],

            "Pearson_r":
                row["Pearson_r"],

            "Pearson_p":
                row["Pearson_p"],

            "Pearson_FDR_q":
                row.get(
                    "Pearson_FDR_q",
                    np.nan
                ),

            "Spearman_rho":
                row["Spearman_rho"],

            "Spearman_p":
                row["Spearman_p"],

            "Spearman_FDR_q":
                row.get(
                    "Spearman_FDR_q",
                    np.nan
                ),

            "Mean_Abs_Correlation":
                row["Mean_Abs_Correlation"]
        }


# ------------------------------------------------------------
# Create an exclusion lookup, if removed_df exists.
# ------------------------------------------------------------

removed_reason_lookup = {}

if "removed_df" in globals():

    if (
        isinstance(
            removed_df,
            pd.DataFrame
        )
        and
        "Descriptor" in removed_df.columns
    ):

        for _, row in removed_df.iterrows():

            descriptor = row[
                "Descriptor"
            ]

            reason = row.get(
                "Reason",
                ""
            )

            details = row.get(
                "Details",
                ""
            )

            removed_reason_lookup[
                descriptor
            ] = (
                f"{reason}. {details}"
            ).strip()


# ------------------------------------------------------------
# Build audit records.
# ------------------------------------------------------------

audit_records = []

for descriptor in all_numeric_columns:

    nonmissing_n = int(
        master_df[
            descriptor
        ].notna().sum()
    )

    missing_n = int(
        master_df[
            descriptor
        ].isna().sum()
    )

    n_unique = int(
        master_df[
            descriptor
        ].dropna().nunique()
    )

    record = {

        "Descriptor":
            descriptor,

        "N_Valid":
            nonmissing_n,

        "N_Missing":
            missing_n,

        "N_Unique":
            n_unique,

        "Status":
            "",

        "Final_Rank":
            np.nan,

        "Pearson_r":
            np.nan,

        "Pearson_p":
            np.nan,

        "Pearson_FDR_q":
            np.nan,

        "Spearman_rho":
            np.nan,

        "Spearman_p":
            np.nan,

        "Spearman_FDR_q":
            np.nan,

        "Mean_Abs_Correlation":
            np.nan,

        "Reason":
            ""
    }


    # --------------------------------------------------------
    # Non-descriptor / metadata.
    # --------------------------------------------------------

    if descriptor in non_descriptor_columns:

        record[
            "Status"
        ] = "NOT_A_CHEMICAL_DESCRIPTOR"

        record[
            "Reason"
        ] = (
            "Metadata, atom index, or QC count; "
            "not eligible for correlation ranking."
        )


    # --------------------------------------------------------
    # Missing values.
    # --------------------------------------------------------

    elif missing_n > 0:

        record[
            "Status"
        ] = "EXCLUDED_MISSING_VALUES"

        record[
            "Reason"
        ] = (
            "One or more ligands lack a valid value. "
            "Excluded from publication correlation analysis."
        )


    # --------------------------------------------------------
    # Constant columns.
    # --------------------------------------------------------

    elif n_unique <= 1:

        record[
            "Status"
        ] = "EXCLUDED_CONSTANT"

        record[
            "Reason"
        ] = (
            "No variation across ligands."
        )


    # --------------------------------------------------------
    # Retained/ranked descriptors.
    # --------------------------------------------------------

    elif descriptor in rank_lookup:

        values = rank_lookup[
            descriptor
        ]

        record[
            "Status"
        ] = "RANKED"

        record[
            "Final_Rank"
        ] = values[
            "Final_Rank"
        ]

        record[
            "Pearson_r"
        ] = values[
            "Pearson_r"
        ]

        record[
            "Pearson_p"
        ] = values[
            "Pearson_p"
        ]

        record[
            "Pearson_FDR_q"
        ] = values[
            "Pearson_FDR_q"
        ]

        record[
            "Spearman_rho"
        ] = values[
            "Spearman_rho"
        ]

        record[
            "Spearman_p"
        ] = values[
            "Spearman_p"
        ]

        record[
            "Spearman_FDR_q"
        ] = values[
            "Spearman_FDR_q"
        ]

        record[
            "Mean_Abs_Correlation"
        ] = values[
            "Mean_Abs_Correlation"
        ]

        record[
            "Reason"
        ] = (
            "Complete descriptor retained for correlation analysis."
        )


    # --------------------------------------------------------
    # Intentionally excluded / redundant descriptors.
    # --------------------------------------------------------

    elif descriptor in removed_reason_lookup:

        record[
            "Status"
        ] = "EXCLUDED_BY_POLICY"

        record[
            "Reason"
        ] = removed_reason_lookup[
            descriptor
        ]


    # --------------------------------------------------------
    # Fallback category.
    # --------------------------------------------------------

    else:

        record[
            "Status"
        ] = "NOT_RANKED"

        record[
            "Reason"
        ] = (
            "Complete numeric variable, but not included in "
            "the final correlation candidate set."
        )


    audit_records.append(
        record
    )


# ------------------------------------------------------------
# Build final audit table.
# ------------------------------------------------------------

descriptor_audit_df = pd.DataFrame(
    audit_records
)


status_order = {

    "RANKED": 1,

    "EXCLUDED_BY_POLICY": 2,

    "EXCLUDED_MISSING_VALUES": 3,

    "EXCLUDED_CONSTANT": 4,

    "NOT_RANKED": 5,

    "NOT_A_CHEMICAL_DESCRIPTOR": 6
}


descriptor_audit_df[
    "_Status_Order"
] = descriptor_audit_df[
    "Status"
].map(
    status_order
).fillna(99)


descriptor_audit_df = (
    descriptor_audit_df
    .sort_values(
        by=[
            "_Status_Order",
            "Final_Rank",
            "Descriptor"
        ]
    )
    .drop(
        columns="_Status_Order"
    )
    .reset_index(
        drop=True
    )
)


# ------------------------------------------------------------
# Display ranked descriptors and all descriptors.
# ------------------------------------------------------------

print("=" * 100)
print("RANKED DESCRIPTORS")
print("=" * 100)

display(
    descriptor_audit_df[
        descriptor_audit_df[
            "Status"
        ] == "RANKED"
    ]
)


print("=" * 100)
print("ALL GENERATED NUMERIC VARIABLES: STATUS AND RANK")
print("=" * 100)

display(
    descriptor_audit_df
)


# ------------------------------------------------------------
# Save to final Excel workbook.
# ------------------------------------------------------------

with pd.ExcelWriter(
    FINAL_EXCEL,
    engine="openpyxl",
    mode="a",
    if_sheet_exists="replace"
) as writer:

    descriptor_audit_df.to_excel(
        writer,
        sheet_name="ALL_DESCRIPTOR_RANKS",
        index=False
    )


print(
    "\nSaved status/rank audit table to:"
)

print(
    FINAL_EXCEL
)
```

```python
# ============================================================
# AUDIT TABLE:
# STATUS AND FINAL RANK OF EVERY GENERATED NUMERIC DESCRIPTOR
#
# Run after master_df, correlation_df, yield_correlation_df,
# constant_columns, missing_columns, and removed_df exist.
# ============================================================

import numpy as np
import pandas as pd


# ------------------------------------------------------------
# Identify every numeric column in the raw descriptor table.
# ------------------------------------------------------------

all_numeric_columns = master_df.select_dtypes(
    include=[np.number]
).columns.tolist()


# ------------------------------------------------------------
# Define metadata / atom-index / QC columns.
# These are not chemical descriptors.
# ------------------------------------------------------------

non_descriptor_columns = {

    "Yield",

    "Pd_Atom",

    "Ligand_Start_Atom",

    "N_Atoms",

    "Donor1_Atom",

    "Donor2_Atom",

    "Number_of_Donor_Candidates",

    "Number_of_Qualifying_Pd_Donors",

    "Donor1_Pyramidalization_Neighbor_Count",

    "Donor2_Pyramidalization_Neighbor_Count"
}


# ------------------------------------------------------------
# Create a lookup from ranked correlation table.
# ------------------------------------------------------------

rank_lookup = {}

if "Rank" in yield_correlation_df.columns:

    for _, row in yield_correlation_df.iterrows():

        rank_lookup[
            row["Descriptor"]
        ] = {

            "Final_Rank":
                row["Rank"],

            "Pearson_r":
                row["Pearson_r"],

            "Pearson_p":
                row["Pearson_p"],

            "Pearson_FDR_q":
                row.get(
                    "Pearson_FDR_q",
                    np.nan
                ),

            "Spearman_rho":
                row["Spearman_rho"],

            "Spearman_p":
                row["Spearman_p"],

            "Spearman_FDR_q":
                row.get(
                    "Spearman_FDR_q",
                    np.nan
                ),

            "Mean_Abs_Correlation":
                row["Mean_Abs_Correlation"]
        }


# ------------------------------------------------------------
# Create an exclusion lookup, if removed_df exists.
# ------------------------------------------------------------

removed_reason_lookup = {}

if "removed_df" in globals():

    if (
        isinstance(
            removed_df,
            pd.DataFrame
        )
        and
        "Descriptor" in removed_df.columns
    ):

        for _, row in removed_df.iterrows():

            descriptor = row[
                "Descriptor"
            ]

            reason = row.get(
                "Reason",
                ""
            )

            details = row.get(
                "Details",
                ""
            )

            removed_reason_lookup[
                descriptor
            ] = (
                f"{reason}. {details}"
            ).strip()


# ------------------------------------------------------------
# Build audit records.
# ------------------------------------------------------------

audit_records = []

for descriptor in all_numeric_columns:

    nonmissing_n = int(
        master_df[
            descriptor
        ].notna().sum()
    )

    missing_n = int(
        master_df[
            descriptor
        ].isna().sum()
    )

    n_unique = int(
        master_df[
            descriptor
        ].dropna().nunique()
    )

    record = {

        "Descriptor":
            descriptor,

        "N_Valid":
            nonmissing_n,

        "N_Missing":
            missing_n,

        "N_Unique":
            n_unique,

        "Status":
            "",

        "Final_Rank":
            np.nan,

        "Pearson_r":
            np.nan,

        "Pearson_p":
            np.nan,

        "Pearson_FDR_q":
            np.nan,

        "Spearman_rho":
            np.nan,

        "Spearman_p":
            np.nan,

        "Spearman_FDR_q":
            np.nan,

        "Mean_Abs_Correlation":
            np.nan,

        "Reason":
            ""
    }


    # --------------------------------------------------------
    # Non-descriptor / metadata.
    # --------------------------------------------------------

    if descriptor in non_descriptor_columns:

        record[
            "Status"
        ] = "NOT_A_CHEMICAL_DESCRIPTOR"

        record[
            "Reason"
        ] = (
            "Metadata, atom index, or QC count; "
            "not eligible for correlation ranking."
        )


    # --------------------------------------------------------
    # Missing values.
    # --------------------------------------------------------

    elif missing_n > 0:

        record[
            "Status"
        ] = "EXCLUDED_MISSING_VALUES"

        record[
            "Reason"
        ] = (
            "One or more ligands lack a valid value. "
            "Excluded from publication correlation analysis."
        )


    # --------------------------------------------------------
    # Constant columns.
    # --------------------------------------------------------

    elif n_unique <= 1:

        record[
            "Status"
        ] = "EXCLUDED_CONSTANT"

        record[
            "Reason"
        ] = (
            "No variation across ligands."
        )


    # --------------------------------------------------------
    # Retained/ranked descriptors.
    # --------------------------------------------------------

    elif descriptor in rank_lookup:

        values = rank_lookup[
            descriptor
        ]

        record[
            "Status"
        ] = "RANKED"

        record[
            "Final_Rank"
        ] = values[
            "Final_Rank"
        ]

        record[
            "Pearson_r"
        ] = values[
            "Pearson_r"
        ]

        record[
            "Pearson_p"
        ] = values[
            "Pearson_p"
        ]

        record[
            "Pearson_FDR_q"
        ] = values[
            "Pearson_FDR_q"
        ]

        record[
            "Spearman_rho"
        ] = values[
            "Spearman_rho"
        ]

        record[
            "Spearman_p"
        ] = values[
            "Spearman_p"
        ]

        record[
            "Spearman_FDR_q"
        ] = values[
            "Spearman_FDR_q"
        ]

        record[
            "Mean_Abs_Correlation"
        ] = values[
            "Mean_Abs_Correlation"
        ]

        record[
            "Reason"
        ] = (
            "Complete descriptor retained for correlation analysis."
        )


    # --------------------------------------------------------
    # Intentionally excluded / redundant descriptors.
    # --------------------------------------------------------

    elif descriptor in removed_reason_lookup:

        record[
            "Status"
        ] = "EXCLUDED_BY_POLICY"

        record[
            "Reason"
        ] = removed_reason_lookup[
            descriptor
        ]


    # --------------------------------------------------------
    # Fallback category.
    # --------------------------------------------------------

    else:

        record[
            "Status"
        ] = "NOT_RANKED"

        record[
            "Reason"
        ] = (
            "Complete numeric variable, but not included in "
            "the final correlation candidate set."
        )


    audit_records.append(
        record
    )


# ------------------------------------------------------------
# Build final audit table.
# ------------------------------------------------------------

descriptor_audit_df = pd.DataFrame(
    audit_records
)


status_order = {

    "RANKED": 1,

    "EXCLUDED_BY_POLICY": 2,

    "EXCLUDED_MISSING_VALUES": 3,

    "EXCLUDED_CONSTANT": 4,

    "NOT_RANKED": 5,

    "NOT_A_CHEMICAL_DESCRIPTOR": 6
}


descriptor_audit_df[
    "_Status_Order"
] = descriptor_audit_df[
    "Status"
].map(
    status_order
).fillna(99)


descriptor_audit_df = (
    descriptor_audit_df
    .sort_values(
        by=[
            "_Status_Order",
            "Final_Rank",
            "Descriptor"
        ]
    )
    .drop(
        columns="_Status_Order"
    )
    .reset_index(
        drop=True
    )
)


# ------------------------------------------------------------
# Display ranked descriptors and all descriptors.
# ------------------------------------------------------------

print("=" * 100)
print("RANKED DESCRIPTORS")
print("=" * 100)

display(
    descriptor_audit_df[
        descriptor_audit_df[
            "Status"
        ] == "RANKED"
    ]
)


print("=" * 100)
print("ALL GENERATED NUMERIC VARIABLES: STATUS AND RANK")
print("=" * 100)

display(
    descriptor_audit_df
)


# ------------------------------------------------------------
# Save to final Excel workbook.
# ------------------------------------------------------------

with pd.ExcelWriter(
    FINAL_EXCEL,
    engine="openpyxl",
    mode="a",
    if_sheet_exists="replace"
) as writer:

    descriptor_audit_df.to_excel(
        writer,
        sheet_name="ALL_DESCRIPTOR_RANKS",
        index=False
    )


print(
    "\nSaved status/rank audit table to:"
)

print(
    FINAL_EXCEL
)
```
