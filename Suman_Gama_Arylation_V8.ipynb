```python
# ============================================================
# Pd γ-ARYLATION
# COMPLETE LIGAND DESCRIPTOR + YIELD CORRELATION ANALYSIS
#
# LIGAND ATOMS : 17 -> LAST
# Pd ATOM       : 1
# Pd-X CUTOFF   : 2.00 - 2.40 Å
#
# PROJECT: Pd γ-arylation ONLY
# ============================================================


# ============================================================
# 1. INSTALL PYTHON PACKAGES
# ============================================================

!pip -q install morfeus-ml pandas numpy scipy openpyxl matplotlib seaborn statsmodels


# ============================================================
# 2. IMPORTS
# ============================================================

import os
import re
import glob
import shutil
import subprocess
import warnings

import numpy as np
import pandas as pd

import matplotlib.pyplot as plt
import seaborn as sns

from scipy.stats import pearsonr, spearmanr

from morfeus import (
    SASA,
    SolidAngle,
    BuriedVolume,
    Sterimol,
    BiteAngle,
    Pyramidalization,
    XTB
)

warnings.filterwarnings("ignore")


# ============================================================
# 3. GOOGLE DRIVE
# ============================================================

from google.colab import drive

drive.mount("/content/drive")


# ============================================================
# 4. USER SETTINGS
# ============================================================

XYZ_FOLDER = "/content/drive/MyDrive/Suman_files"

OUTPUT_FOLDER = (
    "/content/drive/MyDrive/"
    "Pd_gamma_arylation_RESULTS"
)

os.makedirs(
    OUTPUT_FOLDER,
    exist_ok=True
)


# ------------------------------------------------------------
# Output Excel
# ------------------------------------------------------------

FINAL_EXCEL = os.path.join(
    OUTPUT_FOLDER,
    "Pd_gamma_arylation_COMPLETE_ANALYSIS.xlsx"
)


# ============================================================
# CHEMICAL SETTINGS
# ============================================================

PD_ATOM = 1

LIGAND_START = 17

# IMPORTANT:
# EXACT coordination criterion requested
PD_DONOR_MIN = 2.00
PD_DONOR_MAX = 2.40

DONOR_ELEMENTS = {
    "N",
    "P",
    "S",
    "O"
}


# ============================================================
# MORFEUS SETTINGS
# ============================================================

SASA_PROBE_RADIUS = 1.4

# Keep only chemically distinct radii to avoid redundant parameters.
# 3.0 Å = proximal Pd pocket; 4.0 Å = near-ligand shell;
# 5.5 Å = broader/distal ligand environment.
VBUR_RADII = [
    3.0,
    4.0,
    5.5
]

DISTAL_VOLUME_RADIUS = 3.5
VBUR_N_POINTS = 100000
CALCULATE_DIRECTIONAL_VBUR = False
CALCULATE_DISPERSION = True

VBUR_RADII_TYPE = "bondi"

SOLID_ANGLE_RADII_TYPE = "bondi"

STERIMOL_RADII_TYPE = "crc"


# ============================================================
# xTB SETTINGS
# ============================================================

XTB_METHOD = 2                 # GFN2-xTB
XTB_CHARGE = 0
XTB_N_UNPAIRED = 0
XTB_PROCESSES = 2


# ============================================================
# OUTLIER SETTINGS
# ============================================================

OUTLIER_FRACTION_LIMIT = 0.10

IQR_MULTIPLIER = 1.5

MAD_Z_LIMIT = 3.5


# ============================================================
# HIGH CORRELATION THRESHOLD
# ============================================================

HIGH_CORR_THRESHOLD = 0.80


# ============================================================
# 5. YIELD DATA
# ============================================================

YIELD_DATA = {

    "L7": 0,
    "L8": 0,
    "L9": 0,
    "L10": 0,
    "L11": 0,
    "L12": 0,
    "L13": 0,
    "L14": 0,
    "L15": 0,

    "L16": 12,
    "L17": 53,
    "L18": 46,
    "L19": 67,
    "L20": 40,
    "L21": 31,

    "L22": 0,
    "L23": 0,
    "L24": 0,
    "L25": 0
}


# ============================================================
# 6. CREATE PLOT FOLDERS
# ============================================================

PLOT_FOLDER = os.path.join(
    OUTPUT_FOLDER,
    "PLOTS"
)

SCATTER_FOLDER = os.path.join(
    PLOT_FOLDER,
    "Yield_vs_Descriptors"
)

BOX_FOLDER = os.path.join(
    PLOT_FOLDER,
    "Descriptor_Boxplots"
)

os.makedirs(PLOT_FOLDER, exist_ok=True)
os.makedirs(SCATTER_FOLDER, exist_ok=True)
os.makedirs(BOX_FOLDER, exist_ok=True)


# ============================================================
# 7. xTB CHECK / INSTALL
# ============================================================

print("=" * 80)
print("CHECKING xTB")
print("=" * 80)

xtb_path = shutil.which("xtb")

if xtb_path is None:

    print("\n✗ xTB executable not found.")
    print("Installing xTB binary...\n")

    xtb_archive = "/content/xtb.tar.xz"

    subprocess.run(
        [
            "wget",
            "-q",
            "--show-progress",
            "https://github.com/grimme-lab/xtb/releases/latest/download/xtb-6.7.1-linux-x86_64.tar.xz",
            "-O",
            xtb_archive
        ],
        check=True
    )

    subprocess.run(
        [
            "tar",
            "-xf",
            xtb_archive,
            "-C",
            "/content"
        ],
        check=True
    )

    xtb_dirs = sorted([
        d
        for d in os.listdir("/content")
        if d.startswith("xtb-")
        and os.path.isdir(
            os.path.join("/content", d)
        )
    ])

    if not xtb_dirs:

        raise RuntimeError(
            "xTB archive extracted but "
            "xTB directory was not found."
        )

    xtb_dir = os.path.join(
        "/content",
        xtb_dirs[-1]
    )

    xtb_bin = os.path.join(
        xtb_dir,
        "bin"
    )

    xtb_executable = os.path.join(
        xtb_bin,
        "xtb"
    )

    if not os.path.exists(
        xtb_executable
    ):

        raise RuntimeError(
            f"xTB executable not found:\n"
            f"{xtb_executable}"
        )

    os.environ["PATH"] = (
        xtb_bin
        + ":"
        + os.environ["PATH"]
    )

    os.environ["XTBHOME"] = xtb_dir

    print("\n✓ xTB installed")
    print(
        "Executable:",
        xtb_executable
    )


xtb_path = shutil.which("xtb")

if xtb_path is None:

    raise RuntimeError(
        "xTB is still unavailable."
    )


print("\n" + "=" * 80)
print("xTB VERIFICATION")
print("=" * 80)

print(
    "xTB executable:",
    xtb_path
)

version = subprocess.run(
    ["xtb", "--version"],
    capture_output=True,
    text=True
)

print(version.stdout)

print("=" * 80)
print("✓ xTB READY")
print("=" * 80)


# ============================================================
# 8. HELPER FUNCTIONS
# ============================================================

def finite(value):

    try:

        value = float(value)

        if np.isfinite(value):
            return value

    except Exception:
        pass

    return np.nan


def distance(a, b):

    return float(
        np.linalg.norm(
            np.asarray(a)
            -
            np.asarray(b)
        )
    )


def angle_deg(a, b, c):

    v1 = (
        np.asarray(a)
        -
        np.asarray(b)
    )

    v2 = (
        np.asarray(c)
        -
        np.asarray(b)
    )

    n1 = np.linalg.norm(v1)
    n2 = np.linalg.norm(v2)

    if n1 == 0 or n2 == 0:
        return np.nan

    cosine = np.dot(v1, v2) / (
        n1 * n2
    )

    cosine = np.clip(
        cosine,
        -1.0,
        1.0
    )

    return float(
        np.degrees(
            np.arccos(cosine)
        )
    )


def clean_ligand_name(filename):

    base = os.path.basename(filename)

    base = re.sub(
        r"\.xyz$",
        "",
        base,
        flags=re.IGNORECASE
    )

    base = re.sub(
        r"[_-](new|opt|optimized)$",
        "",
        base,
        flags=re.IGNORECASE
    )

    match = re.search(
        r"L\d+",
        base,
        flags=re.IGNORECASE
    )

    if match:
        return match.group(0).upper()

    return base


def safe_filename(text):

    return re.sub(
        r"[^A-Za-z0-9_.-]+",
        "_",
        str(text)
    )


# ============================================================
# 9. XYZ READER
# ============================================================

def read_xyz_file(filepath):

    with open(
        filepath,
        "r",
        encoding="utf-8"
    ) as f:

        lines = [
            x.strip()
            for x in f
            if x.strip()
        ]

    if len(lines) < 3:

        raise ValueError(
            "XYZ file too short."
        )

    n_atoms = int(
        lines[0].split()[0]
    )

    atom_lines = lines[2:]

    if len(atom_lines) < n_atoms:

        raise ValueError(
            "Number of coordinate lines "
            "is smaller than atom count."
        )

    elements = []
    coordinates = []

    for line in atom_lines[:n_atoms]:

        parts = line.split()

        elements.append(
            parts[0]
        )

        coordinates.append([
            float(parts[1]),
            float(parts[2]),
            float(parts[3])
        ])

    coordinates = np.asarray(
        coordinates,
        dtype=float
    )

    return elements, coordinates


# ============================================================
# 10. FIND XYZ FILES
# ============================================================

xyz_files = sorted(
    glob.glob(
        os.path.join(
            XYZ_FOLDER,
            "*.xyz"
        )
    )
)

print("\n" + "=" * 80)
print("XYZ FILES")
print("=" * 80)

print(
    f"Found {len(xyz_files)} XYZ files."
)

if len(xyz_files) == 0:

    raise FileNotFoundError(
        f"No XYZ files found in:\n"
        f"{XYZ_FOLDER}"
    )

for f in xyz_files:
    print(
        " ",
        os.path.basename(f)
    )


# ============================================================
# 11. COVALENT RADII
# ============================================================

COVALENT_RADII = {

    "H": 0.31,
    "C": 0.76,
    "N": 0.71,
    "O": 0.66,
    "F": 0.57,

    "P": 1.07,
    "S": 1.05,

    "Cl": 1.02,
    "Br": 1.20,
    "I": 1.39,

    "Pd": 1.39
}


# ============================================================
# 12. COVALENT NEIGHBORS
# ============================================================

def covalent_neighbors(
    elements,
    coordinates,
    atom_id,
    allowed_indices=None,
    scale=1.25
):

    if allowed_indices is None:

        allowed_indices = range(
            1,
            len(elements) + 1
        )

    atom_element = elements[
        atom_id - 1
    ]

    if atom_element not in COVALENT_RADII:

        return []

    r1 = COVALENT_RADII[
        atom_element
    ]

    result = []

    for j in allowed_indices:

        if j == atom_id:
            continue

        element_j = elements[
            j - 1
        ]

        if element_j not in COVALENT_RADII:
            continue

        r2 = COVALENT_RADII[
            element_j
        ]

        d = distance(
            coordinates[atom_id - 1],
            coordinates[j - 1]
        )

        cutoff = scale * (
            r1 + r2
        )

        if d <= cutoff:

            result.append(
                (j, d)
            )

    return sorted(
        result,
        key=lambda x: x[1]
    )


# ============================================================
# 13. Pd DONOR IDENTIFICATION
#
# ONLY ligand atoms 17 -> LAST
# ONLY N/P/S/O
# ONLY 2.00 -> 2.40 Å
# ============================================================

def find_pd_donors(
    elements,
    coordinates
):

    pd_xyz = coordinates[
        PD_ATOM - 1
    ]

    candidates = []

    for atom_id in range(
        LIGAND_START,
        len(elements) + 1
    ):

        element = elements[
            atom_id - 1
        ]

        if element not in DONOR_ELEMENTS:
            continue

        d = distance(
            pd_xyz,
            coordinates[
                atom_id - 1
            ]
        )

        candidates.append({
            "Atom": atom_id,
            "Element": element,
            "Pd_Distance_A": d
        })

    qualifying = [

        x for x in candidates

        if (
            PD_DONOR_MIN
            <=
            x["Pd_Distance_A"]
            <=
            PD_DONOR_MAX
        )
    ]

    qualifying = sorted(
        qualifying,
        key=lambda x:
        x["Pd_Distance_A"]
    )

    return candidates, qualifying


# ============================================================
# 14. BUILD LIGAND GRAPH
# ============================================================

def build_ligand_graph(
    elements,
    coordinates
):

    ligand_indices = list(
        range(
            LIGAND_START,
            len(elements) + 1
        )
    )

    graph = {
        i: set()
        for i in ligand_indices
    }

    for i in ligand_indices:

        neighbors = covalent_neighbors(
            elements,
            coordinates,
            i,
            allowed_indices=ligand_indices
        )

        for j, d in neighbors:

            graph[i].add(j)
            graph[j].add(i)

    return graph


# ============================================================
# 15. SHORTEST DONOR-DONOR LIGAND PATH
# ============================================================

def shortest_ligand_path(
    graph,
    start,
    end
):

    if start is None or end is None:
        return None

    queue = [
        (
            start,
            [start]
        )
    ]

    visited = {
        start
    }

    while queue:

        current, path = queue.pop(0)

        if current == end:
            return path

        for neighbor in sorted(
            graph.get(current, [])
        ):

            if neighbor not in visited:

                visited.add(
                    neighbor
                )

                queue.append(
                    (
                        neighbor,
                        path + [neighbor]
                    )
                )

    return None


# ============================================================
# 16. PALLADACYCLE CLASSIFICATION
#
# PROJECT DEFINITION:
#
# L7-L15  : Pd-N-C-C-N-Pd
#            ligand path contains 4 atoms
#            => 4-membered project class
#
# L16-L25 : Pd-N-C-C-C-N-Pd
#            ligand path contains 5 atoms
#            => 5-membered project class
# ============================================================

def determine_palladacycle(
    ligand_name,
    elements,
    coordinates,
    donor1,
    donor2
):

    result = {

        "Palladacycle_Type": None,

        "Palladacycle_Ring_Size": np.nan,

        "Palladacycle_Ligand_Path": None,

        "Palladacycle_Path_Bond_Mean_A": np.nan,

        "Palladacycle_Path_Bond_SD_A": np.nan,

        "Palladacycle_Path_Bond_CV": np.nan,

        "Expected_Palladacycle_Type": None,

        "Palladacycle_Type_Check": None
    }

    # --------------------------------------------------------
    # Expected project classification
    # --------------------------------------------------------

    match = re.search(
        r"L(\d+)",
        ligand_name,
        flags=re.IGNORECASE
    )

    if match:

        number = int(
            match.group(1)
        )

        if 7 <= number <= 15:

            result[
                "Expected_Palladacycle_Type"
            ] = "4-membered"

        elif 16 <= number <= 25:

            result[
                "Expected_Palladacycle_Type"
            ] = "5-membered"


    if donor1 is None or donor2 is None:

        result[
            "Palladacycle_Type"
        ] = "NOT_AVAILABLE"

        return result


    graph = build_ligand_graph(
        elements,
        coordinates
    )

    path = shortest_ligand_path(
        graph,
        donor1,
        donor2
    )

    if path is None:

        result[
            "Palladacycle_Type"
        ] = "NOT_FOUND"

        return result


    result[
        "Palladacycle_Ligand_Path"
    ] = "-".join(
        map(str, path)
    )

    path_count = len(path)


    # --------------------------------------------------------
    # IMPORTANT:
    # Project ring-size convention excludes Pd.
    # --------------------------------------------------------

    if path_count == 4:

        result[
            "Palladacycle_Ring_Size"
        ] = 4

        result[
            "Palladacycle_Type"
        ] = "4-membered"

    elif path_count == 5:

        result[
            "Palladacycle_Ring_Size"
        ] = 5

        result[
            "Palladacycle_Type"
        ] = "5-membered"

    else:

        result[
            "Palladacycle_Type"
        ] = f"OTHER_{path_count}"

    # --------------------------------------------------------
    # Path bond geometry
    # --------------------------------------------------------

    bond_lengths = []

    for i in range(
        len(path) - 1
    ):

        a = path[i]
        b = path[i + 1]

        d = distance(
            coordinates[a - 1],
            coordinates[b - 1]
        )

        bond_lengths.append(d)


    if bond_lengths:

        mean_bond = np.mean(
            bond_lengths
        )

        sd_bond = np.std(
            bond_lengths,
            ddof=0
        )

        result[
            "Palladacycle_Path_Bond_Mean_A"
        ] = mean_bond

        result[
            "Palladacycle_Path_Bond_SD_A"
        ] = sd_bond

        if mean_bond != 0:

            result[
                "Palladacycle_Path_Bond_CV"
            ] = (
                sd_bond /
                mean_bond
            )


    expected = result[
        "Expected_Palladacycle_Type"
    ]

    actual = result[
        "Palladacycle_Type"
    ]

    if expected is None:

        result[
            "Palladacycle_Type_Check"
        ] = "NO_EXPECTATION"

    elif expected == actual:

        result[
            "Palladacycle_Type_Check"
        ] = "PASS"

    else:

        result[
            "Palladacycle_Type_Check"
        ] = "CHECK"


    return result


# ============================================================
# 17. STERIC DESCRIPTORS
# ============================================================

def calculate_steric(
    elements,
    coordinates,
    donor1
):

    result = {}

    ligand_indices = list(
        range(
            LIGAND_START,
            len(elements) + 1
        )
    )

    ligand_elements = [
        elements[i - 1]
        for i in ligand_indices
    ]

    ligand_coordinates = coordinates[
        np.array(ligand_indices) - 1
    ]

    # --------------------------------------------------------
    # Maximum ligand extension
    # --------------------------------------------------------

    max_extension = 0.0

    for i in range(
        len(ligand_coordinates)
    ):

        for j in range(
            i + 1,
            len(ligand_coordinates)
        ):

            d = distance(
                ligand_coordinates[i],
                ligand_coordinates[j]
            )

            max_extension = max(
                max_extension,
                d
            )

    result[
        "Ligand_Maximum_Extension_A"
    ] = finite(
        max_extension
    )


    # --------------------------------------------------------
    # SASA
    # --------------------------------------------------------

    try:

        sasa = SASA(
            ligand_elements,
            ligand_coordinates,
            probe_radius=SASA_PROBE_RADIUS
        )

        result[
            "Ligand_SASA_A2"
        ] = finite(
            sasa.area
        )

        result[
            "Ligand_SASA_Volume_A3"
        ] = finite(
            sasa.volume
        )

    except Exception as e:

        result[
            "SASA_Warning"
        ] = str(e)


    # --------------------------------------------------------
    # Solid angle
    #
    # ligand-only:
    # reduced system = Pd + ligand
    # --------------------------------------------------------

    try:

        reduced_elements = (
            ["Pd"]
            +
            ligand_elements
        )

        reduced_coordinates = np.vstack([
            coordinates[
                PD_ATOM - 1
            ],
            ligand_coordinates
        ])

        sa = SolidAngle(
            reduced_elements,
            reduced_coordinates,
            metal_index=1,
            radii_type=SOLID_ANGLE_RADII_TYPE
        )

        result[
            "Ligand_Solid_Angle_sr"
        ] = finite(
            sa.solid_angle
        )

        result[
            "Ligand_Solid_Cone_Angle_deg"
        ] = finite(
            sa.cone_angle
        )

        result[
            "Ligand_Solid_G_percent"
        ] = finite(
            sa.G
        )

    except Exception as e:

        result[
            "SolidAngle_Warning"
        ] = str(e)


    # --------------------------------------------------------
    # Buried volume and nonredundant distal sterics
    # --------------------------------------------------------

    excluded_atoms = [

        i
        for i in range(
            1,
            len(elements) + 1
        )

        if (
            i != PD_ATOM
            and
            i not in ligand_indices
        )
    ]


    for radius in VBUR_RADII:

        key = f"Ligand_%Vbur_{radius:.1f}A"

        try:

            bv = BuriedVolume(
                elements,
                coordinates,
                metal_index=PD_ATOM,
                excluded_atoms=excluded_atoms,
                radius=radius,
                include_hs=False,
                radii_type=VBUR_RADII_TYPE,
                n_points=VBUR_N_POINTS
            )

            result[key] = finite(
                bv.fraction_buried_volume * 100.0
            )

        except Exception as e:

            result[key] = np.nan
            result[f"{key}_Warning"] = str(e)


    # --------------------------------------------------------
    # Distal volume outside a single proximal sphere
    # --------------------------------------------------------

    try:

        bv_distal = BuriedVolume(
            elements,
            coordinates,
            metal_index=PD_ATOM,
            excluded_atoms=excluded_atoms,
            radius=DISTAL_VOLUME_RADIUS,
            include_hs=False,
            radii_type=VBUR_RADII_TYPE,
            n_points=VBUR_N_POINTS
        )

        bv_distal.compute_distal_volume(method="sasa")

        distal = finite(bv_distal.distal_volume)
        molecular = finite(bv_distal.molecular_volume)

        result[
            f"Ligand_Distal_Volume_Outside_{DISTAL_VOLUME_RADIUS:.1f}A_A3"
        ] = distal

        result["Ligand_Molecular_Volume_A3"] = molecular

        result[
            f"Ligand_Distal_Volume_Fraction_Outside_{DISTAL_VOLUME_RADIUS:.1f}A"
        ] = finite(distal / molecular) if (
            pd.notna(distal)
            and pd.notna(molecular)
            and molecular > 0
        ) else np.nan

    except Exception as e:

        result["Distal_Volume_Warning"] = str(e)
        result[
            f"Ligand_Distal_Volume_Outside_{DISTAL_VOLUME_RADIUS:.1f}A_A3"
        ] = np.nan
        result["Ligand_Molecular_Volume_A3"] = np.nan
        result[
            f"Ligand_Distal_Volume_Fraction_Outside_{DISTAL_VOLUME_RADIUS:.1f}A"
        ] = np.nan


    # --------------------------------------------------------
    # Optional directional sterics
    #
    # Disabled by default because quadrant/octet labels require
    # identical chemical orientation across every XYZ structure.
    # --------------------------------------------------------

    if CALCULATE_DIRECTIONAL_VBUR:

        try:

            bv_directional = BuriedVolume(
                elements,
                coordinates,
                metal_index=PD_ATOM,
                excluded_atoms=excluded_atoms,
                radius=DISTAL_VOLUME_RADIUS,
                include_hs=False,
                radii_type=VBUR_RADII_TYPE,
                n_points=VBUR_N_POINTS
            )

            bv_directional.octant_analysis()

            q_values = [
                finite(value * 100.0)
                for value in bv_directional.quadrants.values()
            ]

            o_values = [
                finite(value * 100.0)
                for value in bv_directional.octants.values()
            ]

            q_values = [x for x in q_values if pd.notna(x)]
            o_values = [x for x in o_values if pd.notna(x)]

            if q_values:

                result["Vbur_3.5A_Quadrant_Range_percent"] = finite(
                    max(q_values) - min(q_values)
                )

                result["Vbur_3.5A_Quadrant_SD_percent"] = finite(
                    np.std(q_values, ddof=0)
                )

            if o_values:

                result["Vbur_3.5A_Octant_Range_percent"] = finite(
                    max(o_values) - min(o_values)
                )

                result["Vbur_3.5A_Octant_SD_percent"] = finite(
                    np.std(o_values, ddof=0)
                )

        except Exception as e:

            result["Directional_Vbur_Warning"] = str(e)


    # --------------------------------------------------------
    # Dispersion descriptor, if supported by this Morfeus build
    # --------------------------------------------------------

    if CALCULATE_DISPERSION:

        try:

            from morfeus import Dispersion

            disp = Dispersion(
                ligand_elements,
                ligand_coordinates
            )

            pint = finite(disp.p_int)
            sasa_value = result.get("Ligand_SASA_A2", np.nan)
            volume_value = result.get("Ligand_Molecular_Volume_A3", np.nan)

            result["Ligand_Pint"] = pint
            result["Ligand_Pint_per_SASA"] = finite(
                pint / sasa_value
            ) if pd.notna(sasa_value) and sasa_value > 0 else np.nan
            result["Ligand_Pint_per_Volume"] = finite(
                pint / volume_value
            ) if pd.notna(volume_value) and volume_value > 0 else np.nan

        except Exception as e:

            result["Dispersion_Warning"] = str(e)
            result["Ligand_Pint"] = np.nan
            result["Ligand_Pint_per_SASA"] = np.nan
            result["Ligand_Pint_per_Volume"] = np.nan


    # --------------------------------------------------------
    # Pd-oriented custom Sterimol
    #
    # IMPORTANT:
    # This is NOT standard H-dummy Sterimol.
    # It is a Pd-oriented custom descriptor.
    # --------------------------------------------------------

    if donor1 is not None:

        try:

            reduced_elements = (
                ["Pd"]
                +
                ligand_elements
            )

            reduced_coordinates = np.vstack([
                coordinates[
                    PD_ATOM - 1
                ],
                ligand_coordinates
            ])

            original_to_reduced = {
                original: i + 2
                for i, original
                in enumerate(
                    ligand_indices
                )
            }

            attached_index = (
                original_to_reduced[
                    donor1
                ]
            )

            ster = Sterimol(
                reduced_elements,
                reduced_coordinates,
                dummy_index=1,
                attached_index=attached_index,
                radii_type=STERIMOL_RADII_TYPE
            )

            result[
                "Pd_Oriented_Sterimol_B1_A"
            ] = finite(
                ster.B_1_value
            )

            result[
                "Pd_Oriented_Sterimol_B5_A"
            ] = finite(
                ster.B_5_value
            )

            result[
                "Pd_Oriented_Sterimol_L_A"
            ] = finite(
                ster.L_value
            )

        except Exception as e:

            result[
                "Sterimol_Warning"
            ] = str(e)


    return result


# ============================================================
# 18. COORDINATION DESCRIPTORS
# ============================================================

def calculate_coordination(
    elements,
    coordinates,
    donor1,
    donor2
):

    result = {}

    if donor1 is None:

        return {
            "Pd_Donor1_Distance_A": np.nan,
            "Pd_Donor2_Distance_A": np.nan,
            "Average_Pd_Donor_Distance_A": np.nan,
            "Delta_Pd_Donor_Distance_A": np.nan,
            "Pd_Donor_Distance_Ratio": np.nan,
            "Bite_Angle_deg": np.nan,
            "Donor1_Donor2_Distance_A": np.nan,
            "Pd_Coordination_Distortion_A": np.nan
        }


    pd_xyz = coordinates[
        PD_ATOM - 1
    ]


    d1 = distance(
        pd_xyz,
        coordinates[
            donor1 - 1
        ]
    )


    result[
        "Pd_Donor1_Distance_A"
    ] = d1


    if donor2 is not None:

        d2 = distance(
            pd_xyz,
            coordinates[
                donor2 - 1
            ]
        )

        result[
            "Pd_Donor2_Distance_A"
        ] = d2

        avg = (
            d1 + d2
        ) / 2.0

        delta = abs(
            d1 - d2
        )

        result[
            "Average_Pd_Donor_Distance_A"
        ] = avg

        result[
            "Delta_Pd_Donor_Distance_A"
        ] = delta

        if min(d1, d2) > 0:

            result[
                "Pd_Donor_Distance_Ratio"
            ] = (
                max(d1, d2)
                /
                min(d1, d2)
            )


        result[
            "Pd_Coordination_Distortion_A"
        ] = np.sqrt(
            (
                (d1 - avg) ** 2
                +
                (d2 - avg) ** 2
            ) / 2.0
        )


        result[
            "Donor1_Donor2_Distance_A"
        ] = distance(
            coordinates[
                donor1 - 1
            ],
            coordinates[
                donor2 - 1
            ]
        )

        try:

            bite = BiteAngle(
                coordinates,
                PD_ATOM,
                donor1,
                donor2
            )

            result[
                "Bite_Angle_deg"
            ] = finite(
                bite.angle
            )

        except Exception as e:

            result[
                "Bite_Angle_deg"
            ] = np.nan

            result[
                "Bite_Angle_Warning"
            ] = str(e)

    else:

        result[
            "Pd_Donor2_Distance_A"
        ] = np.nan

        result[
            "Average_Pd_Donor_Distance_A"
        ] = np.nan

        result[
            "Delta_Pd_Donor_Distance_A"
        ] = np.nan

        result[
            "Pd_Donor_Distance_Ratio"
        ] = np.nan

        result[
            "Pd_Coordination_Distortion_A"
        ] = np.nan

        result[
            "Donor1_Donor2_Distance_A"
        ] = np.nan

        result[
            "Bite_Angle_deg"
        ] = np.nan


    return result


# ============================================================
# 19. PYRAMIDALIZATION
# ============================================================

def calculate_pyramidalization(
    elements,
    coordinates,
    donor1,
    donor2
):

    result = {}

    ligand_indices = list(
        range(
            LIGAND_START,
            len(elements) + 1
        )
    )


    def calculate_one(
        donor,
        prefix
    ):

        if donor is None:

            return {

                f"{prefix}_Neighbor_Count":
                    np.nan,

                f"{prefix}_P":
                    np.nan,

                f"{prefix}_P_Angle_deg":
                    np.nan,

                f"{prefix}_P_Warning":
                    "Donor unavailable."
            }


        neighbors = covalent_neighbors(
            elements,
            coordinates,
            donor,
            allowed_indices=ligand_indices
        )

        neighbor_ids = [
            x[0]
            for x in neighbors
        ]

        out = {

            f"{prefix}_Neighbor_Count":
                len(neighbor_ids)
        }


        # ----------------------------------------------------
        # Need at least 3 neighbors
        # ----------------------------------------------------

        if len(neighbor_ids) < 3:

            out[
                f"{prefix}_P"
            ] = np.nan

            out[
                f"{prefix}_P_Angle_deg"
            ] = np.nan

            out[
                f"{prefix}_P_Warning"
            ] = (
                f"{len(neighbor_ids)} "
                "covalent neighbors; "
                "pyramidalization not safely defined."
            )

            return out


        selected = neighbor_ids[:3]


        try:

            pyr = Pyramidalization(
                coordinates,
                donor,
                neighbor_indices=selected
            )

            out[
                f"{prefix}_P"
            ] = finite(
                pyr.P
            )

            out[
                f"{prefix}_P_Angle_deg"
            ] = finite(
                pyr.P_angle
            )

        except Exception as e:

            out[
                f"{prefix}_P"
            ] = np.nan

            out[
                f"{prefix}_P_Angle_deg"
            ] = np.nan

            out[
                f"{prefix}_P_Warning"
            ] = str(e)


        return out


    d1 = calculate_one(
        donor1,
        "Donor1_Pyramidalization"
    )

    d2 = calculate_one(
        donor2,
        "Donor2_Pyramidalization"
    )

    result.update(d1)
    result.update(d2)


    # --------------------------------------------------------
    # Average / delta P
    # --------------------------------------------------------

    p1 = result.get(
        "Donor1_Pyramidalization_P",
        np.nan
    )

    p2 = result.get(
        "Donor2_Pyramidalization_P",
        np.nan
    )

    if (
        pd.notna(p1)
        and
        pd.notna(p2)
    ):

        result[
            "Average_Pyramidalization_P"
        ] = (
            p1 + p2
        ) / 2.0

        result[
            "Delta_Pyramidalization_P"
        ] = abs(
            p1 - p2
        )

    else:

        result[
            "Average_Pyramidalization_P"
        ] = np.nan

        result[
            "Delta_Pyramidalization_P"
        ] = np.nan


    # --------------------------------------------------------
    # Average / delta P angle
    # --------------------------------------------------------

    a1 = result.get(
        "Donor1_Pyramidalization_P_Angle_deg",
        np.nan
    )

    a2 = result.get(
        "Donor2_Pyramidalization_P_Angle_deg",
        np.nan
    )

    if (
        pd.notna(a1)
        and
        pd.notna(a2)
    ):

        result[
            "Average_Pyramidalization_Angle_deg"
        ] = (
            a1 + a2
        ) / 2.0

        result[
            "Delta_Pyramidalization_Angle_deg"
        ] = abs(
            a1 - a2
        )

    else:

        result[
            "Average_Pyramidalization_Angle_deg"
        ] = np.nan

        result[
            "Delta_Pyramidalization_Angle_deg"
        ] = np.nan


    return result


# ============================================================
# 20. xTB ELECTRONIC DESCRIPTORS
#
# IMPORTANT:
# ONLY ligand atoms 17 -> LAST
#
# Pd is NOT included in isolated ligand xTB.
# ============================================================

def calculate_xtb(
    elements,
    coordinates,
    donor1,
    donor2
):

    result = {}

    ligand_indices = list(
        range(
            LIGAND_START,
            len(elements) + 1
        )
    )

    ligand_elements = [
        elements[i - 1]
        for i in ligand_indices
    ]

    ligand_coordinates = coordinates[
        np.array(ligand_indices) - 1
    ]


    try:

        xtb = XTB(
            ligand_elements,
            ligand_coordinates,
            method=XTB_METHOD,
            charge=XTB_CHARGE,
            n_unpaired=XTB_N_UNPAIRED,
            n_processes=XTB_PROCESSES
        )


        # ----------------------------------------------------
        # Charges
        # ----------------------------------------------------

        charges = xtb.get_charges(
            model="Mulliken"
        )

        result[
            "XTB_Ligand_Charge_Sum"
        ] = finite(
            np.sum(charges)
        )


        # ----------------------------------------------------
        # HOMO / LUMO
        # ----------------------------------------------------

        result[
            "XTB_HOMO_eV"
        ] = finite(
            xtb.get_homo(
                unit="eV"
            )
        )

        result[
            "XTB_LUMO_eV"
        ] = finite(
            xtb.get_lumo(
                unit="eV"
            )
        )

        result[
            "XTB_HOMO_LUMO_Gap_eV"
        ] = finite(
            xtb.get_homo_lumo_gap()
        )


        # ----------------------------------------------------
        # Fermi
        # ----------------------------------------------------

        result[
            "XTB_Fermi_Level_eV"
        ] = finite(
            xtb.get_fermi_level()
        )


        # ----------------------------------------------------
        # IP / EA
        # ----------------------------------------------------

        result[
            "XTB_IP_eV"
        ] = finite(
            xtb.get_ip(
                corrected=True
            )
        )

        result[
            "XTB_EA_eV"
        ] = finite(
            xtb.get_ea(
                corrected=True
            )
        )


        # ----------------------------------------------------
        # Chemical potential
        # ----------------------------------------------------

        result[
            "XTB_Chemical_Potential_eV"
        ] = finite(
            xtb.get_chemical_potential(
                corrected=True
            )
        )


        # ----------------------------------------------------
        # Electronegativity
        # ----------------------------------------------------

        result[
            "XTB_Electronegativity_eV"
        ] = finite(
            xtb.get_electronegativity(
                corrected=True
            )
        )


        # ----------------------------------------------------
        # Hardness / softness
        # ----------------------------------------------------

        result[
            "XTB_Hardness_eV"
        ] = finite(
            xtb.get_hardness()
        )

        result[
            "XTB_Softness"
        ] = finite(
            xtb.get_softness()
        )


        # ----------------------------------------------------
        # Global electrophilicity / nucleophilicity
        # ----------------------------------------------------

        try:

            result[
                "XTB_Electrophilicity"
            ] = finite(
                xtb.get_global_descriptor(
                    "electrophilicity",
                    corrected=True
                )
            )

        except Exception:

            result[
                "XTB_Electrophilicity"
            ] = np.nan


        try:

            result[
                "XTB_Nucleophilicity"
            ] = finite(
                xtb.get_global_descriptor(
                    "nucleophilicity",
                    corrected=True
                )
            )

        except Exception:

            result[
                "XTB_Nucleophilicity"
            ] = np.nan


        # ----------------------------------------------------
        # Dipole
        # ----------------------------------------------------

        result[
            "XTB_Dipole_Debye"
        ] = finite(
            xtb.get_dipole_moment(
                unit="debye"
            )
        )


        # ----------------------------------------------------
        # Polarizability
        # ----------------------------------------------------

        try:

            result[
                "XTB_Polarizability"
            ] = finite(
                xtb.get_molecular_polarizability()
            )

        except Exception:

            result[
                "XTB_Polarizability"
            ] = np.nan


        # ----------------------------------------------------
        # NFOD
        # ----------------------------------------------------

        try:

            result[
                "XTB_NFOD"
            ] = finite(
                xtb.get_nfod()
            )

        except Exception:

            result[
                "XTB_NFOD"
            ] = np.nan


        # ----------------------------------------------------
        # Donor charges
        #
        # Original atom ID -> isolated ligand index
        # ----------------------------------------------------

        original_to_xtb = {
            original: i
            for i, original
            in enumerate(
                ligand_indices
            )
        }

        if donor1 is not None:

            idx1 = original_to_xtb[
                donor1
            ]

            result[
                "Donor1_Mulliken_Charge"
            ] = finite(
                charges[idx1]
            )

        else:

            result[
                "Donor1_Mulliken_Charge"
            ] = np.nan


        if donor2 is not None:

            idx2 = original_to_xtb[
                donor2
            ]

            result[
                "Donor2_Mulliken_Charge"
            ] = finite(
                charges[idx2]
            )

        else:

            result[
                "Donor2_Mulliken_Charge"
            ] = np.nan


        # ----------------------------------------------------
        # Donor charge summary
        # ----------------------------------------------------

        q1 = result.get(
            "Donor1_Mulliken_Charge",
            np.nan
        )

        q2 = result.get(
            "Donor2_Mulliken_Charge",
            np.nan
        )

        if (
            pd.notna(q1)
            and
            pd.notna(q2)
        ):

            result[
                "Average_Donor_Charge"
            ] = (
                q1 + q2
            ) / 2.0

            result[
                "Delta_Donor_Charge"
            ] = abs(
                q1 - q2
            )

            result[
                "Donor_Charge_Sum"
            ] = q1 + q2

        else:

            result[
                "Average_Donor_Charge"
            ] = np.nan

            result[
                "Delta_Donor_Charge"
            ] = np.nan

            result[
                "Donor_Charge_Sum"
            ] = np.nan


    except Exception as e:

        result[
            "XTB_Warning"
        ] = str(e)


    return result


# ============================================================
# 22. PROCESS ONE XYZ
# ============================================================

def process_xyz(filepath):

    ligand_name = clean_ligand_name(
        filepath
    )

    elements, coordinates = (
        read_xyz_file(filepath)
    )


    row = {

        "Ligand": ligand_name,

        "XYZ_File":
            os.path.basename(filepath),

        "N_Atoms":
            len(elements),

        "Pd_Atom":
            PD_ATOM,

        "Ligand_Start_Atom":
            LIGAND_START,

        "Pd_Donor_Min_A":
            PD_DONOR_MIN,

        "Pd_Donor_Max_A":
            PD_DONOR_MAX
    }


    # ========================================================
    # DONORS
    # ========================================================

    candidates, qualifying = (
        find_pd_donors(
            elements,
            coordinates
        )
    )


    row[
        "Number_of_Donor_Candidates"
    ] = len(candidates)

    row[
        "Number_of_Qualifying_Pd_Donors"
    ] = len(qualifying)


    # --------------------------------------------------------
    # Donor 1
    # --------------------------------------------------------

    if len(qualifying) >= 1:

        donor1 = qualifying[0]["Atom"]

        row[
            "Donor1_Atom"
        ] = donor1

        row[
            "Donor1_Element"
        ] = qualifying[0]["Element"]

        row[
            "Pd_Donor1_Distance_A"
        ] = qualifying[0][
            "Pd_Distance_A"
        ]

    else:

        donor1 = None

        row[
            "Donor1_Atom"
        ] = np.nan

        row[
            "Donor1_Element"
        ] = None

        row[
            "Pd_Donor1_Distance_A"
        ] = np.nan


    # --------------------------------------------------------
    # Donor 2
    # --------------------------------------------------------

    if len(qualifying) >= 2:

        donor2 = qualifying[1]["Atom"]

        row[
            "Donor2_Atom"
        ] = donor2

        row[
            "Donor2_Element"
        ] = qualifying[1]["Element"]

        row[
            "Pd_Donor2_Distance_A"
        ] = qualifying[1][
            "Pd_Distance_A"
        ]

    else:

        donor2 = None

        row[
            "Donor2_Atom"
        ] = np.nan

        row[
            "Donor2_Element"
        ] = None

        row[
            "Pd_Donor2_Distance_A"
        ] = np.nan


    # --------------------------------------------------------
    # Donor selection QC
    # --------------------------------------------------------

    if len(qualifying) == 2:

        row[
            "Donor_Selection_Check"
        ] = "PASS"

    elif len(qualifying) > 2:

        row[
            "Donor_Selection_Check"
        ] = (
            "CHECK_MULTIPLE_QUALIFYING_DONORS"
        )

    elif len(qualifying) == 1:

        row[
            "Donor_Selection_Check"
        ] = "ONLY_ONE_QUALIFYING_DONOR"

    else:

        row[
            "Donor_Selection_Check"
        ] = "NO_QUALIFYING_DONORS"


    # ========================================================
    # PALLADACYCLE
    # ========================================================

    palladacycle = (
        determine_palladacycle(
            ligand_name,
            elements,
            coordinates,
            donor1,
            donor2
        )
    )

    row.update(
        palladacycle
    )


    # ========================================================
    # STERIC
    # ========================================================

    steric = calculate_steric(
        elements,
        coordinates,
        donor1
    )

    row.update(
        steric
    )


    # ========================================================
    # COORDINATION
    # ========================================================

    coordination = (
        calculate_coordination(
            elements,
            coordinates,
            donor1,
            donor2
        )
    )

    row.update(
        coordination
    )


    # ========================================================
    # PYRAMIDALIZATION
    # ========================================================

    pyr = calculate_pyramidalization(
        elements,
        coordinates,
        donor1,
        donor2
    )

    row.update(
        pyr
    )


    # ========================================================
    # xTB
    # ========================================================

    xtb = calculate_xtb(
        elements,
        coordinates,
        donor1,
        donor2
    )

    row.update(
        xtb
    )


    # ========================================================
    # YIELD
    # ========================================================

    row[
        "Yield"
    ] = YIELD_DATA.get(
        ligand_name,
        np.nan
    )


    # ========================================================
    # WARNINGS
    # ========================================================

    warnings_list = []


    if row[
        "Donor_Selection_Check"
    ] != "PASS":

        warnings_list.append(
            row[
                "Donor_Selection_Check"
            ]
        )


    if (
        row[
            "Palladacycle_Type_Check"
        ]
        ==
        "CHECK"
    ):

        warnings_list.append(
            "Palladacycle type mismatch."
        )


    if pd.isna(
        row.get(
            "Bite_Angle_deg",
            np.nan
        )
    ):

        warnings_list.append(
            "Bite angle unavailable."
        )


    if pd.isna(
        row.get(
            "XTB_HOMO_eV",
            np.nan
        )
    ):

        warnings_list.append(
            "xTB descriptors unavailable."
        )


    row[
        "Warnings"
    ] = " || ".join(
        warnings_list
    )


    row[
        "Overall_Status"
    ] = (
        "PASS"
        if len(warnings_list) == 0
        else "WARNING"
    )


    return row


# ============================================================
# 23. PROCESS ALL XYZ
# ============================================================

print("\n")
print("=" * 80)
print("PROCESSING XYZ FILES")
print("=" * 80)


all_rows = []


for i, filepath in enumerate(
    xyz_files,
    start=1
):

    print(
        f"\n[{i}/{len(xyz_files)}] "
        f"{os.path.basename(filepath)}"
    )

    try:

        row = process_xyz(
            filepath
        )

        all_rows.append(
            row
        )

        print(
            "  Ligand:",
            row["Ligand"]
        )

        print(
            "  Donor 1:",
            row.get(
                "Donor1_Atom"
            ),
            row.get(
                "Donor1_Element"
            ),
            row.get(
                "Pd_Donor1_Distance_A"
            )
        )

        print(
            "  Donor 2:",
            row.get(
                "Donor2_Atom"
            ),
            row.get(
                "Donor2_Element"
            ),
            row.get(
                "Pd_Donor2_Distance_A"
            )
        )

        print(
            "  Palladacycle:",
            row.get(
                "Palladacycle_Type"
            )
        )

        print(
            "  Yield:",
            row.get(
                "Yield"
            )
        )

    except Exception as e:

        print(
            "  ERROR:",
            e
        )

        all_rows.append({

            "Ligand":
                clean_ligand_name(filepath),

            "XYZ_File":
                os.path.basename(filepath),

            "Overall_Status":
                "ERROR",

            "Warnings":
                str(e),

            "Yield":
                YIELD_DATA.get(
                    clean_ligand_name(filepath),
                    np.nan
                )
        })


# ============================================================
# 24. MASTER DATAFRAME
# ============================================================

master_df = pd.DataFrame(
    all_rows
)


# ============================================================
# 25. SORT L7 -> L25
# ============================================================

def ligand_sort_number(x):

    match = re.search(
        r"L(\d+)",
        str(x)
    )

    if match:
        return int(
            match.group(1)
        )

    return 9999


master_df[
    "_Sort"
] = master_df[
    "Ligand"
].apply(
    ligand_sort_number
)

master_df = (
    master_df
    .sort_values("_Sort")
    .drop(
        columns="_Sort"
    )
    .reset_index(
        drop=True
    )
)


# ============================================================
# 26. VERIFY YIELD MATCHING
# ============================================================

print("\n")
print("=" * 80)
print("YIELD MATCHING")
print("=" * 80)

yield_check = master_df[
    [
        "Ligand",
        "Yield"
    ]
]

print(
    yield_check.to_string(
        index=False
    )
)

print(
    "\nTotal ligands:",
    len(master_df)
)

print(
    "Yield available:",
    master_df[
        "Yield"
    ].notna().sum()
)

print(
    "Yield missing:",
    master_df[
        "Yield"
    ].isna().sum()
)


# ============================================================
# 27. DESCRIPTOR CATEGORIES
# ============================================================

steric_cols = [
    c for c in master_df.columns
    if (
        c.startswith("Ligand_SASA")
        or
        c.startswith("Ligand_%Vbur")
        or
        c.startswith("Ligand_Solid")
        or
        c.startswith("Ligand_Maximum")
        or
        c.startswith("Pd_Oriented_Sterimol")
        or
        c.startswith("Ligand_Distal")
        or
        c.startswith("Ligand_Molecular_Volume")
        or
        c.startswith("Ligand_Pint")
        or
        c.startswith("Vbur_")
    )
]


coord_cols = [
    c for c in master_df.columns
    if (
        c.startswith("Pd_Donor")
        or
        c.startswith("Average_Pd")
        or
        c.startswith("Delta_Pd")
        or
        c.startswith("Donor1_Donor2")
        or
        c.startswith("Bite_Angle")
        or
        c.startswith("Pd_Coordination")
        or
        c.startswith("Palladacycle_")
    )
]


donor_cols = [
    c for c in master_df.columns
    if (
        "Donor1_Mulliken" in c
        or
        "Donor2_Mulliken" in c
        or
        "Average_Donor_Charge" in c
        or
        "Delta_Donor_Charge" in c
        or
        "Donor_Charge_Sum" in c
    )
]


pyr_cols = [
    c for c in master_df.columns
    if (
        "Pyramidalization" in c
    )
]


xtb_cols = [
    c for c in master_df.columns
    if c.startswith("XTB_")
]


print("\nDescriptor counts:")

print(
    "A Steric:",
    len(steric_cols)
)

print(
    "B Coordination:",
    len(coord_cols)
)

print(
    "C Donor electronic:",
    len(donor_cols)
)

print(
    "D Pyramidalization:",
    len(pyr_cols)
)

print(
    "E xTB:",
    len(xtb_cols)
)

# ============================================================
# 28. NUMERIC DATASET
# ============================================================

numeric_df = master_df.select_dtypes(
    include=[np.number]
).copy()


# ============================================================
# 29. REMOVE NON-PREDICTOR / QC VARIABLES
#
# These stay in ALL_DESCRIPTORS but are NOT predictors.
# ============================================================

EXCLUDE_FROM_CORRELATION = {

    "Pd_Atom",

    "Ligand_Start_Atom",

    "Pd_Donor_Min_A",

    "Pd_Donor_Max_A",

    "Donor1_Atom",

    "Donor2_Atom",

    # Redundant duplicate of Palladacycle_Ring_Size
    "Palladacycle_Path_Atom_Count",

    "N_Atoms",

    "Number_of_Donor_Candidates",

    "Number_of_Qualifying_Pd_Donors",

    # Pyramidalization QC
    "Donor1_Pyramidalization_Neighbor_Count",

    "Donor2_Pyramidalization_Neighbor_Count",

    # Metadata / settings
    "XTB_Charge",
    "XTB_N_Unpaired",
    "XTB_Method"
}


numeric_df = numeric_df.drop(
    columns=[
        c
        for c in EXCLUDE_FROM_CORRELATION
        if c in numeric_df.columns
    ],
    errors="ignore"
)


# ============================================================
# 30. REMOVE CONSTANT DESCRIPTORS
# ============================================================

constant_columns = []

for col in numeric_df.columns:

    if col == "Yield":
        continue

    values = numeric_df[
        col
    ].dropna()

    if values.nunique() <= 1:

        constant_columns.append(
            col
        )


numeric_df = numeric_df.drop(
    columns=constant_columns,
    errors="ignore"
)


print("\nConstant descriptors removed:")

if constant_columns:

    for c in constant_columns:
        print(" ",
              c)

else:

    print("  None")


# ============================================================
# 31. REQUIRE COMPLETE DESCRIPTORS
#
# We do NOT impute missing descriptor values.
# ============================================================

complete_columns = []

missing_columns = []

for col in numeric_df.columns:

    if numeric_df[
        col
    ].notna().all():

        complete_columns.append(
            col
        )

    else:

        missing_columns.append(
            col
        )


# Yield must remain
if "Yield" not in complete_columns:

    raise ValueError(
        "Yield contains missing values."
    )


correlation_base = numeric_df[
    complete_columns
].copy()


# ============================================================
# 32. OUTLIER SCREENING
#
# Descriptor removed if >10% observations are outliers
# by either IQR or MAD criterion.
#
# With 19 ligands:
# 2 or more outliers => >10%
# ============================================================

outlier_records = []

excluded_outliers = []

retained = []


descriptor_columns = [
    c
    for c in correlation_base.columns
    if c != "Yield"
]


for col in descriptor_columns:

    x = correlation_base[
        col
    ].dropna().astype(float)

    n = len(x)


    # --------------------------------------------------------
    # IQR
    # --------------------------------------------------------

    q1 = x.quantile(
        0.25
    )

    q3 = x.quantile(
        0.75
    )

    iqr = q3 - q1

    if iqr > 0:

        lower = (
            q1
            -
            IQR_MULTIPLIER * iqr
        )

        upper = (
            q3
            +
            IQR_MULTIPLIER * iqr
        )

        iqr_mask = (
            (x < lower)
            |
            (x > upper)
        )

        n_iqr = int(
            iqr_mask.sum()
        )

    else:

        n_iqr = 0


    # --------------------------------------------------------
    # MAD
    # --------------------------------------------------------

    median = x.median()

    mad = np.median(
        np.abs(
            x - median
        )
    )

    if mad > 0:

        modified_z = (
            0.6745
            *
            (x - median)
            /
            mad
        )

        mad_mask = (
            np.abs(
                modified_z
            )
            >
            MAD_Z_LIMIT
        )

        n_mad = int(
            mad_mask.sum()
        )

    else:

        n_mad = 0


    iqr_fraction = (
        n_iqr / n
        if n > 0
        else np.nan
    )

    mad_fraction = (
        n_mad / n
        if n > 0
        else np.nan
    )


    remove = (
        iqr_fraction
        >
        OUTLIER_FRACTION_LIMIT
        or
        mad_fraction
        >
        OUTLIER_FRACTION_LIMIT
    )


    outlier_records.append({

        "Descriptor":
            col,

        "N":
            n,

        "IQR_Outliers":
            n_iqr,

        "IQR_Outlier_Fraction":
            iqr_fraction,

        "MAD_Outliers":
            n_mad,

        "MAD_Outlier_Fraction":
            mad_fraction,

        "Removed":
            remove
    })


    if remove:

        excluded_outliers.append(
            col
        )

    else:

        retained.append(
            col
        )


outlier_df = pd.DataFrame(
    outlier_records
)


# ============================================================
# 33. FINAL FILTERED CORRELATION DATA
#
# Remove exact duplicate descriptors before correlation analysis.
# Ring size and path atom count encode the same project variable;
# ring size is retained as the chemically clearer descriptor.
# ============================================================

if "Palladacycle_Path_Atom_Count" in retained:
    retained.remove("Palladacycle_Path_Atom_Count")



correlation_columns = [
    "Yield"
] + retained


correlation_df = correlation_base[
    correlation_columns
].copy()


print("\n" + "=" * 80)
print("CORRELATION DATASET")
print("=" * 80)

print(
    "Yield + retained descriptors:",
    correlation_df.shape
)

print(
    "\nRetained descriptors:",
    len(retained)
)

print(
    "Removed due to excessive outliers:",
    len(excluded_outliers)
)


if excluded_outliers:

    print("\nRemoved:")

    for c in excluded_outliers:
        print(
            " ",
            c
        )


# ============================================================
# 34. FULL SQUARE PEARSON
# ============================================================

pearson_matrix_full = (
    correlation_df.corr(
        method="pearson"
    )
)


# ============================================================
# 35. FULL SQUARE SPEARMAN
# ============================================================

spearman_matrix_full = (
    correlation_df.corr(
        method="spearman"
    )
)


print(
    "\nFull Pearson matrix:",
    pearson_matrix_full.shape
)

print(
    "Full Spearman matrix:",
    spearman_matrix_full.shape
)


# ============================================================
# 36. ALSO CREATE RAW COMPLETE-DESCRIPTOR MATRICES
#
# Before outlier filtering
# ============================================================

raw_correlation_columns = [
    "Yield"
] + [
    c
    for c in correlation_base.columns
    if c != "Yield"
]

raw_correlation_df = correlation_base[
    raw_correlation_columns
].copy()


pearson_matrix_raw = (
    raw_correlation_df.corr(
        method="pearson"
    )
)


spearman_matrix_raw = (
    raw_correlation_df.corr(
        method="spearman"
    )
)


# ============================================================
# 37–38. PUBLICATION-QUALITY CORRELATION MATRICES
# ============================================================

MATRIX_ANNOTATION_MAX = 30
MATRIX_DPI = 600
MATRIX_BASE_FONT = 10
MATRIX_TICK_FONT = 9
MATRIX_TITLE_FONT = 16


def plot_correlation_matrix(matrix, title, colorbar_label, outfile):

    n = len(matrix)
    side = max(16, min(32, n * 0.68))
    annotation_font = max(8, min(14, 260 / max(n, 1)))

    fig, ax = plt.subplots(figsize=(side, side), constrained_layout=True)

    sns.heatmap(
        matrix,
        ax=ax,
        cmap="RdBu_r",
        center=0,
        vmin=-1,
        vmax=1,
        square=True,
        linewidths=0.35,
        linecolor="white",
        annot=(n <= MATRIX_ANNOTATION_MAX),
        fmt=".2f",
        annot_kws={
            "fontsize": annotation_font,
            "fontweight": "normal"
        },
        cbar_kws={
            "label": colorbar_label,
            "shrink": 0.78,
            "pad": 0.02
        }
    )

    ax.set_title(title, fontsize=MATRIX_TITLE_FONT, pad=16, weight="bold")
    ax.set_xticklabels(ax.get_xticklabels(), rotation=90, ha="center", fontsize=MATRIX_TICK_FONT)
    ax.set_yticklabels(ax.get_yticklabels(), rotation=0, fontsize=MATRIX_TICK_FONT)
    ax.tick_params(length=0)

    cbar = ax.collections[0].colorbar
    cbar.ax.tick_params(labelsize=9)
    cbar.set_label(colorbar_label, fontsize=10)

    fig.savefig(outfile, dpi=MATRIX_DPI, bbox_inches="tight", facecolor="white")
    plt.show()
    plt.close(fig)


plot_correlation_matrix(
    pearson_matrix_full,
    "Pearson Correlation Matrix\nYield + Retained Ligand Descriptors",
    "Pearson r",
    os.path.join(PLOT_FOLDER, "Pearson_FULL_SQUARE_Correlation_Matrix.png")
)

plot_correlation_matrix(
    spearman_matrix_full,
    "Spearman Correlation Matrix\nYield + Retained Ligand Descriptors",
    "Spearman ρ",
    os.path.join(PLOT_FOLDER, "Spearman_FULL_SQUARE_Correlation_Matrix.png")
)


# ============================================================
# 39. YIELD CORRELATIONS
# ============================================================

correlation_results = []


for descriptor in retained:

    temp = correlation_df[
        [
            "Yield",
            descriptor
        ]
    ].dropna()

    n = len(temp)


    if n >= 3:

        try:

            r, p_r = pearsonr(
                temp["Yield"],
                temp[descriptor]
            )

        except Exception:

            r = np.nan
            p_r = np.nan

        try:

            rho, p_s = spearmanr(
                temp["Yield"],
                temp[descriptor]
            )

        except Exception:

            rho = np.nan
            p_s = np.nan

    else:

        r = np.nan
        p_r = np.nan

        rho = np.nan
        p_s = np.nan


    correlation_results.append({

        "Descriptor":
            descriptor,

        "N":
            n,

        "Pearson_r":
            r,

        "Pearson_p":
            p_r,

        "Spearman_rho":
            rho,

        "Spearman_p":
            p_s,

        "Abs_Pearson_r":
            abs(r)
            if pd.notna(r)
            else np.nan,

        "Abs_Spearman_rho":
            abs(rho)
            if pd.notna(rho)
            else np.nan
    })


yield_correlation_df = pd.DataFrame(
    correlation_results
)

# Multiple-testing correction for the yield correlation screen.
# These adjusted p-values are the publication-facing significance values.
try:

    from statsmodels.stats.multitest import multipletests

    if len(yield_correlation_df) > 0:

        yield_correlation_df["Pearson_FDR_p"] = multipletests(
            yield_correlation_df["Pearson_p"].fillna(1.0),
            method="fdr_bh"
        )[1]

        yield_correlation_df["Spearman_FDR_p"] = multipletests(
            yield_correlation_df["Spearman_p"].fillna(1.0),
            method="fdr_bh"
        )[1]

except Exception as e:

    yield_correlation_df["FDR_Warning"] = str(e)


yield_correlation_df[
    "Mean_Abs_Correlation"
] = (
    yield_correlation_df[
        "Abs_Pearson_r"
    ]
    +
    yield_correlation_df[
        "Abs_Spearman_rho"
    ]
) / 2.0


yield_correlation_df = (
    yield_correlation_df
    .sort_values(
        "Mean_Abs_Correlation",
        ascending=False
    )
    .reset_index(
        drop=True
    )
)


# ============================================================
# 40. PRINT YIELD CORRELATIONS
# ============================================================

print("\n")
print("=" * 100)
print("YIELD vs RETAINED DESCRIPTORS")
print("=" * 100)

display(
    yield_correlation_df
)


# ============================================================
# 41. PEARSON YIELD RANKING
# ============================================================

pearson_rank = (
    yield_correlation_df
    .sort_values(
        "Pearson_r",
        ascending=True
    )
)


plt.figure(
    figsize=(
        10,
        max(
            6,
            len(pearson_rank)
            * 0.32
        )
    )
)


plt.barh(
    pearson_rank[
        "Descriptor"
    ],
    pearson_rank[
        "Pearson_r"
    ]
)


plt.axvline(
    0,
    linewidth=1
)

plt.xlabel(
    "Pearson r"
)

plt.ylabel(
    "Descriptor"
)

plt.title(
    "Pearson Correlation with Yield"
)

plt.tight_layout()


plt.savefig(
    os.path.join(
        PLOT_FOLDER,
        "Pearson_Yield_Correlation_Ranking.png"
    ),
    dpi=400,
    bbox_inches="tight"
)

plt.show()

plt.close()


# ============================================================
# 42. SPEARMAN YIELD RANKING
# ============================================================

spearman_rank = (
    yield_correlation_df
    .sort_values(
        "Spearman_rho",
        ascending=True
    )
)


plt.figure(
    figsize=(
        10,
        max(
            6,
            len(spearman_rank)
            * 0.32
        )
    )
)


plt.barh(
    spearman_rank[
        "Descriptor"
    ],
    spearman_rank[
        "Spearman_rho"
    ]
)


plt.axvline(
    0,
    linewidth=1
)

plt.xlabel(
    "Spearman ρ"
)

plt.ylabel(
    "Descriptor"
)

plt.title(
    "Spearman Correlation with Yield"
)

plt.tight_layout()


plt.savefig(
    os.path.join(
        PLOT_FOLDER,
        "Spearman_Yield_Correlation_Ranking.png"
    ),
    dpi=400,
    bbox_inches="tight"
)

plt.show()

plt.close()


# ============================================================
# 43. PEARSON vs SPEARMAN
# ============================================================

plt.figure(
    figsize=(8, 8)
)


plt.scatter(
    yield_correlation_df[
        "Pearson_r"
    ],
    yield_correlation_df[
        "Spearman_rho"
    ],
    s=65,
    alpha=0.85
)


for _, row in (
    yield_correlation_df.iterrows()
):

    if (
        pd.notna(
            row["Pearson_r"]
        )
        and
        pd.notna(
            row["Spearman_rho"]
        )
    ):

        plt.annotate(
            row["Descriptor"],
            (
                row["Pearson_r"],
                row["Spearman_rho"]
            ),
            xytext=(4, 4),
            textcoords="offset points",
            fontsize=7
        )


plt.axhline(
    0,
    linewidth=1
)

plt.axvline(
    0,
    linewidth=1
)

plt.xlabel(
    "Pearson r"
)

plt.ylabel(
    "Spearman ρ"
)

plt.title(
    "Pearson vs Spearman Correlation with Yield"
)

plt.grid(
    alpha=0.25
)

plt.tight_layout()


plt.savefig(
    os.path.join(
        PLOT_FOLDER,
        "Pearson_vs_Spearman_Yield.png"
    ),
    dpi=400,
    bbox_inches="tight"
)

plt.show()

plt.close()


# ============================================================
# 44. YIELD vs EVERY DESCRIPTOR
#
# IMPORTANT:
# NO REGRESSION LINE
# ============================================================

for descriptor in retained:

    temp = correlation_df[
        [
            "Yield",
            descriptor
        ]
    ].dropna()

    if len(temp) < 3:
        continue


    try:

        r, p_r = pearsonr(
            temp[descriptor],
            temp["Yield"]
        )

    except Exception:

        r = np.nan
        p_r = np.nan


    try:

        rho, p_s = spearmanr(
            temp[descriptor],
            temp["Yield"]
        )

    except Exception:

        rho = np.nan
        p_s = np.nan


    plt.figure(
        figsize=(7, 5)
    )


    plt.scatter(
        temp[descriptor],
        temp["Yield"],
        s=55,
        alpha=0.85
    )


    # --------------------------------------------------------
    # Ligand labels
    # --------------------------------------------------------

    for idx in temp.index:

        label = master_df.loc[
            idx,
            "Ligand"
        ]

        if pd.notna(label):

            plt.annotate(
                str(label),
                (
                    temp.loc[
                        idx,
                        descriptor
                    ],
                    temp.loc[
                        idx,
                        "Yield"
                    ]
                ),
                xytext=(4, 4),
                textcoords="offset points",
                fontsize=7
            )


    plt.xlabel(
        descriptor
    )

    plt.ylabel(
        "Yield (%)"
    )

    plt.title(
        f"Yield vs {descriptor}"
    )


    stats_text = (
        f"Pearson r = {r:.3f}\n"
        f"Pearson p = {p_r:.3g}\n"
        f"Spearman ρ = {rho:.3f}\n"
        f"Spearman p = {p_s:.3g}\n"
        f"n = {len(temp)}"
    )


    plt.text(
        0.03,
        0.97,
        stats_text,
        transform=plt.gca().transAxes,
        va="top",
        fontsize=9,
        bbox=dict(
            boxstyle="round",
            alpha=0.15
        )
    )


    plt.grid(
        alpha=0.25
    )

    plt.tight_layout()


    outfile = os.path.join(
        SCATTER_FOLDER,
        "Yield_vs_"
        +
        safe_filename(
            descriptor
        )
        +
        ".png"
    )


    plt.savefig(
        outfile,
        dpi=400,
        bbox_inches="tight"
    )

    plt.close()


# ============================================================
# 45. DESCRIPTOR BOXPLOTS
# ============================================================

for descriptor in retained:

    values = correlation_df[
        descriptor
    ].dropna()

    if len(values) < 3:
        continue


    plt.figure(
        figsize=(6, 4)
    )


    plt.boxplot(
        values
    )

    plt.ylabel(
        descriptor
    )

    plt.title(
        f"Distribution: {descriptor}"
    )

    plt.grid(
        axis="y",
        alpha=0.25
    )

    plt.tight_layout()


    outfile = os.path.join(
        BOX_FOLDER,
        "Boxplot_"
        +
        safe_filename(
            descriptor
        )
        +
        ".png"
    )


    plt.savefig(
        outfile,
        dpi=350,
        bbox_inches="tight"
    )

    plt.close()


# ============================================================
# 46. HIGH CORRELATION BETWEEN DESCRIPTORS
# ============================================================

high_pairs = []


for i in range(
    len(retained)
):

    for j in range(
        i + 1,
        len(retained)
    ):

        d1 = retained[i]
        d2 = retained[j]


        r = pearson_matrix_full.loc[
            d1,
            d2
        ]

        rho = spearman_matrix_full.loc[
            d1,
            d2
        ]

        if (
            abs(r)
            >=
            HIGH_CORR_THRESHOLD
            or
            abs(rho)
            >=
            HIGH_CORR_THRESHOLD
        ):

            high_pairs.append({

                "Descriptor_1":
                    d1,

                "Descriptor_2":
                    d2,

                "Pearson_r":
                    r,

                "Spearman_rho":
                    rho,

                "Max_Abs_Correlation":
                    max(
                        abs(r),
                        abs(rho)
                    )
            })


high_corr_df = pd.DataFrame(
    high_pairs,
    columns=[
        "Descriptor_1",
        "Descriptor_2",
        "Pearson_r",
        "Spearman_rho",
        "Max_Abs_Correlation"
    ]
)


if len(high_corr_df) > 0:

    high_corr_df = (
        high_corr_df
        .sort_values(
            "Max_Abs_Correlation",
            ascending=False
        )
        .reset_index(
            drop=True
        )
    )


# ============================================================
# 47. YIELD BY PALLADACYCLE TYPE
# ============================================================

class_df = master_df[
    [
        "Palladacycle_Type",
        "Yield"
    ]
].dropna()


if len(class_df) > 0:

    plt.figure(
        figsize=(7, 5)
    )


    sns.boxplot(
        data=class_df,
        x="Palladacycle_Type",
        y="Yield"
    )


    sns.stripplot(
        data=class_df,
        x="Palladacycle_Type",
        y="Yield",
        jitter=True,
        alpha=0.7
    )


    plt.xlabel(
        "Palladacycle type"
    )

    plt.ylabel(
        "Yield (%)"
    )

    plt.title(
        "Yield by Palladacycle Type"
    )

    plt.tight_layout()


    plt.savefig(
        os.path.join(
            PLOT_FOLDER,
            "Yield_by_Palladacycle_Type.png"
        ),
        dpi=400,
        bbox_inches="tight"
    )

    plt.show()

    plt.close()


# ============================================================
# 48. SAVE VALIDATION TABLE
# ============================================================

validation_columns = [

    "Ligand",

    "XYZ_File",

    "N_Atoms",

    "Donor1_Atom",
    "Donor1_Element",

    "Donor2_Atom",
    "Donor2_Element",

    "Pd_Donor1_Distance_A",
    "Pd_Donor2_Distance_A",

    "Number_of_Qualifying_Pd_Donors",

    "Donor_Selection_Check",

    "Palladacycle_Type",

    "Expected_Palladacycle_Type",

    "Palladacycle_Type_Check",

    "Palladacycle_Ligand_Path",

    "Overall_Status",

    "Warnings"
]


validation_columns = [
    c
    for c in validation_columns
    if c in master_df.columns
]


validation_df = master_df[
    validation_columns
].copy()


# ============================================================
# 49. FINAL EXCEL WORKBOOK
# ============================================================

print("\n")
print("=" * 80)
print("WRITING FINAL EXCEL WORKBOOK")
print("=" * 80)


with pd.ExcelWriter(
    FINAL_EXCEL,
    engine="openpyxl"
) as writer:


    # --------------------------------------------------------
    # COMPLETE RAW DATA
    # --------------------------------------------------------

    master_df.to_excel(
        writer,
        sheet_name="ALL_DESCRIPTORS",
        index=False
    )


    # --------------------------------------------------------
    # VALIDATION
    # --------------------------------------------------------

    validation_df.to_excel(
        writer,
        sheet_name="VALIDATION",
        index=False
    )


    # --------------------------------------------------------
    # YIELD CORRELATION DATA
    # --------------------------------------------------------

    correlation_df.to_excel(
        writer,
        sheet_name="Correlation_Data",
        index=False
    )


    # --------------------------------------------------------
    # FULL RAW PEARSON
    # --------------------------------------------------------

    pearson_matrix_raw.to_excel(
        writer,
        sheet_name="Pearson_RAW_FULL"
    )


    # --------------------------------------------------------
    # FULL RAW SPEARMAN
    # --------------------------------------------------------

    spearman_matrix_raw.to_excel(
        writer,
        sheet_name="Spearman_RAW_FULL"
    )


    # --------------------------------------------------------
    # FILTERED PEARSON
    # --------------------------------------------------------

    pearson_matrix_full.to_excel(
        writer,
        sheet_name="Pearson_FILTERED_FULL"
    )


    # --------------------------------------------------------
    # FILTERED SPEARMAN
    # --------------------------------------------------------

    spearman_matrix_full.to_excel(
        writer,
        sheet_name="Spearman_FILTERED_FULL"
    )


    # --------------------------------------------------------
    # YIELD CORRELATIONS
    # --------------------------------------------------------

    yield_correlation_df.to_excel(
        writer,
        sheet_name="Yield_Correlations",
        index=False
    )


    # --------------------------------------------------------
    # OUTLIER SCREENING
    # --------------------------------------------------------

    outlier_df.to_excel(
        writer,
        sheet_name="Outlier_Screening",
        index=False
    )


    # --------------------------------------------------------
    # HIGH CORRELATION PAIRS
    # --------------------------------------------------------

    high_corr_df.to_excel(
        writer,
        sheet_name="High_Correlation_Pairs",
        index=False
    )


    # --------------------------------------------------------
    # A - STERIC
    # --------------------------------------------------------

    cols = [
        "Ligand"
    ] + [
        c for c in steric_cols
        if c in master_df.columns
    ] + ["Yield"]

    master_df[
        list(dict.fromkeys(cols))
    ].to_excel(
        writer,
        sheet_name="A_STERIC",
        index=False
    )


    # --------------------------------------------------------
    # B - COORDINATION
    # --------------------------------------------------------

    cols = [
        "Ligand"
    ] + [
        c for c in coord_cols
        if c in master_df.columns
    ] + ["Yield"]

    master_df[
        list(dict.fromkeys(cols))
    ].to_excel(
        writer,
        sheet_name="B_COORDINATION",
        index=False
    )


    # --------------------------------------------------------
    # C - DONOR ELECTRONIC
    # --------------------------------------------------------

    cols = [
        "Ligand"
    ] + [
        c for c in donor_cols
        if c in master_df.columns
    ] + ["Yield"]

    master_df[
        list(dict.fromkeys(cols))
    ].to_excel(
        writer,
        sheet_name="C_DONOR_ELECTRONIC",
        index=False
    )


    # --------------------------------------------------------
    # D - PYRAMIDALIZATION
    # --------------------------------------------------------

    cols = [
        "Ligand"
    ] + [
        c for c in pyr_cols
        if c in master_df.columns
    ] + ["Yield"]

    master_df[
        list(dict.fromkeys(cols))
    ].to_excel(
        writer,
        sheet_name="D_PYRAMIDALIZATION",
        index=False
    )


    # --------------------------------------------------------
    # E - xTB
    # --------------------------------------------------------

    cols = [
        "Ligand"
    ] + [
        c for c in xtb_cols
        if c in master_df.columns
    ] + ["Yield"]

    master_df[
        list(dict.fromkeys(cols))
    ].to_excel(
        writer,
        sheet_name="E_XTB",
        index=False
    )


    # --------------------------------------------------------
    # YIELD DATA
    # --------------------------------------------------------

    yield_table = pd.DataFrame({

        "Ligand":
            list(YIELD_DATA.keys()),

        "Yield":
            list(YIELD_DATA.values())
    })


    yield_table.to_excel(
        writer,
        sheet_name="YIELD_DATA",
        index=False
    )


    # --------------------------------------------------------
    # REMOVED DESCRIPTORS
    # --------------------------------------------------------

    removed_records = []

    for c in constant_columns:

        removed_records.append({

            "Descriptor":
                c,

            "Reason":
                "Constant descriptor"
        })

    for c in missing_columns:

        if c != "Yield":

            removed_records.append({

                "Descriptor":
                    c,

                "Reason":
                    "Contains missing value(s)"
            })

    for c in excluded_outliers:

        removed_records.append({

            "Descriptor":
                c,

            "Reason":
                "More than 10% strong outliers"
        })

    removed_df = pd.DataFrame(
        removed_records
    )

    removed_df.to_excel(
        writer,
        sheet_name="REMOVED_DESCRIPTORS",
        index=False
    )


# ============================================================
# 50. FINAL SUMMARY
# ============================================================

print("\n")
print("=" * 100)
print("FINAL ANALYSIS COMPLETE")
print("=" * 100)

print(
    "\nExcel:"
)

print(
    FINAL_EXCEL
)

print(
    "\nPlots:"
)

print(
    PLOT_FOLDER
)

print(
    "\nNumber of ligands:",
    len(master_df)
)

print(
    "Total descriptors in raw table:",
    len(
        [
            c for c in master_df.columns
            if c not in [
                "Ligand",
                "XYZ_File",
                "Warnings",
                "Overall_Status"
            ]
        ]
    )
)

print(
    "Retained correlation descriptors:",
    len(retained)
)

print(
    "Constant removed:",
    len(constant_columns)
)

print(
    "Missing-value removed:",
    len(
        [
            c
            for c in missing_columns
            if c != "Yield"
        ]
    )
)

print(
    "Outlier removed:",
    len(excluded_outliers)
)

print(
    "\nPearson matrix:",
    pearson_matrix_full.shape
)

print(
    "Spearman matrix:",
    spearman_matrix_full.shape
)

print(
    "\nTop 10 Yield correlations:"
)

display(
    yield_correlation_df.head(10)
)

print(
    "\n" + "=" * 100
)
print(
    "✓ ALL PROCESSING FINISHED"
)
print(
    "=" * 100
)
```

```python
# ============================================================
# FINAL PUBLICATION CORRELATION ANALYSIS
# Pd-Catalyzed γ-Arylation
#
# PURPOSE
# ------------------------------------------------------------
# 1. FULL correlation matrices for Supporting Information
# 2. TOP-10 descriptor correlation matrix for Main Manuscript
#
# FIXED ANALYSIS PROTOCOL
# ------------------------------------------------------------
# - n = 19 ligands
# - Yield values are fixed for L7-L25
# - ALL valid numerical descriptors are considered
# - NO IQR filtering
# - NO MAD filtering
# - NO numerical-range filtering
# - NO scaling
# - NO normalization
# - NO imputation
# - Constant descriptors removed only
# - Descriptors with missing values are excluded from the
#   SAME-N full square matrix
# - Categorical descriptors excluded
# - Palladacycle_Path_Atom_Count excluded
# - NBO descriptors excluded
# - Top 10 ranked by Mean Absolute Correlation
#   = (|Pearson r| + |Spearman rho|) / 2
#
# MAIN MANUSCRIPT:
#   Top-10 SPEARMAN correlation matrix
#
# SUPPORTING INFORMATION:
#   Full Pearson + Full Spearman matrices
# ============================================================


# ============================================================
# I. GOOGLE DRIVE
# ============================================================

from google.colab import drive

drive.mount("/content/drive")


# ============================================================
# 1. IMPORTS
# ============================================================

import os
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns

from scipy.stats import pearsonr, spearmanr


# ============================================================
# 2. CHECK MASTER DATAFRAME
# ============================================================

if "master_df" not in globals():

    raise NameError(
        "\nmaster_df was not found.\n"
        "Please run the complete descriptor-calculation code first."
    )


raw_df = master_df.copy()


print("\n")
print("=" * 110)
print("FINAL PUBLICATION CORRELATION ANALYSIS")
print("=" * 110)

print(
    f"Number of rows in master_df: {len(raw_df)}"
)


# ============================================================
# 3. FIXED EXPERIMENTAL YIELD DATA
# ============================================================

YIELD_DATA = {

    "L7": 0,
    "L8": 0,
    "L9": 0,
    "L10": 0,
    "L11": 0,
    "L12": 0,
    "L13": 0,
    "L14": 0,
    "L15": 0,

    "L16": 12,
    "L17": 53,
    "L18": 46,
    "L19": 67,
    "L20": 40,
    "L21": 31,

    "L22": 0,
    "L23": 0,
    "L24": 0,
    "L25": 0
}


# ============================================================
# 4. IDENTIFY LIGAND COLUMN
# ============================================================

if "Ligand" in raw_df.columns:

    ligand_column = "Ligand"

elif "Ligand_ID" in raw_df.columns:

    ligand_column = "Ligand_ID"

else:

    raise KeyError(
        "Neither 'Ligand' nor 'Ligand_ID' exists in master_df."
    )


# ============================================================
# 5. ASSIGN YIELD
# ============================================================

raw_df["Yield"] = (

    raw_df[ligand_column]
    .astype(str)
    .map(YIELD_DATA)

)


# ============================================================
# 6. VERIFY YIELD
# ============================================================

if raw_df["Yield"].isna().any():

    missing_ligands = (

        raw_df.loc[
            raw_df["Yield"].isna(),
            ligand_column
        ]
        .astype(str)
        .tolist()

    )

    raise ValueError(
        "Yield is missing for ligand(s): "
        + str(missing_ligands)
    )


# ============================================================
# 7. VERIFY L7-L25 DATASET
# ============================================================

expected_ligands = [

    "L7",
    "L8",
    "L9",
    "L10",
    "L11",
    "L12",
    "L13",
    "L14",
    "L15",
    "L16",
    "L17",
    "L18",
    "L19",
    "L20",
    "L21",
    "L22",
    "L23",
    "L24",
    "L25"

]


observed_ligands = (

    raw_df[ligand_column]
    .astype(str)
    .tolist()

)


missing_expected = [

    ligand
    for ligand in expected_ligands
    if ligand not in observed_ligands

]


if missing_expected:

    raise ValueError(
        "Expected ligand(s) missing from master_df: "
        + str(missing_expected)
    )


print("\nYield assignment verified.")


# ============================================================
# 8. PRINT YIELD TABLE
# ============================================================

print("\n")
print("=" * 110)
print("YIELD DATA USED FOR CORRELATION")
print("=" * 110)

print(

    raw_df[
        [
            ligand_column,
            "Yield"
        ]
    ]
    .to_string(index=False)

)


# ============================================================
# 9. EXCLUDE NON-PREDICTOR COLUMNS
# ============================================================

EXCLUDE_FROM_CORRELATION = {

    # --------------------------------------------------------
    # Identification
    # --------------------------------------------------------

    "Ligand",
    "Ligand_ID",
    "XYZ_File",
    "Filename",
    "File",

    "Warnings",
    "Overall_Status",


    # --------------------------------------------------------
    # Atom indexing / bookkeeping
    # --------------------------------------------------------

    "Pd_Atom",
    "Ligand_Start_Atom",

    "Donor1_Atom",
    "Donor2_Atom",

    "N_Atoms",
    "Total_Atoms",
    "Ligand_Atom_Count",

    "Number_of_Donor_Candidates",
    "Number_of_Qualifying_Pd_Donors",


    # --------------------------------------------------------
    # Coordination settings
    # --------------------------------------------------------

    "Pd_Donor_Min_A",
    "Pd_Donor_Max_A",


    # --------------------------------------------------------
    # Pyramidalization QC
    # --------------------------------------------------------

    "Donor1_Pyramidalization_Neighbor_Count",
    "Donor2_Pyramidalization_Neighbor_Count",


    # --------------------------------------------------------
    # xTB metadata
    # --------------------------------------------------------

    "XTB_Charge",
    "XTB_N_Unpaired",
    "XTB_Method",
    "XTB_Charge_Sum",


    # --------------------------------------------------------
    # DUPLICATE / REDUNDANT PROJECT FIELD
    # --------------------------------------------------------

    "Palladacycle_Path_Atom_Count",


    # --------------------------------------------------------
    # CATEGORICAL PALLADACYCLE VARIABLES
    # --------------------------------------------------------

    "Palladacycle_Type",
    "Expected_Palladacycle_Type",
    "Palladacycle_Type_Check",

    "Palladacycle_Ligand_Path",


    # --------------------------------------------------------
    # NBO
    # --------------------------------------------------------

    "NBO_Donor1_to_Pd_E2_kcal",
    "NBO_Donor2_to_Pd_E2_kcal",
    "NBO_Total_Ligand_to_Pd_E2_kcal",
    "NBO_Max_Ligand_to_Pd_E2_kcal",
    "NBO_Donor_Orbital_Gap_eV"

}


# ============================================================
# 10. BUILD NUMERICAL DATASET
# ============================================================

numeric_df = pd.DataFrame(index=raw_df.index)


# Yield first

numeric_df["Yield"] = pd.to_numeric(
    raw_df["Yield"],
    errors="coerce"
)


# Add every valid numerical descriptor

for column in raw_df.columns:

    if column == "Yield":
        continue

    if column in EXCLUDE_FROM_CORRELATION:
        continue


    converted = pd.to_numeric(
        raw_df[column],
        errors="coerce"
    )


    # At least 3 numerical observations required

    if converted.notna().sum() >= 3:

        numeric_df[column] = converted


# ============================================================
# 11. REMOVE ONLY CONSTANT DESCRIPTORS
# ============================================================

constant_descriptors = []


for column in numeric_df.columns:

    if column == "Yield":
        continue


    values = numeric_df[column].dropna()


    if values.nunique() <= 1:

        constant_descriptors.append(column)


if constant_descriptors:

    print("\n")
    print("Constant descriptors removed:")

    for descriptor in constant_descriptors:

        print(
            "   ",
            descriptor
        )


    numeric_df = numeric_df.drop(
        columns=constant_descriptors
    )


# ============================================================
# 12. KEEP ONLY COMPLETE DESCRIPTORS
# ============================================================
#
# IMPORTANT:
#
# This is NOT outlier filtering.
# This is NOT range filtering.
#
# It simply guarantees that the full square matrix has
# exactly the same N for every descriptor.
#
# No imputation is performed.
# ============================================================

complete_columns = []
missing_descriptors = []


for column in numeric_df.columns:

    if numeric_df[column].notna().all():

        complete_columns.append(column)

    else:

        missing_descriptors.append(column)


raw_complete_df = numeric_df[
    complete_columns
].copy()


# ============================================================
# 13. FORCE YIELD TO FIRST COLUMN
# ============================================================

descriptor_columns = [

    column
    for column in raw_complete_df.columns
    if column != "Yield"

]


raw_complete_df = raw_complete_df[
    ["Yield"] + descriptor_columns
].copy()


# ============================================================
# 14. FINAL SAME-N CHECK
# ============================================================

if raw_complete_df["Yield"].isna().any():

    raise ValueError(
        "Yield contains missing values."
    )


N_FINAL = len(raw_complete_df)


print("\n")
print("=" * 110)
print("FINAL CORRELATION DATASET")
print("=" * 110)

print(
    "Number of ligands:",
    N_FINAL
)

print(
    "Number of numerical descriptors:",
    len(descriptor_columns)
)

print(
    "Number of complete descriptors excluded:",
    len(missing_descriptors)
)


# ============================================================
# 15. CHECK EXPECTED N = 19
# ============================================================

if N_FINAL != 19:

    raise ValueError(
        f"Expected N=19 ligands, but correlation dataset "
        f"contains N={N_FINAL}."
    )


# ============================================================
# 16. CHECK IMPORTANT DESCRIPTORS
# ============================================================

IMPORTANT_DESCRIPTORS = [

    # Sterimol

    "Pd_Oriented_Sterimol_B1_A",
    "Pd_Oriented_Sterimol_B5_A",
    "Pd_Oriented_Sterimol_L_A",


    # Buried volume

    "Ligand_%Vbur_3.0A",
    "Ligand_%Vbur_3.5A",
    "Ligand_%Vbur_4.0A",
    "Ligand_%Vbur_4.5A",
    "Ligand_%Vbur_5.0A",


    # Distal / molecular volume

    "Ligand_Distal_Volume_Outside_3.5A_A3",
    "Ligand_Distal_Volume_Fraction_Outside_3.5A",
    "Ligand_Molecular_Volume_A3",


    # Surface

    "Ligand_SASA_A2",
    "Ligand_SASA_Volume_A3",


    # Dispersion

    "Ligand_Pint",


    # Solid angle

    "Ligand_Solid_Angle",
    "Ligand_Solid_Cone_Angle_deg",
    "Ligand_G"

]


print("\n")
print("=" * 110)
print("IMPORTANT DESCRIPTOR VERIFICATION")
print("=" * 110)


for descriptor in IMPORTANT_DESCRIPTORS:

    if descriptor in raw_complete_df.columns:

        values = raw_complete_df[descriptor]

        print(
            f"✓ {descriptor}"
        )

        print(
            f"    N = {values.notna().sum()}"
        )

        print(
            f"    Min = {values.min():.6f}"
        )

        print(
            f"    Max = {values.max():.6f}"
        )

    elif descriptor in raw_df.columns:

        valid_n = pd.to_numeric(
            raw_df[descriptor],
            errors="coerce"
        ).notna().sum()

        print(
            f"⚠ {descriptor}"
        )

        print(
            f"    Present but incomplete: "
            f"{valid_n}/{N_FINAL}"
        )

    else:

        print(
            f"✗ {descriptor} NOT FOUND"
        )


# ============================================================
# 17. FULL PEARSON MATRIX
# ============================================================

pearson_matrix_full = (

    raw_complete_df
    .corr(method="pearson")

)


# ============================================================
# 18. FULL SPEARMAN MATRIX
# ============================================================

spearman_matrix_full = (

    raw_complete_df
    .corr(method="spearman")

)


# ============================================================
# 19. YIELD vs ALL DESCRIPTORS
# ============================================================

yield_results = []


y = raw_complete_df["Yield"].to_numpy(
    dtype=float
)


for descriptor in descriptor_columns:

    x = raw_complete_df[descriptor].to_numpy(
        dtype=float
    )


    if (
        len(x) >= 3
        and np.std(x) > 0
        and np.std(y) > 0
    ):

        r, p_pearson = pearsonr(
            x,
            y
        )


        rho, p_spearman = spearmanr(
            x,
            y
        )


        mean_abs = (

            abs(r)
            +
            abs(rho)

        ) / 2.0


        yield_results.append({

            "Descriptor":
                descriptor,

            "N":
                len(x),

            "Pearson_r":
                r,

            "Pearson_p":
                p_pearson,

            "Spearman_rho":
                rho,

            "Spearman_p":
                p_spearman,

            "Abs_Pearson_r":
                abs(r),

            "Abs_Spearman_rho":
                abs(rho),

            "Mean_Abs_Correlation":
                mean_abs

        })


# ============================================================
# 20. YIELD CORRELATION RANKING
# ============================================================

yield_correlation_df = pd.DataFrame(
    yield_results
)


yield_correlation_df = (

    yield_correlation_df
    .sort_values(
        "Mean_Abs_Correlation",
        ascending=False
    )
    .reset_index(drop=True)

)


# ============================================================
# 21. PRINT FINAL RANKING
# ============================================================

print("\n")
print("=" * 110)
print("FINAL YIELD CORRELATION RANKING")
print("=" * 110)

display(
    yield_correlation_df
)


# ============================================================
# 22. SELECT TOP 10
# ============================================================
#
# FIXED RULE:
#
# Mean absolute correlation =
#
# (|Pearson r| + |Spearman rho|) / 2
#
# ============================================================

TOP_N = 10


top10_descriptors = (

    yield_correlation_df[
        "Descriptor"
    ]
    .head(TOP_N)
    .tolist()

)


top10_ranking = yield_correlation_df[
    yield_correlation_df[
        "Descriptor"
    ].isin(top10_descriptors)
].copy()


# Preserve exact ranking order

top10_ranking["Rank"] = (

    top10_ranking[
        "Descriptor"
    ]
    .map(
        {
            descriptor: i + 1
            for i, descriptor
            in enumerate(top10_descriptors)
        }
    )

)


top10_ranking = (

    top10_ranking
    .sort_values("Rank")
    .reset_index(drop=True)

)


# ============================================================
# 23. PRINT TOP 10
# ============================================================

print("\n")
print("=" * 110)
print("TOP 10 DESCRIPTORS FOR MAIN MANUSCRIPT")
print("=" * 110)

display(
    top10_ranking
)


print("\nTop-10 descriptors:")

for i, descriptor in enumerate(
    top10_descriptors,
    start=1
):

    print(
        f"{i:2d}. {descriptor}"
    )


# ============================================================
# 24. TOP-10 PEARSON MATRIX
# ============================================================

top10_columns = [
    "Yield"
] + top10_descriptors


pearson_matrix_top10 = (

    raw_complete_df[
        top10_columns
    ]
    .corr(method="pearson")

)


# ============================================================
# 25. TOP-10 SPEARMAN MATRIX
# ============================================================

spearman_matrix_top10 = (

    raw_complete_df[
        top10_columns
    ]
    .corr(method="spearman")

)


# ============================================================
# 26. REORDER TOP-10 MATRIX EXACTLY BY RANK
# ============================================================

top10_matrix_order = (

    ["Yield"]
    + top10_descriptors

)


pearson_matrix_top10 = (

    pearson_matrix_top10
    .loc[
        top10_matrix_order,
        top10_matrix_order
    ]

)


spearman_matrix_top10 = (

    spearman_matrix_top10
    .loc[
        top10_matrix_order,
        top10_matrix_order
    ]

)


# ============================================================
# 27. SHORT DISPLAY NAMES
# ============================================================

DISPLAY_NAMES = {

    "Yield":
        "Yield",

    "Pd_Donor1_Distance_A":
        "Pd–Donor1 distance",

    "Palladacycle_Ring_Size":
        "Palladacycle ring size",

    "Donor1_Donor2_Distance_A":
        "Donor1–Donor2 distance",

    "Bite_Angle_deg":
        "Bite angle",

    "Pd_Oriented_Sterimol_B5_A":
        "Sterimol B5",

    "Pd_Oriented_Sterimol_B1_A":
        "Sterimol B1",

    "Pd_Oriented_Sterimol_L_A":
        "Sterimol L",

    "Ligand_%Vbur_3.0A":
        "%Vbur (3.0 Å)",

    "Ligand_%Vbur_3.5A":
        "%Vbur (3.5 Å)",

    "Ligand_%Vbur_4.0A":
        "%Vbur (4.0 Å)",

    "Ligand_%Vbur_4.5A":
        "%Vbur (4.5 Å)",

    "Ligand_%Vbur_5.0A":
        "%Vbur (5.0 Å)",

    "Ligand_Pint":
        "Ligand P_int",

    "Ligand_Molecular_Volume_A3":
        "Molecular volume",

    "Ligand_SASA_A2":
        "SASA area",

    "Ligand_SASA_Volume_A3":
        "SASA volume",

    "Ligand_Distal_Volume_Outside_3.5A_A3":
        "Distal volume",

    "Ligand_Distal_Volume_Fraction_Outside_3.5A":
        "Distal volume fraction",

    "Pd_Donor_Distance_Ratio":
        "Pd–donor distance ratio"

}


def rename_matrix_for_plot(matrix):

    return matrix.rename(
        index={
            column:
            DISPLAY_NAMES.get(
                column,
                column
            )
            for column in matrix.index
        },
        columns={
            column:
            DISPLAY_NAMES.get(
                column,
                column
            )
            for column in matrix.columns
        }
    )


# ============================================================
# 28. PUBLICATION-QUALITY HEATMAP FUNCTION
# ============================================================

def plot_publication_heatmap(
    matrix,
    title,
    colorbar_label,
    outfile,
    annotation=True
):

    plot_matrix = rename_matrix_for_plot(
        matrix
    )


    n = len(plot_matrix)


    fig_size = max(
        10,
        n * 0.95
    )


    plt.figure(
        figsize=(
            fig_size,
            fig_size
        )
    )


    sns.heatmap(

        plot_matrix,

        annot=annotation,

        fmt=".2f",

        cmap="RdBu_r",

        center=0,

        vmin=-1,

        vmax=1,

        square=True,

        linewidths=0.5,

        linecolor="white",

        annot_kws={
            "fontsize": 9
        },

        cbar_kws={
            "label": colorbar_label,
            "shrink": 0.80
        }

    )


    plt.title(
        title,
        fontsize=15,
        fontweight="bold",
        pad=15
    )


    plt.xticks(
        rotation=65,
        ha="right",
        fontsize=9
    )


    plt.yticks(
        rotation=0,
        fontsize=9
    )


    plt.tight_layout()


    plt.savefig(
        outfile,
        dpi=600,
        bbox_inches="tight"
    )


    plt.show()

    plt.close()


# ============================================================
# 29. CREATE OUTPUT FOLDER
# ============================================================

if "OUTPUT_FOLDER" in globals():

    CORRELATION_OUTPUT = OUTPUT_FOLDER

else:

    CORRELATION_OUTPUT = (
        "/content/drive/MyDrive/"
        "Pd_gamma_arylation_Correlation"
    )


os.makedirs(
    CORRELATION_OUTPUT,
    exist_ok=True
)


# ============================================================
# 30. MAIN MANUSCRIPT:
#     TOP-10 SPEARMAN MATRIX
# ============================================================

TOP10_SPEARMAN_PNG = os.path.join(

    CORRELATION_OUTPUT,

    "MAIN_MANUSCRIPT_TOP10_SPEARMAN_Correlation.png"

)


plot_publication_heatmap(

    spearman_matrix_top10,

    "Spearman Rank Correlation\n"
    "Top 10 Ligand Descriptors + Yield",

    "Spearman ρ",

    TOP10_SPEARMAN_PNG,

    annotation=True

)


# ============================================================
# 31. TOP-10 PEARSON MATRIX
# ============================================================
#
# This is produced for validation.
#
# The Spearman version above is the recommended
# single matrix for the Main Manuscript.
# ============================================================

TOP10_PEARSON_PNG = os.path.join(

    CORRELATION_OUTPUT,

    "TOP10_PEARSON_Correlation.png"

)


plot_publication_heatmap(

    pearson_matrix_top10,

    "Pearson Correlation\n"
    "Top 10 Ligand Descriptors + Yield",

    "Pearson r",

    TOP10_PEARSON_PNG,

    annotation=True

)


# ============================================================
# 32. FULL PEARSON MATRIX
#     SUPPORTING INFORMATION
# ============================================================

FULL_PEARSON_PNG = os.path.join(

    CORRELATION_OUTPUT,

    "SI_FULL_PEARSON_Correlation_Matrix.png"

)


plot_publication_heatmap(

    pearson_matrix_full,

    "Full Pearson Correlation Matrix\n"
    "Yield + All Final Numerical Descriptors",

    "Pearson r",

    FULL_PEARSON_PNG,

    annotation=(
        len(pearson_matrix_full) <= 30
    )

)


# ============================================================
# 33. FULL SPEARMAN MATRIX
#     SUPPORTING INFORMATION
# ============================================================

FULL_SPEARMAN_PNG = os.path.join(

    CORRELATION_OUTPUT,

    "SI_FULL_SPEARMAN_Correlation_Matrix.png"

)


plot_publication_heatmap(

    spearman_matrix_full,

    "Full Spearman Correlation Matrix\n"
    "Yield + All Final Numerical Descriptors",

    "Spearman ρ",

    FULL_SPEARMAN_PNG,

    annotation=(
        len(spearman_matrix_full) <= 30
    )

)


# ============================================================
# 34. SAVE FINAL EXCEL WORKBOOK
# ============================================================

FINAL_CORRELATION_EXCEL = os.path.join(

    CORRELATION_OUTPUT,

    "Pd_gamma_arylation_FINAL_Publication_Correlation.xlsx"

)


with pd.ExcelWriter(

    FINAL_CORRELATION_EXCEL,

    engine="openpyxl"

) as writer:


    # --------------------------------------------------------
    # Raw complete numerical dataset
    # --------------------------------------------------------

    raw_complete_df.to_excel(

        writer,

        sheet_name="RAW_COMPLETE_DATA",

        index=False

    )


    # --------------------------------------------------------
    # Full Pearson
    # --------------------------------------------------------

    pearson_matrix_full.to_excel(

        writer,

        sheet_name="FULL_PEARSON"

    )


    # --------------------------------------------------------
    # Full Spearman
    # --------------------------------------------------------

    spearman_matrix_full.to_excel(

        writer,

        sheet_name="FULL_SPEARMAN"

    )


    # --------------------------------------------------------
    # Full Yield ranking
    # --------------------------------------------------------

    yield_correlation_df.to_excel(

        writer,

        sheet_name="YIELD_RANKING",

        index=False

    )


    # --------------------------------------------------------
    # Top 10 ranking
    # --------------------------------------------------------

    top10_ranking.to_excel(

        writer,

        sheet_name="TOP10_RANKING",

        index=False

    )


    # --------------------------------------------------------
    # Top 10 Pearson
    # --------------------------------------------------------

    pearson_matrix_top10.to_excel(

        writer,

        sheet_name="TOP10_PEARSON"

    )


    # --------------------------------------------------------
    # Top 10 Spearman
    # --------------------------------------------------------

    spearman_matrix_top10.to_excel(

        writer,

        sheet_name="TOP10_SPEARMAN"

    )


    # --------------------------------------------------------
    # Missing descriptors
    # --------------------------------------------------------

    pd.DataFrame({

        "Descriptor":
            missing_descriptors

    }).to_excel(

        writer,

        sheet_name="EXCLUDED_MISSING",

        index=False

    )


    # --------------------------------------------------------
    # Constant descriptors
    # --------------------------------------------------------

    pd.DataFrame({

        "Descriptor":
            constant_descriptors

    }).to_excel(

        writer,

        sheet_name="EXCLUDED_CONSTANT",

        index=False

    )


    # --------------------------------------------------------
    # Analysis protocol
    # --------------------------------------------------------

    protocol_df = pd.DataFrame({

        "Parameter": [

            "Dataset size",
            "Yield mapping",
            "Correlation type",
            "Outlier filtering",
            "MAD filtering",
            "Range filtering",
            "Scaling",
            "Normalization",
            "Imputation",
            "Constant descriptors",
            "Missing descriptors",
            "Ranking criterion",
            "Main manuscript matrix",
            "Supporting Information matrices",
            "Duplicate path-count descriptor",
            "NBO descriptors"

        ],

        "Value": [

            "19 ligands (L7-L25)",
            "Fixed experimental yields",
            "Pearson + Spearman",
            "None",
            "None",
            "None",
            "None",
            "None",
            "None",
            "Removed",
            "Excluded from same-N matrix",
            "Mean absolute correlation",
            "Top-10 Spearman",
            "Full Pearson + Full Spearman",
            "Excluded",
            "Excluded"

        ]

    })


    protocol_df.to_excel(

        writer,

        sheet_name="ANALYSIS_PROTOCOL",

        index=False

    )


# ============================================================
# 35. FINAL VERIFICATION
# ============================================================

print("\n")
print("=" * 110)
print("FINAL VERIFICATION")
print("=" * 110)


print(
    f"Final N = {N_FINAL}"
)


print(
    f"Final descriptor count = {len(descriptor_columns)}"
)


print(
    f"Full Pearson matrix = "
    f"{pearson_matrix_full.shape}"
)


print(
    f"Full Spearman matrix = "
    f"{spearman_matrix_full.shape}"
)


print(
    f"Top-10 Pearson matrix = "
    f"{pearson_matrix_top10.shape}"
)


print(
    f"Top-10 Spearman matrix = "
    f"{spearman_matrix_top10.shape}"
)


print("\n")
print("TOP 10 — FROZEN FINAL RANKING")

for i, descriptor in enumerate(
    top10_descriptors,
    start=1
):

    row = top10_ranking[
        top10_ranking[
            "Descriptor"
        ] == descriptor
    ].iloc[0]


    print(

        f"{i:2d}. "
        f"{descriptor} | "
        f"Pearson r = {row['Pearson_r']:.6f} | "
        f"Spearman ρ = {row['Spearman_rho']:.6f} | "
        f"Mean |r| = {row['Mean_Abs_Correlation']:.6f}"

    )


print("\n")
print("FILES CREATED")
print("=" * 110)

print(
    "\nMain manuscript — Top-10 Spearman:"
)

print(
    TOP10_SPEARMAN_PNG
)


print(
    "\nSupporting Information — Full Pearson:"
)

print(
    FULL_PEARSON_PNG
)


print(
    "\nSupporting Information — Full Spearman:"
)

print(
    FULL_SPEARMAN_PNG
)


print(
    "\nExcel workbook:"
)

print(
    FINAL_CORRELATION_EXCEL
)


print("\n")
print("=" * 110)
print("ANALYSIS PROTOCOL CONFIRMED")
print("=" * 110)

print(
    "✓ N = 19"
)

print(
    "✓ All real zero yields retained"
)

print(
    "✓ No IQR filtering"
)

print(
    "✓ No MAD filtering"
)

print(
    "✓ No numerical-range filtering"
)

print(
    "✓ No scaling or normalization"
)

print(
    "✓ No imputation"
)

print(
    "✓ Constant descriptors only removed"
)

print(
    "✓ Incomplete descriptors excluded from same-N matrix"
)

print(
    "✓ Palladacycle_Path_Atom_Count excluded"
)

print(
    "✓ NBO descriptors excluded"
)

print(
    "✓ Sterimol B1/B5/L retained when complete"
)

print(
    "✓ %Vbur descriptors retained when complete"
)

print(
    "✓ Distal Volume retained when complete"
)

print(
    "✓ P_int retained when complete"
)

print(
    "✓ Molecular Volume retained when complete"
)

print(
    "✓ Top 10 determined automatically"
)

print(
    "✓ Ranking criterion frozen:"
)

print(
    "  Mean |correlation| = "
    "(|Pearson r| + |Spearman ρ|) / 2"
)

print(
    "✓ Main manuscript = Top-10 Spearman"
)

print(
    "✓ SI = Full Pearson + Full Spearman"
)

print("\n")
print("=" * 110)
print("FINAL PUBLICATION CORRELATION ANALYSIS COMPLETE")
print("=" * 110)
```

```python
# ============================================================
# Pd γ-ARYLATION PROJECT
# FINAL PUBLICATION CORRELATION ANALYSIS
#
# PRIMARY:
#   - RAW / FULL dataset
#   - No IQR filtering
#   - No MAD filtering
#   - No outlier removal
#   - No scaling / normalization
#   - No imputation
#
# OUTPUT:
#   1. Full Pearson correlation matrix          -> SI
#   2. Full Spearman correlation matrix         -> SI
#   3. Top-10 descriptor ranking                -> Main/SI
#   4. Top-10 Spearman heatmap                  -> Main manuscript
#   5. Top-10 Pearson heatmap                   -> SI
#   6. Full Pearson heatmap                     -> SI
#   7. Full Spearman heatmap                    -> SI
#   8. Leave-one-ligand-out robustness analysis -> SI
#
# IMPORTANT:
#   - Palladacycle_Path_Atom_Count is COMPLETELY excluded
#   - NBO descriptors are COMPLETELY excluded
#   - Sterimol B1/B5/L are retained when available
#   - %Vbur / Distal Volume / P_int retained when available
# ============================================================


# ============================================================
# 1. INSTALL
# ============================================================

!pip -q install pandas numpy scipy openpyxl matplotlib seaborn


# ============================================================
# 2. IMPORTS
# ============================================================

import os
import re
import warnings
import numpy as np
import pandas as pd
import seaborn as sns
import matplotlib.pyplot as plt

from scipy.stats import pearsonr, spearmanr

warnings.filterwarnings("ignore")


# ============================================================
# 3. PATHS
# ============================================================

INPUT_EXCEL = "/content/drive/MyDrive/Pd_gamma_arylation_Descriptors.xlsx"

OUTPUT_DIR = "/content/drive/MyDrive/Pd_gamma_arylation_Correlation"

os.makedirs(OUTPUT_DIR, exist_ok=True)

print("=" * 80)
print("Pd γ-ARYLATION FINAL CORRELATION ANALYSIS")
print("=" * 80)

print("\nInput:")
print(INPUT_EXCEL)

print("\nOutput directory:")
print(OUTPUT_DIR)


# ============================================================
# 4. LOAD EXCEL
# ============================================================

if not os.path.exists(INPUT_EXCEL):
    raise FileNotFoundError(
        f"\nERROR: Excel file not found:\n{INPUT_EXCEL}\n\n"
        "Please change INPUT_EXCEL to your actual descriptor Excel file."
    )

xls = pd.ExcelFile(INPUT_EXCEL)

print("\nAvailable Excel sheets:")
for s in xls.sheet_names:
    print("  -", s)


# ============================================================
# 5. FIND MAIN DESCRIPTOR SHEET
# ============================================================

preferred_sheets = [
    "Master_Descriptors",
    "Master",
    "All_Descriptors",
    "Descriptors",
    "Final_Descriptors"
]

selected_sheet = None

for s in preferred_sheets:
    if s in xls.sheet_names:
        selected_sheet = s
        break

if selected_sheet is None:
    selected_sheet = xls.sheet_names[0]

print("\nSelected descriptor sheet:")
print(selected_sheet)


# ============================================================
# 6. READ DATA
# ============================================================

master_df = pd.read_excel(INPUT_EXCEL, sheet_name=selected_sheet)

print("\nOriginal dataframe shape:")
print(master_df.shape)

print("\nOriginal columns:")
for i, c in enumerate(master_df.columns, start=1):
    print(f"{i:3d}. {c}")


# ============================================================
# 7. CLEAN COLUMN NAMES
# ============================================================

master_df.columns = [
    str(c).strip()
    for c in master_df.columns
]


# ============================================================
# 8. IDENTIFY LIGAND COLUMN
# ============================================================

ligand_candidates = [
    "Ligand",
    "Ligand_ID",
    "Ligand ID",
    "ligand",
    "ligand_id",
    "ID"
]

ligand_col = None

for c in ligand_candidates:
    if c in master_df.columns:
        ligand_col = c
        break

if ligand_col is None:
    raise ValueError(
        "\nERROR: Could not find ligand identifier column.\n"
        "Expected one of:\n"
        + "\n".join(ligand_candidates)
    )

print("\nLigand identifier column:")
print(ligand_col)


# ============================================================
# 9. YIELD DATA
#
# IMPORTANT:
# Zero values are REAL experimental yields here.
# They are NOT missing values.
# ============================================================

YIELD_DATA = {
    "L7": 0,
    "L8": 0,
    "L9": 0,
    "L10": 0,
    "L11": 0,
    "L12": 0,
    "L13": 0,
    "L14": 0,
    "L15": 0,
    "L16": 12,
    "L17": 53,
    "L18": 46,
    "L19": 67,
    "L20": 40,
    "L21": 31,
    "L22": 0,
    "L23": 0,
    "L24": 0,
    "L25": 0
}


# ============================================================
# 10. STANDARDIZE LIGAND IDS
# ============================================================

def normalize_ligand_id(x):

    if pd.isna(x):
        return np.nan

    s = str(x).strip()

    # Examples:
    # L7 -> L7
    # l7 -> L7
    # L07 -> L7

    m = re.search(r"L\s*0*(\d+)", s, flags=re.IGNORECASE)

    if m:
        return f"L{int(m.group(1))}"

    return s


master_df["_Ligand_Normalized"] = (
    master_df[ligand_col]
    .apply(normalize_ligand_id)
)


# ============================================================
# 11. ADD / REPLACE YIELD
# ============================================================

master_df["Yield"] = (
    master_df["_Ligand_Normalized"]
    .map(YIELD_DATA)
)


# ============================================================
# 12. VERIFY YIELD MAPPING
# ============================================================

print("\nYield mapping check:")

yield_check = master_df[
    [ligand_col, "_Ligand_Normalized", "Yield"]
].copy()

print(yield_check.to_string(index=False))


missing_yield_ids = (
    master_df.loc[
        master_df["Yield"].isna(),
        "_Ligand_Normalized"
    ]
    .dropna()
    .unique()
)

if len(missing_yield_ids) > 0:
    raise ValueError(
        "\nERROR: Yield missing for ligand IDs:\n"
        + ", ".join(map(str, missing_yield_ids))
    )


# ============================================================
# 13. VERIFY EXACTLY 19 LIGANDS
# ============================================================

n_ligands = len(master_df)

print("\nNumber of rows / ligands:", n_ligands)

if n_ligands != 19:
    print(
        "\nWARNING: Expected 19 ligand rows, "
        f"but found {n_ligands}."
    )


# ============================================================
# 14. CHECK DUPLICATE LIGANDS
# ============================================================

duplicate_ligands = (
    master_df["_Ligand_Normalized"]
    [master_df["_Ligand_Normalized"].duplicated(keep=False)]
    .dropna()
    .unique()
)

if len(duplicate_ligands) > 0:

    print("\nWARNING: Duplicate ligand IDs detected:")
    print(duplicate_ligands)

else:

    print("\nNo duplicate ligand IDs detected.")


# ============================================================
# 15. COMPLETELY EXCLUDE UNWANTED VARIABLES
# ============================================================

EXCLUDE_EXACT = {

    # Identifier/helper
    ligand_col,
    "_Ligand_Normalized",

    # Duplicate mathematical variable
    "Palladacycle_Path_Atom_Count",

    # Metadata
    "n_unpaired",
    "N_Unpaired",
    "Charge",
    "Multiplicity",

    # NBO descriptors
    "NBO",
    "NBO_Charge",
    "NBO_Bond",
    "NBO_Bond_Order",
    "NBO_Energy",
    "NBO_Occupancy",
    "F_NBO"
}


# Exclude anything containing NBO in the name
EXCLUDE_NBO_CONTAINING = []

for c in master_df.columns:

    if "NBO" in str(c).upper():

        EXCLUDE_NBO_CONTAINING.append(c)


EXCLUDED_COLUMNS = set(EXCLUDE_EXACT)

EXCLUDED_COLUMNS.update(EXCLUDE_NBO_CONTAINING)

print("\nColumns explicitly excluded:")

for c in sorted(EXCLUDED_COLUMNS):
    if c in master_df.columns:
        print("  -", c)


# ============================================================
# 16. PREPARE NUMERIC DATAFRAME
# ============================================================

analysis_candidates = []

for c in master_df.columns:

    if c in EXCLUDED_COLUMNS:
        continue

    if c == "Yield":
        continue

    analysis_candidates.append(c)


numeric_df = pd.DataFrame(index=master_df.index)

for c in analysis_candidates:

    converted = pd.to_numeric(
        master_df[c],
        errors="coerce"
    )

    # Keep column only if at least one numerical value exists
    if converted.notna().sum() > 0:

        numeric_df[c] = converted


# Add Yield
numeric_df["Yield"] = pd.to_numeric(
    master_df["Yield"],
    errors="coerce"
)


# ============================================================
# 17. REMOVE CONSTANT DESCRIPTORS ONLY
#
# This is NOT outlier filtering.
#
# A constant descriptor cannot correlate with Yield.
# ============================================================

constant_columns = []

for c in numeric_df.columns:

    if c == "Yield":
        continue

    if numeric_df[c].dropna().nunique() <= 1:

        constant_columns.append(c)


if constant_columns:

    print("\nConstant descriptors removed:")

    for c in constant_columns:
        print("  -", c)

    numeric_df = numeric_df.drop(
        columns=constant_columns
    )

else:

    print("\nNo constant descriptors found.")


# ============================================================
# 18. REMOVE COLUMNS WITH INCOMPLETE DATA
#
# IMPORTANT:
# For the FULL square correlation matrix, every descriptor
# must have all 19 observations.
#
# NO IMPUTATION.
# ============================================================

incomplete_columns = []

for c in numeric_df.columns:

    n_valid = numeric_df[c].notna().sum()

    if n_valid != len(master_df):

        incomplete_columns.append(
            (c, n_valid)
        )


print("\nIncomplete descriptors excluded from FULL matrix:")

if incomplete_columns:

    for c, n in incomplete_columns:
        print(
            f"  - {c}: {n}/{len(master_df)} valid"
        )

else:

    print("  None")


# ============================================================
# 19. BUILD COMPLETE-CASE DATAFRAME
# ============================================================

complete_columns = []

for c in numeric_df.columns:

    if numeric_df[c].notna().sum() == len(master_df):

        complete_columns.append(c)


correlation_df = numeric_df[
    complete_columns
].copy()


# ============================================================
# 20. ENSURE YIELD EXISTS
# ============================================================

if "Yield" not in correlation_df.columns:

    raise ValueError(
        "Yield was not available in the complete dataframe."
    )


# ============================================================
# 21. VERIFY YIELD
# ============================================================

print("\nYield statistics:")
print(
    correlation_df["Yield"].describe()
)


print("\nYield values:")
print(
    correlation_df["Yield"].tolist()
)


# ============================================================
# 22. PRINT DESCRIPTOR AVAILABILITY
# ============================================================

print("\n" + "=" * 80)
print("IMPORTANT DESCRIPTOR AVAILABILITY")
print("=" * 80)

important_patterns = [
    "Sterimol",
    "Vbur",
    "Buried",
    "Distal",
    "Pint",
    "SASA",
    "Molecular_Volume",
    "Ring_Size",
    "Bite_Angle",
    "Donor1_Donor2",
    "Pd_Donor1"
]

for pattern in important_patterns:

    matches = [
        c for c in correlation_df.columns
        if pattern.lower() in c.lower()
    ]

    print(f"\n{pattern}:")

    if matches:
        for c in matches:
            print("  ✓", c)
    else:
        print("  --- not available ---")


# ============================================================
# 23. CHECK THAT PATH COUNT IS ABSENT
# ============================================================

path_count_columns = [
    c for c in master_df.columns
    if "Palladacycle_Path_Atom_Count" in str(c)
]

if path_count_columns:

    raise RuntimeError(
        "\nERROR: Palladacycle_Path_Atom_Count still exists."
    )

print(
    "\n✓ Palladacycle_Path_Atom_Count completely excluded."
)


# ============================================================
# 24. CHECK THAT NBO IS ABSENT
# ============================================================

nbo_columns_remaining = [
    c for c in correlation_df.columns
    if "NBO" in str(c).upper()
]

if nbo_columns_remaining:

    raise RuntimeError(
        "\nERROR: NBO descriptors still present:\n"
        + "\n".join(nbo_columns_remaining)
    )

print("✓ NBO descriptors completely excluded.")


# ============================================================
# 25. FINAL FULL MATRIX DATASET
# ============================================================

print("\n" + "=" * 80)
print("FULL CORRELATION DATASET")
print("=" * 80)

print(
    f"Rows: {correlation_df.shape[0]}"
)

print(
    f"Numeric variables including Yield: "
    f"{correlation_df.shape[1]}"
)

print("\nVariables included:")

for c in correlation_df.columns:

    print("  -", c)


# ============================================================
# 26. PEARSON CORRELATION MATRIX
# ============================================================

pearson_matrix = correlation_df.corr(
    method="pearson"
)


# ============================================================
# 27. SPEARMAN CORRELATION MATRIX
# ============================================================

spearman_matrix = correlation_df.corr(
    method="spearman"
)


# ============================================================
# 28. VERIFY MATRIX DIMENSIONS
# ============================================================

print("\nPearson matrix:")
print(pearson_matrix.shape)

print("Spearman matrix:")
print(spearman_matrix.shape)


# ============================================================
# 29. SAVE FULL MATRICES
# ============================================================

pearson_matrix.to_excel(
    os.path.join(
        OUTPUT_DIR,
        "SI_FULL_PEARSON_Correlation_Matrix.xlsx"
    )
)

spearman_matrix.to_excel(
    os.path.join(
        OUTPUT_DIR,
        "SI_FULL_SPEARMAN_Correlation_Matrix.xlsx"
    )
)


# ============================================================
# 30. CORRELATION WITH YIELD
# ============================================================

yield_pearson = []
yield_spearman = []

predictor_columns = [
    c for c in correlation_df.columns
    if c != "Yield"
]


for descriptor in predictor_columns:

    x = correlation_df[descriptor]
    y = correlation_df["Yield"]

    # Pearson
    r, p_r = pearsonr(x, y)

    # Spearman
    rho, p_rho = spearmanr(x, y)

    yield_pearson.append({
        "Descriptor": descriptor,
        "N": len(x),
        "Pearson_r": r,
        "Pearson_p": p_r
    })

    yield_spearman.append({
        "Descriptor": descriptor,
        "N": len(x),
        "Spearman_rho": rho,
        "Spearman_p": p_rho
    })


pearson_yield_df = pd.DataFrame(
    yield_pearson
)

spearman_yield_df = pd.DataFrame(
    yield_spearman
)


# ============================================================
# 31. MERGE PEARSON + SPEARMAN
# ============================================================

ranking_df = pd.merge(
    pearson_yield_df,
    spearman_yield_df,
    on=["Descriptor", "N"],
    how="inner"
)


# ============================================================
# 32. MEAN ABSOLUTE CORRELATION
# ============================================================

ranking_df["Mean_Abs_Correlation"] = (
    ranking_df["Pearson_r"].abs()
    +
    ranking_df["Spearman_rho"].abs()
) / 2.0


# ============================================================
# 33. SORT BY MEAN ABSOLUTE CORRELATION
# ============================================================

ranking_df = ranking_df.sort_values(
    by="Mean_Abs_Correlation",
    ascending=False
).reset_index(drop=True)


ranking_df.insert(
    0,
    "Rank",
    np.arange(
        1,
        len(ranking_df) + 1
    )
)


# ============================================================
# 34. SAVE COMPLETE RANKING
# ============================================================

ranking_path = os.path.join(
    OUTPUT_DIR,
    "FULL_Yield_Correlation_Ranking.xlsx"
)

ranking_df.to_excel(
    ranking_path,
    index=False
)


# ============================================================
# 35. DISPLAY TOP 20
# ============================================================

print("\n" + "=" * 80)
print("TOP DESCRIPTORS ASSOCIATED WITH YIELD")
print("=" * 80)

display(
    ranking_df.head(20).round(6)
)


# ============================================================
# 36. TOP 10
# ============================================================

TOP_N = 10

top10_df = ranking_df.head(TOP_N).copy()


print("\n" + "=" * 80)
print("TOP 10 DESCRIPTORS")
print("=" * 80)

display(
    top10_df.round(6)
)


# ============================================================
# 37. TOP-10 DESCRIPTOR LIST
# ============================================================

top10_descriptors = (
    top10_df["Descriptor"]
    .tolist()
)

print("\nTop-10 descriptor list:")

for i, descriptor in enumerate(
    top10_descriptors,
    start=1
):

    print(
        f"{i:2d}. {descriptor}"
    )


# ============================================================
# 38. TOP-10 SPEARMAN MATRIX
# ============================================================

top10_spearman_columns = [
    "Yield"
] + top10_descriptors


top10_spearman_matrix = spearman_matrix.loc[
    top10_spearman_columns,
    top10_spearman_columns
]


# ============================================================
# 39. TOP-10 PEARSON MATRIX
# ============================================================

top10_pearson_matrix = pearson_matrix.loc[
    top10_spearman_columns,
    top10_spearman_columns
]


# ============================================================
# 40. SAVE TOP-10 MATRICES
# ============================================================

top10_spearman_matrix.to_excel(
    os.path.join(
        OUTPUT_DIR,
        "MAIN_TOP10_SPEARMAN_Correlation_Matrix.xlsx"
    )
)

top10_pearson_matrix.to_excel(
    os.path.join(
        OUTPUT_DIR,
        "SI_TOP10_PEARSON_Correlation_Matrix.xlsx"
    )
)


# ============================================================
# 41. HEATMAP FUNCTION
# ============================================================

def make_heatmap(
    matrix,
    title,
    filename,
    figsize=(12, 10)
):

    plt.figure(
        figsize=figsize
    )

    sns.heatmap(
        matrix,
        cmap="RdBu_r",
        vmin=-1,
        vmax=1,
        center=0,
        square=True,
        annot=True,
        fmt=".2f",
        linewidths=0.5,
        cbar_kws={
            "label": "Correlation coefficient"
        }
    )

    plt.title(
        title,
        fontsize=15,
        fontweight="bold",
        pad=15
    )

    plt.xticks(
        rotation=70,
        ha="right",
        fontsize=9
    )

    plt.yticks(
        rotation=0,
        fontsize=9
    )

    plt.tight_layout()

    path = os.path.join(
        OUTPUT_DIR,
        filename
    )

    plt.savefig(
        path,
        dpi=600,
        bbox_inches="tight"
    )

    plt.show()

    print(
        "\nSaved:",
        path
    )


# ============================================================
# 42. MAIN MANUSCRIPT:
# TOP-10 SPEARMAN HEATMAP
# ============================================================

make_heatmap(
    top10_spearman_matrix,
    "Spearman Correlation: Yield and Top-10 Ligand Descriptors",
    "MAIN_MANUSCRIPT_TOP10_SPEARMAN_Correlation.png",
    figsize=(12, 10)
)


# ============================================================
# 43. SI:
# TOP-10 PEARSON HEATMAP
# ============================================================

make_heatmap(
    top10_pearson_matrix,
    "Pearson Correlation: Yield and Top-10 Ligand Descriptors",
    "SI_TOP10_PEARSON_Correlation.png",
    figsize=(12, 10)
)


# ============================================================
# 44. SI:
# FULL PEARSON HEATMAP
# ============================================================

make_heatmap(
    pearson_matrix,
    "Full Pearson Correlation Matrix",
    "SI_FULL_PEARSON_Correlation_Matrix.png",
    figsize=(
        max(14, len(pearson_matrix.columns) * 0.35),
        max(12, len(pearson_matrix.columns) * 0.35)
    )
)


# ============================================================
# 45. SI:
# FULL SPEARMAN HEATMAP
# ============================================================

make_heatmap(
    spearman_matrix,
    "Full Spearman Correlation Matrix",
    "SI_FULL_SPEARMAN_Correlation_Matrix.png",
    figsize=(
        max(14, len(spearman_matrix.columns) * 0.35),
        max(12, len(spearman_matrix.columns) * 0.35)
    )
)


# ============================================================
# 46. LEAVE-ONE-LIGAND-OUT ROBUSTNESS ANALYSIS
#
# This is the preferred robustness test.
#
# For each descriptor:
#   remove L7 -> calculate correlation
#   remove L8 -> calculate correlation
#   ...
#   remove L25
#
# Then determine:
#   - minimum correlation
#   - maximum correlation
#   - mean
#   - median
#   - standard deviation
#   - sign changes
# ============================================================

print("\n" + "=" * 80)
print("LEAVE-ONE-LIGAND-OUT ROBUSTNESS ANALYSIS")
print("=" * 80)


loo_results = []


for descriptor in predictor_columns:

    full_r = ranking_df.loc[
        ranking_df["Descriptor"] == descriptor,
        "Pearson_r"
    ].iloc[0]

    full_rho = ranking_df.loc[
        ranking_df["Descriptor"] == descriptor,
        "Spearman_rho"
    ].iloc[0]

    loo_pearson = []
    loo_spearman = []

    for idx in correlation_df.index:

        subset = correlation_df.drop(
            index=idx
        )

        x = subset[descriptor]
        y = subset["Yield"]

        r, _ = pearsonr(
            x,
            y
        )

        rho, _ = spearmanr(
            x,
            y
        )

        loo_pearson.append(r)
        loo_spearman.append(rho)


    loo_pearson = np.array(
        loo_pearson
    )

    loo_spearman = np.array(
        loo_spearman
    )


    # Sign changes relative to full dataset
    pearson_sign_changes = np.sum(
        np.sign(loo_pearson)
        !=
        np.sign(full_r)
    )

    spearman_sign_changes = np.sum(
        np.sign(loo_spearman)
        !=
        np.sign(full_rho)
    )


    loo_results.append({

        "Descriptor": descriptor,

        "Full_Pearson_r": full_r,

        "LOO_Pearson_Min": np.min(
            loo_pearson
        ),

        "LOO_Pearson_Max": np.max(
            loo_pearson
        ),

        "LOO_Pearson_Mean": np.mean(
            loo_pearson
        ),

        "LOO_Pearson_Median": np.median(
            loo_pearson
        ),

        "LOO_Pearson_SD": np.std(
            loo_pearson,
            ddof=1
        ),

        "Pearson_Sign_Changes": pearson_sign_changes,

        "Full_Spearman_rho": full_rho,

        "LOO_Spearman_Min": np.min(
            loo_spearman
        ),

        "LOO_Spearman_Max": np.max(
            loo_spearman
        ),

        "LOO_Spearman_Mean": np.mean(
            loo_spearman
        ),

        "LOO_Spearman_Median": np.median(
            loo_spearman
        ),

        "LOO_Spearman_SD": np.std(
            loo_spearman,
            ddof=1
        ),

        "Spearman_Sign_Changes": spearman_sign_changes
    })


loo_df = pd.DataFrame(
    loo_results
)


# ============================================================
# 47. ROBUSTNESS SCORE
#
# A simple descriptive score:
#
# mean of absolute LOO Pearson and Spearman correlations
#
# This is NOT a new statistical test.
# It is only a stability summary.
# ============================================================

loo_df["LOO_Mean_Abs_Pearson"] = (
    loo_df["LOO_Pearson_Mean"].abs()
)

loo_df["LOO_Mean_Abs_Spearman"] = (
    loo_df["LOO_Spearman_Mean"].abs()
)

loo_df["LOO_Robustness_Mean_Abs"] = (
    loo_df["LOO_Mean_Abs_Pearson"]
    +
    loo_df["LOO_Mean_Abs_Spearman"]
) / 2.0


# ============================================================
# 48. SORT ROBUSTNESS RESULTS
# ============================================================

loo_df = loo_df.sort_values(
    by="LOO_Robustness_Mean_Abs",
    ascending=False
).reset_index(drop=True)


# ============================================================
# 49. SAVE LOO RESULTS
# ============================================================

loo_path = os.path.join(
    OUTPUT_DIR,
    "SI_LEAVE_ONE_LIGAND_OUT_Robustness.xlsx"
)

loo_df.to_excel(
    loo_path,
    index=False
)


print("\nLeave-one-ligand-out results saved:")
print(loo_path)


# ============================================================
# 50. DISPLAY ROBUSTNESS OF TOP-10
# ============================================================

top10_robustness = loo_df[
    loo_df["Descriptor"].isin(
        top10_descriptors
    )
].copy()


print("\n" + "=" * 80)
print("ROBUSTNESS OF TOP-10 DESCRIPTORS")
print("=" * 80)

display(
    top10_robustness.round(5)
)


# ============================================================
# 51. CREATE TOP-10 ROBUSTNESS SUMMARY
# ============================================================

top10_summary = pd.merge(
    top10_df,
    top10_robustness,
    on="Descriptor",
    how="left",
    suffixes=(
        "_Full",
        "_LOO"
    )
)


# ============================================================
# 52. SAVE TOP-10 SUMMARY
# ============================================================

top10_summary_path = os.path.join(
    OUTPUT_DIR,
    "TOP10_Descriptor_Summary_and_Robustness.xlsx"
)

top10_summary.to_excel(
    top10_summary_path,
    index=False
)


# ============================================================
# 53. OPTIONAL:
# FDR CORRECTION
#
# We do NOT use FDR to rank descriptors.
#
# It is reported separately because many descriptors
# are tested simultaneously.
# ============================================================

try:

    from statsmodels.stats.multitest import multipletests

except ImportError:

    !pip -q install statsmodels

    from statsmodels.stats.multitest import multipletests


# Pearson FDR
pearson_fdr = multipletests(
    ranking_df["Pearson_p"],
    method="fdr_bh"
)

ranking_df["Pearson_FDR_q"] = (
    pearson_fdr[1]
)


# Spearman FDR
spearman_fdr = multipletests(
    ranking_df["Spearman_p"],
    method="fdr_bh"
)

ranking_df["Spearman_FDR_q"] = (
    spearman_fdr[1]
)


# ============================================================
# 54. RE-SORT AFTER ADDING FDR
#
# Ranking remains based ONLY on Mean_Abs_Correlation.
# ============================================================

ranking_df = ranking_df.sort_values(
    by="Mean_Abs_Correlation",
    ascending=False
).reset_index(drop=True)

ranking_df["Rank"] = np.arange(
    1,
    len(ranking_df) + 1
)


# ============================================================
# 55. SAVE FINAL RANKING WITH FDR
# ============================================================

ranking_df.to_excel(
    os.path.join(
        OUTPUT_DIR,
        "FULL_Yield_Correlation_Ranking_with_FDR.xlsx"
    ),
    index=False
)


# ============================================================
# 56. UPDATE TOP-10 AFTER FDR ADDITION
# ============================================================

top10_df = ranking_df.head(
    TOP_N
).copy()


# ============================================================
# 57. PRINT FINAL TOP-10 TABLE
# ============================================================

print("\n" + "=" * 80)
print("FINAL TOP-10 — RAW / FULL DATASET")
print("=" * 80)

display(
    top10_df[
        [
            "Rank",
            "Descriptor",
            "Pearson_r",
            "Pearson_p",
            "Pearson_FDR_q",
            "Spearman_rho",
            "Spearman_p",
            "Spearman_FDR_q",
            "Mean_Abs_Correlation"
        ]
    ].round(6)
)


# ============================================================
# 58. IDENTIFY STATISTICALLY NOTABLE DESCRIPTORS
# ============================================================

print("\n" + "=" * 80)
print("DESCRIPTORS WITH SPEARMAN q < 0.05")
print("=" * 80)

sig_spearman = ranking_df[
    ranking_df["Spearman_FDR_q"] < 0.05
]

if len(sig_spearman) > 0:

    display(
        sig_spearman[
            [
                "Rank",
                "Descriptor",
                "Spearman_rho",
                "Spearman_p",
                "Spearman_FDR_q"
            ]
        ].round(6)
    )

else:

    print(
        "No descriptor passes Spearman FDR < 0.05."
    )


# ============================================================
# 59. FINAL EXCEL WORKBOOK
# ============================================================

final_excel = os.path.join(
    OUTPUT_DIR,
    "Pd_gamma_arylation_FINAL_Publication_Correlation.xlsx"
)


# ------------------------------------------------------------
# Raw complete data
# ------------------------------------------------------------

raw_complete_output = master_df.copy()


# ------------------------------------------------------------
# Excluded descriptor table
# ------------------------------------------------------------

excluded_rows = []

for c, n in incomplete_columns:

    excluded_rows.append({

        "Descriptor": c,

        "Reason":
            "Incomplete values; excluded from strict full matrix",

        "Valid_N": n,

        "Expected_N": len(master_df)

    })


for c in constant_columns:

    excluded_rows.append({

        "Descriptor": c,

        "Reason":
            "Constant descriptor",

        "Valid_N":
            numeric_df[c].notna().sum()
            if c in numeric_df.columns
            else 0,

        "Expected_N": len(master_df)

    })


for c in sorted(
    set(EXCLUDED_COLUMNS)
    & set(master_df.columns)
):

    excluded_rows.append({

        "Descriptor": c,

        "Reason":
            "Explicitly excluded by protocol",

        "Valid_N":
            master_df[c].notna().sum(),

        "Expected_N": len(master_df)

    })


excluded_df = pd.DataFrame(
    excluded_rows
)


# ------------------------------------------------------------
# Protocol
# ------------------------------------------------------------

protocol_df = pd.DataFrame({

    "Item": [

        "Project",

        "Dataset",

        "Number of ligands",

        "Yield values",

        "Primary correlation",

        "Secondary correlation",

        "Outlier filtering",

        "IQR filtering",

        "MAD filtering",

        "Scaling",

        "Normalization",

        "Imputation",

        "Constant descriptors",

        "Incomplete descriptors",

        "Duplicate path-count descriptor",

        "NBO descriptors",

        "Main manuscript figure",

        "Supporting Information",

        "Ranking criterion",

        "Robustness analysis"

    ],

    "Description": [

        "Pd γ-arylation",

        "Ligand-dependent descriptor dataset",

        str(len(master_df)),

        "Experimental yields; zero values retained as real observations",

        "Spearman",

        "Pearson",

        "None",

        "None",

        "None",

        "None",

        "None",

        "None",

        "Removed",

        "Excluded from strict complete-case full matrix",

        "Palladacycle_Path_Atom_Count completely excluded",

        "Completely excluded",

        "Top-10 Spearman correlation heatmap",

        "Full Pearson + Full Spearman matrices",

        "Mean absolute correlation = (|Pearson| + |Spearman|) / 2",

        "Leave-one-ligand-out analysis"

    ]

})


# ============================================================
# 60. WRITE FINAL MULTI-SHEET EXCEL
# ============================================================

with pd.ExcelWriter(
    final_excel,
    engine="openpyxl"
) as writer:

    raw_complete_output.to_excel(
        writer,
        sheet_name="Raw_Complete_Data",
        index=False
    )

    correlation_df.to_excel(
        writer,
        sheet_name="Correlation_Data",
        index=False
    )

    pearson_matrix.to_excel(
        writer,
        sheet_name="FULL_Pearson",
        index=True
    )

    spearman_matrix.to_excel(
        writer,
        sheet_name="FULL_Spearman",
        index=True
    )

    ranking_df.to_excel(
        writer,
        sheet_name="Yield_Ranking",
        index=False
    )

    top10_df.to_excel(
        writer,
        sheet_name="TOP10",
        index=False
    )

    top10_spearman_matrix.to_excel(
        writer,
        sheet_name="TOP10_Spearman",
        index=True
    )

    top10_pearson_matrix.to_excel(
        writer,
        sheet_name="TOP10_Pearson",
        index=True
    )

    loo_df.to_excel(
        writer,
        sheet_name="LOO_Robustness",
        index=False
    )

    top10_summary.to_excel(
        writer,
        sheet_name="TOP10_Robustness",
        index=False
    )

    excluded_df.to_excel(
        writer,
        sheet_name="Excluded_Descriptors",
        index=False
    )

    protocol_df.to_excel(
        writer,
        sheet_name="Analysis_Protocol",
        index=False
    )


# ============================================================
# 61. FINAL SUMMARY
# ============================================================

print("\n\n")
print("=" * 80)
print("ANALYSIS COMPLETE")
print("=" * 80)

print("\nOutput directory:")
print(OUTPUT_DIR)

print("\nFinal Excel:")
print(final_excel)

print("\nMain manuscript figure:")
print(
    os.path.join(
        OUTPUT_DIR,
        "MAIN_MANUSCRIPT_TOP10_SPEARMAN_Correlation.png"
    )
)

print("\nSupporting Information figures:")

print(
    os.path.join(
        OUTPUT_DIR,
        "SI_TOP10_PEARSON_Correlation.png"
    )
)

print(
    os.path.join(
        OUTPUT_DIR,
        "SI_FULL_PEARSON_Correlation_Matrix.png"
    )
)

print(
    os.path.join(
        OUTPUT_DIR,
        "SI_FULL_SPEARMAN_Correlation_Matrix.png"
    )

)

print("\nRobustness:")
print(
    os.path.join(
        OUTPUT_DIR,
        "SI_LEAVE_ONE_LIGAND_OUT_Robustness.xlsx"
    )
)

print("\n" + "=" * 80)
print("FINAL TOP-10")
print("=" * 80)

for _, row in top10_df.iterrows():

    print(
        f"{int(row['Rank']):2d}. "
        f"{row['Descriptor']} | "
        f"Pearson r = {row['Pearson_r']:.3f} | "
        f"Spearman ρ = {row['Spearman_rho']:.3f} | "
        f"Mean |r| = {row['Mean_Abs_Correlation']:.3f}"
    )

print("\n" + "=" * 80)
print("PROTOCOL CHECK")
print("=" * 80)

print("✓ RAW / FULL dataset")
print("✓ No IQR filtering")
print("✓ No MAD filtering")
print("✓ No outlier removal")
print("✓ No scaling")
print("✓ No normalization")
print("✓ No imputation")
print("✓ Constant descriptors removed only")
print("✓ Incomplete descriptors excluded from strict full matrix")
print("✓ Palladacycle_Path_Atom_Count excluded")
print("✓ NBO descriptors excluded")
print("✓ Sterimol retained when available")
print("✓ %Vbur retained when available")
print("✓ Distal Volume retained when available")
print("✓ P_int retained when available")
print("✓ Full Pearson matrix")
print("✓ Full Spearman matrix")
print("✓ Top-10 ranking")
print("✓ Top-10 Spearman heatmap")
print("✓ Leave-one-ligand-out robustness")

print("\nDONE.")
```

```python
# ============================================================
# Pd γ-ARYLATION PROJECT
# FINAL PUBLICATION CORRELATION ANALYSIS
#
# PRIMARY:
#   - RAW / FULL dataset
#   - No IQR filtering
#   - No MAD filtering
#   - No outlier removal
#   - No scaling / normalization
#   - No imputation
#
# OUTPUT:
#   1. Full Pearson correlation matrix          -> SI
#   2. Full Spearman correlation matrix         -> SI
#   3. Top-10 descriptor ranking                -> Main/SI
#   4. Top-10 Spearman heatmap                  -> Main manuscript
#   5. Top-10 Pearson heatmap                   -> SI
#   6. Full Pearson heatmap                     -> SI
#   7. Full Spearman heatmap                    -> SI
#   8. Leave-one-ligand-out robustness analysis -> SI
#
# IMPORTANT:
#   - Palladacycle_Path_Atom_Count is COMPLETELY excluded
#   - NBO descriptors are COMPLETELY excluded
#   - Sterimol B1/B5/L are retained when available
#   - %Vbur / Distal Volume / P_int retained when available
# ============================================================


# ============================================================
# 1. INSTALL
# ============================================================

!pip -q install pandas numpy scipy openpyxl matplotlib seaborn


# ============================================================
# 2. IMPORTS
# ============================================================

import os
import re
import warnings
import numpy as np
import pandas as pd
import seaborn as sns
import matplotlib.pyplot as plt

from scipy.stats import pearsonr, spearmanr

warnings.filterwarnings("ignore")


# ============================================================
# 3. PATHS
# ============================================================

INPUT_EXCEL = "/content/drive/MyDrive/Pd_gamma_arylation_RESULTS/Pd_gamma_arylation_COMPLETE_ANALYSIS.xlsx"

OUTPUT_DIR = "/content/drive/MyDrive/Pd_gamma_arylation_Correlation"

os.makedirs(OUTPUT_DIR, exist_ok=True)

print("=" * 80)
print("Pd γ-ARYLATION FINAL CORRELATION ANALYSIS")
print("=" * 80)

print("\nInput:")
print(INPUT_EXCEL)

print("\nOutput directory:")
print(OUTPUT_DIR)


# ============================================================
# 4. LOAD EXCEL
# ============================================================

if not os.path.exists(INPUT_EXCEL):
    raise FileNotFoundError(
        f"\nERROR: Excel file not found:\n{INPUT_EXCEL}\n\n"
        "Please change INPUT_EXCEL to your actual descriptor Excel file."
    )

xls = pd.ExcelFile(INPUT_EXCEL)

print("\nAvailable Excel sheets:")
for s in xls.sheet_names:
    print("  -", s)


# ============================================================
# 5. FIND MAIN DESCRIPTOR SHEET
# ============================================================

preferred_sheets = [
    "Master_Descriptors",
    "Master",
    "All_Descriptors",
    "Descriptors",
    "Final_Descriptors"
]

selected_sheet = None

for s in preferred_sheets:
    if s in xls.sheet_names:
        selected_sheet = s
        break

if selected_sheet is None:
    selected_sheet = xls.sheet_names[0]

print("\nSelected descriptor sheet:")
print(selected_sheet)


# ============================================================
# 6. READ DATA
# ============================================================

master_df = pd.read_excel(INPUT_EXCEL, sheet_name=selected_sheet)

print("\nOriginal dataframe shape:")
print(master_df.shape)

print("\nOriginal columns:")
for i, c in enumerate(master_df.columns, start=1):
    print(f"{i:3d}. {c}")


# ============================================================
# 7. CLEAN COLUMN NAMES
# ============================================================

master_df.columns = [
    str(c).strip()
    for c in master_df.columns
]


# ============================================================
# 8. IDENTIFY LIGAND COLUMN
# ============================================================

ligand_candidates = [
    "Ligand",
    "Ligand_ID",
    "Ligand ID",
    "ligand",
    "ligand_id",
    "ID"
]

ligand_col = None

for c in ligand_candidates:
    if c in master_df.columns:
        ligand_col = c
        break

if ligand_col is None:
    raise ValueError(
        "\nERROR: Could not find ligand identifier column.\n"
        "Expected one of:\n"
        + "\n".join(ligand_candidates)
    )

print("\nLigand identifier column:")
print(ligand_col)


# ============================================================
# 9. YIELD DATA
#
# IMPORTANT:
# Zero values are REAL experimental yields here.
# They are NOT missing values.
# ============================================================

YIELD_DATA = {
    "L7": 0,
    "L8": 0,
    "L9": 0,
    "L10": 0,
    "L11": 0,
    "L12": 0,
    "L13": 0,
    "L14": 0,
    "L15": 0,
    "L16": 12,
    "L17": 53,
    "L18": 46,
    "L19": 67,
    "L20": 40,
    "L21": 31,
    "L22": 0,
    "L23": 0,
    "L24": 0,
    "L25": 0
}


# ============================================================
# 10. STANDARDIZE LIGAND IDS
# ============================================================

def normalize_ligand_id(x):

    if pd.isna(x):
        return np.nan

    s = str(x).strip()

    # Examples:
    # L7 -> L7
    # l7 -> L7
    # L07 -> L7

    m = re.search(r"L\s*0*(\d+)", s, flags=re.IGNORECASE)

    if m:
        return f"L{int(m.group(1))}"

    return s


master_df["_Ligand_Normalized"] = (
    master_df[ligand_col]
    .apply(normalize_ligand_id)
)


# ============================================================
# 11. ADD / REPLACE YIELD
# ============================================================

master_df["Yield"] = (
    master_df["_Ligand_Normalized"]
    .map(YIELD_DATA)
)


# ============================================================
# 12. VERIFY YIELD MAPPING
# ============================================================

print("\nYield mapping check:")

yield_check = master_df[
    [ligand_col, "_Ligand_Normalized", "Yield"]
].copy()

print(yield_check.to_string(index=False))


missing_yield_ids = (
    master_df.loc[
        master_df["Yield"].isna(),
        "_Ligand_Normalized"
    ]
    .dropna()
    .unique()
)

if len(missing_yield_ids) > 0:
    raise ValueError(
        "\nERROR: Yield missing for ligand IDs:\n"
        + ", ".join(map(str, missing_yield_ids))
    )


# ============================================================
# 13. VERIFY EXACTLY 19 LIGANDS
# ============================================================

n_ligands = len(master_df)

print("\nNumber of rows / ligands:", n_ligands)

if n_ligands != 19:
    print(
        "\nWARNING: Expected 19 ligand rows, "
        f"but found {n_ligands}."
    )


# ============================================================
# 14. CHECK DUPLICATE LIGANDS
# ============================================================

duplicate_ligands = (
    master_df["_Ligand_Normalized"]
    [master_df["_Ligand_Normalized"].duplicated(keep=False)]
    .dropna()
    .unique()
)

if len(duplicate_ligands) > 0:

    print("\nWARNING: Duplicate ligand IDs detected:")
    print(duplicate_ligands)

else:

    print("\nNo duplicate ligand IDs detected.")


# ============================================================
# 15. COMPLETELY EXCLUDE UNWANTED VARIABLES
# ============================================================

EXCLUDE_EXACT = {

    # Identifier/helper
    ligand_col,
    "_Ligand_Normalized",

    # Duplicate mathematical variable
    "Palladacycle_Path_Atom_Count",

    # Metadata
    "n_unpaired",
    "N_Unpaired",
    "Charge",
    "Multiplicity",

    # NBO descriptors
    "NBO",
    "NBO_Charge",
    "NBO_Bond",
    "NBO_Bond_Order",
    "NBO_Energy",
    "NBO_Occupancy",
    "F_NBO"
}


# Exclude anything containing NBO in the name
EXCLUDE_NBO_CONTAINING = []

for c in master_df.columns:

    if "NBO" in str(c).upper():

        EXCLUDE_NBO_CONTAINING.append(c)


EXCLUDED_COLUMNS = set(EXCLUDE_EXACT)

EXCLUDED_COLUMNS.update(EXCLUDE_NBO_CONTAINING)

print("\nColumns explicitly excluded:")

for c in sorted(EXCLUDED_COLUMNS):
    if c in master_df.columns:
        print("  -", c)


# ============================================================
# 16. PREPARE NUMERIC DATAFRAME
# ============================================================

analysis_candidates = []

for c in master_df.columns:

    if c in EXCLUDED_COLUMNS:
        continue

    if c == "Yield":
        continue

    analysis_candidates.append(c)


numeric_df = pd.DataFrame(index=master_df.index)

for c in analysis_candidates:

    converted = pd.to_numeric(
        master_df[c],
        errors="coerce"
    )

    # Keep column only if at least one numerical value exists
    if converted.notna().sum() > 0:

        numeric_df[c] = converted


# Add Yield
numeric_df["Yield"] = pd.to_numeric(
    master_df["Yield"],
    errors="coerce"
)


# ============================================================
# 17. REMOVE CONSTANT DESCRIPTORS ONLY
#
# This is NOT outlier filtering.
#
# A constant descriptor cannot correlate with Yield.
# ============================================================

constant_columns = []

for c in numeric_df.columns:

    if c == "Yield":
        continue

    if numeric_df[c].dropna().nunique() <= 1:

        constant_columns.append(c)


if constant_columns:

    print("\nConstant descriptors removed:")

    for c in constant_columns:
        print("  -", c)

    numeric_df = numeric_df.drop(
        columns=constant_columns
    )

else:

    print("\nNo constant descriptors found.")


# ============================================================
# 18. REMOVE COLUMNS WITH INCOMPLETE DATA
#
# IMPORTANT:
# For the FULL square correlation matrix, every descriptor
# must have all 19 observations.
#
# NO IMPUTATION.
# ============================================================

incomplete_columns = []

for c in numeric_df.columns:

    n_valid = numeric_df[c].notna().sum()

    if n_valid != len(master_df):

        incomplete_columns.append(
            (c, n_valid)
        )


print("\nIncomplete descriptors excluded from FULL matrix:")

if incomplete_columns:

    for c, n in incomplete_columns:
        print(
            f"  - {c}: {n}/{len(master_df)} valid"
        )

else:

    print("  None")


# ============================================================
# 19. BUILD COMPLETE-CASE DATAFRAME
# ============================================================

complete_columns = []

for c in numeric_df.columns:

    if numeric_df[c].notna().sum() == len(master_df):

        complete_columns.append(c)


correlation_df = numeric_df[
    complete_columns
].copy()


# ============================================================
# 20. ENSURE YIELD EXISTS
# ============================================================

if "Yield" not in correlation_df.columns:

    raise ValueError(
        "Yield was not available in the complete dataframe."
    )


# ============================================================
# 21. VERIFY YIELD
# ============================================================

print("\nYield statistics:")
print(
    correlation_df["Yield"].describe()
)


print("\nYield values:")
print(
    correlation_df["Yield"].tolist()
)


# ============================================================
# 22. PRINT DESCRIPTOR AVAILABILITY
# ============================================================

print("\n" + "=" * 80)
print("IMPORTANT DESCRIPTOR AVAILABILITY")
print("=" * 80)

important_patterns = [
    "Sterimol",
    "Vbur",
    "Buried",
    "Distal",
    "Pint",
    "SASA",
    "Molecular_Volume",
    "Ring_Size",
    "Bite_Angle",
    "Donor1_Donor2",
    "Pd_Donor1"
]

for pattern in important_patterns:

    matches = [
        c for c in correlation_df.columns
        if pattern.lower() in c.lower()
    ]

    print(f"\n{pattern}:")

    if matches:
        for c in matches:
            print("  ✓", c)
    else:
        print("  --- not available ---")


# ============================================================
# 23. CHECK THAT PATH COUNT IS ABSENT
# ============================================================

path_count_columns = [
    c for c in master_df.columns
    if "Palladacycle_Path_Atom_Count" in str(c)
]

if path_count_columns:

    raise RuntimeError(
        "\nERROR: Palladacycle_Path_Atom_Count still exists."
    )

print(
    "\n✓ Palladacycle_Path_Atom_Count completely excluded."
)


# ============================================================
# 24. CHECK THAT NBO IS ABSENT
# ============================================================

nbo_columns_remaining = [
    c for c in correlation_df.columns
    if "NBO" in str(c).upper()
]

if nbo_columns_remaining:

    raise RuntimeError(
        "\nERROR: NBO descriptors still present:\n"
        + "\n".join(nbo_columns_remaining)
    )

print("✓ NBO descriptors completely excluded.")


# ============================================================
# 25. FINAL FULL MATRIX DATASET
# ============================================================

print("\n" + "=" * 80)
print("FULL CORRELATION DATASET")
print("=" * 80)

print(
    f"Rows: {correlation_df.shape[0]}"
)

print(
    f"Numeric variables including Yield: "
    f"{correlation_df.shape[1]}"
)

print("\nVariables included:")

for c in correlation_df.columns:

    print("  -", c)


# ============================================================
# 26. PEARSON CORRELATION MATRIX
# ============================================================

pearson_matrix = correlation_df.corr(
    method="pearson"
)


# ============================================================
# 27. SPEARMAN CORRELATION MATRIX
# ============================================================

spearman_matrix = correlation_df.corr(
    method="spearman"
)


# ============================================================
# 28. VERIFY MATRIX DIMENSIONS
# ============================================================

print("\nPearson matrix:")
print(pearson_matrix.shape)

print("Spearman matrix:")
print(spearman_matrix.shape)


# ============================================================
# 29. SAVE FULL MATRICES
# ============================================================

pearson_matrix.to_excel(
    os.path.join(
        OUTPUT_DIR,
        "SI_FULL_PEARSON_Correlation_Matrix.xlsx"
    )
)

spearman_matrix.to_excel(
    os.path.join(
        OUTPUT_DIR,
        "SI_FULL_SPEARMAN_Correlation_Matrix.xlsx"
    )
)


# ============================================================
# 30. CORRELATION WITH YIELD
# ============================================================

yield_pearson = []
yield_spearman = []

predictor_columns = [
    c for c in correlation_df.columns
    if c != "Yield"
]


for descriptor in predictor_columns:

    x = correlation_df[descriptor]
    y = correlation_df["Yield"]

    # Pearson
    r, p_r = pearsonr(x, y)

    # Spearman
    rho, p_rho = spearmanr(x, y)

    yield_pearson.append({
        "Descriptor": descriptor,
        "N": len(x),
        "Pearson_r": r,
        "Pearson_p": p_r
    })

    yield_spearman.append({
        "Descriptor": descriptor,
        "N": len(x),
        "Spearman_rho": rho,
        "Spearman_p": p_rho
    })


pearson_yield_df = pd.DataFrame(
    yield_pearson
)

spearman_yield_df = pd.DataFrame(
    yield_spearman
)


# ============================================================
# 31. MERGE PEARSON + SPEARMAN
# ============================================================

ranking_df = pd.merge(
    pearson_yield_df,
    spearman_yield_df,
    on=["Descriptor", "N"],
    how="inner"
)


# ============================================================
# 32. MEAN ABSOLUTE CORRELATION
# ============================================================

ranking_df["Mean_Abs_Correlation"] = (
    ranking_df["Pearson_r"].abs()
    +
    ranking_df["Spearman_rho"].abs()
) / 2.0


# ============================================================
# 33. SORT BY MEAN ABSOLUTE CORRELATION
# ============================================================

ranking_df = ranking_df.sort_values(
    by="Mean_Abs_Correlation",
    ascending=False
).reset_index(drop=True)


ranking_df.insert(
    0,
    "Rank",
    np.arange(
        1,
        len(ranking_df) + 1
    )
)


# ============================================================
# 34. SAVE COMPLETE RANKING
# ============================================================

ranking_path = os.path.join(
    OUTPUT_DIR,
    "FULL_Yield_Correlation_Ranking.xlsx"
)

ranking_df.to_excel(
    ranking_path,
    index=False
)


# ============================================================
# 35. DISPLAY TOP 20
# ============================================================

print("\n" + "=" * 80)
print("TOP DESCRIPTORS ASSOCIATED WITH YIELD")
print("=" * 80)

display(
    ranking_df.head(20).round(6)
)


# ============================================================
# 36. TOP 10
# ============================================================

TOP_N = 10

top10_df = ranking_df.head(TOP_N).copy()


print("\n" + "=" * 80)
print("TOP 10 DESCRIPTORS")
print("=" * 80)

display(
    top10_df.round(6)
)


# ============================================================
# 37. TOP-10 DESCRIPTOR LIST
# ============================================================

top10_descriptors = (
    top10_df["Descriptor"]
    .tolist()
)

print("\nTop-10 descriptor list:")

for i, descriptor in enumerate(
    top10_descriptors,
    start=1
):

    print(
        f"{i:2d}. {descriptor}"
    )


# ============================================================
# 38. TOP-10 SPEARMAN MATRIX
# ============================================================

top10_spearman_columns = [
    "Yield"
] + top10_descriptors


top10_spearman_matrix = spearman_matrix.loc[
    top10_spearman_columns,
    top10_spearman_columns
]


# ============================================================
# 39. TOP-10 PEARSON MATRIX
# ============================================================

top10_pearson_matrix = pearson_matrix.loc[
    top10_spearman_columns,
    top10_spearman_columns
]


# ============================================================
# 40. SAVE TOP-10 MATRICES
# ============================================================

top10_spearman_matrix.to_excel(
    os.path.join(
        OUTPUT_DIR,
        "MAIN_TOP10_SPEARMAN_Correlation_Matrix.xlsx"
    )
)

top10_pearson_matrix.to_excel(
    os.path.join(
        OUTPUT_DIR,
        "SI_TOP10_PEARSON_Correlation_Matrix.xlsx"
    )
)


# ============================================================
# 41. HEATMAP FUNCTION
# ============================================================

def make_heatmap(
    matrix,
    title,
    filename,
    figsize=(12, 10)
):

    plt.figure(
        figsize=figsize
    )

    sns.heatmap(
        matrix,
        cmap="RdBu_r",
        vmin=-1,
        vmax=1,
        center=0,
        square=True,
        annot=True,
        fmt=".2f",
        linewidths=0.5,
        cbar_kws={
            "label": "Correlation coefficient"
        }
    )

    plt.title(
        title,
        fontsize=15,
        fontweight="bold",
        pad=15
    )

    plt.xticks(
        rotation=70,
        ha="right",
        fontsize=9
    )

    plt.yticks(
        rotation=0,
        fontsize=9
    )

    plt.tight_layout()

    path = os.path.join(
        OUTPUT_DIR,
        filename
    )

    plt.savefig(
        path,
        dpi=600,
        bbox_inches="tight"
    )

    plt.show()

    print(
        "\nSaved:",
        path
    )


# ============================================================
# 42. MAIN MANUSCRIPT:
# TOP-10 SPEARMAN HEATMAP
# ============================================================

make_heatmap(
    top10_spearman_matrix,
    "Spearman Correlation: Yield and Top-10 Ligand Descriptors",
    "MAIN_MANUSCRIPT_TOP10_SPEARMAN_Correlation.png",
    figsize=(12, 10)
)


# ============================================================
# 43. SI:
# TOP-10 PEARSON HEATMAP
# ============================================================

make_heatmap(
    top10_pearson_matrix,
    "Pearson Correlation: Yield and Top-10 Ligand Descriptors",
    "SI_TOP10_PEARSON_Correlation.png",
    figsize=(12, 10)
)


# ============================================================
# 44. SI:
# FULL PEARSON HEATMAP
# ============================================================

make_heatmap(
    pearson_matrix,
    "Full Pearson Correlation Matrix",
    "SI_FULL_PEARSON_Correlation_Matrix.png",
    figsize=(
        max(14, len(pearson_matrix.columns) * 0.35),
        max(12, len(pearson_matrix.columns) * 0.35)
    )
)


# ============================================================
# 45. SI:
# FULL SPEARMAN HEATMAP
# ============================================================

make_heatmap(
    spearman_matrix,
    "Full Spearman Correlation Matrix",
    "SI_FULL_SPEARMAN_Correlation_Matrix.png",
    figsize=(
        max(14, len(spearman_matrix.columns) * 0.35),
        max(12, len(spearman_matrix.columns) * 0.35)
    )
)


# ============================================================
# 46. LEAVE-ONE-LIGAND-OUT ROBUSTNESS ANALYSIS
#
# This is the preferred robustness test.
#
# For each descriptor:
#   remove L7 -> calculate correlation
#   remove L8 -> calculate correlation
#   ...
#   remove L25
#
# Then determine:
#   - minimum correlation
#   - maximum correlation
#   - mean
#   - median
#   - standard deviation
#   - sign changes
# ============================================================

print("\n" + "=" * 80)
print("LEAVE-ONE-LIGAND-OUT ROBUSTNESS ANALYSIS")
print("=" * 80)


loo_results = []


for descriptor in predictor_columns:

    full_r = ranking_df.loc[
        ranking_df["Descriptor"] == descriptor,
        "Pearson_r"
    ].iloc[0]

    full_rho = ranking_df.loc[
        ranking_df["Descriptor"] == descriptor,
        "Spearman_rho"
    ].iloc[0]

    loo_pearson = []
    loo_spearman = []

    for idx in correlation_df.index:

        subset = correlation_df.drop(
            index=idx
        )

        x = subset[descriptor]
        y = subset["Yield"]

        r, _ = pearsonr(
            x,
            y
        )

        rho, _ = spearmanr(
            x,
            y
        )

        loo_pearson.append(r)
        loo_spearman.append(rho)


    loo_pearson = np.array(
        loo_pearson
    )

    loo_spearman = np.array(
        loo_spearman
    )


    # Sign changes relative to full dataset
    pearson_sign_changes = np.sum(
        np.sign(loo_pearson)
        !=
        np.sign(full_r)
    )

    spearman_sign_changes = np.sum(
        np.sign(loo_spearman)
        !=
        np.sign(full_rho)
    )


    loo_results.append({

        "Descriptor": descriptor,

        "Full_Pearson_r": full_r,

        "LOO_Pearson_Min": np.min(
            loo_pearson
        ),

        "LOO_Pearson_Max": np.max(
            loo_pearson
        ),

        "LOO_Pearson_Mean": np.mean(
            loo_pearson
        ),

        "LOO_Pearson_Median": np.median(
            loo_pearson
        ),

        "LOO_Pearson_SD": np.std(
            loo_pearson,
            ddof=1
        ),

        "Pearson_Sign_Changes": pearson_sign_changes,

        "Full_Spearman_rho": full_rho,

        "LOO_Spearman_Min": np.min(
            loo_spearman
        ),

        "LOO_Spearman_Max": np.max(
            loo_spearman
        ),

        "LOO_Spearman_Mean": np.mean(
            loo_spearman
        ),

        "LOO_Spearman_Median": np.median(
            loo_spearman
        ),

        "LOO_Spearman_SD": np.std(
            loo_spearman,
            ddof=1
        ),

        "Spearman_Sign_Changes": spearman_sign_changes
    })


loo_df = pd.DataFrame(
    loo_results
)


# ============================================================
# 47. ROBUSTNESS SCORE
#
# A simple descriptive score:
#
# mean of absolute LOO Pearson and Spearman correlations
#
# This is NOT a new statistical test.
# It is only a stability summary.
# ============================================================

loo_df["LOO_Mean_Abs_Pearson"] = (
    loo_df["LOO_Pearson_Mean"].abs()
)

loo_df["LOO_Mean_Abs_Spearman"] = (
    loo_df["LOO_Spearman_Mean"].abs()
)

loo_df["LOO_Robustness_Mean_Abs"] = (
    loo_df["LOO_Mean_Abs_Pearson"]
    +
    loo_df["LOO_Mean_Abs_Spearman"]
) / 2.0


# ============================================================
# 48. SORT ROBUSTNESS RESULTS
# ============================================================

loo_df = loo_df.sort_values(
    by="LOO_Robustness_Mean_Abs",
    ascending=False
).reset_index(drop=True)


# ============================================================
# 49. SAVE LOO RESULTS
# ============================================================

loo_path = os.path.join(
    OUTPUT_DIR,
    "SI_LEAVE_ONE_LIGAND_OUT_Robustness.xlsx"
)

loo_df.to_excel(
    loo_path,
    index=False
)


print("\nLeave-one-ligand-out results saved:")
print(loo_path)


# ============================================================
# 50. DISPLAY ROBUSTNESS OF TOP-10
# ============================================================

top10_robustness = loo_df[
    loo_df["Descriptor"].isin(
        top10_descriptors
    )
].copy()


print("\n" + "=" * 80)
print("ROBUSTNESS OF TOP-10 DESCRIPTORS")
print("=" * 80)

display(
    top10_robustness.round(5)
)


# ============================================================
# 51. CREATE TOP-10 ROBUSTNESS SUMMARY
# ============================================================

top10_summary = pd.merge(
    top10_df,
    top10_robustness,
    on="Descriptor",
    how="left",
    suffixes=(
        "_Full",
        "_LOO"
    )
)


# ============================================================
# 52. SAVE TOP-10 SUMMARY
# ============================================================

top10_summary_path = os.path.join(
    OUTPUT_DIR,
    "TOP10_Descriptor_Summary_and_Robustness.xlsx"
)

top10_summary.to_excel(
    top10_summary_path,
    index=False
)


# ============================================================
# 53. OPTIONAL:
# FDR CORRECTION
#
# We do NOT use FDR to rank descriptors.
#
# It is reported separately because many descriptors
# are tested simultaneously.
# ============================================================

try:

    from statsmodels.stats.multitest import multipletests

except ImportError:

    !pip -q install statsmodels

    from statsmodels.stats.multitest import multipletests


# Pearson FDR
pearson_fdr = multipletests(
    ranking_df["Pearson_p"],
    method="fdr_bh"
)

ranking_df["Pearson_FDR_q"] = (
    pearson_fdr[1]
)


# Spearman FDR
spearman_fdr = multipletests(
    ranking_df["Spearman_p"],
    method="fdr_bh"
)

ranking_df["Spearman_FDR_q"] = (
    spearman_fdr[1]
)


# ============================================================
# 54. RE-SORT AFTER ADDING FDR
#
# Ranking remains based ONLY on Mean_Abs_Correlation.
# ============================================================

ranking_df = ranking_df.sort_values(
    by="Mean_Abs_Correlation",
    ascending=False
).reset_index(drop=True)

ranking_df["Rank"] = np.arange(
    1,
    len(ranking_df) + 1
)


# ============================================================
# 55. SAVE FINAL RANKING WITH FDR
# ============================================================

ranking_df.to_excel(
    os.path.join(
        OUTPUT_DIR,
        "FULL_Yield_Correlation_Ranking_with_FDR.xlsx"
    ),
    index=False
)


# ============================================================
# 56. UPDATE TOP-10 AFTER FDR ADDITION
# ============================================================

top10_df = ranking_df.head(
    TOP_N
).copy()


# ============================================================
# 57. PRINT FINAL TOP-10 TABLE
# ============================================================

print("\n" + "=" * 80)
print("FINAL TOP-10 — RAW / FULL DATASET")
print("=" * 80)

display(
    top10_df[
        [
            "Rank",
            "Descriptor",
            "Pearson_r",
            "Pearson_p",
            "Pearson_FDR_q",
            "Spearman_rho",
            "Spearman_p",
            "Spearman_FDR_q",
            "Mean_Abs_Correlation"
        ]
    ].round(6)
)


# ============================================================
# 58. IDENTIFY STATISTICALLY NOTABLE DESCRIPTORS
# ============================================================

print("\n" + "=" * 80)
print("DESCRIPTORS WITH SPEARMAN q < 0.05")
print("=" * 80)

sig_spearman = ranking_df[
    ranking_df["Spearman_FDR_q"] < 0.05
]

if len(sig_spearman) > 0:

    display(
        sig_spearman[
            [
                "Rank",
                "Descriptor",
                "Spearman_rho",
                "Spearman_p",
                "Spearman_FDR_q"
            ]
        ].round(6)
    )

else:

    print(
        "No descriptor passes Spearman FDR < 0.05."
    )


# ============================================================
# 59. FINAL EXCEL WORKBOOK
# ============================================================

final_excel = os.path.join(
    OUTPUT_DIR,
    "Pd_gamma_arylation_FINAL_Publication_Correlation.xlsx"
)


# ------------------------------------------------------------
# Raw complete data
# ------------------------------------------------------------

raw_complete_output = master_df.copy()


# ------------------------------------------------------------
# Excluded descriptor table
# ------------------------------------------------------------

excluded_rows = []

for c, n in incomplete_columns:

    excluded_rows.append({

        "Descriptor": c,

        "Reason":
            "Incomplete values; excluded from strict full matrix",

        "Valid_N": n,

        "Expected_N": len(master_df)

    })


for c in constant_columns:

    excluded_rows.append({

        "Descriptor": c,

        "Reason":
            "Constant descriptor",

        "Valid_N":
            numeric_df[c].notna().sum()
            if c in numeric_df.columns
            else 0,

        "Expected_N": len(master_df)

    })


for c in sorted(
    set(EXCLUDED_COLUMNS)
    & set(master_df.columns)
):

    excluded_rows.append({

        "Descriptor": c,

        "Reason":
            "Explicitly excluded by protocol",

        "Valid_N":
            master_df[c].notna().sum(),

        "Expected_N": len(master_df)

    })


excluded_df = pd.DataFrame(
    excluded_rows
)


# ------------------------------------------------------------
# Protocol
# ------------------------------------------------------------

protocol_df = pd.DataFrame({

    "Item": [

        "Project",

        "Dataset",

        "Number of ligands",

        "Yield values",

        "Primary correlation",

        "Secondary correlation",

        "Outlier filtering",

        "IQR filtering",

        "MAD filtering",

        "Scaling",

        "Normalization",

        "Imputation",

        "Constant descriptors",

        "Incomplete descriptors",

        "Duplicate path-count descriptor",

        "NBO descriptors",

        "Main manuscript figure",

        "Supporting Information",

        "Ranking criterion",

        "Robustness analysis"

    ],

    "Description": [

        "Pd γ-arylation",

        "Ligand-dependent descriptor dataset",

        str(len(master_df)),

        "Experimental yields; zero values retained as real observations",

        "Spearman",

        "Pearson",

        "None",

        "None",

        "None",

        "None",

        "None",

        "None",

        "Removed",

        "Excluded from strict complete-case full matrix",

        "Palladacycle_Path_Atom_Count completely excluded",

        "Completely excluded",

        "Top-10 Spearman correlation heatmap",

        "Full Pearson + Full Spearman matrices",

        "Mean absolute correlation = (|Pearson| + |Spearman|) / 2",

        "Leave-one-ligand-out analysis"

    ]

})


# ============================================================
# 60. WRITE FINAL MULTI-SHEET EXCEL
# ============================================================

with pd.ExcelWriter(
    final_excel,
    engine="openpyxl"
) as writer:

    raw_complete_output.to_excel(
        writer,
        sheet_name="Raw_Complete_Data",
        index=False
    )

    correlation_df.to_excel(
        writer,
        sheet_name="Correlation_Data",
        index=False
    )

    pearson_matrix.to_excel(
        writer,
        sheet_name="FULL_Pearson",
        index=True
    )

    spearman_matrix.to_excel(
        writer,
        sheet_name="FULL_Spearman",
        index=True
    )

    ranking_df.to_excel(
        writer,
        sheet_name="Yield_Ranking",
        index=False
    )

    top10_df.to_excel(
        writer,
        sheet_name="TOP10",
        index=False
    )

    top10_spearman_matrix.to_excel(
        writer,
        sheet_name="TOP10_Spearman",
        index=True
    )

    top10_pearson_matrix.to_excel(
        writer,
        sheet_name="TOP10_Pearson",
        index=True
    )

    loo_df.to_excel(
        writer,
        sheet_name="LOO_Robustness",
        index=False
    )

    top10_summary.to_excel(
        writer,
        sheet_name="TOP10_Robustness",
        index=False
    )

    excluded_df.to_excel(
        writer,
        sheet_name="Excluded_Descriptors",
        index=False
    )

    protocol_df.to_excel(
        writer,
        sheet_name="Analysis_Protocol",
        index=False
    )


# ============================================================
# 61. FINAL SUMMARY
# ============================================================

print("\n\n")
print("=" * 80)
print("ANALYSIS COMPLETE")
print("=" * 80)

print("\nOutput directory:")
print(OUTPUT_DIR)

print("\nFinal Excel:")
print(final_excel)

print("\nMain manuscript figure:")
print(
    os.path.join(
        OUTPUT_DIR,
        "MAIN_MANUSCRIPT_TOP10_SPEARMAN_Correlation.png"
    )
)

print("\nSupporting Information figures:")

print(
    os.path.join(
        OUTPUT_DIR,
        "SI_TOP10_PEARSON_Correlation.png"
    )
)

print(
    os.path.join(
        OUTPUT_DIR,
        "SI_FULL_PEARSON_Correlation_Matrix.png"
    )
)

print(
    os.path.join(
        OUTPUT_DIR,
        "SI_FULL_SPEARMAN_Correlation_Matrix.png"
    )

)

print("\nRobustness:")
print(
    os.path.join(
        OUTPUT_DIR,
        "SI_LEAVE_ONE_LIGAND_OUT_Robustness.xlsx"
    )
)

print("\n" + "=" * 80)
print("FINAL TOP-10")
print("=" * 80)

for _, row in top10_df.iterrows():

    print(
        f"{int(row['Rank']):2d}. "
        f"{row['Descriptor']} | "
        f"Pearson r = {row['Pearson_r']:.3f} | "
        f"Spearman ρ = {row['Spearman_rho']:.3f} | "
        f"Mean |r| = {row['Mean_Abs_Correlation']:.3f}"
    )

print("\n" + "=" * 80)
print("PROTOCOL CHECK")
print("=" * 80)

print("✓ RAW / FULL dataset")
print("✓ No IQR filtering")
print("✓ No MAD filtering")
print("✓ No outlier removal")
print("✓ No scaling")
print("✓ No normalization")
print("✓ No imputation")
print("✓ Constant descriptors removed only")
print("✓ Incomplete descriptors excluded from strict full matrix")
print("✓ Palladacycle_Path_Atom_Count excluded")
print("✓ NBO descriptors excluded")
print("✓ Sterimol retained when available")
print("✓ %Vbur retained when available")
print("✓ Distal Volume retained when available")
print("✓ P_int retained when available")
print("✓ Full Pearson matrix")
print("✓ Full Spearman matrix")
print("✓ Top-10 ranking")
print("✓ Top-10 Spearman heatmap")
print("✓ Leave-one-ligand-out robustness")

print("\nDONE.")
```
